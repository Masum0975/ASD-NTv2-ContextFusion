```python
# ======================================================================================
# ASD-ContextFusion — BOOTSTRAP RECOVERY
# GIT-FREE EXACT DHINDSA 2025 REPOSITORY DOWNLOAD
#
# Use this because Git is not installed on Windows.
#
# What this cell does:
#   1. Finds repository default branch through GitHub API
#   2. Retrieves exact latest commit SHA
#   3. Downloads THAT EXACT COMMIT as ZIP
#   4. Extracts to E:\ASD_ContextFusion_Q1
#   5. Verifies monoallelic ASD files
#   6. Locks the exact 190 ASD genes
#   7. Finishes project manifest/config
#
# No Git required.
# ======================================================================================

from pathlib import Path
import os
import json
import hashlib
import shutil
import urllib.request
import zipfile
import textwrap
import pandas as pd


# ======================================================================================
# 0. PROJECT PATHS
# ======================================================================================

ROOT = Path(r"E:\ASD_ContextFusion_Q1")

if not ROOT.exists():
    raise RuntimeError(
        r"E:\ASD_ContextFusion_Q1 does not exist."
    )

os.chdir(ROOT)


ADMIN_DIR = ROOT / "00_admin"

RAW_DHINDSA_DIR = (
    ROOT /
    "01_raw" /
    "dhindsa_2025"
)

NT_DIR = (
    ROOT /
    "03_features" /
    "nt_v2"
)

NOTEBOOK_DIR = (
    ROOT /
    "notebooks"
)

for d in [
    ADMIN_DIR,
    RAW_DHINDSA_DIR,
    NT_DIR,
    NOTEBOOK_DIR
]:
    d.mkdir(
        parents=True,
        exist_ok=True
    )


print("=" * 110)
print("ASD-CONTEXTFUSION — GIT-FREE BOOTSTRAP RECOVERY")
print("=" * 110)

print("\nProject root:")
print(ROOT.resolve())


# ======================================================================================
# 1. GITHUB REPOSITORY INFO
# ======================================================================================

OWNER = "bweido"
REPO_NAME = "Mantis-ml-NDD"

REPO_API = (
    f"https://api.github.com/repos/"
    f"{OWNER}/{REPO_NAME}"
)


def fetch_json(url):

    request = urllib.request.Request(
        url,
        headers={
            "User-Agent":
                "ASD-ContextFusion-Jupyter"
        }
    )

    with urllib.request.urlopen(
        request,
        timeout=60
    ) as response:

        return json.loads(
            response.read().decode(
                "utf-8"
            )
        )


print("\n[1/8] QUERYING GITHUB")
print("-" * 110)


repo_info = fetch_json(
    REPO_API
)


default_branch = (
    repo_info[
        "default_branch"
    ]
)


print(
    "Repository     :",
    repo_info[
        "full_name"
    ]
)

print(
    "Default branch :",
    default_branch
)


# ======================================================================================
# 2. LOCK EXACT COMMIT SHA
# ======================================================================================

commit_info = fetch_json(
    f"{REPO_API}/commits/{default_branch}"
)


COMMIT_SHA = (
    commit_info[
        "sha"
    ]
)


print("\n[2/8] EXACT REPOSITORY VERSION")
print("-" * 110)

print(
    "Commit SHA:"
)

print(
    COMMIT_SHA
)


# ======================================================================================
# 3. DOWNLOAD EXACT COMMIT ZIP
# ======================================================================================

ZIP_FILE = (
    RAW_DHINDSA_DIR /
    f"Mantis-ml-NDD_{COMMIT_SHA[:12]}.zip"
)


DOWNLOAD_URL = (
    f"https://github.com/"
    f"{OWNER}/{REPO_NAME}/archive/"
    f"{COMMIT_SHA}.zip"
)


print("\n[3/8] DOWNLOADING EXACT COMMIT")
print("-" * 110)


if not ZIP_FILE.exists():

    request = urllib.request.Request(
        DOWNLOAD_URL,
        headers={
            "User-Agent":
                "ASD-ContextFusion-Jupyter"
        }
    )

    with urllib.request.urlopen(
        request,
        timeout=180
    ) as response:

        data = response.read()


    ZIP_FILE.write_bytes(
        data
    )


    print(
        "Downloaded ✅"
    )

else:

    print(
        "ZIP already exists ✅"
    )


print(
    "ZIP size:",
    f"{ZIP_FILE.stat().st_size / 1024 / 1024:.2f} MB"
)


# ======================================================================================
# 4. EXTRACT REPOSITORY
# ======================================================================================

FINAL_REPO = (
    RAW_DHINDSA_DIR /
    "Mantis-ml-NDD"
)


EXPECTED_SEED_RELATIVE = (
    Path(
        "Input_Files"
    )
    /
    "asd_monoallelic.tsv"
)


if (
    FINAL_REPO /
    EXPECTED_SEED_RELATIVE
).exists():

    print("\n[4/8] REPOSITORY EXTRACTION")
    print("-" * 110)

    print(
        "Repository already extracted ✅"
    )

else:

    # Remove incomplete previous directory only
    if FINAL_REPO.exists():

        shutil.rmtree(
            FINAL_REPO
        )


    TEMP_EXTRACT = (
        RAW_DHINDSA_DIR /
        "_temp_extract"
    )


    if TEMP_EXTRACT.exists():

        shutil.rmtree(
            TEMP_EXTRACT
        )


    TEMP_EXTRACT.mkdir(
        parents=True,
        exist_ok=True
    )


    with zipfile.ZipFile(
        ZIP_FILE,
        "r"
    ) as zf:

        zf.extractall(
            TEMP_EXTRACT
        )


    extracted_dirs = [
        p
        for p in TEMP_EXTRACT.iterdir()
        if p.is_dir()
    ]


    if len(
        extracted_dirs
    ) != 1:

        raise RuntimeError(
            "Unexpected ZIP structure."
        )


    shutil.move(
        str(
            extracted_dirs[0]
        ),
        str(
            FINAL_REPO
        )
    )


    shutil.rmtree(
        TEMP_EXTRACT
    )


    print("\n[4/8] REPOSITORY EXTRACTION")
    print("-" * 110)

    print(
        "Exact repository commit extracted ✅"
    )


print(
    "Repository path:"
)

print(
    FINAL_REPO
)


# ======================================================================================
# 5. VERIFY OFFICIAL ASD FILES
# ======================================================================================

ASD_SEED_FILE = (
    FINAL_REPO /
    "Input_Files" /
    "asd_monoallelic.tsv"
)


ASD_CONFIG_FILE = (
    FINAL_REPO /
    "mantis_ml" /
    "conf" /
    "asd_monoallelic_config.yaml"
)


assert ASD_SEED_FILE.exists(), (
    f"Missing ASD seed file:\n"
    f"{ASD_SEED_FILE}"
)


assert ASD_CONFIG_FILE.exists(), (
    f"Missing ASD config:\n"
    f"{ASD_CONFIG_FILE}"
)


print("\n[5/8] OFFICIAL ASD INPUTS")
print("-" * 110)

print(
    "Seed file   :",
    ASD_SEED_FILE
)

print(
    "Config file :",
    ASD_CONFIG_FILE
)


# ======================================================================================
# 6. LOCK EXACT 190 MONOALLELIC ASD GENES
# ======================================================================================

seed_raw = pd.read_csv(
    ASD_SEED_FILE,
    sep="\t",
    header=None,
    comment="#"
)


# Use first column as official gene column
values = (
    seed_raw.iloc[:, 0]
    .dropna()
    .astype(str)
    .str.strip()
)


HEADER_LIKE = {
    "",
    "gene",
    "Gene",
    "Gene_Name",
    "gene_name",
    "symbol",
    "Symbol"
}


asd_genes = sorted(
    set(
        gene
        for gene in values
        if gene not in HEADER_LIKE
    )
)


print("\n[6/8] ASD SEED AUDIT")
print("-" * 110)

print(
    "Detected ASD genes:",
    len(
        asd_genes
    )
)

print(
    "Unique ASD genes  :",
    len(
        set(
            asd_genes
        )
    )
)

print(
    "\nFirst 15 genes:"
)

print(
    asd_genes[:15]
)


if len(
    asd_genes
) != 190:

    print(
        "\nSeed file preview:"
    )

    print(
        seed_raw.head(
            20
        )
    )

    raise RuntimeError(
        f"Expected exactly 190 monoallelic ASD genes, "
        f"but detected {len(asd_genes)}.\n"
        "STOPPING — benchmark will not be silently altered."
    )


print(
    "\n190/190 MONOALLELIC ASD GENES VERIFIED ✅"
)


# ======================================================================================
# 7. SAVE LOCKED SEEDS + MANIFEST
# ======================================================================================

LOCKED_SEED_FILE = (
    ADMIN_DIR /
    "dhindsa_2025_monoallelic_ASD_190.tsv"
)


pd.DataFrame(
    {
        "gene":
            asd_genes
    }
).to_csv(
    LOCKED_SEED_FILE,
    sep="\t",
    index=False
)


SEED_SHA256 = hashlib.sha256(
    LOCKED_SEED_FILE.read_bytes()
).hexdigest()


ZIP_SHA256 = hashlib.sha256(
    ZIP_FILE.read_bytes()
).hexdigest()


BENCHMARK_MANIFEST = {

    "benchmark":
        "Dhindsa_et_al_2025_monoallelic_ASD",

    "repository":
        f"https://github.com/{OWNER}/{REPO_NAME}",

    "acquisition_method":
        "GitHub exact-commit ZIP; no Git client",

    "default_branch":
        default_branch,

    "locked_commit_sha":
        COMMIT_SHA,

    "repository_zip_sha256":
        ZIP_SHA256,

    "asd_seed_count":
        190,

    "asd_seed_sha256":
        SEED_SHA256,

    "asd_seed_file":
        str(
            ASD_SEED_FILE.resolve()
        ),

    "asd_config_file":
        str(
            ASD_CONFIG_FILE.resolve()
        ),

    "local_repository":
        str(
            FINAL_REPO.resolve()
        )
}


MANIFEST_FILE = (
    ADMIN_DIR /
    "dhindsa_2025_benchmark_manifest.json"
)


with open(
    MANIFEST_FILE,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        BENCHMARK_MANIFEST,
        f,
        indent=2
    )


print("\n[7/8] REPRODUCIBILITY LOCK")
print("-" * 110)

print(
    "Repository commit:"
)

print(
    COMMIT_SHA
)

print(
    "\nRepository ZIP SHA256:"
)

print(
    ZIP_SHA256
)

print(
    "\nASD seed SHA256:"
)

print(
    SEED_SHA256
)


# ======================================================================================
# 8. FINAL PROJECT STATUS
# ======================================================================================

README_FILE = (
    ROOT /
    "README_START_HERE.txt"
)


README_FILE.write_text(
    textwrap.dedent(
        f"""
        ASD-ContextFusion — Q1 Project
        =================================

        Project root:
        {ROOT}

        Benchmark:
        Dhindsa et al. 2025 monoallelic ASD

        Exact repository commit:
        {COMMIT_SHA}

        ASD seed genes:
        190

        ACTIVE ANALYSIS STRATEGY
        ------------------------
        1. Exact Dhindsa benchmark/context backbone
        2. Developmental-expression modality
        3. Genetic/context modality
        4. GENCODE v50 transcript modality
        5. NT-v2 512D contextual nucleotide embeddings
        6. Length+GC and k-mer sequence controls
        7. Cross-fitted modality-preserving late fusion
        8. Leakage-sensitive robustness analysis
        9. Dhindsa-style confirmatory PU evaluation
        10. Temporal/external ASD validation

        IMPORTANT
        ---------
        Raw Context+NT feature concatenation is not the proposed model.
        Model selection must occur only inside training/inner CV.
        Final held-out evaluation cannot be used for model tuning.
        """
    ),
    encoding="utf-8"
)


print("\n[8/8] PROJECT STATUS")
print("-" * 110)

print(
    "Project root      :",
    ROOT
)

print(
    "Repository        :",
    FINAL_REPO
)

print(
    "ASD seeds         : 190"
)

print(
    "Benchmark manifest:",
    MANIFEST_FILE
)

print(
    "README            :",
    README_FILE
)


print("\nNext NT feature location:")
print(
    NT_DIR
)


print("\n" + "=" * 110)
print("BOOTSTRAP RECOVERY COMPLETE ✅")
print("GIT IS NOT REQUIRED ✅")
print("DHINDSA 2025 EXACT BENCHMARK LOCKED ✅")
print("=" * 110)
```

```python
# ======================================================================================
# ASD-ContextFusion
# WINDOWS PATH-LENGTH RECOVERY
#
# Uses the already-downloaded Dhindsa ZIP.
# Does NOT re-download anything.
#
# Extracts ONLY scientifically required repository components:
#   - Input_Files/
#   - mantis_ml/
#   - root config/documentation files
#
# Skips:
#   - misc/
#   - long historical analysis paths
#   - irrelevant repository artifacts
# ======================================================================================

from pathlib import Path, PurePosixPath
import os
import shutil
import zipfile
import hashlib
import json
import pandas as pd


# ======================================================================================
# 0. PATHS
# ======================================================================================

ROOT = Path(r"E:\ASD_ContextFusion_Q1")

RAW_DIR = (
    ROOT /
    "01_raw" /
    "dhindsa_2025"
)

ADMIN_DIR = (
    ROOT /
    "00_admin"
)

FINAL_REPO = (
    RAW_DIR /
    "Mantis-ml-NDD"
)

ADMIN_DIR.mkdir(
    parents=True,
    exist_ok=True
)

FINAL_REPO.mkdir(
    parents=True,
    exist_ok=True
)


print("=" * 110)
print("ASD-CONTEXTFUSION — WINDOWS SELECTIVE EXTRACTION RECOVERY")
print("=" * 110)


# ======================================================================================
# 1. FIND ALREADY DOWNLOADED ZIP
# ======================================================================================

zip_candidates = sorted(
    RAW_DIR.glob(
        "Mantis-ml-NDD_*.zip"
    ),
    key=lambda p: p.stat().st_mtime,
    reverse=True
)


if not zip_candidates:

    raise RuntimeError(
        "Previously downloaded Dhindsa ZIP was not found."
    )


ZIP_FILE = (
    zip_candidates[0]
)


print("\n[1/7] EXISTING ZIP")
print("-" * 110)

print(
    "ZIP:"
)

print(
    ZIP_FILE
)

print(
    "Size:",
    f"{ZIP_FILE.stat().st_size / 1024 / 1024:.2f} MB"
)


# ======================================================================================
# 2. OPTIONAL CLEANUP OF FAILED TEMP EXTRACTION
#
# \\?\ prefix allows Windows long-path deletion.
# ======================================================================================

print("\n[2/7] CLEANING FAILED TEMP EXTRACTION")
print("-" * 110)


OLD_TEMP = (
    RAW_DIR /
    "_temp_extract"
)


def windows_long_path(path):

    absolute = str(
        Path(path).resolve()
    )

    if os.name == "nt":
        return (
            "\\\\?\\" +
            absolute
        )

    return absolute


if OLD_TEMP.exists():

    try:

        shutil.rmtree(
            windows_long_path(
                OLD_TEMP
            )
        )

        print(
            "Old failed temp folder removed ✅"
        )

    except Exception as e:

        print(
            "Could not fully remove old temp folder."
        )

        print(
            "This does NOT block the analysis."
        )

        print(
            "Reason:",
            e
        )

else:

    print(
        "No failed temp folder found ✅"
    )


# ======================================================================================
# 3. DEFINE WHAT WE ACTUALLY NEED
# ======================================================================================

KEEP_PREFIXES = [

    "Input_Files/",
    "mantis_ml/",
]


KEEP_ROOT_FILES = {

    "README.md",
    "LICENSE",
    "setup.py",
    "setup.cfg",
    "requirements.txt",
    "environment.yml",
    "environment.yaml",
    "pyproject.toml",
    "MANIFEST.in",
}


print("\n[3/7] SELECTIVE EXTRACTION")
print("-" * 110)

print(
    "Keeping:"
)

for x in KEEP_PREFIXES:
    print(" -", x)

print(
    "\nSkipping misc/ and unrelated long-path analyses ✅"
)


# ======================================================================================
# 4. EXTRACT ONLY REQUIRED FILES
# ======================================================================================

n_files = 0
n_skipped = 0


with zipfile.ZipFile(
    ZIP_FILE,
    "r"
) as zf:


    members = zf.infolist()


    if not members:

        raise RuntimeError(
            "ZIP appears empty."
        )


    # GitHub ZIP top-level directory:
    # Mantis-ml-NDD-<commit>/
    first_parts = PurePosixPath(
        members[0].filename
    ).parts


    if not first_parts:

        raise RuntimeError(
            "Unexpected ZIP structure."
        )


    archive_root = (
        first_parts[0]
    )


    print(
        "Archive root:",
        archive_root
    )


    for info in members:


        if info.is_dir():
            continue


        posix_path = PurePosixPath(
            info.filename
        )


        parts = (
            posix_path.parts
        )


        if (
            len(parts) < 2
            or
            parts[0] != archive_root
        ):
            continue


        relative = PurePosixPath(
            *parts[1:]
        )


        relative_str = (
            relative.as_posix()
        )


        keep = False


        # --------------------------------------------------------------
        # Important folders
        # --------------------------------------------------------------

        if any(
            relative_str.startswith(
                prefix
            )
            for prefix
            in KEEP_PREFIXES
        ):

            keep = True


        # --------------------------------------------------------------
        # Useful root files
        # --------------------------------------------------------------

        elif (
            len(relative.parts) == 1
            and
            relative.name
            in KEEP_ROOT_FILES
        ):

            keep = True


        if not keep:

            n_skipped += 1
            continue


        target = (
            FINAL_REPO /
            Path(
                *relative.parts
            )
        )


        target.parent.mkdir(
            parents=True,
            exist_ok=True
        )


        # Stream instead of extractall()
        with zf.open(
            info,
            "r"
        ) as src:

            with open(
                target,
                "wb"
            ) as dst:

                shutil.copyfileobj(
                    src,
                    dst
                )


        n_files += 1


print(
    "\nFiles extracted:",
    n_files
)

print(
    "Unnecessary files skipped:",
    n_skipped
)


# ======================================================================================
# 5. VERIFY CRITICAL DHINDSA FILES
# ======================================================================================

print("\n[4/7] VERIFYING DHINDSA ASD INPUTS")
print("-" * 110)


ASD_SEED_FILE = (
    FINAL_REPO /
    "Input_Files" /
    "asd_monoallelic.tsv"
)


ASD_CONFIG_FILE = (
    FINAL_REPO /
    "mantis_ml" /
    "conf" /
    "asd_monoallelic_config.yaml"
)


assert ASD_SEED_FILE.exists(), (
    f"Missing ASD seed file:\n"
    f"{ASD_SEED_FILE}"
)


assert ASD_CONFIG_FILE.exists(), (
    f"Missing ASD config:\n"
    f"{ASD_CONFIG_FILE}"
)


print(
    "ASD seed file ✅"
)

print(
    ASD_SEED_FILE
)


print(
    "\nASD configuration ✅"
)

print(
    ASD_CONFIG_FILE
)


# ======================================================================================
# 6. READ EXACT 190 ASD SEEDS
# ======================================================================================

print("\n[5/7] ASD SEED LOCK")
print("-" * 110)


seed_df = pd.read_csv(
    ASD_SEED_FILE,
    sep="\t",
    header=None,
    comment="#"
)


values = (
    seed_df.iloc[:, 0]
    .dropna()
    .astype(str)
    .str.strip()
)


HEADER_LIKE = {

    "",
    "gene",
    "Gene",
    "Gene_Name",
    "gene_name",
    "symbol",
    "Symbol"
}


asd_genes = sorted(
    {
        x
        for x in values
        if x not in HEADER_LIKE
    }
)


print(
    "Detected monoallelic ASD genes:",
    len(
        asd_genes
    )
)


print(
    "Unique genes:",
    len(
        set(
            asd_genes
        )
    )
)


print(
    "First 15:"
)

print(
    asd_genes[:15]
)


assert len(
    asd_genes
) == 190, (
    f"Expected 190 ASD genes, "
    f"found {len(asd_genes)}"
)


print(
    "\n190/190 MONOALLELIC ASD GENES VERIFIED ✅"
)


# ======================================================================================
# 7. SAVE LOCKED SEED LIST
# ======================================================================================

LOCKED_SEED_FILE = (
    ADMIN_DIR /
    "dhindsa_2025_monoallelic_ASD_190.tsv"
)


pd.DataFrame(
    {
        "gene":
            asd_genes
    }
).to_csv(
    LOCKED_SEED_FILE,
    sep="\t",
    index=False
)


seed_sha = hashlib.sha256(
    LOCKED_SEED_FILE.read_bytes()
).hexdigest()


zip_sha = hashlib.sha256(
    ZIP_FILE.read_bytes()
).hexdigest()


print("\n[6/7] REPRODUCIBILITY HASHES")
print("-" * 110)

print(
    "Seed SHA256:"
)

print(
    seed_sha
)


print(
    "\nRepository ZIP SHA256:"
)

print(
    zip_sha
)


# ======================================================================================
# 8. SAVE RECOVERY MANIFEST
# ======================================================================================

manifest = {

    "project":
        "ASD-ContextFusion",

    "benchmark":
        "Dhindsa et al. 2025 monoallelic ASD",

    "repository_source":
        "bweido/Mantis-ml-NDD",

    "repository_acquisition":
        "exact-commit GitHub ZIP",

    "local_repository":
        str(
            FINAL_REPO.resolve()
        ),

    "selective_extraction":
        True,

    "kept_repository_components":
        KEEP_PREFIXES,

    "excluded_repository_component":
        "misc and unrelated long-path analysis directories",

    "asd_seed_count":
        190,

    "asd_seed_sha256":
        seed_sha,

    "repository_zip_sha256":
        zip_sha
}


MANIFEST_FILE = (
    ADMIN_DIR /
    "dhindsa_selective_extraction_manifest.json"
)


with open(
    MANIFEST_FILE,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        manifest,
        f,
        indent=2
    )


print("\n[7/7] FINAL STATUS")
print("-" * 110)

print(
    "Repository:"
)

print(
    FINAL_REPO
)

print(
    "\nLocked ASD seed list:"
)

print(
    LOCKED_SEED_FILE
)

print(
    "\nManifest:"
)

print(
    MANIFEST_FILE
)


print("\n" + "=" * 110)
print("SELECTIVE EXTRACTION COMPLETE ✅")
print("WINDOWS PATH-LENGTH PROBLEM BYPASSED ✅")
print("190 ASD BENCHMARK READY ✅")
print("=" * 110)
```

```python
# ======================================================================================
# ASD-ContextFusion
# NOTEBOOK 01 / CELL 1
#
# DHINDSA 2025 — EXACT ASD CONFIG + FEATURE PROVENANCE AUDIT
#
# CATEGORY:
#   BENCHMARK / DATA-PROVENANCE AUDIT
#
# Goals:
#   1. Read exact monoallelic ASD configuration
#   2. Inventory all repository input files
#   3. Detect tabular gene-level feature files
#   4. Measure overlap with the locked 190 ASD genes
#   5. Identify likely feature families
#   6. Flag potentially seed-derived / circular features
#   7. Save a locked audit for later model construction
#
# IMPORTANT:
#   NO model training in this cell.
# ======================================================================================

from pathlib import Path
import sys
import json
import re
import hashlib
import importlib.util
import subprocess
import warnings

import numpy as np
import pandas as pd

warnings.filterwarnings("ignore")


# ======================================================================================
# 0. PATHS
# ======================================================================================

ROOT = Path(r"E:\ASD_ContextFusion_Q1")

ADMIN_DIR = (
    ROOT /
    "00_admin"
)

REPO = (
    ROOT /
    "01_raw" /
    "dhindsa_2025" /
    "Mantis-ml-NDD"
)

INPUT_DIR = (
    REPO /
    "Input_Files"
)

CONFIG_FILE = (
    REPO /
    "mantis_ml" /
    "conf" /
    "asd_monoallelic_config.yaml"
)

LOCKED_SEED_FILE = (
    ADMIN_DIR /
    "dhindsa_2025_monoallelic_ASD_190.tsv"
)

AUDIT_DIR = (
    ROOT /
    "02_interim" /
    "dhindsa_feature_audit"
)

AUDIT_DIR.mkdir(
    parents=True,
    exist_ok=True
)


for path in [
    REPO,
    INPUT_DIR,
    CONFIG_FILE,
    LOCKED_SEED_FILE
]:
    assert path.exists(), f"Missing required path:\n{path}"


print("=" * 115)
print("ASD-CONTEXTFUSION — DHINDSA 2025 FEATURE PROVENANCE AUDIT")
print("=" * 115)


# ======================================================================================
# 1. LOAD LOCKED ASD SEEDS
# ======================================================================================

print("\n[1/9] LOCKED ASD SEEDS")
print("-" * 115)


seed_df = pd.read_csv(
    LOCKED_SEED_FILE,
    sep="\t"
)


assert "gene" in seed_df.columns


asd_genes = (
    seed_df["gene"]
    .astype(str)
    .str.strip()
    .tolist()
)


asd_set = set(
    asd_genes
)


assert len(asd_set) == 190


print(
    "Locked ASD genes:",
    len(asd_set)
)

print(
    "First 15:"
)

print(
    sorted(asd_set)[:15]
)


# ======================================================================================
# 2. PYAML CHECK
# ======================================================================================

if importlib.util.find_spec(
    "yaml"
) is None:

    print(
        "\nInstalling PyYAML..."
    )

    subprocess.run(
        [
            sys.executable,
            "-m",
            "pip",
            "install",
            "pyyaml"
        ],
        check=True
    )


import yaml


# ======================================================================================
# 3. READ EXACT ASD CONFIG
# ======================================================================================

print("\n[2/9] EXACT ASD CONFIGURATION")
print("-" * 115)


config_text = CONFIG_FILE.read_text(
    encoding="utf-8",
    errors="replace"
)


with open(
    CONFIG_FILE,
    "r",
    encoding="utf-8"
) as f:

    config = yaml.safe_load(
        f
    )


print(
    "Config path:"
)

print(
    CONFIG_FILE
)


print(
    "\nConfig SHA256:"
)


config_sha = hashlib.sha256(
    CONFIG_FILE.read_bytes()
).hexdigest()


print(
    config_sha
)


print(
    "\nTop-level config keys:"
)

if isinstance(
    config,
    dict
):

    for key in config.keys():
        print(" -", key)

else:

    print(
        "Config parsed as:",
        type(config)
    )


# Save exact text snapshot
CONFIG_SNAPSHOT = (
    AUDIT_DIR /
    "asd_monoallelic_config_exact_snapshot.yaml"
)

CONFIG_SNAPSHOT.write_text(
    config_text,
    encoding="utf-8"
)


# ======================================================================================
# 4. SEARCH CONFIG FOR FILE / FEATURE REFERENCES
# ======================================================================================

print("\n[3/9] CONFIG FILE REFERENCES")
print("-" * 115)


REFERENCE_SUFFIXES = (
    ".tsv",
    ".csv",
    ".txt",
    ".bed",
    ".gz",
    ".pkl",
    ".pickle",
    ".h5",
    ".hdf5"
)


def flatten_object(
    obj,
    prefix=""
):

    rows = []

    if isinstance(
        obj,
        dict
    ):

        for key, value in obj.items():

            new_prefix = (
                f"{prefix}.{key}"
                if prefix
                else str(key)
            )

            rows.extend(
                flatten_object(
                    value,
                    new_prefix
                )
            )

    elif isinstance(
        obj,
        list
    ):

        for i, value in enumerate(
            obj
        ):

            new_prefix = (
                f"{prefix}[{i}]"
            )

            rows.extend(
                flatten_object(
                    value,
                    new_prefix
                )
            )

    else:

        rows.append(
            {
                "config_key":
                    prefix,

                "value":
                    obj
            }
        )

    return rows


flat_config = pd.DataFrame(
    flatten_object(
        config
    )
)


flat_config[
    "value_string"
] = (
    flat_config[
        "value"
    ]
    .astype(str)
)


file_ref_mask = (
    flat_config[
        "value_string"
    ]
    .str.lower()
    .str.endswith(
        REFERENCE_SUFFIXES
    )
)


config_file_refs = (
    flat_config[
        file_ref_mask
    ]
    .copy()
)


print(
    "Explicit file-like references in config:",
    len(
        config_file_refs
    )
)


if len(
    config_file_refs
) > 0:

    print(
        config_file_refs[
            [
                "config_key",
                "value_string"
            ]
        ]
        .to_string(
            index=False
        )
    )

else:

    print(
        "No simple direct filename values detected."
    )


# ======================================================================================
# 5. INVENTORY INPUT FILES
# ======================================================================================

print("\n[4/9] INPUT FILE INVENTORY")
print("-" * 115)


all_files = sorted(
    [
        p
        for p in INPUT_DIR.rglob("*")
        if p.is_file()
    ]
)


inventory_rows = []


for path in all_files:

    suffixes = "".join(
        path.suffixes
    ).lower()

    inventory_rows.append(
        {
            "relative_path":
                str(
                    path.relative_to(
                        REPO
                    )
                ),

            "filename":
                path.name,

            "suffix":
                suffixes,

            "size_mb":
                path.stat().st_size /
                1024 /
                1024
        }
    )


inventory = pd.DataFrame(
    inventory_rows
)


print(
    "Input files detected:",
    len(inventory)
)


print(
    "\nLargest 30 input files:"
)


print(
    inventory
    .sort_values(
        "size_mb",
        ascending=False
    )
    .head(30)
    .round(
        {
            "size_mb": 3
        }
    )
    .to_string(
        index=False
    )
)


# ======================================================================================
# 6. INSPECT TABULAR FILES
#
# We do not assume filenames.
# We attempt to identify gene-level feature tables empirically.
# ======================================================================================

print("\n[5/9] TABULAR FEATURE DISCOVERY")
print("-" * 115)


TABULAR_SUFFIX_PATTERNS = (
    ".tsv",
    ".csv",
    ".txt",
    ".tsv.gz",
    ".csv.gz",
    ".txt.gz"
)


def is_tabular_file(
    path
):

    name = path.name.lower()

    return any(
        name.endswith(
            suffix
        )
        for suffix
        in TABULAR_SUFFIX_PATTERNS
    )


def try_read_table(
    path,
    max_rows=10000
):

    name = (
        path.name.lower()
    )


    if (
        name.endswith(
            ".tsv"
        )
        or
        name.endswith(
            ".tsv.gz"
        )
    ):

        separators = [
            "\t"
        ]

    elif (
        name.endswith(
            ".csv"
        )
        or
        name.endswith(
            ".csv.gz"
        )
    ):

        separators = [
            ","
        ]

    else:

        separators = [
            "\t",
            ","
        ]


    for sep in separators:

        try:

            df = pd.read_csv(
                path,
                sep=sep,
                nrows=max_rows,
                low_memory=False,
                comment="#"
            )

            if df.shape[1] >= 1:

                return df, sep

        except Exception:

            pass


    return None, None


tabular_rows = []


for path in all_files:

    if not is_tabular_file(
        path
    ):
        continue


    df, sep = try_read_table(
        path
    )


    if df is None:
        continue


    best_overlap = 0
    best_column = None


    for column in df.columns:

        try:

            values = set(
                df[
                    column
                ]
                .dropna()
                .astype(str)
                .str.strip()
            )

        except Exception:

            continue


        overlap = len(
            values &
            asd_set
        )


        if overlap > best_overlap:

            best_overlap = overlap
            best_column = str(
                column
            )


    tabular_rows.append(
        {
            "relative_path":
                str(
                    path.relative_to(
                        REPO
                    )
                ),

            "filename":
                path.name,

            "sample_rows_read":
                len(df),

            "n_columns":
                df.shape[1],

            "separator":
                "\\t"
                if sep == "\t"
                else sep,

            "best_gene_column":
                best_column,

            "asd_overlap_sample":
                best_overlap
        }
    )


tabular_audit = pd.DataFrame(
    tabular_rows
)


if len(
    tabular_audit
) == 0:

    raise RuntimeError(
        "No readable tabular files detected."
    )


print(
    "Readable tabular files:",
    len(
        tabular_audit
    )
)


print(
    "\nTop files by ASD-gene overlap:"
)


print(
    tabular_audit
    .sort_values(
        [
            "asd_overlap_sample",
            "n_columns"
        ],
        ascending=[
            False,
            False
        ]
    )
    .head(40)
    .to_string(
        index=False
    )
)


# ======================================================================================
# 7. IDENTIFY HIGH-VALUE GENE-LEVEL TABLE CANDIDATES
# ======================================================================================

print("\n[6/9] HIGH-VALUE FEATURE TABLE CANDIDATES")
print("-" * 115)


high_value = (
    tabular_audit[
        (
            tabular_audit[
                "asd_overlap_sample"
            ]
            >= 100
        )
        |
        (
            tabular_audit[
                "n_columns"
            ]
            >= 20
        )
    ]
    .copy()
)


high_value = (
    high_value
    .sort_values(
        [
            "asd_overlap_sample",
            "n_columns"
        ],
        ascending=[
            False,
            False
        ]
    )
)


print(
    "Candidate tables:",
    len(
        high_value
    )
)


if len(
    high_value
) > 0:

    print(
        high_value
        .to_string(
            index=False
        )
    )


# ======================================================================================
# 8. FEATURE-FAMILY / LEAKAGE KEYWORD AUDIT
#
# This is column-name screening only.
# It does NOT yet decide final exclusion.
# ======================================================================================

print("\n[7/9] FEATURE FAMILY / LEAKAGE SCREEN")
print("-" * 115)


SAFE_FAMILY_PATTERNS = {

    "constraint_variation": [

        r"gnomad",
        r"exac",
        r"loeuf",
        r"\bpli\b",
        r"rvis",
        r"gevir",
        r"constraint",
        r"intolerance",
        r"missense",
        r"lof",
        r"mutation"
    ],


    "developmental_expression": [

        r"single.?cell",
        r"cortex",
        r"fetal",
        r"brain",
        r"expression",
        r"protein.?atlas",
        r"hpa",
        r"gtex"
    ],


    "gene_properties": [

        r"gene.?size",
        r"gene.?length",
        r"gc",
        r"cds",
        r"essential"
    ]
}


RISK_PATTERNS = {

    "label_or_seed": [

        r"label",
        r"known.?gene",
        r"seed",
        r"target",
        r"asd",
        r"autism"
    ],


    "ontology_pathway": [

        r"gene.?ontology",
        r"(^|_)go(_|$)",
        r"msigdb",
        r"pathway",
        r"hpo"
    ],


    "network_seed_related": [

        r"ppi",
        r"inweb",
        r"network",
        r"interactor",
        r"neighbor"
    ],


    "disease_annotation": [

        r"disease",
        r"gwas",
        r"mgi"
    ]
}


def matches_patterns(
    text,
    patterns
):

    text = str(
        text
    ).lower()

    return any(
        re.search(
            pattern,
            text
        )
        is not None
        for pattern in patterns
    )


feature_rows = []


for row in high_value.itertuples(
    index=False
):

    path = (
        REPO /
        row.relative_path
    )


    df, _ = try_read_table(
        path,
        max_rows=1000
    )


    if df is None:
        continue


    for column in df.columns:

        safe_groups = [

            family
            for family, patterns
            in SAFE_FAMILY_PATTERNS.items()
            if matches_patterns(
                column,
                patterns
            )
        ]


        risk_groups = [

            family
            for family, patterns
            in RISK_PATTERNS.items()
            if matches_patterns(
                column,
                patterns
            )
        ]


        feature_rows.append(
            {
                "source_file":
                    row.relative_path,

                "feature":
                    str(
                        column
                    ),

                "safe_family_candidates":
                    ";".join(
                        safe_groups
                    ),

                "risk_flags":
                    ";".join(
                        risk_groups
                    )
            }
        )


feature_audit = pd.DataFrame(
    feature_rows
)


print(
    "Feature-column records audited:",
    len(
        feature_audit
    )
)


risk_features = (
    feature_audit[
        feature_audit[
            "risk_flags"
        ]
        .astype(str)
        .str.len()
        > 0
    ]
)


print(
    "Columns carrying review flags:",
    len(
        risk_features
    )
)


if len(
    risk_features
) > 0:

    print(
        "\nFirst 60 flagged columns:"
    )

    print(
        risk_features
        .head(60)
        .to_string(
            index=False
        )
    )


# ======================================================================================
# 9. SEARCH SOURCE CODE FOR ASD CONFIG USAGE
#
# Helps identify preprocessing entrypoints.
# ======================================================================================

print("\n[8/9] ASD CONFIG CODE USAGE")
print("-" * 115)


code_files = list(
    (
        REPO /
        "mantis_ml"
    ).rglob(
        "*.py"
    )
)


usage_rows = []


search_terms = [

    "asd_monoallelic",
    "balanced",
    "unlabeled",
    "feature_table",
    "processed_feature",
    "complete_feature"
]


for code_file in code_files:

    try:

        text = code_file.read_text(
            encoding="utf-8",
            errors="replace"
        )

    except Exception:

        continue


    lower = text.lower()


    hits = [

        term
        for term in search_terms
        if term.lower()
        in lower
    ]


    if hits:

        usage_rows.append(
            {
                "file":
                    str(
                        code_file.relative_to(
                            REPO
                        )
                    ),

                "matched_terms":
                    ";".join(
                        hits
                    )
            }
        )


code_usage = pd.DataFrame(
    usage_rows
)


print(
    "Relevant source files:",
    len(
        code_usage
    )
)


if len(
    code_usage
) > 0:

    print(
        code_usage
        .to_string(
            index=False
        )
    )


# ======================================================================================
# 10. SAVE EVERYTHING
# ======================================================================================

print("\n[9/9] SAVING AUDIT")
print("-" * 115)


INVENTORY_FILE = (
    AUDIT_DIR /
    "repository_input_file_inventory.tsv"
)

TABULAR_FILE = (
    AUDIT_DIR /
    "tabular_input_audit.tsv"
)

HIGH_VALUE_FILE = (
    AUDIT_DIR /
    "high_value_feature_table_candidates.tsv"
)

FEATURE_AUDIT_FILE = (
    AUDIT_DIR /
    "feature_family_and_leakage_screen.tsv"
)

CONFIG_FLAT_FILE = (
    AUDIT_DIR /
    "asd_config_flattened.tsv"
)

CODE_USAGE_FILE = (
    AUDIT_DIR /
    "preprocessing_code_usage_audit.tsv"
)


inventory.to_csv(
    INVENTORY_FILE,
    sep="\t",
    index=False
)


tabular_audit.to_csv(
    TABULAR_FILE,
    sep="\t",
    index=False
)


high_value.to_csv(
    HIGH_VALUE_FILE,
    sep="\t",
    index=False
)


feature_audit.to_csv(
    FEATURE_AUDIT_FILE,
    sep="\t",
    index=False
)


flat_config.to_csv(
    CONFIG_FLAT_FILE,
    sep="\t",
    index=False
)


code_usage.to_csv(
    CODE_USAGE_FILE,
    sep="\t",
    index=False
)


# ======================================================================================
# FINAL REPORT
# ======================================================================================

print("\n" + "=" * 115)
print("DHINDSA FEATURE-PROVENANCE AUDIT COMPLETE ✅")
print("=" * 115)


print(
    "\nLocked ASD genes       :",
    len(
        asd_set
    )
)


print(
    "Repository input files:",
    len(
        inventory
    )
)


print(
    "Readable tables       :",
    len(
        tabular_audit
    )
)


print(
    "High-value candidates :",
    len(
        high_value
    )
)


print(
    "Audited feature cols  :",
    len(
        feature_audit
    )
)


print(
    "Risk-flagged columns  :",
    len(
        risk_features
    )
)


print(
    "\nSaved audit directory:"
)

print(
    AUDIT_DIR
)


print("\nNEXT DECISION:")
print(
    "Use this audit to identify the authors' exact compiled/processed "
    "gene-level feature table and split it into:"
)

print(
    "  A. Developmental-expression modality"
)

print(
    "  B. Genetic/constraint modality"
)

print(
    "  C. Full Dhindsa benchmark modality"
)

print(
    "  D. Leakage-risk / sensitivity-only features"
)

print("=" * 115)
```

```python
# ======================================================================================
# ASD-ContextFusion
# NOTEBOOK 01 / CELL 1B
#
# COMPLETE DHINDSA REPOSITORY RECOVERY
# Extract EVERYTHING scientifically relevant from the exact downloaded commit,
# while skipping only misc/ (the Windows long-path problem source).
#
# Then:
#   - verify main.nf
#   - inventory repository
#   - search preprocessing code for generated feature tables
#   - search download/data-source URLs
#   - identify feature-compilation entrypoints
#
# NO MODEL TRAINING.
# ======================================================================================

from pathlib import Path, PurePosixPath
import os
import re
import json
import shutil
import zipfile
import hashlib
import pandas as pd


# ======================================================================================
# 0. PATHS
# ======================================================================================

ROOT = Path(r"E:\ASD_ContextFusion_Q1")

RAW_DIR = (
    ROOT /
    "01_raw" /
    "dhindsa_2025"
)

REPO = (
    RAW_DIR /
    "Mantis-ml-NDD"
)

ADMIN_DIR = (
    ROOT /
    "00_admin"
)

AUDIT_DIR = (
    ROOT /
    "02_interim" /
    "dhindsa_repository_audit"
)

for d in [
    REPO,
    ADMIN_DIR,
    AUDIT_DIR
]:
    d.mkdir(
        parents=True,
        exist_ok=True
    )


print("=" * 115)
print("ASD-CONTEXTFUSION — COMPLETE DHINDSA REPOSITORY RECOVERY")
print("=" * 115)


# ======================================================================================
# 1. FIND EXACT DOWNLOADED ZIP
# ======================================================================================

zip_candidates = sorted(
    RAW_DIR.glob(
        "Mantis-ml-NDD_*.zip"
    ),
    key=lambda p:
        p.stat().st_mtime,
    reverse=True
)


if not zip_candidates:

    raise RuntimeError(
        "Exact Dhindsa ZIP is missing."
    )


ZIP_FILE = zip_candidates[0]


print("\n[1/9] EXACT REPOSITORY ZIP")
print("-" * 115)

print("ZIP:")
print(ZIP_FILE)

print(
    "Size:",
    f"{ZIP_FILE.stat().st_size / 1024 / 1024:.2f} MB"
)


ZIP_SHA256 = hashlib.sha256(
    ZIP_FILE.read_bytes()
).hexdigest()


print("\nSHA256:")
print(ZIP_SHA256)


# ======================================================================================
# 2. EXTRACT ALL NON-MISC REPOSITORY CONTENT
#
# This fixes our earlier overly-selective extraction.
#
# Skip ONLY:
#   misc/
#
# Everything else is retained unless Windows itself reports an impossible path.
# ======================================================================================

print("\n[2/9] EXTRACTING COMPLETE NON-MISC REPOSITORY")
print("-" * 115)


extracted = 0
already_present = 0
skipped_misc = 0
skipped_long = []


with zipfile.ZipFile(
    ZIP_FILE,
    "r"
) as zf:

    members = zf.infolist()

    file_members = [
        x
        for x in members
        if not x.is_dir()
    ]

    if not file_members:
        raise RuntimeError(
            "ZIP contains no files."
        )


    archive_root = (
        PurePosixPath(
            file_members[0].filename
        ).parts[0]
    )


    print(
        "Archive root:",
        archive_root
    )


    for info in file_members:

        p = PurePosixPath(
            info.filename
        )

        parts = p.parts


        if (
            len(parts) < 2
            or
            parts[0] != archive_root
        ):
            continue


        relative = PurePosixPath(
            *parts[1:]
        )

        relative_posix = (
            relative.as_posix()
        )


        # --------------------------------------------------------------
        # Skip ONLY misc/
        # --------------------------------------------------------------

        if (
            relative_posix == "misc"
            or
            relative_posix.startswith(
                "misc/"
            )
        ):

            skipped_misc += 1
            continue


        target = (
            REPO /
            Path(
                *relative.parts
            )
        )


        # --------------------------------------------------------------
        # Conservative Windows path guard.
        # Scientifically relevant paths should normally remain below this.
        # --------------------------------------------------------------

        if (
            os.name == "nt"
            and
            len(
                str(
                    target.resolve()
                )
            ) > 245
        ):

            skipped_long.append(
                relative_posix
            )

            continue


        target.parent.mkdir(
            parents=True,
            exist_ok=True
        )


        # Existing file with matching size -> retain
        if (
            target.exists()
            and
            target.stat().st_size
            ==
            info.file_size
        ):

            already_present += 1
            continue


        with zf.open(
            info,
            "r"
        ) as src:

            with open(
                target,
                "wb"
            ) as dst:

                shutil.copyfileobj(
                    src,
                    dst
                )


        extracted += 1


print(
    "New files extracted  :",
    extracted
)

print(
    "Already present      :",
    already_present
)

print(
    "misc/ files skipped  :",
    skipped_misc
)

print(
    "Other long paths skip:",
    len(
        skipped_long
    )
)


# ======================================================================================
# 3. ROOT STRUCTURE AUDIT
# ======================================================================================

print("\n[3/9] REPOSITORY ROOT")
print("-" * 115)


root_entries = sorted(
    REPO.iterdir(),
    key=lambda x:
        (
            not x.is_dir(),
            x.name.lower()
        )
)


for p in root_entries:

    marker = (
        "DIR "
        if p.is_dir()
        else "FILE"
    )

    print(
        f"{marker:4s}  {p.name}"
    )


# ======================================================================================
# 4. VERIFY PIPELINE / PACKAGE FILES
# ======================================================================================

print("\n[4/9] PIPELINE FILE CHECK")
print("-" * 115)


important_candidates = [

    REPO / "main.nf",
    REPO / "nextflow.config",
    REPO / "setup.py",
    REPO / "requirements.txt",
    REPO / "README.md",

    REPO /
    "mantis_ml" /
    "modules" /
    "pre_processing" /
    "feature_table_compiler.py",

    REPO /
    "mantis_ml" /
    "modules" /
    "pre_processing" /
    "data_compilation" /
    "process_generic_features.py"
]


for path in important_candidates:

    print(
        (
            "✅"
            if path.exists()
            else "—"
        ),
        path.name,
        "->",
        path
    )


MAIN_NF = REPO / "main.nf"


if not MAIN_NF.exists():

    print(
        "\nNOTE: main.nf was not present in this repository snapshot."
    )

else:

    print(
        "\nmain.nf recovered ✅"
    )


# ======================================================================================
# 5. COMPLETE FILE INVENTORY
# ======================================================================================

print("\n[5/9] COMPLETE REPOSITORY INVENTORY")
print("-" * 115)


all_files = sorted(
    [
        p
        for p in REPO.rglob("*")
        if p.is_file()
    ]
)


inventory_rows = []


for path in all_files:

    inventory_rows.append(
        {
            "relative_path":
                str(
                    path.relative_to(
                        REPO
                    )
                ),

            "size_mb":
                (
                    path.stat().st_size
                    /
                    1024
                    /
                    1024
                ),

            "suffix":
                "".join(
                    path.suffixes
                ).lower()
        }
    )


inventory = pd.DataFrame(
    inventory_rows
)


print(
    "Total retained repository files:",
    len(
        inventory
    )
)


print(
    "\nLargest 25:"
)


print(
    inventory
    .sort_values(
        "size_mb",
        ascending=False
    )
    .head(25)
    .round(
        {
            "size_mb": 3
        }
    )
    .to_string(
        index=False
    )
)


# ======================================================================================
# 6. SEARCH CODE FOR GENERATED FEATURE-TABLE FILENAMES
# ======================================================================================

print("\n[6/9] FEATURE-TABLE PIPELINE DISCOVERY")
print("-" * 115)


TEXT_SUFFIXES = {

    ".py",
    ".nf",
    ".config",
    ".yaml",
    ".yml",
    ".txt",
    ".md",
    ".json",
    ".sh"
}


feature_terms = [

    "complete_feature_table",
    "processed_feature_table",
    "compiled_feature",
    "processed-feature",
    "feature_table",
    "balanced_train_df",
    "balanced_test_df",
    "full_train_df",
    "full_test_df"
]


feature_hits = []


for path in all_files:

    if path.suffix.lower() not in TEXT_SUFFIXES:
        continue


    try:

        text = path.read_text(
            encoding="utf-8",
            errors="replace"
        )

    except Exception:

        continue


    lines = text.splitlines()


    for line_no, line in enumerate(
        lines,
        start=1
    ):

        lower_line = line.lower()


        matched = [
            term
            for term in feature_terms
            if term.lower()
            in lower_line
        ]


        if matched:

            feature_hits.append(
                {
                    "file":
                        str(
                            path.relative_to(
                                REPO
                            )
                        ),

                    "line":
                        line_no,

                    "matched":
                        ";".join(
                            matched
                        ),

                    "text":
                        line.strip()[:500]
                }
            )


feature_hits_df = pd.DataFrame(
    feature_hits
)


print(
    "Feature-table references:",
    len(
        feature_hits_df
    )
)


if len(
    feature_hits_df
) > 0:

    print(
        "\nFirst 80 references:"
    )

    print(
        feature_hits_df
        .head(80)
        .to_string(
            index=False
        )
    )


# ======================================================================================
# 7. SEARCH DATA-COMPILATION CODE FOR URLs / EXTERNAL SOURCES
# ======================================================================================

print("\n[7/9] EXTERNAL DATA SOURCE DISCOVERY")
print("-" * 115)


url_pattern = re.compile(
    r'https?://[^\s\'"<>]+'
)


source_keyword_pattern = re.compile(
    r'(?i)'
    r'(protein.?atlas|'
    r'gnomad|'
    r'exac|'
    r'gevir|'
    r'rvis|'
    r'single.?cell|'
    r'cortex|'
    r'fetal|'
    r'brainspan|'
    r'inweb|'
    r'ppi|'
    r'gene.?ontology|'
    r'go_|'
    r'msigdb|'
    r'gtex)'
)


source_hits = []


for path in all_files:

    if path.suffix.lower() not in TEXT_SUFFIXES:
        continue


    try:

        text = path.read_text(
            encoding="utf-8",
            errors="replace"
        )

    except Exception:

        continue


    for line_no, line in enumerate(
        text.splitlines(),
        start=1
    ):

        urls = url_pattern.findall(
            line
        )

        keyword_match = (
            source_keyword_pattern.search(
                line
            )
        )


        if urls or keyword_match:

            source_hits.append(
                {
                    "file":
                        str(
                            path.relative_to(
                                REPO
                            )
                        ),

                    "line":
                        line_no,

                    "urls":
                        ";".join(
                            urls
                        ),

                    "text":
                        line.strip()[:600]
                }
            )


source_hits_df = pd.DataFrame(
    source_hits
)


print(
    "Potential source references:",
    len(
        source_hits_df
    )
)


if len(
    source_hits_df
) > 0:

    print(
        "\nFirst 100:"
    )

    print(
        source_hits_df
        .head(100)
        .to_string(
            index=False
        )
    )


# ======================================================================================
# 8. MAIN.NF / PREPROCESSING COMMAND AUDIT
# ======================================================================================

print("\n[8/9] PIPELINE EXECUTION AUDIT")
print("-" * 115)


execution_rows = []


execution_keywords = [

    "python",
    "mantis",
    "pre_processing",
    "feature_table",
    "compile",
    "process",
    "nextflow"
]


candidate_execution_files = [

    p
    for p in all_files
    if (
        p.suffix.lower()
        in {
            ".nf",
            ".py",
            ".sh"
        }
    )
]


for path in candidate_execution_files:

    try:

        lines = path.read_text(
            encoding="utf-8",
            errors="replace"
        ).splitlines()

    except Exception:

        continue


    for line_no, line in enumerate(
        lines,
        start=1
    ):

        line_lower = line.lower()


        if any(
            keyword in line_lower
            for keyword
            in execution_keywords
        ):

            if (
                "feature" in line_lower
                or
                "pre_process" in line_lower
                or
                "compile" in line_lower
                or
                "main.nf" in line_lower
            ):

                execution_rows.append(
                    {
                        "file":
                            str(
                                path.relative_to(
                                    REPO
                                )
                            ),

                        "line":
                            line_no,

                        "text":
                            line.strip()[:600]
                    }
                )


execution_df = pd.DataFrame(
    execution_rows
)


print(
    "Relevant execution references:",
    len(
        execution_df
    )
)


if len(
    execution_df
) > 0:

    print(
        execution_df
        .head(100)
        .to_string(
            index=False
        )
    )


# ======================================================================================
# 9. SAVE AUDIT
# ======================================================================================

print("\n[9/9] SAVING")
print("-" * 115)


inventory.to_csv(
    AUDIT_DIR /
    "complete_repository_inventory.tsv",
    sep="\t",
    index=False
)


feature_hits_df.to_csv(
    AUDIT_DIR /
    "feature_table_pipeline_references.tsv",
    sep="\t",
    index=False
)


source_hits_df.to_csv(
    AUDIT_DIR /
    "external_data_source_references.tsv",
    sep="\t",
    index=False
)


execution_df.to_csv(
    AUDIT_DIR /
    "pipeline_execution_references.tsv",
    sep="\t",
    index=False
)


if skipped_long:

    pd.DataFrame(
        {
            "skipped_path":
                skipped_long
        }
    ).to_csv(
        AUDIT_DIR /
        "non_misc_long_paths_skipped.tsv",
        sep="\t",
        index=False
    )


manifest = {

    "repository_zip":
        str(
            ZIP_FILE
        ),

    "repository_zip_sha256":
        ZIP_SHA256,

    "extraction_policy":
        "retain_all_repository_content_except_misc",

    "new_files_extracted":
        extracted,

    "already_present":
        already_present,

    "misc_files_skipped":
        skipped_misc,

    "other_long_paths_skipped":
        len(
            skipped_long
        ),

    "total_repository_files_retained":
        len(
            inventory
        )
}


with open(
    AUDIT_DIR /
    "repository_recovery_manifest.json",
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        manifest,
        f,
        indent=2
    )


print("\n" + "=" * 115)
print("REPOSITORY RECOVERY COMPLETE ✅")
print("=" * 115)

print(
    "Retained repository files :",
    len(
        inventory
    )
)

print(
    "Feature-table references  :",
    len(
        feature_hits_df
    )
)

print(
    "External-source references:",
    len(
        source_hits_df
    )
)

print(
    "Pipeline references       :",
    len(
        execution_df
    )
)

print(
    "\nAudit directory:"
)

print(
    AUDIT_DIR
)


if MAIN_NF.exists():

    print(
        "\nMAIN PIPELINE RECOVERED ✅"
    )

else:

    print(
        "\nmain.nf not present — inspect root listing above."
    )


print("=" * 115)
```

```python
# ======================================================================================
# ASD-ContextFusion
# NOTEBOOK 01 / CELL 2
#
# EXACT DHINDSA FEATURE TABLE REBUILD
#
# CATEGORY:
#   BENCHMARK DATA CONSTRUCTION
#
# Uses the AUTHORS' OWN:
#   - Config class
#   - ProcessGenericFeatures
#   - ProcessFeaturesFilteredByDisease
#   - FeatureTableCompiler
#
# Goal:
#   Rebuild:
#       generic_feature_table.tsv
#       filtered_by_disease_feature_table.tsv
#       complete_feature_table.tsv
#
# Then verify:
#   - gene universe
#   - 190 / 190 ASD seeds
#   - feature count
#
# NO MODEL TRAINING.
# ======================================================================================

from pathlib import Path
import os
import sys
import re
import json
import inspect
import warnings
import traceback

import numpy as np
import pandas as pd

warnings.filterwarnings("ignore")


# ======================================================================================
# 0. PATHS
# ======================================================================================

ROOT = Path(
    r"E:\ASD_ContextFusion_Q1"
)

REPO = (
    ROOT /
    "01_raw" /
    "dhindsa_2025" /
    "Mantis-ml-NDD"
)

ADMIN_DIR = (
    ROOT /
    "00_admin"
)

OUT_ROOT = (
    ROOT /
    "02_interim" /
    "dhindsa_exact_rebuild"
)

COMPILED_DIR = (
    OUT_ROOT /
    "data" /
    "compiled_feature_tables"
)

PROCESSED_DIR = (
    OUT_ROOT /
    "processed-feature-tables"
)

ASD_CONFIG = (
    REPO /
    "mantis_ml" /
    "conf" /
    "asd_monoallelic_config.yaml"
)

ASD_SEEDS = (
    ADMIN_DIR /
    "dhindsa_2025_monoallelic_ASD_190.tsv"
)

for p in [
    REPO,
    ASD_CONFIG,
    ASD_SEEDS
]:
    assert p.exists(), f"Missing:\n{p}"


OUT_ROOT.mkdir(
    parents=True,
    exist_ok=True
)

COMPILED_DIR.mkdir(
    parents=True,
    exist_ok=True
)

PROCESSED_DIR.mkdir(
    parents=True,
    exist_ok=True
)


print("=" * 115)
print("ASD-CONTEXTFUSION — EXACT DHINDSA FEATURE TABLE REBUILD")
print("=" * 115)

print("\nRepository:")
print(REPO)

print("\nOutput:")
print(OUT_ROOT)


# ======================================================================================
# 1. LEGACY NUMPY / PANDAS COMPATIBILITY
#
# Authors' code was written for Python 3.7-era libraries.
# These shims preserve old API names without altering data logic.
# ======================================================================================

print("\n[1/8] LEGACY COMPATIBILITY")
print("-" * 115)


legacy_numpy_aliases = {
    "int": int,
    "float": float,
    "bool": bool,
    "object": object,
    "str": str
}


for name, value in legacy_numpy_aliases.items():

    if not hasattr(
        np,
        name
    ):

        setattr(
            np,
            name,
            value
        )


if not hasattr(
    pd,
    "np"
):

    pd.np = np


if not hasattr(
    pd.Series,
    "iteritems"
):

    pd.Series.iteritems = (
        pd.Series.items
    )


if not hasattr(
    pd.DataFrame,
    "iteritems"
):

    pd.DataFrame.iteritems = (
        pd.DataFrame.items
    )


# DataFrame.append was removed in modern pandas
if not hasattr(
    pd.DataFrame,
    "append"
):

    def _legacy_append(
        self,
        other,
        ignore_index=False,
        verify_integrity=False,
        sort=False
    ):

        if isinstance(
            other,
            dict
        ):

            other = pd.DataFrame(
                [other]
            )

        elif isinstance(
            other,
            pd.Series
        ):

            other = other.to_frame().T


        return pd.concat(
            [
                self,
                other
            ],
            ignore_index=ignore_index,
            verify_integrity=verify_integrity,
            sort=sort
        )


    pd.DataFrame.append = (
        _legacy_append
    )


print(
    "NumPy version :",
    np.__version__
)

print(
    "Pandas version:",
    pd.__version__
)

print(
    "Compatibility shims loaded ✅"
)


# ======================================================================================
# 2. ADD AUTHORS' REPOSITORY TO PYTHON PATH
# ======================================================================================

print("\n[2/8] LOADING AUTHORS' SOURCE CODE")
print("-" * 115)


repo_string = str(
    REPO.resolve()
)


if repo_string not in sys.path:

    sys.path.insert(
        0,
        repo_string
    )


os.chdir(
    REPO
)


# ======================================================================================
# 3. IMPORT CONFIG MODULE
# ======================================================================================

try:

    import mantis_ml.config_class as config_module

except Exception as e:

    print(
        "\nCould not import config module."
    )

    print(
        type(e).__name__,
        ":",
        e
    )

    raise


# Find config-like classes dynamically
config_classes = []

for name, obj in inspect.getmembers(
    config_module,
    inspect.isclass
):

    if (
        "config" in name.lower()
        and
        obj.__module__
        ==
        config_module.__name__
    ):

        config_classes.append(
            (
                name,
                obj
            )
        )


if not config_classes:

    raise RuntimeError(
        "No Config-like class detected "
        "in mantis_ml.config_class"
    )


print(
    "Detected config classes:"
)

for name, cls in config_classes:

    print(
        " -",
        name,
        inspect.signature(
            cls
        )
    )


# Prefer literal Config if available
CONFIG_CLASS = None


for name, cls in config_classes:

    if name.lower() == "config":

        CONFIG_CLASS = cls
        break


if CONFIG_CLASS is None:

    CONFIG_CLASS = (
        config_classes[0][1]
    )


print(
    "\nUsing config class:",
    CONFIG_CLASS.__name__
)


# ======================================================================================
# 4. CREATE CONFIG OBJECT DYNAMICALLY
#
# Handles slightly different mantis-ml constructor signatures.
# ======================================================================================

print("\n[3/8] INITIALIZING DHINDSA ASD CONFIGURATION")
print("-" * 115)


sig = inspect.signature(
    CONFIG_CLASS
)


kwargs = {}
positional = []
unknown_required = []


for name, param in sig.parameters.items():

    lname = name.lower()


    if name == "self":
        continue


    # --------------------------------------------------------------
    # Disease/phenotype yaml
    # --------------------------------------------------------------

    if (
        "config" in lname
        or
        "conf_file" in lname
        or
        lname in {
            "conf",
            "yaml_file",
            "yaml_path"
        }
    ):

        kwargs[
            name
        ] = str(
            ASD_CONFIG
        )


    # --------------------------------------------------------------
    # Output root
    # --------------------------------------------------------------

    elif (
        "out" in lname
        or
        "output" in lname
    ):

        kwargs[
            name
        ] = str(
            OUT_ROOT
        )


    # --------------------------------------------------------------
    # Seed / known-gene file
    # --------------------------------------------------------------

    elif (
        "seed" in lname
        or
        "known_gene" in lname
        or
        "known_genes" in lname
        or
        "gene_file" in lname
    ):

        kwargs[
            name
        ] = str(
            ASD_SEEDS
        )


    # --------------------------------------------------------------
    # Optional parameter -> keep author's default
    # --------------------------------------------------------------

    elif (
        param.default
        is not
        inspect._empty
    ):

        continue


    else:

        unknown_required.append(
            name
        )


if unknown_required:

    print(
        "Required parameters not recognized automatically:",
        unknown_required
    )

    print(
        "\nConstructor signature:"
    )

    print(
        sig
    )

    raise RuntimeError(
        "Config constructor contains an unexpected required parameter. "
        "Stopping rather than guessing."
    )


print(
    "Config arguments:"
)

for key, value in kwargs.items():

    print(
        f"  {key} = {value}"
    )


cfg = CONFIG_CLASS(
    **kwargs
)


print(
    "\nConfig initialized ✅"
)


# ======================================================================================
# 5. FORCE PROJECT OUTPUT LOCATION TO E DRIVE
#
# We preserve authors' feature logic but prevent output from escaping
# our clean project folder.
# ======================================================================================

print("\n[4/8] LOCKING OUTPUT PATHS")
print("-" * 115)


# Known config attributes from authors' pipeline
attribute_overrides = {

    "out_root":
        OUT_ROOT,

    "out_data_dir":
        OUT_ROOT / "data",

    "compiled_data_dir":
        COMPILED_DIR,

    "generic_feature_table":
        COMPILED_DIR /
        "generic_feature_table.tsv",

    "filtered_by_disease_feature_table":
        COMPILED_DIR /
        "filtered_by_disease_feature_table.tsv",

    "complete_feature_table":
        COMPILED_DIR /
        "complete_feature_table.tsv",

    "processed_data_dir":
        PROCESSED_DIR
}


for attr, value in attribute_overrides.items():

    if hasattr(
        cfg,
        attr
    ):

        value = Path(
            value
        )

        value.parent.mkdir(
            parents=True,
            exist_ok=True
        )

        setattr(
            cfg,
            attr,
            value
        )

        print(
            f"{attr:36s} -> {value}"
        )


# Ensure data_dir points to AUTHORS' bundled data
AUTHORS_DATA_DIR = (
    REPO /
    "mantis_ml" /
    "data"
)


if hasattr(
    cfg,
    "data_dir"
):

    cfg.data_dir = (
        AUTHORS_DATA_DIR
    )

    print(
        f"{'data_dir':36s} -> {AUTHORS_DATA_DIR}"
    )


assert AUTHORS_DATA_DIR.exists()


# ======================================================================================
# 6. IMPORT AND RUN AUTHORS' FEATURE TABLE COMPILER
# ======================================================================================

print("\n[5/8] RUNNING AUTHORS' FEATURE COMPILER")
print("=" * 115)


from mantis_ml.modules.pre_processing.feature_table_compiler import (
    FeatureTableCompiler
)


print(
    "FeatureTableCompiler source:"
)

print(
    inspect.getfile(
        FeatureTableCompiler
    )
)


compiler = (
    FeatureTableCompiler(
        cfg
    )
)


print(
    "\nStarting exact authors' preprocessing..."
)


try:

    compiler.run()

except Exception as e:

    print("\n" + "=" * 115)
    print("FEATURE COMPILER STOPPED")
    print("=" * 115)

    print(
        "\nError type:",
        type(e).__name__
    )

    print(
        "Message:",
        e
    )

    print(
        "\nFull traceback:"
    )

    traceback.print_exc()

    raise


print(
    "\nAuthors' FeatureTableCompiler completed ✅"
)


# ======================================================================================
# 7. LOCATE OUTPUT TABLES
# ======================================================================================

print("\n[6/8] LOCATING COMPILED TABLES")
print("-" * 115)


# First use expected paths
GENERIC_FILE = (
    COMPILED_DIR /
    "generic_feature_table.tsv"
)

FILTERED_FILE = (
    COMPILED_DIR /
    "filtered_by_disease_feature_table.tsv"
)

COMPLETE_FILE = (
    COMPILED_DIR /
    "complete_feature_table.tsv"
)


# If authors' object saved elsewhere, locate it
if not COMPLETE_FILE.exists():

    candidates = list(
        OUT_ROOT.rglob(
            "complete_feature_table.tsv"
        )
    )


    if len(
        candidates
    ) == 1:

        COMPLETE_FILE = (
            candidates[0]
        )


if not GENERIC_FILE.exists():

    candidates = list(
        OUT_ROOT.rglob(
            "generic_feature_table.tsv"
        )
    )

    if len(
        candidates
    ) == 1:

        GENERIC_FILE = (
            candidates[0]
        )


if not FILTERED_FILE.exists():

    candidates = list(
        OUT_ROOT.rglob(
            "filtered_by_disease_feature_table.tsv"
        )
    )

    if len(
        candidates
    ) == 1:

        FILTERED_FILE = (
            candidates[0]
        )


print(
    "Generic:"
)

print(
    GENERIC_FILE,
    "✅"
    if GENERIC_FILE.exists()
    else "❌"
)


print(
    "\nDisease-filtered:"
)

print(
    FILTERED_FILE,
    "✅"
    if FILTERED_FILE.exists()
    else "❌"
)


print(
    "\nComplete:"
)

print(
    COMPLETE_FILE,
    "✅"
    if COMPLETE_FILE.exists()
    else "❌"
)


if not COMPLETE_FILE.exists():

    raise RuntimeError(
        "Feature compiler finished but "
        "complete_feature_table.tsv was not found."
    )


# ======================================================================================
# 8. FINAL TABLE AUDIT
# ======================================================================================

print("\n[7/8] COMPLETE FEATURE TABLE AUDIT")
print("-" * 115)


complete = pd.read_csv(
    COMPLETE_FILE,
    sep="\t",
    low_memory=False
)


print(
    "Complete table shape:",
    complete.shape
)


print(
    "First 20 columns:"
)


for column in complete.columns[:20]:

    print(
        " -",
        column
    )


# --------------------------------------------------------------
# Detect gene column
# --------------------------------------------------------------

seed_table = pd.read_csv(
    ASD_SEEDS,
    sep="\t"
)


seed_set = set(
    seed_table[
        "gene"
    ]
    .astype(str)
    .str.strip()
)


best_gene_column = None
best_overlap = -1


for column in complete.columns:

    try:

        values = set(
            complete[
                column
            ]
            .dropna()
            .astype(str)
            .str.strip()
        )

    except Exception:

        continue


    overlap = len(
        values &
        seed_set
    )


    if overlap > best_overlap:

        best_overlap = (
            overlap
        )

        best_gene_column = (
            column
        )


print(
    "\nDetected gene column:",
    best_gene_column
)

print(
    "ASD seed overlap:",
    best_overlap,
    "/ 190"
)


if best_overlap != 190:

    raise RuntimeError(
        f"Expected 190/190 ASD genes, "
        f"found {best_overlap}/190. "
        "Stopping before model construction."
    )


# --------------------------------------------------------------
# Gene universe
# --------------------------------------------------------------

gene_values = (
    complete[
        best_gene_column
    ]
    .astype(str)
    .str.strip()
)


print(
    "Rows             :",
    len(
        complete
    )
)

print(
    "Unique genes     :",
    gene_values.nunique()
)

print(
    "Feature columns  :",
    complete.shape[1]
)


# ======================================================================================
# 9. SAVE LOCKED COPY + MANIFEST
# ======================================================================================

print("\n[8/8] LOCKING REBUILT DHINDSA TABLE")
print("-" * 115)


LOCKED_COMPLETE = (
    ROOT /
    "03_features" /
    "dhindsa_full_context" /
    "dhindsa_complete_feature_table_locked.tsv"
)


LOCKED_COMPLETE.parent.mkdir(
    parents=True,
    exist_ok=True
)


complete.to_csv(
    LOCKED_COMPLETE,
    sep="\t",
    index=False
)


import hashlib


complete_sha = hashlib.sha256(
    LOCKED_COMPLETE.read_bytes()
).hexdigest()


manifest = {

    "source":
        "Dhindsa et al. 2025 Mantis-ml-NDD",

    "construction":
        "authors FeatureTableCompiler",

    "asd_positive_genes":
        190,

    "gene_column":
        str(
            best_gene_column
        ),

    "asd_seed_overlap":
        int(
            best_overlap
        ),

    "rows":
        int(
            complete.shape[0]
        ),

    "columns":
        int(
            complete.shape[1]
        ),

    "unique_genes":
        int(
            gene_values.nunique()
        ),

    "complete_table":
        str(
            LOCKED_COMPLETE
        ),

    "sha256":
        complete_sha
}


MANIFEST = (
    ADMIN_DIR /
    "dhindsa_complete_feature_table_manifest.json"
)


with open(
    MANIFEST,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        manifest,
        f,
        indent=2
    )


print(
    "Locked table:"
)

print(
    LOCKED_COMPLETE
)


print(
    "\nSHA256:"
)

print(
    complete_sha
)


print("\n" + "=" * 115)
print("DHINDSA EXACT FEATURE REBUILD COMPLETE ✅")
print("=" * 115)

print(
    f"Rows        : {complete.shape[0]}"
)

print(
    f"Columns     : {complete.shape[1]}"
)

print(
    f"Unique genes: {gene_values.nunique()}"
)

print(
    f"ASD genes   : {best_overlap}/190"
)

print("=" * 115)
```

```python
# ======================================================================================
# ASD-ContextFusion
# ROBUST LEGACY ENVIRONMENT RECOVERY
#
# Strategy:
#   Conda -> ONLY Python 3.7 + pip
#   pip   -> pinned legacy scientific stack
#
# No Jupyter kernel switch required.
# Current Python 3.13 remains untouched.
# ======================================================================================

from pathlib import Path
import subprocess
import shutil
import sys
import os
import json


ROOT = Path(r"E:\ASD_ContextFusion_Q1")

ENV_DIR = (
    ROOT /
    "envs" /
    "mantis37"
)

ADMIN_DIR = (
    ROOT /
    "00_admin"
)

ADMIN_DIR.mkdir(
    parents=True,
    exist_ok=True
)


CONDA = Path(
    r"E:\Python\Library\bin\conda.BAT"
)


print("=" * 110)
print("ASD-CONTEXTFUSION — ROBUST LEGACY ENVIRONMENT RECOVERY")
print("=" * 110)


# ======================================================================================
# 1. VERIFY CONDA
# ======================================================================================

print("\n[1/6] CONDA")
print("-" * 110)


if not CONDA.exists():

    raise RuntimeError(
        f"Conda not found:\n{CONDA}"
    )


print(
    "Conda:",
    CONDA
)


# ======================================================================================
# 2. REMOVE ONLY INCOMPLETE LEGACY ENV
# ======================================================================================

print("\n[2/6] CLEANING INCOMPLETE ENVIRONMENT")
print("-" * 110)


ENV_PYTHON = (
    ENV_DIR /
    "python.exe"
)


# Previous failed conda solve may have left a partial folder.
if ENV_DIR.exists():

    if not ENV_PYTHON.exists():

        print(
            "Incomplete environment detected."
        )

        shutil.rmtree(
            ENV_DIR,
            ignore_errors=True
        )

        print(
            "Incomplete environment removed."
        )

    else:

        print(
            "Existing Python detected; environment will be verified."
        )

else:

    print(
        "No previous environment detected."
    )


# ======================================================================================
# 3. CREATE MINIMAL PYTHON 3.7 ENVIRONMENT
#
# IMPORTANT:
# Do NOT ask conda to solve pandas/scikit-learn/xgboost simultaneously.
# ======================================================================================

print("\n[3/6] CREATING MINIMAL PYTHON 3.7 ENVIRONMENT")
print("-" * 110)


if not ENV_PYTHON.exists():

    conda_command = (
        f'call "{CONDA}" create -y '
        f'-p "{ENV_DIR}" '
        f'--override-channels '
        f'-c conda-forge '
        f'--repodata-fn repodata.json '
        f'python=3.7.12 pip'
    )


    print(
        "Creating only Python 3.7 + pip..."
    )


    result = subprocess.run(
        [
            "cmd.exe",
            "/d",
            "/s",
            "/c",
            conda_command
        ],
        stdout=subprocess.PIPE,
        stderr=subprocess.STDOUT,
        text=True
    )


    print(
        result.stdout[-12000:]
    )


    if result.returncode != 0:

        raise RuntimeError(
            "\nMinimal Python 3.7 environment creation failed.\n"
            "The actual Conda solver output is printed immediately above.\n"
            "Do not rerun the old environment cell."
        )


if not ENV_PYTHON.exists():

    raise RuntimeError(
        "Conda returned success but python.exe was not found."
    )


print(
    "\nPython 3.7 environment created."
)

print(
    ENV_PYTHON
)


# ======================================================================================
# 4. PIN PIP / SETUP TO PYTHON-3.7-COMPATIBLE VERSIONS
# ======================================================================================

print("\n[4/6] PREPARING LEGACY PIP")
print("-" * 110)


bootstrap_packages = [

    "pip==23.1.2",
    "setuptools==67.8.0",
    "wheel==0.41.3"
]


result = subprocess.run(
    [
        str(
            ENV_PYTHON
        ),
        "-m",
        "pip",
        "install",
        "--upgrade",
        *bootstrap_packages
    ],
    stdout=subprocess.PIPE,
    stderr=subprocess.STDOUT,
    text=True
)


print(
    result.stdout[-8000:]
)


if result.returncode != 0:

    raise RuntimeError(
        "Legacy pip bootstrap failed."
    )


# ======================================================================================
# 5. INSTALL COMPATIBLE SCIENTIFIC STACK
#
# These versions support Python 3.7 and retain the pandas behavior
# expected by the original Mantis source.
# ======================================================================================

print("\n[5/6] INSTALLING LEGACY SCIENTIFIC STACK")
print("-" * 110)


legacy_packages = [

    # Core numerical stack
    "numpy==1.21.6",
    "pandas==1.3.5",
    "scipy==1.7.3",

    # ML
    "scikit-learn==0.24.2",
    "xgboost==1.5.2",

    # Data / config
    "PyYAML==6.0.1",
    "xlrd==1.2.0",
    "openpyxl==3.0.10",

    # Utilities
    "joblib==1.1.1",
    "tqdm==4.64.1",

    # Other common Mantis dependencies
    "networkx==2.6.3",
    "statsmodels==0.12.2"
]


result = subprocess.run(
    [
        str(
            ENV_PYTHON
        ),
        "-m",
        "pip",
        "install",
        "--only-binary=:all:",
        *legacy_packages
    ],
    stdout=subprocess.PIPE,
    stderr=subprocess.STDOUT,
    text=True
)


print(
    result.stdout[-15000:]
)


if result.returncode != 0:

    raise RuntimeError(
        "\nLegacy package installation failed.\n"
        "The pip error is printed directly above."
    )


# ======================================================================================
# 6. VERIFY ENVIRONMENT
# ======================================================================================

print("\n[6/6] FINAL VERIFICATION")
print("-" * 110)


verify_code = r'''
import sys
import numpy
import pandas
import scipy
import sklearn
import xgboost
import yaml
import networkx
import statsmodels

print("Python       :", sys.version.split()[0])
print("NumPy        :", numpy.__version__)
print("Pandas       :", pandas.__version__)
print("SciPy        :", scipy.__version__)
print("scikit-learn :", sklearn.__version__)
print("XGBoost      :", xgboost.__version__)
print("PyYAML       :", yaml.__version__)
print("NetworkX     :", networkx.__version__)
print("statsmodels  :", statsmodels.__version__)
'''


result = subprocess.run(
    [
        str(
            ENV_PYTHON
        ),
        "-c",
        verify_code
    ],
    stdout=subprocess.PIPE,
    stderr=subprocess.STDOUT,
    text=True
)


print(
    result.stdout
)


if result.returncode != 0:

    raise RuntimeError(
        "Environment verification failed."
    )


# ======================================================================================
# SAVE MANIFEST
# ======================================================================================

manifest = {

    "environment":
        "ASD-ContextFusion Mantis compatibility environment",

    "path":
        str(
            ENV_DIR
        ),

    "python_executable":
        str(
            ENV_PYTHON
        ),

    "python":
        "3.7.12",

    "numpy":
        "1.21.6",

    "pandas":
        "1.3.5",

    "scipy":
        "1.7.3",

    "scikit_learn":
        "0.24.2",

    "xgboost":
        "1.5.2",

    "execution_strategy":
        "Legacy compiler executed as subprocess from main Python 3.13 Jupyter environment"
}


MANIFEST = (
    ADMIN_DIR /
    "mantis37_compatibility_environment.json"
)


with open(
    MANIFEST,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        manifest,
        f,
        indent=2
    )


print("=" * 110)
print("LEGACY COMPATIBILITY ENVIRONMENT READY")
print("=" * 110)

print(
    "Legacy Python:"
)

print(
    ENV_PYTHON
)

print(
    "\nNO JUPYTER KERNEL CHANGE REQUIRED."
)

print(
    "Next step = run original Dhindsa compiler "
    "through this Python executable."
)

print("=" * 110)
```

```python
# ======================================================================================
# ASD-ContextFusion
# LEGACY ENVIRONMENT — WINDOWS .BAT QUOTING FIX
#
# Fixes:
#   conda.BAT quoting problem
#
# Strategy:
#   1. Create ONLY Python 3.7 + pip with Conda
#   2. Install legacy scientific packages with pip
#   3. Verify environment
#
# Main Python 3.13 remains untouched.
# ======================================================================================

from pathlib import Path
import subprocess
import shutil
import json


# ======================================================================================
# 0. PATHS
# ======================================================================================

ROOT = Path(r"E:\ASD_ContextFusion_Q1")

CONDA = Path(
    r"E:\Python\Library\bin\conda.BAT"
)

ENV_DIR = (
    ROOT /
    "envs" /
    "mantis37"
)

ADMIN_DIR = (
    ROOT /
    "00_admin"
)

ADMIN_DIR.mkdir(
    parents=True,
    exist_ok=True
)

ENV_DIR.parent.mkdir(
    parents=True,
    exist_ok=True
)


ENV_PYTHON = (
    ENV_DIR /
    "python.exe"
)


print("=" * 110)
print("ASD-CONTEXTFUSION — WINDOWS LEGACY ENVIRONMENT FIX")
print("=" * 110)


# ======================================================================================
# 1. VERIFY CONDA
# ======================================================================================

print("\n[1/7] CONDA CHECK")
print("-" * 110)


if not CONDA.exists():

    raise RuntimeError(
        f"Conda not found:\n{CONDA}"
    )


print(
    "Conda:"
)

print(
    CONDA
)


version_result = subprocess.run(
    [
        str(CONDA),
        "--version"
    ],
    capture_output=True,
    text=True,
    shell=False
)


print(
    version_result.stdout.strip()
)


if version_result.returncode != 0:

    print(
        version_result.stderr
    )

    raise RuntimeError(
        "Conda executable check failed."
    )


# ======================================================================================
# 2. CLEAN ONLY INCOMPLETE ENVIRONMENT
# ======================================================================================

print("\n[2/7] ENVIRONMENT STATUS")
print("-" * 110)


if ENV_DIR.exists() and not ENV_PYTHON.exists():

    print(
        "Incomplete previous environment found."
    )

    shutil.rmtree(
        ENV_DIR,
        ignore_errors=True
    )

    print(
        "Incomplete environment removed ✅"
    )


elif ENV_PYTHON.exists():

    print(
        "Existing Python environment detected."
    )


else:

    print(
        "No existing environment detected."
    )


# ======================================================================================
# 3. CREATE MINIMAL PYTHON 3.7 ENVIRONMENT
#
# IMPORTANT:
# Direct .BAT execution.
# No cmd.exe.
# No nested quoting.
# ======================================================================================

print("\n[3/7] CREATE PYTHON 3.7")
print("-" * 110)


if not ENV_PYTHON.exists():

    create_cmd = [

        str(CONDA),

        "create",

        "-y",

        "-p",
        str(ENV_DIR),

        "--override-channels",

        "-c",
        "conda-forge",

        "--solver",
        "classic",

        "python=3.7.12",

        "pip"
    ]


    print(
        "Creating minimal environment..."
    )


    result = subprocess.run(
        create_cmd,
        stdout=subprocess.PIPE,
        stderr=subprocess.STDOUT,
        text=True,
        shell=False
    )


    print(
        result.stdout[-15000:]
    )


    if result.returncode != 0:

        raise RuntimeError(
            "\nConda environment creation failed.\n"
            "This time the output above is the REAL Conda solver message."
        )


else:

    print(
        "Environment already exists — creation skipped ✅"
    )


if not ENV_PYTHON.exists():

    raise RuntimeError(
        f"Python executable not found:\n{ENV_PYTHON}"
    )


# ======================================================================================
# 4. VERIFY PYTHON FIRST
# ======================================================================================

print("\n[4/7] PYTHON VERIFICATION")
print("-" * 110)


result = subprocess.run(
    [
        str(ENV_PYTHON),
        "--version"
    ],
    stdout=subprocess.PIPE,
    stderr=subprocess.STDOUT,
    text=True
)


print(
    result.stdout
)


if result.returncode != 0:

    raise RuntimeError(
        "Legacy Python failed to start."
    )


# ======================================================================================
# 5. PREPARE PYTHON-3.7-COMPATIBLE PIP
# ======================================================================================

print("\n[5/7] LEGACY PIP")
print("-" * 110)


pip_bootstrap = [

    "pip==23.1.2",
    "setuptools==67.8.0",
    "wheel==0.41.3"
]


result = subprocess.run(
    [
        str(ENV_PYTHON),
        "-m",
        "pip",
        "install",
        "--upgrade",
        *pip_bootstrap
    ],
    stdout=subprocess.PIPE,
    stderr=subprocess.STDOUT,
    text=True
)


print(
    result.stdout[-10000:]
)


if result.returncode != 0:

    raise RuntimeError(
        "pip bootstrap failed."
    )


# ======================================================================================
# 6. INSTALL LEGACY SCIENTIFIC STACK
#
# Install in small groups so one dependency conflict
# does not hide the actual problem.
# ======================================================================================

print("\n[6/7] INSTALLING COMPATIBLE PACKAGES")
print("-" * 110)


package_groups = [

    [
        "numpy==1.21.6",
        "pandas==1.3.5",
        "scipy==1.7.3"
    ],

    [
        "scikit-learn==0.24.2",
        "joblib==1.1.1"
    ],

    [
        "xgboost==1.5.2"
    ],

    [
        "PyYAML==6.0.1",
        "xlrd==1.2.0",
        "openpyxl==3.0.10",
        "tqdm==4.64.1"
    ],

    [
        "networkx==2.6.3",
        "statsmodels==0.12.2"
    ]
]


for i, packages in enumerate(
    package_groups,
    start=1
):

    print(
        f"\nInstalling group {i}/{len(package_groups)}:"
    )

    print(
        packages
    )


    result = subprocess.run(
        [
            str(ENV_PYTHON),
            "-m",
            "pip",
            "install",
            "--only-binary=:all:",
            *packages
        ],
        stdout=subprocess.PIPE,
        stderr=subprocess.STDOUT,
        text=True
    )


    print(
        result.stdout[-8000:]
    )


    if result.returncode != 0:

        raise RuntimeError(
            f"\nPackage group {i} failed.\n"
            f"Failed group: {packages}\n"
            "The exact pip error is printed immediately above."
        )


# ======================================================================================
# 7. FINAL ENVIRONMENT AUDIT
# ======================================================================================

print("\n[7/7] FINAL VERIFICATION")
print("-" * 110)


verify_code = r'''
import sys
import numpy
import pandas
import scipy
import sklearn
import xgboost
import yaml
import networkx
import statsmodels
import xlrd
import openpyxl

print("Python       :", sys.version.split()[0])
print("NumPy        :", numpy.__version__)
print("Pandas       :", pandas.__version__)
print("SciPy        :", scipy.__version__)
print("scikit-learn :", sklearn.__version__)
print("XGBoost      :", xgboost.__version__)
print("PyYAML       :", yaml.__version__)
print("NetworkX     :", networkx.__version__)
print("statsmodels  :", statsmodels.__version__)
print("xlrd         :", xlrd.__version__)
print("openpyxl     :", openpyxl.__version__)
'''


result = subprocess.run(
    [
        str(ENV_PYTHON),
        "-c",
        verify_code
    ],
    stdout=subprocess.PIPE,
    stderr=subprocess.STDOUT,
    text=True
)


print(
    result.stdout
)


if result.returncode != 0:

    raise RuntimeError(
        "Final environment verification failed."
    )


# ======================================================================================
# SAVE MANIFEST
# ======================================================================================

manifest = {

    "environment":
        "ASD-ContextFusion Dhindsa/Mantis legacy",

    "environment_path":
        str(ENV_DIR),

    "python_executable":
        str(ENV_PYTHON),

    "python":
        "3.7.12",

    "numpy":
        "1.21.6",

    "pandas":
        "1.3.5",

    "scipy":
        "1.7.3",

    "scikit_learn":
        "0.24.2",

    "xgboost":
        "1.5.2",

    "purpose":
        "Run original Dhindsa/Mantis preprocessing without modern Pandas incompatibility",

    "main_environment":
        "Python 3.13 remains primary ASD-ContextFusion analysis environment"
}


MANIFEST_FILE = (
    ADMIN_DIR /
    "mantis37_environment_manifest.json"
)


with open(
    MANIFEST_FILE,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        manifest,
        f,
        indent=2
    )


print("\n" + "=" * 110)
print("LEGACY MANTIS ENVIRONMENT READY ✅")
print("=" * 110)

print(
    "\nLegacy Python:"
)

print(
    ENV_PYTHON
)

print(
    "\nIMPORTANT:"
)

print(
    "Do NOT change your Jupyter kernel."
)

print(
    "Next Dhindsa preprocessing will be launched "
    "from this notebook as a subprocess."
)

print("=" * 110)
```

```python
# ======================================================================================
# ASD-ContextFusion
# NOTEBOOK 01 / CELL 3
#
# ORIGINAL DHINDSA FEATURE COMPILATION
# Executed in isolated Python 3.7 legacy environment
#
# Main Jupyter:
#   Python 3.13
#
# Dhindsa preprocessing:
#   Python 3.7.12
#   Pandas 1.3.5
#   NumPy 1.21.6
#
# OUTPUT:
#   E:\ASD_ContextFusion_Q1\02_interim\dhindsa_exact_rebuild
#
# NO MODEL TRAINING.
# ======================================================================================

from pathlib import Path
import subprocess
import shutil
import hashlib
import json
import os
import sys
import pandas as pd


# ======================================================================================
# 0. PROJECT PATHS
# ======================================================================================

ROOT = Path(r"E:\ASD_ContextFusion_Q1")

REPO = (
    ROOT /
    "01_raw" /
    "dhindsa_2025" /
    "Mantis-ml-NDD"
)

LEGACY_PYTHON = (
    ROOT /
    "envs" /
    "mantis37" /
    "python.exe"
)

CONFIG_FILE = (
    REPO /
    "mantis_ml" /
    "conf" /
    "asd_monoallelic_config.yaml"
)

SEED_FILE = (
    ROOT /
    "00_admin" /
    "dhindsa_2025_monoallelic_ASD_190.tsv"
)

RUNNER_FILE = (
    ROOT /
    "00_admin" /
    "run_dhindsa_feature_compiler_legacy.py"
)

OUT_ROOT = (
    ROOT /
    "02_interim" /
    "dhindsa_exact_rebuild"
)

LOG_DIR = (
    ROOT /
    "08_logs"
)

LOG_FILE = (
    LOG_DIR /
    "dhindsa_exact_feature_rebuild.log"
)

LOCK_DIR = (
    ROOT /
    "03_features" /
    "dhindsa_full_context"
)

MANIFEST_FILE = (
    ROOT /
    "00_admin" /
    "dhindsa_exact_feature_rebuild_manifest.json"
)


for p in [
    REPO,
    LEGACY_PYTHON,
    CONFIG_FILE,
    SEED_FILE
]:
    assert p.exists(), f"Missing required file:\n{p}"


LOG_DIR.mkdir(
    parents=True,
    exist_ok=True
)

LOCK_DIR.mkdir(
    parents=True,
    exist_ok=True
)


print("=" * 115)
print("ASD-CONTEXTFUSION — ORIGINAL DHINDSA FEATURE COMPILATION")
print("=" * 115)

print("\nMain Jupyter Python:")
print(sys.version.split()[0])

print("\nLegacy compiler Python:")
print(LEGACY_PYTHON)


# ======================================================================================
# 1. VERIFY LEGACY ENVIRONMENT ONE LAST TIME
# ======================================================================================

print("\n[1/8] LEGACY ENVIRONMENT VERIFICATION")
print("-" * 115)


verify_code = r'''
import sys
import numpy
import pandas
import scipy
import sklearn

print("Python       :", sys.version.split()[0])
print("NumPy        :", numpy.__version__)
print("Pandas       :", pandas.__version__)
print("SciPy        :", scipy.__version__)
print("scikit-learn :", sklearn.__version__)
'''


verify = subprocess.run(
    [
        str(LEGACY_PYTHON),
        "-c",
        verify_code
    ],
    stdout=subprocess.PIPE,
    stderr=subprocess.STDOUT,
    text=True
)


print(
    verify.stdout
)


if verify.returncode != 0:
    raise RuntimeError(
        "Legacy Python verification failed."
    )


# ======================================================================================
# 2. VERIFY ORIGINAL SOURCE WAS RESTORED
# ======================================================================================

print("\n[2/8] SOURCE INTEGRITY CHECK")
print("-" * 115)


GENERIC_SOURCE = (
    REPO /
    "mantis_ml" /
    "modules" /
    "pre_processing" /
    "data_compilation" /
    "process_generic_features.py"
)


source_text = GENERIC_SOURCE.read_text(
    encoding="utf-8",
    errors="replace"
)


# This string was introduced only by our temporary Pandas-3 patch.
TEMP_PATCH_MARKER = (
    "_cols = hits_per_gene_df.columns.tolist()"
)


if TEMP_PATCH_MARKER in source_text:

    raise RuntimeError(
        "Temporary Pandas-3 source patch is still present.\n"
        "Do not continue because we want the original Dhindsa source."
    )


print(
    "Temporary modern-Pandas patch absent ✅"
)

print(
    "Original source path:"
)

print(
    GENERIC_SOURCE
)


# ======================================================================================
# 3. CLEAN ONLY FAILED/INCOMPLETE REBUILD OUTPUT
#
# We do NOT touch:
#   raw repository
#   seeds
#   NT data
#   project config
# ======================================================================================

print("\n[3/8] CLEAN OUTPUT DIRECTORY")
print("-" * 115)


if OUT_ROOT.exists():

    shutil.rmtree(
        OUT_ROOT
    )

    print(
        "Previous partial rebuild removed ✅"
    )

else:

    print(
        "No previous rebuild output detected ✅"
    )


OUT_ROOT.mkdir(
    parents=True,
    exist_ok=True
)


# ======================================================================================
# 4. CREATE LEGACY RUNNER SCRIPT
#
# This script executes INSIDE Python 3.7.
# No source-code modifications are made.
# ======================================================================================

print("\n[4/8] BUILD LEGACY COMPILER RUNNER")
print("-" * 115)


runner_code = f'''
from __future__ import print_function

import os
import sys
from pathlib import Path

REPO = Path(r"{REPO}")
CONFIG_FILE = Path(r"{CONFIG_FILE}")
OUT_ROOT = Path(r"{OUT_ROOT}")

os.chdir(str(REPO))
sys.path.insert(0, str(REPO))

print("=" * 100)
print("DHINDSA / MANTIS LEGACY FEATURE COMPILER")
print("=" * 100)
print("Python :", sys.version)
print("Repo   :", REPO)
print("Config :", CONFIG_FILE)
print("Output :", OUT_ROOT)

from mantis_ml.config_class import Config

from mantis_ml.modules.pre_processing.feature_table_compiler import (
    FeatureTableCompiler
)

cfg = Config(
    str(CONFIG_FILE),
    str(OUT_ROOT),
    verbose=False
)

print("")
print("Configuration initialized.")
print("P/U balancing ratio:", cfg.balancing_ratio)

compiler = FeatureTableCompiler(
    cfg
)

print("")
print("Starting FeatureTableCompiler.run()")
print("=" * 100)

compiler.run()

print("")
print("=" * 100)
print("DHINDSA FEATURE COMPILER FINISHED SUCCESSFULLY")
print("=" * 100)

print("Complete table:")
print(cfg.complete_feature_table)

print("Processed data dir:")
print(cfg.processed_data_dir)
'''


RUNNER_FILE.write_text(
    runner_code,
    encoding="utf-8"
)


print(
    "Runner:"
)

print(
    RUNNER_FILE
)


# ======================================================================================
# 5. RUN ORIGINAL COMPILER WITH LIVE OUTPUT + LOGGING
# ======================================================================================

print("\n[5/8] RUNNING DHINDSA COMPILER")
print("=" * 115)

print(
    "This may take several minutes."
)

print(
    "All output is also being saved to:"
)

print(
    LOG_FILE
)

print("=" * 115)


env = os.environ.copy()

env["PYTHONPATH"] = (
    str(REPO)
)


with open(
    LOG_FILE,
    "w",
    encoding="utf-8"
) as log_handle:

    process = subprocess.Popen(
        [
            str(LEGACY_PYTHON),
            "-u",
            str(RUNNER_FILE)
        ],
        cwd=str(REPO),
        env=env,
        stdout=subprocess.PIPE,
        stderr=subprocess.STDOUT,
        text=True,
        bufsize=1
    )


    for line in process.stdout:

        print(
            line,
            end=""
        )

        log_handle.write(
            line
        )

        log_handle.flush()


    return_code = process.wait()


if return_code != 0:

    print("\n" + "=" * 115)
    print("DHINDSA COMPILER FAILED")
    print("=" * 115)

    print(
        "Exit code:",
        return_code
    )

    print(
        "\nFull log:"
    )

    print(
        LOG_FILE
    )

    raise RuntimeError(
        "Legacy Dhindsa compiler stopped. "
        "Send me the LAST ~50 lines of the log/output."
    )


print("\nCompiler exit code: 0 ✅")


# ======================================================================================
# 6. LOCATE AUTHORS' OUTPUT TABLES
# ======================================================================================

print("\n[6/8] LOCATING FEATURE TABLES")
print("-" * 115)


complete_candidates = list(
    OUT_ROOT.rglob(
        "complete_feature_table.tsv"
    )
)


generic_candidates = list(
    OUT_ROOT.rglob(
        "generic_feature_table.tsv"
    )
)


filtered_candidates = list(
    OUT_ROOT.rglob(
        "filtered_by_disease_feature_table.tsv"
    )
)


print(
    "Complete tables found:",
    len(complete_candidates)
)

print(
    "Generic tables found :",
    len(generic_candidates)
)

print(
    "Filtered tables found:",
    len(filtered_candidates)
)


if len(
    complete_candidates
) != 1:

    raise RuntimeError(
        f"Expected exactly one complete_feature_table.tsv, "
        f"found {len(complete_candidates)}"
    )


COMPLETE_FILE = (
    complete_candidates[0]
)


GENERIC_FILE = (
    generic_candidates[0]
    if len(generic_candidates) == 1
    else None
)


FILTERED_FILE = (
    filtered_candidates[0]
    if len(filtered_candidates) == 1
    else None
)


print(
    "\nComplete:"
)

print(
    COMPLETE_FILE
)


# ======================================================================================
# 7. COMPLETE TABLE AUDIT
# ======================================================================================

print("\n[7/8] COMPLETE FEATURE TABLE AUDIT")
print("-" * 115)


complete = pd.read_csv(
    COMPLETE_FILE,
    sep="\t",
    low_memory=False
)


seeds = (
    pd.read_csv(
        SEED_FILE,
        sep="\t"
    )["gene"]
    .astype(str)
    .str.strip()
)


seed_set = set(
    seeds
)


assert len(seed_set) == 190


# --------------------------------------------------------------
# Detect gene column empirically
# --------------------------------------------------------------

best_gene_col = None
best_overlap = -1


for column in complete.columns:

    try:

        values = set(
            complete[column]
            .dropna()
            .astype(str)
            .str.strip()
        )

        overlap = len(
            values &
            seed_set
        )

        if overlap > best_overlap:

            best_overlap = overlap
            best_gene_col = column

    except Exception:

        continue


if best_gene_col is None:

    raise RuntimeError(
        "Could not determine gene-symbol column."
    )


genes = (
    complete[
        best_gene_col
    ]
    .astype(str)
    .str.strip()
)


n_rows = (
    complete.shape[0]
)

n_columns = (
    complete.shape[1]
)

n_unique_genes = (
    genes.nunique()
)


print(
    "Complete table shape :",
    complete.shape
)

print(
    "Gene column          :",
    best_gene_col
)

print(
    "Unique genes         :",
    n_unique_genes
)

print(
    "Locked ASD overlap   :",
    f"{best_overlap}/190"
)


if best_overlap != 190:

    missing = sorted(
        seed_set -
        set(genes)
    )

    print(
        "\nMissing ASD genes:"
    )

    print(
        missing
    )

    raise RuntimeError(
        "190/190 ASD retention failed. "
        "Stopping before any modeling."
    )


print(
    "\n190/190 ASD GENES RETAINED ✅"
)


# ======================================================================================
# 8. CREATE OUR LOCKED ANALYSIS COPY
#
# Important:
# Authors' original compiled output remains untouched.
#
# We add our OWN exact locked ASD label based solely on the
# previously verified 190 monoallelic ASD seed list.
# ======================================================================================

print("\n[8/8] LOCKING ANALYSIS DATASET")
print("-" * 115)


analysis_table = (
    complete.copy()
)


analysis_table[
    "ASD_ContextFusion_label"
] = (
    genes.isin(
        seed_set
    )
    .astype(
        "int8"
    )
)


assert int(
    analysis_table[
        "ASD_ContextFusion_label"
    ].sum()
) == 190


LOCKED_COMPLETE = (
    LOCK_DIR /
    "dhindsa_complete_feature_table_locked.tsv"
)


analysis_table.to_csv(
    LOCKED_COMPLETE,
    sep="\t",
    index=False
)


# --------------------------------------------------------------
# Also preserve authors' raw compiled table unchanged
# --------------------------------------------------------------

AUTHORS_COMPLETE_COPY = (
    LOCK_DIR /
    "dhindsa_complete_feature_table_authors_unmodified.tsv"
)


shutil.copy2(
    COMPLETE_FILE,
    AUTHORS_COMPLETE_COPY
)


# --------------------------------------------------------------
# Generic + disease-filtered copies if present
# --------------------------------------------------------------

LOCKED_GENERIC = None

if GENERIC_FILE is not None:

    LOCKED_GENERIC = (
        LOCK_DIR /
        "dhindsa_generic_feature_table_authors.tsv"
    )

    shutil.copy2(
        GENERIC_FILE,
        LOCKED_GENERIC
    )


LOCKED_FILTERED = None

if FILTERED_FILE is not None:

    LOCKED_FILTERED = (
        LOCK_DIR /
        "dhindsa_filtered_by_disease_feature_table_authors.tsv"
    )

    shutil.copy2(
        FILTERED_FILE,
        LOCKED_FILTERED
    )


# --------------------------------------------------------------
# SHA256 helper
# --------------------------------------------------------------

def sha256_file(
    path,
    chunk_size=1024 * 1024
):

    h = hashlib.sha256()

    with open(
        path,
        "rb"
    ) as f:

        while True:

            chunk = f.read(
                chunk_size
            )

            if not chunk:
                break

            h.update(
                chunk
            )

    return h.hexdigest()


authors_sha = sha256_file(
    AUTHORS_COMPLETE_COPY
)

locked_sha = sha256_file(
    LOCKED_COMPLETE
)


# --------------------------------------------------------------
# Manifest
# --------------------------------------------------------------

manifest = {

    "project":
        "ASD-ContextFusion",

    "benchmark":
        "Dhindsa et al. 2025 monoallelic ASD",

    "compiler_execution":
        "Original FeatureTableCompiler in isolated legacy Python subprocess",

    "legacy_python":
        "3.7.12",

    "legacy_numpy":
        "1.21.6",

    "legacy_pandas":
        "1.3.5",

    "repository_commit":
        "79aef13b6db251f00907fc3daa84ddde9c188b8b",

    "gene_column":
        str(
            best_gene_col
        ),

    "rows":
        int(
            n_rows
        ),

    "authors_columns":
        int(
            n_columns
        ),

    "locked_analysis_columns":
        int(
            analysis_table.shape[1]
        ),

    "unique_genes":
        int(
            n_unique_genes
        ),

    "locked_asd_positive_genes":
        190,

    "asd_overlap":
        int(
            best_overlap
        ),

    "label_definition":
        "1 = member of locked Dhindsa monoallelic ASD 190 seed set; 0 = unlabeled/background",

    "important_label_note":
        "0 denotes unlabeled/background, not confirmed non-ASD",

    "authors_unmodified_complete_table":
        str(
            AUTHORS_COMPLETE_COPY
        ),

    "authors_unmodified_sha256":
        authors_sha,

    "analysis_locked_table":
        str(
            LOCKED_COMPLETE
        ),

    "analysis_locked_sha256":
        locked_sha,

    "compiler_log":
        str(
            LOG_FILE
        )
}


with open(
    MANIFEST_FILE,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        manifest,
        f,
        indent=2
    )


# ======================================================================================
# FINAL REPORT
# ======================================================================================

print("\n" + "=" * 115)
print("DHINDSA EXACT FEATURE REBUILD COMPLETE ✅")
print("=" * 115)

print(
    "Rows                :",
    n_rows
)

print(
    "Authors' columns    :",
    n_columns
)

print(
    "Unique genes        :",
    n_unique_genes
)

print(
    "ASD positive genes  :",
    "190/190"
)

print(
    "Background/unlabeled:",
    n_unique_genes - 190
)

print(
    "\nAuthors' unmodified table:"
)

print(
    AUTHORS_COMPLETE_COPY
)

print(
    "\nLocked analysis table:"
)

print(
    LOCKED_COMPLETE
)

print(
    "\nAuthors SHA256:"
)

print(
    authors_sha
)

print(
    "\nLocked SHA256:"
)

print(
    locked_sha
)

print(
    "\nCompiler log:"
)

print(
    LOG_FILE
)

print("\n" + "=" * 115)

print(
    "GO / NO-GO:"
)

print(
    "GO ✅"
    if best_overlap == 190
    else
    "NO-GO ❌"
)

print("=" * 115)
```

```python
# ======================================================================================
# ASD-ContextFusion
# TMSB15B DUPLICATE PROVENANCE AUDIT
#
# PURPOSE:
#   Determine exactly WHY TMSB15B appears twice in the Dhindsa processed table.
#
# NO ROW REMOVAL
# NO AVERAGING
# NO MODELING
# NO DATA MODIFICATION
# ======================================================================================

from pathlib import Path
import json
import numpy as np
import pandas as pd


ROOT = Path(r"E:\ASD_ContextFusion_Q1")

PROCESSED = (
    ROOT /
    "03_features" /
    "dhindsa_processed_context" /
    "dhindsa_asd_monoallelic_processed_authors.tsv"
)

COMPILED_DIR = (
    ROOT /
    "02_interim" /
    "asd_monoallelic" /
    "data" /
    "compiled_feature_tables"
)

COMPLETE = COMPILED_DIR / "complete_feature_table.tsv"
GENERIC = COMPILED_DIR / "generic_feature_table.tsv"
FILTERED = COMPILED_DIR / "filtered_by_disease_feature_table.tsv"

BCM = (
    ROOT /
    "01_raw" /
    "dhindsa_2025" /
    "Mantis-ml-NDD" /
    "mantis_ml" /
    "data" /
    "bcm_single_cell" /
    "bcm_df.tsv"
)

NT_META = (
    ROOT /
    "03_features" /
    "nt_v2" /
    "NTv2_embedding_row_metadata.tsv"
)

OUT_DIR = ROOT / "02_interim" / "duplicate_audit"
OUT_DIR.mkdir(parents=True, exist_ok=True)

TARGET = "TMSB15B"


for p in [
    PROCESSED,
    COMPLETE,
    GENERIC,
    FILTERED,
    BCM,
    NT_META
]:
    assert p.exists(), f"Missing:\n{p}"


print("=" * 118)
print("TMSB15B DUPLICATE PROVENANCE AUDIT")
print("=" * 118)


# ======================================================================================
# 1. LOAD TABLES
# ======================================================================================

processed = pd.read_csv(
    PROCESSED,
    sep="\t",
    low_memory=False
)

complete = pd.read_csv(
    COMPLETE,
    sep="\t",
    low_memory=False
)

generic = pd.read_csv(
    GENERIC,
    sep="\t",
    low_memory=False
)

filtered = pd.read_csv(
    FILTERED,
    sep="\t",
    low_memory=False
)

bcm = pd.read_csv(
    BCM,
    sep="\t",
    low_memory=False
)

nt_meta = pd.read_csv(
    NT_META,
    sep="\t",
    low_memory=False
)


print("\n[1/7] TABLE SHAPES")
print("-" * 118)

print("Processed :", processed.shape)
print("Complete  :", complete.shape)
print("Generic   :", generic.shape)
print("Filtered  :", filtered.shape)
print("BCM raw   :", bcm.shape)
print("NT meta   :", nt_meta.shape)


# ======================================================================================
# 2. COUNT TMSB15B THROUGH EACH DHINDSA STAGE
# ======================================================================================

print("\n[2/7] TMSB15B ROW COUNTS THROUGH PIPELINE")
print("-" * 118)


tables = {
    "processed": processed,
    "complete": complete,
    "generic": generic,
    "filtered": filtered
}


for name, df in tables.items():

    assert "Gene_Name" in df.columns

    n = int(
        (
            df["Gene_Name"]
            .astype(str)
            .str.strip()
            == TARGET
        ).sum()
    )

    print(
        f"{name:12s}: {n} row(s)"
    )


# ======================================================================================
# 3. EXACTLY WHICH PROCESSED FEATURES DIFFER?
# ======================================================================================

print("\n[3/7] DIFFERING PROCESSED FEATURES")
print("-" * 118)


p = (
    processed.loc[
        processed["Gene_Name"]
        .astype(str)
        .str.strip()
        .eq(TARGET)
    ]
    .reset_index(drop=True)
)


assert len(p) == 2


def same_value(a, b):

    if pd.isna(a) and pd.isna(b):
        return True

    try:
        return bool(
            np.isclose(
                float(a),
                float(b),
                equal_nan=True
            )
        )
    except Exception:
        return str(a) == str(b)


diff_rows = []


for col in processed.columns:

    a = p.loc[0, col]
    b = p.loc[1, col]

    if not same_value(a, b):

        diff_rows.append(
            {
                "feature": col,
                "row_1": a,
                "row_2": b
            }
        )


diff_df = pd.DataFrame(
    diff_rows
)


print(
    "Number of differing processed columns:",
    len(diff_df)
)


if len(diff_df):

    print(
        diff_df.to_string(
            index=False,
            max_rows=500
        )
    )


# ======================================================================================
# 4. MAP DIFFERING FEATURES TO AUTHORS' SOURCE BLOCKS
# ======================================================================================

print("\n[4/7] SOURCE-BLOCK MAPPING")
print("-" * 118)


generic_cols = set(
    generic.columns
)

filtered_cols = set(
    filtered.columns
)

bcm_cols = set(
    bcm.columns
)


source_rows = []


for feature in diff_df["feature"].tolist():

    memberships = []

    if feature in bcm_cols:
        memberships.append("BCM_raw")

    if feature in generic_cols:
        memberships.append("generic")

    if feature in filtered_cols:
        memberships.append("phenotype_filtered")

    source_rows.append(
        {
            "feature":
                feature,

            "source_membership":
                ";".join(memberships)
                if memberships
                else
                "not_directly_identified"
        }
    )


source_df = pd.DataFrame(
    source_rows
)


print(
    source_df.to_string(
        index=False,
        max_rows=500
    )
)


print("\nSource membership counts:")

print(
    source_df[
        "source_membership"
    ]
    .value_counts()
    .to_string()
)


# ======================================================================================
# 5. FIND TMSB15B IN RAW BCM TABLE
# ======================================================================================

print("\n[5/7] RAW BCM TMSB15B RECORDS")
print("-" * 118)


# First prefer canonical Gene_Name if present.
if "Gene_Name" in bcm.columns:

    bcm_gene_col = "Gene_Name"

else:

    bcm_gene_col = None

    for col in bcm.columns:

        values = (
            bcm[col]
            .astype(str)
            .str.strip()
        )

        if TARGET in set(values):

            bcm_gene_col = col
            break


print(
    "BCM gene column:",
    bcm_gene_col
)


if bcm_gene_col is None:

    raise RuntimeError(
        "Could not identify TMSB15B in raw BCM table."
    )


bcm_target = (
    bcm.loc[
        bcm[bcm_gene_col]
        .astype(str)
        .str.strip()
        .eq(TARGET)
    ]
    .copy()
    .reset_index(drop=True)
)


print(
    "Raw BCM TMSB15B rows:",
    len(bcm_target)
)


# Print likely identifier fields first.
identifier_cols = [
    c
    for c in bcm_target.columns
    if any(
        token in c.lower()
        for token in [
            "gene",
            "ensembl",
            "id",
            "symbol"
        ]
    )
]


if identifier_cols:

    print("\nIdentifier columns:")

    print(
        bcm_target[
            identifier_cols
        ].to_string(
            index=False
        )
    )


# Compare raw BCM rows.
if len(bcm_target) > 1:

    raw_diff = []

    for col in bcm_target.columns:

        vals = (
            bcm_target[col]
            .astype(object)
            .where(
                pd.notna(
                    bcm_target[col]
                ),
                "__NA__"
            )
            .astype(str)
            .unique()
        )

        if len(vals) > 1:

            raw_diff.append(col)


    print(
        "\nRaw BCM columns varying across TMSB15B rows:",
        len(raw_diff)
    )

    print(
        raw_diff
    )


# ======================================================================================
# 6. CHECK TMSB15B IN NT METADATA
# ======================================================================================

print("\n[6/7] NT-v2 REPRESENTATION")
print("-" * 118)


nt_hits = []


for col in nt_meta.columns:

    vals = (
        nt_meta[col]
        .astype(str)
        .str.strip()
    )

    matches = np.where(
        vals.eq(TARGET)
    )[0]

    if len(matches):

        nt_hits.append(
            (
                col,
                matches.tolist()
            )
        )


print(
    "NT metadata exact TMSB15B matches:"
)

print(
    nt_hits
)


# ======================================================================================
# 7. SAVE AUDIT
# ======================================================================================

print("\n[7/7] SAVE AUDIT")
print("-" * 118)


DIFF_OUT = (
    OUT_DIR /
    "TMSB15B_processed_differences.tsv"
)

SOURCE_OUT = (
    OUT_DIR /
    "TMSB15B_difference_source_mapping.tsv"
)

BCM_OUT = (
    OUT_DIR /
    "TMSB15B_raw_BCM_rows.tsv"
)


diff_df.to_csv(
    DIFF_OUT,
    sep="\t",
    index=False
)

source_df.to_csv(
    SOURCE_OUT,
    sep="\t",
    index=False
)

bcm_target.to_csv(
    BCM_OUT,
    sep="\t",
    index=False
)


summary = {
    "target_gene":
        TARGET,

    "processed_rows":
        int(
            (
                processed["Gene_Name"]
                .astype(str)
                .str.strip()
                == TARGET
            ).sum()
        ),

    "complete_rows":
        int(
            (
                complete["Gene_Name"]
                .astype(str)
                .str.strip()
                == TARGET
            ).sum()
        ),

    "generic_rows":
        int(
            (
                generic["Gene_Name"]
                .astype(str)
                .str.strip()
                == TARGET
            ).sum()
        ),

    "filtered_rows":
        int(
            (
                filtered["Gene_Name"]
                .astype(str)
                .str.strip()
                == TARGET
            ).sum()
        ),

    "raw_bcm_rows":
        int(
            len(
                bcm_target
            )
        ),

    "processed_differing_features":
        int(
            len(
                diff_df
            )
        ),

    "NT_exact_hits":
        nt_hits
}


SUMMARY_OUT = (
    OUT_DIR /
    "TMSB15B_duplicate_audit_summary.json"
)


with open(
    SUMMARY_OUT,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        summary,
        f,
        indent=2
    )


print("\n" + "=" * 118)
print("AUDIT COMPLETE ✅")
print("=" * 118)

print(
    "Processed differing features:",
    len(diff_df)
)

print(
    "Raw BCM TMSB15B rows:",
    len(bcm_target)
)

print(
    "NT exact matches:",
    nt_hits
)

print("\nNO DATA WERE MODIFIED.")
print("=" * 118)
```

```python
# ======================================================================================
# ASD-ContextFusion
# FRAMEWORK LOCK + ROW-PRESERVING DHINDSA / NT-v2 MASTER TABLE
#
# IMPORTANT
# ---------
# - Dhindsa processed rows are preserved exactly.
# - TMSB15B is NOT collapsed or averaged.
# - One NT embedding may map to multiple Dhindsa rows of the same gene.
# - No PU sampling yet.
# - No model training yet.
# ======================================================================================

from pathlib import Path
import json
import hashlib

import numpy as np
import pandas as pd


# ======================================================================================
# 0. PATHS
# ======================================================================================

ROOT = Path(r"E:\ASD_ContextFusion_Q1")

CONTEXT_FILE = (
    ROOT /
    "03_features" /
    "dhindsa_processed_context" /
    "dhindsa_asd_monoallelic_processed_authors.tsv"
)

SEED_FILE = (
    ROOT /
    "00_admin" /
    "dhindsa_2025_monoallelic_ASD_190.tsv"
)

NT_NPY = (
    ROOT /
    "03_features" /
    "nt_v2" /
    "NTv2_50m_full_transcript_embeddings_18252x512.npy"
)

NT_META_FILE = (
    ROOT /
    "03_features" /
    "nt_v2" /
    "NTv2_embedding_row_metadata.tsv"
)

OUT_DIR = (
    ROOT /
    "03_features" /
    "multimodal_master"
)

ADMIN_DIR = ROOT / "00_admin"

OUT_DIR.mkdir(parents=True, exist_ok=True)
ADMIN_DIR.mkdir(parents=True, exist_ok=True)


for p in [CONTEXT_FILE, SEED_FILE, NT_NPY, NT_META_FILE]:
    assert p.exists(), f"Missing:\n{p}"


print("=" * 118)
print("ASD-CONTEXTFUSION — FINAL FRAMEWORK + MULTIMODAL MASTER LOCK")
print("=" * 118)


# ======================================================================================
# 1. LOAD EXACT AUTHORS' PROCESSED CONTEXT
# ======================================================================================

print("\n[1/8] DHINDSA PROCESSED DATA")
print("-" * 118)

context = pd.read_csv(
    CONTEXT_FILE,
    sep="\t",
    low_memory=False
)

assert context.shape == (18627, 319)
assert "Gene_Name" in context.columns

context["Gene_Name"] = (
    context["Gene_Name"]
    .astype(str)
    .str.strip()
)

print("Rows        :", len(context))
print("Columns     :", context.shape[1])
print("Unique genes:", context["Gene_Name"].nunique())

assert context["Gene_Name"].nunique() == 18626


# Give every AUTHORS' row a permanent ID.
context.insert(
    0,
    "dhindsa_row_id",
    np.arange(len(context), dtype=np.int64)
)


# ======================================================================================
# 2. LOCK ASD SEEDS
# ======================================================================================

print("\n[2/8] ASD SEED SET")
print("-" * 118)

seeds = pd.read_csv(SEED_FILE, sep="\t")

seed_set = set(
    seeds["gene"]
    .astype(str)
    .str.strip()
)

assert len(seed_set) == 190

context["ASD_label"] = (
    context["Gene_Name"]
    .isin(seed_set)
    .astype(np.int8)
)

assert (
    context.loc[
        context["ASD_label"] == 1,
        "Gene_Name"
    ].nunique()
    == 190
)

print("Monoallelic ASD seeds:", 190)
print("Label definition     : locked Dhindsa ASD seed list")


# ======================================================================================
# 3. CONTEXT PREDICTOR DEFINITION
# ======================================================================================

print("\n[3/8] CONTEXT PREDICTORS")
print("-" * 118)

NON_PREDICTORS = {
    "dhindsa_row_id",
    "Gene_Name",
    "known_gene",
    "ASD_label",
    "ASD_ContextFusion_label"
}

context_predictors = [
    c for c in context.columns
    if c not in NON_PREDICTORS
]

assert len(context_predictors) == 317

print("Dhindsa processed predictors:", len(context_predictors))
print("known_gene used as feature   : NO")


# ======================================================================================
# 4. LOAD VERIFIED NT-v2
# ======================================================================================

print("\n[4/8] VERIFIED NT-v2")
print("-" * 118)

X_nt = np.load(
    NT_NPY,
    mmap_mode="r"
)

nt_meta = pd.read_csv(
    NT_META_FILE,
    sep="\t",
    low_memory=False
)

assert X_nt.shape == (18252, 512)
assert len(nt_meta) == 18252


EXPECTED_NT_SHA = (
    "91ca0065d01b45e220084c72da08dd0ee8b8fd066385c4444ad914a580f41f06"
)


def sha256_file(path):
    h = hashlib.sha256()

    with open(path, "rb") as f:
        while True:
            block = f.read(1024 * 1024)
            if not block:
                break
            h.update(block)

    return h.hexdigest()


nt_sha = sha256_file(NT_NPY)

assert nt_sha == EXPECTED_NT_SHA

print("NT shape :", X_nt.shape)
print("SHA256   : MATCH ✅")


# ======================================================================================
# 5. USE THE LOCKED GENCODE GENE SYMBOL
#
# Our transcript mapping was based on GENCODE, therefore
# gencode_gene_symbol is the prespecified mapping identifier.
# ======================================================================================

print("\n[5/8] NT GENE MAPPING")
print("-" * 118)

assert "gencode_gene_symbol" in nt_meta.columns

nt_meta["Gene_Name"] = (
    nt_meta["gencode_gene_symbol"]
    .astype(str)
    .str.strip()
)

assert nt_meta["Gene_Name"].is_unique


nt_lookup = pd.DataFrame({
    "Gene_Name": nt_meta["Gene_Name"],
    "NT_row_index": np.arange(
        len(nt_meta),
        dtype=np.int64
    )
})


print(
    "Unique NT genes:",
    nt_lookup["Gene_Name"].nunique()
)


# ======================================================================================
# 6. ROW-PRESERVING MERGE
#
# IMPORTANT:
# Context is LEFT SIDE.
# We do NOT group by Gene_Name.
# We do NOT drop TMSB15B.
# Thus two TMSB15B context rows receive the SAME NT_row_index.
# ======================================================================================

print("\n[6/8] ROW-PRESERVING MULTIMODAL MERGE")
print("-" * 118)

master = context.merge(
    nt_lookup,
    on="Gene_Name",
    how="inner",
    validate="many_to_one"
)


n_rows = len(master)
n_genes = master["Gene_Name"].nunique()

n_asd_rows = int(master["ASD_label"].sum())

n_asd_genes = (
    master.loc[
        master["ASD_label"] == 1,
        "Gene_Name"
    ].nunique()
)


print("Multimodal rows        :", n_rows)
print("Unique multimodal genes:", n_genes)
print("ASD rows               :", n_asd_rows)
print("Unique ASD genes       :", n_asd_genes)


# Expected:
# 18,252 unique NT-covered genes
# + one additional context row because TMSB15B has two Dhindsa rows.
assert n_genes == 18252
assert n_rows == 18253
assert n_asd_genes == 190


# TMSB15B must be preserved twice.
tmsb = master.loc[
    master["Gene_Name"] == "TMSB15B",
    ["dhindsa_row_id", "Gene_Name", "NT_row_index"]
]

print("\nTMSB15B mapping:")
print(tmsb.to_string(index=False))

assert len(tmsb) == 2
assert tmsb["NT_row_index"].nunique() == 1


print(
    "\nAuthors' duplicate rows preserved; "
    "single gene-level NT embedding referenced by both ✅"
)


# ======================================================================================
# 7. SAVE MASTER TABLE
# ======================================================================================

print("\n[7/8] SAVE MULTIMODAL MASTER")
print("-" * 118)

MASTER_FILE = (
    OUT_DIR /
    "dhindsa_NTv2_row_preserved_master.tsv"
)

FEATURE_FILE = (
    OUT_DIR /
    "dhindsa_317_context_predictors.tsv"
)


master.to_csv(
    MASTER_FILE,
    sep="\t",
    index=False
)

pd.DataFrame({
    "feature": context_predictors
}).to_csv(
    FEATURE_FILE,
    sep="\t",
    index=False
)


master_sha = sha256_file(MASTER_FILE)


print("Master:")
print(MASTER_FILE)

print("\nSHA256:")
print(master_sha)


# ======================================================================================
# 8. LOCK THE STUDY FRAMEWORK
# ======================================================================================

print("\n[8/8] FRAMEWORK PROTOCOL LOCK")
print("-" * 118)

protocol = {

    "project":
        "ASD-ContextFusion",

    "study_role_of_dhindsa_2025":
        (
            "benchmark dataset, contextual-feature backbone, "
            "and stochastic positive-unlabeled sampling framework"
        ),

    "novel_extension":
        (
            "contextual transcript-sequence representation using frozen "
            "Nucleotide Transformer v2 with modality-preserving late fusion"
        ),

    "dataset": {
        "monoallelic_ASD_seed_genes": 190,

        "dhindsa_processed_rows": 18627,

        "dhindsa_processed_unique_genes": 18626,

        "dhindsa_context_predictors": 317,

        "multimodal_rows": int(n_rows),

        "multimodal_unique_genes": int(n_genes),

        "multimodal_ASD_unique_genes": int(n_asd_genes),

        "background_definition":
            "unlabeled genes; not confirmed non-ASD"
    },

    "duplicate_policy": {
        "TMSB15B":
            (
                "Preserve both released Dhindsa context rows. "
                "Map the same gene-level NT-v2 embedding to both rows. "
                "No averaging, row deletion, or arbitrary collapse."
            )
    },

    "sequence": {
        "primary":
            "frozen NT-v2 50M 512-dimensional transcript embeddings",

        "comparators": [
            "transcript length + GC",
            "prespecified conventional 4-mer representation"
        ]
    },

    "models": {
        "full_context_benchmark":
            "XGBoost",

        "NTv2_branch":
            "L2-regularized logistic regression",

        "fusion":
            "cross-fitted logistic late stacking",

        "raw_context_plus_NT_concatenation":
            "supplementary ablation only; not proposed model"
    },

    "biological_ablation_policy":
        (
            "Developmental-expression and genetic/constraint analyses "
            "must use source-defined feature families from the Dhindsa "
            "paper/repository; no keyword-based or outcome-driven feature grouping."
        ),

    "PU_sampling": {
        "source":
            "Dhindsa 2025 / mantis-ml",

        "positive_seed_fraction":
            0.80,

        "unlabeled_to_positive_ratio":
            1.5,

        "cross_validation":
            "10-fold out-of-bag framework",

        "stochastic_iterations":
            10,

        "balanced_dataset_count":
            "derived by the authors' sampling machinery; not manually chosen",

        "same_saved_draws_for_all_modalities":
            True
    },

    "evaluation": {
        "primary_metric":
            "Average Precision / AUPRC",

        "secondary_metric":
            "AUROC",

        "additional":
            [
                "top-ranked gene enrichment",
                "gene-level paired bootstrap confidence intervals",
                "external/non-seed ASD validation"
            ]
    },

    "leakage_policy":
        (
            "Full Dhindsa feature set retained as benchmark. "
            "Seed-derived GO/PPI/disease-context features are evaluated "
            "separately in a leakage-sensitive robustness analysis."
        ),

    "prohibited": [
        "arbitrary fixed non-ASD controls",
        "manual hard-negative ratios",
        "result-driven model hunting",
        "test-fold-driven feature selection",
        "arbitrary duplicate-row deletion",
        "unprespecified raw feature concatenation as the proposed model"
    ],

    "NT_embedding_sha256":
        nt_sha,

    "multimodal_master_sha256":
        master_sha
}


PROTOCOL_FILE = (
    ADMIN_DIR /
    "ASD_ContextFusion_LOCKED_ANALYSIS_PROTOCOL.json"
)


with open(
    PROTOCOL_FILE,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        protocol,
        f,
        indent=2
    )


print("Protocol:")
print(PROTOCOL_FILE)


print("\n" + "=" * 118)
print("FRAMEWORK + MULTIMODAL MASTER LOCKED ✅")
print("=" * 118)

print("Dhindsa full rows       : 18,627")
print("Dhindsa predictors      : 317")
print("NT-v2 genes             : 18,252")
print("Multimodal rows         :", n_rows)
print("Multimodal unique genes :", n_genes)
print("ASD genes               :", n_asd_genes)
print("TMSB15B context rows    :", len(tmsb))
print("NT dimensions           : 512")

print("\nNEXT:")
print(
    "Generate ONE Dhindsa-style PU sampling/split manifest "
    "and reuse it identically for Context, NT, 4-mer and Fusion."
)

print("\nGO / NO-GO: GO ✅")
print("=" * 118)
```

```python
# ======================================================================================
# ASD-ContextFusion
# CORRECTED FINAL MULTIMODAL MASTER LOCK
#
# CRITICAL CORRECTION:
#   Dhindsa gene universe <-> NT metadata mapping uses ORIGINAL_GENE_SYMBOL.
#
# WHY:
#   original_gene_symbol = locked Dhindsa / source gene identity
#   gencode_gene_symbol  = GENCODE-resolved symbol after exact/rescue mapping
#
# Expected:
#   Direct exact GENCODE overlap : 16,991
#   HGNC/alias-rescued genes     :  1,261
#   Final NT-covered genes       : 18,252
#
# NO modeling
# NO control sampling
# NO feature selection
# NO duplicate collapsing
# ======================================================================================

from pathlib import Path
import hashlib
import json

import numpy as np
import pandas as pd


# ======================================================================================
# 0. PATHS
# ======================================================================================

ROOT = Path(r"E:\ASD_ContextFusion_Q1")

CONTEXT_FILE = (
    ROOT /
    "03_features" /
    "dhindsa_processed_context" /
    "dhindsa_asd_monoallelic_processed_authors.tsv"
)

SEED_FILE = (
    ROOT /
    "00_admin" /
    "dhindsa_2025_monoallelic_ASD_190.tsv"
)

NT_NPY = (
    ROOT /
    "03_features" /
    "nt_v2" /
    "NTv2_50m_full_transcript_embeddings_18252x512.npy"
)

NT_META_FILE = (
    ROOT /
    "03_features" /
    "nt_v2" /
    "NTv2_embedding_row_metadata.tsv"
)

OUT_DIR = (
    ROOT /
    "03_features" /
    "multimodal_master"
)

ADMIN_DIR = (
    ROOT /
    "00_admin"
)

OUT_DIR.mkdir(parents=True, exist_ok=True)
ADMIN_DIR.mkdir(parents=True, exist_ok=True)

for p in [
    CONTEXT_FILE,
    SEED_FILE,
    NT_NPY,
    NT_META_FILE
]:
    assert p.exists(), f"Missing:\n{p}"


print("=" * 118)
print("ASD-CONTEXTFUSION — CORRECTED FINAL MULTIMODAL MASTER")
print("=" * 118)


# ======================================================================================
# 1. LOAD AUTHORS' EXACT PROCESSED CONTEXT
# ======================================================================================

print("\n[1/9] DHINDSA PROCESSED CONTEXT")
print("-" * 118)

context = pd.read_csv(
    CONTEXT_FILE,
    sep="\t",
    low_memory=False
)

assert context.shape == (18627, 319)
assert "Gene_Name" in context.columns

context["Gene_Name"] = (
    context["Gene_Name"]
    .astype(str)
    .str.strip()
)

assert context["Gene_Name"].nunique() == 18626

print("Rows        :", len(context))
print("Columns     :", context.shape[1])
print("Unique genes:", context["Gene_Name"].nunique())


# permanent row ID
context.insert(
    0,
    "dhindsa_row_id",
    np.arange(
        len(context),
        dtype=np.int64
    )
)


# ======================================================================================
# 2. LOCK 190 MONOALLELIC ASD SEEDS
# ======================================================================================

print("\n[2/9] ASD SEED LOCK")
print("-" * 118)

seed_df = pd.read_csv(
    SEED_FILE,
    sep="\t"
)

assert "gene" in seed_df.columns

seed_set = set(
    seed_df["gene"]
    .astype(str)
    .str.strip()
)

assert len(seed_set) == 190

context["ASD_label"] = (
    context["Gene_Name"]
    .isin(seed_set)
    .astype(np.int8)
)

n_seed_genes = (
    context.loc[
        context["ASD_label"] == 1,
        "Gene_Name"
    ]
    .nunique()
)

assert n_seed_genes == 190

print("ASD seed genes:", n_seed_genes)


# ======================================================================================
# 3. CONTEXT PREDICTORS
# ======================================================================================

print("\n[3/9] DHINDSA PREDICTORS")
print("-" * 118)

NON_PREDICTORS = {
    "dhindsa_row_id",
    "Gene_Name",
    "known_gene",
    "ASD_label",
    "ASD_ContextFusion_label"
}

context_predictors = [
    c
    for c in context.columns
    if c not in NON_PREDICTORS
]

assert len(context_predictors) == 317

print(
    "Processed predictors:",
    len(context_predictors)
)

print(
    "known_gene excluded:",
    "YES ✅"
)


# ======================================================================================
# 4. LOAD VERIFIED NT-v2
# ======================================================================================

print("\n[4/9] VERIFIED NT-v2")
print("-" * 118)

X_nt = np.load(
    NT_NPY,
    mmap_mode="r"
)

nt_meta = pd.read_csv(
    NT_META_FILE,
    sep="\t",
    low_memory=False
)

assert X_nt.shape == (18252, 512)
assert len(nt_meta) == 18252

assert "original_gene_symbol" in nt_meta.columns
assert "gencode_gene_symbol" in nt_meta.columns


def sha256_file(path):

    h = hashlib.sha256()

    with open(path, "rb") as f:

        while True:

            block = f.read(
                1024 * 1024
            )

            if not block:
                break

            h.update(block)

    return h.hexdigest()


EXPECTED_NT_SHA = (
    "91ca0065d01b45e220084c72da08dd0ee8b8fd066385c4444ad914a580f41f06"
)

nt_sha = sha256_file(
    NT_NPY
)

assert nt_sha == EXPECTED_NT_SHA

print("NT matrix:", X_nt.shape)
print("SHA256   : MATCH ✅")


# ======================================================================================
# 5. AUDIT ORIGINAL vs GENCODE SYMBOL MAPPING
# ======================================================================================

print("\n[5/9] SYMBOL-MAPPING AUDIT")
print("-" * 118)

dhindsa_gene_set = set(
    context["Gene_Name"]
)


original_symbols = (
    nt_meta["original_gene_symbol"]
    .astype(str)
    .str.strip()
)

gencode_symbols = (
    nt_meta["gencode_gene_symbol"]
    .astype(str)
    .str.strip()
)


original_overlap = len(
    set(original_symbols) &
    dhindsa_gene_set
)

gencode_overlap = len(
    set(gencode_symbols) &
    dhindsa_gene_set
)


print(
    "original_gene_symbol overlap:",
    original_overlap
)

print(
    "gencode_gene_symbol overlap :",
    gencode_overlap
)

print(
    "Difference                  :",
    original_overlap - gencode_overlap
)


# This is the critical provenance check.
assert original_overlap == 18252, (
    f"Expected 18,252 locked original symbols, "
    f"found {original_overlap}"
)

assert gencode_overlap == 16991, (
    f"Expected 16,991 direct GENCODE-symbol overlap, "
    f"found {gencode_overlap}"
)

assert (
    original_overlap -
    gencode_overlap
) == 1261


print("\nMapping provenance reproduced:")
print("Direct/current GENCODE symbols : 16,991")
print("Alias/rescue contribution      :  1,261")
print("Final locked NT universe       : 18,252 ✅")


# ======================================================================================
# 6. LOCK THE CORRECT MERGE IDENTIFIER
# ======================================================================================

print("\n[6/9] CORRECT NT GENE IDENTIFIER")
print("-" * 118)

nt_meta["Gene_Name"] = (
    original_symbols
)


# one embedding per original locked gene
assert nt_meta["Gene_Name"].is_unique


nt_lookup = pd.DataFrame(
    {
        "Gene_Name":
            nt_meta["Gene_Name"],

        "NT_row_index":
            np.arange(
                len(nt_meta),
                dtype=np.int64
            )
    }
)


assert len(nt_lookup) == 18252
assert nt_lookup["Gene_Name"].nunique() == 18252


print(
    "Merge identifier: original_gene_symbol ✅"
)

print(
    "Unique NT genes  :",
    nt_lookup["Gene_Name"].nunique()
)


# ======================================================================================
# 7. ROW-PRESERVING MULTIMODAL MERGE
#
# Dhindsa rows stay exactly as released.
#
# TMSB15B:
#   two context rows
#   one gene-level NT embedding
#   same NT_row_index mapped to both context rows
#
# ======================================================================================

print("\n[7/9] ROW-PRESERVING MERGE")
print("-" * 118)

master = context.merge(
    nt_lookup,
    on="Gene_Name",
    how="inner",
    validate="many_to_one",
    sort=False
)


n_rows = len(master)

n_genes = (
    master["Gene_Name"]
    .nunique()
)

n_asd_genes = (
    master.loc[
        master["ASD_label"] == 1,
        "Gene_Name"
    ]
    .nunique()
)

n_unlabeled_genes = (
    n_genes -
    n_asd_genes
)


print(
    "Multimodal rows        :",
    n_rows
)

print(
    "Unique multimodal genes:",
    n_genes
)

print(
    "Unique ASD genes       :",
    n_asd_genes
)

print(
    "Unique unlabeled genes :",
    n_unlabeled_genes
)


# Expected:
# 18,252 unique genes
# plus one extra released context row for duplicated TMSB15B
assert n_genes == 18252
assert n_rows == 18253
assert n_asd_genes == 190
assert n_unlabeled_genes == 18062


# ======================================================================================
# 8. TMSB15B INTEGRITY CHECK
# ======================================================================================

print("\n[8/9] TMSB15B ROW-PRESERVATION CHECK")
print("-" * 118)

tmsb = (
    master.loc[
        master["Gene_Name"] == "TMSB15B",
        [
            "dhindsa_row_id",
            "Gene_Name",
            "NT_row_index",
            "ASD_label"
        ]
    ]
    .copy()
)


print(
    tmsb.to_string(
        index=False
    )
)


assert len(tmsb) == 2
assert tmsb["NT_row_index"].nunique() == 1


print(
    "\nTwo released context rows retained ✅"
)

print(
    "Same gene-level NT embedding mapped to both ✅"
)


# ======================================================================================
# 9. SAVE FINAL MASTER + PROTOCOL MANIFEST
# ======================================================================================

print("\n[9/9] SAVE FINAL LOCKED MASTER")
print("-" * 118)

MASTER_FILE = (
    OUT_DIR /
    "dhindsa_NTv2_row_preserved_master.tsv"
)

FEATURE_FILE = (
    OUT_DIR /
    "dhindsa_317_context_predictors.tsv"
)

ROW_MAP_FILE = (
    OUT_DIR /
    "dhindsa_NTv2_row_mapping.tsv"
)


master.to_csv(
    MASTER_FILE,
    sep="\t",
    index=False
)


pd.DataFrame(
    {
        "feature":
            context_predictors
    }
).to_csv(
    FEATURE_FILE,
    sep="\t",
    index=False
)


master[
    [
        "dhindsa_row_id",
        "Gene_Name",
        "ASD_label",
        "NT_row_index"
    ]
].to_csv(
    ROW_MAP_FILE,
    sep="\t",
    index=False
)


master_sha = sha256_file(
    MASTER_FILE
)

row_map_sha = sha256_file(
    ROW_MAP_FILE
)


manifest = {

    "project":
        "ASD-ContextFusion",

    "stage":
        "final row-preserving Dhindsa + NTv2 multimodal lock",

    "dhindsa": {

        "processed_rows":
            18627,

        "unique_genes":
            18626,

        "processed_predictors":
            317,

        "ASD_seeds":
            190
    },

    "NTv2": {

        "embedding_shape":
            [18252, 512],

        "embedding_sha256":
            nt_sha,

        "merge_identifier":
            "original_gene_symbol",

        "gencode_gene_symbol_role":
            (
                "GENCODE-resolved transcript-mapping provenance; "
                "not the Dhindsa merge identifier"
            ),

        "original_gene_symbol_role":
            (
                "locked source/Dhindsa gene identity used for "
                "cross-modal harmonization"
            ),

        "direct_gencode_symbol_overlap":
            16991,

        "rescued_symbol_count":
            1261,

        "final_locked_gene_count":
            18252
    },

    "multimodal": {

        "rows":
            int(n_rows),

        "unique_genes":
            int(n_genes),

        "ASD_unique_genes":
            int(n_asd_genes),

        "unlabeled_unique_genes":
            int(n_unlabeled_genes)
    },

    "TMSB15B": {

        "released_context_rows":
            2,

        "NT_embeddings":
            1,

        "policy":
            (
                "Preserve both released Dhindsa rows and map the "
                "same gene-level NT-v2 embedding to each row."
            )
    },

    "files": {

        "master":
            str(MASTER_FILE),

        "context_feature_list":
            str(FEATURE_FILE),

        "row_mapping":
            str(ROW_MAP_FILE)
    },

    "sha256": {

        "master":
            master_sha,

        "row_mapping":
            row_map_sha
    }
}


MANIFEST_FILE = (
    ADMIN_DIR /
    "ASD_ContextFusion_multimodal_master_manifest.json"
)


with open(
    MANIFEST_FILE,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        manifest,
        f,
        indent=2
    )


print("\n" + "=" * 118)
print("FINAL MULTIMODAL MASTER LOCKED ✅")
print("=" * 118)

print("Dhindsa processed rows : 18,627")
print("Dhindsa predictors     : 317")
print("NT-v2 unique genes     : 18,252")
print("NT dimensions          : 512")
print("Multimodal rows        :", n_rows)
print("Multimodal unique genes:", n_genes)
print("ASD genes              :", n_asd_genes)
print("Unlabeled genes        :", n_unlabeled_genes)
print("TMSB15B rows           :", len(tmsb))

print("\nMaster:")
print(MASTER_FILE)

print("\nManifest:")
print(MANIFEST_FILE)

print("\nGO / NO-GO: GO ✅")
print("=" * 118)
```

```python
# ======================================================================================
# ASD-ContextFusion
# SOURCE-LOCK: EXACT DHINDSA / MANTIS PU SAMPLING IMPLEMENTATION
#
# PURPOSE
# -------
# Find the EXACT repository code implementing:
#   - P:U balancing ratio = 1:1.5
#   - 80% positive seed sampling
#   - balanced dataset generation
#   - stratified 10-fold CV
#   - out-of-bag prediction
#   - stochastic iterations
#
# NO MODEL TRAINING
# NO NEW SAMPLING
# NO DATA MODIFICATION
# ======================================================================================

from pathlib import Path
import re
import json

ROOT = Path(r"E:\ASD_ContextFusion_Q1")

REPO = (
    ROOT /
    "01_raw" /
    "dhindsa_2025" /
    "Mantis-ml-NDD"
)

OUT_DIR = (
    ROOT /
    "00_admin"
)

OUT_DIR.mkdir(
    parents=True,
    exist_ok=True
)

assert REPO.exists()


print("=" * 118)
print("ASD-CONTEXTFUSION — EXACT DHINDSA PU SOURCE LOCK")
print("=" * 118)


# ======================================================================================
# 1. SEARCH TERMS
# ======================================================================================

search_patterns = [
    r"balancing_ratio",
    r"Balancing ratio",
    r"0\.8",
    r"80%",
    r"train_test",
    r"balanced",
    r"unlabel",
    r"known_gene",
    r"known_genes",
    r"StratifiedKFold",
    r"n_splits",
    r"kfold",
    r"out.of.bag",
    r"iteration",
    r"random_state",
    r"sample\(",
    r"shuffle"
]


# ======================================================================================
# 2. SEARCH ALL SCIENTIFIC PYTHON SOURCE FILES
# ======================================================================================

print("\n[1/4] SEARCHING REPOSITORY SOURCE")
print("-" * 118)


hits = []

python_files = list(
    REPO.rglob("*.py")
)


for path in python_files:

    # Ignore irrelevant cache/build content
    path_lower = str(path).lower()

    if any(
        x in path_lower
        for x in [
            "__pycache__",
            ".git",
            "site-packages"
        ]
    ):
        continue


    try:

        lines = path.read_text(
            encoding="utf-8",
            errors="replace"
        ).splitlines()

    except Exception:
        continue


    for line_no, line in enumerate(
        lines,
        start=1
    ):

        matched = [
            pattern
            for pattern in search_patterns
            if re.search(
                pattern,
                line,
                flags=re.IGNORECASE
            )
        ]


        if matched:

            hits.append(
                {
                    "file":
                        str(
                            path.relative_to(REPO)
                        ),

                    "line":
                        line_no,

                    "text":
                        line.strip(),

                    "patterns":
                        matched
                }
            )


print(
    "Python files scanned:",
    len(python_files)
)

print(
    "Relevant source hits:",
    len(hits)
)


# ======================================================================================
# 3. RANK LIKELY PU / SUPERVISED-LEARNING FILES
# ======================================================================================

print("\n[2/4] LIKELY PU / SAMPLING MODULES")
print("-" * 118)


file_scores = {}


strong_terms = [
    "balancing_ratio",
    "balanced",
    "train_test",
    "StratifiedKFold",
    "known_genes",
    "unlabeled",
    "iteration"
]


for h in hits:

    score = 0

    text = h["text"].lower()

    for term in strong_terms:

        if term.lower() in text:
            score += 1


    file_scores[
        h["file"]
    ] = (
        file_scores.get(
            h["file"],
            0
        )
        +
        score
    )


ranked_files = sorted(
    file_scores.items(),
    key=lambda x: x[1],
    reverse=True
)


for file, score in ranked_files[:20]:

    print(
        f"{score:3d}  {file}"
    )


# ======================================================================================
# 4. PRINT SOURCE CONTEXT AROUND HIGH-VALUE LINES
# ======================================================================================

print("\n[3/4] EXACT SOURCE SNIPPETS")
print("=" * 118)


high_value_regex = re.compile(
    r"balancing_ratio|balanced|StratifiedKFold|"
    r"known_genes|known_gene|unlabel|train_test|"
    r"0\.8|iteration",
    flags=re.IGNORECASE
)


selected_files = [
    file
    for file, score
    in ranked_files[:12]
    if score > 0
]


source_report = []


for relative_file in selected_files:

    path = (
        REPO /
        relative_file
    )

    lines = path.read_text(
        encoding="utf-8",
        errors="replace"
    ).splitlines()


    important_lines = []

    for i, line in enumerate(
        lines
    ):

        if high_value_regex.search(
            line
        ):

            important_lines.append(i)


    if not important_lines:
        continue


    # Merge nearby matches into ranges
    ranges = []

    for idx in important_lines:

        start = max(
            0,
            idx - 8
        )

        end = min(
            len(lines),
            idx + 12
        )


        if (
            ranges
            and
            start <= ranges[-1][1]
        ):

            ranges[-1] = (
                ranges[-1][0],
                max(
                    ranges[-1][1],
                    end
                )
            )

        else:

            ranges.append(
                (
                    start,
                    end
                )
            )


    print("\n" + "=" * 118)
    print(relative_file)
    print("=" * 118)


    file_entries = []


    for start, end in ranges[:15]:

        print(
            f"\n--- lines {start + 1}-{end} ---"
        )


        snippet_lines = []


        for j in range(
            start,
            end
        ):

            rendered = (
                f"{j + 1:5d}: "
                f"{lines[j]}"
            )

            print(
                rendered
            )

            snippet_lines.append(
                rendered
            )


        file_entries.append(
            {
                "start_line":
                    start + 1,

                "end_line":
                    end,

                "text":
                    "\n".join(
                        snippet_lines
                    )
            }
        )


    source_report.append(
        {
            "file":
                relative_file,

            "snippets":
                file_entries
        }
    )


# ======================================================================================
# 5. ALSO PRINT ASD CONFIG
# ======================================================================================

print("\n" + "=" * 118)
print("[ASD MONOALLELIC CONFIG]")
print("=" * 118)


CONFIG_FILE = (
    REPO /
    "mantis_ml" /
    "conf" /
    "asd_monoallelic_config.yaml"
)


config_text = CONFIG_FILE.read_text(
    encoding="utf-8",
    errors="replace"
)


print(
    config_text
)


# ======================================================================================
# 6. SAVE AUDIT
# ======================================================================================

print("\n[4/4] SAVING SOURCE-LOCK REPORT")
print("-" * 118)


REPORT_FILE = (
    OUT_DIR /
    "dhindsa_exact_PU_source_lock.json"
)


payload = {

    "repository":
        "bweido/Mantis-ml-NDD",

    "commit":
        "79aef13b6db251f00907fc3daa84ddde9c188b8b",

    "purpose":
        (
            "Locate exact code implementing Dhindsa/Mantis "
            "positive-unlabeled stochastic sampling and CV."
        ),

    "ranked_files":
        [
            {
                "file": f,
                "score": int(s)
            }
            for f, s
            in ranked_files
        ],

    "source_snippets":
        source_report,

    "config_file":
        str(CONFIG_FILE),

    "config_text":
        config_text
}


with open(
    REPORT_FILE,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        payload,
        f,
        indent=2
    )


print(
    "Saved:"
)

print(
    REPORT_FILE
)


print("\n" + "=" * 118)
print("PU SOURCE-LOCK AUDIT COMPLETE ✅")
print("=" * 118)

print(
    "NO SAMPLING WAS GENERATED."
)

print(
    "NO MODEL WAS TRAINED."
)

print(
    "\nNext step:"
)

print(
    "Call the authors' exact balanced-dataset generator "
    "and persist its generated gene/row IDs for reuse "
    "across Context, NT-v2, 4-mer and late fusion."
)

print("=" * 118)
```

```python
# ======================================================================================
# ASD-ContextFusion
# FINAL CORRECTED MULTIMODAL MASTER REBUILD
#
# INPUT:
#   Correct Dhindsa 190-seed processed table
#   18,627 rows x 318 columns
#   316 context predictors
#
# OUTPUT:
#   Row-preserved Dhindsa + verified NT-v2 master
#
# NO MODELING
# NO PU SAMPLING
# ======================================================================================

from pathlib import Path
import hashlib
import json

import numpy as np
import pandas as pd


ROOT = Path(r"E:\ASD_ContextFusion_Q1")

CONTEXT_FILE = (
    ROOT /
    "03_features" /
    "dhindsa_processed_context" /
    "dhindsa_asd_monoallelic_processed_authors.tsv"
)

NT_NPY = (
    ROOT /
    "03_features" /
    "nt_v2" /
    "NTv2_50m_full_transcript_embeddings_18252x512.npy"
)

NT_META_FILE = (
    ROOT /
    "03_features" /
    "nt_v2" /
    "NTv2_embedding_row_metadata.tsv"
)

OUT_DIR = (
    ROOT /
    "03_features" /
    "multimodal_master"
)

OUT_DIR.mkdir(
    parents=True,
    exist_ok=True
)


print("=" * 118)
print("ASD-CONTEXTFUSION — FINAL CORRECTED MULTIMODAL MASTER")
print("=" * 118)


# ======================================================================================
# 1. CORRECT DHINDSA CONTEXT
# ======================================================================================

context = pd.read_csv(
    CONTEXT_FILE,
    sep="\t",
    low_memory=False
)

assert context.shape == (18627, 318)

assert "Gene_Name" in context.columns
assert "known_gene" in context.columns

context["Gene_Name"] = (
    context["Gene_Name"]
    .astype(str)
    .str.strip()
)

context["known_gene"] = (
    pd.to_numeric(
        context["known_gene"],
        errors="raise"
    )
    .astype(np.int8)
)


assert context["Gene_Name"].nunique() == 18626

assert (
    context.loc[
        context["known_gene"] == 1,
        "Gene_Name"
    ].nunique()
    == 190
)


context.insert(
    0,
    "dhindsa_row_id",
    np.arange(
        len(context),
        dtype=np.int64
    )
)


context["ASD_label"] = (
    context["known_gene"]
    .astype(np.int8)
)


context_predictors = [
    c
    for c in context.columns
    if c not in [
        "dhindsa_row_id",
        "Gene_Name",
        "known_gene",
        "ASD_label"
    ]
]


assert len(context_predictors) == 316


print("\n[1/5] CORRECT DHINDSA CONTEXT")
print("-" * 118)

print("Rows       :", len(context))
print("Columns    :", context.shape[1])
print("Unique genes:", context["Gene_Name"].nunique())
print("ASD genes  :", 190)
print("Predictors :", len(context_predictors))


# ======================================================================================
# 2. VERIFIED NT-v2
# ======================================================================================

X_nt = np.load(
    NT_NPY,
    mmap_mode="r"
)

nt_meta = pd.read_csv(
    NT_META_FILE,
    sep="\t",
    low_memory=False
)


assert X_nt.shape == (18252, 512)
assert len(nt_meta) == 18252


def sha256_file(path):

    h = hashlib.sha256()

    with open(path, "rb") as f:

        while True:

            block = f.read(
                1024 * 1024
            )

            if not block:
                break

            h.update(block)

    return h.hexdigest()


EXPECTED_NT_SHA = (
    "91ca0065d01b45e220084c72da08dd0ee8b8fd066385c4444ad914a580f41f06"
)

nt_sha = sha256_file(
    NT_NPY
)

assert nt_sha == EXPECTED_NT_SHA


print("\n[2/5] VERIFIED NT-v2")
print("-" * 118)

print("Shape :", X_nt.shape)
print("SHA256: MATCH ✅")


# ======================================================================================
# 3. CORRECT ORIGINAL-GENE MAPPING
# ======================================================================================

assert "original_gene_symbol" in nt_meta.columns

nt_meta["Gene_Name"] = (
    nt_meta["original_gene_symbol"]
    .astype(str)
    .str.strip()
)

assert nt_meta["Gene_Name"].is_unique


nt_lookup = pd.DataFrame(
    {
        "Gene_Name":
            nt_meta["Gene_Name"],

        "NT_row_index":
            np.arange(
                len(nt_meta),
                dtype=np.int64
            )
    }
)


context_gene_set = set(
    context["Gene_Name"]
)

nt_gene_set = set(
    nt_lookup["Gene_Name"]
)


assert len(
    context_gene_set &
    nt_gene_set
) == 18252


print("\n[3/5] NT GENE MAPPING")
print("-" * 118)

print("NT unique genes:", len(nt_gene_set))
print("Context overlap:", 18252)


# ======================================================================================
# 4. ROW-PRESERVING MERGE
# ======================================================================================

master = context.merge(
    nt_lookup,
    on="Gene_Name",
    how="inner",
    validate="many_to_one",
    sort=False
)


n_rows = len(master)

n_genes = (
    master["Gene_Name"]
    .nunique()
)

n_asd = (
    master.loc[
        master["ASD_label"] == 1,
        "Gene_Name"
    ]
    .nunique()
)


assert n_rows == 18253
assert n_genes == 18252
assert n_asd == 190


tmsb = master.loc[
    master["Gene_Name"] == "TMSB15B",
    [
        "dhindsa_row_id",
        "Gene_Name",
        "NT_row_index",
        "ASD_label"
    ]
]


assert len(tmsb) == 2
assert tmsb["NT_row_index"].nunique() == 1


print("\n[4/5] ROW-PRESERVING MULTIMODAL MASTER")
print("-" * 118)

print("Rows        :", n_rows)
print("Unique genes:", n_genes)
print("ASD genes   :", n_asd)
print("Unlabeled   :", n_genes - n_asd)

print("\nTMSB15B:")
print(tmsb.to_string(index=False))


# ======================================================================================
# 5. SAVE FINAL MASTER
# ======================================================================================

MASTER_FILE = (
    OUT_DIR /
    "dhindsa_NTv2_row_preserved_master.tsv"
)

FEATURE_FILE = (
    OUT_DIR /
    "dhindsa_316_context_predictors.tsv"
)

ROW_MAP_FILE = (
    OUT_DIR /
    "dhindsa_NTv2_row_mapping.tsv"
)


master.to_csv(
    MASTER_FILE,
    sep="\t",
    index=False
)


pd.DataFrame(
    {
        "feature":
            context_predictors
    }
).to_csv(
    FEATURE_FILE,
    sep="\t",
    index=False
)


master[
    [
        "dhindsa_row_id",
        "Gene_Name",
        "known_gene",
        "ASD_label",
        "NT_row_index"
    ]
].to_csv(
    ROW_MAP_FILE,
    sep="\t",
    index=False
)


MASTER_SHA = sha256_file(
    MASTER_FILE
)


manifest = {

    "project":
        "ASD-ContextFusion",

    "context_preprocessing":
        "Dhindsa/Mantis exact released workflow with explicit released 190 monoallelic ASD seeds",

    "context_processed_shape":
        [18627, 318],

    "context_predictors":
        316,

    "context_sha256":
        "89a5061bb706034d3425f14469bea3179b1576aa23b955acc8225bf53d655098",

    "NT_shape":
        [18252, 512],

    "NT_sha256":
        nt_sha,

    "multimodal_rows":
        int(n_rows),

    "multimodal_unique_genes":
        int(n_genes),

    "ASD_genes":
        int(n_asd),

    "unlabelled_genes":
        int(n_genes - n_asd),

    "TMSB15B_policy":
        "two Dhindsa rows retained; one gene-level NT embedding mapped to both",

    "master_sha256":
        MASTER_SHA
}


MANIFEST_FILE = (
    ROOT /
    "00_admin" /
    "ASD_ContextFusion_FINAL_multimodal_master_manifest.json"
)


with open(
    MANIFEST_FILE,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        manifest,
        f,
        indent=2
    )


# Remove invalidation marker only after successful rebuild
INVALID_MARKER = (
    OUT_DIR /
    "INVALIDATED_BY_EXACT_190_SEED_REBUILD.txt"
)

if INVALID_MARKER.exists():
    INVALID_MARKER.unlink()


print("\n[5/5] FINAL SAVE")
print("-" * 118)

print("Master:")
print(MASTER_FILE)

print("\nPredictor list:")
print(FEATURE_FILE)

print("\nSHA256:")
print(MASTER_SHA)


print("\n" + "=" * 118)
print("FINAL MULTIMODAL MASTER REBUILT ✅")
print("=" * 118)

print("Context predictors :", 316)
print("NT dimensions      :", 512)
print("Multimodal rows    :", n_rows)
print("Unique genes       :", n_genes)
print("ASD genes          :", n_asd)
print("Unlabeled genes    :", n_genes - n_asd)
print("TMSB15B rows       :", len(tmsb))

print("\nGO / NO-GO: GO ✅")
print("=" * 118)
```

```python
# ======================================================================================
# ASD-ContextFusion
# SCIKIT-LEARN COMPATIBILITY SHIM FOR AUTHORS' STRATIFIEDKFOLD CALL
#
# Does NOT change:
#   - shuffle=False
#   - number of folds
#   - sampling
#   - PU ratio
#   - seed sampling
#
# It only removes a random_state argument that has no effect when shuffle=False.
# ======================================================================================

from pathlib import Path
import subprocess
import os

ROOT = Path(r"E:\ASD_ContextFusion_Q1")

REPO = (
    ROOT /
    "01_raw" /
    "dhindsa_2025" /
    "Mantis-ml-NDD"
)

LEGACY_PYTHON = (
    ROOT /
    "envs" /
    "mantis37" /
    "python.exe"
)

RUNNER = (
    ROOT /
    "00_admin" /
    "run_exact_dhindsa_PU_common_universe.py"
)

LOG_FILE = (
    ROOT /
    "08_logs" /
    "dhindsa_PU_common_universe.log"
)

PARTITION_DIR = (
    ROOT /
    "04_models" /
    "checkpoints" /
    "dhindsa_PU_common_universe"
)

assert RUNNER.exists()
assert LEGACY_PYTHON.exists()


print("=" * 110)
print("APPLYING SCIKIT-LEARN COMPATIBILITY SHIM")
print("=" * 110)


text = RUNNER.read_text(
    encoding="utf-8"
)

PATCH_MARKER = "# ASD_CONTEXTFUSION_SKLEARN_COMPAT_PATCH"


if PATCH_MARKER not in text:

    target = (
        "from mantis_ml.modules.supervised_learn.core.prepare_train_test_sets "
        "import PrepareTrainTestSets"
    )

    assert target in text, (
        "Could not locate PrepareTrainTestSets import in runner."
    )

    patch = r'''

# ASD_CONTEXTFUSION_SKLEARN_COMPAT_PATCH
# ----------------------------------------------------------------------
# Authors' released code calls:
#
# StratifiedKFold(
#     n_splits=...,
#     random_state=<value>,
#     shuffle=False
# )
#
# Some scikit-learn versions reject random_state when shuffle=False.
# Since random_state has NO effect when shuffle=False, we discard only
# that ineffective argument while preserving the authors' split logic.
# ----------------------------------------------------------------------

import mantis_ml.modules.supervised_learn.core.balanced_datasets_random_generator as _bdg

_original_StratifiedKFold = _bdg.StratifiedKFold


def _compat_StratifiedKFold(*args, **kwargs):

    if kwargs.get("shuffle", False) is False:
        kwargs["random_state"] = None

    return _original_StratifiedKFold(
        *args,
        **kwargs
    )


_bdg.StratifiedKFold = _compat_StratifiedKFold

print(
    "scikit-learn StratifiedKFold compatibility shim active [OK]"
)
# ----------------------------------------------------------------------
'''

    text = text.replace(
        target,
        target + patch
    )

    RUNNER.write_text(
        text,
        encoding="utf-8"
    )

    print("Compatibility patch inserted: [OK]")

else:

    print("Compatibility patch already present: [OK]")


# ======================================================================================
# Remove ONLY partial partition artifacts from this failed iteration-1 run
# ======================================================================================

if PARTITION_DIR.exists():

    for pattern in [
        "iteration_*_test_fold_membership.tsv.gz",
        "iteration_*_balanced_dataset_audit.tsv",
        "PU_iteration_summary.tsv"
    ]:

        for p in PARTITION_DIR.glob(pattern):
            p.unlink()


print("Partial partition outputs cleared: [OK]")


# ======================================================================================
# RERUN
# ======================================================================================

env = os.environ.copy()

env["PYTHONIOENCODING"] = "utf-8"
env["PYTHONUTF8"] = "1"


print("\n" + "=" * 110)
print("RESTARTING EXACT DHINDSA/MANTIS PU PARTITION GENERATOR")
print("=" * 110)


with open(
    LOG_FILE,
    "w",
    encoding="utf-8"
) as log:

    process = subprocess.Popen(
        [
            str(LEGACY_PYTHON),
            "-u",
            str(RUNNER)
        ],
        cwd=str(REPO),
        stdout=subprocess.PIPE,
        stderr=subprocess.STDOUT,
        text=True,
        encoding="utf-8",
        errors="replace",
        bufsize=1,
        env=env
    )


    for line in process.stdout:

        print(
            line,
            end=""
        )

        log.write(
            line
        )

        log.flush()


    return_code = process.wait()


print("\n" + "=" * 110)

if return_code == 0:

    print("PU PARTITION GENERATION COMPLETED [OK]")

else:

    print(
        "PU PARTITION GENERATION STOPPED "
        f"(exit code {return_code})"
    )

    print("\nPaste the FINAL traceback block only.")

print("=" * 110)
```

```python
# ======================================================================================
# ASD-ContextFusion
# PRIMARY MODELING STAGE
#
# Branch A: Dhindsa Context -> AUTHORS' RELEASED XGBOOST PARAMETERS
# Branch B: Frozen NT-v2 512D -> STANDARDIZED L2 LOGISTIC LINEAR PROBE
#
# SAME FROZEN PU PARTITIONS FOR BOTH BRANCHES
#
# - no new sampling
# - no feature selection
# - no NT regeneration
# - no model hunting
# - checkpoint/resume by stochastic iteration
# ======================================================================================

from pathlib import Path
import subprocess
import os
import textwrap

ROOT = Path(r"E:\ASD_ContextFusion_Q1")

REPO = (
    ROOT /
    "01_raw" /
    "dhindsa_2025" /
    "Mantis-ml-NDD"
)

LEGACY_PYTHON = (
    ROOT /
    "envs" /
    "mantis37" /
    "python.exe"
)

MASTER_FILE = (
    ROOT /
    "03_features" /
    "multimodal_master" /
    "dhindsa_NTv2_row_preserved_master.tsv"
)

PREDICTOR_FILE = (
    ROOT /
    "03_features" /
    "multimodal_master" /
    "dhindsa_316_context_predictors.tsv"
)

NT_FILE = (
    ROOT /
    "03_features" /
    "nt_v2" /
    "NTv2_50m_full_transcript_embeddings_18252x512.npy"
)

PARTITION_DIR = (
    ROOT /
    "04_models" /
    "checkpoints" /
    "dhindsa_PU_common_universe"
)

OUT_DIR = (
    ROOT /
    "05_results" /
    "proposed" /
    "primary_context_NTv2"
)

CHECKPOINT_DIR = (
    ROOT /
    "04_models" /
    "checkpoints" /
    "primary_context_NTv2"
)

RUNNER = (
    ROOT /
    "00_admin" /
    "run_primary_context_NTv2_frozen_PU.py"
)

LOG_FILE = (
    ROOT /
    "08_logs" /
    "primary_context_NTv2_frozen_PU.log"
)


for p in [
    REPO,
    LEGACY_PYTHON,
    MASTER_FILE,
    PREDICTOR_FILE,
    NT_FILE,
    PARTITION_DIR
]:
    assert p.exists(), f"Missing: {p}"


OUT_DIR.mkdir(
    parents=True,
    exist_ok=True
)

CHECKPOINT_DIR.mkdir(
    parents=True,
    exist_ok=True
)

LOG_FILE.parent.mkdir(
    parents=True,
    exist_ok=True
)


runner_code = rf'''
from __future__ import print_function

from pathlib import Path
import warnings
import json
import time

warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd

import xgboost as xgb

from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    roc_auc_score,
    average_precision_score
)


ROOT = Path(r"{ROOT}")

MASTER_FILE = Path(r"{MASTER_FILE}")
PREDICTOR_FILE = Path(r"{PREDICTOR_FILE}")
NT_FILE = Path(r"{NT_FILE}")
PARTITION_DIR = Path(r"{PARTITION_DIR}")

OUT_DIR = Path(r"{OUT_DIR}")
CHECKPOINT_DIR = Path(r"{CHECKPOINT_DIR}")


OUT_DIR.mkdir(
    parents=True,
    exist_ok=True
)

CHECKPOINT_DIR.mkdir(
    parents=True,
    exist_ok=True
)


print("=" * 112)
print("ASD-CONTEXTFUSION - PRIMARY CONTEXT + NT-v2 MODELING")
print("=" * 112)


# ==================================================================================
# 1. LOAD LOCKED MASTER
# ==================================================================================

master = pd.read_csv(
    str(MASTER_FILE),
    sep="\t",
    low_memory=False
)

predictor_df = pd.read_csv(
    str(PREDICTOR_FILE),
    sep="\t"
)

context_features = (
    predictor_df["feature"]
    .astype(str)
    .tolist()
)


assert len(master) == 18253
assert master["Gene_Name"].nunique() == 18252
assert len(context_features) == 316

assert all(
    c in master.columns
    for c in context_features
)

assert "dhindsa_row_id" in master.columns
assert "NT_row_index" in master.columns
assert "known_gene" in master.columns


master["dhindsa_row_id"] = (
    master["dhindsa_row_id"]
    .astype(int)
)

master["NT_row_index"] = (
    master["NT_row_index"]
    .astype(int)
)

master["known_gene"] = (
    master["known_gene"]
    .astype(int)
)


assert master["dhindsa_row_id"].is_unique


master_idx = master.set_index(
    "dhindsa_row_id",
    drop=False
)


X_nt = np.load(
    str(NT_FILE),
    mmap_mode="r"
)

assert X_nt.shape == (18252, 512)


print("")
print("[DATA]")
print("Multimodal rows    :", len(master))
print("Unique genes       :", master["Gene_Name"].nunique())
print("Context predictors :", len(context_features))
print("NT-v2 dimensions   :", X_nt.shape[1])
print("ASD genes          :", master.loc[master.known_gene == 1, "Gene_Name"].nunique())


# ==================================================================================
# 2. LOCK MODEL SPECIFICATIONS
# ==================================================================================
#
# EXACT released Mantis XGBoost values from ensemble_lib.py.
#
# No tuning and no result-driven modifications.
# ==================================================================================

XGB_PARAMS = {{
    "learning_rate": 0.01,
    "n_estimators": 300,
    "max_depth": 5,
    "min_child_weight": 3,
    "gamma": 0,
    "subsample": 0.8,
    "colsample_bytree": 0.8,
    "objective": "binary:logistic",
    "nthread": 1,
    "scale_pos_weight": 1
}}


# NT-v2 prespecified linear probe:
#
# StandardScaler:
#   fitted ONLY on training fold.
#
# Logistic Regression:
#   L2 regularization
#   C=1.0 (standard sklearn regularization strength)
#   no tuning against results.
#
NT_LOGISTIC_PARAMS = {{
    "penalty": "l2",
    "C": 1.0,
    "solver": "lbfgs",
    "max_iter": 2000
}}


print("")
print("[MODELS]")
print("Context : XGBoost - authors released parameters")
print("NT-v2   : StandardScaler + L2 logistic linear probe")
print("")
print("XGBoost parameters:")
print(XGB_PARAMS)
print("")
print("NT parameters:")
print(NT_LOGISTIC_PARAMS)


# ==================================================================================
# 3. TRAIN OVER EXACT FROZEN PU PARTITIONS
# ==================================================================================

all_metric_files = []
all_prediction_files = []


for iteration in range(1, 11):

    membership_file = (
        PARTITION_DIR /
        (
            "iteration_%02d_test_fold_membership.tsv.gz"
            % iteration
        )
    )

    pred_file = (
        CHECKPOINT_DIR /
        (
            "iteration_%02d_OOF_predictions.tsv.gz"
            % iteration
        )
    )

    metric_file = (
        CHECKPOINT_DIR /
        (
            "iteration_%02d_fold_metrics.tsv"
            % iteration
        )
    )


    assert membership_file.exists()


    # --------------------------------------------------------------------------
    # RESUME MODE
    # --------------------------------------------------------------------------

    if pred_file.exists() and metric_file.exists():

        existing_pred = pd.read_csv(
            str(pred_file),
            sep="\t",
            compression="gzip"
        )

        existing_metric = pd.read_csv(
            str(metric_file),
            sep="\t"
        )

        if (
            len(existing_metric) == 790
            and
            existing_metric["balanced_dataset"].nunique() == 79
            and
            existing_metric["fold"].nunique() == 10
            and
            len(existing_pred) > 0
        ):

            print("")
            print(
                "Iteration %02d already complete - SKIPPING"
                % iteration
            )

            all_metric_files.append(
                metric_file
            )

            all_prediction_files.append(
                pred_file
            )

            continue


    print("")
    print("=" * 112)
    print(
        "STOCHASTIC ITERATION %02d"
        % iteration
    )
    print("=" * 112)


    membership = pd.read_csv(
        str(membership_file),
        sep="\t",
        compression="gzip"
    )


    assert membership["iteration"].nunique() == 1
    assert int(membership["iteration"].iloc[0]) == iteration

    assert membership["balanced_dataset"].nunique() == 79
    assert membership["fold"].nunique() == 10


    iteration_predictions = []
    iteration_metrics = []

    split_counter = 0

    iter_start = time.time()


    # ==========================================================================
    # EACH BALANCED DATASET
    # ==========================================================================

    for balanced_dataset, bal in membership.groupby(
        "balanced_dataset",
        sort=True
    ):

        balanced_dataset = int(
            balanced_dataset
        )


        # Across its 10 test folds, this is the complete balanced dataset.
        balanced_ids = (
            bal["dhindsa_row_id"]
            .astype(int)
            .tolist()
        )


        assert len(balanced_ids) == len(
            set(balanced_ids)
        )

        assert 380 <= len(balanced_ids) <= 383


        # ======================================================================
        # EACH STRATIFIED FOLD
        # ======================================================================

        for fold in range(10):

            split_counter += 1


            test_part = bal.loc[
                bal["fold"] == fold
            ].copy()


            test_ids = (
                test_part[
                    "dhindsa_row_id"
                ]
                .astype(int)
                .tolist()
            )


            test_id_set = set(
                test_ids
            )


            train_ids = [
                rid
                for rid in balanced_ids
                if rid not in test_id_set
            ]


            assert (
                set(train_ids)
                .isdisjoint(
                    test_id_set
                )
            )


            train_df = master_idx.loc[
                train_ids
            ]

            test_df = master_idx.loc[
                test_ids
            ]


            y_train = (
                train_df["known_gene"]
                .to_numpy(
                    dtype=np.int32
                )
            )

            y_test = (
                test_df["known_gene"]
                .to_numpy(
                    dtype=np.int32
                )
            )


            assert set(
                np.unique(y_train)
            ) == {{0, 1}}

            assert set(
                np.unique(y_test)
            ) == {{0, 1}}


            # ==================================================================
            # A. AUTHORS' CONTEXT XGBOOST
            # ==================================================================

            Xc_train = (
                train_df[
                    context_features
                ]
                .to_numpy(
                    dtype=np.float32
                )
            )

            Xc_test = (
                test_df[
                    context_features
                ]
                .to_numpy(
                    dtype=np.float32
                )
            )


            context_model = (
                xgb.XGBClassifier(
                    **XGB_PARAMS
                )
            )


            context_model.fit(
                Xc_train,
                y_train
            )


            context_prob = (
                context_model
                .predict_proba(
                    Xc_test
                )[:, 1]
            )


            # ==================================================================
            # B. NT-v2 L2 LINEAR PROBE
            # ==================================================================

            nt_train_index = (
                train_df[
                    "NT_row_index"
                ]
                .to_numpy(
                    dtype=np.int64
                )
            )

            nt_test_index = (
                test_df[
                    "NT_row_index"
                ]
                .to_numpy(
                    dtype=np.int64
                )
            )


            Xn_train = np.asarray(
                X_nt[
                    nt_train_index
                ],
                dtype=np.float32
            )

            Xn_test = np.asarray(
                X_nt[
                    nt_test_index
                ],
                dtype=np.float32
            )


            scaler = StandardScaler()

            Xn_train_scaled = (
                scaler.fit_transform(
                    Xn_train
                )
            )

            Xn_test_scaled = (
                scaler.transform(
                    Xn_test
                )
            )


            nt_model = (
                LogisticRegression(
                    **NT_LOGISTIC_PARAMS
                )
            )


            nt_model.fit(
                Xn_train_scaled,
                y_train
            )


            nt_prob = (
                nt_model
                .predict_proba(
                    Xn_test_scaled
                )[:, 1]
            )


            # ==================================================================
            # METRICS
            # ==================================================================

            context_roc = roc_auc_score(
                y_test,
                context_prob
            )

            context_ap = average_precision_score(
                y_test,
                context_prob
            )

            nt_roc = roc_auc_score(
                y_test,
                nt_prob
            )

            nt_ap = average_precision_score(
                y_test,
                nt_prob
            )


            iteration_metrics.append(
                {{
                    "iteration":
                        iteration,

                    "balanced_dataset":
                        balanced_dataset,

                    "fold":
                        fold,

                    "n_train":
                        len(train_ids),

                    "n_test":
                        len(test_ids),

                    "test_positive":
                        int(y_test.sum()),

                    "test_unlabelled":
                        int(
                            len(y_test)
                            - y_test.sum()
                        ),

                    "Context_AUROC":
                        float(context_roc),

                    "Context_AP":
                        float(context_ap),

                    "NT_AUROC":
                        float(nt_roc),

                    "NT_AP":
                        float(nt_ap),

                    "Delta_AUROC_NT_minus_Context":
                        float(
                            nt_roc
                            - context_roc
                        ),

                    "Delta_AP_NT_minus_Context":
                        float(
                            nt_ap
                            - context_ap
                        )
                }}
            )


            # ==================================================================
            # SAVE EVERY OOF TEST PREDICTION
            # ==================================================================

            test_meta = test_df[
                [
                    "dhindsa_row_id",
                    "Gene_Name",
                    "known_gene",
                    "NT_row_index"
                ]
            ].copy()


            for j in range(
                len(test_meta)
            ):

                row = (
                    test_meta.iloc[j]
                )

                iteration_predictions.append(
                    {{
                        "iteration":
                            iteration,

                        "balanced_dataset":
                            balanced_dataset,

                        "fold":
                            fold,

                        "dhindsa_row_id":
                            int(
                                row[
                                    "dhindsa_row_id"
                                ]
                            ),

                        "Gene_Name":
                            str(
                                row[
                                    "Gene_Name"
                                ]
                            ),

                        "known_gene":
                            int(
                                row[
                                    "known_gene"
                                ]
                            ),

                        "NT_row_index":
                            int(
                                row[
                                    "NT_row_index"
                                ]
                            ),

                        "Context_probability":
                            float(
                                context_prob[j]
                            ),

                        "NT_probability":
                            float(
                                nt_prob[j]
                            )
                    }}
                )


            if (
                split_counter % 100
                == 0
            ):

                elapsed = (
                    time.time()
                    - iter_start
                )

                print(
                    "Iteration %02d | %d / 790 splits | %.1f min"
                    % (
                        iteration,
                        split_counter,
                        elapsed / 60.0
                    )
                )


    # ==========================================================================
    # ITERATION CHECKPOINT
    # ==========================================================================

    metrics_df = pd.DataFrame(
        iteration_metrics
    )

    predictions_df = pd.DataFrame(
        iteration_predictions
    )


    assert len(metrics_df) == 790

    assert (
        metrics_df[
            "balanced_dataset"
        ].nunique()
        == 79
    )

    assert (
        metrics_df[
            "fold"
        ].nunique()
        == 10
    )


    # Every unlabeled row is OOF tested once per stochastic iteration.
    unlabeled_pred = (
        predictions_df.loc[
            predictions_df[
                "known_gene"
            ] == 0
        ]
    )

    assert len(
        unlabeled_pred
    ) == 18063

    assert (
        unlabeled_pred[
            "dhindsa_row_id"
        ].nunique()
        == 18063
    )


    predictions_df.to_csv(
        str(pred_file),
        sep="\t",
        index=False,
        compression="gzip"
    )

    metrics_df.to_csv(
        str(metric_file),
        sep="\t",
        index=False
    )


    all_metric_files.append(
        metric_file
    )

    all_prediction_files.append(
        pred_file
    )


    print("")
    print(
        "Iteration %02d completed."
        % iteration
    )

    print(
        "Context AUROC: %.4f | AP: %.4f"
        % (
            metrics_df[
                "Context_AUROC"
            ].mean(),

            metrics_df[
                "Context_AP"
            ].mean()
        )
    )

    print(
        "NT-v2   AUROC: %.4f | AP: %.4f"
        % (
            metrics_df[
                "NT_AUROC"
            ].mean(),

            metrics_df[
                "NT_AP"
            ].mean()
        )
    )

    print(
        "Checkpoint saved:",
        pred_file
    )


# ==================================================================================
# 4. COMBINE ALL 10 ITERATIONS
# ==================================================================================

print("")
print("=" * 112)
print("COMBINING ALL FROZEN PU RESULTS")
print("=" * 112)


all_metrics = pd.concat(
    [
        pd.read_csv(
            str(p),
            sep="\t"
        )
        for p in all_metric_files
    ],
    ignore_index=True
)


all_predictions = pd.concat(
    [
        pd.read_csv(
            str(p),
            sep="\t",
            compression="gzip"
        )
        for p in all_prediction_files
    ],
    ignore_index=True
)


assert len(all_metrics) == 7900


ALL_METRIC_FILE = (
    OUT_DIR /
    "Context_NTv2_all_7900_fold_metrics.tsv"
)

ALL_PRED_FILE = (
    OUT_DIR /
    "Context_NTv2_all_OOF_predictions.tsv.gz"
)


all_metrics.to_csv(
    str(ALL_METRIC_FILE),
    sep="\t",
    index=False
)

all_predictions.to_csv(
    str(ALL_PRED_FILE),
    sep="\t",
    index=False,
    compression="gzip"
)


# ==================================================================================
# 5. SPLIT-LEVEL SUMMARY
# ==================================================================================

split_summary = pd.DataFrame(
    [
        {{
            "Model":
                "Dhindsa_Context_XGBoost",

            "AUROC_mean":
                all_metrics[
                    "Context_AUROC"
                ].mean(),

            "AUROC_sd":
                all_metrics[
                    "Context_AUROC"
                ].std(),

            "AP_mean":
                all_metrics[
                    "Context_AP"
                ].mean(),

            "AP_sd":
                all_metrics[
                    "Context_AP"
                ].std()
        }},

        {{
            "Model":
                "NTv2_L2_Logistic",

            "AUROC_mean":
                all_metrics[
                    "NT_AUROC"
                ].mean(),

            "AUROC_sd":
                all_metrics[
                    "NT_AUROC"
                ].std(),

            "AP_mean":
                all_metrics[
                    "NT_AP"
                ].mean(),

            "AP_sd":
                all_metrics[
                    "NT_AP"
                ].std()
        }}
    ]
)


SPLIT_SUMMARY_FILE = (
    OUT_DIR /
    "Context_NTv2_split_level_summary.tsv"
)


split_summary.to_csv(
    str(SPLIT_SUMMARY_FILE),
    sep="\t",
    index=False
)


# ==================================================================================
# 6. GENE-LEVEL OUT-OF-FOLD AGGREGATION
# ==================================================================================

gene_predictions = (
    all_predictions
    .groupby(
        [
            "Gene_Name",
            "known_gene"
        ],
        as_index=False
    )
    .agg(
        Context_probability=(
            "Context_probability",
            "mean"
        ),

        NT_probability=(
            "NT_probability",
            "mean"
        ),

        OOF_prediction_count=(
            "Context_probability",
            "size"
        )
    )
)


assert (
    gene_predictions[
        "Gene_Name"
    ].nunique()
    == 18252
)

assert (
    gene_predictions.loc[
        gene_predictions[
            "known_gene"
        ] == 1,
        "Gene_Name"
    ].nunique()
    == 190
)


GENE_PRED_FILE = (
    OUT_DIR /
    "Context_NTv2_gene_level_OOF_scores.tsv"
)


gene_predictions.to_csv(
    str(GENE_PRED_FILE),
    sep="\t",
    index=False
)


y_gene = (
    gene_predictions[
        "known_gene"
    ]
    .to_numpy(
        dtype=np.int32
    )
)


gene_context_roc = roc_auc_score(
    y_gene,
    gene_predictions[
        "Context_probability"
    ]
)

gene_context_ap = average_precision_score(
    y_gene,
    gene_predictions[
        "Context_probability"
    ]
)

gene_nt_roc = roc_auc_score(
    y_gene,
    gene_predictions[
        "NT_probability"
    ]
)

gene_nt_ap = average_precision_score(
    y_gene,
    gene_predictions[
        "NT_probability"
    ]
)


gene_summary = pd.DataFrame(
    [
        {{
            "Model":
                "Dhindsa_Context_XGBoost",

            "Gene_level_AUROC":
                gene_context_roc,

            "Gene_level_AP":
                gene_context_ap
        }},

        {{
            "Model":
                "NTv2_L2_Logistic",

            "Gene_level_AUROC":
                gene_nt_roc,

            "Gene_level_AP":
                gene_nt_ap
        }}
    ]
)


GENE_SUMMARY_FILE = (
    OUT_DIR /
    "Context_NTv2_gene_level_summary.tsv"
)


gene_summary.to_csv(
    str(GENE_SUMMARY_FILE),
    sep="\t",
    index=False
)


# ==================================================================================
# 7. PROVENANCE
# ==================================================================================

manifest = {{
    "context_model":
        "XGBoost",

    "context_feature_count":
        316,

    "feature_selection":
        None,

    "run_boruta":
        False,

    "XGBoost_parameters":
        XGB_PARAMS,

    "NT_model":
        "StandardScaler + L2 LogisticRegression",

    "NT_dimensions":
        512,

    "NT_Logistic_parameters":
        NT_LOGISTIC_PARAMS,

    "PU_iterations":
        10,

    "splits_per_iteration":
        790,

    "total_splits":
        7900,

    "same_partitions_across_modalities":
        True,

    "primary_metric":
        "Average Precision",

    "secondary_metric":
        "AUROC",

    "model_selection_from_results":
        False
}}


with open(
    str(
        OUT_DIR /
        "primary_modeling_manifest.json"
    ),
    "w"
) as f:

    json.dump(
        manifest,
        f,
        indent=2
    )


# ==================================================================================
# FINAL OUTPUT
# ==================================================================================

print("")
print("=" * 112)
print("PRIMARY MODELING COMPLETE")
print("=" * 112)

print("")
print("SPLIT-LEVEL - 7,900 FROZEN TEST FOLDS")
print(split_summary.to_string(index=False))

print("")
print("GENE-LEVEL AGGREGATED OOF PERFORMANCE")
print(gene_summary.to_string(index=False))

print("")
print("Genes scored :", len(gene_predictions))
print("ASD genes    :", int((gene_predictions.known_gene == 1).sum()))

print("")
print("Files:")
print(ALL_METRIC_FILE)
print(ALL_PRED_FILE)
print(GENE_PRED_FILE)
print(GENE_SUMMARY_FILE)

print("")
print("GO / NO-GO: GO")
print("=" * 112)
'''


RUNNER.write_text(
    runner_code,
    encoding="utf-8"
)


print("=" * 115)
print("STARTING PRIMARY MODELING")
print("=" * 115)

print("Runner:")
print(RUNNER)

print("\nThis is checkpointed iteration-by-iteration.")
print("Completed iterations will automatically be skipped on rerun.\n")


env = os.environ.copy()
env["PYTHONIOENCODING"] = "utf-8"
env["PYTHONUTF8"] = "1"


with open(
    LOG_FILE,
    "a",
    encoding="utf-8"
) as log:

    process = subprocess.Popen(
        [
            str(LEGACY_PYTHON),
            "-u",
            str(RUNNER)
        ],
        cwd=str(REPO),
        stdout=subprocess.PIPE,
        stderr=subprocess.STDOUT,
        text=True,
        encoding="utf-8",
        errors="replace",
        bufsize=1,
        env=env
    )


    for line in process.stdout:

        print(
            line,
            end=""
        )

        log.write(
            line
        )

        log.flush()


    return_code = process.wait()


print("\n" + "=" * 115)

if return_code == 0:

    print("PRIMARY MODELING FINISHED [OK]")

else:

    print(
        "MODELING STOPPED - exit code:",
        return_code
    )

    print(
        "Do NOT delete checkpoints."
    )

    print(
        "Paste the final traceback; rerun will resume from completed iteration."
    )

print("=" * 115)
```

```python
# ======================================================================================
# ASD-ContextFusion
# MEMORY-SAFE RESUME
#
# Strategy:
#   ONE stochastic iteration = ONE fresh legacy Python process
#
# This DOES NOT change:
#   - frozen PU partitions
#   - Context predictors
#   - XGBoost parameters
#   - NT-v2 embeddings
#   - logistic probe
#   - metrics
#
# Completed iterations are never repeated.
# ======================================================================================

from pathlib import Path
import subprocess
import os
import pandas as pd


ROOT = Path(r"E:\ASD_ContextFusion_Q1")

REPO = (
    ROOT /
    "01_raw" /
    "dhindsa_2025" /
    "Mantis-ml-NDD"
)

LEGACY_PYTHON = (
    ROOT /
    "envs" /
    "mantis37" /
    "python.exe"
)

RUNNER = (
    ROOT /
    "00_admin" /
    "run_primary_context_NTv2_frozen_PU.py"
)

CHECKPOINT_DIR = (
    ROOT /
    "04_models" /
    "checkpoints" /
    "primary_context_NTv2"
)

LOG_FILE = (
    ROOT /
    "08_logs" /
    "primary_context_NTv2_memory_safe_resume.log"
)


assert RUNNER.exists()
assert LEGACY_PYTHON.exists()

CHECKPOINT_DIR.mkdir(
    parents=True,
    exist_ok=True
)


print("=" * 115)
print("ASD-CONTEXTFUSION - MEMORY-SAFE MODELING RESUME")
print("=" * 115)


# ======================================================================================
# 1. PATCH EXISTING RUNNER TO SUPPORT ONE ITERATION PER PROCESS
# ======================================================================================

text = RUNNER.read_text(
    encoding="utf-8"
)

PATCH_MARKER = "# ASD_CF_MEMORY_SAFE_ITERATION_PATCH"


if PATCH_MARKER not in text:

    # --------------------------------------------------------------------------
    # Add os/sys imports
    # --------------------------------------------------------------------------

    if "import os\n" not in text:

        text = text.replace(
            "import time\n",
            "import time\nimport os\nimport sys\n"
        )


    # --------------------------------------------------------------------------
    # Replace normal 1..10 loop with optional single-iteration mode
    # --------------------------------------------------------------------------

    old_loop = "for iteration in range(1, 11):"

    new_loop = r'''
# ASD_CF_MEMORY_SAFE_ITERATION_PATCH
TARGET_ITERATION = int(
    os.environ.get(
        "ASD_CF_TARGET_ITERATION",
        "0"
    )
)

if TARGET_ITERATION:

    assert 1 <= TARGET_ITERATION <= 10

    ITERATION_RANGE = [
        TARGET_ITERATION
    ]

    print("")
    print(
        "MEMORY-SAFE SINGLE ITERATION MODE:",
        TARGET_ITERATION
    )

else:

    ITERATION_RANGE = range(
        1,
        11
    )


for iteration in ITERATION_RANGE:'''


    assert old_loop in text, (
        "Could not find main iteration loop."
    )

    text = text.replace(
        old_loop,
        new_loop,
        1
    )


    # --------------------------------------------------------------------------
    # Stop single-iteration process BEFORE global aggregation
    # --------------------------------------------------------------------------

    aggregation_marker = (
        "# 4. COMBINE ALL 10 ITERATIONS"
    )

    assert aggregation_marker in text


    exit_block = r'''
# ==================================================================================
# MEMORY-SAFE PROCESS BOUNDARY
# ==================================================================================

if TARGET_ITERATION:

    print("")
    print("=" * 112)

    print(
        "ITERATION %02d CHECKPOINT COMPLETE - PROCESS EXITING"
        % TARGET_ITERATION
    )

    print(
        "RAM will now be released before the next iteration."
    )

    print("=" * 112)

    sys.exit(0)


'''


    text = text.replace(
        aggregation_marker,
        exit_block + aggregation_marker,
        1
    )


    RUNNER.write_text(
        text,
        encoding="utf-8"
    )

    print(
        "Memory-safe runner patch inserted: [OK]"
    )

else:

    print(
        "Memory-safe runner patch already present: [OK]"
    )


# ======================================================================================
# 2. CHECK EXISTING VALID CHECKPOINTS
# ======================================================================================

def iteration_complete(i):

    pred_file = (
        CHECKPOINT_DIR /
        f"iteration_{i:02d}_OOF_predictions.tsv.gz"
    )

    metric_file = (
        CHECKPOINT_DIR /
        f"iteration_{i:02d}_fold_metrics.tsv"
    )


    if not (
        pred_file.exists()
        and
        metric_file.exists()
    ):
        return False


    try:

        m = pd.read_csv(
            metric_file,
            sep="\t"
        )


        return (
            len(m) == 790
            and
            m["balanced_dataset"].nunique() == 79
            and
            m["fold"].nunique() == 10
        )

    except Exception:

        return False


completed_before = [
    i
    for i in range(1, 11)
    if iteration_complete(i)
]


print("\nAlready completed:")
print(completed_before)


assert 1 in completed_before, (
    "Iteration 01 checkpoint was expected but not found."
)


# ======================================================================================
# 3. RUN EACH INCOMPLETE ITERATION IN A FRESH PROCESS
# ======================================================================================

env_base = os.environ.copy()

env_base["PYTHONIOENCODING"] = "utf-8"
env_base["PYTHONUTF8"] = "1"

# Reduce hidden BLAS/OpenMP thread memory.
# This does not alter the model specification.
env_base["OMP_NUM_THREADS"] = "1"
env_base["MKL_NUM_THREADS"] = "1"
env_base["OPENBLAS_NUM_THREADS"] = "1"
env_base["NUMEXPR_NUM_THREADS"] = "1"


with open(
    LOG_FILE,
    "a",
    encoding="utf-8"
) as log:


    for iteration in range(1, 11):


        if iteration_complete(
            iteration
        ):

            print(
                f"\nIteration {iteration:02d}: "
                "already COMPLETE - SKIPPING [OK]"
            )

            continue


        print("\n" + "=" * 115)

        print(
            f"STARTING ITERATION {iteration:02d} "
            "IN FRESH PROCESS"
        )

        print("=" * 115)


        env = env_base.copy()

        env[
            "ASD_CF_TARGET_ITERATION"
        ] = str(
            iteration
        )


        process = subprocess.Popen(
            [
                str(
                    LEGACY_PYTHON
                ),
                "-u",
                str(
                    RUNNER
                )
            ],
            cwd=str(
                REPO
            ),
            stdout=subprocess.PIPE,
            stderr=subprocess.STDOUT,
            text=True,
            encoding="utf-8",
            errors="replace",
            bufsize=1,
            env=env
        )


        for line in process.stdout:

            # Always save everything to log
            log.write(
                line
            )

            log.flush()


            # Hide repetitive XGBoost warning from notebook display only
            if (
                "default evaluation metric used"
                in line
            ):
                continue


            print(
                line,
                end=""
            )


        return_code = (
            process.wait()
        )


        if return_code != 0:

            print("\n" + "=" * 115)

            print(
                f"ITERATION {iteration:02d} STOPPED "
                f"(exit code {return_code})"
            )

            print(
                "Previously completed checkpoints are safe."
            )

            print(
                "Paste the final traceback from the log/output."
            )

            print("=" * 115)

            raise RuntimeError(
                f"Iteration {iteration:02d} failed."
            )


        # ----------------------------------------------------------------------
        # Mandatory checkpoint verification
        # ----------------------------------------------------------------------

        assert iteration_complete(
            iteration
        ), (
            f"Iteration {iteration:02d} process exited "
            "without a valid checkpoint."
        )


        metric_file = (
            CHECKPOINT_DIR /
            f"iteration_{iteration:02d}_fold_metrics.tsv"
        )

        m = pd.read_csv(
            metric_file,
            sep="\t"
        )


        print(
            f"\nIteration {iteration:02d}: VERIFIED [OK]"
        )

        print(
            "Context AP:",
            round(
                m[
                    "Context_AP"
                ].mean(),
                4
            )
        )

        print(
            "NT-v2 AP :",
            round(
                m[
                    "NT_AP"
                ].mean(),
                4
            )
        )

        print(
            "Fresh process closed -> RAM released."
        )


# ======================================================================================
# 4. VERIFY 10/10 BEFORE FINAL AGGREGATION
# ======================================================================================

completed_after = [
    i
    for i in range(1, 11)
    if iteration_complete(i)
]


print("\n" + "=" * 115)

print(
    "COMPLETED ITERATIONS:",
    completed_after
)

print("=" * 115)


assert completed_after == list(
    range(
        1,
        11
    )
)


# ======================================================================================
# 5. FINAL AGGREGATION ONLY
#
# Run same runner without TARGET_ITERATION.
# It will SKIP all 10 trained iterations and only create final combined outputs.
# ======================================================================================

print("\n")
print("=" * 115)
print("ALL 10 ITERATIONS COMPLETE")
print("RUNNING FINAL AGGREGATION")
print("=" * 115)


final_env = env_base.copy()

final_env.pop(
    "ASD_CF_TARGET_ITERATION",
    None
)


process = subprocess.Popen(
    [
        str(
            LEGACY_PYTHON
        ),
        "-u",
        str(
            RUNNER
        )
    ],
    cwd=str(
        REPO
    ),
    stdout=subprocess.PIPE,
    stderr=subprocess.STDOUT,
    text=True,
    encoding="utf-8",
    errors="replace",
    bufsize=1,
    env=final_env
)


for line in process.stdout:

    print(
        line,
        end=""
    )


return_code = (
    process.wait()
)


print("\n" + "=" * 115)

if return_code == 0:

    print(
        "PRIMARY MODELING + FINAL AGGREGATION COMPLETE [OK]"
    )

else:

    print(
        "FINAL AGGREGATION STOPPED - "
        f"exit code {return_code}"
    )

    print(
        "Training checkpoints remain safe."
    )

print("=" * 115)
```

```python
# ======================================================================================
# ASD-ContextFusion
# POST-PRIMARY LOCK + SEQUENCE-CONTROL READINESS AUDIT
#
# PURPOSE
# -------
# 1. Verify completed 7,900-split primary modeling outputs
# 2. Verify frozen PU partitions remain intact
# 3. Verify multimodal master
# 4. Inspect NT-v2 metadata for transcript/sequence provenance
# 5. Locate the exact existing GENCODE/transcript sequence assets
#
# NO TRAINING
# NO NEW SAMPLING
# NO FEATURE GENERATION
# NO DATA MODIFICATION
# ======================================================================================

from pathlib import Path
import pandas as pd
import numpy as np
import json
import re


# ======================================================================================
# 0. PATHS
# ======================================================================================

ROOT = Path(r"E:\ASD_ContextFusion_Q1")

PRIMARY_DIR = (
    ROOT /
    "05_results" /
    "proposed" /
    "primary_context_NTv2"
)

PARTITION_DIR = (
    ROOT /
    "04_models" /
    "checkpoints" /
    "dhindsa_PU_common_universe"
)

MASTER_FILE = (
    ROOT /
    "03_features" /
    "multimodal_master" /
    "dhindsa_NTv2_row_preserved_master.tsv"
)

NT_META_FILE = (
    ROOT /
    "03_features" /
    "nt_v2" /
    "NTv2_embedding_row_metadata.tsv"
)

NT_NPY = (
    ROOT /
    "03_features" /
    "nt_v2" /
    "NTv2_50m_full_transcript_embeddings_18252x512.npy"
)


ALL_METRICS = (
    PRIMARY_DIR /
    "Context_NTv2_all_7900_fold_metrics.tsv"
)

ALL_PRED = (
    PRIMARY_DIR /
    "Context_NTv2_all_OOF_predictions.tsv.gz"
)

GENE_SCORES = (
    PRIMARY_DIR /
    "Context_NTv2_gene_level_OOF_scores.tsv"
)

GENE_SUMMARY = (
    PRIMARY_DIR /
    "Context_NTv2_gene_level_summary.tsv"
)

SPLIT_SUMMARY = (
    PRIMARY_DIR /
    "Context_NTv2_split_level_summary.tsv"
)


print("=" * 120)
print("ASD-CONTEXTFUSION - POST-PRIMARY / SEQUENCE-CONTROL READINESS AUDIT")
print("=" * 120)


# ======================================================================================
# 1. PRIMARY RESULT FILES
# ======================================================================================

print("\n[1/7] PRIMARY MODELING ARTIFACTS")
print("-" * 120)


critical_files = [
    ALL_METRICS,
    ALL_PRED,
    GENE_SCORES,
    GENE_SUMMARY,
    MASTER_FILE,
    NT_META_FILE,
    NT_NPY
]


for p in critical_files:

    print(
        p.name,
        "->",
        "EXISTS [OK]" if p.exists() else "MISSING"
    )


missing = [
    p
    for p in critical_files
    if not p.exists()
]


assert not missing, (
    "Critical completed artifact(s) missing:\n"
    + "\n".join(str(x) for x in missing)
)


print(
    "\nSplit-summary file:",
    "EXISTS [OK]"
    if SPLIT_SUMMARY.exists()
    else "not separately present"
)


# ======================================================================================
# 2. INDEPENDENTLY VERIFY 7,900 SPLIT RESULTS
# ======================================================================================

print("\n[2/7] 7,900-SPLIT RESULT VERIFICATION")
print("-" * 120)


metrics = pd.read_csv(
    ALL_METRICS,
    sep="\t"
)


print("Rows:", len(metrics))
print("Columns:", len(metrics.columns))
print("Metric columns:")
print(metrics.columns.tolist())


assert len(metrics) == 7900, (
    f"Expected exactly 7900 split-level records; "
    f"found {len(metrics)}"
)


expected_metric_columns = [
    "Context_AUROC",
    "Context_AP",
    "NT_AUROC",
    "NT_AP"
]


for c in expected_metric_columns:

    assert c in metrics.columns, (
        f"Missing metric column: {c}"
    )


observed = {
    "Context_AUROC":
        float(metrics["Context_AUROC"].mean()),

    "Context_AP":
        float(metrics["Context_AP"].mean()),

    "NT_AUROC":
        float(metrics["NT_AUROC"].mean()),

    "NT_AP":
        float(metrics["NT_AP"].mean())
}


print("\nRecomputed means:")

for key, value in observed.items():

    print(
        f"{key:16s}: {value:.6f}"
    )


expected = {
    "Context_AUROC": 0.942180,
    "Context_AP":    0.909122,
    "NT_AUROC":      0.862039,
    "NT_AP":         0.808142
}


for key in expected:

    assert abs(
        observed[key] -
        expected[key]
    ) < 5e-6, (
        f"{key} does not reproduce the completed "
        "aggregation result."
    )


print(
    "\n7,900 split-level aggregation independently verified [OK]"
)


# ======================================================================================
# 3. GENE-LEVEL RESULT VERIFICATION
# ======================================================================================

print("\n[3/7] GENE-LEVEL OOF RESULTS")
print("-" * 120)


gene_scores = pd.read_csv(
    GENE_SCORES,
    sep="\t",
    low_memory=False
)

gene_summary = pd.read_csv(
    GENE_SUMMARY,
    sep="\t"
)


print(
    "Gene-score rows:",
    len(gene_scores)
)

print(
    "Gene-score columns:",
    gene_scores.columns.tolist()
)


assert len(gene_scores) == 18252, (
    f"Expected 18,252 scored genes; "
    f"found {len(gene_scores)}"
)


if "Gene_Name" in gene_scores.columns:

    assert (
        gene_scores["Gene_Name"].nunique()
        == 18252
    )


if "known_gene" in gene_scores.columns:

    assert int(
        gene_scores["known_gene"].sum()
    ) == 190


print("\nSaved gene-level summary:")
print(
    gene_summary.to_string(
        index=False
    )
)


print(
    "\nGene-level primary outputs verified [OK]"
)


# ======================================================================================
# 4. FROZEN PU PARTITIONS
# ======================================================================================

print("\n[4/7] FROZEN PU PARTITIONS")
print("-" * 120)


partition_files = sorted(
    PARTITION_DIR.glob(
        "iteration_*_test_fold_membership.tsv.gz"
    )
)


print(
    "Membership files:",
    len(partition_files)
)


assert len(partition_files) == 10, (
    "Expected exactly 10 frozen PU membership files."
)


summary_file = (
    PARTITION_DIR /
    "PU_iteration_summary.tsv"
)


assert summary_file.exists()


pu_summary = pd.read_csv(
    summary_file,
    sep="\t"
)


assert len(pu_summary) == 10


print(
    pu_summary.to_string(
        index=False
    )
)


if "train_test_splits" in pu_summary.columns:

    total_splits = int(
        pu_summary[
            "train_test_splits"
        ].sum()
    )

    print(
        "\nTotal frozen train/test splits:",
        total_splits
    )

    assert total_splits == 7900


print(
    "\nFrozen PU partitions intact [OK]"
)

print(
    "NO NEW DOWNSTREAM SAMPLING IS PERMITTED."
)


# ======================================================================================
# 5. MULTIMODAL MASTER
# ======================================================================================

print("\n[5/7] MULTIMODAL MASTER")
print("-" * 120)


master = pd.read_csv(
    MASTER_FILE,
    sep="\t",
    low_memory=False
)


print(
    "Rows        :",
    len(master)
)

print(
    "Unique genes:",
    master["Gene_Name"].nunique()
)

print(
    "ASD rows    :",
    int(
        (master["known_gene"] == 1).sum()
    )
)

print(
    "ASD genes   :",
    master.loc[
        master["known_gene"] == 1,
        "Gene_Name"
    ].nunique()
)


assert len(master) == 18253

assert (
    master["Gene_Name"].nunique()
    == 18252
)

assert (
    master.loc[
        master["known_gene"] == 1,
        "Gene_Name"
    ].nunique()
    == 190
)


tmsb = master.loc[
    master["Gene_Name"] == "TMSB15B"
]


print(
    "\nTMSB15B rows:",
    len(tmsb)
)


assert len(tmsb) == 2


print(
    "Row-preserving master intact [OK]"
)


# ======================================================================================
# 6. NT METADATA / TRANSCRIPT PROVENANCE AUDIT
# ======================================================================================

print("\n[6/7] NT-v2 TRANSCRIPT METADATA")
print("-" * 120)


nt_meta = pd.read_csv(
    NT_META_FILE,
    sep="\t",
    low_memory=False
)


assert len(nt_meta) == 18252


print(
    "Metadata rows:",
    len(nt_meta)
)

print(
    "Metadata columns:",
    len(nt_meta.columns)
)


print("\nALL NT metadata columns:")

for i, col in enumerate(
    nt_meta.columns,
    start=1
):

    print(
        f"{i:02d}. {col}"
    )


# Find likely transcript/sequence-related metadata fields
pattern = re.compile(
    r"(sequence|seq|transcript|gencode|ensembl|"
    r"length|gc|fasta|chrom|strand)",
    re.IGNORECASE
)


sequence_like_columns = [
    c
    for c in nt_meta.columns
    if pattern.search(
        str(c)
    )
]


print(
    "\nSequence/transcript-related columns:"
)

if sequence_like_columns:

    for c in sequence_like_columns:
        print(" -", c)

else:

    print(
        "No obvious sequence-related metadata columns detected."
    )


# Small safe preview
preview_cols = list(
    dict.fromkeys(
        [
            c
            for c in [
                "original_gene_symbol",
                "gencode_gene_symbol"
            ]
            if c in nt_meta.columns
        ]
        +
        sequence_like_columns
    )
)


if preview_cols:

    print(
        "\nMetadata preview:"
    )

    print(
        nt_meta[
            preview_cols
        ]
        .head(10)
        .to_string(
            index=False
        )
    )


# ======================================================================================
# 7. LOCATE EXISTING SEQUENCE / GENCODE ASSETS
# ======================================================================================

print("\n[7/7] EXISTING SEQUENCE ASSET SEARCH")
print("-" * 120)


search_roots = [
    ROOT / "01_raw",
    ROOT / "02_interim",
    ROOT / "03_features"
]


sequence_suffixes = (
    ".fa",
    ".fasta",
    ".fna",
    ".fa.gz",
    ".fasta.gz",
    ".fna.gz"
)

annotation_suffixes = (
    ".gtf",
    ".gff",
    ".gff3",
    ".gtf.gz",
    ".gff.gz",
    ".gff3.gz"
)


sequence_files = []
annotation_files = []


for base in search_roots:

    if not base.exists():
        continue

    for p in base.rglob("*"):

        if not p.is_file():
            continue

        name = p.name.lower()

        if name.endswith(
            sequence_suffixes
        ):

            sequence_files.append(
                p
            )

        elif name.endswith(
            annotation_suffixes
        ):

            annotation_files.append(
                p
            )


print(
    "\nSequence FASTA-like files found:",
    len(sequence_files)
)


for p in sequence_files[:50]:

    print(
        "SEQ :",
        p
    )


if len(sequence_files) > 50:

    print(
        "... additional sequence files omitted from display"
    )


print(
    "\nAnnotation files found:",
    len(annotation_files)
)


for p in annotation_files[:50]:

    print(
        "ANNO:",
        p
    )


if len(annotation_files) > 50:

    print(
        "... additional annotation files omitted from display"
    )


# Files whose names mention GENCODE/transcript even if extension differs
name_hits = []


name_pattern = re.compile(
    r"(gencode|transcript|canonical|mane)",
    re.IGNORECASE
)


for base in search_roots:

    if not base.exists():
        continue

    for p in base.rglob("*"):

        if (
            p.is_file()
            and
            name_pattern.search(
                p.name
            )
        ):

            name_hits.append(
                p
            )


name_hits = sorted(
    set(name_hits)
)


print(
    "\nGENCODE/transcript-named files:",
    len(name_hits)
)


for p in name_hits[:100]:

    print(
        "HIT :",
        p
    )


# ======================================================================================
# FINAL STATUS
# ======================================================================================

print("\n" + "=" * 120)
print("POST-PRIMARY AUDIT COMPLETE")
print("=" * 120)

print(
    "Primary Context + NT-v2 modeling : LOCKED [OK]"
)

print(
    "7,900 frozen split results       : VERIFIED [OK]"
)

print(
    "18,252 gene-level scores         : VERIFIED [OK]"
)

print(
    "PU partitions                    : FROZEN [OK]"
)

print(
    "\nNEXT SCIENTIFIC STAGE:"
)

print(
    "1. Build exact transcript Length + GC control"
)

print(
    "2. Build prespecified 4-mer representation"
)

print(
    "3. Evaluate both on THE SAME 7,900 frozen folds"
)

print(
    "4. Only then run cross-fitted logistic late fusion"
)

print("=" * 120)
```

```python
# ======================================================================================
# ASD-ContextFusion
# EXACT GENCODE v50 TRANSCRIPT RECOVERY + SEQUENCE CONTROL FEATURE GENERATION
#
# SCIENTIFIC RULES
# ----------------
# - Use the exact representative transcript_id already locked in NT-v2 metadata.
# - Use GENCODE Release 50 transcript sequences (GRCh38.p14).
# - Transcript ID + version MUST match.
# - FASTA sequence length MUST match locked sequence_length_nt.
# - Length+GC and 4-mer are computed from the SAME transcript used for NT-v2.
#
# 4-mer definition:
# - all oriented DNA 4-mers over A,C,G,T
# - 4^4 = 256 features
# - lexicographic order using alphabet A,C,G,T
# - counts normalized by number of valid A/C/G/T 4-mer windows
# - ambiguous-base-containing windows are excluded
#
# NO MODEL TRAINING
# NO PU SAMPLING
# NO PARTITION MODIFICATION
# NO NT REGENERATION
# ======================================================================================

from pathlib import Path
from itertools import product

import urllib.request
import gzip
import hashlib
import json
import shutil

import numpy as np
import pandas as pd


# ======================================================================================
# 0. PATHS
# ======================================================================================

ROOT = Path(r"E:\ASD_ContextFusion_Q1")

NT_META_FILE = (
    ROOT /
    "03_features" /
    "nt_v2" /
    "NTv2_embedding_row_metadata.tsv"
)

OUT_DIR = (
    ROOT /
    "03_features" /
    "sequence_controls"
)

RAW_SEQ_DIR = (
    ROOT /
    "01_raw" /
    "gencode_v50"
)

OUT_DIR.mkdir(
    parents=True,
    exist_ok=True
)

RAW_SEQ_DIR.mkdir(
    parents=True,
    exist_ok=True
)


# Exact GENCODE Release 50 transcript FASTA.
GENCODE_URL = (
    "https://ftp.ebi.ac.uk/pub/databases/gencode/"
    "Gencode_human/release_50/"
    "gencode.v50.transcripts.fa.gz"
)

GENCODE_FASTA = (
    RAW_SEQ_DIR /
    "gencode.v50.transcripts.fa.gz"
)

SELECTED_FASTA = (
    OUT_DIR /
    "GENCODE_v50_selected_18252_representative_transcripts.fa.gz"
)

LENGTH_GC_FILE = (
    OUT_DIR /
    "GENCODE_v50_selected_18252_Length_GC.tsv"
)

KMER_NPY = (
    OUT_DIR /
    "GENCODE_v50_selected_18252_4mer_frequency_18252x256.npy"
)

KMER_NAMES_FILE = (
    OUT_DIR /
    "GENCODE_v50_4mer_feature_names.tsv"
)

ROW_META_FILE = (
    OUT_DIR /
    "GENCODE_v50_sequence_control_row_metadata.tsv"
)

MANIFEST_FILE = (
    OUT_DIR /
    "sequence_control_manifest.json"
)


print("=" * 122)
print("ASD-CONTEXTFUSION - EXACT GENCODE v50 SEQUENCE CONTROL BUILD")
print("=" * 122)


# ======================================================================================
# 1. HELPERS
# ======================================================================================

def sha256_file(path, chunk_size=1024 * 1024):

    h = hashlib.sha256()

    with open(path, "rb") as f:

        while True:

            block = f.read(chunk_size)

            if not block:
                break

            h.update(block)

    return h.hexdigest()


def download_atomic(url, destination):

    destination = Path(destination)

    temp = destination.with_suffix(
        destination.suffix + ".part"
    )

    if temp.exists():
        temp.unlink()

    request = urllib.request.Request(
        url,
        headers={
            "User-Agent":
                "Mozilla/5.0 ASD-ContextFusion reproducible research"
        }
    )

    print("\nDownloading:")
    print(url)

    print("\nDestination:")
    print(destination)

    downloaded = 0

    with urllib.request.urlopen(
        request,
        timeout=120
    ) as response:

        total_header = response.headers.get(
            "Content-Length"
        )

        total = (
            int(total_header)
            if total_header is not None
            else None
        )

        with open(
            temp,
            "wb"
        ) as out:

            while True:

                block = response.read(
                    1024 * 1024
                )

                if not block:
                    break

                out.write(block)

                downloaded += len(block)

                if (
                    downloaded %
                    (50 * 1024 * 1024)
                    <
                    1024 * 1024
                ):

                    if total:

                        print(
                            f"Downloaded "
                            f"{downloaded / 1024**2:.1f} / "
                            f"{total / 1024**2:.1f} MB"
                        )

                    else:

                        print(
                            f"Downloaded "
                            f"{downloaded / 1024**2:.1f} MB"
                        )

    temp.replace(
        destination
    )

    print(
        "\nDownload complete [OK]"
    )


# ======================================================================================
# 2. LOAD LOCKED NT METADATA
# ======================================================================================

print("\n[1/8] LOCKED NT-v2 TRANSCRIPT METADATA")
print("-" * 122)


assert NT_META_FILE.exists(), (
    f"Missing:\n{NT_META_FILE}"
)


meta = pd.read_csv(
    NT_META_FILE,
    sep="\t",
    low_memory=False
)


required_columns = {
    "embedding_row",
    "original_gene_symbol",
    "gencode_gene_symbol",
    "transcript_id",
    "sequence_length",
    "sequence_length_nt"
}


assert required_columns.issubset(
    set(meta.columns)
), (
    "Required NT transcript provenance columns are missing."
)


assert len(meta) == 18252


meta["transcript_id"] = (
    meta["transcript_id"]
    .astype(str)
    .str.strip()
)

meta["original_gene_symbol"] = (
    meta["original_gene_symbol"]
    .astype(str)
    .str.strip()
)


assert meta[
    "transcript_id"
].is_unique, (
    "Locked representative transcript IDs are not unique."
)


assert meta[
    "embedding_row"
].is_unique


# Preserve exact NT embedding row order.
meta = (
    meta
    .sort_values(
        "embedding_row"
    )
    .reset_index(
        drop=True
    )
)


assert np.array_equal(
    meta["embedding_row"].to_numpy(),
    np.arange(
        18252
    )
)


expected_lengths = (
    pd.to_numeric(
        meta["sequence_length_nt"],
        errors="raise"
    )
    .astype(int)
    .to_numpy()
)


expected_lengths_alt = (
    pd.to_numeric(
        meta["sequence_length"],
        errors="raise"
    )
    .astype(int)
    .to_numpy()
)


assert np.array_equal(
    expected_lengths,
    expected_lengths_alt
), (
    "sequence_length and sequence_length_nt "
    "do not agree in locked metadata."
)


print(
    "Metadata rows       :",
    len(meta)
)

print(
    "Unique transcripts  :",
    meta["transcript_id"].nunique()
)

print(
    "Unique source genes :",
    meta["original_gene_symbol"].nunique()
)

print(
    "Length metadata     : CONSISTENT [OK]"
)


# ======================================================================================
# 3. DOWNLOAD EXACT GENCODE RELEASE 50 FASTA
# ======================================================================================

print("\n[2/8] GENCODE RELEASE 50 TRANSCRIPT FASTA")
print("-" * 122)


if GENCODE_FASTA.exists():

    print(
        "Existing FASTA found:"
    )

    print(
        GENCODE_FASTA
    )

    print(
        "\nExisting file will be validated; "
        "it will NOT be redownloaded."
    )

else:

    download_atomic(
        GENCODE_URL,
        GENCODE_FASTA
    )


assert GENCODE_FASTA.exists()

assert (
    GENCODE_FASTA.stat().st_size
    > 0
)


gencode_sha = sha256_file(
    GENCODE_FASTA
)


print(
    "\nGENCODE FASTA size:",
    round(
        GENCODE_FASTA.stat().st_size /
        1024**2,
        2
    ),
    "MB"
)

print(
    "\nGENCODE FASTA SHA256:"
)

print(
    gencode_sha
)


# ======================================================================================
# 4. PRESPECIFY 4-MER VOCABULARY
# ======================================================================================

print("\n[3/8] PRESPECIFIED 4-MER VOCABULARY")
print("-" * 122)


DNA = "ACGT"

KMERS = [
    "".join(x)
    for x in product(
        DNA,
        repeat=4
    )
]


assert len(KMERS) == 256

assert len(set(KMERS)) == 256


KMER_TO_INDEX = {
    kmer: i
    for i, kmer in enumerate(
        KMERS
    )
}


pd.DataFrame({
    "kmer_index":
        np.arange(
            256,
            dtype=int
        ),
    "kmer":
        KMERS
}).to_csv(
    KMER_NAMES_FILE,
    sep="\t",
    index=False
)


print(
    "4-mer features:",
    len(KMERS)
)

print(
    "First 10:",
    KMERS[:10]
)

print(
    "Last 10 :",
    KMERS[-10:]
)


# ======================================================================================
# 5. PREALLOCATE FEATURES
# ======================================================================================

n = len(meta)


actual_length = np.zeros(
    n,
    dtype=np.int32
)

gc_fraction = np.full(
    n,
    np.nan,
    dtype=np.float64
)

valid_base_count = np.zeros(
    n,
    dtype=np.int32
)

ambiguous_base_count = np.zeros(
    n,
    dtype=np.int32
)

valid_4mer_windows = np.zeros(
    n,
    dtype=np.int32
)

invalid_4mer_windows = np.zeros(
    n,
    dtype=np.int32
)

X_4mer = np.zeros(
    (
        n,
        256
    ),
    dtype=np.float32
)


transcript_to_row = {
    tid: i
    for i, tid in enumerate(
        meta["transcript_id"]
    )
}


found = np.zeros(
    n,
    dtype=bool
)


# ======================================================================================
# 6. FEATURE FUNCTION
# ======================================================================================

def compute_sequence_features(
    seq,
    row_index
):

    seq = (
        seq
        .upper()
        .replace(
            "U",
            "T"
        )
    )

    L = len(seq)

    actual_length[
        row_index
    ] = L


    # ------------------------------------------------------------------
    # GC
    # ------------------------------------------------------------------

    a = seq.count("A")
    c = seq.count("C")
    g = seq.count("G")
    t = seq.count("T")

    valid_bases = (
        a +
        c +
        g +
        t
    )

    invalid_bases = (
        L -
        valid_bases
    )

    valid_base_count[
        row_index
    ] = valid_bases

    ambiguous_base_count[
        row_index
    ] = invalid_bases


    if valid_bases > 0:

        gc_fraction[
            row_index
        ] = (
            g + c
        ) / valid_bases


    # ------------------------------------------------------------------
    # 4-mer frequencies
    # ------------------------------------------------------------------

    counts = np.zeros(
        256,
        dtype=np.float64
    )

    valid_windows = 0
    invalid_windows = 0


    if L >= 4:

        for pos in range(
            L - 3
        ):

            kmer = seq[
                pos:
                pos + 4
            ]

            idx = KMER_TO_INDEX.get(
                kmer
            )

            if idx is None:

                invalid_windows += 1

            else:

                counts[idx] += 1.0
                valid_windows += 1


    valid_4mer_windows[
        row_index
    ] = valid_windows

    invalid_4mer_windows[
        row_index
    ] = invalid_windows


    if valid_windows > 0:

        counts /= (
            float(
                valid_windows
            )
        )


    X_4mer[
        row_index,
        :
    ] = counts.astype(
        np.float32
    )


# ======================================================================================
# 7. STREAM GENCODE FASTA AND EXTRACT ONLY LOCKED TRANSCRIPTS
# ======================================================================================

print("\n[4/8] EXACT TRANSCRIPT EXTRACTION")
print("-" * 122)


if SELECTED_FASTA.exists():

    SELECTED_FASTA.unlink()


current_id = None
current_target_row = None
sequence_chunks = []


def finish_record(
    transcript_id,
    row_index,
    chunks,
    selected_handle
):

    if (
        transcript_id is None
        or
        row_index is None
    ):
        return


    assert not found[
        row_index
    ], (
        f"Duplicate FASTA record for "
        f"{transcript_id}"
    )


    seq = "".join(
        chunks
    )


    compute_sequence_features(
        seq,
        row_index
    )


    found[
        row_index
    ] = True


    selected_handle.write(
        ">" +
        transcript_id +
        "\n"
    )


    width = 80

    for start in range(
        0,
        len(seq),
        width
    ):

        selected_handle.write(
            seq[
                start:
                start + width
            ] +
            "\n"
        )


with gzip.open(
    GENCODE_FASTA,
    "rt",
    encoding="utf-8",
    errors="strict"
) as fasta_in:

    with gzip.open(
        SELECTED_FASTA,
        "wt",
        encoding="utf-8"
    ) as selected_out:

        for line in fasta_in:

            if line.startswith(">"):

                # Finish previous record.
                finish_record(
                    current_id,
                    current_target_row,
                    sequence_chunks,
                    selected_out
                )


                header = (
                    line[1:]
                    .strip()
                )


                # GENCODE transcript FASTA:
                # first pipe-delimited field is transcript ID with version.
                fasta_tid = (
                    header
                    .split("|", 1)[0]
                    .split()[0]
                    .strip()
                )


                current_id = fasta_tid

                current_target_row = (
                    transcript_to_row.get(
                        fasta_tid
                    )
                )

                sequence_chunks = []

            else:

                if current_target_row is not None:

                    sequence_chunks.append(
                        line.strip()
                    )


        # Last FASTA record
        finish_record(
            current_id,
            current_target_row,
            sequence_chunks,
            selected_out
        )


n_found = int(
    found.sum()
)


print(
    "Required transcripts:",
    n
)

print(
    "Recovered transcripts:",
    n_found
)


if n_found != n:

    missing_ids = (
        meta.loc[
            ~found,
            [
                "original_gene_symbol",
                "gencode_gene_symbol",
                "transcript_id",
                "sequence_length_nt"
            ]
        ]
        .copy()
    )

    missing_file = (
        OUT_DIR /
        "ERROR_missing_GENCODE_v50_transcripts.tsv"
    )

    missing_ids.to_csv(
        missing_file,
        sep="\t",
        index=False
    )

    print(
        "\nMissing transcript IDs saved:"
    )

    print(
        missing_file
    )

    print(
        "\nFirst missing records:"
    )

    print(
        missing_ids
        .head(30)
        .to_string(
            index=False
        )
    )

    raise RuntimeError(
        "\nSTOP: GENCODE v50 FASTA did not recover "
        "all 18,252 locked transcript IDs.\n"
        "Do NOT build sequence controls until this is resolved."
    )


print(
    "\n18,252 / 18,252 exact transcript IDs recovered [OK]"
)


# ======================================================================================
# 8. MANDATORY SEQUENCE LENGTH VERIFICATION
# ======================================================================================

print("\n[5/8] EXACT SEQUENCE LENGTH VERIFICATION")
print("-" * 122)


length_match = (
    actual_length
    ==
    expected_lengths
)


n_length_match = int(
    length_match.sum()
)


print(
    "Exact length matches:",
    f"{n_length_match}/{n}"
)


if n_length_match != n:

    mismatch_df = meta.loc[
        ~length_match,
        [
            "embedding_row",
            "original_gene_symbol",
            "gencode_gene_symbol",
            "transcript_id",
            "sequence_length",
            "sequence_length_nt"
        ]
    ].copy()


    mismatch_df[
        "actual_fasta_length"
    ] = actual_length[
        ~length_match
    ]


    mismatch_file = (
        OUT_DIR /
        "ERROR_GENCODE_v50_sequence_length_mismatch.tsv"
    )


    mismatch_df.to_csv(
        mismatch_file,
        sep="\t",
        index=False
    )


    print(
        "\nLength mismatches saved:"
    )

    print(
        mismatch_file
    )


    print(
        mismatch_df
        .head(30)
        .to_string(
            index=False
        )
    )


    raise RuntimeError(
        "\nSTOP: transcript IDs matched but "
        "sequence lengths do not exactly match "
        "the locked NT-v2 transcript metadata."
    )


print(
    "\nSEQUENCE IDENTITY PROVENANCE CHECK PASSED [OK]"
)

print(
    "Every locked transcript ID has exactly "
    "the expected GENCODE v50 sequence length."
)


# ======================================================================================
# 9. FEATURE QC
# ======================================================================================

print("\n[6/8] LENGTH + GC + 4-MER QC")
print("-" * 122)


assert np.isfinite(
    gc_fraction
).all()


assert np.isfinite(
    X_4mer
).all()


assert (
    actual_length > 0
).all()


assert (
    valid_base_count > 0
).all()


assert (
    valid_4mer_windows > 0
).all()


row_sums = (
    X_4mer.sum(
        axis=1
    )
)


assert np.allclose(
    row_sums,
    1.0,
    atol=1e-5
), (
    "At least one normalized 4-mer row "
    "does not sum to 1."
)


print(
    "Length range:",
    int(
        actual_length.min()
    ),
    "-",
    int(
        actual_length.max()
    )
)


print(
    "Mean length:",
    round(
        float(
            actual_length.mean()
        ),
        2
    )
)


print(
    "GC range:",
    round(
        float(
            gc_fraction.min()
        ),
        6
    ),
    "-",
    round(
        float(
            gc_fraction.max()
        ),
        6
    )
)


print(
    "Mean GC:",
    round(
        float(
            gc_fraction.mean()
        ),
        6
    )
)


print(
    "Transcripts with ambiguous bases:",
    int(
        (
            ambiguous_base_count > 0
        ).sum()
    )
)


print(
    "Total ambiguous bases:",
    int(
        ambiguous_base_count.sum()
    )
)


print(
    "Total invalid 4-mer windows:",
    int(
        invalid_4mer_windows.sum()
    )
)


print(
    "4-mer matrix shape:",
    X_4mer.shape
)


# ======================================================================================
# 10. SAVE LENGTH+GC
# ======================================================================================

print("\n[7/8] SAVE EXACT SEQUENCE CONTROL FEATURES")
print("-" * 122)


length_gc = pd.DataFrame({

    "embedding_row":
        meta[
            "embedding_row"
        ].astype(int),

    "Gene_Name":
        meta[
            "original_gene_symbol"
        ],

    "gencode_gene_symbol":
        meta[
            "gencode_gene_symbol"
        ],

    "transcript_id":
        meta[
            "transcript_id"
        ],

    "transcript_length_nt":
        actual_length,

    "GC_fraction":
        gc_fraction,

    "valid_ACGT_bases":
        valid_base_count,

    "ambiguous_bases":
        ambiguous_base_count,

    "valid_4mer_windows":
        valid_4mer_windows,

    "invalid_4mer_windows":
        invalid_4mer_windows
})


length_gc.to_csv(
    LENGTH_GC_FILE,
    sep="\t",
    index=False
)


np.save(
    KMER_NPY,
    X_4mer
)


row_meta = meta[
    [
        "embedding_row",
        "original_gene_symbol",
        "gencode_gene_symbol",
        "gene_id",
        "transcript_id",
        "chromosome",
        "strand",
        "selection_basis",
        "sequence_length_nt",
        "ntv2_n_chunks"
    ]
].copy()


row_meta = row_meta.rename(
    columns={
        "original_gene_symbol":
            "Gene_Name"
    }
)


row_meta.to_csv(
    ROW_META_FILE,
    sep="\t",
    index=False
)


print(
    "Length+GC:"
)

print(
    LENGTH_GC_FILE
)


print(
    "\n4-mer matrix:"
)

print(
    KMER_NPY
)


print(
    "\n4-mer names:"
)

print(
    KMER_NAMES_FILE
)


print(
    "\nRow metadata:"
)

print(
    ROW_META_FILE
)


print(
    "\nSelected exact transcripts:"
)

print(
    SELECTED_FASTA
)


# ======================================================================================
# 11. FINAL SAVED-ARTIFACT VERIFICATION
# ======================================================================================

print("\n[8/8] FINAL INTEGRITY LOCK")
print("-" * 122)


X_saved = np.load(
    KMER_NPY,
    mmap_mode="r"
)


lg_saved = pd.read_csv(
    LENGTH_GC_FILE,
    sep="\t"
)


rows_saved = pd.read_csv(
    ROW_META_FILE,
    sep="\t"
)


assert X_saved.shape == (
    18252,
    256
)

assert len(
    lg_saved
) == 18252

assert len(
    rows_saved
) == 18252


assert (
    lg_saved[
        "Gene_Name"
    ].tolist()
    ==
    meta[
        "original_gene_symbol"
    ].tolist()
)


assert (
    rows_saved[
        "transcript_id"
    ].tolist()
    ==
    meta[
        "transcript_id"
    ].tolist()
)


selected_fasta_sha = sha256_file(
    SELECTED_FASTA
)

length_gc_sha = sha256_file(
    LENGTH_GC_FILE
)

kmer_sha = sha256_file(
    KMER_NPY
)

row_meta_sha = sha256_file(
    ROW_META_FILE
)

kmer_names_sha = sha256_file(
    KMER_NAMES_FILE
)


manifest = {

    "project":
        "ASD-ContextFusion",

    "stage":
        "Exact transcript sequence controls",

    "status":
        "completed_and_locked",

    "source": {

        "annotation":
            "GENCODE Release 50",

        "assembly":
            "GRCh38.p14",

        "source_url":
            GENCODE_URL,

        "source_fasta":
            str(
                GENCODE_FASTA
            ),

        "source_fasta_sha256":
            gencode_sha
    },

    "transcript_identity": {

        "source":
            "locked NT-v2 row metadata",

        "target_transcripts":
            18252,

        "exact_transcript_ids_recovered":
            int(
                found.sum()
            ),

        "exact_sequence_length_matches":
            int(
                length_match.sum()
            ),

        "transcript_version_preserved":
            True,

        "sequence_reselection_performed":
            False
    },

    "length_gc_control": {

        "features": [
            "transcript_length_nt",
            "GC_fraction"
        ],

        "length_definition":
            (
                "full selected transcript nucleotide "
                "sequence length"
            ),

        "gc_definition":
            (
                "(G+C)/(A+C+G+T); ambiguous "
                "bases excluded from denominator"
            )
    },

    "four_mer_control": {

        "k":
            4,

        "alphabet":
            "ACGT",

        "dimensions":
            256,

        "orientation":
            "transcript orientation; no reverse-complement collapsing",

        "feature_order":
            (
                "lexicographic Cartesian product using "
                "alphabet A,C,G,T"
            ),

        "normalization":
            (
                "count / number of valid ACGT 4-mer windows "
                "within each selected transcript"
            ),

        "ambiguous_window_policy":
            (
                "windows containing non-ACGT characters excluded"
            )
    },

    "qc": {

        "transcripts_with_ambiguous_bases":
            int(
                (
                    ambiguous_base_count > 0
                ).sum()
            ),

        "total_ambiguous_bases":
            int(
                ambiguous_base_count.sum()
            ),

        "total_invalid_4mer_windows":
            int(
                invalid_4mer_windows.sum()
            ),

        "all_4mer_rows_normalized":
            bool(
                np.allclose(
                    row_sums,
                    1.0,
                    atol=1e-5
                )
            )
    },

    "outputs": {

        "selected_transcript_fasta":
            str(
                SELECTED_FASTA
            ),

        "length_gc":
            str(
                LENGTH_GC_FILE
            ),

        "four_mer_matrix":
            str(
                KMER_NPY
            ),

        "four_mer_feature_names":
            str(
                KMER_NAMES_FILE
            ),

        "row_metadata":
            str(
                ROW_META_FILE
            )
    },

    "sha256": {

        "selected_transcript_fasta":
            selected_fasta_sha,

        "length_gc":
            length_gc_sha,

        "four_mer_matrix":
            kmer_sha,

        "four_mer_feature_names":
            kmer_names_sha,

        "row_metadata":
            row_meta_sha
    },

    "downstream_rule":
        (
            "Length+GC and 4-mer models must use the "
            "already frozen Dhindsa/Mantis 7,900 PU folds. "
            "No new sampling is permitted."
        )
}


with open(
    MANIFEST_FILE,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        manifest,
        f,
        indent=2
    )


print(
    "Manifest:"
)

print(
    MANIFEST_FILE
)


print(
    "\nSource FASTA SHA256:"
)

print(
    gencode_sha
)


print(
    "\nSelected FASTA SHA256:"
)

print(
    selected_fasta_sha
)


print(
    "\nLength+GC SHA256:"
)

print(
    length_gc_sha
)


print(
    "\n4-mer SHA256:"
)

print(
    kmer_sha
)


print("\n" + "=" * 122)
print("EXACT GENCODE v50 SEQUENCE CONTROLS LOCKED [OK]")
print("=" * 122)

print(
    "Exact transcript IDs      :",
    f"{int(found.sum())}/18252"
)

print(
    "Exact sequence lengths    :",
    f"{int(length_match.sum())}/18252"
)

print(
    "Length+GC matrix          :",
    "18252 x 2"
)

print(
    "4-mer matrix              :",
    X_saved.shape
)

print(
    "New PU sampling           : NO"
)

print(
    "Primary models modified   : NO"
)

print(
    "NT embeddings regenerated : NO"
)

print(
    "\nNEXT:"
)

print(
    "Evaluate Length+GC and 4-mer on the SAME "
    "7,900 frozen PU folds."
)

print("=" * 122)
```

```python
# ======================================================================================
# ASD-ContextFusion
# RESUME INTERRUPTED GENCODE v50 FASTA DOWNLOAD
#
# - Keeps existing .part file
# - Resumes from already downloaded bytes
# - Retries transient network failures
# - Verifies gzip integrity before accepting file
# - Does NOT touch modeling / PU partitions / NT embeddings
# ======================================================================================

from pathlib import Path
import subprocess
import shutil
import gzip
import os


ROOT = Path(r"E:\ASD_ContextFusion_Q1")

RAW_SEQ_DIR = (
    ROOT /
    "01_raw" /
    "gencode_v50"
)

RAW_SEQ_DIR.mkdir(
    parents=True,
    exist_ok=True
)


GENCODE_URL = (
    "https://ftp.ebi.ac.uk/pub/databases/gencode/"
    "Gencode_human/release_50/"
    "gencode.v50.transcripts.fa.gz"
)


GENCODE_FASTA = (
    RAW_SEQ_DIR /
    "gencode.v50.transcripts.fa.gz"
)


# This is exactly the temporary filename created
# by the previous download_atomic() function.
PART_FILE = (
    GENCODE_FASTA.with_suffix(
        GENCODE_FASTA.suffix + ".part"
    )
)


print("=" * 118)
print("GENCODE v50 DOWNLOAD RECOVERY")
print("=" * 118)


# ======================================================================================
# 1. CURRENT STATE
# ======================================================================================

if GENCODE_FASTA.exists():

    print("\nFinal FASTA already exists:")
    print(GENCODE_FASTA)

    print(
        "Size:",
        round(
            GENCODE_FASTA.stat().st_size / 1024**2,
            2
        ),
        "MB"
    )

else:

    print("\nFinal FASTA not yet complete [EXPECTED]")


if PART_FILE.exists():

    part_mb = (
        PART_FILE.stat().st_size /
        1024**2
    )

    print("\nPartial download found [OK]")
    print(PART_FILE)

    print(
        "Already downloaded:",
        round(
            part_mb,
            2
        ),
        "MB"
    )

else:

    print(
        "\nNo .part file found."
    )

    print(
        "Download will start from byte 0."
    )


# ======================================================================================
# 2. USE WINDOWS CURL RESUME SUPPORT
#
# curl -C -:
#   resumes automatically from local file size.
#
# This is much safer than keeping one long urllib connection.
# ======================================================================================

if not GENCODE_FASTA.exists():

    curl_exe = (
        shutil.which("curl.exe")
        or
        shutil.which("curl")
    )


    if curl_exe is None:

        raise RuntimeError(
            "curl was not found on this Windows installation."
        )


    print("\nUsing:")
    print(curl_exe)


    print("\nResuming GENCODE download...")
    print(GENCODE_URL)


    command = [
        curl_exe,

        "-L",

        "--fail",

        "--retry",
        "30",

        "--retry-delay",
        "3",

        "--retry-all-errors",

        "--connect-timeout",
        "30",

        "-C",
        "-",

        "-o",
        str(PART_FILE),

        GENCODE_URL
    ]


    result = subprocess.run(
        command
    )


    if result.returncode != 0:

        raise RuntimeError(
            "\nGENCODE download did not complete.\n"
            f"curl exit code: {result.returncode}\n\n"
            "IMPORTANT: do NOT delete the .part file.\n"
            "Run THIS SAME CELL again; curl will resume "
            "from the newest downloaded byte."
        )


    print(
        "\nNetwork transfer completed [OK]"
    )


# ======================================================================================
# 3. CHOOSE FILE TO VALIDATE
# ======================================================================================

candidate = (
    GENCODE_FASTA
    if GENCODE_FASTA.exists()
    else
    PART_FILE
)


assert candidate.exists(), (
    f"Downloaded file missing:\n{candidate}"
)


print("\nDownloaded compressed size:")

print(
    round(
        candidate.stat().st_size /
        1024**2,
        2
    ),
    "MB"
)


# ======================================================================================
# 4. FULL GZIP INTEGRITY TEST
#
# Reading through the complete gzip stream forces Python
# to validate the gzip trailer / CRC.
# ======================================================================================

print("\nValidating complete gzip stream...")


uncompressed_bytes = 0


try:

    with gzip.open(
        candidate,
        "rb"
    ) as f:

        while True:

            block = f.read(
                8 * 1024 * 1024
            )

            if not block:
                break

            uncompressed_bytes += len(
                block
            )


except Exception as e:

    print(
        "\nGZIP VALIDATION FAILED"
    )

    print(
        type(e).__name__,
        ":",
        e
    )

    raise RuntimeError(
        "\nThe downloaded file is still incomplete/corrupted.\n"
        "DO NOT delete the .part file.\n"
        "Run this same recovery cell again."
    )


print(
    "GZIP integrity: PASS [OK]"
)

print(
    "Uncompressed data read:",
    round(
        uncompressed_bytes /
        1024**2,
        2
    ),
    "MB"
)


# ======================================================================================
# 5. ATOMICALLY PROMOTE .part -> FINAL FASTA
# ======================================================================================

if candidate == PART_FILE:

    if GENCODE_FASTA.exists():

        GENCODE_FASTA.unlink()


    os.replace(
        PART_FILE,
        GENCODE_FASTA
    )


print("\nFinal GENCODE FASTA:")
print(GENCODE_FASTA)

print(
    "Compressed size:",
    round(
        GENCODE_FASTA.stat().st_size /
        1024**2,
        2
    ),
    "MB"
)


assert GENCODE_FASTA.exists()

assert not PART_FILE.exists()


print("\n" + "=" * 118)
print("GENCODE v50 FASTA DOWNLOAD RECOVERED [OK]")
print("=" * 118)

print(
    "Existing partial bytes reused : YES"
)

print(
    "Full gzip integrity verified  : YES"
)

print(
    "Primary modeling modified     : NO"
)

print(
    "PU partitions modified        : NO"
)

print(
    "NT embeddings modified        : NO"
)

print("\nNEXT:")
print(
    "Rerun the previous EXACT GENCODE v50 "
    "SEQUENCE CONTROL BUILD cell."
)

print(
    "It will detect the completed FASTA and "
    "skip downloading it."
)

print("=" * 118)
```

```python
# ======================================================================================
# ASD-ContextFusion
# EXACT GENCODE v50 TRANSCRIPT RECOVERY + SEQUENCE CONTROL FEATURE GENERATION
#
# SCIENTIFIC RULES
# ----------------
# - Use the exact representative transcript_id already locked in NT-v2 metadata.
# - Use GENCODE Release 50 transcript sequences (GRCh38.p14).
# - Transcript ID + version MUST match.
# - FASTA sequence length MUST match locked sequence_length_nt.
# - Length+GC and 4-mer are computed from the SAME transcript used for NT-v2.
#
# 4-mer definition:
# - all oriented DNA 4-mers over A,C,G,T
# - 4^4 = 256 features
# - lexicographic order using alphabet A,C,G,T
# - counts normalized by number of valid A/C/G/T 4-mer windows
# - ambiguous-base-containing windows are excluded
#
# NO MODEL TRAINING
# NO PU SAMPLING
# NO PARTITION MODIFICATION
# NO NT REGENERATION
# ======================================================================================

from pathlib import Path
from itertools import product

import urllib.request
import gzip
import hashlib
import json

import numpy as np
import pandas as pd


# ======================================================================================
# 0. PATHS
# ======================================================================================

ROOT = Path(r"E:\ASD_ContextFusion_Q1")

NT_META_FILE = (
    ROOT /
    "03_features" /
    "nt_v2" /
    "NTv2_embedding_row_metadata.tsv"
)

OUT_DIR = (
    ROOT /
    "03_features" /
    "sequence_controls"
)

RAW_SEQ_DIR = (
    ROOT /
    "01_raw" /
    "gencode_v50"
)

OUT_DIR.mkdir(
    parents=True,
    exist_ok=True
)

RAW_SEQ_DIR.mkdir(
    parents=True,
    exist_ok=True
)


GENCODE_URL = (
    "https://ftp.ebi.ac.uk/pub/databases/gencode/"
    "Gencode_human/release_50/"
    "gencode.v50.transcripts.fa.gz"
)

GENCODE_FASTA = (
    RAW_SEQ_DIR /
    "gencode.v50.transcripts.fa.gz"
)

SELECTED_FASTA = (
    OUT_DIR /
    "GENCODE_v50_selected_18252_representative_transcripts.fa.gz"
)

LENGTH_GC_FILE = (
    OUT_DIR /
    "GENCODE_v50_selected_18252_Length_GC.tsv"
)

KMER_NPY = (
    OUT_DIR /
    "GENCODE_v50_selected_18252_4mer_frequency_18252x256.npy"
)

KMER_NAMES_FILE = (
    OUT_DIR /
    "GENCODE_v50_4mer_feature_names.tsv"
)

ROW_META_FILE = (
    OUT_DIR /
    "GENCODE_v50_sequence_control_row_metadata.tsv"
)

MANIFEST_FILE = (
    OUT_DIR /
    "sequence_control_manifest.json"
)


print("=" * 122)
print("ASD-CONTEXTFUSION - EXACT GENCODE v50 SEQUENCE CONTROL BUILD")
print("=" * 122)


# ======================================================================================
# 1. HELPERS
# ======================================================================================

def sha256_file(path, chunk_size=1024 * 1024):

    h = hashlib.sha256()

    with open(path, "rb") as f:

        while True:

            block = f.read(chunk_size)

            if not block:
                break

            h.update(block)

    return h.hexdigest()


def download_atomic(url, destination):

    destination = Path(destination)

    temp = destination.with_suffix(
        destination.suffix + ".part"
    )

    request = urllib.request.Request(
        url,
        headers={
            "User-Agent":
                "Mozilla/5.0 ASD-ContextFusion reproducible research"
        }
    )

    print("\nDownloading:")
    print(url)

    print("\nDestination:")
    print(destination)

    downloaded = 0

    with urllib.request.urlopen(
        request,
        timeout=120
    ) as response:

        total_header = response.headers.get(
            "Content-Length"
        )

        total = (
            int(total_header)
            if total_header is not None
            else None
        )

        with open(
            temp,
            "wb"
        ) as out:

            while True:

                block = response.read(
                    1024 * 1024
                )

                if not block:
                    break

                out.write(block)

                downloaded += len(block)

                if (
                    downloaded %
                    (50 * 1024 * 1024)
                    <
                    1024 * 1024
                ):

                    if total:

                        print(
                            f"Downloaded "
                            f"{downloaded / 1024**2:.1f} / "
                            f"{total / 1024**2:.1f} MB"
                        )

                    else:

                        print(
                            f"Downloaded "
                            f"{downloaded / 1024**2:.1f} MB"
                        )

    temp.replace(
        destination
    )

    print(
        "\nDownload complete [OK]"
    )


# ======================================================================================
# 2. LOAD LOCKED NT METADATA
# ======================================================================================

print("\n[1/8] LOCKED NT-v2 TRANSCRIPT METADATA")
print("-" * 122)


assert NT_META_FILE.exists(), (
    f"Missing:\n{NT_META_FILE}"
)


meta = pd.read_csv(
    NT_META_FILE,
    sep="\t",
    low_memory=False
)


required_columns = {
    "embedding_row",
    "original_gene_symbol",
    "gencode_gene_symbol",
    "transcript_id",
    "sequence_length",
    "sequence_length_nt"
}


assert required_columns.issubset(
    set(meta.columns)
), (
    "Required NT transcript provenance columns are missing."
)


assert len(meta) == 18252


meta["transcript_id"] = (
    meta["transcript_id"]
    .astype(str)
    .str.strip()
)

meta["original_gene_symbol"] = (
    meta["original_gene_symbol"]
    .astype(str)
    .str.strip()
)


assert meta[
    "transcript_id"
].is_unique, (
    "Locked representative transcript IDs are not unique."
)


assert meta[
    "embedding_row"
].is_unique


# Preserve exact NT embedding row order.
meta = (
    meta
    .sort_values(
        "embedding_row"
    )
    .reset_index(
        drop=True
    )
)


assert np.array_equal(
    meta["embedding_row"].to_numpy(),
    np.arange(
        18252
    )
)


expected_lengths = (
    pd.to_numeric(
        meta["sequence_length_nt"],
        errors="raise"
    )
    .astype(int)
    .to_numpy()
)


expected_lengths_alt = (
    pd.to_numeric(
        meta["sequence_length"],
        errors="raise"
    )
    .astype(int)
    .to_numpy()
)


assert np.array_equal(
    expected_lengths,
    expected_lengths_alt
), (
    "sequence_length and sequence_length_nt "
    "do not agree in locked metadata."
)


print(
    "Metadata rows       :",
    len(meta)
)

print(
    "Unique transcripts  :",
    meta["transcript_id"].nunique()
)

print(
    "Unique source genes :",
    meta["original_gene_symbol"].nunique()
)

print(
    "Length metadata     : CONSISTENT [OK]"
)


# ======================================================================================
# 3. USE EXISTING GENCODE RELEASE 50 FASTA
# ======================================================================================

print("\n[2/8] GENCODE RELEASE 50 TRANSCRIPT FASTA")
print("-" * 122)


if GENCODE_FASTA.exists():

    print(
        "Existing FASTA found:"
    )

    print(
        GENCODE_FASTA
    )

    print(
        "\nExisting file will be validated; "
        "it will NOT be redownloaded."
    )

else:

    download_atomic(
        GENCODE_URL,
        GENCODE_FASTA
    )


assert GENCODE_FASTA.exists()

assert (
    GENCODE_FASTA.stat().st_size
    > 0
)


# Full gzip integrity check
print("\nValidating GENCODE gzip integrity...")


with gzip.open(
    GENCODE_FASTA,
    "rb"
) as f:

    while True:

        block = f.read(
            8 * 1024 * 1024
        )

        if not block:
            break


print(
    "GENCODE gzip integrity: PASS [OK]"
)


gencode_sha = sha256_file(
    GENCODE_FASTA
)


print(
    "\nGENCODE FASTA size:",
    round(
        GENCODE_FASTA.stat().st_size /
        1024**2,
        2
    ),
    "MB"
)

print(
    "\nGENCODE FASTA SHA256:"
)

print(
    gencode_sha
)


# ======================================================================================
# 4. PRESPECIFY 4-MER VOCABULARY
# ======================================================================================

print("\n[3/8] PRESPECIFIED 4-MER VOCABULARY")
print("-" * 122)


DNA = "ACGT"

KMERS = [
    "".join(x)
    for x in product(
        DNA,
        repeat=4
    )
]


assert len(KMERS) == 256
assert len(set(KMERS)) == 256


KMER_TO_INDEX = {
    kmer: i
    for i, kmer in enumerate(
        KMERS
    )
}


pd.DataFrame({
    "kmer_index":
        np.arange(
            256,
            dtype=int
        ),
    "kmer":
        KMERS
}).to_csv(
    KMER_NAMES_FILE,
    sep="\t",
    index=False
)


print(
    "4-mer features:",
    len(KMERS)
)

print(
    "First 10:",
    KMERS[:10]
)

print(
    "Last 10 :",
    KMERS[-10:]
)


# ======================================================================================
# 5. PREALLOCATE FEATURES
# ======================================================================================

n = len(meta)


actual_length = np.zeros(
    n,
    dtype=np.int32
)

gc_fraction = np.full(
    n,
    np.nan,
    dtype=np.float64
)

valid_base_count = np.zeros(
    n,
    dtype=np.int32
)

ambiguous_base_count = np.zeros(
    n,
    dtype=np.int32
)

valid_4mer_windows = np.zeros(
    n,
    dtype=np.int32
)

invalid_4mer_windows = np.zeros(
    n,
    dtype=np.int32
)

X_4mer = np.zeros(
    (
        n,
        256
    ),
    dtype=np.float32
)


transcript_to_row = {
    tid: i
    for i, tid in enumerate(
        meta["transcript_id"]
    )
}


found = np.zeros(
    n,
    dtype=bool
)


# ======================================================================================
# 6. FEATURE FUNCTION
# ======================================================================================

def compute_sequence_features(
    seq,
    row_index
):

    seq = (
        seq
        .upper()
        .replace(
            "U",
            "T"
        )
    )

    L = len(seq)

    actual_length[
        row_index
    ] = L


    # ------------------------------------------------------------------
    # GC
    # ------------------------------------------------------------------

    a = seq.count("A")
    c = seq.count("C")
    g = seq.count("G")
    t = seq.count("T")

    valid_bases = (
        a +
        c +
        g +
        t
    )

    invalid_bases = (
        L -
        valid_bases
    )

    valid_base_count[
        row_index
    ] = valid_bases

    ambiguous_base_count[
        row_index
    ] = invalid_bases


    if valid_bases > 0:

        gc_fraction[
            row_index
        ] = (
            g + c
        ) / valid_bases


    # ------------------------------------------------------------------
    # 4-mer frequencies
    # ------------------------------------------------------------------

    counts = np.zeros(
        256,
        dtype=np.float64
    )

    valid_windows = 0
    invalid_windows = 0


    if L >= 4:

        for pos in range(
            L - 3
        ):

            kmer = seq[
                pos:
                pos + 4
            ]

            idx = KMER_TO_INDEX.get(
                kmer
            )

            if idx is None:

                invalid_windows += 1

            else:

                counts[idx] += 1.0
                valid_windows += 1


    valid_4mer_windows[
        row_index
    ] = valid_windows

    invalid_4mer_windows[
        row_index
    ] = invalid_windows


    if valid_windows > 0:

        counts /= (
            float(
                valid_windows
            )
        )


    X_4mer[
        row_index,
        :
    ] = counts.astype(
        np.float32
    )


# ======================================================================================
# 7. STREAM GENCODE FASTA AND EXTRACT ONLY LOCKED TRANSCRIPTS
# ======================================================================================

print("\n[4/8] EXACT TRANSCRIPT EXTRACTION")
print("-" * 122)


if SELECTED_FASTA.exists():

    SELECTED_FASTA.unlink()


current_id = None
current_target_row = None
sequence_chunks = []


def finish_record(
    transcript_id,
    row_index,
    chunks,
    selected_handle
):

    if (
        transcript_id is None
        or
        row_index is None
    ):
        return


    assert not found[
        row_index
    ], (
        f"Duplicate FASTA record for "
        f"{transcript_id}"
    )


    seq = "".join(
        chunks
    )


    compute_sequence_features(
        seq,
        row_index
    )


    found[
        row_index
    ] = True


    selected_handle.write(
        ">" +
        transcript_id +
        "\n"
    )


    width = 80

    for start in range(
        0,
        len(seq),
        width
    ):

        selected_handle.write(
            seq[
                start:
                start + width
            ] +
            "\n"
        )


with gzip.open(
    GENCODE_FASTA,
    "rt",
    encoding="utf-8",
    errors="strict"
) as fasta_in:

    with gzip.open(
        SELECTED_FASTA,
        "wt",
        encoding="utf-8"
    ) as selected_out:

        for line in fasta_in:

            if line.startswith(">"):

                finish_record(
                    current_id,
                    current_target_row,
                    sequence_chunks,
                    selected_out
                )


                header = (
                    line[1:]
                    .strip()
                )


                # First GENCODE FASTA field = transcript ID + version.
                fasta_tid = (
                    header
                    .split("|", 1)[0]
                    .split()[0]
                    .strip()
                )


                current_id = fasta_tid

                current_target_row = (
                    transcript_to_row.get(
                        fasta_tid
                    )
                )

                sequence_chunks = []

            else:

                if current_target_row is not None:

                    sequence_chunks.append(
                        line.strip()
                    )


        finish_record(
            current_id,
            current_target_row,
            sequence_chunks,
            selected_out
        )


n_found = int(
    found.sum()
)


print(
    "Required transcripts:",
    n
)

print(
    "Recovered transcripts:",
    n_found
)


if n_found != n:

    missing_ids = (
        meta.loc[
            ~found,
            [
                "original_gene_symbol",
                "gencode_gene_symbol",
                "transcript_id",
                "sequence_length_nt"
            ]
        ]
        .copy()
    )

    missing_file = (
        OUT_DIR /
        "ERROR_missing_GENCODE_v50_transcripts.tsv"
    )

    missing_ids.to_csv(
        missing_file,
        sep="\t",
        index=False
    )

    print(
        "\nMissing transcript IDs saved:"
    )

    print(
        missing_file
    )

    print(
        "\nFirst missing records:"
    )

    print(
        missing_ids
        .head(30)
        .to_string(
            index=False
        )
    )

    raise RuntimeError(
        "\nSTOP: GENCODE v50 FASTA did not recover "
        "all 18,252 locked transcript IDs.\n"
        "Do NOT build sequence controls until this is resolved."
    )


print(
    "\n18,252 / 18,252 exact transcript IDs recovered [OK]"
)


# ======================================================================================
# 8. MANDATORY SEQUENCE LENGTH VERIFICATION
# ======================================================================================

print("\n[5/8] EXACT SEQUENCE LENGTH VERIFICATION")
print("-" * 122)


length_match = (
    actual_length
    ==
    expected_lengths
)


n_length_match = int(
    length_match.sum()
)


print(
    "Exact length matches:",
    f"{n_length_match}/{n}"
)


if n_length_match != n:

    mismatch_df = meta.loc[
        ~length_match,
        [
            "embedding_row",
            "original_gene_symbol",
            "gencode_gene_symbol",
            "transcript_id",
            "sequence_length",
            "sequence_length_nt"
        ]
    ].copy()


    mismatch_df[
        "actual_fasta_length"
    ] = actual_length[
        ~length_match
    ]


    mismatch_file = (
        OUT_DIR /
        "ERROR_GENCODE_v50_sequence_length_mismatch.tsv"
    )


    mismatch_df.to_csv(
        mismatch_file,
        sep="\t",
        index=False
    )


    print(
        "\nLength mismatches saved:"
    )

    print(
        mismatch_file
    )


    print(
        mismatch_df
        .head(30)
        .to_string(
            index=False
        )
    )


    raise RuntimeError(
        "\nSTOP: transcript IDs matched but "
        "sequence lengths do not exactly match "
        "the locked NT-v2 transcript metadata."
    )


print(
    "\nSEQUENCE IDENTITY PROVENANCE CHECK PASSED [OK]"
)

print(
    "Every locked transcript ID has exactly "
    "the expected GENCODE v50 sequence length."
)


# ======================================================================================
# 9. FEATURE QC
# ======================================================================================

print("\n[6/8] LENGTH + GC + 4-MER QC")
print("-" * 122)


assert np.isfinite(
    gc_fraction
).all()


assert np.isfinite(
    X_4mer
).all()


assert (
    actual_length > 0
).all()


assert (
    valid_base_count > 0
).all()


assert (
    valid_4mer_windows > 0
).all()


row_sums = (
    X_4mer.sum(
        axis=1
    )
)


assert np.allclose(
    row_sums,
    1.0,
    atol=1e-5
), (
    "At least one normalized 4-mer row "
    "does not sum to 1."
)


print(
    "Length range:",
    int(
        actual_length.min()
    ),
    "-",
    int(
        actual_length.max()
    )
)


print(
    "Mean length:",
    round(
        float(
            actual_length.mean()
        ),
        2
    )
)


print(
    "GC range:",
    round(
        float(
            gc_fraction.min()
        ),
        6
    ),
    "-",
    round(
        float(
            gc_fraction.max()
        ),
        6
    )
)


print(
    "Mean GC:",
    round(
        float(
            gc_fraction.mean()
        ),
        6
    )
)


print(
    "Transcripts with ambiguous bases:",
    int(
        (
            ambiguous_base_count > 0
        ).sum()
    )
)


print(
    "Total ambiguous bases:",
    int(
        ambiguous_base_count.sum()
    )
)


print(
    "Total invalid 4-mer windows:",
    int(
        invalid_4mer_windows.sum()
    )
)


print(
    "4-mer matrix shape:",
    X_4mer.shape
)


# ======================================================================================
# 10. SAVE LENGTH+GC
# ======================================================================================

print("\n[7/8] SAVE EXACT SEQUENCE CONTROL FEATURES")
print("-" * 122)


length_gc = pd.DataFrame({

    "embedding_row":
        meta[
            "embedding_row"
        ].astype(int),

    "Gene_Name":
        meta[
            "original_gene_symbol"
        ],

    "gencode_gene_symbol":
        meta[
            "gencode_gene_symbol"
        ],

    "transcript_id":
        meta[
            "transcript_id"
        ],

    "transcript_length_nt":
        actual_length,

    "GC_fraction":
        gc_fraction,

    "valid_ACGT_bases":
        valid_base_count,

    "ambiguous_bases":
        ambiguous_base_count,

    "valid_4mer_windows":
        valid_4mer_windows,

    "invalid_4mer_windows":
        invalid_4mer_windows
})


length_gc.to_csv(
    LENGTH_GC_FILE,
    sep="\t",
    index=False
)


np.save(
    KMER_NPY,
    X_4mer
)


row_meta = meta[
    [
        "embedding_row",
        "original_gene_symbol",
        "gencode_gene_symbol",
        "gene_id",
        "transcript_id",
        "chromosome",
        "strand",
        "selection_basis",
        "sequence_length_nt",
        "ntv2_n_chunks"
    ]
].copy()


row_meta = row_meta.rename(
    columns={
        "original_gene_symbol":
            "Gene_Name"
    }
)


row_meta.to_csv(
    ROW_META_FILE,
    sep="\t",
    index=False
)


print(
    "Length+GC:"
)

print(
    LENGTH_GC_FILE
)


print(
    "\n4-mer matrix:"
)

print(
    KMER_NPY
)


print(
    "\n4-mer names:"
)

print(
    KMER_NAMES_FILE
)


print(
    "\nRow metadata:"
)

print(
    ROW_META_FILE
)


print(
    "\nSelected exact transcripts:"
)

print(
    SELECTED_FASTA
)


# ======================================================================================
# 11. FINAL SAVED-ARTIFACT VERIFICATION
# ======================================================================================

print("\n[8/8] FINAL INTEGRITY LOCK")
print("-" * 122)


X_saved = np.load(
    KMER_NPY,
    mmap_mode="r"
)


lg_saved = pd.read_csv(
    LENGTH_GC_FILE,
    sep="\t"
)


rows_saved = pd.read_csv(
    ROW_META_FILE,
    sep="\t"
)


assert X_saved.shape == (
    18252,
    256
)

assert len(
    lg_saved
) == 18252

assert len(
    rows_saved
) == 18252


assert (
    lg_saved[
        "Gene_Name"
    ].tolist()
    ==
    meta[
        "original_gene_symbol"
    ].tolist()
)


assert (
    rows_saved[
        "transcript_id"
    ].tolist()
    ==
    meta[
        "transcript_id"
    ].tolist()
)


selected_fasta_sha = sha256_file(
    SELECTED_FASTA
)

length_gc_sha = sha256_file(
    LENGTH_GC_FILE
)

kmer_sha = sha256_file(
    KMER_NPY
)

row_meta_sha = sha256_file(
    ROW_META_FILE
)

kmer_names_sha = sha256_file(
    KMER_NAMES_FILE
)


manifest = {

    "project":
        "ASD-ContextFusion",

    "stage":
        "Exact transcript sequence controls",

    "status":
        "completed_and_locked",

    "source": {

        "annotation":
            "GENCODE Release 50",

        "assembly":
            "GRCh38.p14",

        "source_url":
            GENCODE_URL,

        "source_fasta":
            str(
                GENCODE_FASTA
            ),

        "source_fasta_sha256":
            gencode_sha
    },

    "transcript_identity": {

        "source":
            "locked NT-v2 row metadata",

        "target_transcripts":
            18252,

        "exact_transcript_ids_recovered":
            int(
                found.sum()
            ),

        "exact_sequence_length_matches":
            int(
                length_match.sum()
            ),

        "transcript_version_preserved":
            True,

        "sequence_reselection_performed":
            False
    },

    "length_gc_control": {

        "features": [
            "transcript_length_nt",
            "GC_fraction"
        ],

        "length_definition":
            (
                "full selected transcript nucleotide "
                "sequence length"
            ),

        "gc_definition":
            (
                "(G+C)/(A+C+G+T); ambiguous "
                "bases excluded from denominator"
            )
    },

    "four_mer_control": {

        "k":
            4,

        "alphabet":
            "ACGT",

        "dimensions":
            256,

        "orientation":
            "transcript orientation; no reverse-complement collapsing",

        "feature_order":
            (
                "lexicographic Cartesian product using "
                "alphabet A,C,G,T"
            ),

        "normalization":
            (
                "count / number of valid ACGT 4-mer windows "
                "within each selected transcript"
            ),

        "ambiguous_window_policy":
            (
                "windows containing non-ACGT characters excluded"
            )
    },

    "qc": {

        "transcripts_with_ambiguous_bases":
            int(
                (
                    ambiguous_base_count > 0
                ).sum()
            ),

        "total_ambiguous_bases":
            int(
                ambiguous_base_count.sum()
            ),

        "total_invalid_4mer_windows":
            int(
                invalid_4mer_windows.sum()
            ),

        "all_4mer_rows_normalized":
            bool(
                np.allclose(
                    row_sums,
                    1.0,
                    atol=1e-5
                )
            )
    },

    "outputs": {

        "selected_transcript_fasta":
            str(
                SELECTED_FASTA
            ),

        "length_gc":
            str(
                LENGTH_GC_FILE
            ),

        "four_mer_matrix":
            str(
                KMER_NPY
            ),

        "four_mer_feature_names":
            str(
                KMER_NAMES_FILE
            ),

        "row_metadata":
            str(
                ROW_META_FILE
            )
    },

    "sha256": {

        "selected_transcript_fasta":
            selected_fasta_sha,

        "length_gc":
            length_gc_sha,

        "four_mer_matrix":
            kmer_sha,

        "four_mer_feature_names":
            kmer_names_sha,

        "row_metadata":
            row_meta_sha
    },

    "downstream_rule":
        (
            "Length+GC and 4-mer models must use the "
            "already frozen Dhindsa/Mantis 7,900 PU folds. "
            "No new sampling is permitted."
        )
}


with open(
    MANIFEST_FILE,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        manifest,
        f,
        indent=2
    )


print(
    "Manifest:"
)

print(
    MANIFEST_FILE
)


print(
    "\nSource FASTA SHA256:"
)

print(
    gencode_sha
)


print(
    "\nSelected FASTA SHA256:"
)

print(
    selected_fasta_sha
)


print(
    "\nLength+GC SHA256:"
)

print(
    length_gc_sha
)


print(
    "\n4-mer SHA256:"
)

print(
    kmer_sha
)


print("\n" + "=" * 122)
print("EXACT GENCODE v50 SEQUENCE CONTROLS LOCKED [OK]")
print("=" * 122)

print(
    "Exact transcript IDs      :",
    f"{int(found.sum())}/18252"
)

print(
    "Exact sequence lengths    :",
    f"{int(length_match.sum())}/18252"
)

print(
    "Length+GC matrix          :",
    "18252 x 2"
)

print(
    "4-mer matrix              :",
    X_saved.shape
)

print(
    "New PU sampling           : NO"
)

print(
    "Primary models modified   : NO"
)

print(
    "NT embeddings regenerated : NO"
)

print(
    "\nNEXT:"
)

print(
    "Evaluate Length+GC and 4-mer on the SAME "
    "7,900 frozen PU folds."
)

print("=" * 122)
```

```python
# ======================================================================================
# ASD-ContextFusion
# SEQUENCE CONTROLS ON EXACT SAME 7,900 FROZEN PU FOLDS
#
# CONTROLS
# --------
# 1. Transcript Length + GC
# 2. Prespecified 256-dimensional oriented 4-mer frequencies
#
# MODEL FOR BOTH CONTROLS
# -----------------------
# StandardScaler fitted ONLY on training fold
# + L2 LogisticRegression
# penalty='l2', C=1.0, solver='lbfgs', max_iter=2000
#
# RATIONALE
# ---------
# Same linear-probe model used for NT-v2 so representation quality is compared
# without changing classifier capacity.
#
# CRITICAL
# --------
# - SAME frozen Dhindsa/Mantis PU membership
# - SAME 10 stochastic iterations
# - SAME 79 balanced datasets / iteration
# - SAME 10 folds / balanced dataset
# - SAME 7,900 test folds
# - NO new sampling
# - NO model selection
# - NO hyperparameter tuning
# - Primary Context / NT results untouched
#
# MEMORY SAFE
# -----------
# One stochastic iteration = one fresh legacy Python process.
# Completed iterations automatically skipped.
# ======================================================================================

from pathlib import Path
import subprocess
import os
import pandas as pd
import textwrap


# ======================================================================================
# 0. PATHS
# ======================================================================================

ROOT = Path(r"E:\ASD_ContextFusion_Q1")

LEGACY_PYTHON = (
    ROOT /
    "envs" /
    "mantis37" /
    "python.exe"
)

RUNNER = (
    ROOT /
    "00_admin" /
    "run_sequence_controls_frozen_PU.py"
)

CHECKPOINT_DIR = (
    ROOT /
    "04_models" /
    "checkpoints" /
    "sequence_controls"
)

RESULT_DIR = (
    ROOT /
    "05_results" /
    "proposed" /
    "sequence_controls"
)

LOG_FILE = (
    ROOT /
    "08_logs" /
    "sequence_controls_frozen_PU.log"
)


CHECKPOINT_DIR.mkdir(
    parents=True,
    exist_ok=True
)

RESULT_DIR.mkdir(
    parents=True,
    exist_ok=True
)

LOG_FILE.parent.mkdir(
    parents=True,
    exist_ok=True
)


assert LEGACY_PYTHON.exists(), LEGACY_PYTHON


print("=" * 122)
print("ASD-CONTEXTFUSION - FROZEN-PU SEQUENCE CONTROL EVALUATION")
print("=" * 122)


# ======================================================================================
# 1. BUILD LOCKED LEGACY RUNNER
# ======================================================================================

runner_code = r'''
from __future__ import print_function

from pathlib import Path
import os
import sys
import time
import json
import hashlib

import numpy as np
import pandas as pd

from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    roc_auc_score,
    average_precision_score
)


# ==================================================================================
# 0. PATHS
# ==================================================================================

ROOT = Path(r"E:\ASD_ContextFusion_Q1")

MASTER_FILE = (
    ROOT /
    "03_features" /
    "multimodal_master" /
    "dhindsa_NTv2_row_preserved_master.tsv"
)

LENGTH_GC_FILE = (
    ROOT /
    "03_features" /
    "sequence_controls" /
    "GENCODE_v50_selected_18252_Length_GC.tsv"
)

KMER_FILE = (
    ROOT /
    "03_features" /
    "sequence_controls" /
    "GENCODE_v50_selected_18252_4mer_frequency_18252x256.npy"
)

SEQ_MANIFEST = (
    ROOT /
    "03_features" /
    "sequence_controls" /
    "sequence_control_manifest.json"
)

PARTITION_DIR = (
    ROOT /
    "04_models" /
    "checkpoints" /
    "dhindsa_PU_common_universe"
)

CHECKPOINT_DIR = (
    ROOT /
    "04_models" /
    "checkpoints" /
    "sequence_controls"
)

OUT_DIR = (
    ROOT /
    "05_results" /
    "proposed" /
    "sequence_controls"
)

PRIMARY_DIR = (
    ROOT /
    "05_results" /
    "proposed" /
    "primary_context_NTv2"
)


CHECKPOINT_DIR.mkdir(
    parents=True,
    exist_ok=True
)

OUT_DIR.mkdir(
    parents=True,
    exist_ok=True
)


print("=" * 116)
print("ASD-CONTEXTFUSION - LENGTH+GC + 4-MER FROZEN-PU EVALUATION")
print("=" * 116)


# ==================================================================================
# 1. LOAD LOCKED DATA
# ==================================================================================

for p in [
    MASTER_FILE,
    LENGTH_GC_FILE,
    KMER_FILE,
    SEQ_MANIFEST
]:
    assert p.exists(), "Missing:\n%s" % str(p)


master = pd.read_csv(
    str(MASTER_FILE),
    sep="\t",
    low_memory=False
)


length_gc = pd.read_csv(
    str(LENGTH_GC_FILE),
    sep="\t",
    low_memory=False
)


X_4mer = np.load(
    str(KMER_FILE),
    mmap_mode="r"
)


with open(
    str(SEQ_MANIFEST),
    "r"
) as f:
    seq_manifest = json.load(f)


# Exact structural locks.
assert len(master) == 18253
assert master["Gene_Name"].nunique() == 18252

assert len(length_gc) == 18252
assert X_4mer.shape == (18252, 256)

assert "dhindsa_row_id" in master.columns
assert "NT_row_index" in master.columns
assert "known_gene" in master.columns

assert "embedding_row" in length_gc.columns
assert "Gene_Name" in length_gc.columns
assert "transcript_length_nt" in length_gc.columns
assert "GC_fraction" in length_gc.columns


master["dhindsa_row_id"] = (
    master["dhindsa_row_id"]
    .astype(int)
)

master["NT_row_index"] = (
    master["NT_row_index"]
    .astype(int)
)

master["known_gene"] = (
    master["known_gene"]
    .astype(int)
)


length_gc["embedding_row"] = (
    length_gc["embedding_row"]
    .astype(int)
)


# Sequence-control rows MUST remain in exact NT embedding order.
assert np.array_equal(
    length_gc["embedding_row"].to_numpy(),
    np.arange(18252)
)


# Exact source-gene identity by NT row.
nt_gene_lookup = (
    length_gc["Gene_Name"]
    .astype(str)
    .to_numpy()
)


for row in master[
    [
        "Gene_Name",
        "NT_row_index"
    ]
].itertuples(index=False):

    assert (
        str(row.Gene_Name)
        ==
        str(
            nt_gene_lookup[
                int(row.NT_row_index)
            ]
        )
    )


master_idx = master.set_index(
    "dhindsa_row_id",
    drop=False
)


# Pre-create control matrices in exact NT row order.
X_length_gc = (
    length_gc[
        [
            "transcript_length_nt",
            "GC_fraction"
        ]
    ]
    .to_numpy(
        dtype=np.float32
    )
)


assert X_length_gc.shape == (
    18252,
    2
)

assert np.isfinite(
    X_length_gc
).all()

assert np.isfinite(
    np.asarray(X_4mer)
).all()


print("")
print("[DATA]")
print("Multimodal rows       :", len(master))
print("Unique genes          :", master["Gene_Name"].nunique())
print("ASD genes             :", master.loc[master.known_gene == 1, "Gene_Name"].nunique())
print("Length+GC dimensions  :", X_length_gc.shape[1])
print("4-mer dimensions      :", X_4mer.shape[1])


# ==================================================================================
# 2. LOCK MODEL SPECIFICATION
# ==================================================================================

LOGISTIC_PARAMS = {
    "penalty":
        "l2",

    "C":
        1.0,

    "solver":
        "lbfgs",

    "max_iter":
        2000
}


print("")
print("[MODELS]")
print(
    "Length+GC : StandardScaler + "
    "L2 LogisticRegression"
)

print(
    "4-mer     : StandardScaler + "
    "L2 LogisticRegression"
)

print("")
print("Parameters:")
print(LOGISTIC_PARAMS)

print("")
print(
    "No tuning. Same linear-probe specification "
    "as the NT-v2 branch."
)


# ==================================================================================
# 3. CHECKPOINT VALIDATION
# ==================================================================================

def checkpoint_valid(iteration):

    pred_file = (
        CHECKPOINT_DIR /
        (
            "iteration_%02d_OOF_predictions.tsv.gz"
            % iteration
        )
    )

    metric_file = (
        CHECKPOINT_DIR /
        (
            "iteration_%02d_fold_metrics.tsv"
            % iteration
        )
    )

    if not (
        pred_file.exists()
        and
        metric_file.exists()
    ):
        return False

    try:

        m = pd.read_csv(
            str(metric_file),
            sep="\t"
        )

        p = pd.read_csv(
            str(pred_file),
            sep="\t",
            compression="gzip"
        )

        if len(m) != 790:
            return False

        if (
            m["balanced_dataset"].nunique()
            != 79
        ):
            return False

        if (
            m["fold"].nunique()
            != 10
        ):
            return False


        required_metric = {
            "LengthGC_AUROC",
            "LengthGC_AP",
            "Fourmer_AUROC",
            "Fourmer_AP"
        }

        if not required_metric.issubset(
            set(m.columns)
        ):
            return False


        required_pred = {
            "dhindsa_row_id",
            "Gene_Name",
            "known_gene",
            "NT_row_index",
            "LengthGC_probability",
            "Fourmer_probability"
        }

        if not required_pred.issubset(
            set(p.columns)
        ):
            return False


        unl = p.loc[
            p["known_gene"] == 0
        ]

        # Authors' frozen sampling guarantees each
        # unlabeled ROW is tested exactly once per iteration.
        if len(unl) != 18063:
            return False

        if (
            unl["dhindsa_row_id"].nunique()
            != 18063
        ):
            return False


        return True

    except Exception:

        return False


# ==================================================================================
# 4. SINGLE-ITERATION MODE
# ==================================================================================

TARGET = os.environ.get(
    "ASD_CF_SEQUENCE_TARGET_ITERATION"
)


if TARGET is not None:

    iteration = int(TARGET)

    assert 1 <= iteration <= 10


    if checkpoint_valid(iteration):

        print("")
        print(
            "Iteration %02d already complete - SKIPPING"
            % iteration
        )

        sys.exit(0)


    membership_file = (
        PARTITION_DIR /
        (
            "iteration_%02d_test_fold_membership.tsv.gz"
            % iteration
        )
    )

    assert membership_file.exists()


    print("")
    print("=" * 116)
    print(
        "STOCHASTIC ITERATION %02d"
        % iteration
    )
    print("=" * 116)


    membership = pd.read_csv(
        str(membership_file),
        sep="\t",
        compression="gzip"
    )


    required_membership = {
        "iteration",
        "balanced_dataset",
        "fold",
        "dhindsa_row_id",
        "Gene_Name",
        "known_gene"
    }


    assert required_membership.issubset(
        set(membership.columns)
    )

    assert membership[
        "iteration"
    ].nunique() == 1

    assert int(
        membership[
            "iteration"
        ].iloc[0]
    ) == iteration

    assert (
        membership[
            "balanced_dataset"
        ].nunique()
        == 79
    )

    assert (
        membership[
            "fold"
        ].nunique()
        == 10
    )


    iteration_predictions = []
    iteration_metrics = []

    split_counter = 0
    iter_start = time.time()


    # ==============================================================================
    # EACH BALANCED DATASET
    # ==============================================================================

    for balanced_dataset, bal in membership.groupby(
        "balanced_dataset",
        sort=True
    ):

        balanced_dataset = int(
            balanced_dataset
        )


        # Across all 10 folds, this is the complete
        # authors' balanced dataset.
        balanced_ids = (
            bal[
                "dhindsa_row_id"
            ]
            .astype(int)
            .tolist()
        )


        assert len(
            balanced_ids
        ) == len(
            set(balanced_ids)
        )

        assert (
            380
            <= len(balanced_ids)
            <= 383
        )


        # ==========================================================================
        # EACH OF THE EXACT 10 FOLDS
        # ==========================================================================

        for fold in range(10):

            split_counter += 1


            test_part = bal.loc[
                bal["fold"] == fold
            ].copy()


            test_ids = (
                test_part[
                    "dhindsa_row_id"
                ]
                .astype(int)
                .tolist()
            )


            test_id_set = set(
                test_ids
            )


            train_ids = [
                rid
                for rid in balanced_ids
                if rid not in test_id_set
            ]


            assert (
                set(train_ids)
                .isdisjoint(
                    test_id_set
                )
            )


            train_df = master_idx.loc[
                train_ids
            ]

            test_df = master_idx.loc[
                test_ids
            ]


            y_train = (
                train_df[
                    "known_gene"
                ]
                .to_numpy(
                    dtype=np.int32
                )
            )

            y_test = (
                test_df[
                    "known_gene"
                ]
                .to_numpy(
                    dtype=np.int32
                )
            )


            assert set(
                np.unique(
                    y_train
                )
            ) == {0, 1}

            assert set(
                np.unique(
                    y_test
                )
            ) == {0, 1}


            nt_train_idx = (
                train_df[
                    "NT_row_index"
                ]
                .to_numpy(
                    dtype=np.int64
                )
            )

            nt_test_idx = (
                test_df[
                    "NT_row_index"
                ]
                .to_numpy(
                    dtype=np.int64
                )
            )


            # ==================================================================
            # A. LENGTH + GC
            # ==================================================================

            Xlg_train = np.asarray(
                X_length_gc[
                    nt_train_idx
                ],
                dtype=np.float32
            )

            Xlg_test = np.asarray(
                X_length_gc[
                    nt_test_idx
                ],
                dtype=np.float32
            )


            scaler_lg = StandardScaler()

            Xlg_train_scaled = (
                scaler_lg.fit_transform(
                    Xlg_train
                )
            )

            Xlg_test_scaled = (
                scaler_lg.transform(
                    Xlg_test
                )
            )


            model_lg = LogisticRegression(
                **LOGISTIC_PARAMS
            )


            model_lg.fit(
                Xlg_train_scaled,
                y_train
            )


            lg_prob = (
                model_lg
                .predict_proba(
                    Xlg_test_scaled
                )[:, 1]
            )


            # ==================================================================
            # B. PRESPECIFIED ORIENTED 4-MER
            # ==================================================================

            X4_train = np.asarray(
                X_4mer[
                    nt_train_idx
                ],
                dtype=np.float32
            )

            X4_test = np.asarray(
                X_4mer[
                    nt_test_idx
                ],
                dtype=np.float32
            )


            scaler_4 = StandardScaler()

            X4_train_scaled = (
                scaler_4.fit_transform(
                    X4_train
                )
            )

            X4_test_scaled = (
                scaler_4.transform(
                    X4_test
                )
            )


            model_4 = LogisticRegression(
                **LOGISTIC_PARAMS
            )


            model_4.fit(
                X4_train_scaled,
                y_train
            )


            fourmer_prob = (
                model_4
                .predict_proba(
                    X4_test_scaled
                )[:, 1]
            )


            # ==================================================================
            # C. SPLIT METRICS
            # ==================================================================

            lg_roc = roc_auc_score(
                y_test,
                lg_prob
            )

            lg_ap = average_precision_score(
                y_test,
                lg_prob
            )

            fourmer_roc = roc_auc_score(
                y_test,
                fourmer_prob
            )

            fourmer_ap = average_precision_score(
                y_test,
                fourmer_prob
            )


            iteration_metrics.append(
                {
                    "iteration":
                        iteration,

                    "balanced_dataset":
                        balanced_dataset,

                    "fold":
                        fold,

                    "n_train":
                        len(train_ids),

                    "n_test":
                        len(test_ids),

                    "test_positive":
                        int(
                            y_test.sum()
                        ),

                    "test_unlabelled":
                        int(
                            len(y_test)
                            -
                            y_test.sum()
                        ),

                    "LengthGC_AUROC":
                        float(
                            lg_roc
                        ),

                    "LengthGC_AP":
                        float(
                            lg_ap
                        ),

                    "Fourmer_AUROC":
                        float(
                            fourmer_roc
                        ),

                    "Fourmer_AP":
                        float(
                            fourmer_ap
                        ),

                    "Delta_AUROC_Fourmer_minus_LengthGC":
                        float(
                            fourmer_roc
                            -
                            lg_roc
                        ),

                    "Delta_AP_Fourmer_minus_LengthGC":
                        float(
                            fourmer_ap
                            -
                            lg_ap
                        )
                }
            )


            # ==================================================================
            # D. SAVE EVERY OOF TEST PREDICTION
            # ==================================================================

            test_meta = test_df[
                [
                    "dhindsa_row_id",
                    "Gene_Name",
                    "known_gene",
                    "NT_row_index"
                ]
            ].copy()


            for j in range(
                len(test_meta)
            ):

                row = test_meta.iloc[j]

                iteration_predictions.append(
                    {
                        "iteration":
                            iteration,

                        "balanced_dataset":
                            balanced_dataset,

                        "fold":
                            fold,

                        "dhindsa_row_id":
                            int(
                                row[
                                    "dhindsa_row_id"
                                ]
                            ),

                        "Gene_Name":
                            str(
                                row[
                                    "Gene_Name"
                                ]
                            ),

                        "known_gene":
                            int(
                                row[
                                    "known_gene"
                                ]
                            ),

                        "NT_row_index":
                            int(
                                row[
                                    "NT_row_index"
                                ]
                            ),

                        "LengthGC_probability":
                            float(
                                lg_prob[j]
                            ),

                        "Fourmer_probability":
                            float(
                                fourmer_prob[j]
                            )
                    }
                )


            if (
                split_counter % 100
                == 0
            ):

                elapsed = (
                    time.time()
                    -
                    iter_start
                )

                print(
                    "Iteration %02d | %d / 790 splits | %.1f min"
                    % (
                        iteration,
                        split_counter,
                        elapsed / 60.0
                    )
                )


    # ==============================================================================
    # ITERATION CHECKPOINT
    # ==============================================================================

    metrics_df = pd.DataFrame(
        iteration_metrics
    )

    predictions_df = pd.DataFrame(
        iteration_predictions
    )


    assert len(
        metrics_df
    ) == 790

    assert (
        metrics_df[
            "balanced_dataset"
        ].nunique()
        == 79
    )

    assert (
        metrics_df[
            "fold"
        ].nunique()
        == 10
    )


    unlabeled_pred = (
        predictions_df.loc[
            predictions_df[
                "known_gene"
            ] == 0
        ]
    )


    assert len(
        unlabeled_pred
    ) == 18063

    assert (
        unlabeled_pred[
            "dhindsa_row_id"
        ].nunique()
        == 18063
    )


    # All 190 positive genes must be observed
    # across the stochastic iteration.
    assert (
        predictions_df.loc[
            predictions_df[
                "known_gene"
            ] == 1,
            "Gene_Name"
        ].nunique()
        == 190
    )


    pred_file = (
        CHECKPOINT_DIR /
        (
            "iteration_%02d_OOF_predictions.tsv.gz"
            % iteration
        )
    )

    metric_file = (
        CHECKPOINT_DIR /
        (
            "iteration_%02d_fold_metrics.tsv"
            % iteration
        )
    )


    predictions_df.to_csv(
        str(pred_file),
        sep="\t",
        index=False,
        compression="gzip"
    )

    metrics_df.to_csv(
        str(metric_file),
        sep="\t",
        index=False
    )


    print("")
    print(
        "Iteration %02d completed."
        % iteration
    )

    print(
        "Length+GC AUROC: %.4f | AP: %.4f"
        % (
            metrics_df[
                "LengthGC_AUROC"
            ].mean(),

            metrics_df[
                "LengthGC_AP"
            ].mean()
        )
    )

    print(
        "4-mer     AUROC: %.4f | AP: %.4f"
        % (
            metrics_df[
                "Fourmer_AUROC"
            ].mean(),

            metrics_df[
                "Fourmer_AP"
            ].mean()
        )
    )

    print(
        "Checkpoint saved:",
        pred_file
    )

    print("")
    print("=" * 116)
    print(
        "ITERATION %02d CHECKPOINT COMPLETE - PROCESS EXITING"
        % iteration
    )
    print(
        "RAM will now be released before the next iteration."
    )
    print("=" * 116)

    sys.exit(0)


# ==================================================================================
# 5. AGGREGATION-ONLY MODE
#
# No target iteration -> ALL 10 checkpoints must already exist.
# This mode DOES NOT train models.
# ==================================================================================

print("")
print("=" * 116)
print("SEQUENCE-CONTROL AGGREGATION-ONLY MODE")
print("=" * 116)


metric_files = []
prediction_files = []


for iteration in range(
    1,
    11
):

    assert checkpoint_valid(
        iteration
    ), (
        "Iteration %02d is not a valid completed checkpoint. "
        "Aggregation stopped."
        % iteration
    )


    metric_files.append(
        CHECKPOINT_DIR /
        (
            "iteration_%02d_fold_metrics.tsv"
            % iteration
        )
    )

    prediction_files.append(
        CHECKPOINT_DIR /
        (
            "iteration_%02d_OOF_predictions.tsv.gz"
            % iteration
        )
    )


print(
    "All 10 sequence-control checkpoints verified [OK]"
)


# ==================================================================================
# 6. COMBINE ALL 10 ITERATIONS
# ==================================================================================

all_metrics = pd.concat(
    [
        pd.read_csv(
            str(p),
            sep="\t"
        )
        for p in metric_files
    ],
    ignore_index=True
)


all_predictions = pd.concat(
    [
        pd.read_csv(
            str(p),
            sep="\t",
            compression="gzip"
        )
        for p in prediction_files
    ],
    ignore_index=True
)


assert len(
    all_metrics
) == 7900


ALL_METRIC_FILE = (
    OUT_DIR /
    "SequenceControls_all_7900_fold_metrics.tsv"
)

ALL_PRED_FILE = (
    OUT_DIR /
    "SequenceControls_all_OOF_predictions.tsv.gz"
)


all_metrics.to_csv(
    str(
        ALL_METRIC_FILE
    ),
    sep="\t",
    index=False
)


all_predictions.to_csv(
    str(
        ALL_PRED_FILE
    ),
    sep="\t",
    index=False,
    compression="gzip"
)


# ==================================================================================
# 7. SPLIT-LEVEL SUMMARY
# ==================================================================================

split_summary = pd.DataFrame(
    [
        {
            "Model":
                "LengthGC_L2_Logistic",

            "AUROC_mean":
                all_metrics[
                    "LengthGC_AUROC"
                ].mean(),

            "AUROC_sd":
                all_metrics[
                    "LengthGC_AUROC"
                ].std(),

            "AP_mean":
                all_metrics[
                    "LengthGC_AP"
                ].mean(),

            "AP_sd":
                all_metrics[
                    "LengthGC_AP"
                ].std()
        },

        {
            "Model":
                "Fourmer_L2_Logistic",

            "AUROC_mean":
                all_metrics[
                    "Fourmer_AUROC"
                ].mean(),

            "AUROC_sd":
                all_metrics[
                    "Fourmer_AUROC"
                ].std(),

            "AP_mean":
                all_metrics[
                    "Fourmer_AP"
                ].mean(),

            "AP_sd":
                all_metrics[
                    "Fourmer_AP"
                ].std()
        }
    ]
)


SPLIT_SUMMARY_FILE = (
    OUT_DIR /
    "SequenceControls_split_level_summary.tsv"
)


split_summary.to_csv(
    str(
        SPLIT_SUMMARY_FILE
    ),
    sep="\t",
    index=False
)


# ==================================================================================
# 8. GENE-LEVEL AGGREGATED OOF SCORES
# ==================================================================================

gene_predictions = (
    all_predictions
    .groupby(
        [
            "Gene_Name",
            "known_gene"
        ],
        as_index=False
    )
    .agg(
        LengthGC_probability=(
            "LengthGC_probability",
            "mean"
        ),

        Fourmer_probability=(
            "Fourmer_probability",
            "mean"
        ),

        OOF_prediction_count=(
            "LengthGC_probability",
            "size"
        )
    )
)


assert (
    gene_predictions[
        "Gene_Name"
    ].nunique()
    == 18252
)


assert (
    gene_predictions.loc[
        gene_predictions[
            "known_gene"
        ] == 1,
        "Gene_Name"
    ].nunique()
    == 190
)


GENE_PRED_FILE = (
    OUT_DIR /
    "SequenceControls_gene_level_OOF_scores.tsv"
)


gene_predictions.to_csv(
    str(
        GENE_PRED_FILE
    ),
    sep="\t",
    index=False
)


y_gene = (
    gene_predictions[
        "known_gene"
    ]
    .to_numpy(
        dtype=np.int32
    )
)


gene_lg_roc = roc_auc_score(
    y_gene,
    gene_predictions[
        "LengthGC_probability"
    ]
)

gene_lg_ap = average_precision_score(
    y_gene,
    gene_predictions[
        "LengthGC_probability"
    ]
)


gene_4_roc = roc_auc_score(
    y_gene,
    gene_predictions[
        "Fourmer_probability"
    ]
)

gene_4_ap = average_precision_score(
    y_gene,
    gene_predictions[
        "Fourmer_probability"
    ]
)


gene_summary = pd.DataFrame(
    [
        {
            "Model":
                "LengthGC_L2_Logistic",

            "Gene_level_AUROC":
                gene_lg_roc,

            "Gene_level_AP":
                gene_lg_ap
        },

        {
            "Model":
                "Fourmer_L2_Logistic",

            "Gene_level_AUROC":
                gene_4_roc,

            "Gene_level_AP":
                gene_4_ap
        }
    ]
)


GENE_SUMMARY_FILE = (
    OUT_DIR /
    "SequenceControls_gene_level_summary.tsv"
)


gene_summary.to_csv(
    str(
        GENE_SUMMARY_FILE
    ),
    sep="\t",
    index=False
)


# ==================================================================================
# 9. VERIFY / MERGE WITH EXISTING PRIMARY CONTEXT + NT RESULTS
# ==================================================================================

PRIMARY_METRIC_FILE = (
    PRIMARY_DIR /
    "Context_NTv2_all_7900_fold_metrics.tsv"
)

PRIMARY_GENE_FILE = (
    PRIMARY_DIR /
    "Context_NTv2_gene_level_OOF_scores.tsv"
)


assert PRIMARY_METRIC_FILE.exists()
assert PRIMARY_GENE_FILE.exists()


primary_metrics = pd.read_csv(
    str(
        PRIMARY_METRIC_FILE
    ),
    sep="\t"
)


primary_gene = pd.read_csv(
    str(
        PRIMARY_GENE_FILE
    ),
    sep="\t"
)


assert len(
    primary_metrics
) == 7900


KEYS = [
    "iteration",
    "balanced_dataset",
    "fold"
]


combined_metrics = primary_metrics.merge(
    all_metrics,
    on=KEYS,
    how="inner",
    suffixes=(
        "_primary",
        "_control"
    ),
    validate="one_to_one"
)


assert len(
    combined_metrics
) == 7900


# Exact split-size/label consistency.
for col in [
    "n_train",
    "n_test",
    "test_positive",
    "test_unlabelled"
]:

    left = (
        col +
        "_primary"
    )

    right = (
        col +
        "_control"
    )

    assert np.array_equal(
        combined_metrics[
            left
        ].to_numpy(),

        combined_metrics[
            right
        ].to_numpy()
    )


COMBINED_METRIC_FILE = (
    OUT_DIR /
    "Context_NTv2_LengthGC_4mer_all_7900_fold_metrics.tsv"
)


combined_metrics.to_csv(
    str(
        COMBINED_METRIC_FILE
    ),
    sep="\t",
    index=False
)


combined_gene = primary_gene.merge(
    gene_predictions[
        [
            "Gene_Name",
            "known_gene",
            "LengthGC_probability",
            "Fourmer_probability"
        ]
    ],
    on=[
        "Gene_Name",
        "known_gene"
    ],
    how="inner",
    validate="one_to_one"
)


assert len(
    combined_gene
) == 18252

assert (
    combined_gene[
        "Gene_Name"
    ].nunique()
    == 18252
)


COMBINED_GENE_FILE = (
    OUT_DIR /
    "Context_NTv2_LengthGC_4mer_gene_level_OOF_scores.tsv"
)


combined_gene.to_csv(
    str(
        COMBINED_GENE_FILE
    ),
    sep="\t",
    index=False
)


# ==================================================================================
# 10. FOUR-MODEL COMPARISON SUMMARY
# ==================================================================================

split_all_models = pd.DataFrame(
    [
        {
            "Model":
                "Dhindsa_Context_XGBoost",

            "AUROC_mean":
                primary_metrics[
                    "Context_AUROC"
                ].mean(),

            "AUROC_sd":
                primary_metrics[
                    "Context_AUROC"
                ].std(),

            "AP_mean":
                primary_metrics[
                    "Context_AP"
                ].mean(),

            "AP_sd":
                primary_metrics[
                    "Context_AP"
                ].std()
        },

        {
            "Model":
                "NTv2_L2_Logistic",

            "AUROC_mean":
                primary_metrics[
                    "NT_AUROC"
                ].mean(),

            "AUROC_sd":
                primary_metrics[
                    "NT_AUROC"
                ].std(),

            "AP_mean":
                primary_metrics[
                    "NT_AP"
                ].mean(),

            "AP_sd":
                primary_metrics[
                    "NT_AP"
                ].std()
        },

        {
            "Model":
                "LengthGC_L2_Logistic",

            "AUROC_mean":
                all_metrics[
                    "LengthGC_AUROC"
                ].mean(),

            "AUROC_sd":
                all_metrics[
                    "LengthGC_AUROC"
                ].std(),

            "AP_mean":
                all_metrics[
                    "LengthGC_AP"
                ].mean(),

            "AP_sd":
                all_metrics[
                    "LengthGC_AP"
                ].std()
        },

        {
            "Model":
                "Fourmer_L2_Logistic",

            "AUROC_mean":
                all_metrics[
                    "Fourmer_AUROC"
                ].mean(),

            "AUROC_sd":
                all_metrics[
                    "Fourmer_AUROC"
                ].std(),

            "AP_mean":
                all_metrics[
                    "Fourmer_AP"
                ].mean(),

            "AP_sd":
                all_metrics[
                    "Fourmer_AP"
                ].std()
        }
    ]
)


FOUR_MODEL_SPLIT_SUMMARY = (
    OUT_DIR /
    "FourModel_split_level_summary.tsv"
)


split_all_models.to_csv(
    str(
        FOUR_MODEL_SPLIT_SUMMARY
    ),
    sep="\t",
    index=False
)


y_combined = (
    combined_gene[
        "known_gene"
    ]
    .to_numpy(
        dtype=np.int32
    )
)


model_probability_columns = [
    (
        "Dhindsa_Context_XGBoost",
        "Context_probability"
    ),

    (
        "NTv2_L2_Logistic",
        "NT_probability"
    ),

    (
        "LengthGC_L2_Logistic",
        "LengthGC_probability"
    ),

    (
        "Fourmer_L2_Logistic",
        "Fourmer_probability"
    )
]


gene_rows = []


for model_name, prob_col in model_probability_columns:

    gene_rows.append(
        {
            "Model":
                model_name,

            "Gene_level_AUROC":
                roc_auc_score(
                    y_combined,
                    combined_gene[
                        prob_col
                    ]
                ),

            "Gene_level_AP":
                average_precision_score(
                    y_combined,
                    combined_gene[
                        prob_col
                    ]
                )
        }
    )


four_model_gene_summary = pd.DataFrame(
    gene_rows
)


FOUR_MODEL_GENE_SUMMARY = (
    OUT_DIR /
    "FourModel_gene_level_summary.tsv"
)


four_model_gene_summary.to_csv(
    str(
        FOUR_MODEL_GENE_SUMMARY
    ),
    sep="\t",
    index=False
)


# ==================================================================================
# 11. PROVENANCE MANIFEST
# ==================================================================================

manifest = {
    "project":
        "ASD-ContextFusion",

    "stage":
        "Sequence control evaluation",

    "controls": {
        "LengthGC":
            [
                "transcript_length_nt",
                "GC_fraction"
            ],

        "Fourmer":
            "256 oriented normalized 4-mer frequencies"
    },

    "control_model":
        "StandardScaler + L2 LogisticRegression",

    "control_model_parameters":
        LOGISTIC_PARAMS,

    "design_reason":
        (
            "Same linear-probe specification as NT-v2 "
            "to compare sequence representations without "
            "changing classifier capacity."
        ),

    "PU_iterations":
        10,

    "balanced_datasets_per_iteration":
        79,

    "folds_per_balanced_dataset":
        10,

    "total_test_folds":
        7900,

    "same_frozen_partitions_as_primary_context_NTv2":
        True,

    "new_sampling":
        False,

    "hyperparameter_tuning":
        False,

    "result_driven_model_selection":
        False,

    "scaler_fit_scope":
        "training fold only",

    "primary_metric":
        "Average Precision / AUPRC",

    "secondary_metric":
        "AUROC",

    "sequence_feature_manifest":
        str(
            SEQ_MANIFEST
        )
}


with open(
    str(
        OUT_DIR /
        "sequence_control_evaluation_manifest.json"
    ),
    "w"
) as f:

    json.dump(
        manifest,
        f,
        indent=2
    )


# ==================================================================================
# FINAL
# ==================================================================================

print("")
print("=" * 116)
print("SEQUENCE CONTROL EVALUATION COMPLETE")
print("=" * 116)

print("")
print("SPLIT-LEVEL - EXACT SAME 7,900 FROZEN TEST FOLDS")
print(
    split_all_models.to_string(
        index=False
    )
)

print("")
print("GENE-LEVEL AGGREGATED OOF PERFORMANCE")
print(
    four_model_gene_summary.to_string(
        index=False
    )
)

print("")
print("Genes scored :", len(combined_gene))
print(
    "ASD genes    :",
    int(
        (
            combined_gene[
                "known_gene"
            ] == 1
        ).sum()
    )
)

print("")
print("Files:")
print(ALL_METRIC_FILE)
print(ALL_PRED_FILE)
print(GENE_PRED_FILE)
print(COMBINED_METRIC_FILE)
print(COMBINED_GENE_FILE)
print(FOUR_MODEL_SPLIT_SUMMARY)
print(FOUR_MODEL_GENE_SUMMARY)

print("")
print("New PU sampling       : NO")
print("Primary models rerun  : NO")
print("Model tuning          : NO")
print("Same 7,900 folds      : YES")

print("")
print("GO / NO-GO: GO")
print("=" * 116)
'''


RUNNER.write_text(
    runner_code,
    encoding="utf-8"
)


print("\nRunner written:")
print(RUNNER)


# ======================================================================================
# 2. NOTEBOOK-SIDE CHECKPOINT VALIDATOR
# ======================================================================================

def checkpoint_valid(i):

    pred_file = (
        CHECKPOINT_DIR /
        f"iteration_{i:02d}_OOF_predictions.tsv.gz"
    )

    metric_file = (
        CHECKPOINT_DIR /
        f"iteration_{i:02d}_fold_metrics.tsv"
    )


    if not (
        pred_file.exists()
        and
        metric_file.exists()
    ):
        return False


    try:

        m = pd.read_csv(
            metric_file,
            sep="\t"
        )

        p = pd.read_csv(
            pred_file,
            sep="\t",
            compression="gzip"
        )


        if not (
            len(m) == 790
            and
            m[
                "balanced_dataset"
            ].nunique() == 79
            and
            m[
                "fold"
            ].nunique() == 10
        ):
            return False


        unl = p.loc[
            p[
                "known_gene"
            ] == 0
        ]


        if not (
            len(unl) == 18063
            and
            unl[
                "dhindsa_row_id"
            ].nunique() == 18063
        ):
            return False


        return True


    except Exception:

        return False


# ======================================================================================
# 3. CURRENT STATUS
# ======================================================================================

completed_before = [
    i
    for i in range(
        1,
        11
    )
    if checkpoint_valid(i)
]


print("\nCompleted valid sequence-control iterations:")
print(completed_before)


for i in range(
    1,
    11
):

    if checkpoint_valid(i):

        m = pd.read_csv(
            CHECKPOINT_DIR /
            f"iteration_{i:02d}_fold_metrics.tsv",
            sep="\t"
        )

        print(
            f"Iteration {i:02d}: COMPLETE [OK] | "
            f"Length+GC AP={m['LengthGC_AP'].mean():.4f} | "
            f"4-mer AP={m['Fourmer_AP'].mean():.4f}"
        )

    else:

        print(
            f"Iteration {i:02d}: INCOMPLETE"
        )


# ======================================================================================
# 4. MEMORY-SAFE PROCESS ENVIRONMENT
# ======================================================================================

env_base = os.environ.copy()

env_base[
    "PYTHONIOENCODING"
] = "utf-8"

env_base[
    "PYTHONUTF8"
] = "1"


# Engineering/resource control only.
env_base[
    "OMP_NUM_THREADS"
] = "1"

env_base[
    "MKL_NUM_THREADS"
] = "1"

env_base[
    "OPENBLAS_NUM_THREADS"
] = "1"

env_base[
    "NUMEXPR_NUM_THREADS"
] = "1"


# ======================================================================================
# 5. RUN ONLY INCOMPLETE ITERATIONS
# ======================================================================================

with open(
    LOG_FILE,
    "a",
    encoding="utf-8"
) as log:


    for iteration in range(
        1,
        11
    ):


        if checkpoint_valid(
            iteration
        ):

            print(
                f"\nIteration {iteration:02d}: "
                "already COMPLETE -> SKIPPING [OK]"
            )

            continue


        print(
            "\n" +
            "=" * 122
        )

        print(
            f"STARTING SEQUENCE-CONTROL ITERATION "
            f"{iteration:02d} IN FRESH PROCESS"
        )

        print(
            "=" * 122
        )


        env = env_base.copy()

        env[
            "ASD_CF_SEQUENCE_TARGET_ITERATION"
        ] = str(
            iteration
        )


        process = subprocess.Popen(
            [
                str(
                    LEGACY_PYTHON
                ),
                "-u",
                str(
                    RUNNER
                )
            ],

            stdout=subprocess.PIPE,
            stderr=subprocess.STDOUT,

            text=True,
            encoding="utf-8",
            errors="replace",

            bufsize=1,
            env=env
        )


        for line in process.stdout:

            print(
                line,
                end=""
            )

            log.write(
                line
            )

            log.flush()


        return_code = (
            process.wait()
        )


        if return_code != 0:

            print(
                "\n" +
                "=" * 122
            )

            print(
                f"ITERATION {iteration:02d} FAILED"
            )

            print(
                "Existing valid sequence-control checkpoints "
                "remain untouched."
            )

            print(
                "Primary Context + NT results remain untouched."
            )

            print(
                "=" * 122
            )

            raise RuntimeError(
                f"Sequence-control iteration "
                f"{iteration:02d} failed."
            )


        assert checkpoint_valid(
            iteration
        ), (
            f"Iteration {iteration:02d} process ended "
            "without a valid frozen-fold checkpoint."
        )


        m = pd.read_csv(
            CHECKPOINT_DIR /
            f"iteration_{iteration:02d}_fold_metrics.tsv",
            sep="\t"
        )


        print(
            f"\nIteration {iteration:02d}: "
            "VERIFIED [OK]"
        )

        print(
            "Length+GC AUROC:",
            f"{m['LengthGC_AUROC'].mean():.4f}"
        )

        print(
            "Length+GC AP   :",
            f"{m['LengthGC_AP'].mean():.4f}"
        )

        print(
            "4-mer AUROC   :",
            f"{m['Fourmer_AUROC'].mean():.4f}"
        )

        print(
            "4-mer AP      :",
            f"{m['Fourmer_AP'].mean():.4f}"
        )

        print(
            "Fresh process closed -> RAM released."
        )


# ======================================================================================
# 6. VERIFY 10 / 10
# ======================================================================================

completed_after = [
    i
    for i in range(
        1,
        11
    )
    if checkpoint_valid(i)
]


print(
    "\n" +
    "=" * 122
)

print(
    "SEQUENCE-CONTROL ITERATION CHECK"
)

print(
    "=" * 122
)


print(
    "Completed:",
    completed_after
)


assert completed_after == list(
    range(
        1,
        11
    )
), (
    "Not all 10 iterations are complete. "
    "Aggregation will NOT run."
)


print(
    "\nALL 10 SEQUENCE-CONTROL ITERATIONS VERIFIED [OK]"
)


# ======================================================================================
# 7. AGGREGATION ONLY
# ======================================================================================

print(
    "\n" +
    "=" * 122
)

print(
    "RUNNING FINAL SEQUENCE-CONTROL AGGREGATION"
)

print(
    "=" * 122
)


final_env = env_base.copy()

final_env.pop(
    "ASD_CF_SEQUENCE_TARGET_ITERATION",
    None
)


process = subprocess.Popen(
    [
        str(
            LEGACY_PYTHON
        ),
        "-u",
        str(
            RUNNER
        )
    ],

    stdout=subprocess.PIPE,
    stderr=subprocess.STDOUT,

    text=True,
    encoding="utf-8",
    errors="replace",

    bufsize=1,
    env=final_env
)


with open(
    LOG_FILE,
    "a",
    encoding="utf-8"
) as log:


    for line in process.stdout:

        print(
            line,
            end=""
        )

        log.write(
            line
        )

        log.flush()


return_code = (
    process.wait()
)


if return_code != 0:

    raise RuntimeError(
        "Final sequence-control aggregation failed. "
        "All 10 iteration checkpoints remain safe."
    )


# ======================================================================================
# 8. FINAL NOTEBOOK-SIDE RESULT AUDIT
# ======================================================================================

FINAL_SPLIT = (
    RESULT_DIR /
    "FourModel_split_level_summary.tsv"
)

FINAL_GENE = (
    RESULT_DIR /
    "FourModel_gene_level_summary.tsv"
)

FINAL_METRICS = (
    RESULT_DIR /
    "Context_NTv2_LengthGC_4mer_all_7900_fold_metrics.tsv"
)

FINAL_GENE_SCORES = (
    RESULT_DIR /
    "Context_NTv2_LengthGC_4mer_gene_level_OOF_scores.tsv"
)


for p in [
    FINAL_SPLIT,
    FINAL_GENE,
    FINAL_METRICS,
    FINAL_GENE_SCORES
]:

    assert p.exists(), (
        f"Missing final output:\n{p}"
    )


metrics_final = pd.read_csv(
    FINAL_METRICS,
    sep="\t"
)

genes_final = pd.read_csv(
    FINAL_GENE_SCORES,
    sep="\t"
)


assert len(
    metrics_final
) == 7900

assert len(
    genes_final
) == 18252

assert (
    genes_final[
        "Gene_Name"
    ].nunique()
    == 18252
)

assert (
    genes_final.loc[
        genes_final[
            "known_gene"
        ] == 1,
        "Gene_Name"
    ].nunique()
    == 190
)


print(
    "\n" +
    "=" * 122
)

print(
    "SEQUENCE CONTROLS + FOUR-MODEL COMPARISON COMPLETE [OK]"
)

print(
    "=" * 122
)


print(
    "\nSPLIT-LEVEL SUMMARY:"
)

print(
    pd.read_csv(
        FINAL_SPLIT,
        sep="\t"
    ).to_string(
        index=False
    )
)


print(
    "\nGENE-LEVEL SUMMARY:"
)

print(
    pd.read_csv(
        FINAL_GENE,
        sep="\t"
    ).to_string(
        index=False
    )
)


print(
    "\nFrozen test folds : 7900 [OK]"
)

print(
    "Genes scored      : 18252 [OK]"
)

print(
    "ASD genes         : 190 [OK]"
)

print(
    "New PU sampling   : NO"
)

print(
    "Primary rerun     : NO"
)

print(
    "Hyperparam tuning : NO"
)

print(
    "\nNEXT SCIENTIFIC STAGE:"
)

print(
    "Cross-fitted modality-preserving logistic late fusion."
)

print(
    "=" * 122
)
```

```python
# ======================================================================================
# ASD-ContextFusion
# STRICT CROSS-FITTED MODALITY-PRESERVING LATE LOGISTIC STACKING
#
# PROPOSED FUSION:
#   Dhindsa Context XGBoost probability
#   +
#   NT-v2 L2 logistic probability
#   ->
#   Logistic meta-learner
#
# IMPORTANT
# ---------
# Length+GC and 4-mer are CONTROLS / COMPARATORS only.
# They are NOT included in the proposed fusion.
#
# STRICT NESTED CROSS-FITTING
# ---------------------------
# For each frozen outer fold f:
#
#   1. Outer-test Context/NT predictions:
#        reuse the already-saved primary OOF predictions.
#        These were trained without outer fold f.
#
#   2. Meta-training predictions:
#        MUST NOT use ordinary saved OOF predictions from the other folds,
#        because those base models may have been trained using outer fold f.
#
#      Instead:
#        for every inner validation fold g != f,
#        train Context + NT models excluding BOTH f and g,
#        predict g,
#        collect these leakage-free inner OOF predictions.
#
#   3. Fit logistic meta-learner on those inner OOF predictions.
#
#   4. Apply meta-learner to the saved outer-test Context + NT OOF predictions.
#
# COMPUTATIONAL OPTIMIZATION WITHOUT METHODOLOGICAL CHANGE
# --------------------------------------------------------
# For a pair of excluded folds {a,b}, the same base model can:
#   - predict fold a for outer=b
#   - predict fold b for outer=a
#
# Therefore only C(10,2)=45 inner base fits per balanced dataset are needed,
# instead of 90 ordered fits.
#
# ENGINEERING CHECKPOINTING
# -------------------------
# One balanced dataset = one fresh Python process.
# There are:
#   10 iterations x 79 balanced datasets = 790 checkpointable tasks.
#
# Completed tasks are automatically skipped after restart/crash.
#
# NO:
# - new PU sampling
# - partition regeneration
# - NT regeneration
# - primary model modification
# - result-driven tuning
# - test-fold-driven selection
# - raw feature concatenation
# ======================================================================================

from pathlib import Path
import subprocess
import os
import pandas as pd
import textwrap


# ======================================================================================
# 0. PATHS
# ======================================================================================

ROOT = Path(r"E:\ASD_ContextFusion_Q1")

REPO = (
    ROOT /
    "01_raw" /
    "dhindsa_2025" /
    "Mantis-ml-NDD"
)

LEGACY_PYTHON = (
    ROOT /
    "envs" /
    "mantis37" /
    "python.exe"
)

RUNNER = (
    ROOT /
    "00_admin" /
    "run_strict_crossfitted_context_NTv2_fusion.py"
)

CHECKPOINT_DIR = (
    ROOT /
    "04_models" /
    "checkpoints" /
    "strict_context_NTv2_fusion"
)

RESULT_DIR = (
    ROOT /
    "05_results" /
    "proposed" /
    "strict_context_NTv2_fusion"
)

LOG_FILE = (
    ROOT /
    "08_logs" /
    "strict_context_NTv2_fusion.log"
)


CHECKPOINT_DIR.mkdir(
    parents=True,
    exist_ok=True
)

RESULT_DIR.mkdir(
    parents=True,
    exist_ok=True
)

LOG_FILE.parent.mkdir(
    parents=True,
    exist_ok=True
)


for p in [
    REPO,
    LEGACY_PYTHON
]:
    assert p.exists(), p


print("=" * 124)
print("ASD-CONTEXTFUSION - STRICT CROSS-FITTED CONTEXT + NT-v2 LATE FUSION")
print("=" * 124)


# ======================================================================================
# 1. WRITE STRICT NESTED-FUSION RUNNER
# ======================================================================================

runner_code = r'''
from __future__ import print_function

from pathlib import Path
import os
import sys
import json
import time
import gc

import numpy as np
import pandas as pd

import xgboost as xgb

from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    roc_auc_score,
    average_precision_score
)


# ==================================================================================
# PATHS
# ==================================================================================

ROOT = Path(r"E:\ASD_ContextFusion_Q1")


MASTER_FILE = (
    ROOT /
    "03_features" /
    "multimodal_master" /
    "dhindsa_NTv2_row_preserved_master.tsv"
)

PREDICTOR_FILE = (
    ROOT /
    "03_features" /
    "multimodal_master" /
    "dhindsa_316_context_predictors.tsv"
)

NT_FILE = (
    ROOT /
    "03_features" /
    "nt_v2" /
    "NTv2_50m_full_transcript_embeddings_18252x512.npy"
)

PARTITION_DIR = (
    ROOT /
    "04_models" /
    "checkpoints" /
    "dhindsa_PU_common_universe"
)

PRIMARY_CHECKPOINT_DIR = (
    ROOT /
    "04_models" /
    "checkpoints" /
    "primary_context_NTv2"
)

CHECKPOINT_DIR = (
    ROOT /
    "04_models" /
    "checkpoints" /
    "strict_context_NTv2_fusion"
)

OUT_DIR = (
    ROOT /
    "05_results" /
    "proposed" /
    "strict_context_NTv2_fusion"
)

PRIMARY_RESULT_DIR = (
    ROOT /
    "05_results" /
    "proposed" /
    "primary_context_NTv2"
)

CONTROL_RESULT_DIR = (
    ROOT /
    "05_results" /
    "proposed" /
    "sequence_controls"
)


CHECKPOINT_DIR.mkdir(
    parents=True,
    exist_ok=True
)

OUT_DIR.mkdir(
    parents=True,
    exist_ok=True
)


# ==================================================================================
# 1. EXACT BASE-MODEL SPECIFICATIONS
# ==================================================================================

# Exact released Mantis/Dhindsa Context XGBoost parameters.
XGB_PARAMS = {
    "learning_rate": 0.01,
    "n_estimators": 300,
    "max_depth": 5,
    "min_child_weight": 3,
    "gamma": 0,
    "subsample": 0.8,
    "colsample_bytree": 0.8,
    "objective": "binary:logistic",
    "nthread": 1,
    "scale_pos_weight": 1
}


# Exact already-locked NT-v2 linear probe.
NT_LOGISTIC_PARAMS = {
    "penalty": "l2",
    "C": 1.0,
    "solver": "lbfgs",
    "max_iter": 2000
}


# Fixed logistic meta learner.
#
# No tuning.
# No scaler is used here because both inputs are already probabilities
# on the common [0,1] scale.
META_LOGISTIC_PARAMS = {
    "penalty": "l2",
    "C": 1.0,
    "solver": "lbfgs",
    "max_iter": 2000
}


# ==================================================================================
# 2. LOAD LOCKED DATA
# ==================================================================================

for p in [
    MASTER_FILE,
    PREDICTOR_FILE,
    NT_FILE
]:
    assert p.exists(), "Missing:\n%s" % str(p)


master = pd.read_csv(
    str(MASTER_FILE),
    sep="\t",
    low_memory=False
)


predictor_df = pd.read_csv(
    str(PREDICTOR_FILE),
    sep="\t"
)


context_features = (
    predictor_df[
        "feature"
    ]
    .astype(str)
    .tolist()
)


assert len(master) == 18253
assert master["Gene_Name"].nunique() == 18252
assert len(context_features) == 316

assert all(
    x in master.columns
    for x in context_features
)


master["dhindsa_row_id"] = (
    master[
        "dhindsa_row_id"
    ]
    .astype(int)
)

master["NT_row_index"] = (
    master[
        "NT_row_index"
    ]
    .astype(int)
)

master["known_gene"] = (
    master[
        "known_gene"
    ]
    .astype(int)
)


assert master[
    "dhindsa_row_id"
].is_unique


master_idx = master.set_index(
    "dhindsa_row_id",
    drop=False
)


X_nt = np.load(
    str(NT_FILE),
    mmap_mode="r"
)


assert X_nt.shape == (
    18252,
    512
)


# ==================================================================================
# 3. CHECKPOINT HELPERS
# ==================================================================================

def prediction_file(iteration, balanced_dataset):

    return (
        CHECKPOINT_DIR /
        (
            "iteration_%02d_balanced_%02d_fusion_predictions.tsv.gz"
            % (
                iteration,
                balanced_dataset
            )
        )
    )


def metric_file(iteration, balanced_dataset):

    return (
        CHECKPOINT_DIR /
        (
            "iteration_%02d_balanced_%02d_fusion_metrics.tsv"
            % (
                iteration,
                balanced_dataset
            )
        )
    )


def coefficient_file(iteration, balanced_dataset):

    return (
        CHECKPOINT_DIR /
        (
            "iteration_%02d_balanced_%02d_meta_coefficients.tsv"
            % (
                iteration,
                balanced_dataset
            )
        )
    )


def task_valid(iteration, balanced_dataset):

    pf = prediction_file(
        iteration,
        balanced_dataset
    )

    mf = metric_file(
        iteration,
        balanced_dataset
    )

    cf = coefficient_file(
        iteration,
        balanced_dataset
    )


    if not (
        pf.exists()
        and
        mf.exists()
        and
        cf.exists()
    ):
        return False


    try:

        p = pd.read_csv(
            str(pf),
            sep="\t",
            compression="gzip"
        )

        m = pd.read_csv(
            str(mf),
            sep="\t"
        )

        c = pd.read_csv(
            str(cf),
            sep="\t"
        )


        if len(m) != 10:
            return False

        if m["fold"].nunique() != 10:
            return False

        if len(c) != 10:
            return False

        if c["fold"].nunique() != 10:
            return False

        if p[
            "dhindsa_row_id"
        ].nunique() != len(p):
            return False

        if not {
            "Fusion_probability",
            "Context_probability",
            "NT_probability"
        }.issubset(
            set(p.columns)
        ):
            return False


        membership_file = (
            PARTITION_DIR /
            (
                "iteration_%02d_test_fold_membership.tsv.gz"
                % iteration
            )
        )


        membership = pd.read_csv(
            str(membership_file),
            sep="\t",
            compression="gzip"
        )


        target = membership.loc[
            membership[
                "balanced_dataset"
            ] == balanced_dataset
        ]


        expected_ids = set(
            target[
                "dhindsa_row_id"
            ]
            .astype(int)
        )

        observed_ids = set(
            p[
                "dhindsa_row_id"
            ]
            .astype(int)
        )


        if observed_ids != expected_ids:
            return False


        return True


    except Exception:

        return False


# ==================================================================================
# 4. SINGLE BALANCED-DATASET MODE
# ==================================================================================

TARGET_ITER = os.environ.get(
    "ASD_CF_FUSION_ITERATION"
)

TARGET_BAL = os.environ.get(
    "ASD_CF_FUSION_BALANCED_DATASET"
)


if (
    TARGET_ITER is not None
    and
    TARGET_BAL is not None
):

    iteration = int(
        TARGET_ITER
    )

    balanced_dataset = int(
        TARGET_BAL
    )


    assert 1 <= iteration <= 10
    assert 0 <= balanced_dataset <= 78


    if task_valid(
        iteration,
        balanced_dataset
    ):

        print(
            "Iteration %02d / balanced %02d already complete - SKIPPING"
            % (
                iteration,
                balanced_dataset
            )
        )

        sys.exit(0)


    # ------------------------------------------------------------------------------
    # Membership
    # ------------------------------------------------------------------------------

    membership_file = (
        PARTITION_DIR /
        (
            "iteration_%02d_test_fold_membership.tsv.gz"
            % iteration
        )
    )


    assert membership_file.exists()


    membership = pd.read_csv(
        str(membership_file),
        sep="\t",
        compression="gzip"
    )


    bal = (
        membership.loc[
            membership[
                "balanced_dataset"
            ] == balanced_dataset
        ]
        .copy()
        .sort_values(
            "dhindsa_row_id"
        )
        .reset_index(
            drop=True
        )
    )


    assert 380 <= len(bal) <= 383
    assert bal["fold"].nunique() == 10
    assert bal["dhindsa_row_id"].is_unique


    row_ids = (
        bal[
            "dhindsa_row_id"
        ]
        .astype(int)
        .to_numpy()
    )

    fold_id = (
        bal[
            "fold"
        ]
        .astype(int)
        .to_numpy()
    )


    # ------------------------------------------------------------------------------
    # TMSB15B duplicate leakage protection
    # ------------------------------------------------------------------------------

    tmsb = bal.loc[
        bal[
            "Gene_Name"
        ] == "TMSB15B"
    ]


    if len(tmsb) == 2:

        assert (
            tmsb[
                "fold"
            ].nunique()
            == 1
        ), (
            "TMSB15B duplicate rows crossed folds."
        )


    # ------------------------------------------------------------------------------
    # Aligned master arrays
    # ------------------------------------------------------------------------------

    bal_master = master_idx.loc[
        row_ids
    ].copy()


    assert np.array_equal(
        bal_master[
            "dhindsa_row_id"
        ].to_numpy(
            dtype=np.int64
        ),
        row_ids
    )


    y = (
        bal_master[
            "known_gene"
        ]
        .to_numpy(
            dtype=np.int32
        )
    )


    X_context = (
        bal_master[
            context_features
        ]
        .to_numpy(
            dtype=np.float32
        )
    )


    nt_row_index = (
        bal_master[
            "NT_row_index"
        ]
        .to_numpy(
            dtype=np.int64
        )
    )


    X_nt_bal = np.asarray(
        X_nt[
            nt_row_index
        ],
        dtype=np.float32
    )


    assert X_context.shape == (
        len(bal),
        316
    )

    assert X_nt_bal.shape == (
        len(bal),
        512
    )


    # ------------------------------------------------------------------------------
    # Already-saved OUTER OOF Context / NT probabilities
    # ------------------------------------------------------------------------------

    primary_pred_file = (
        PRIMARY_CHECKPOINT_DIR /
        (
            "iteration_%02d_OOF_predictions.tsv.gz"
            % iteration
        )
    )


    assert primary_pred_file.exists()


    primary = pd.read_csv(
        str(primary_pred_file),
        sep="\t",
        compression="gzip"
    )


    primary_bal = (
        primary.loc[
            primary[
                "balanced_dataset"
            ] == balanced_dataset
        ]
        .copy()
        .sort_values(
            "dhindsa_row_id"
        )
        .reset_index(
            drop=True
        )
    )


    assert len(primary_bal) == len(bal)


    assert np.array_equal(
        primary_bal[
            "dhindsa_row_id"
        ].to_numpy(
            dtype=np.int64
        ),
        row_ids
    )


    assert np.array_equal(
        primary_bal[
            "fold"
        ].to_numpy(
            dtype=np.int32
        ),
        fold_id
    )


    assert np.array_equal(
        primary_bal[
            "known_gene"
        ].to_numpy(
            dtype=np.int32
        ),
        y
    )


    outer_context_prob = (
        primary_bal[
            "Context_probability"
        ]
        .to_numpy(
            dtype=np.float64
        )
    )

    outer_nt_prob = (
        primary_bal[
            "NT_probability"
        ]
        .to_numpy(
            dtype=np.float64
        )
    )


    assert np.isfinite(
        outer_context_prob
    ).all()

    assert np.isfinite(
        outer_nt_prob
    ).all()


    # ==================================================================================
    # 5. STRICT INNER CROSS-FITTING
    #
    # inner_context[outer_fold, row_position]
    # inner_nt[outer_fold, row_position]
    #
    # For outer fold f, values are filled ONLY for rows where fold != f.
    # Every such prediction was produced by a model trained without:
    #   - outer fold f
    #   - that row's own validation fold g
    # ==================================================================================

    n_rows = len(
        bal
    )


    inner_context = np.full(
        (
            10,
            n_rows
        ),
        np.nan,
        dtype=np.float64
    )

    inner_nt = np.full(
        (
            10,
            n_rows
        ),
        np.nan,
        dtype=np.float64
    )


    start = time.time()

    pair_counter = 0


    # Only 45 unordered excluded-fold pairs.
    for fold_a in range(10):

        for fold_b in range(
            fold_a + 1,
            10
        ):

            pair_counter += 1


            train_mask = (
                (fold_id != fold_a)
                &
                (fold_id != fold_b)
            )

            val_a_mask = (
                fold_id == fold_a
            )

            val_b_mask = (
                fold_id == fold_b
            )


            y_inner_train = y[
                train_mask
            ]


            assert set(
                np.unique(
                    y_inner_train
                )
            ) == {
                0,
                1
            }


            # ----------------------------------------------------------------------
            # A. CONTEXT XGBOOST
            # ----------------------------------------------------------------------

            context_model = (
                xgb.XGBClassifier(
                    **XGB_PARAMS
                )
            )


            context_model.fit(
                X_context[
                    train_mask
                ],
                y_inner_train
            )


            context_prob_a = (
                context_model
                .predict_proba(
                    X_context[
                        val_a_mask
                    ]
                )[:, 1]
            )


            context_prob_b = (
                context_model
                .predict_proba(
                    X_context[
                        val_b_mask
                    ]
                )[:, 1]
            )


            # Model excludes a & b.
            #
            # Predict a -> usable as inner validation when OUTER=b
            inner_context[
                fold_b,
                val_a_mask
            ] = context_prob_a

            # Predict b -> usable as inner validation when OUTER=a
            inner_context[
                fold_a,
                val_b_mask
            ] = context_prob_b


            del context_model


            # ----------------------------------------------------------------------
            # B. NT-v2 LINEAR PROBE
            # ----------------------------------------------------------------------

            scaler_nt = StandardScaler()


            Xn_train_scaled = (
                scaler_nt.fit_transform(
                    X_nt_bal[
                        train_mask
                    ]
                )
            )


            Xn_a_scaled = (
                scaler_nt.transform(
                    X_nt_bal[
                        val_a_mask
                    ]
                )
            )


            Xn_b_scaled = (
                scaler_nt.transform(
                    X_nt_bal[
                        val_b_mask
                    ]
                )
            )


            nt_model = (
                LogisticRegression(
                    **NT_LOGISTIC_PARAMS
                )
            )


            nt_model.fit(
                Xn_train_scaled,
                y_inner_train
            )


            nt_prob_a = (
                nt_model
                .predict_proba(
                    Xn_a_scaled
                )[:, 1]
            )

            nt_prob_b = (
                nt_model
                .predict_proba(
                    Xn_b_scaled
                )[:, 1]
            )


            inner_nt[
                fold_b,
                val_a_mask
            ] = nt_prob_a

            inner_nt[
                fold_a,
                val_b_mask
            ] = nt_prob_b


            del nt_model
            del scaler_nt

            del Xn_train_scaled
            del Xn_a_scaled
            del Xn_b_scaled

            gc.collect()


            if (
                pair_counter % 5
                == 0
                or
                pair_counter == 45
            ):

                elapsed = (
                    time.time()
                    -
                    start
                )

                print(
                    "Iteration %02d | balanced %02d | "
                    "inner pairs %02d / 45 | %.1f min"
                    % (
                        iteration,
                        balanced_dataset,
                        pair_counter,
                        elapsed / 60.0
                    )
                )


    assert pair_counter == 45


    # ==================================================================================
    # 6. OUTER-FOLD META LEARNER
    # ==================================================================================

    prediction_rows = []
    metric_rows = []
    coefficient_rows = []


    for outer_fold in range(10):

        meta_train_mask = (
            fold_id != outer_fold
        )

        outer_test_mask = (
            fold_id == outer_fold
        )


        # Strict leakage check:
        # every meta-training row must have both inner OOF predictions.
        assert np.isfinite(
            inner_context[
                outer_fold,
                meta_train_mask
            ]
        ).all()


        assert np.isfinite(
            inner_nt[
                outer_fold,
                meta_train_mask
            ]
        ).all()


        # Outer test positions should NEVER have inner meta-training predictions.
        assert np.isnan(
            inner_context[
                outer_fold,
                outer_test_mask
            ]
        ).all()


        assert np.isnan(
            inner_nt[
                outer_fold,
                outer_test_mask
            ]
        ).all()


        X_meta_train = np.column_stack(
            [
                inner_context[
                    outer_fold,
                    meta_train_mask
                ],

                inner_nt[
                    outer_fold,
                    meta_train_mask
                ]
            ]
        )


        y_meta_train = y[
            meta_train_mask
        ]


        assert X_meta_train.shape[1] == 2

        assert set(
            np.unique(
                y_meta_train
            )
        ) == {
            0,
            1
        }


        # No scaler:
        # both meta features are probabilities in [0,1].
        meta_model = (
            LogisticRegression(
                **META_LOGISTIC_PARAMS
            )
        )


        meta_model.fit(
            X_meta_train,
            y_meta_train
        )


        X_outer_test_meta = np.column_stack(
            [
                outer_context_prob[
                    outer_test_mask
                ],

                outer_nt_prob[
                    outer_test_mask
                ]
            ]
        )


        y_outer_test = y[
            outer_test_mask
        ]


        fusion_prob = (
            meta_model
            .predict_proba(
                X_outer_test_meta
            )[:, 1]
        )


        assert np.isfinite(
            fusion_prob
        ).all()


        fusion_roc = roc_auc_score(
            y_outer_test,
            fusion_prob
        )

        fusion_ap = average_precision_score(
            y_outer_test,
            fusion_prob
        )


        context_roc = roc_auc_score(
            y_outer_test,
            outer_context_prob[
                outer_test_mask
            ]
        )

        context_ap = average_precision_score(
            y_outer_test,
            outer_context_prob[
                outer_test_mask
            ]
        )


        nt_roc = roc_auc_score(
            y_outer_test,
            outer_nt_prob[
                outer_test_mask
            ]
        )

        nt_ap = average_precision_score(
            y_outer_test,
            outer_nt_prob[
                outer_test_mask
            ]
        )


        metric_rows.append(
            {
                "iteration":
                    iteration,

                "balanced_dataset":
                    balanced_dataset,

                "fold":
                    outer_fold,

                "n_meta_train":
                    int(
                        meta_train_mask.sum()
                    ),

                "n_outer_test":
                    int(
                        outer_test_mask.sum()
                    ),

                "test_positive":
                    int(
                        y_outer_test.sum()
                    ),

                "test_unlabelled":
                    int(
                        len(y_outer_test)
                        -
                        y_outer_test.sum()
                    ),

                "Context_AUROC":
                    float(
                        context_roc
                    ),

                "Context_AP":
                    float(
                        context_ap
                    ),

                "NT_AUROC":
                    float(
                        nt_roc
                    ),

                "NT_AP":
                    float(
                        nt_ap
                    ),

                "Fusion_AUROC":
                    float(
                        fusion_roc
                    ),

                "Fusion_AP":
                    float(
                        fusion_ap
                    ),

                "Delta_AUROC_Fusion_minus_Context":
                    float(
                        fusion_roc
                        -
                        context_roc
                    ),

                "Delta_AP_Fusion_minus_Context":
                    float(
                        fusion_ap
                        -
                        context_ap
                    ),

                "Delta_AUROC_Fusion_minus_NT":
                    float(
                        fusion_roc
                        -
                        nt_roc
                    ),

                "Delta_AP_Fusion_minus_NT":
                    float(
                        fusion_ap
                        -
                        nt_ap
                    )
            }
        )


        coefficient_rows.append(
            {
                "iteration":
                    iteration,

                "balanced_dataset":
                    balanced_dataset,

                "fold":
                    outer_fold,

                "Context_coefficient":
                    float(
                        meta_model.coef_[
                            0,
                            0
                        ]
                    ),

                "NT_coefficient":
                    float(
                        meta_model.coef_[
                            0,
                            1
                        ]
                    ),

                "intercept":
                    float(
                        meta_model.intercept_[
                            0
                        ]
                    )
            }
        )


        test_positions = np.where(
            outer_test_mask
        )[0]


        for local_j, pos in enumerate(
            test_positions
        ):

            prediction_rows.append(
                {
                    "iteration":
                        iteration,

                    "balanced_dataset":
                        balanced_dataset,

                    "fold":
                        outer_fold,

                    "dhindsa_row_id":
                        int(
                            row_ids[
                                pos
                            ]
                        ),

                    "Gene_Name":
                        str(
                            bal_master.iloc[
                                pos
                            ][
                                "Gene_Name"
                            ]
                        ),

                    "known_gene":
                        int(
                            y[
                                pos
                            ]
                        ),

                    "NT_row_index":
                        int(
                            nt_row_index[
                                pos
                            ]
                        ),

                    "Context_probability":
                        float(
                            outer_context_prob[
                                pos
                            ]
                        ),

                    "NT_probability":
                        float(
                            outer_nt_prob[
                                pos
                            ]
                        ),

                    "Fusion_probability":
                        float(
                            fusion_prob[
                                local_j
                            ]
                        )
                }
            )


        del meta_model
        gc.collect()


    # ==================================================================================
    # 7. SAVE BALANCED-DATASET CHECKPOINT
    # ==================================================================================

    pred_df = pd.DataFrame(
        prediction_rows
    )

    metric_df = pd.DataFrame(
        metric_rows
    )

    coef_df = pd.DataFrame(
        coefficient_rows
    )


    assert len(
        pred_df
    ) == len(
        bal
    )

    assert pred_df[
        "dhindsa_row_id"
    ].nunique() == len(
        bal
    )

    assert len(
        metric_df
    ) == 10

    assert metric_df[
        "fold"
    ].nunique() == 10

    assert len(
        coef_df
    ) == 10


    pf = prediction_file(
        iteration,
        balanced_dataset
    )

    mf = metric_file(
        iteration,
        balanced_dataset
    )

    cf = coefficient_file(
        iteration,
        balanced_dataset
    )


    pred_df.to_csv(
        str(pf),
        sep="\t",
        index=False,
        compression="gzip"
    )

    metric_df.to_csv(
        str(mf),
        sep="\t",
        index=False
    )

    coef_df.to_csv(
        str(cf),
        sep="\t",
        index=False
    )


    assert task_valid(
        iteration,
        balanced_dataset
    )


    print("")
    print("=" * 116)

    print(
        "ITERATION %02d / BALANCED %02d COMPLETE"
        % (
            iteration,
            balanced_dataset
        )
    )

    print("=" * 116)

    print(
        "Context AP : %.4f"
        % metric_df[
            "Context_AP"
        ].mean()
    )

    print(
        "NT AP      : %.4f"
        % metric_df[
            "NT_AP"
        ].mean()
    )

    print(
        "Fusion AP  : %.4f"
        % metric_df[
            "Fusion_AP"
        ].mean()
    )

    print(
        "Fusion AUROC: %.4f"
        % metric_df[
            "Fusion_AUROC"
        ].mean()
    )

    print("")
    print(
        "Checkpoint:",
        pf
    )

    print(
        "Fresh process can now exit."
    )

    print("=" * 116)

    sys.exit(0)


# ==================================================================================
# 8. AGGREGATION-ONLY MODE
# ==================================================================================

print("=" * 116)
print("STRICT FUSION AGGREGATION-ONLY MODE")
print("=" * 116)


all_pred_files = []
all_metric_files = []
all_coef_files = []


for iteration in range(
    1,
    11
):

    for balanced_dataset in range(
        79
    ):

        assert task_valid(
            iteration,
            balanced_dataset
        ), (
            "Missing/invalid task: iteration %02d balanced %02d"
            % (
                iteration,
                balanced_dataset
            )
        )


        all_pred_files.append(
            prediction_file(
                iteration,
                balanced_dataset
            )
        )

        all_metric_files.append(
            metric_file(
                iteration,
                balanced_dataset
            )
        )

        all_coef_files.append(
            coefficient_file(
                iteration,
                balanced_dataset
            )
        )


assert len(
    all_metric_files
) == 790


# ==================================================================================
# 9. COMBINE ALL OUTER FOLDS
# ==================================================================================

all_metrics = pd.concat(
    [
        pd.read_csv(
            str(p),
            sep="\t"
        )
        for p in all_metric_files
    ],
    ignore_index=True
)


all_predictions = pd.concat(
    [
        pd.read_csv(
            str(p),
            sep="\t",
            compression="gzip"
        )
        for p in all_pred_files
    ],
    ignore_index=True
)


all_coefficients = pd.concat(
    [
        pd.read_csv(
            str(p),
            sep="\t"
        )
        for p in all_coef_files
    ],
    ignore_index=True
)


assert len(
    all_metrics
) == 7900

assert len(
    all_coefficients
) == 7900


# Unlabelled row must be tested once per stochastic iteration.
for iteration in range(
    1,
    11
):

    tmp = all_predictions.loc[
        (
            all_predictions[
                "iteration"
            ] == iteration
        )
        &
        (
            all_predictions[
                "known_gene"
            ] == 0
        )
    ]


    assert len(
        tmp
    ) == 18063

    assert tmp[
        "dhindsa_row_id"
    ].nunique() == 18063


ALL_METRIC_FILE = (
    OUT_DIR /
    "StrictFusion_all_7900_fold_metrics.tsv"
)

ALL_PRED_FILE = (
    OUT_DIR /
    "StrictFusion_all_OOF_predictions.tsv.gz"
)

ALL_COEF_FILE = (
    OUT_DIR /
    "StrictFusion_all_7900_meta_coefficients.tsv"
)


all_metrics.to_csv(
    str(
        ALL_METRIC_FILE
    ),
    sep="\t",
    index=False
)


all_predictions.to_csv(
    str(
        ALL_PRED_FILE
    ),
    sep="\t",
    index=False,
    compression="gzip"
)


all_coefficients.to_csv(
    str(
        ALL_COEF_FILE
    ),
    sep="\t",
    index=False
)


# ==================================================================================
# 10. SPLIT-LEVEL SUMMARY
# ==================================================================================

split_summary = pd.DataFrame(
    [
        {
            "Model":
                "Dhindsa_Context_XGBoost",

            "AUROC_mean":
                all_metrics[
                    "Context_AUROC"
                ].mean(),

            "AUROC_sd":
                all_metrics[
                    "Context_AUROC"
                ].std(),

            "AP_mean":
                all_metrics[
                    "Context_AP"
                ].mean(),

            "AP_sd":
                all_metrics[
                    "Context_AP"
                ].std()
        },

        {
            "Model":
                "NTv2_L2_Logistic",

            "AUROC_mean":
                all_metrics[
                    "NT_AUROC"
                ].mean(),

            "AUROC_sd":
                all_metrics[
                    "NT_AUROC"
                ].std(),

            "AP_mean":
                all_metrics[
                    "NT_AP"
                ].mean(),

            "AP_sd":
                all_metrics[
                    "NT_AP"
                ].std()
        },

        {
            "Model":
                "Strict_Context_NTv2_LateFusion",

            "AUROC_mean":
                all_metrics[
                    "Fusion_AUROC"
                ].mean(),

            "AUROC_sd":
                all_metrics[
                    "Fusion_AUROC"
                ].std(),

            "AP_mean":
                all_metrics[
                    "Fusion_AP"
                ].mean(),

            "AP_sd":
                all_metrics[
                    "Fusion_AP"
                ].std()
        }
    ]
)


SPLIT_SUMMARY_FILE = (
    OUT_DIR /
    "StrictFusion_split_level_summary.tsv"
)


split_summary.to_csv(
    str(
        SPLIT_SUMMARY_FILE
    ),
    sep="\t",
    index=False
)


# ==================================================================================
# 11. GENE-LEVEL AGGREGATION
# ==================================================================================

gene_predictions = (
    all_predictions
    .groupby(
        [
            "Gene_Name",
            "known_gene"
        ],
        as_index=False
    )
    .agg(
        Context_probability=(
            "Context_probability",
            "mean"
        ),

        NT_probability=(
            "NT_probability",
            "mean"
        ),

        Fusion_probability=(
            "Fusion_probability",
            "mean"
        ),

        OOF_prediction_count=(
            "Fusion_probability",
            "size"
        )
    )
)


assert gene_predictions[
    "Gene_Name"
].nunique() == 18252


assert gene_predictions.loc[
    gene_predictions[
        "known_gene"
    ] == 1,
    "Gene_Name"
].nunique() == 190


y_gene = (
    gene_predictions[
        "known_gene"
    ]
    .to_numpy(
        dtype=np.int32
    )
)


gene_summary = pd.DataFrame(
    [
        {
            "Model":
                "Dhindsa_Context_XGBoost",

            "Gene_level_AUROC":
                roc_auc_score(
                    y_gene,
                    gene_predictions[
                        "Context_probability"
                    ]
                ),

            "Gene_level_AP":
                average_precision_score(
                    y_gene,
                    gene_predictions[
                        "Context_probability"
                    ]
                )
        },

        {
            "Model":
                "NTv2_L2_Logistic",

            "Gene_level_AUROC":
                roc_auc_score(
                    y_gene,
                    gene_predictions[
                        "NT_probability"
                    ]
                ),

            "Gene_level_AP":
                average_precision_score(
                    y_gene,
                    gene_predictions[
                        "NT_probability"
                    ]
                )
        },

        {
            "Model":
                "Strict_Context_NTv2_LateFusion",

            "Gene_level_AUROC":
                roc_auc_score(
                    y_gene,
                    gene_predictions[
                        "Fusion_probability"
                    ]
                ),

            "Gene_level_AP":
                average_precision_score(
                    y_gene,
                    gene_predictions[
                        "Fusion_probability"
                    ]
                )
        }
    ]
)


GENE_PRED_FILE = (
    OUT_DIR /
    "StrictFusion_gene_level_OOF_scores.tsv"
)

GENE_SUMMARY_FILE = (
    OUT_DIR /
    "StrictFusion_gene_level_summary.tsv"
)


gene_predictions.to_csv(
    str(
        GENE_PRED_FILE
    ),
    sep="\t",
    index=False
)


gene_summary.to_csv(
    str(
        GENE_SUMMARY_FILE
    ),
    sep="\t",
    index=False
)


# ==================================================================================
# 12. META-COEFFICIENT SUMMARY
# ==================================================================================

coefficient_summary = pd.DataFrame(
    [
        {
            "Parameter":
                "Context_coefficient",

            "Mean":
                all_coefficients[
                    "Context_coefficient"
                ].mean(),

            "SD":
                all_coefficients[
                    "Context_coefficient"
                ].std(),

            "Median":
                all_coefficients[
                    "Context_coefficient"
                ].median()
        },

        {
            "Parameter":
                "NT_coefficient",

            "Mean":
                all_coefficients[
                    "NT_coefficient"
                ].mean(),

            "SD":
                all_coefficients[
                    "NT_coefficient"
                ].std(),

            "Median":
                all_coefficients[
                    "NT_coefficient"
                ].median()
        },

        {
            "Parameter":
                "Intercept",

            "Mean":
                all_coefficients[
                    "intercept"
                ].mean(),

            "SD":
                all_coefficients[
                    "intercept"
                ].std(),

            "Median":
                all_coefficients[
                    "intercept"
                ].median()
        }
    ]
)


COEF_SUMMARY_FILE = (
    OUT_DIR /
    "StrictFusion_meta_coefficient_summary.tsv"
)


coefficient_summary.to_csv(
    str(
        COEF_SUMMARY_FILE
    ),
    sep="\t",
    index=False
)


# ==================================================================================
# 13. ADD SEQUENCE CONTROLS TO FINAL COMPARISON TABLE
#     (comparison only; NOT fusion inputs)
# ==================================================================================

CONTROL_SPLIT_FILE = (
    CONTROL_RESULT_DIR /
    "FourModel_split_level_summary.tsv"
)

CONTROL_GENE_FILE = (
    CONTROL_RESULT_DIR /
    "FourModel_gene_level_summary.tsv"
)


assert CONTROL_SPLIT_FILE.exists()
assert CONTROL_GENE_FILE.exists()


control_split = pd.read_csv(
    str(
        CONTROL_SPLIT_FILE
    ),
    sep="\t"
)

control_gene = pd.read_csv(
    str(
        CONTROL_GENE_FILE
    ),
    sep="\t"
)


# Add fusion as fifth model.
fusion_split_only = split_summary.loc[
    split_summary[
        "Model"
    ] == "Strict_Context_NTv2_LateFusion"
]


fusion_gene_only = gene_summary.loc[
    gene_summary[
        "Model"
    ] == "Strict_Context_NTv2_LateFusion"
]


five_model_split = pd.concat(
    [
        control_split,
        fusion_split_only
    ],
    ignore_index=True
)


five_model_gene = pd.concat(
    [
        control_gene,
        fusion_gene_only
    ],
    ignore_index=True
)


FIVE_SPLIT_FILE = (
    OUT_DIR /
    "FiveModel_split_level_summary.tsv"
)

FIVE_GENE_FILE = (
    OUT_DIR /
    "FiveModel_gene_level_summary.tsv"
)


five_model_split.to_csv(
    str(
        FIVE_SPLIT_FILE
    ),
    sep="\t",
    index=False
)


five_model_gene.to_csv(
    str(
        FIVE_GENE_FILE
    ),
    sep="\t",
    index=False
)


# ==================================================================================
# 14. PROVENANCE
# ==================================================================================

manifest = {
    "project":
        "ASD-ContextFusion",

    "stage":
        "Strict cross-fitted modality-preserving late fusion",

    "proposed_fusion_inputs": [
        "Dhindsa Context XGBoost probability",
        "NT-v2 L2 logistic probability"
    ],

    "sequence_controls_in_fusion":
        False,

    "sequence_controls_role":
        (
            "Comparator/baseline only; "
            "Length+GC and 4-mer are not fusion inputs."
        ),

    "fusion_model":
        "L2 LogisticRegression",

    "fusion_parameters":
        META_LOGISTIC_PARAMS,

    "meta_feature_scaling":
        False,

    "meta_feature_scale_reason":
        (
            "Both meta-features are probabilities on [0,1]."
        ),

    "outer_framework":
        (
            "Exact frozen Dhindsa/Mantis 10-fold membership"
        ),

    "outer_iterations":
        10,

    "balanced_datasets_per_iteration":
        79,

    "outer_splits":
        7900,

    "inner_cross_fitting":
        (
            "For outer fold f and inner validation fold g, "
            "base models are trained excluding both f and g."
        ),

    "inner_folds_per_outer_split":
        9,

    "inner_fit_optimization":
        (
            "45 unordered leave-two-fold-out fits per "
            "balanced dataset; each model predicts both excluded folds. "
            "This is mathematically equivalent to 90 ordered inner fits."
        ),

    "outer_test_base_predictions":
        (
            "Reused from already-saved primary Context/NT OOF predictions; "
            "each was trained without its outer test fold."
        ),

    "new_PU_sampling":
        False,

    "partition_regeneration":
        False,

    "hyperparameter_tuning":
        False,

    "result_driven_model_selection":
        False,

    "raw_feature_concatenation":
        False,

    "primary_metric":
        "Average Precision / AUPRC",

    "secondary_metric":
        "AUROC"
}


with open(
    str(
        OUT_DIR /
        "strict_crossfitted_fusion_manifest.json"
    ),
    "w"
) as f:

    json.dump(
        manifest,
        f,
        indent=2
    )


# ==================================================================================
# FINAL
# ==================================================================================

print("")
print("=" * 116)
print("STRICT CROSS-FITTED LATE FUSION COMPLETE")
print("=" * 116)

print("")
print("SPLIT-LEVEL PERFORMANCE")
print(
    split_summary.to_string(
        index=False
    )
)

print("")
print("GENE-LEVEL PERFORMANCE")
print(
    gene_summary.to_string(
        index=False
    )
)

print("")
print("META-LEARNER COEFFICIENTS")
print(
    coefficient_summary.to_string(
        index=False
    )
)

print("")
print("Five-model comparison:")
print(
    five_model_split.to_string(
        index=False
    )
)

print("")
print(
    "Outer frozen folds       : 7900"
)

print(
    "New PU sampling          : NO"
)

print(
    "Sequence controls fused  : NO"
)

print(
    "Raw concatenation        : NO"
)

print(
    "Strict inner cross-fit   : YES"
)

print("")
print("GO / NO-GO: GO")
print("=" * 116)
'''


RUNNER.write_text(
    runner_code,
    encoding="utf-8"
)


print("\nRunner:")
print(RUNNER)


# ======================================================================================
# 2. NOTEBOOK-SIDE CHECKPOINT VALIDATION
# ======================================================================================

def task_valid(
    iteration,
    balanced_dataset
):

    pf = (
        CHECKPOINT_DIR /
        f"iteration_{iteration:02d}_balanced_{balanced_dataset:02d}_fusion_predictions.tsv.gz"
    )

    mf = (
        CHECKPOINT_DIR /
        f"iteration_{iteration:02d}_balanced_{balanced_dataset:02d}_fusion_metrics.tsv"
    )

    cf = (
        CHECKPOINT_DIR /
        f"iteration_{iteration:02d}_balanced_{balanced_dataset:02d}_meta_coefficients.tsv"
    )


    if not (
        pf.exists()
        and
        mf.exists()
        and
        cf.exists()
    ):
        return False


    try:

        m = pd.read_csv(
            mf,
            sep="\t"
        )

        c = pd.read_csv(
            cf,
            sep="\t"
        )

        p = pd.read_csv(
            pf,
            sep="\t",
            compression="gzip"
        )


        return (
            len(m) == 10
            and
            m[
                "fold"
            ].nunique() == 10
            and
            len(c) == 10
            and
            c[
                "fold"
            ].nunique() == 10
            and
            p[
                "dhindsa_row_id"
            ].is_unique
        )


    except Exception:

        return False


# ======================================================================================
# 3. CURRENT STATUS
# ======================================================================================

completed_tasks = []


for iteration in range(
    1,
    11
):

    for balanced_dataset in range(
        79
    ):

        if task_valid(
            iteration,
            balanced_dataset
        ):

            completed_tasks.append(
                (
                    iteration,
                    balanced_dataset
                )
            )


print(
    "\nCompleted valid fusion tasks:",
    len(
        completed_tasks
    ),
    "/ 790"
)


if completed_tasks:

    print(
        "Last completed:",
        completed_tasks[
            -1
        ]
    )


# ======================================================================================
# 4. PROCESS ENVIRONMENT
# ======================================================================================

env_base = os.environ.copy()

env_base[
    "PYTHONIOENCODING"
] = "utf-8"

env_base[
    "PYTHONUTF8"
] = "1"

# Engineering/resource controls only.
env_base[
    "OMP_NUM_THREADS"
] = "1"

env_base[
    "MKL_NUM_THREADS"
] = "1"

env_base[
    "OPENBLAS_NUM_THREADS"
] = "1"

env_base[
    "NUMEXPR_NUM_THREADS"
] = "1"


# ======================================================================================
# 5. RUN ONLY INCOMPLETE BALANCED-DATASET TASKS
# ======================================================================================

with open(
    LOG_FILE,
    "a",
    encoding="utf-8"
) as log:


    for iteration in range(
        1,
        11
    ):


        print(
            "\n" +
            "=" * 124
        )

        print(
            f"FUSION STOCHASTIC ITERATION {iteration:02d}"
        )

        print(
            "=" * 124
        )


        for balanced_dataset in range(
            79
        ):


            if task_valid(
                iteration,
                balanced_dataset
            ):

                print(
                    f"Iteration {iteration:02d} | "
                    f"balanced {balanced_dataset:02d}: "
                    f"COMPLETE -> SKIP"
                )

                continue


            print(
                "\n" +
                "-" * 124
            )

            print(
                f"STARTING iteration {iteration:02d} | "
                f"balanced {balanced_dataset:02d} "
                f"IN FRESH PROCESS"
            )

            print(
                "-" * 124
            )


            env = (
                env_base.copy()
            )

            env[
                "ASD_CF_FUSION_ITERATION"
            ] = str(
                iteration
            )

            env[
                "ASD_CF_FUSION_BALANCED_DATASET"
            ] = str(
                balanced_dataset
            )


            process = subprocess.Popen(
                [
                    str(
                        LEGACY_PYTHON
                    ),
                    "-u",
                    str(
                        RUNNER
                    )
                ],

                cwd=str(
                    REPO
                ),

                stdout=subprocess.PIPE,
                stderr=subprocess.STDOUT,

                text=True,
                encoding="utf-8",
                errors="replace",

                bufsize=1,
                env=env
            )


            for line in process.stdout:

                # Keep all messages in log.
                log.write(
                    line
                )

                log.flush()


                # Hide only repetitive XGBoost warning in notebook.
                if (
                    "default evaluation metric used"
                    in line
                ):
                    continue


                print(
                    line,
                    end=""
                )


            return_code = (
                process.wait()
            )


            if return_code != 0:

                print(
                    "\n" +
                    "=" * 124
                )

                print(
                    f"FUSION TASK FAILED: "
                    f"iteration {iteration:02d}, "
                    f"balanced {balanced_dataset:02d}"
                )

                print(
                    "All previously completed balanced-dataset "
                    "checkpoints remain safe."
                )

                print(
                    "Rerun THIS SAME CELL after fixing/restarting."
                )

                print(
                    "=" * 124
                )


                raise RuntimeError(
                    "Strict fusion task failed."
                )


            assert task_valid(
                iteration,
                balanced_dataset
            ), (
                "Process exited without a valid checkpoint."
            )


            mf = (
                CHECKPOINT_DIR /
                f"iteration_{iteration:02d}_balanced_{balanced_dataset:02d}_fusion_metrics.tsv"
            )


            m = pd.read_csv(
                mf,
                sep="\t"
            )


            print(
                f"\nVERIFIED [OK] | "
                f"Iteration {iteration:02d} | "
                f"Balanced {balanced_dataset:02d}"
            )

            print(
                "Context AP:",
                f"{m['Context_AP'].mean():.4f}"
            )

            print(
                "NT AP     :",
                f"{m['NT_AP'].mean():.4f}"
            )

            print(
                "Fusion AP :",
                f"{m['Fusion_AP'].mean():.4f}"
            )

            print(
                "Fresh process exited -> RAM released."
            )


# ======================================================================================
# 6. VERIFY ALL 790 CHECKPOINTS
# ======================================================================================

valid_count = sum(
    task_valid(
        iteration,
        balanced_dataset
    )
    for iteration in range(
        1,
        11
    )
    for balanced_dataset in range(
        79
    )
)


print(
    "\n" +
    "=" * 124
)

print(
    "STRICT FUSION CHECKPOINT AUDIT"
)

print(
    "=" * 124
)

print(
    "Valid balanced-dataset tasks:",
    valid_count,
    "/ 790"
)


assert valid_count == 790, (
    "All 790 fusion tasks are not complete. "
    "Final aggregation will NOT run."
)


print(
    "ALL 790 STRICT-FUSION TASKS VERIFIED [OK]"
)


# ======================================================================================
# 7. FINAL AGGREGATION
# ======================================================================================

print(
    "\n" +
    "=" * 124
)

print(
    "RUNNING STRICT-FUSION FINAL AGGREGATION"
)

print(
    "=" * 124
)


final_env = (
    env_base.copy()
)

final_env.pop(
    "ASD_CF_FUSION_ITERATION",
    None
)

final_env.pop(
    "ASD_CF_FUSION_BALANCED_DATASET",
    None
)


process = subprocess.Popen(
    [
        str(
            LEGACY_PYTHON
        ),
        "-u",
        str(
            RUNNER
        )
    ],

    cwd=str(
        REPO
    ),

    stdout=subprocess.PIPE,
    stderr=subprocess.STDOUT,

    text=True,
    encoding="utf-8",
    errors="replace",

    bufsize=1,
    env=final_env
)


with open(
    LOG_FILE,
    "a",
    encoding="utf-8"
) as log:


    for line in process.stdout:

        log.write(
            line
        )

        log.flush()

        print(
            line,
            end=""
        )


return_code = (
    process.wait()
)


if return_code != 0:

    raise RuntimeError(
        "Strict fusion aggregation failed. "
        "All 790 balanced-dataset checkpoints remain safe."
    )


# ======================================================================================
# 8. NOTEBOOK-SIDE FINAL AUDIT
# ======================================================================================

FINAL_METRICS = (
    RESULT_DIR /
    "StrictFusion_all_7900_fold_metrics.tsv"
)

FINAL_GENE = (
    RESULT_DIR /
    "StrictFusion_gene_level_OOF_scores.tsv"
)

FINAL_GENE_SUMMARY = (
    RESULT_DIR /
    "StrictFusion_gene_level_summary.tsv"
)

FINAL_FIVE_SPLIT = (
    RESULT_DIR /
    "FiveModel_split_level_summary.tsv"
)

FINAL_FIVE_GENE = (
    RESULT_DIR /
    "FiveModel_gene_level_summary.tsv"
)


for p in [
    FINAL_METRICS,
    FINAL_GENE,
    FINAL_GENE_SUMMARY,
    FINAL_FIVE_SPLIT,
    FINAL_FIVE_GENE
]:

    assert p.exists(), (
        f"Missing:\n{p}"
    )


final_metrics = pd.read_csv(
    FINAL_METRICS,
    sep="\t"
)

final_gene = pd.read_csv(
    FINAL_GENE,
    sep="\t"
)


assert len(
    final_metrics
) == 7900

assert len(
    final_gene
) == 18252

assert final_gene[
    "Gene_Name"
].nunique() == 18252

assert final_gene.loc[
    final_gene[
        "known_gene"
    ] == 1,
    "Gene_Name"
].nunique() == 190


print(
    "\n" +
    "=" * 124
)

print(
    "STRICT CROSS-FITTED CONTEXT + NT-v2 FUSION COMPLETE [OK]"
)

print(
    "=" * 124
)


print(
    "\nFIVE-MODEL SPLIT-LEVEL SUMMARY:"
)

print(
    pd.read_csv(
        FINAL_FIVE_SPLIT,
        sep="\t"
    ).to_string(
        index=False
    )
)


print(
    "\nFIVE-MODEL GENE-LEVEL SUMMARY:"
)

print(
    pd.read_csv(
        FINAL_FIVE_GENE,
        sep="\t"
    ).to_string(
        index=False
    )
)


print(
    "\nFrozen outer test folds : 7900 [OK]"
)

print(
    "Genes scored            : 18252 [OK]"
)

print(
    "ASD genes               : 190 [OK]"
)

print(
    "New PU sampling         : NO"
)

print(
    "Raw concatenation       : NO"
)

print(
    "Controls fused          : NO"
)

print(
    "Strict nested cross-fit : YES"
)

print(
    "\nNEXT:"
)

print(
    "Paired statistical comparison + "
    "leakage-sensitive robustness analysis."
)

print(
    "=" * 124
)
```

```python
# ======================================================================================
# ASD-ContextFusion
# STRICT FUSION - 6 WORKER PARALLEL RESUME
#
# USE AFTER JUPYTER KERNEL RESTART
#
# IMPORTANT:
#   This cell DOES NOT rewrite the scientific runner.
#   It resumes the EXISTING strict-fusion runner:
#
#   E:\ASD_ContextFusion_Q1\00_admin\
#   run_strict_crossfitted_context_NTv2_fusion.py
#
# SCIENTIFIC METHOD: UNCHANGED
# ENGINEERING CHANGE ONLY: 6 independent balanced-dataset tasks run in parallel.
#
# SAFE TO RERUN:
#   Valid completed checkpoints are automatically skipped.
# ======================================================================================

from pathlib import Path
import os
import subprocess
import pandas as pd
import numpy as np
import threading
import time

from concurrent.futures import (
    ThreadPoolExecutor,
    as_completed
)


# ======================================================================================
# 0. SETTINGS
# ======================================================================================

ROOT = Path(r"E:\ASD_ContextFusion_Q1")

REPO = (
    ROOT /
    "01_raw" /
    "dhindsa_2025" /
    "Mantis-ml-NDD"
)

LEGACY_PYTHON = (
    ROOT /
    "envs" /
    "mantis37" /
    "python.exe"
)

RUNNER = (
    ROOT /
    "00_admin" /
    "run_strict_crossfitted_context_NTv2_fusion.py"
)

PARTITION_DIR = (
    ROOT /
    "04_models" /
    "checkpoints" /
    "dhindsa_PU_common_universe"
)

CHECKPOINT_DIR = (
    ROOT /
    "04_models" /
    "checkpoints" /
    "strict_context_NTv2_fusion"
)

RESULT_DIR = (
    ROOT /
    "05_results" /
    "proposed" /
    "strict_context_NTv2_fusion"
)

PARALLEL_LOG_DIR = (
    ROOT /
    "08_logs" /
    "strict_context_NTv2_fusion_parallel"
)


# ----------------------------------------------------------------------
# Recommended for your 32 GB RAM machine.
# If RAM usage becomes excessive, change ONLY this to 4.
# This is engineering-only and does not change scientific results.
# ----------------------------------------------------------------------

MAX_WORKERS = 6


CHECKPOINT_DIR.mkdir(
    parents=True,
    exist_ok=True
)

RESULT_DIR.mkdir(
    parents=True,
    exist_ok=True
)

PARALLEL_LOG_DIR.mkdir(
    parents=True,
    exist_ok=True
)


# ======================================================================================
# 1. REQUIRED FILE CHECK
# ======================================================================================

required = [
    REPO,
    LEGACY_PYTHON,
    RUNNER,
    PARTITION_DIR
]


for p in required:

    assert p.exists(), (
        f"\nMISSING REQUIRED PATH:\n{p}\n"
    )


print("=" * 124)
print("ASD-CONTEXTFUSION - 6 WORKER STRICT-FUSION PARALLEL RESUME")
print("=" * 124)

print("\nRunner:")
print(RUNNER)

print("\nWorkers:", MAX_WORKERS)

print(
    "\nScientific model/folds unchanged. "
    "Only independent task scheduling is parallelized."
)


# ======================================================================================
# 2. LOAD FROZEN MEMBERSHIP ONCE
#
# Used ONLY to validate checkpoint identity.
# NO partitions are generated or modified.
# ======================================================================================

print(
    "\n[1/5] Loading frozen PU membership for checkpoint validation..."
)


expected_ids = {}


for iteration in range(1, 11):

    membership_file = (
        PARTITION_DIR /
        f"iteration_{iteration:02d}_test_fold_membership.tsv.gz"
    )


    assert membership_file.exists(), membership_file


    membership = pd.read_csv(
        membership_file,
        sep="\t",
        compression="gzip",
        usecols=[
            "balanced_dataset",
            "dhindsa_row_id"
        ]
    )


    assert (
        membership[
            "balanced_dataset"
        ].nunique()
        == 79
    )


    for balanced_dataset, df in membership.groupby(
        "balanced_dataset",
        sort=True
    ):

        expected_ids[
            (
                iteration,
                int(
                    balanced_dataset
                )
            )
        ] = set(
            df[
                "dhindsa_row_id"
            ]
            .astype(int)
            .tolist()
        )


assert len(
    expected_ids
) == 790


print(
    "Frozen tasks loaded:",
    len(expected_ids),
    "[OK]"
)


# ======================================================================================
# 3. STRICT CHECKPOINT VALIDATOR
# ======================================================================================

def pred_file(
    iteration,
    balanced_dataset
):

    return (
        CHECKPOINT_DIR /
        (
            f"iteration_{iteration:02d}_balanced_"
            f"{balanced_dataset:02d}_fusion_predictions.tsv.gz"
        )
    )


def metric_file(
    iteration,
    balanced_dataset
):

    return (
        CHECKPOINT_DIR /
        (
            f"iteration_{iteration:02d}_balanced_"
            f"{balanced_dataset:02d}_fusion_metrics.tsv"
        )
    )


def coef_file(
    iteration,
    balanced_dataset
):

    return (
        CHECKPOINT_DIR /
        (
            f"iteration_{iteration:02d}_balanced_"
            f"{balanced_dataset:02d}_meta_coefficients.tsv"
        )
    )


def task_valid(
    iteration,
    balanced_dataset
):

    pf = pred_file(
        iteration,
        balanced_dataset
    )

    mf = metric_file(
        iteration,
        balanced_dataset
    )

    cf = coef_file(
        iteration,
        balanced_dataset
    )


    if not (
        pf.exists()
        and
        mf.exists()
        and
        cf.exists()
    ):

        return False


    try:

        metrics = pd.read_csv(
            mf,
            sep="\t"
        )

        coefficients = pd.read_csv(
            cf,
            sep="\t"
        )

        predictions = pd.read_csv(
            pf,
            sep="\t",
            compression="gzip"
        )


        # --------------------------------------------------------------
        # Expected 10 outer folds
        # --------------------------------------------------------------

        if len(metrics) != 10:
            return False

        if (
            metrics[
                "fold"
            ].nunique()
            != 10
        ):
            return False


        if len(coefficients) != 10:
            return False

        if (
            coefficients[
                "fold"
            ].nunique()
            != 10
        ):
            return False


        # --------------------------------------------------------------
        # Required output columns
        # --------------------------------------------------------------

        required_prediction_columns = {
            "iteration",
            "balanced_dataset",
            "fold",
            "dhindsa_row_id",
            "Gene_Name",
            "known_gene",
            "Context_probability",
            "NT_probability",
            "Fusion_probability"
        }


        if not required_prediction_columns.issubset(
            set(
                predictions.columns
            )
        ):

            return False


        # --------------------------------------------------------------
        # Exactly one OOF prediction per row in this balanced dataset
        # --------------------------------------------------------------

        if not predictions[
            "dhindsa_row_id"
        ].is_unique:

            return False


        observed_ids = set(
            predictions[
                "dhindsa_row_id"
            ]
            .astype(int)
            .tolist()
        )


        if observed_ids != expected_ids[
            (
                iteration,
                balanced_dataset
            )
        ]:

            return False


        # --------------------------------------------------------------
        # Finite probabilities
        # --------------------------------------------------------------

        probability_columns = [
            "Context_probability",
            "NT_probability",
            "Fusion_probability"
        ]


        for col in probability_columns:

            values = predictions[
                col
            ].to_numpy(
                dtype=float
            )

            if not np.isfinite(
                values
            ).all():

                return False


        return True


    except Exception:

        return False


# ======================================================================================
# 4. FIND COMPLETED + REMAINING TASKS
# ======================================================================================

print(
    "\n[2/5] Auditing existing strict-fusion checkpoints..."
)


all_tasks = [
    (
        iteration,
        balanced_dataset
    )
    for iteration in range(
        1,
        11
    )
    for balanced_dataset in range(
        79
    )
]


completed = []

remaining = []


for task in all_tasks:

    iteration, balanced_dataset = task


    if task_valid(
        iteration,
        balanced_dataset
    ):

        completed.append(
            task
        )

    else:

        remaining.append(
            task
        )


print(
    "Completed valid tasks :",
    len(completed),
    "/ 790"
)

print(
    "Remaining tasks       :",
    len(remaining)
)


if completed:

    print(
        "Last valid checkpoint:",
        completed[-1]
    )


if not remaining:

    print(
        "\nAll 790 tasks are already complete."
    )


# ======================================================================================
# 5. BASE ENVIRONMENT
#
# IMPORTANT:
# Each scientific model remains single-threaded.
# We parallelize ACROSS independent balanced datasets.
# ======================================================================================

env_base = os.environ.copy()


env_base[
    "PYTHONIOENCODING"
] = "utf-8"

env_base[
    "PYTHONUTF8"
] = "1"


# Engineering/resource controls only.
env_base[
    "OMP_NUM_THREADS"
] = "1"

env_base[
    "MKL_NUM_THREADS"
] = "1"

env_base[
    "OPENBLAS_NUM_THREADS"
] = "1"

env_base[
    "NUMEXPR_NUM_THREADS"
] = "1"


# ======================================================================================
# 6. PARALLEL TASK FUNCTION
# ======================================================================================

print_lock = threading.Lock()


def run_one_task(
    task
):

    iteration, balanced_dataset = task


    # Recheck immediately before launch.
    if task_valid(
        iteration,
        balanced_dataset
    ):

        return {
            "iteration":
                iteration,

            "balanced_dataset":
                balanced_dataset,

            "status":
                "SKIPPED",

            "return_code":
                0
        }


    env = env_base.copy()


    env[
        "ASD_CF_FUSION_ITERATION"
    ] = str(
        iteration
    )

    env[
        "ASD_CF_FUSION_BALANCED_DATASET"
    ] = str(
        balanced_dataset
    )


    task_log = (
        PARALLEL_LOG_DIR /
        (
            f"iteration_{iteration:02d}_"
            f"balanced_{balanced_dataset:02d}.log"
        )
    )


    start = time.time()


    # Each task writes to its OWN log file.
    # Therefore parallel processes do not corrupt a shared log.
    with open(
        task_log,
        "w",
        encoding="utf-8"
    ) as log:


        process = subprocess.Popen(
            [
                str(
                    LEGACY_PYTHON
                ),
                "-u",
                str(
                    RUNNER
                )
            ],

            cwd=str(
                REPO
            ),

            stdout=log,
            stderr=subprocess.STDOUT,

            env=env
        )


        return_code = (
            process.wait()
        )


    elapsed = (
        time.time()
        -
        start
    )


    if return_code != 0:

        return {
            "iteration":
                iteration,

            "balanced_dataset":
                balanced_dataset,

            "status":
                "FAILED",

            "return_code":
                return_code,

            "minutes":
                elapsed / 60.0,

            "log":
                str(
                    task_log
                )
        }


    # Scientific checkpoint must pass after process exits.
    if not task_valid(
        iteration,
        balanced_dataset
    ):

        return {
            "iteration":
                iteration,

            "balanced_dataset":
                balanced_dataset,

            "status":
                "INVALID_CHECKPOINT",

            "return_code":
                return_code,

            "minutes":
                elapsed / 60.0,

            "log":
                str(
                    task_log
                )
        }


    metrics = pd.read_csv(
        metric_file(
            iteration,
            balanced_dataset
        ),
        sep="\t"
    )


    return {
        "iteration":
            iteration,

        "balanced_dataset":
            balanced_dataset,

        "status":
            "COMPLETE",

        "return_code":
            0,

        "minutes":
            elapsed / 60.0,

        "Context_AP":
            metrics[
                "Context_AP"
            ].mean(),

        "NT_AP":
            metrics[
                "NT_AP"
            ].mean(),

        "Fusion_AP":
            metrics[
                "Fusion_AP"
            ].mean(),

        "log":
            str(
                task_log
            )
    }


# ======================================================================================
# 7. RUN REMAINING TASKS WITH 6 WORKERS
# ======================================================================================

print(
    "\n[3/5] Starting parallel strict-fusion resume..."
)

print(
    "Parallel workers:",
    MAX_WORKERS
)

print(
    "Each XGBoost model still uses nthread=1."
)

print(
    "Each completed task is checkpointed independently."
)


start_global = time.time()

newly_completed = 0

failed_tasks = []


if remaining:


    with ThreadPoolExecutor(
        max_workers=MAX_WORKERS
    ) as executor:


        future_to_task = {
            executor.submit(
                run_one_task,
                task
            ):
            task

            for task in remaining
        }


        total_remaining = len(
            remaining
        )


        for future in as_completed(
            future_to_task
        ):


            task = future_to_task[
                future
            ]


            try:

                result = (
                    future.result()
                )


            except Exception as e:

                failed_tasks.append(
                    (
                        task,
                        repr(e)
                    )
                )

                with print_lock:

                    print(
                        "\nFAILED:",
                        task,
                        repr(e)
                    )

                continue


            status = result[
                "status"
            ]


            if status in [
                "COMPLETE",
                "SKIPPED"
            ]:

                newly_completed += 1


            else:

                failed_tasks.append(
                    (
                        task,
                        result
                    )
                )


            with print_lock:


                done_now = (
                    len(completed)
                    +
                    newly_completed
                )


                elapsed_hours = (
                    time.time()
                    -
                    start_global
                ) / 3600.0


                if status == "COMPLETE":

                    print(
                        "\n"
                        f"[{done_now}/790] "
                        f"ITER {result['iteration']:02d} | "
                        f"BAL {result['balanced_dataset']:02d} "
                        f"COMPLETE [OK]"
                    )

                    print(
                        "   Task time :",
                        f"{result['minutes']:.1f} min"
                    )

                    print(
                        "   Context AP:",
                        f"{result['Context_AP']:.4f}"
                    )

                    print(
                        "   NT AP     :",
                        f"{result['NT_AP']:.4f}"
                    )

                    print(
                        "   Fusion AP :",
                        f"{result['Fusion_AP']:.4f}"
                    )

                    print(
                        "   Parallel elapsed:",
                        f"{elapsed_hours:.2f} h"
                    )


                elif status == "SKIPPED":

                    print(
                        f"\nITER {result['iteration']:02d} | "
                        f"BAL {result['balanced_dataset']:02d}: "
                        "already valid -> SKIPPED"
                    )


                else:

                    print(
                        "\nTASK ERROR:",
                        result
                    )


# ======================================================================================
# 8. FINAL CHECKPOINT AUDIT
# ======================================================================================

print(
    "\n" +
    "=" * 124
)

print(
    "[4/5] FINAL 790-TASK CHECKPOINT AUDIT"
)

print(
    "=" * 124
)


valid_tasks = []


for iteration in range(
    1,
    11
):

    for balanced_dataset in range(
        79
    ):

        if task_valid(
            iteration,
            balanced_dataset
        ):

            valid_tasks.append(
                (
                    iteration,
                    balanced_dataset
                )
            )


valid_count = len(
    valid_tasks
)


print(
    "Valid strict-fusion tasks:",
    valid_count,
    "/ 790"
)


if failed_tasks:

    print(
        "\nFailed/invalid tasks detected:",
        len(
            failed_tasks
        )
    )

    for item in failed_tasks[
        :10
    ]:

        print(
            item
        )


if valid_count != 790:

    print(
        "\nNOT ALL TASKS COMPLETE."
    )

    print(
        "Do NOT delete checkpoints."
    )

    print(
        "Rerun THIS SAME CELL."
    )

    print(
        "All valid completed tasks will be skipped."
    )

    raise RuntimeError(
        f"Only {valid_count}/790 tasks are valid."
    )


print(
    "\nALL 790 STRICT-FUSION TASKS VERIFIED [OK]"
)


# ======================================================================================
# 9. RUN AGGREGATION ONLY
#
# No iteration/balanced environment variables are provided.
# Therefore runner enters its aggregation-only mode.
# ======================================================================================

print(
    "\n" +
    "=" * 124
)

print(
    "[5/5] RUNNING FINAL STRICT-FUSION AGGREGATION"
)

print(
    "=" * 124
)


aggregation_log = (
    PARALLEL_LOG_DIR /
    "FINAL_AGGREGATION.log"
)


final_env = (
    env_base.copy()
)


final_env.pop(
    "ASD_CF_FUSION_ITERATION",
    None
)

final_env.pop(
    "ASD_CF_FUSION_BALANCED_DATASET",
    None
)


with open(
    aggregation_log,
    "w",
    encoding="utf-8"
) as log:


    process = subprocess.Popen(
        [
            str(
                LEGACY_PYTHON
            ),
            "-u",
            str(
                RUNNER
            )
        ],

        cwd=str(
            REPO
        ),

        stdout=subprocess.PIPE,
        stderr=subprocess.STDOUT,

        text=True,
        encoding="utf-8",
        errors="replace",

        bufsize=1,
        env=final_env
    )


    for line in process.stdout:

        log.write(
            line
        )

        log.flush()

        print(
            line,
            end=""
        )


    return_code = (
        process.wait()
    )


if return_code != 0:

    raise RuntimeError(
        "\nAggregation failed.\n"
        "ALL 790 task checkpoints are still safe.\n"
        f"Check log:\n{aggregation_log}"
    )


# ======================================================================================
# 10. FINAL OUTPUT VERIFICATION
# ======================================================================================

FINAL_SPLIT = (
    RESULT_DIR /
    "FiveModel_split_level_summary.tsv"
)

FINAL_GENE = (
    RESULT_DIR /
    "FiveModel_gene_level_summary.tsv"
)

FINAL_METRICS = (
    RESULT_DIR /
    "StrictFusion_all_7900_fold_metrics.tsv"
)

FINAL_SCORES = (
    RESULT_DIR /
    "StrictFusion_gene_level_OOF_scores.tsv"
)


for p in [
    FINAL_SPLIT,
    FINAL_GENE,
    FINAL_METRICS,
    FINAL_SCORES
]:

    assert p.exists(), (
        f"Missing final output:\n{p}"
    )


metrics = pd.read_csv(
    FINAL_METRICS,
    sep="\t"
)

gene_scores = pd.read_csv(
    FINAL_SCORES,
    sep="\t"
)


assert len(
    metrics
) == 7900

assert len(
    gene_scores
) == 18252

assert (
    gene_scores[
        "Gene_Name"
    ].nunique()
    == 18252
)

assert (
    gene_scores.loc[
        gene_scores[
            "known_gene"
        ] == 1,
        "Gene_Name"
    ].nunique()
    == 190
)


print(
    "\n" +
    "=" * 124
)

print(
    "STRICT CONTEXT + NT-v2 FUSION COMPLETE [OK]"
)

print(
    "=" * 124
)


print(
    "\nFIVE-MODEL SPLIT-LEVEL SUMMARY:"
)

print(
    pd.read_csv(
        FINAL_SPLIT,
        sep="\t"
    ).to_string(
        index=False
    )
)


print(
    "\nFIVE-MODEL GENE-LEVEL SUMMARY:"
)

print(
    pd.read_csv(
        FINAL_GENE,
        sep="\t"
    ).to_string(
        index=False
    )
)


print(
    "\nFinal integrity:"
)

print(
    "Strict fusion tasks     : 790/790 [OK]"
)

print(
    "Frozen outer folds      : 7900 [OK]"
)

print(
    "Genes scored            : 18252 [OK]"
)

print(
    "ASD genes               : 190 [OK]"
)

print(
    "New PU sampling         : NO"
)

print(
    "Primary models modified : NO"
)

print(
    "Raw concatenation       : NO"
)

print(
    "Sequence controls fused : NO"
)

print(
    "Strict nested cross-fit : YES"
)

print(
    "Parallel workers        :",
    MAX_WORKERS
)

print(
    "\nNEXT SCIENTIFIC STAGE:"
)

print(
    "Paired statistical comparison + leakage-sensitive robustness."
)

print(
    "=" * 124
)
```

```python
# ======================================================================================
# ASD-ContextFusion
# STRICT FUSION - 6 WORKER PARALLEL RESUME AFTER KERNEL RESTART
#
# SCIENTIFIC METHOD: UNCHANGED
# - same frozen 7,900 PU folds
# - same strict nested cross-fitting
# - same Context XGBoost parameters
# - same NT-v2 logistic parameters
# - same existing scientific runner
# - no new PU sampling
# - no tuning
#
# ENGINEERING ONLY:
# - resume valid checkpoints
# - 6 independent balanced-dataset tasks in parallel
# ======================================================================================

from pathlib import Path
import os
import subprocess
import pandas as pd
import numpy as np
import threading
import time

from concurrent.futures import ThreadPoolExecutor, as_completed


# ======================================================================================
# 0. PATHS / SETTINGS
# ======================================================================================

ROOT = Path(r"E:\ASD_ContextFusion_Q1")

REPO = (
    ROOT /
    "01_raw" /
    "dhindsa_2025" /
    "Mantis-ml-NDD"
)

LEGACY_PYTHON = (
    ROOT /
    "envs" /
    "mantis37" /
    "python.exe"
)

# IMPORTANT:
# Existing scientific runner is reused.
# It is NOT rewritten.
RUNNER = (
    ROOT /
    "00_admin" /
    "run_strict_crossfitted_context_NTv2_fusion.py"
)

PARTITION_DIR = (
    ROOT /
    "04_models" /
    "checkpoints" /
    "dhindsa_PU_common_universe"
)

CHECKPOINT_DIR = (
    ROOT /
    "04_models" /
    "checkpoints" /
    "strict_context_NTv2_fusion"
)

RESULT_DIR = (
    ROOT /
    "05_results" /
    "proposed" /
    "strict_context_NTv2_fusion"
)

PARALLEL_LOG_DIR = (
    ROOT /
    "08_logs" /
    "strict_context_NTv2_fusion_parallel"
)

MAX_WORKERS = 6


CHECKPOINT_DIR.mkdir(parents=True, exist_ok=True)
RESULT_DIR.mkdir(parents=True, exist_ok=True)
PARALLEL_LOG_DIR.mkdir(parents=True, exist_ok=True)


for p in [
    REPO,
    LEGACY_PYTHON,
    RUNNER,
    PARTITION_DIR
]:
    assert p.exists(), f"MISSING:\n{p}"


print("=" * 124)
print("ASD-CONTEXTFUSION - STRICT FUSION PARALLEL RESUME")
print("=" * 124)

print("\nRunner:")
print(RUNNER)

print("\nParallel workers:", MAX_WORKERS)

print(
    "\nScientific model/folds unchanged. "
    "Only incomplete independent tasks will be executed."
)


# ======================================================================================
# 1. LOAD EXACT FROZEN TASK MEMBERSHIP
# ======================================================================================

print("\n[1/5] Loading frozen PU membership...")


expected_ids = {}


for iteration in range(1, 11):

    membership_file = (
        PARTITION_DIR /
        f"iteration_{iteration:02d}_test_fold_membership.tsv.gz"
    )

    assert membership_file.exists(), membership_file

    membership = pd.read_csv(
        membership_file,
        sep="\t",
        compression="gzip",
        usecols=[
            "balanced_dataset",
            "dhindsa_row_id"
        ]
    )

    assert membership["balanced_dataset"].nunique() == 79

    for balanced_dataset, df in membership.groupby(
        "balanced_dataset",
        sort=True
    ):

        expected_ids[
            (
                iteration,
                int(balanced_dataset)
            )
        ] = set(
            df["dhindsa_row_id"]
            .astype(int)
            .tolist()
        )


assert len(expected_ids) == 790

print("Frozen tasks:", len(expected_ids), "[OK]")


# ======================================================================================
# 2. CHECKPOINT PATH HELPERS
# ======================================================================================

def pred_file(iteration, balanced_dataset):

    return (
        CHECKPOINT_DIR /
        (
            f"iteration_{iteration:02d}_balanced_"
            f"{balanced_dataset:02d}_fusion_predictions.tsv.gz"
        )
    )


def metric_file(iteration, balanced_dataset):

    return (
        CHECKPOINT_DIR /
        (
            f"iteration_{iteration:02d}_balanced_"
            f"{balanced_dataset:02d}_fusion_metrics.tsv"
        )
    )


def coef_file(iteration, balanced_dataset):

    return (
        CHECKPOINT_DIR /
        (
            f"iteration_{iteration:02d}_balanced_"
            f"{balanced_dataset:02d}_meta_coefficients.tsv"
        )
    )


# ======================================================================================
# 3. STRICT VALIDATION OF EXISTING CHECKPOINT
# ======================================================================================

def task_valid(iteration, balanced_dataset):

    pf = pred_file(iteration, balanced_dataset)
    mf = metric_file(iteration, balanced_dataset)
    cf = coef_file(iteration, balanced_dataset)

    if not (
        pf.exists()
        and mf.exists()
        and cf.exists()
    ):
        return False

    try:

        metrics = pd.read_csv(
            mf,
            sep="\t"
        )

        coefficients = pd.read_csv(
            cf,
            sep="\t"
        )

        predictions = pd.read_csv(
            pf,
            sep="\t",
            compression="gzip"
        )


        # --------------------------------------------------------------
        # Exactly 10 frozen outer folds
        # --------------------------------------------------------------

        if len(metrics) != 10:
            return False

        if metrics["fold"].nunique() != 10:
            return False

        if len(coefficients) != 10:
            return False

        if coefficients["fold"].nunique() != 10:
            return False


        # --------------------------------------------------------------
        # Required columns
        # --------------------------------------------------------------

        required_prediction_columns = {
            "iteration",
            "balanced_dataset",
            "fold",
            "dhindsa_row_id",
            "Gene_Name",
            "known_gene",
            "Context_probability",
            "NT_probability",
            "Fusion_probability"
        }

        if not required_prediction_columns.issubset(
            set(predictions.columns)
        ):
            return False


        # --------------------------------------------------------------
        # Exactly one prediction per Dhindsa row
        # --------------------------------------------------------------

        if not predictions["dhindsa_row_id"].is_unique:
            return False


        observed_ids = set(
            predictions["dhindsa_row_id"]
            .astype(int)
            .tolist()
        )

        expected = expected_ids[
            (
                iteration,
                balanced_dataset
            )
        ]

        if observed_ids != expected:
            return False


        # --------------------------------------------------------------
        # Finite probabilities
        # --------------------------------------------------------------

        for col in [
            "Context_probability",
            "NT_probability",
            "Fusion_probability"
        ]:

            x = predictions[col].to_numpy(
                dtype=float
            )

            if not np.isfinite(x).all():
                return False


        return True


    except Exception:

        return False


# ======================================================================================
# 4. CURRENT RESUME STATUS
# ======================================================================================

print("\n[2/5] Auditing saved checkpoints...")


all_tasks = [
    (
        iteration,
        balanced_dataset
    )
    for iteration in range(1, 11)
    for balanced_dataset in range(79)
]


completed = []
remaining = []


for task in all_tasks:

    iteration, balanced_dataset = task

    if task_valid(
        iteration,
        balanced_dataset
    ):

        completed.append(task)

    else:

        remaining.append(task)


print("\n" + "=" * 80)

print(
    "Completed valid tasks :",
    len(completed),
    "/ 790"
)

print(
    "Remaining tasks       :",
    len(remaining)
)

print(
    "Progress              :",
    f"{100 * len(completed) / 790:.2f}%"
)

if completed:
    print(
        "Last valid task       :",
        completed[-1]
    )

print("=" * 80)


# Based on your previous supplied log,
# approximately 478/790 should already be valid.
# But we trust DISK CHECKPOINT VALIDATION, not the previous console text.


# ======================================================================================
# 5. SINGLE-THREAD EACH SCIENTIFIC MODEL
# ======================================================================================

env_base = os.environ.copy()

env_base["PYTHONIOENCODING"] = "utf-8"
env_base["PYTHONUTF8"] = "1"

# Engineering/resource controls only.
env_base["OMP_NUM_THREADS"] = "1"
env_base["MKL_NUM_THREADS"] = "1"
env_base["OPENBLAS_NUM_THREADS"] = "1"
env_base["NUMEXPR_NUM_THREADS"] = "1"


# Remove stale task selectors if inherited somehow.
env_base.pop(
    "ASD_CF_FUSION_ITERATION",
    None
)

env_base.pop(
    "ASD_CF_FUSION_BALANCED_DATASET",
    None
)


# ======================================================================================
# 6. RUN ONE INCOMPLETE TASK
# ======================================================================================

print_lock = threading.Lock()


def run_one_task(task):

    iteration, balanced_dataset = task


    # Recheck just before launching.
    if task_valid(
        iteration,
        balanced_dataset
    ):

        return {
            "iteration": iteration,
            "balanced_dataset": balanced_dataset,
            "status": "SKIPPED"
        }


    env = env_base.copy()

    env[
        "ASD_CF_FUSION_ITERATION"
    ] = str(iteration)

    env[
        "ASD_CF_FUSION_BALANCED_DATASET"
    ] = str(balanced_dataset)


    task_log = (
        PARALLEL_LOG_DIR /
        (
            f"iteration_{iteration:02d}_"
            f"balanced_{balanced_dataset:02d}.log"
        )
    )


    start = time.time()


    with open(
        task_log,
        "w",
        encoding="utf-8"
    ) as log:

        process = subprocess.Popen(
            [
                str(LEGACY_PYTHON),
                "-u",
                str(RUNNER)
            ],
            cwd=str(REPO),
            stdout=log,
            stderr=subprocess.STDOUT,
            env=env
        )

        return_code = process.wait()


    elapsed_minutes = (
        time.time() - start
    ) / 60.0


    if return_code != 0:

        return {
            "iteration": iteration,
            "balanced_dataset": balanced_dataset,
            "status": "FAILED",
            "return_code": return_code,
            "minutes": elapsed_minutes,
            "log": str(task_log)
        }


    if not task_valid(
        iteration,
        balanced_dataset
    ):

        return {
            "iteration": iteration,
            "balanced_dataset": balanced_dataset,
            "status": "INVALID_CHECKPOINT",
            "minutes": elapsed_minutes,
            "log": str(task_log)
        }


    metrics = pd.read_csv(
        metric_file(
            iteration,
            balanced_dataset
        ),
        sep="\t"
    )


    return {
        "iteration": iteration,
        "balanced_dataset": balanced_dataset,
        "status": "COMPLETE",
        "minutes": elapsed_minutes,

        "Context_AP":
            float(
                metrics["Context_AP"].mean()
            ),

        "NT_AP":
            float(
                metrics["NT_AP"].mean()
            ),

        "Fusion_AP":
            float(
                metrics["Fusion_AP"].mean()
            ),

        "log":
            str(task_log)
    }


# ======================================================================================
# 7. PARALLEL RESUME
# ======================================================================================

print("\n[3/5] Starting remaining strict-fusion tasks...")

print("Parallel workers:", MAX_WORKERS)

print(
    "Each base model remains single-threaded."
)


resume_start = time.time()

newly_done = 0

failed_tasks = []


if remaining:

    with ThreadPoolExecutor(
        max_workers=MAX_WORKERS
    ) as executor:


        future_to_task = {
            executor.submit(
                run_one_task,
                task
            ): task

            for task in remaining
        }


        for future in as_completed(
            future_to_task
        ):

            task = future_to_task[
                future
            ]


            try:

                result = future.result()

            except Exception as e:

                failed_tasks.append(
                    (
                        task,
                        repr(e)
                    )
                )

                with print_lock:

                    print(
                        "\nFAILED:",
                        task,
                        repr(e)
                    )

                continue


            status = result["status"]


            if status in [
                "COMPLETE",
                "SKIPPED"
            ]:

                newly_done += 1

            else:

                failed_tasks.append(
                    (
                        task,
                        result
                    )
                )


            with print_lock:

                current_valid_estimate = (
                    len(completed)
                    +
                    newly_done
                )

                elapsed_h = (
                    time.time()
                    -
                    resume_start
                ) / 3600.0


                if status == "COMPLETE":

                    print(
                        "\n"
                        f"[{current_valid_estimate}/790] "
                        f"ITER {result['iteration']:02d} | "
                        f"BAL {result['balanced_dataset']:02d} "
                        f"COMPLETE [OK]"
                    )

                    print(
                        "   Task time :",
                        f"{result['minutes']:.1f} min"
                    )

                    print(
                        "   Context AP:",
                        f"{result['Context_AP']:.4f}"
                    )

                    print(
                        "   NT AP     :",
                        f"{result['NT_AP']:.4f}"
                    )

                    print(
                        "   Fusion AP :",
                        f"{result['Fusion_AP']:.4f}"
                    )

                    print(
                        "   Resume elapsed:",
                        f"{elapsed_h:.2f} h"
                    )


                elif status == "SKIPPED":

                    print(
                        f"\nITER {result['iteration']:02d} | "
                        f"BAL {result['balanced_dataset']:02d}: "
                        "already valid -> SKIP"
                    )


                else:

                    print(
                        "\nTASK ERROR:",
                        result
                    )


# ======================================================================================
# 8. FINAL 790-TASK AUDIT
# ======================================================================================

print("\n" + "=" * 124)
print("[4/5] FINAL STRICT-FUSION CHECKPOINT AUDIT")
print("=" * 124)


valid_tasks = []


for iteration in range(1, 11):

    for balanced_dataset in range(79):

        if task_valid(
            iteration,
            balanced_dataset
        ):

            valid_tasks.append(
                (
                    iteration,
                    balanced_dataset
                )
            )


valid_count = len(valid_tasks)


print(
    "Valid strict-fusion tasks:",
    valid_count,
    "/ 790"
)


if failed_tasks:

    print(
        "\nFailed/invalid during this run:",
        len(failed_tasks)
    )

    for item in failed_tasks[:10]:
        print(item)


if valid_count != 790:

    print("\nNOT ALL TASKS COMPLETE.")

    print(
        "Do NOT delete any checkpoint."
    )

    print(
        "Simply rerun THIS SAME CELL."
    )

    print(
        "All valid completed tasks will be skipped."
    )

    raise RuntimeError(
        f"Only {valid_count}/790 tasks valid."
    )


print(
    "\nALL 790 STRICT-FUSION TASKS VERIFIED [OK]"
)


# ======================================================================================
# 9. FINAL AGGREGATION
#
# No target iteration / balanced dataset variable supplied.
# Existing scientific runner enters aggregation-only mode.
# ======================================================================================

print("\n" + "=" * 124)
print("[5/5] RUNNING FINAL STRICT-FUSION AGGREGATION")
print("=" * 124)


aggregation_log = (
    PARALLEL_LOG_DIR /
    "FINAL_AGGREGATION.log"
)


final_env = env_base.copy()

final_env.pop(
    "ASD_CF_FUSION_ITERATION",
    None
)

final_env.pop(
    "ASD_CF_FUSION_BALANCED_DATASET",
    None
)


with open(
    aggregation_log,
    "w",
    encoding="utf-8"
) as log:

    process = subprocess.Popen(
        [
            str(LEGACY_PYTHON),
            "-u",
            str(RUNNER)
        ],
        cwd=str(REPO),
        stdout=subprocess.PIPE,
        stderr=subprocess.STDOUT,
        text=True,
        encoding="utf-8",
        errors="replace",
        bufsize=1,
        env=final_env
    )


    for line in process.stdout:

        log.write(line)
        log.flush()

        print(
            line,
            end=""
        )


    return_code = process.wait()


if return_code != 0:

    raise RuntimeError(
        "\nFinal aggregation failed.\n"
        "All 790 checkpoints remain safe.\n"
        f"Log:\n{aggregation_log}"
    )


# ======================================================================================
# 10. FINAL SCIENTIFIC OUTPUT VERIFICATION
# ======================================================================================

FINAL_SPLIT = (
    RESULT_DIR /
    "FiveModel_split_level_summary.tsv"
)

FINAL_GENE = (
    RESULT_DIR /
    "FiveModel_gene_level_summary.tsv"
)

FINAL_METRICS = (
    RESULT_DIR /
    "StrictFusion_all_7900_fold_metrics.tsv"
)

FINAL_SCORES = (
    RESULT_DIR /
    "StrictFusion_gene_level_OOF_scores.tsv"
)


for p in [
    FINAL_SPLIT,
    FINAL_GENE,
    FINAL_METRICS,
    FINAL_SCORES
]:

    assert p.exists(), (
        f"Missing final result:\n{p}"
    )


metrics = pd.read_csv(
    FINAL_METRICS,
    sep="\t"
)

gene_scores = pd.read_csv(
    FINAL_SCORES,
    sep="\t"
)


assert len(metrics) == 7900

assert len(gene_scores) == 18252

assert (
    gene_scores["Gene_Name"].nunique()
    == 18252
)

assert (
    gene_scores.loc[
        gene_scores["known_gene"] == 1,
        "Gene_Name"
    ].nunique()
    == 190
)


print("\n" + "=" * 124)
print("STRICT CONTEXT + NT-v2 FUSION COMPLETE [OK]")
print("=" * 124)


print("\nFIVE-MODEL SPLIT-LEVEL SUMMARY:")

print(
    pd.read_csv(
        FINAL_SPLIT,
        sep="\t"
    ).to_string(index=False)
)


print("\nFIVE-MODEL GENE-LEVEL SUMMARY:")

print(
    pd.read_csv(
        FINAL_GENE,
        sep="\t"
    ).to_string(index=False)
)


print("\nFINAL INTEGRITY:")

print("Strict fusion tasks     : 790/790 [OK]")
print("Frozen outer folds      : 7900 [OK]")
print("Genes scored            : 18252 [OK]")
print("ASD genes               : 190 [OK]")
print("New PU sampling         : NO")
print("Primary models modified : NO")
print("Raw concatenation       : NO")
print("Sequence controls fused : NO")
print("Strict nested cross-fit : YES")
print("Parallel workers        :", MAX_WORKERS)

print("\nNEXT SCIENTIFIC STAGE:")
print(
    "Paired statistical comparison + "
    "leakage-sensitive robustness analysis."
)

print("=" * 124)
```

```python
# ==========================================================================================
# ASD-ContextFusion
# STAGE: PAIRED STATISTICAL COMPARISON
#
# PRIMARY INFERENCE:
#   Gene-level paired stratified bootstrap
#   - 18,252 unique genes
#   - 190 ASD positives
#   - 18,062 unlabeled genes
#   - same resampled genes applied to both models in each comparison
#   - class counts preserved in every bootstrap replicate
#
# SUPPORTING INFERENCE:
#   10 stochastic-iteration paired exact sign-flip test
#
# IMPORTANT:
#   - NO model fitting
#   - NO new PU sampling
#   - NO tuning
#   - NO feature changes
#   - NO naive 7,900-fold independence assumption
#
# PRIMARY METRIC: AP / AUPRC
# SECONDARY: AUROC
# ==========================================================================================

from pathlib import Path
from itertools import product
import hashlib
import json
import time

import numpy as np
import pandas as pd

from sklearn.metrics import (
    roc_auc_score,
    average_precision_score
)


# ==========================================================================================
# 0. SETTINGS
# ==========================================================================================

ROOT = Path(r"E:\ASD_ContextFusion_Q1")

FUSION_DIR = (
    ROOT /
    "05_results" /
    "proposed" /
    "strict_context_NTv2_fusion"
)

CONTROL_DIR = (
    ROOT /
    "05_results" /
    "proposed" /
    "sequence_controls"
)

OUT_DIR = (
    ROOT /
    "05_results" /
    "proposed" /
    "paired_statistics"
)

ADMIN_DIR = ROOT / "00_admin"

OUT_DIR.mkdir(
    parents=True,
    exist_ok=True
)


# ----------------------------------------------------------------------
# Existing frozen inputs ONLY
# ----------------------------------------------------------------------

FUSION_METRICS_FILE = (
    FUSION_DIR /
    "StrictFusion_all_7900_fold_metrics.tsv"
)

FUSION_GENE_FILE = (
    FUSION_DIR /
    "StrictFusion_gene_level_OOF_scores.tsv"
)

CONTROL_METRICS_FILE = (
    CONTROL_DIR /
    "Context_NTv2_LengthGC_4mer_all_7900_fold_metrics.tsv"
)

CONTROL_GENE_FILE = (
    CONTROL_DIR /
    "Context_NTv2_LengthGC_4mer_gene_level_OOF_scores.tsv"
)

META_FILE = (
    FUSION_DIR /
    "StrictFusion_all_7900_meta_coefficients.tsv"
)


# ----------------------------------------------------------------------
# Reproducible bootstrap
#
# 5000 = publication-grade paired bootstrap.
# Checkpointed every 100 replicates.
# ----------------------------------------------------------------------

N_BOOTSTRAP = 5000
BOOTSTRAP_SEED = 20260830
BOOTSTRAP_BATCH = 100


BOOTSTRAP_REPLICATE_FILE = (
    OUT_DIR /
    "GeneLevel_paired_bootstrap_replicates.tsv.gz"
)

GENE_BOOT_SUMMARY_FILE = (
    OUT_DIR /
    "GeneLevel_paired_bootstrap_summary.tsv"
)

ITERATION_DELTA_FILE = (
    OUT_DIR /
    "IterationLevel_paired_deltas.tsv"
)

ITERATION_STATS_FILE = (
    OUT_DIR /
    "IterationLevel_exact_signflip_summary.tsv"
)

TASK_DESCRIPTIVE_FILE = (
    OUT_DIR /
    "BalancedDataset_descriptive_comparison.tsv"
)

OBSERVED_GENE_FILE = (
    OUT_DIR /
    "GeneLevel_observed_performance.tsv"
)

META_SUMMARY_FILE = (
    OUT_DIR /
    "MetaLearner_coefficient_descriptive_summary.tsv"
)

MANIFEST_FILE = (
    ADMIN_DIR /
    "ASD_ContextFusion_paired_statistics_manifest.json"
)


# ==========================================================================================
# 1. BASIC HELPERS
# ==========================================================================================

def sha256_file(path, chunk_size=1024 * 1024):

    h = hashlib.sha256()

    with open(path, "rb") as f:

        while True:

            chunk = f.read(chunk_size)

            if not chunk:
                break

            h.update(chunk)

    return h.hexdigest()


def normalized_name(x):

    return "".join(
        ch.lower()
        for ch in str(x)
        if ch.isalnum()
    )


def pick_column(
    df,
    candidates,
    label,
    fuzzy_tokens=None
):

    # Exact match first
    for c in candidates:

        if c in df.columns:
            return c


    # Case/punctuation-insensitive match
    normalized_map = {}

    for c in df.columns:

        normalized_map.setdefault(
            normalized_name(c),
            []
        ).append(c)


    for c in candidates:

        key = normalized_name(c)

        if key in normalized_map:

            hits = normalized_map[key]

            if len(hits) == 1:
                return hits[0]


    # Controlled fuzzy fallback
    if fuzzy_tokens is not None:

        hits = []

        for c in df.columns:

            nc = normalized_name(c)

            if all(
                any(
                    token in nc
                    for token in token_group
                )
                for token_group in fuzzy_tokens
            ):

                hits.append(c)

        hits = list(dict.fromkeys(hits))

        if len(hits) == 1:
            return hits[0]


    raise RuntimeError(
        f"\nCould not uniquely identify column for: {label}\n"
        f"Available columns:\n{list(df.columns)}"
    )


def exact_sign_flip_pvalue(differences):

    """
    Exact two-sided paired randomization/sign-flip test.

    With 10 stochastic iterations:
    2^10 = 1024 exact sign assignments.

    Null:
    paired model labels are exchangeable at the stochastic-iteration level.
    """

    d = np.asarray(
        differences,
        dtype=float
    )

    d = d[np.isfinite(d)]

    if len(d) == 0:
        return np.nan

    n = len(d)

    observed = abs(
        d.mean()
    )

    signs = np.asarray(
        list(
            product(
                [-1.0, 1.0],
                repeat=n
            )
        ),
        dtype=float
    )

    null_statistics = np.abs(
        (
            signs *
            d.reshape(1, -1)
        ).mean(axis=1)
    )

    p = np.mean(
        null_statistics
        >=
        observed - 1e-15
    )

    return float(p)


def bh_adjust(pvalues):

    """
    Benjamini-Hochberg FDR adjustment.
    Used only for the family of supporting iteration-level tests.
    """

    p = np.asarray(
        pvalues,
        dtype=float
    )

    m = len(p)

    order = np.argsort(p)

    ranked = p[order]

    adjusted = (
        ranked *
        m /
        np.arange(
            1,
            m + 1
        )
    )

    adjusted = np.minimum.accumulate(
        adjusted[::-1]
    )[::-1]

    adjusted = np.clip(
        adjusted,
        0.0,
        1.0
    )

    out = np.empty_like(
        adjusted
    )

    out[order] = adjusted

    return out


# ==========================================================================================
# 2. REQUIRED FILE AUDIT
# ==========================================================================================

print("=" * 118)
print("ASD-CONTEXTFUSION - PAIRED STATISTICAL COMPARISON")
print("=" * 118)

print("\n[1/9] REQUIRED INPUT AUDIT")


required_files = [
    FUSION_METRICS_FILE,
    FUSION_GENE_FILE,
    CONTROL_METRICS_FILE,
    CONTROL_GENE_FILE
]


for p in required_files:

    assert p.exists(), (
        f"\nMISSING REQUIRED FROZEN RESULT:\n{p}"
    )

    print(
        "[OK]",
        p
    )


# ==========================================================================================
# 3. LOAD FROZEN RESULTS
# ==========================================================================================

print("\n[2/9] LOADING EXISTING RESULTS - NO MODEL FITTING")


fusion_metrics = pd.read_csv(
    FUSION_METRICS_FILE,
    sep="\t"
)

fusion_gene = pd.read_csv(
    FUSION_GENE_FILE,
    sep="\t"
)

control_metrics = pd.read_csv(
    CONTROL_METRICS_FILE,
    sep="\t"
)

control_gene = pd.read_csv(
    CONTROL_GENE_FILE,
    sep="\t"
)


print(
    "Fusion metrics:",
    fusion_metrics.shape
)

print(
    "Control metrics:",
    control_metrics.shape
)

print(
    "Fusion genes:",
    fusion_gene.shape
)

print(
    "Control genes:",
    control_gene.shape
)


print("\nFusion metric columns:")
print(
    list(
        fusion_metrics.columns
    )
)

print("\nControl metric columns:")
print(
    list(
        control_metrics.columns
    )
)

print("\nFusion gene columns:")
print(
    list(
        fusion_gene.columns
    )
)

print("\nControl gene columns:")
print(
    list(
        control_gene.columns
    )
)


# ==========================================================================================
# 4. IDENTIFY / VALIDATE COLUMN INTERFACE
# ==========================================================================================

print("\n[3/9] COLUMN + IDENTITY VALIDATION")


KEYS = [
    "iteration",
    "balanced_dataset",
    "fold"
]


for k in KEYS:

    assert k in fusion_metrics.columns, (
        f"Missing fusion key: {k}"
    )

    assert k in control_metrics.columns, (
        f"Missing control key: {k}"
    )


# ----------------------------------------------------------------------
# Split-level metric columns
# ----------------------------------------------------------------------

FC_AUC = pick_column(
    fusion_metrics,
    [
        "Context_AUROC"
    ],
    "Fusion-file Context AUROC"
)

FC_AP = pick_column(
    fusion_metrics,
    [
        "Context_AP"
    ],
    "Fusion-file Context AP"
)

FN_AUC = pick_column(
    fusion_metrics,
    [
        "NT_AUROC"
    ],
    "Fusion-file NT AUROC"
)

FN_AP = pick_column(
    fusion_metrics,
    [
        "NT_AP"
    ],
    "Fusion-file NT AP"
)

FF_AUC = pick_column(
    fusion_metrics,
    [
        "Fusion_AUROC"
    ],
    "Fusion AUROC",
    fuzzy_tokens=[
        ["fusion"],
        ["auroc", "rocauc"]
    ]
)

FF_AP = pick_column(
    fusion_metrics,
    [
        "Fusion_AP"
    ],
    "Fusion AP"
)


CC_AUC = pick_column(
    control_metrics,
    [
        "Context_AUROC"
    ],
    "Control-file Context AUROC"
)

CC_AP = pick_column(
    control_metrics,
    [
        "Context_AP"
    ],
    "Control-file Context AP"
)

CN_AUC = pick_column(
    control_metrics,
    [
        "NT_AUROC"
    ],
    "Control-file NT AUROC"
)

CN_AP = pick_column(
    control_metrics,
    [
        "NT_AP"
    ],
    "Control-file NT AP"
)

CL_AUC = pick_column(
    control_metrics,
    [
        "LengthGC_AUROC",
        "Length_GC_AUROC"
    ],
    "Length+GC AUROC",
    fuzzy_tokens=[
        ["lengthgc", "lengthandgc"],
        ["auroc", "rocauc"]
    ]
)

CL_AP = pick_column(
    control_metrics,
    [
        "LengthGC_AP",
        "Length_GC_AP"
    ],
    "Length+GC AP",
    fuzzy_tokens=[
        ["lengthgc", "lengthandgc"],
        ["ap", "averageprecision"]
    ]
)

C4_AUC = pick_column(
    control_metrics,
    [
        "Fourmer_AUROC",
        "FourMer_AUROC",
        "4mer_AUROC"
    ],
    "4-mer AUROC",
    fuzzy_tokens=[
        ["fourmer", "4mer"],
        ["auroc", "rocauc"]
    ]
)

C4_AP = pick_column(
    control_metrics,
    [
        "Fourmer_AP",
        "FourMer_AP",
        "4mer_AP"
    ],
    "4-mer AP",
    fuzzy_tokens=[
        ["fourmer", "4mer"],
        ["ap", "averageprecision"]
    ]
)


# ----------------------------------------------------------------------
# Gene-level probability columns
# ----------------------------------------------------------------------

FG_CONTEXT = pick_column(
    fusion_gene,
    [
        "Context_probability"
    ],
    "Gene Context probability"
)

FG_NT = pick_column(
    fusion_gene,
    [
        "NT_probability"
    ],
    "Gene NT probability"
)

FG_FUSION = pick_column(
    fusion_gene,
    [
        "Fusion_probability"
    ],
    "Gene Fusion probability"
)


CG_LENGTH = pick_column(
    control_gene,
    [
        "LengthGC_probability",
        "Length_GC_probability",
        "LengthGC_score",
        "Length_GC_score"
    ],
    "Gene Length+GC probability",
    fuzzy_tokens=[
        ["lengthgc", "lengthandgc"],
        ["probability", "prob", "score"]
    ]
)

CG_FOURMER = pick_column(
    control_gene,
    [
        "Fourmer_probability",
        "FourMer_probability",
        "4mer_probability",
        "Fourmer_score",
        "FourMer_score",
        "4mer_score"
    ],
    "Gene 4-mer probability",
    fuzzy_tokens=[
        ["fourmer", "4mer"],
        ["probability", "prob", "score"]
    ]
)


# Optional duplicate Context / NT columns in four-model gene file
CG_CONTEXT = None
CG_NT = None

if "Context_probability" in control_gene.columns:
    CG_CONTEXT = "Context_probability"

if "NT_probability" in control_gene.columns:
    CG_NT = "NT_probability"


print("\nResolved split columns:")
print(
    {
        "Context_AUROC": FC_AUC,
        "Context_AP": FC_AP,
        "NT_AUROC": FN_AUC,
        "NT_AP": FN_AP,
        "Fusion_AUROC": FF_AUC,
        "Fusion_AP": FF_AP,
        "LengthGC_AUROC": CL_AUC,
        "LengthGC_AP": CL_AP,
        "Fourmer_AUROC": C4_AUC,
        "Fourmer_AP": C4_AP
    }
)

print("\nResolved gene columns:")
print(
    {
        "Context": FG_CONTEXT,
        "NTv2": FG_NT,
        "Fusion": FG_FUSION,
        "LengthGC": CG_LENGTH,
        "Fourmer": CG_FOURMER
    }
)


# ==========================================================================================
# 5. SPLIT-LEVEL INTEGRITY
# ==========================================================================================

print("\n[4/9] SPLIT-LEVEL FROZEN-FOLD INTEGRITY")


assert len(
    fusion_metrics
) == 7900

assert len(
    control_metrics
) == 7900


assert not fusion_metrics[
    KEYS
].duplicated().any()

assert not control_metrics[
    KEYS
].duplicated().any()


fusion_keys = set(
    map(
        tuple,
        fusion_metrics[
            KEYS
        ].to_numpy()
    )
)

control_keys = set(
    map(
        tuple,
        control_metrics[
            KEYS
        ].to_numpy()
    )
)


assert (
    fusion_keys
    ==
    control_keys
), "Frozen fold identities differ between fusion and controls."


split = fusion_metrics[
    KEYS +
    [
        FC_AUC,
        FC_AP,
        FN_AUC,
        FN_AP,
        FF_AUC,
        FF_AP
    ]
].copy()


split = split.rename(
    columns={
        FC_AUC: "Context_AUROC",
        FC_AP: "Context_AP",
        FN_AUC: "NTv2_AUROC",
        FN_AP: "NTv2_AP",
        FF_AUC: "Fusion_AUROC",
        FF_AP: "Fusion_AP"
    }
)


control_part = control_metrics[
    KEYS +
    [
        CC_AUC,
        CC_AP,
        CN_AUC,
        CN_AP,
        CL_AUC,
        CL_AP,
        C4_AUC,
        C4_AP
    ]
].copy()


control_part = control_part.rename(
    columns={
        CC_AUC: "ControlCheck_Context_AUROC",
        CC_AP: "ControlCheck_Context_AP",
        CN_AUC: "ControlCheck_NTv2_AUROC",
        CN_AP: "ControlCheck_NTv2_AP",
        CL_AUC: "LengthGC_AUROC",
        CL_AP: "LengthGC_AP",
        C4_AUC: "Fourmer_AUROC",
        C4_AP: "Fourmer_AP"
    }
)


split = split.merge(
    control_part,
    on=KEYS,
    how="inner",
    validate="one_to_one"
)


assert len(
    split
) == 7900


# Context + NT values must be exactly the same scientific folds/results
for a, b in [
    (
        "Context_AUROC",
        "ControlCheck_Context_AUROC"
    ),
    (
        "Context_AP",
        "ControlCheck_Context_AP"
    ),
    (
        "NTv2_AUROC",
        "ControlCheck_NTv2_AUROC"
    ),
    (
        "NTv2_AP",
        "ControlCheck_NTv2_AP"
    )
]:

    assert np.allclose(
        split[a].to_numpy(float),
        split[b].to_numpy(float),
        rtol=0.0,
        atol=1e-12,
        equal_nan=True
    ), (
        f"Frozen result mismatch: {a} vs {b}"
    )


split = split.drop(
    columns=[
        "ControlCheck_Context_AUROC",
        "ControlCheck_Context_AP",
        "ControlCheck_NTv2_AUROC",
        "ControlCheck_NTv2_AP"
    ]
)


print(
    "Frozen split identities:",
    len(split),
    "[OK]"
)

print(
    "Iterations:",
    split["iteration"].nunique()
)

print(
    "Balanced datasets:",
    split[
        [
            "iteration",
            "balanced_dataset"
        ]
    ].drop_duplicates().shape[0]
)

print(
    "Folds per balanced dataset:",
    split.groupby(
        [
            "iteration",
            "balanced_dataset"
        ]
    )["fold"].nunique().min(),
    "-",
    split.groupby(
        [
            "iteration",
            "balanced_dataset"
        ]
    )["fold"].nunique().max()
)


# ==========================================================================================
# 6. GENE-LEVEL INTEGRITY + RECOMPUTE ALL FIVE MODELS
# ==========================================================================================

print("\n[5/9] GENE-LEVEL INTEGRITY + PERFORMANCE RECOMPUTATION")


for df_name, df in [
    (
        "fusion_gene",
        fusion_gene
    ),
    (
        "control_gene",
        control_gene
    )
]:

    assert "Gene_Name" in df.columns
    assert "known_gene" in df.columns

    assert len(
        df
    ) == 18252

    assert (
        df[
            "Gene_Name"
        ].nunique()
        == 18252
    )

    assert (
        df.loc[
            df["known_gene"] == 1,
            "Gene_Name"
        ].nunique()
        == 190
    )


gene = fusion_gene[
    [
        "Gene_Name",
        "known_gene",
        FG_CONTEXT,
        FG_NT,
        FG_FUSION
    ]
].copy()


gene = gene.rename(
    columns={
        FG_CONTEXT: "Context",
        FG_NT: "NTv2",
        FG_FUSION: "Fusion"
    }
)


control_gene_part = control_gene[
    [
        "Gene_Name",
        "known_gene",
        CG_LENGTH,
        CG_FOURMER
    ]
    +
    (
        [CG_CONTEXT]
        if CG_CONTEXT is not None
        else []
    )
    +
    (
        [CG_NT]
        if CG_NT is not None
        else []
    )
].copy()


rename_control_gene = {
    CG_LENGTH: "LengthGC",
    CG_FOURMER: "Fourmer"
}

if CG_CONTEXT is not None:
    rename_control_gene[
        CG_CONTEXT
    ] = "ControlCheck_Context"

if CG_NT is not None:
    rename_control_gene[
        CG_NT
    ] = "ControlCheck_NTv2"


control_gene_part = (
    control_gene_part
    .rename(
        columns=
            rename_control_gene
    )
)


gene = gene.merge(
    control_gene_part,
    on=[
        "Gene_Name",
        "known_gene"
    ],
    how="inner",
    validate="one_to_one"
)


assert len(
    gene
) == 18252


if "ControlCheck_Context" in gene.columns:

    assert np.allclose(
        gene["Context"],
        gene["ControlCheck_Context"],
        rtol=0.0,
        atol=1e-12
    )

    gene = gene.drop(
        columns=[
            "ControlCheck_Context"
        ]
    )


if "ControlCheck_NTv2" in gene.columns:

    assert np.allclose(
        gene["NTv2"],
        gene["ControlCheck_NTv2"],
        rtol=0.0,
        atol=1e-12
    )

    gene = gene.drop(
        columns=[
            "ControlCheck_NTv2"
        ]
    )


MODEL_COLUMNS = {
    "Context": "Context",
    "NTv2": "NTv2",
    "LengthGC": "LengthGC",
    "Fourmer": "Fourmer",
    "Fusion": "Fusion"
}


for col in MODEL_COLUMNS.values():

    assert np.isfinite(
        gene[col].to_numpy(
            dtype=float
        )
    ).all()

    assert (
        gene[col].between(
            0.0,
            1.0
        )
    ).all()


y_gene = gene[
    "known_gene"
].to_numpy(
    dtype=int
)


assert int(
    y_gene.sum()
) == 190

assert len(
    y_gene
) == 18252


observed_rows = []


for model, col in MODEL_COLUMNS.items():

    scores = gene[
        col
    ].to_numpy(
        dtype=float
    )

    auc = roc_auc_score(
        y_gene,
        scores
    )

    ap = average_precision_score(
        y_gene,
        scores
    )

    observed_rows.append(
        {
            "Model": model,
            "Gene_level_AUROC": auc,
            "Gene_level_AP": ap
        }
    )


observed_gene = pd.DataFrame(
    observed_rows
)


# ----------------------------------------------------------------------
# Verify against already-frozen published console values.
# Tolerance reflects six-decimal printed summaries.
# ----------------------------------------------------------------------

EXPECTED_GENE = {
    "Context": {
        "AUROC": 0.947076,
        "AP": 0.182366
    },
    "NTv2": {
        "AUROC": 0.899628,
        "AP": 0.099593
    },
    "LengthGC": {
        "AUROC": 0.832049,
        "AP": 0.052922
    },
    "Fourmer": {
        "AUROC": 0.779947,
        "AP": 0.039135
    },
    "Fusion": {
        "AUROC": 0.951591,
        "AP": 0.196525
    }
}


for _, row in observed_gene.iterrows():

    model = row["Model"]

    assert np.isclose(
        row["Gene_level_AUROC"],
        EXPECTED_GENE[model]["AUROC"],
        atol=5e-6,
        rtol=0.0
    ), (
        f"{model} gene AUROC does not reproduce frozen result."
    )

    assert np.isclose(
        row["Gene_level_AP"],
        EXPECTED_GENE[model]["AP"],
        atol=5e-6,
        rtol=0.0
    ), (
        f"{model} gene AP does not reproduce frozen result."
    )


observed_gene.to_csv(
    OBSERVED_GENE_FILE,
    sep="\t",
    index=False
)


print("\nRecomputed five-model gene-level performance:")
print(
    observed_gene.to_string(
        index=False
    )
)

print(
    "\nGene-level frozen results reproduced [OK]"
)


# ==========================================================================================
# 7. DEFINE PRESPECIFIED PAIRED COMPARISONS
# ==========================================================================================

COMPARISONS = [
    (
        "Fusion_vs_Context",
        "Fusion",
        "Context"
    ),
    (
        "Fusion_vs_NTv2",
        "Fusion",
        "NTv2"
    ),
    (
        "NTv2_vs_LengthGC",
        "NTv2",
        "LengthGC"
    ),
    (
        "NTv2_vs_Fourmer",
        "NTv2",
        "Fourmer"
    )
]


# ==========================================================================================
# 8. SUPPORTING: ITERATION-LEVEL PAIRED EXACT TEST
#
# IMPORTANT:
# We DO NOT run a primary p-value test treating 7,900 folds as independent.
# Instead:
#  - summarize 790 folds inside each stochastic iteration
#  - obtain 10 paired stochastic-iteration performance values
#  - exact paired sign-flip over those 10 differences
# ==========================================================================================

print("\n[6/9] STOCHASTIC-ITERATION PAIRED INFERENCE")


iteration_mean = (
    split
    .groupby(
        "iteration",
        as_index=False
    )
    .agg(
        Context_AUROC=(
            "Context_AUROC",
            "mean"
        ),
        Context_AP=(
            "Context_AP",
            "mean"
        ),
        NTv2_AUROC=(
            "NTv2_AUROC",
            "mean"
        ),
        NTv2_AP=(
            "NTv2_AP",
            "mean"
        ),
        LengthGC_AUROC=(
            "LengthGC_AUROC",
            "mean"
        ),
        LengthGC_AP=(
            "LengthGC_AP",
            "mean"
        ),
        Fourmer_AUROC=(
            "Fourmer_AUROC",
            "mean"
        ),
        Fourmer_AP=(
            "Fourmer_AP",
            "mean"
        ),
        Fusion_AUROC=(
            "Fusion_AUROC",
            "mean"
        ),
        Fusion_AP=(
            "Fusion_AP",
            "mean"
        )
    )
)


assert len(
    iteration_mean
) == 10


iteration_delta_rows = []

iteration_summary_rows = []


for comparison_name, model_a, model_b in COMPARISONS:

    for metric in [
        "AP",
        "AUROC"
    ]:

        col_a = (
            f"{model_a}_{metric}"
        )

        col_b = (
            f"{model_b}_{metric}"
        )

        delta = (
            iteration_mean[col_a]
            -
            iteration_mean[col_b]
        ).to_numpy(
            dtype=float
        )


        for iteration_value, delta_value in zip(
            iteration_mean["iteration"],
            delta
        ):

            iteration_delta_rows.append(
                {
                    "comparison":
                        comparison_name,

                    "metric":
                        metric,

                    "iteration":
                        int(
                            iteration_value
                        ),

                    "delta_A_minus_B":
                        float(
                            delta_value
                        )
                }
            )


        p_exact = exact_sign_flip_pvalue(
            delta
        )


        iteration_summary_rows.append(
            {
                "comparison":
                    comparison_name,

                "metric":
                    metric,

                "model_A":
                    model_a,

                "model_B":
                    model_b,

                "n_stochastic_iterations":
                    len(delta),

                "mean_delta":
                    float(
                        np.mean(delta)
                    ),

                "sd_delta":
                    float(
                        np.std(
                            delta,
                            ddof=1
                        )
                    ),

                "median_delta":
                    float(
                        np.median(
                            delta
                        )
                    ),

                "min_delta":
                    float(
                        np.min(
                            delta
                        )
                    ),

                "max_delta":
                    float(
                        np.max(
                            delta
                        )
                    ),

                "positive_iterations":
                    int(
                        np.sum(
                            delta > 0
                        )
                    ),

                "negative_iterations":
                    int(
                        np.sum(
                            delta < 0
                        )
                    ),

                "zero_iterations":
                    int(
                        np.sum(
                            delta == 0
                        )
                    ),

                "exact_two_sided_signflip_p":
                    p_exact
            }
        )


iteration_deltas = pd.DataFrame(
    iteration_delta_rows
)

iteration_stats = pd.DataFrame(
    iteration_summary_rows
)


iteration_stats[
    "BH_FDR_q_across_8_supporting_tests"
] = bh_adjust(
    iteration_stats[
        "exact_two_sided_signflip_p"
    ].to_numpy(
        float
    )
)


iteration_deltas.to_csv(
    ITERATION_DELTA_FILE,
    sep="\t",
    index=False
)

iteration_stats.to_csv(
    ITERATION_STATS_FILE,
    sep="\t",
    index=False
)


print("\nIteration-level exact paired results:")
print(
    iteration_stats.to_string(
        index=False
    )
)


# ==========================================================================================
# 9. DESCRIPTIVE BALANCED-DATASET CONSISTENCY
#
# 790 balanced datasets = descriptive ONLY.
# Not treated as 790 independent biological replicates for primary inference.
# ==========================================================================================

task_mean = (
    split
    .groupby(
        [
            "iteration",
            "balanced_dataset"
        ],
        as_index=False
    )
    .mean(
        numeric_only=True
    )
)


assert len(
    task_mean
) == 790


task_rows = []


for comparison_name, model_a, model_b in COMPARISONS:

    for metric in [
        "AP",
        "AUROC"
    ]:

        delta = (
            task_mean[
                f"{model_a}_{metric}"
            ]
            -
            task_mean[
                f"{model_b}_{metric}"
            ]
        ).to_numpy(
            dtype=float
        )


        task_rows.append(
            {
                "comparison":
                    comparison_name,

                "metric":
                    metric,

                "n_balanced_datasets":
                    len(delta),

                "mean_delta":
                    float(
                        np.mean(delta)
                    ),

                "median_delta":
                    float(
                        np.median(delta)
                    ),

                "sd_delta":
                    float(
                        np.std(
                            delta,
                            ddof=1
                        )
                    ),

                "fraction_A_gt_B":
                    float(
                        np.mean(
                            delta > 0
                        )
                    ),

                "fraction_A_ge_B":
                    float(
                        np.mean(
                            delta >= 0
                        )
                    )
            }
        )


task_descriptive = pd.DataFrame(
    task_rows
)


task_descriptive.to_csv(
    TASK_DESCRIPTIVE_FILE,
    sep="\t",
    index=False
)


# ==========================================================================================
# 10. PRIMARY: GENE-LEVEL PAIRED STRATIFIED BOOTSTRAP
#
# Resampling:
#  - ASD genes sampled WITH replacement from the 190 ASD genes
#  - unlabeled genes sampled WITH replacement from the 18,062 unlabeled genes
#  - class counts therefore stay fixed in each bootstrap
#  - SAME gene indices used for every model => paired comparison
#
# This preserves the benchmark prevalence and avoids unpaired CIs.
# ==========================================================================================

print("\n[7/9] PRIMARY GENE-LEVEL PAIRED STRATIFIED BOOTSTRAP")


positive_idx = np.flatnonzero(
    y_gene == 1
)

unlabeled_idx = np.flatnonzero(
    y_gene == 0
)


assert len(
    positive_idx
) == 190

assert len(
    unlabeled_idx
) == 18062


MODEL_SCORE_ARRAYS = {
    model:
        gene[col].to_numpy(
            dtype=float
        )
    for model, col in MODEL_COLUMNS.items()
}


BOOT_COLUMNS = [
    "replicate"
]

for comparison_name, _, _ in COMPARISONS:

    BOOT_COLUMNS.extend(
        [
            f"{comparison_name}__AP_delta",
            f"{comparison_name}__AUROC_delta"
        ]
    )


# ----------------------------------------------------------------------
# Safe resume
# ----------------------------------------------------------------------

if BOOTSTRAP_REPLICATE_FILE.exists():

    existing_boot = pd.read_csv(
        BOOTSTRAP_REPLICATE_FILE,
        sep="\t",
        compression="gzip"
    )


    missing_boot_cols = set(
        BOOT_COLUMNS
    ) - set(
        existing_boot.columns
    )


    if missing_boot_cols:

        raise RuntimeError(
            "\nExisting bootstrap checkpoint has incompatible columns:\n"
            f"{missing_boot_cols}\n"
            "Do NOT overwrite blindly. Inspect it first."
        )


    existing_boot = (
        existing_boot[
            BOOT_COLUMNS
        ]
        .sort_values(
            "replicate"
        )
        .reset_index(
            drop=True
        )
    )


    assert existing_boot[
        "replicate"
    ].is_unique


    expected_replicates = np.arange(
        len(
            existing_boot
        )
    )


    assert np.array_equal(
        existing_boot[
            "replicate"
        ].to_numpy(
            dtype=int
        ),
        expected_replicates
    ), (
        "Bootstrap checkpoint replicate sequence is not contiguous."
    )


    if len(
        existing_boot
    ) > N_BOOTSTRAP:

        raise RuntimeError(
            "Existing bootstrap checkpoint contains more replicates "
            "than current N_BOOTSTRAP setting."
        )


    boot_df = (
        existing_boot
        .copy()
    )


    start_replicate = len(
        boot_df
    )


    print(
        "Existing valid bootstrap checkpoint:",
        start_replicate,
        "/",
        N_BOOTSTRAP
    )


else:

    boot_df = pd.DataFrame(
        columns=
            BOOT_COLUMNS
    )

    start_replicate = 0

    print(
        "No prior bootstrap checkpoint. Starting at replicate 0."
    )


bootstrap_start_time = time.time()


for batch_start in range(
    start_replicate,
    N_BOOTSTRAP,
    BOOTSTRAP_BATCH
):

    batch_end = min(
        batch_start
        +
        BOOTSTRAP_BATCH,
        N_BOOTSTRAP
    )

    batch_rows = []


    for b in range(
        batch_start,
        batch_end
    ):

        # --------------------------------------------------------------
        # Replicate-specific deterministic RNG.
        #
        # This means interrupted runs can resume at any replicate
        # WITHOUT reproducing all previous random draws.
        # --------------------------------------------------------------

        rng = np.random.default_rng(
            np.random.SeedSequence(
                [
                    BOOTSTRAP_SEED,
                    b
                ]
            )
        )


        sampled_positive = rng.choice(
            positive_idx,
            size=len(
                positive_idx
            ),
            replace=True
        )


        sampled_unlabeled = rng.choice(
            unlabeled_idx,
            size=len(
                unlabeled_idx
            ),
            replace=True
        )


        sampled_idx = np.concatenate(
            [
                sampled_positive,
                sampled_unlabeled
            ]
        )


        y_b = y_gene[
            sampled_idx
        ]


        metrics_b = {}


        # Compute each model ONCE per bootstrap replicate
        for model in MODEL_SCORE_ARRAYS:

            score_b = (
                MODEL_SCORE_ARRAYS[
                    model
                ][
                    sampled_idx
                ]
            )


            metrics_b[
                (
                    model,
                    "AUROC"
                )
            ] = roc_auc_score(
                y_b,
                score_b
            )


            metrics_b[
                (
                    model,
                    "AP"
                )
            ] = average_precision_score(
                y_b,
                score_b
            )


        row = {
            "replicate":
                b
        }


        for comparison_name, model_a, model_b in COMPARISONS:

            row[
                f"{comparison_name}__AP_delta"
            ] = (
                metrics_b[
                    (
                        model_a,
                        "AP"
                    )
                ]
                -
                metrics_b[
                    (
                        model_b,
                        "AP"
                    )
                ]
            )


            row[
                f"{comparison_name}__AUROC_delta"
            ] = (
                metrics_b[
                    (
                        model_a,
                        "AUROC"
                    )
                ]
                -
                metrics_b[
                    (
                        model_b,
                        "AUROC"
                    )
                ]
            )


        batch_rows.append(
            row
        )


    batch_df = pd.DataFrame(
        batch_rows
    )


    boot_df = pd.concat(
        [
            boot_df,
            batch_df
        ],
        ignore_index=True
    )


    # Save complete checkpoint after every batch
    boot_df.to_csv(
        BOOTSTRAP_REPLICATE_FILE,
        sep="\t",
        index=False,
        compression="gzip"
    )


    elapsed_minutes = (
        time.time()
        -
        bootstrap_start_time
    ) / 60.0


    print(
        f"Bootstrap {batch_end:4d}/{N_BOOTSTRAP} "
        f"complete | session elapsed {elapsed_minutes:.1f} min"
    )


assert len(
    boot_df
) == N_BOOTSTRAP

assert boot_df[
    "replicate"
].nunique() == N_BOOTSTRAP


# ==========================================================================================
# 11. BOOTSTRAP SUMMARY
# ==========================================================================================

bootstrap_summary_rows = []


# observed model metric lookup
observed_lookup = (
    observed_gene
    .set_index(
        "Model"
    )
)


for comparison_name, model_a, model_b in COMPARISONS:

    for metric in [
        "AP",
        "AUROC"
    ]:

        if metric == "AP":

            observed_a = float(
                observed_lookup.loc[
                    model_a,
                    "Gene_level_AP"
                ]
            )

            observed_b = float(
                observed_lookup.loc[
                    model_b,
                    "Gene_level_AP"
                ]
            )

        else:

            observed_a = float(
                observed_lookup.loc[
                    model_a,
                    "Gene_level_AUROC"
                ]
            )

            observed_b = float(
                observed_lookup.loc[
                    model_b,
                    "Gene_level_AUROC"
                ]
            )


        observed_delta = (
            observed_a
            -
            observed_b
        )


        bootstrap_delta = boot_df[
            f"{comparison_name}__{metric}_delta"
        ].to_numpy(
            dtype=float
        )


        ci_low = float(
            np.quantile(
                bootstrap_delta,
                0.025
            )
        )

        ci_high = float(
            np.quantile(
                bootstrap_delta,
                0.975
            )
        )


        support_positive = float(
            np.mean(
                bootstrap_delta > 0
            )
        )


        relative_improvement = np.nan

        if (
            metric == "AP"
            and
            observed_b != 0
        ):

            relative_improvement = (
                observed_delta
                /
                observed_b
            )


        bootstrap_summary_rows.append(
            {
                "comparison":
                    comparison_name,

                "metric":
                    metric,

                "model_A":
                    model_a,

                "model_B":
                    model_b,

                "model_A_observed":
                    observed_a,

                "model_B_observed":
                    observed_b,

                "observed_delta_A_minus_B":
                    observed_delta,

                "bootstrap_mean_delta":
                    float(
                        np.mean(
                            bootstrap_delta
                        )
                    ),

                "bootstrap_median_delta":
                    float(
                        np.median(
                            bootstrap_delta
                        )
                    ),

                "bootstrap_SD_delta":
                    float(
                        np.std(
                            bootstrap_delta,
                            ddof=1
                        )
                    ),

                "CI95_percentile_low":
                    ci_low,

                "CI95_percentile_high":
                    ci_high,

                "bootstrap_fraction_delta_gt_0":
                    support_positive,

                "CI_excludes_zero":
                    bool(
                        (
                            ci_low > 0
                        )
                        or
                        (
                            ci_high < 0
                        )
                    ),

                "relative_improvement_A_over_B":
                    relative_improvement,

                "n_bootstrap":
                    N_BOOTSTRAP,

                "bootstrap_seed":
                    BOOTSTRAP_SEED
            }
        )


bootstrap_summary = pd.DataFrame(
    bootstrap_summary_rows
)


bootstrap_summary.to_csv(
    GENE_BOOT_SUMMARY_FILE,
    sep="\t",
    index=False
)


print("\nPRIMARY GENE-LEVEL PAIRED BOOTSTRAP:")
print(
    bootstrap_summary.to_string(
        index=False
    )
)


# ==========================================================================================
# 12. META-LEARNER COEFFICIENTS - DESCRIPTIVE ONLY
#
# No coefficient "significance" claim here.
# ==========================================================================================

print("\n[8/9] META-LEARNER COEFFICIENT STABILITY")


meta_summary = None


if META_FILE.exists():

    meta = pd.read_csv(
        META_FILE,
        sep="\t"
    )


    print(
        "Meta coefficient rows:",
        len(meta)
    )

    print(
        "Meta columns:",
        list(
            meta.columns
        )
    )


    parameter_candidates = {
        "Context_coefficient":
            [
                "Context_coefficient"
            ],

        "NT_coefficient":
            [
                "NT_coefficient"
            ],

        "Intercept":
            [
                "Intercept"
            ]
    }


    meta_rows = []


    for parameter_name, candidates in parameter_candidates.items():

        col = pick_column(
            meta,
            candidates,
            parameter_name
        )


        values = meta[
            col
        ].to_numpy(
            dtype=float
        )


        meta_rows.append(
            {
                "Parameter":
                    parameter_name,

                "N":
                    len(values),

                "Mean":
                    float(
                        np.mean(
                            values
                        )
                    ),

                "SD":
                    float(
                        np.std(
                            values,
                            ddof=1
                        )
                    ),

                "Median":
                    float(
                        np.median(
                            values
                        )
                    ),

                "Min":
                    float(
                        np.min(
                            values
                        )
                    ),

                "Max":
                    float(
                        np.max(
                            values
                        )
                    ),

                "Fraction_positive":
                    float(
                        np.mean(
                            values > 0
                        )
                    )
            }
        )


    meta_summary = pd.DataFrame(
        meta_rows
    )


    meta_summary.to_csv(
        META_SUMMARY_FILE,
        sep="\t",
        index=False
    )


    print(
        meta_summary.to_string(
            index=False
        )
    )


else:

    print(
        "Meta file not found -> coefficient descriptive stage skipped."
    )


# ==========================================================================================
# 13. FINAL INTEGRITY + MANIFEST
# ==========================================================================================

print("\n[9/9] SAVE REPRODUCIBILITY MANIFEST")


manifest = {
    "project":
        "ASD-ContextFusion",

    "stage":
        "paired_statistical_comparison",

    "scientific_status":
        "analysis-only; no model fitting or tuning",

    "primary_metric":
        "Average Precision / AUPRC",

    "secondary_metric":
        "AUROC",

    "primary_inference":
        (
            "Gene-level paired stratified bootstrap on 18,252 "
            "aggregated OOF gene scores; positive and unlabeled "
            "classes resampled separately with replacement and the "
            "same sampled genes used for all compared models."
        ),

    "supporting_inference":
        (
            "Exact paired sign-flip test over 10 stochastic-iteration "
            "mean performance differences."
        ),

    "explicitly_not_used_as_primary":
        (
            "Naive inferential testing treating all 7,900 frozen "
            "folds as independent observations."
        ),

    "comparisons":
        [
            {
                "name":
                    x[0],
                "model_A":
                    x[1],
                "model_B":
                    x[2]
            }
            for x in COMPARISONS
        ],

    "gene_count":
        int(
            len(
                gene
            )
        ),

    "ASD_gene_count":
        int(
            gene[
                "known_gene"
            ].sum()
        ),

    "unlabeled_gene_count":
        int(
            (
                gene[
                    "known_gene"
                ] == 0
            ).sum()
        ),

    "frozen_fold_count":
        int(
            len(
                split
            )
        ),

    "stochastic_iterations":
        int(
            iteration_mean[
                "iteration"
            ].nunique()
        ),

    "balanced_dataset_count_total":
        int(
            task_mean.shape[
                0
            ]
        ),

    "bootstrap_replicates":
        int(
            N_BOOTSTRAP
        ),

    "bootstrap_seed":
        int(
            BOOTSTRAP_SEED
        ),

    "new_PU_sampling":
        False,

    "models_refit":
        False,

    "models_tuned":
        False,

    "input_files":
        {
            str(
                FUSION_METRICS_FILE
            ):
                sha256_file(
                    FUSION_METRICS_FILE
                ),

            str(
                FUSION_GENE_FILE
            ):
                sha256_file(
                    FUSION_GENE_FILE
                ),

            str(
                CONTROL_METRICS_FILE
            ):
                sha256_file(
                    CONTROL_METRICS_FILE
                ),

            str(
                CONTROL_GENE_FILE
            ):
                sha256_file(
                    CONTROL_GENE_FILE
                )
        },

    "outputs":
        [
            str(
                OBSERVED_GENE_FILE
            ),
            str(
                BOOTSTRAP_REPLICATE_FILE
            ),
            str(
                GENE_BOOT_SUMMARY_FILE
            ),
            str(
                ITERATION_DELTA_FILE
            ),
            str(
                ITERATION_STATS_FILE
            ),
            str(
                TASK_DESCRIPTIVE_FILE
            ),
            str(
                META_SUMMARY_FILE
            )
            if META_FILE.exists()
            else None
        ]
}


with open(
    MANIFEST_FILE,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        manifest,
        f,
        indent=2
    )


# ==========================================================================================
# 14. FINAL CONSOLE SUMMARY
# ==========================================================================================

print("\n" + "=" * 118)
print("PAIRED STATISTICAL COMPARISON COMPLETE [OK]")
print("=" * 118)


print("\nPRIMARY RESULT TABLE:")
print(
    bootstrap_summary[
        [
            "comparison",
            "metric",
            "model_A_observed",
            "model_B_observed",
            "observed_delta_A_minus_B",
            "CI95_percentile_low",
            "CI95_percentile_high",
            "bootstrap_fraction_delta_gt_0",
            "CI_excludes_zero"
        ]
    ].to_string(
        index=False
    )
)


print("\nSUPPORTING ITERATION-LEVEL EXACT TEST:")
print(
    iteration_stats[
        [
            "comparison",
            "metric",
            "mean_delta",
            "positive_iterations",
            "negative_iterations",
            "exact_two_sided_signflip_p",
            "BH_FDR_q_across_8_supporting_tests"
        ]
    ].to_string(
        index=False
    )
)


fusion_context_ap = bootstrap_summary.loc[
    (
        bootstrap_summary[
            "comparison"
        ] == "Fusion_vs_Context"
    )
    &
    (
        bootstrap_summary[
            "metric"
        ] == "AP"
    )
].iloc[0]


fusion_context_auc = bootstrap_summary.loc[
    (
        bootstrap_summary[
            "comparison"
        ] == "Fusion_vs_Context"
    )
    &
    (
        bootstrap_summary[
            "metric"
        ] == "AUROC"
    )
].iloc[0]


print("\nKEY PRESPECIFIED COMPARISON: FUSION vs CONTEXT")

print(
    "Gene AP delta:",
    f"{fusion_context_ap['observed_delta_A_minus_B']:.6f}",
    "| 95% CI:",
    (
        f"[{fusion_context_ap['CI95_percentile_low']:.6f}, "
        f"{fusion_context_ap['CI95_percentile_high']:.6f}]"
    )
)

print(
    "Gene AUROC delta:",
    f"{fusion_context_auc['observed_delta_A_minus_B']:.6f}",
    "| 95% CI:",
    (
        f"[{fusion_context_auc['CI95_percentile_low']:.6f}, "
        f"{fusion_context_auc['CI95_percentile_high']:.6f}]"
    )
)


print("\nINTEGRITY:")
print("Models rerun              : NO")
print("New PU sampling           : NO")
print("Hyperparameter tuning     : NO")
print("Frozen outer folds        : 7900")
print("Unique genes              : 18252")
print("ASD genes                 : 190")
print("Primary inference         : paired gene bootstrap")
print("Naive 7900-fold p-values  : NOT USED")


print("\nSaved to:")
print(OUT_DIR)

print("\nManifest:")
print(MANIFEST_FILE)


print("\nNEXT SCIENTIFIC STAGE:")
print(
    "Leakage-sensitive robustness analysis using "
    "source-defined Dhindsa/Mantis feature families."
)

print("=" * 118)
```

```python
# ==========================================================================================
# ASD-ContextFusion
# STAGE 8A — DHINDSA SOURCE-PROVENANCE / FEATURE-FAMILY LOCK
#
# PURPOSE:
#   Identify the authors' SOURCE-DEFINED feature families BEFORE robustness modeling.
#
# IMPORTANT:
#   - NO model fitting
#   - NO new PU sampling
#   - NO tuning
#   - NO feature selection by outcome
#   - NO keyword-based predictor classification
#
# We use source-defined terms ONLY to locate repository evidence/code/data sources.
# We DO NOT assign a predictor to a family merely because its NAME contains a keyword.
#
# Outputs will be used to lock:
#   A. Developmental-expression family
#   B. Genetic / constraint family
#   C. Full Context benchmark
#   D. Leakage-sensitive seed-derived GO/PPI/disease-context family
#
# Run in normal Anaconda Jupyter Python.
# ==========================================================================================

from pathlib import Path
import pandas as pd
import numpy as np
import hashlib
import json
import re


# ==========================================================================================
# 0. PATHS
# ==========================================================================================

ROOT = Path(r"E:\ASD_ContextFusion_Q1")

REPO = (
    ROOT /
    "01_raw" /
    "dhindsa_2025" /
    "Mantis-ml-NDD"
)

OUT_DIR = (
    ROOT /
    "02_interim" /
    "dhindsa_source_feature_family_lock"
)

ADMIN_DIR = (
    ROOT /
    "00_admin"
)

OUT_DIR.mkdir(
    parents=True,
    exist_ok=True
)

assert REPO.exists(), (
    f"Repository missing:\n{REPO}"
)


# ==========================================================================================
# 1. HELPERS
# ==========================================================================================

def sha256_file(path, chunk_size=1024 * 1024):

    h = hashlib.sha256()

    with open(path, "rb") as f:

        while True:

            chunk = f.read(chunk_size)

            if not chunk:
                break

            h.update(chunk)

    return h.hexdigest()


def safe_read_text(path):

    for encoding in [
        "utf-8",
        "utf-8-sig",
        "latin-1"
    ]:

        try:

            return path.read_text(
                encoding=encoding,
                errors="strict"
            )

        except Exception:

            pass

    try:

        return path.read_text(
            encoding="utf-8",
            errors="replace"
        )

    except Exception:

        return None


def safe_header(path):

    """
    Read ONLY table header.
    Does not load full data.
    """

    name = path.name.lower()

    try:

        if (
            name.endswith(".tsv")
            or
            name.endswith(".tsv.gz")
        ):

            df = pd.read_csv(
                path,
                sep="\t",
                nrows=0
            )

        elif (
            name.endswith(".csv")
            or
            name.endswith(".csv.gz")
        ):

            df = pd.read_csv(
                path,
                sep=",",
                nrows=0
            )

        elif name.endswith(".txt"):

            # Attempt tab-delimited text only.
            df = pd.read_csv(
                path,
                sep="\t",
                nrows=0
            )

        else:

            return None

        return [
            str(x)
            for x in df.columns
        ]

    except Exception:

        return None


# ==========================================================================================
# 2. LOCATE FINAL LOCKED 316-PREDICTOR FILE
# ==========================================================================================

print("=" * 124)
print("ASD-CONTEXTFUSION — STAGE 8A SOURCE-PROVENANCE LOCK")
print("=" * 124)

print("\n[1/9] LOCATING FINAL LOCKED CONTEXT PREDICTOR LIST")


predictor_candidates = sorted(
    ROOT.rglob(
        "dhindsa_316_context_predictors.tsv"
    )
)


if not predictor_candidates:

    raise RuntimeError(
        "\nCould not find dhindsa_316_context_predictors.tsv"
    )


valid_predictor_files = []


for p in predictor_candidates:

    try:

        df = pd.read_csv(
            p,
            sep="\t"
        )

        if "feature" in df.columns:

            values = (
                df["feature"]
                .astype(str)
                .str.strip()
            )

        elif df.shape[1] == 1:

            values = (
                df.iloc[:, 0]
                .astype(str)
                .str.strip()
            )

        else:

            continue


        if (
            len(values) == 316
            and
            values.nunique() == 316
        ):

            valid_predictor_files.append(
                (
                    p,
                    values.tolist()
                )
            )

    except Exception:

        pass


if not valid_predictor_files:

    raise RuntimeError(
        "\nNo valid 316-feature predictor file found."
    )


# Prefer 03_features path if multiple valid copies exist.
valid_predictor_files = sorted(
    valid_predictor_files,
    key=lambda x:
        (
            0
            if "03_features" in str(x[0])
            else 1,
            len(str(x[0]))
        )
)


PREDICTOR_FILE, predictors = (
    valid_predictor_files[0]
)


predictor_set = set(
    predictors
)


assert len(
    predictors
) == 316

assert len(
    predictor_set
) == 316


print("Predictor file:")
print(PREDICTOR_FILE)

print("\nPredictors:", len(predictors), "[OK]")

print("\nSHA256:")
print(
    sha256_file(
        PREDICTOR_FILE
    )
)


# ==========================================================================================
# 3. FINAL MASTER IDENTITY AUDIT
# ==========================================================================================

print("\n[2/9] FINAL MULTIMODAL MASTER AUDIT")


master_candidates = sorted(
    ROOT.rglob(
        "dhindsa_NTv2_row_preserved_master.tsv"
    )
)


valid_master_files = []


for p in master_candidates:

    try:

        header = safe_header(
            p
        )

        if header is None:
            continue

        if (
            "dhindsa_row_id" in header
            and
            "Gene_Name" in header
            and
            "known_gene" in header
            and
            "NT_row_index" in header
        ):

            valid_master_files.append(
                p
            )

    except Exception:

        pass


if not valid_master_files:

    raise RuntimeError(
        "\nCould not locate final row-preserved multimodal master."
    )


valid_master_files = sorted(
    valid_master_files,
    key=lambda p:
        (
            0
            if "03_features" in str(p)
            else 1,
            len(str(p))
        )
)


MASTER_FILE = (
    valid_master_files[0]
)


master_identity = pd.read_csv(
    MASTER_FILE,
    sep="\t",
    usecols=[
        "dhindsa_row_id",
        "Gene_Name",
        "known_gene",
        "NT_row_index"
    ]
)


assert len(
    master_identity
) == 18253

assert (
    master_identity[
        "Gene_Name"
    ].nunique()
    == 18252
)

assert (
    master_identity.loc[
        master_identity[
            "known_gene"
        ] == 1,
        "Gene_Name"
    ].nunique()
    == 190
)


tmsb = master_identity.loc[
    master_identity[
        "Gene_Name"
    ] == "TMSB15B"
]


assert len(
    tmsb
) == 2

assert (
    tmsb[
        "NT_row_index"
    ].nunique()
    == 1
)


print("Master:")
print(MASTER_FILE)

print("Rows        :", len(master_identity))
print("Unique genes:", master_identity["Gene_Name"].nunique())
print("ASD genes   :", master_identity.loc[master_identity["known_gene"] == 1, "Gene_Name"].nunique())
print("TMSB15B rows:", len(tmsb))

print("Master identity [OK]")


# ==========================================================================================
# 4. SOURCE-DEFINED CATEGORIES FROM AUTHORS' REPOSITORY
#
# These expressions are used ONLY to FIND source documentation/code.
# They are NOT used to classify final predictor names.
# ==========================================================================================

print("\n[3/9] AUTHORS' SOURCE-DEFINED EVIDENCE SEARCH")


SOURCE_PATTERNS = {

    "developing_cortex_scRNA":
        [
            r"single[-_ ]?cell",
            r"\bscrna\b",
            r"fetal",
            r"developing human cortex"
        ],

    "human_protein_atlas_expression":
        [
            r"human protein atlas",
            r"protein[-_ ]?atlas",
            r"\bhpa\b"
        ],

    "GeVIR":
        [
            r"\bgevir\b"
        ],

    "GO_functional_context":
        [
            r"gene ontology",
            r"\bgo[_ -]?term",
            r"\bgo[_-]",
            r"go enrichment"
        ],

    "protein_protein_interaction":
        [
            r"protein[- ]protein",
            r"\bppi\b",
            r"\binweb\b"
        ],

    "genic_constraint_intolerance":
        [
            r"constraint",
            r"intolerance",
            r"\bloeuf\b",
            r"\bpli\b",
            r"\brvis\b",
            r"\bgnomad\b",
            r"\bexac\b"
        ]
}


SOURCE_EXTENSIONS = {
    ".py",
    ".yaml",
    ".yml",
    ".nf",
    ".md",
    ".txt",
    ".json"
}


source_files = []


for p in REPO.rglob("*"):

    try:

        if (
            p.is_file()
            and
            p.suffix.lower()
            in SOURCE_EXTENSIONS
        ):

            source_files.append(
                p
            )

    except OSError:

        pass


print(
    "Repository text/code files:",
    len(source_files)
)


source_hits = []


for path in source_files:

    text = safe_read_text(
        path
    )

    if text is None:
        continue


    relative_path = str(
        path.relative_to(
            REPO
        )
    )


    for line_no, line in enumerate(
        text.splitlines(),
        start=1
    ):

        lower_line = (
            line.lower()
        )


        for family, patterns in SOURCE_PATTERNS.items():

            matched_patterns = []


            for pattern in patterns:

                if re.search(
                    pattern,
                    lower_line,
                    flags=re.I
                ):

                    matched_patterns.append(
                        pattern
                    )


            if matched_patterns:

                source_hits.append(
                    {
                        "source_defined_family":
                            family,

                        "repository_file":
                            relative_path,

                        "line_number":
                            line_no,

                        "matched_locator_patterns":
                            " | ".join(
                                matched_patterns
                            ),

                        "source_line":
                            line.strip()
                    }
                )


source_hits_df = pd.DataFrame(
    source_hits
)


SOURCE_HITS_FILE = (
    OUT_DIR /
    "repository_source_defined_evidence.tsv"
)


source_hits_df.to_csv(
    SOURCE_HITS_FILE,
    sep="\t",
    index=False
)


print(
    "Source-defined evidence hits:",
    len(source_hits_df)
)


if len(
    source_hits_df
) > 0:

    source_summary = (
        source_hits_df
        .groupby(
            "source_defined_family"
        )
        .agg(
            evidence_lines=(
                "source_line",
                "size"
            ),
            repository_files=(
                "repository_file",
                "nunique"
            )
        )
        .reset_index()
        .sort_values(
            "source_defined_family"
        )
    )

    print("\nSource evidence summary:")

    print(
        source_summary.to_string(
            index=False
        )
    )


# ==========================================================================================
# 5. README — AUTHOR-DEFINED ENHANCEMENTS
# ==========================================================================================

print("\n[4/9] README AUTHOR-DEFINED FEATURE ENHANCEMENTS")


README = (
    REPO /
    "README.md"
)


readme_lines_df = pd.DataFrame()


if README.exists():

    readme_text = safe_read_text(
        README
    )

    readme_rows = []


    for line_no, line in enumerate(
        readme_text.splitlines(),
        start=1
    ):

        if any(
            re.search(
                pattern,
                line,
                flags=re.I
            )
            for patterns
            in SOURCE_PATTERNS.values()
            for pattern
            in patterns
        ):

            readme_rows.append(
                {
                    "line_number":
                        line_no,

                    "text":
                        line.strip()
                }
            )


    readme_lines_df = pd.DataFrame(
        readme_rows
    )


    README_EVIDENCE_FILE = (
        OUT_DIR /
        "README_source_defined_feature_evidence.tsv"
    )


    readme_lines_df.to_csv(
        README_EVIDENCE_FILE,
        sep="\t",
        index=False
    )


    if len(
        readme_lines_df
    ):

        print(
            readme_lines_df.to_string(
                index=False
            )
        )


else:

    print(
        "README.md not found."
    )


# ==========================================================================================
# 6. EXACT DATA-SOURCE REFERENCES FROM CODE
#
# Extract URLs/file-like references WITHOUT assigning families.
# ==========================================================================================

print("\n[5/9] CODE DATA-SOURCE / FILE REFERENCE AUDIT")


URL_REGEX = re.compile(
    r"https?://[^\s\"\'\)\]]+",
    flags=re.I
)


FILE_REGEX = re.compile(
    r"""['"]([^'"]+\.(?:csv|tsv|txt|gz|xlsx|xls|bed|gff|gtf|parquet|pkl|pickle))['"]""",
    flags=re.I
)


reference_rows = []


for path in source_files:

    text = safe_read_text(
        path
    )

    if text is None:
        continue


    relative_path = str(
        path.relative_to(
            REPO
        )
    )


    for line_no, line in enumerate(
        text.splitlines(),
        start=1
    ):

        urls = URL_REGEX.findall(
            line
        )

        files = FILE_REGEX.findall(
            line
        )


        for value in urls:

            reference_rows.append(
                {
                    "repository_file":
                        relative_path,

                    "line_number":
                        line_no,

                    "reference_type":
                        "URL",

                    "reference":
                        value,

                    "source_line":
                        line.strip()
                }
            )


        for value in files:

            reference_rows.append(
                {
                    "repository_file":
                        relative_path,

                    "line_number":
                        line_no,

                    "reference_type":
                        "FILE",

                    "reference":
                        value,

                    "source_line":
                        line.strip()
                }
            )


references_df = pd.DataFrame(
    reference_rows
)


REFERENCES_FILE = (
    OUT_DIR /
    "repository_data_source_references.tsv"
)


references_df.to_csv(
    REFERENCES_FILE,
    sep="\t",
    index=False
)


print(
    "Explicit source/file references:",
    len(references_df)
)


# ==========================================================================================
# 7. EXACT TABULAR HEADER PROVENANCE
#
# KEY PRINCIPLE:
# A predictor is linked to a source table ONLY when its exact final predictor
# name appears in that source table's header.
#
# NO keyword-based assignment.
# ==========================================================================================

print("\n[6/9] EXACT TABULAR HEADER -> PREDICTOR PROVENANCE")


SCAN_ROOTS = [
    REPO,

    ROOT /
    "02_interim" /
    "dhindsa_repository_audit",

    ROOT /
    "02_interim" /
    "dhindsa_feature_audit"
]


tabular_files = []


for scan_root in SCAN_ROOTS:

    if not scan_root.exists():
        continue


    for p in scan_root.rglob("*"):

        try:

            if not p.is_file():
                continue


            lower_name = (
                p.name.lower()
            )


            if (
                lower_name.endswith(".tsv")
                or
                lower_name.endswith(".tsv.gz")
                or
                lower_name.endswith(".csv")
                or
                lower_name.endswith(".csv.gz")
                or
                lower_name.endswith(".txt")
            ):

                tabular_files.append(
                    p
                )

        except OSError:

            pass


# Deduplicate resolved paths
unique_tabular_files = []

seen_tabular = set()


for p in tabular_files:

    try:

        key = str(
            p.resolve()
        )

    except Exception:

        key = str(
            p
        )


    if key not in seen_tabular:

        seen_tabular.add(
            key
        )

        unique_tabular_files.append(
            p
        )


header_provenance_rows = []


for path in unique_tabular_files:

    # Avoid using the already-final 316 predictor file itself
    # as fake provenance evidence.
    if path.resolve() == PREDICTOR_FILE.resolve():
        continue


    # Avoid final model/result folders if encountered indirectly.
    path_string_lower = str(
        path
    ).lower()


    if any(
        blocked in path_string_lower
        for blocked in [
            "\\04_models\\",
            "\\05_results\\",
            "\\08_logs\\"
        ]
    ):

        continue


    header = safe_header(
        path
    )


    if header is None:
        continue


    overlap = sorted(
        predictor_set.intersection(
            set(
                header
            )
        )
    )


    if not overlap:
        continue


    if path.is_relative_to(
        REPO
    ):

        provenance_root = (
            "Dhindsa_repository"
        )

        relative_path = str(
            path.relative_to(
                REPO
            )
        )

    else:

        provenance_root = (
            "project_interim_audit"
        )

        try:

            relative_path = str(
                path.relative_to(
                    ROOT
                )
            )

        except Exception:

            relative_path = str(
                path
            )


    header_provenance_rows.append(
        {
            "provenance_root":
                provenance_root,

            "source_table":
                relative_path,

            "table_column_count":
                len(
                    header
                ),

            "exact_final_predictor_overlap_count":
                len(
                    overlap
                ),

            "exact_final_predictors":
                " | ".join(
                    overlap
                )
        }
    )


header_provenance_df = pd.DataFrame(
    header_provenance_rows
)


HEADER_PROVENANCE_FILE = (
    OUT_DIR /
    "exact_table_header_predictor_provenance.tsv"
)


header_provenance_df.to_csv(
    HEADER_PROVENANCE_FILE,
    sep="\t",
    index=False
)


print(
    "Tables with exact final-predictor overlap:",
    len(
        header_provenance_df
    )
)


if len(
    header_provenance_df
):

    print("\nTOP SOURCE TABLE OVERLAPS:")

    print(
        header_provenance_df
        .sort_values(
            "exact_final_predictor_overlap_count",
            ascending=False
        )
        .head(
            30
        )
        .to_string(
            index=False
        )
    )


# ==========================================================================================
# 8. EXACT PREDICTOR LITERAL OCCURRENCES IN AUTHORS' CODE
#
# Again: this is evidence only.
# We do NOT classify by the predictor name.
# ==========================================================================================

print("\n[7/9] EXACT FINAL-PREDICTOR OCCURRENCES IN AUTHORS' CODE")


CODE_EXTENSIONS = {
    ".py",
    ".yaml",
    ".yml",
    ".nf",
    ".json"
}


code_files = [
    p
    for p in source_files
    if p.suffix.lower()
    in CODE_EXTENSIONS
]


code_cache = {}


for path in code_files:

    text = safe_read_text(
        path
    )

    if text is not None:

        code_cache[
            path
        ] = text.splitlines()


literal_rows = []


for predictor in predictors:

    for path, lines in code_cache.items():

        relative_path = str(
            path.relative_to(
                REPO
            )
        )


        for line_no, line in enumerate(
            lines,
            start=1
        ):

            if predictor in line:

                literal_rows.append(
                    {
                        "predictor":
                            predictor,

                        "repository_file":
                            relative_path,

                        "line_number":
                            line_no,

                        "source_line":
                            line.strip()
                    }
                )


literal_df = pd.DataFrame(
    literal_rows
)


LITERAL_FILE = (
    OUT_DIR /
    "exact_predictor_code_literal_evidence.tsv"
)


literal_df.to_csv(
    LITERAL_FILE,
    sep="\t",
    index=False
)


if len(
    literal_df
):

    code_covered_predictors = set(
        literal_df[
            "predictor"
        ]
    )

else:

    code_covered_predictors = set()


print(
    "Final predictors with direct literal code evidence:",
    len(
        code_covered_predictors
    ),
    "/ 316"
)


# ==========================================================================================
# 9. BUILD PREDICTOR EVIDENCE STATUS
#
# IMPORTANT:
# This does NOT classify predictors into biological families.
# It only records whether exact provenance evidence has been located.
# ==========================================================================================

print("\n[8/9] FINAL PROVENANCE-EVIDENCE MATRIX")


table_predictor_sources = {}


if len(
    header_provenance_df
):

    for _, row in header_provenance_df.iterrows():

        predictor_names = [
            x.strip()
            for x in str(
                row[
                    "exact_final_predictors"
                ]
            ).split("|")
            if x.strip()
        ]


        for predictor in predictor_names:

            table_predictor_sources.setdefault(
                predictor,
                []
            ).append(
                str(
                    row[
                        "source_table"
                    ]
                )
            )


code_predictor_sources = {}


if len(
    literal_df
):

    for predictor, group in literal_df.groupby(
        "predictor"
    ):

        code_predictor_sources[
            predictor
        ] = sorted(
            group[
                "repository_file"
            ]
            .astype(str)
            .unique()
            .tolist()
        )


evidence_rows = []


for predictor in predictors:

    table_sources = sorted(
        set(
            table_predictor_sources.get(
                predictor,
                []
            )
        )
    )

    code_sources = sorted(
        set(
            code_predictor_sources.get(
                predictor,
                []
            )
        )
    )


    evidence_rows.append(
        {
            "predictor":
                predictor,

            "exact_source_table_evidence":
                bool(
                    table_sources
                ),

            "exact_code_literal_evidence":
                bool(
                    code_sources
                ),

            "source_tables":
                " | ".join(
                    table_sources
                ),

            "source_code_files":
                " | ".join(
                    code_sources
                ),

            "provenance_evidence_found":
                bool(
                    table_sources
                    or
                    code_sources
                )
        }
    )


evidence_matrix = pd.DataFrame(
    evidence_rows
)


EVIDENCE_MATRIX_FILE = (
    OUT_DIR /
    "final_316_predictor_provenance_evidence_matrix.tsv"
)


evidence_matrix.to_csv(
    EVIDENCE_MATRIX_FILE,
    sep="\t",
    index=False
)


resolved_count = int(
    evidence_matrix[
        "provenance_evidence_found"
    ].sum()
)


unresolved = evidence_matrix.loc[
    ~evidence_matrix[
        "provenance_evidence_found"
    ]
].copy()


UNRESOLVED_FILE = (
    OUT_DIR /
    "unresolved_final_predictors.tsv"
)


unresolved.to_csv(
    UNRESOLVED_FILE,
    sep="\t",
    index=False
)


print(
    "Predictors with direct provenance evidence:",
    resolved_count,
    "/ 316"
)

print(
    "Predictors still unresolved:",
    len(
        unresolved
    )
)


# ==========================================================================================
# 10. SOURCE-EVIDENCE SUMMARY BY REPOSITORY FILE
# ==========================================================================================

file_evidence_rows = []


if len(
    literal_df
):

    for file_name, group in literal_df.groupby(
        "repository_file"
    ):

        file_evidence_rows.append(
            {
                "repository_file":
                    file_name,

                "exact_predictors_mentioned":
                    group[
                        "predictor"
                    ].nunique(),

                "predictors":
                    " | ".join(
                        sorted(
                            group[
                                "predictor"
                            ]
                            .unique()
                            .tolist()
                        )
                    )
            }
        )


file_evidence_df = pd.DataFrame(
    file_evidence_rows
)


FILE_EVIDENCE_FILE = (
    OUT_DIR /
    "repository_file_predictor_evidence_summary.tsv"
)


file_evidence_df.to_csv(
    FILE_EVIDENCE_FILE,
    sep="\t",
    index=False
)


# ==========================================================================================
# 11. REPRODUCIBILITY MANIFEST
# ==========================================================================================

manifest = {

    "project":
        "ASD-ContextFusion",

    "stage":
        "8A_source_provenance_feature_family_lock",

    "purpose":
        (
            "Identify authors' source-defined Dhindsa/Mantis "
            "feature provenance before leakage-sensitive robustness modeling."
        ),

    "model_fitting":
        False,

    "new_PU_sampling":
        False,

    "feature_selection_by_outcome":
        False,

    "keyword_based_predictor_classification":
        False,

    "important_rule":
        (
            "Source-defined search expressions are used only to locate "
            "repository documentation/code evidence. Final predictors are "
            "not assigned to families from their names."
        ),

    "locked_context_predictor_count":
        316,

    "predictor_file":
        str(
            PREDICTOR_FILE
        ),

    "predictor_file_sha256":
        sha256_file(
            PREDICTOR_FILE
        ),

    "master_file":
        str(
            MASTER_FILE
        ),

    "multimodal_rows":
        18253,

    "unique_genes":
        18252,

    "ASD_genes":
        190,

    "TMSB15B_policy":
        (
            "Two context rows retained; both reference "
            "one gene-level NT embedding."
        ),

    "source_defined_locator_categories":
        list(
            SOURCE_PATTERNS.keys()
        ),

    "predictors_with_direct_provenance_evidence":
        resolved_count,

    "predictors_unresolved":
        int(
            len(
                unresolved
            )
        ),

    "outputs":
        {
            "repository_source_defined_evidence":
                str(
                    SOURCE_HITS_FILE
                ),

            "repository_data_source_references":
                str(
                    REFERENCES_FILE
                ),

            "exact_table_header_predictor_provenance":
                str(
                    HEADER_PROVENANCE_FILE
                ),

            "exact_predictor_code_literal_evidence":
                str(
                    LITERAL_FILE
                ),

            "final_predictor_provenance_matrix":
                str(
                    EVIDENCE_MATRIX_FILE
                ),

            "unresolved_predictors":
                str(
                    UNRESOLVED_FILE
                ),

            "repository_file_predictor_summary":
                str(
                    FILE_EVIDENCE_FILE
                )
        }
}


MANIFEST_FILE = (
    ADMIN_DIR /
    "ASD_ContextFusion_stage8A_source_provenance_manifest.json"
)


with open(
    MANIFEST_FILE,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        manifest,
        f,
        indent=2
    )


# ==========================================================================================
# 12. FINAL REPORT
# ==========================================================================================

print("\n[9/9] STAGE 8A AUDIT COMPLETE")


print("\n" + "=" * 124)
print("ASD-CONTEXTFUSION — SOURCE-PROVENANCE LOCK AUDIT COMPLETE [OK]")
print("=" * 124)


print("\nFINAL LOCKED INPUT:")
print("Context predictors :", len(predictors))
print("Multimodal rows     :", len(master_identity))
print("Unique genes        :", master_identity["Gene_Name"].nunique())
print("ASD genes           :", master_identity.loc[master_identity["known_gene"] == 1, "Gene_Name"].nunique())
print("TMSB15B rows        :", len(tmsb))


print("\nPROVENANCE COVERAGE:")
print("Predictors total                 :", len(predictors))
print("With direct provenance evidence :", resolved_count)
print("Still unresolved                 :", len(unresolved))


if len(
    header_provenance_df
):

    print("\nTOP EXACT SOURCE-TABLE OVERLAPS:")

    print(
        header_provenance_df
        .sort_values(
            "exact_final_predictor_overlap_count",
            ascending=False
        )
        .head(
            20
        )
        [
            [
                "provenance_root",
                "source_table",
                "exact_final_predictor_overlap_count"
            ]
        ]
        .to_string(
            index=False
        )
    )


if len(
    source_hits_df
):

    print("\nSOURCE-DEFINED FAMILY EVIDENCE COUNTS:")

    print(
        source_hits_df
        .groupby(
            "source_defined_family"
        )
        .agg(
            evidence_lines=(
                "source_line",
                "size"
            ),
            repository_files=(
                "repository_file",
                "nunique"
            )
        )
        .reset_index()
        .to_string(
            index=False
        )
    )


if len(
    file_evidence_df
):

    print("\nTOP REPOSITORY FILES WITH EXACT PREDICTOR EVIDENCE:")

    print(
        file_evidence_df
        .sort_values(
            "exact_predictors_mentioned",
            ascending=False
        )
        .head(
            20
        )
        [
            [
                "repository_file",
                "exact_predictors_mentioned"
            ]
        ]
        .to_string(
            index=False
        )
    )


print("\nSAVED AUDIT DIRECTORY:")
print(OUT_DIR)

print("\nMANIFEST:")
print(MANIFEST_FILE)


print("\nIMPORTANT:")
print(
    "NO robustness model has been fitted yet."
)

print(
    "NO feature family has been assigned from predictor-name keywords."
)

print(
    "The next step is to use the authors' exact source provenance "
    "from this audit to freeze the feature-family manifest."
)


print("\nNEXT:")
print(
    "Paste the final console output from this cell here. "
    "Then we will lock the exact developmental-expression, "
    "genetic/constraint, and leakage-sensitive feature families "
    "BEFORE any robustness result is observed."
)

print("=" * 124)
```

```python
# ==========================================================================================
# ASD-ContextFusion
# STAGE 8A.1 — CORRECTED SOURCE-DEFINED FEATURE MANIFEST LOCK
# VERSION: lineage-aware categorical transformation handling
#
# SCIENTIFIC PURPOSE
# ------------------
# Freeze the authors' SOURCE-DEFINED Context partition BEFORE robustness modeling:
#
#   A. Authors' generic feature class
#   B. Authors' filtered-by-disease feature class
#
# IMPORTANT CORRECTION
# --------------------
# Some FINAL model predictors do not exist verbatim in complete_feature_table.tsv
# because the authors' EDA creates them by one-hot encoding categorical raw features.
#
# Specifically, the pipeline documents:
#   ProteinAtlas_gene_expr_levels
#       -> ProteinAtlas_gene_expr_levels_High
#       -> ProteinAtlas_gene_expr_levels_Low
#       -> ProteinAtlas_gene_expr_levels_Medium
#       -> ProteinAtlas_gene_expr_levels_Not_detected
#
# This code does NOT assume that mapping.
# It PROVES the lineage from:
#   1. raw complete table
#   2. raw source-class table
#   3. raw category values
#   4. regenerated pandas dummy names
#   5. exact processed-feature-table columns
#
# ONLY THEN does a derived predictor inherit its raw parent's source class.
#
# NO:
#   - model fitting
#   - XGBoost
#   - NT model
#   - fusion
#   - new PU sampling
#   - outcome-based selection
#   - predictor-name keyword grouping
#
# If ANY lineage remains unresolved, STOP.
# ==========================================================================================


from pathlib import Path
import hashlib
import json

import numpy as np
import pandas as pd


# ==========================================================================================
# 0. PATHS
# ==========================================================================================

ROOT = Path(r"E:\ASD_ContextFusion_Q1")

ADMIN_DIR = (
    ROOT /
    "00_admin"
)

FEATURE_DIR = (
    ROOT /
    "03_features" /
    "multimodal_master"
)

EXACT190_ROOT = (
    ROOT /
    "02_interim" /
    "exact190_rebuild" /
    "asd_monoallelic"
)

COMPILED_DIR = (
    EXACT190_ROOT /
    "data" /
    "compiled_feature_tables"
)

PROCESSED_DIR = (
    EXACT190_ROOT /
    "processed-feature-tables"
)

OUT_DIR = (
    ROOT /
    "02_interim" /
    "dhindsa_source_feature_family_lock"
)

OUT_DIR.mkdir(
    parents=True,
    exist_ok=True
)


FINAL_PREDICTOR_FILE = (
    FEATURE_DIR /
    "dhindsa_316_context_predictors.tsv"
)

LOCKED_SEED_FILE = (
    ADMIN_DIR /
    "dhindsa_2025_monoallelic_ASD_190.tsv"
)

GENERIC_FILE = (
    COMPILED_DIR /
    "generic_feature_table.tsv"
)

FILTERED_FILE = (
    COMPILED_DIR /
    "filtered_by_disease_feature_table.tsv"
)

COMPLETE_FILE = (
    COMPILED_DIR /
    "complete_feature_table.tsv"
)

PREFERRED_PROCESSED_FILE = (
    PROCESSED_DIR /
    "processed_feature_table.tsv"
)


# ==========================================================================================
# 1. HELPERS
# ==========================================================================================

def sha256_file(
    path,
    chunk_size=1024 * 1024
):

    h = hashlib.sha256()

    with open(
        path,
        "rb"
    ) as f:

        while True:

            chunk = f.read(
                chunk_size
            )

            if not chunk:
                break

            h.update(
                chunk
            )

    return h.hexdigest()


def read_header(
    path
):

    return list(
        pd.read_csv(
            path,
            sep="\t",
            nrows=0
        ).columns
    )


def read_feature_list(
    path
):

    df = pd.read_csv(
        path,
        sep="\t"
    )

    if "feature" in df.columns:

        values = (
            df[
                "feature"
            ]
            .dropna()
            .astype(str)
            .str.strip()
            .tolist()
        )

    elif df.shape[1] == 1:

        values = (
            df.iloc[
                :,
                0
            ]
            .dropna()
            .astype(str)
            .str.strip()
            .tolist()
        )

    else:

        raise RuntimeError(
            "Could not identify feature column in:\n"
            f"{path}"
        )


    if len(
        values
    ) != len(
        set(
            values
        )
    ):

        raise RuntimeError(
            "Duplicate predictors detected in locked feature file."
        )


    return values


def read_locked_seed_set(
    path
):

    # ------------------------------------------------------------------
    # Headered attempt
    # ------------------------------------------------------------------

    df = pd.read_csv(
        path,
        sep="\t"
    )


    candidates = [
        "Gene_Name",
        "Gene",
        "gene",
        "gene_symbol",
        "symbol"
    ]


    for col in candidates:

        if col in df.columns:

            genes = set(
                df[
                    col
                ]
                .dropna()
                .astype(str)
                .str.strip()
            )

            genes.discard(
                ""
            )

            if len(
                genes
            ) == 190:

                return genes


    # ------------------------------------------------------------------
    # Single-column headered fallback
    # ------------------------------------------------------------------

    if df.shape[1] == 1:

        genes = set(
            df.iloc[
                :,
                0
            ]
            .dropna()
            .astype(str)
            .str.strip()
        )

        genes.discard(
            ""
        )

        if len(
            genes
        ) == 190:

            return genes


    # ------------------------------------------------------------------
    # Headerless fallback
    # ------------------------------------------------------------------

    df = pd.read_csv(
        path,
        sep="\t",
        header=None
    )

    genes = set(
        df.iloc[
            :,
            0
        ]
        .dropna()
        .astype(str)
        .str.strip()
    )

    genes.discard(
        ""
    )


    if len(
        genes
    ) != 190:

        raise RuntimeError(
            "Could not reproduce exact 190-gene locked ASD seed set. "
            f"Observed: {len(genes)}"
        )


    return genes


# ==========================================================================================
# 2. REQUIRED FILE AUDIT
# ==========================================================================================

print(
    "=" * 126
)

print(
    "ASD-CONTEXTFUSION — STAGE 8A.1 CORRECTED SOURCE-DEFINED FEATURE LOCK"
)

print(
    "=" * 126
)


print(
    "\n[1/10] REQUIRED FILE AUDIT"
)


required_files = [
    FINAL_PREDICTOR_FILE,
    LOCKED_SEED_FILE,
    GENERIC_FILE,
    FILTERED_FILE,
    COMPLETE_FILE
]


for path in required_files:

    if not path.exists():

        raise RuntimeError(
            f"Required file missing:\n{path}"
        )

    print(
        "[OK]",
        path
    )


# ==========================================================================================
# 3. LOCATE EXACT190 PROCESSED FEATURE TABLE
# ==========================================================================================

print(
    "\n[2/10] EXACT190 PROCESSED TABLE"
)


if PREFERRED_PROCESSED_FILE.exists():

    PROCESSED_FILE = (
        PREFERRED_PROCESSED_FILE
    )

else:

    processed_candidates = sorted(
        EXACT190_ROOT.rglob(
            "processed_feature_table.tsv"
        )
    )


    if len(
        processed_candidates
    ) != 1:

        raise RuntimeError(
            "Could not uniquely identify exact190 "
            "processed_feature_table.tsv.\n"
            f"Candidates: {processed_candidates}"
        )


    PROCESSED_FILE = (
        processed_candidates[
            0
        ]
    )


# Hard safety:
# never silently use pre-exact190 result.
assert (
    "exact190_rebuild"
    in str(
        PROCESSED_FILE
    ).lower()
)


print(
    "Processed table:"
)

print(
    PROCESSED_FILE
)


# ==========================================================================================
# 4. FINAL 316-PREDICTOR LOCK
# ==========================================================================================

print(
    "\n[3/10] FINAL 316-PREDICTOR LOCK"
)


final_predictors = (
    read_feature_list(
        FINAL_PREDICTOR_FILE
    )
)

final_set = set(
    final_predictors
)


assert len(
    final_predictors
) == 316

assert len(
    final_set
) == 316

assert "known_gene" not in final_set

assert "Gene_Name" not in final_set


print(
    "Final predictors:",
    len(
        final_predictors
    ),
    "[OK]"
)

print(
    "SHA256:",
    sha256_file(
        FINAL_PREDICTOR_FILE
    )
)


# ==========================================================================================
# 5. VERIFY EXACT 190 ASD SEED IDENTITY
# ==========================================================================================

print(
    "\n[4/10] EXACT 190 ASD SEED IDENTITY"
)


complete_identity = pd.read_csv(
    COMPLETE_FILE,
    sep="\t",
    usecols=[
        "Gene_Name",
        "known_gene"
    ]
)


complete_identity[
    "Gene_Name"
] = (
    complete_identity[
        "Gene_Name"
    ]
    .astype(str)
    .str.strip()
)


complete_seed_set = set(
    complete_identity.loc[
        complete_identity[
            "known_gene"
        ] == 1,
        "Gene_Name"
    ]
)


locked_seed_set = (
    read_locked_seed_set(
        LOCKED_SEED_FILE
    )
)


print(
    "Complete-table positives:",
    len(
        complete_seed_set
    )
)

print(
    "Locked ASD seeds:",
    len(
        locked_seed_set
    )
)


assert len(
    complete_seed_set
) == 190

assert len(
    locked_seed_set
) == 190


if (
    complete_seed_set
    !=
    locked_seed_set
):

    raise RuntimeError(
        "\nExact ASD seed identity mismatch.\n"
        "Only complete table:\n"
        f"{sorted(complete_seed_set - locked_seed_set)}\n"
        "Only locked seed file:\n"
        f"{sorted(locked_seed_set - complete_seed_set)}"
    )


print(
    "Exact seed identity: MATCH [OK]"
)


# ==========================================================================================
# 6. AUTHORS' RAW SOURCE CLASSES
# ==========================================================================================

print(
    "\n[5/10] AUTHORS' RAW SOURCE-CLASS AUDIT"
)


NON_FEATURE_COLUMNS = {
    "Gene_Name",
    "known_gene"
}


generic_header = (
    read_header(
        GENERIC_FILE
    )
)

filtered_header = (
    read_header(
        FILTERED_FILE
    )
)

complete_header = (
    read_header(
        COMPLETE_FILE
    )
)

processed_header = (
    read_header(
        PROCESSED_FILE
    )
)


generic_raw = (
    set(
        generic_header
    )
    -
    NON_FEATURE_COLUMNS
)

filtered_raw = (
    set(
        filtered_header
    )
    -
    NON_FEATURE_COLUMNS
)

complete_raw = (
    set(
        complete_header
    )
    -
    NON_FEATURE_COLUMNS
)


raw_overlap = (
    generic_raw
    &
    filtered_raw
)


print(
    "Generic raw features:",
    len(
        generic_raw
    )
)

print(
    "Filtered-by-disease raw features:",
    len(
        filtered_raw
    )
)

print(
    "Complete raw features:",
    len(
        complete_raw
    )
)

print(
    "Generic/filtered raw overlap:",
    len(
        raw_overlap
    )
)


# Authors' source classes should be disjoint except identity columns,
# which were already removed.
if raw_overlap:

    raise RuntimeError(
        "\nAuthors' generic and filtered-by-disease raw feature "
        "classes overlap unexpectedly:\n"
        f"{sorted(raw_overlap)}"
    )


if (
    generic_raw
    |
    filtered_raw
) != complete_raw:

    only_complete = (
        complete_raw
        -
        (
            generic_raw
            |
            filtered_raw
        )
    )

    outside_complete = (
        (
            generic_raw
            |
            filtered_raw
        )
        -
        complete_raw
    )


    raise RuntimeError(
        "\nAuthors' raw source classes do not reconstruct "
        "complete feature table.\n"
        f"Only complete: {sorted(only_complete)}\n"
        f"Outside complete: {sorted(outside_complete)}"
    )


print(
    "Raw source classes exactly reconstruct complete table [OK]"
)


# ==========================================================================================
# 7. VERIFY FINAL PREDICTORS ARE ACTUAL PROCESSED MODEL FEATURES
# ==========================================================================================

print(
    "\n[6/10] FINAL-vs-PROCESSED FEATURE IDENTITY"
)


processed_set = set(
    processed_header
)


missing_from_processed = sorted(
    final_set
    -
    processed_set
)


if missing_from_processed:

    raise RuntimeError(
        "\nFinal predictor(s) absent from exact190 processed table:\n"
        f"{missing_from_processed}"
    )


print(
    "All 316 final predictors exist in exact190 processed table [OK]"
)


# ==========================================================================================
# 8. DIRECT RAW-SOURCE MAPPING
# ==========================================================================================

direct_generic = []

direct_filtered = []

raw_unresolved = []


for feature in final_predictors:

    in_generic = (
        feature
        in generic_raw
    )

    in_filtered = (
        feature
        in filtered_raw
    )


    if (
        in_generic
        and
        in_filtered
    ):

        raise RuntimeError(
            "Ambiguous final predictor appears in BOTH raw classes:\n"
            f"{feature}"
        )


    elif in_generic:

        direct_generic.append(
            feature
        )


    elif in_filtered:

        direct_filtered.append(
            feature
        )


    else:

        raw_unresolved.append(
            feature
        )


print(
    "Direct GENERIC final predictors:",
    len(
        direct_generic
    )
)

print(
    "Direct FILTERED final predictors:",
    len(
        direct_filtered
    )
)

print(
    "Not verbatim in raw complete table:",
    len(
        raw_unresolved
    )
)


if raw_unresolved:

    print(
        "\nRaw-unresolved final predictors:"
    )

    for feature in raw_unresolved:

        print(
            " -",
            feature
        )


# ==========================================================================================
# 9. PROVE AUTHOR-GENERATED CATEGORICAL LINEAGE
#
# We are NOT classifying by feature-name keyword.
#
# The parent is source-provenance documented:
#   ProteinAtlas_gene_expr_levels
#
# We verify:
#   A. raw parent exists in COMPLETE
#   B. parent belongs to exactly one author source class
#   C. parent raw categories regenerate the exact unresolved dummy names
#   D. all generated dummy columns exist in processed_feature_table.tsv
#
# If any of these fail -> STOP.
# ==========================================================================================

print(
    "\n[7/10] AUTHOR ONE-HOT TRANSFORMATION LINEAGE"
)


CATEGORICAL_PARENT = (
    "ProteinAtlas_gene_expr_levels"
)


if CATEGORICAL_PARENT not in complete_raw:

    raise RuntimeError(
        "Expected authors' categorical parent is absent from "
        "complete feature table:\n"
        f"{CATEGORICAL_PARENT}"
    )


parent_in_generic = (
    CATEGORICAL_PARENT
    in generic_raw
)

parent_in_filtered = (
    CATEGORICAL_PARENT
    in filtered_raw
)


print(
    "Parent:",
    CATEGORICAL_PARENT
)

print(
    "Parent in generic:",
    parent_in_generic
)

print(
    "Parent in filtered-by-disease:",
    parent_in_filtered
)


# Must belong to exactly one source class.
if (
    int(
        parent_in_generic
    )
    +
    int(
        parent_in_filtered
    )
) != 1:

    raise RuntimeError(
        "Categorical parent does not map uniquely to an "
        "authors' source class."
    )


parent_source_class = (
    "generic"
    if parent_in_generic
    else
    "filtered_by_disease"
)


# ----------------------------------------------------------------------
# Read exact raw parent values from authors' complete table
# ----------------------------------------------------------------------

raw_parent_df = pd.read_csv(
    COMPLETE_FILE,
    sep="\t",
    usecols=[
        "Gene_Name",
        CATEGORICAL_PARENT
    ]
)


if raw_parent_df[
    CATEGORICAL_PARENT
].isna().any():

    raise RuntimeError(
        "Raw categorical parent contains missing values. "
        "Lineage recreation would require additional imputation audit."
    )


raw_categories = sorted(
    raw_parent_df[
        CATEGORICAL_PARENT
    ]
    .astype(str)
    .unique()
    .tolist()
)


print(
    "\nRaw parent categories:"
)

for category in raw_categories:

    print(
        " -",
        category
    )


# ----------------------------------------------------------------------
# Regenerate pandas one-hot names from the actual raw values.
#
# We do not use these dummies for modeling.
# This is provenance verification ONLY.
# ----------------------------------------------------------------------

recreated_dummy_df = pd.get_dummies(
    raw_parent_df[
        CATEGORICAL_PARENT
    ].astype(str),
    prefix=CATEGORICAL_PARENT
)


recreated_dummy_names = set(
    recreated_dummy_df.columns
)


print(
    "\nRecreated dummy columns:"
)

for feature in sorted(
    recreated_dummy_names
):

    print(
        " -",
        feature
    )


# ----------------------------------------------------------------------
# Which FINAL predictors are provably derived from this raw parent?
# ----------------------------------------------------------------------

derived_final = [
    feature
    for feature
    in final_predictors
    if feature
    in recreated_dummy_names
]


derived_final_set = set(
    derived_final
)


raw_unresolved_set = set(
    raw_unresolved
)


print(
    "\nDerived FINAL predictors recovered:",
    len(
        derived_final
    )
)


for feature in derived_final:

    print(
        " -",
        feature
    )


# CRITICAL:
# every previously raw-unresolved final predictor must be explained
# exactly by this verified one-hot transformation.
if (
    raw_unresolved_set
    !=
    derived_final_set
):

    unexplained = sorted(
        raw_unresolved_set
        -
        derived_final_set
    )

    unexpected_derived = sorted(
        derived_final_set
        -
        raw_unresolved_set
    )


    raise RuntimeError(
        "\nCategorical lineage does NOT completely explain "
        "raw-unresolved final predictors.\n"
        f"Still unexplained: {unexplained}\n"
        f"Unexpected derived: {unexpected_derived}"
    )


# Every derived final predictor must exist in processed table.
missing_derived_processed = sorted(
    derived_final_set
    -
    processed_set
)


if missing_derived_processed:

    raise RuntimeError(
        "\nVerified dummy predictor absent from processed model table:\n"
        f"{missing_derived_processed}"
    )


print(
    "\nAll raw-unresolved predictors explained by "
    "author one-hot transformation [OK]"
)

print(
    "Derived predictors inherit source class:",
    parent_source_class
)


# ==========================================================================================
# 10. BUILD FINAL LINEAGE-AWARE SOURCE CLASSIFICATION
# ==========================================================================================

print(
    "\n[8/10] FINAL SOURCE-CLASS ASSIGNMENT"
)


source_rows = []


direct_generic_set = set(
    direct_generic
)

direct_filtered_set = set(
    direct_filtered
)


for order, feature in enumerate(
    final_predictors
):

    if feature in direct_generic_set:

        source_class = (
            "generic"
        )

        raw_parent = (
            feature
        )

        lineage_type = (
            "direct_raw_column"
        )

        classification_basis = (
            "Exact final predictor name equals a column in "
            "authors' generic_feature_table.tsv."
        )


    elif feature in direct_filtered_set:

        source_class = (
            "filtered_by_disease"
        )

        raw_parent = (
            feature
        )

        lineage_type = (
            "direct_raw_column"
        )

        classification_basis = (
            "Exact final predictor name equals a column in "
            "authors' filtered_by_disease_feature_table.tsv."
        )


    elif feature in derived_final_set:

        source_class = (
            parent_source_class
        )

        raw_parent = (
            CATEGORICAL_PARENT
        )

        lineage_type = (
            "authors_EDA_one_hot_derived"
        )

        classification_basis = (
            "Final predictor regenerated exactly from the raw "
            "ProteinAtlas_gene_expr_levels categories by one-hot "
            "encoding; raw parent source class inherited."
        )


    else:

        raise RuntimeError(
            f"Unresolved final predictor: {feature}"
        )


    source_rows.append(
        {
            "final_feature_order":
                int(
                    order
                ),

            "feature":
                feature,

            "authors_source_class":
                source_class,

            "raw_source_parent":
                raw_parent,

            "lineage_type":
                lineage_type,

            "classification_basis":
                classification_basis
        }
    )


source_manifest = pd.DataFrame(
    source_rows
)


assert len(
    source_manifest
) == 316

assert (
    source_manifest[
        "feature"
    ].nunique()
    == 316
)


generic_final = (
    source_manifest.loc[
        source_manifest[
            "authors_source_class"
        ] == "generic",
        "feature"
    ]
    .tolist()
)


filtered_final = (
    source_manifest.loc[
        source_manifest[
            "authors_source_class"
        ] == "filtered_by_disease",
        "feature"
    ]
    .tolist()
)


classified_set = (
    set(
        generic_final
    )
    |
    set(
        filtered_final
    )
)


classified_overlap = (
    set(
        generic_final
    )
    &
    set(
        filtered_final
    )
)


unresolved_final = (
    final_set
    -
    classified_set
)


assert len(
    classified_overlap
) == 0

assert len(
    unresolved_final
) == 0

assert classified_set == final_set


print(
    "Final GENERIC predictors:",
    len(
        generic_final
    )
)

print(
    "Final FILTERED-BY-DISEASE predictors:",
    len(
        filtered_final
    )
)

print(
    "Final source overlap:",
    len(
        classified_overlap
    )
)

print(
    "Final unresolved:",
    len(
        unresolved_final
    )
)


# Expected based on actual preceding audit:
assert len(
    generic_final
) == 266, (
    "Unexpected generic final-feature count."
)

assert len(
    filtered_final
) == 50, (
    "Unexpected filtered-by-disease final-feature count."
)


print(
    "\n316 = 266 generic + 50 filtered-by-disease [OK]"
)


# ==========================================================================================
# 11. SCIENTIFIC ROBUSTNESS DEFINITION
#
# IMPORTANT WORDING:
#
# We do NOT call generic-only Context "leakage-free".
#
# We call it:
#   disease-context-reduced / generic-only sensitivity Context
#
# because all authors' phenotype/disease-filtered features are removed,
# but this does not prove that every remaining generic feature is
# completely free from all possible information overlap.
# ==========================================================================================

print(
    "\n[9/10] FREEZING ROBUSTNESS DEFINITION"
)


source_manifest[
    "in_full_context"
] = True


source_manifest[
    "in_generic_only_sensitivity"
] = (
    source_manifest[
        "authors_source_class"
    ] == "generic"
)


source_manifest[
    "removed_in_disease_context_sensitivity"
] = (
    source_manifest[
        "authors_source_class"
    ] == "filtered_by_disease"
)


GENERIC_ONLY_LABEL = (
    "Dhindsa_GenericOnly_DiseaseContextReduced"
)


ROBUSTNESS_DEFINITION = {
    "benchmark":
        (
            "Existing frozen full 316-feature Dhindsa Context model."
        ),

    "primary_sensitivity":
        (
            "Same Context XGBoost specification and exact same frozen "
            "PU partitions using only the 266 final predictors inherited "
            "from the authors' generic_feature_table.tsv."
        ),

    "removed_family":
        (
            "All 50 final predictors inherited from the authors' "
            "filtered_by_disease_feature_table.tsv, including four "
            "one-hot predictors derived by the authors' EDA from "
            "ProteinAtlas_gene_expr_levels."
        ),

    "interpretation":
        (
            "Conservative disease-context-reduced sensitivity analysis; "
            "NOT claimed to be completely leakage-free."
        )
}


print(
    "Full benchmark features       : 316"
)

print(
    "Generic-only sensitivity      :",
    len(
        generic_final
    )
)

print(
    "Disease-filtered features out :",
    len(
        filtered_final
    )
)

print(
    "Sensitivity label             :",
    GENERIC_ONLY_LABEL
)


# ==========================================================================================
# 12. SAVE LOCKED FEATURE FILES
# ==========================================================================================

FINAL_MANIFEST_FILE = (
    OUT_DIR /
    "FINAL_316_source_defined_feature_manifest_LINEAGE_AWARE.tsv"
)

GENERIC_LIST_FILE = (
    OUT_DIR /
    "GENERIC_ONLY_266_context_predictors_LOCKED.tsv"
)

FILTERED_LIST_FILE = (
    OUT_DIR /
    "FILTERED_BY_DISEASE_50_context_predictors_EXCLUDED.tsv"
)

DERIVED_LINEAGE_FILE = (
    OUT_DIR /
    "ProteinAtlas_onehot_lineage_audit.tsv"
)

SOURCE_TABLE_AUDIT_FILE = (
    OUT_DIR /
    "authors_source_table_final_feature_counts.tsv"
)

HASH_FILE = (
    OUT_DIR /
    "stage8A1_locked_output_SHA256.tsv"
)


source_manifest.to_csv(
    FINAL_MANIFEST_FILE,
    sep="\t",
    index=False
)


pd.DataFrame(
    {
        "feature":
            generic_final
    }
).to_csv(
    GENERIC_LIST_FILE,
    sep="\t",
    index=False
)


pd.DataFrame(
    {
        "feature":
            filtered_final
    }
).to_csv(
    FILTERED_LIST_FILE,
    sep="\t",
    index=False
)


derived_lineage = pd.DataFrame(
    {
        "derived_final_feature":
            derived_final,

        "raw_parent":
            CATEGORICAL_PARENT,

        "raw_parent_source_class":
            parent_source_class,

        "transformation":
            "authors_EDA_one_hot_encoding",

        "exists_in_exact190_processed_table":
            True
    }
)


derived_lineage.to_csv(
    DERIVED_LINEAGE_FILE,
    sep="\t",
    index=False
)


source_table_audit = pd.DataFrame(
    [
        {
            "source_class":
                "generic",

            "raw_source_features":
                len(
                    generic_raw
                ),

            "direct_final_features":
                len(
                    direct_generic
                ),

            "derived_final_features":
                int(
                    (
                        derived_lineage[
                            "raw_parent_source_class"
                        ] == "generic"
                    ).sum()
                ),

            "final_model_features":
                len(
                    generic_final
                ),

            "sensitivity_role":
                "retained"
        },

        {
            "source_class":
                "filtered_by_disease",

            "raw_source_features":
                len(
                    filtered_raw
                ),

            "direct_final_features":
                len(
                    direct_filtered
                ),

            "derived_final_features":
                int(
                    (
                        derived_lineage[
                            "raw_parent_source_class"
                        ] == "filtered_by_disease"
                    ).sum()
                ),

            "final_model_features":
                len(
                    filtered_final
                ),

            "sensitivity_role":
                "removed"
        }
    ]
)


source_table_audit.to_csv(
    SOURCE_TABLE_AUDIT_FILE,
    sep="\t",
    index=False
)


# ==========================================================================================
# 13. SAVE SCIENTIFIC LOCK JSON
# ==========================================================================================

LOCK_JSON = (
    ADMIN_DIR /
    "ASD_ContextFusion_STAGE8A1_SOURCE_DEFINED_ROBUSTNESS_LOCK.json"
)


lock = {

    "project":
        "ASD-ContextFusion",

    "stage":
        "8A.1",

    "lock_version":
        "lineage_aware_source_defined_v2",

    "scientific_status":
        (
            "Pre-robustness outcome lock. "
            "No robustness model has yet been evaluated."
        ),

    "full_context_feature_count":
        316,

    "generic_only_sensitivity_feature_count":
        int(
            len(
                generic_final
            )
        ),

    "filtered_by_disease_removed_feature_count":
        int(
            len(
                filtered_final
            )
        ),

    "direct_generic_final_features":
        int(
            len(
                direct_generic
            )
        ),

    "direct_filtered_final_features":
        int(
            len(
                direct_filtered
            )
        ),

    "onehot_derived_feature_count":
        int(
            len(
                derived_final
            )
        ),

    "onehot_parent":
        CATEGORICAL_PARENT,

    "onehot_parent_source_class":
        parent_source_class,

    "onehot_raw_categories":
        raw_categories,

    "onehot_derived_final_features":
        derived_final,

    "source_overlap":
        int(
            len(
                classified_overlap
            )
        ),

    "source_unresolved":
        int(
            len(
                unresolved_final
            )
        ),

    "classification_rule":
        (
            "Direct final predictors classified by exact membership "
            "in authors' generic or filtered-by-disease compiled tables. "
            "Post-EDA dummy predictors classified only after exact "
            "transformation lineage was verified from the raw categorical "
            "parent and exact processed-table membership."
        ),

    "keyword_based_predictor_classification":
        False,

    "outcome_based_feature_selection":
        False,

    "new_PU_sampling":
        False,

    "model_fitting":
        False,

    "robustness_definition":
        ROBUSTNESS_DEFINITION,

    "claim_guardrail":
        (
            "Generic-only model is described as disease-context-reduced "
            "or generic-only sensitivity analysis, not as completely "
            "leakage-free."
        ),

    "locked_seed_count":
        190,

    "exact_seed_identity_match":
        True,

    "input_files":
        {
            "final_predictor_file":
                {
                    "path":
                        str(
                            FINAL_PREDICTOR_FILE
                        ),

                    "sha256":
                        sha256_file(
                            FINAL_PREDICTOR_FILE
                        )
                },

            "generic_feature_table":
                {
                    "path":
                        str(
                            GENERIC_FILE
                        ),

                    "sha256":
                        sha256_file(
                            GENERIC_FILE
                        )
                },

            "filtered_by_disease_feature_table":
                {
                    "path":
                        str(
                            FILTERED_FILE
                        ),

                    "sha256":
                        sha256_file(
                            FILTERED_FILE
                        )
                },

            "complete_feature_table":
                {
                    "path":
                        str(
                            COMPLETE_FILE
                        ),

                    "sha256":
                        sha256_file(
                            COMPLETE_FILE
                        )
                },

            "processed_feature_table":
                {
                    "path":
                        str(
                            PROCESSED_FILE
                        ),

                    "sha256":
                        sha256_file(
                            PROCESSED_FILE
                        )
                },

            "locked_seed_file":
                {
                    "path":
                        str(
                            LOCKED_SEED_FILE
                        ),

                    "sha256":
                        sha256_file(
                            LOCKED_SEED_FILE
                        )
                }
        },

    "outputs":
        {
            "final_source_manifest":
                str(
                    FINAL_MANIFEST_FILE
                ),

            "generic_only_predictor_list":
                str(
                    GENERIC_LIST_FILE
                ),

            "filtered_by_disease_predictor_list":
                str(
                    FILTERED_LIST_FILE
                ),

            "onehot_lineage_audit":
                str(
                    DERIVED_LINEAGE_FILE
                ),

            "source_table_audit":
                str(
                    SOURCE_TABLE_AUDIT_FILE
                )
        },

    "future_modeling_rule":
        (
            "These source-defined feature lists are frozen before "
            "observing Stage 8B robustness performance and must not "
            "be altered based on downstream results."
        )
}


with open(
    LOCK_JSON,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        lock,
        f,
        indent=2
    )


# ==========================================================================================
# 14. OUTPUT HASH LOCK
# ==========================================================================================

hash_rows = []


for path in [
    FINAL_MANIFEST_FILE,
    GENERIC_LIST_FILE,
    FILTERED_LIST_FILE,
    DERIVED_LINEAGE_FILE,
    SOURCE_TABLE_AUDIT_FILE,
    LOCK_JSON
]:

    hash_rows.append(
        {
            "file":
                str(
                    path
                ),

            "sha256":
                sha256_file(
                    path
                )
        }
    )


hash_df = pd.DataFrame(
    hash_rows
)


hash_df.to_csv(
    HASH_FILE,
    sep="\t",
    index=False
)


# ==========================================================================================
# 15. FINAL LOCK AUDIT
# ==========================================================================================

print(
    "\n[10/10] FINAL PRE-MODELING LOCK AUDIT"
)


print(
    "\n" +
    "=" * 126
)

print(
    "STAGE 8A.1 CORRECTED SOURCE-DEFINED ROBUSTNESS LOCK COMPLETE [OK]"
)

print(
    "=" * 126
)


print(
    "\nSOURCE CLASS COUNTS"
)

print(
    "Full final predictors               :",
    len(
        final_predictors
    )
)

print(
    "Generic direct predictors           :",
    len(
        direct_generic
    )
)

print(
    "Filtered direct predictors          :",
    len(
        direct_filtered
    )
)

print(
    "Author one-hot derived predictors   :",
    len(
        derived_final
    )
)

print(
    "Final GENERIC model predictors      :",
    len(
        generic_final
    )
)

print(
    "Final FILTERED-BY-DISEASE predictors:",
    len(
        filtered_final
    )
)

print(
    "Source overlap                     :",
    len(
        classified_overlap
    )
)

print(
    "Source unresolved                  :",
    len(
        unresolved_final
    )
)


print(
    "\nONE-HOT LINEAGE"
)

print(
    "Raw parent:",
    CATEGORICAL_PARENT
)

print(
    "Raw parent source class:",
    parent_source_class
)

print(
    "Raw categories:",
    raw_categories
)

print(
    "Derived final columns:"
)

for feature in derived_final:

    print(
        " -",
        feature
    )


print(
    "\nIDENTITY / INTEGRITY"
)

print(
    "Exact ASD seeds                    : 190 [OK]"
)

print(
    "Exact seed identity                : MATCH [OK]"
)

print(
    "All 316 in exact processed table  : YES [OK]"
)

print(
    "Keyword-based feature assignment   : NO"
)

print(
    "Outcome-driven feature selection   : NO"
)

print(
    "Robustness model fitted            : NO"
)

print(
    "New PU sampling                    : NO"
)


print(
    "\nSCIENTIFIC DEFINITION FROZEN"
)

print(
    "Full Context:"
)

print(
    "  Existing frozen 316-feature benchmark."
)

print(
    "\nGeneric-only disease-context-reduced sensitivity:"
)

print(
    " ",
    len(
        generic_final
    ),
    "authors' generic-source final predictors."
)

print(
    "\nRemoved disease-filtered family:"
)

print(
    " ",
    len(
        filtered_final
    ),
    "authors' filtered-by-disease final predictors."
)

print(
    "\nClaim guardrail:"
)

print(
    "  Do NOT call the generic-only model completely leakage-free."
)


print(
    "\nSAVED:"
)

print(
    "Feature manifest:"
)

print(
    FINAL_MANIFEST_FILE
)

print(
    "\nGeneric-only list:"
)

print(
    GENERIC_LIST_FILE
)

print(
    "\nFiltered-by-disease list:"
)

print(
    FILTERED_LIST_FILE
)

print(
    "\nOne-hot lineage audit:"
)

print(
    DERIVED_LINEAGE_FILE
)

print(
    "\nScientific lock:"
)

print(
    LOCK_JSON
)

print(
    "\nHash lock:"
)

print(
    HASH_FILE
)


print(
    "\nNEXT SCIENTIFIC STAGE:"
)

print(
    "Stage 8B — same frozen PU partitions, same frozen XGBoost "
    "specification, generic-only Context sensitivity + strict "
    "Context/NT-v2 cross-fitted late fusion."
)

print(
    "\nIMPORTANT: do not change these feature lists after Stage 8B results are observed."
)

print(
    "=" * 126
)
```

```python
# ==========================================================================================
# ASD-ContextFusion
# STAGE 8B
# SOURCE-DEFINED DISEASE-CONTEXT-REDUCED ROBUSTNESS
#
# Scientific design
# -----------------
# Full Context benchmark:
#   ALREADY FROZEN — NOT RERUN
#
# Robustness Context:
#   266 final predictors inherited from authors' GENERIC feature class
#
# NT-v2:
#   UNCHANGED
#
# Fusion:
#   SAME strict nested cross-fitted logistic late fusion
#
# PU partitions:
#   EXACT SAME frozen 10 iterations x 79 balanced datasets x 10 folds
#
# IMPORTANT
# ---------
# - no new PU sampling
# - no tuning
# - no model hunting
# - no change to XGBoost parameters
# - no change to NT probe
# - no change to meta learner
# - no raw concatenation
# - no controls fused
# - no feature-list changes after Stage 8A.1 lock
#
# Engineering:
#   790 independent checkpointed tasks
#   6 subprocess workers
#   safe to rerun after kernel/power interruption
# ==========================================================================================

from pathlib import Path
import os
import sys
import json
import time
import hashlib
import subprocess
import threading
import textwrap

import numpy as np
import pandas as pd

from concurrent.futures import (
    ThreadPoolExecutor,
    as_completed
)


# ==========================================================================================
# 0. PATHS / SETTINGS
# ==========================================================================================

ROOT = Path(r"E:\ASD_ContextFusion_Q1")

REPO = (
    ROOT /
    "01_raw" /
    "dhindsa_2025" /
    "Mantis-ml-NDD"
)

LEGACY_PYTHON = (
    ROOT /
    "envs" /
    "mantis37" /
    "python.exe"
)

MASTER_FILE = (
    ROOT /
    "03_features" /
    "multimodal_master" /
    "dhindsa_NTv2_row_preserved_master.tsv"
)

NT_FILE = (
    ROOT /
    "03_features" /
    "nt_v2" /
    "NTv2_50m_full_transcript_embeddings_18252x512.npy"
)

PARTITION_DIR = (
    ROOT /
    "04_models" /
    "checkpoints" /
    "dhindsa_PU_common_universe"
)

FEATURE_LOCK_DIR = (
    ROOT /
    "02_interim" /
    "dhindsa_source_feature_family_lock"
)

GENERIC_FEATURE_FILE = (
    FEATURE_LOCK_DIR /
    "GENERIC_ONLY_266_context_predictors_LOCKED.tsv"
)

FEATURE_MANIFEST_FILE = (
    FEATURE_LOCK_DIR /
    "FINAL_316_source_defined_feature_manifest_LINEAGE_AWARE.tsv"
)

SCIENTIFIC_LOCK_FILE = (
    ROOT /
    "00_admin" /
    "ASD_ContextFusion_STAGE8A1_SOURCE_DEFINED_ROBUSTNESS_LOCK.json"
)

# Existing fully-verified strict fusion checkpoints.
# We reuse ONLY the frozen outer-test NT probabilities from these.
FULL_STRICT_CHECKPOINT_DIR = (
    ROOT /
    "04_models" /
    "checkpoints" /
    "strict_context_NTv2_fusion"
)

FULL_STRICT_RESULT_DIR = (
    ROOT /
    "05_results" /
    "proposed" /
    "strict_context_NTv2_fusion"
)

# New robustness outputs
CHECKPOINT_DIR = (
    ROOT /
    "04_models" /
    "checkpoints" /
    "stage8B_generic_context_strict_fusion"
)

RESULT_DIR = (
    ROOT /
    "05_results" /
    "proposed" /
    "stage8B_generic_context_strict_fusion"
)

LOG_DIR = (
    ROOT /
    "08_logs" /
    "stage8B_generic_context_strict_fusion_parallel"
)

RUNNER = (
    ROOT /
    "00_admin" /
    "run_stage8B_generic_context_strict_fusion.py"
)

RUNNER_MANIFEST = (
    ROOT /
    "00_admin" /
    "ASD_ContextFusion_STAGE8B_runner_manifest.json"
)

MAX_WORKERS = 6

RUNNER_VERSION = "ASD_CF_STAGE8B_GENERIC266_STRICT_V1"


for d in [
    CHECKPOINT_DIR,
    RESULT_DIR,
    LOG_DIR
]:
    d.mkdir(
        parents=True,
        exist_ok=True
    )


# ==========================================================================================
# 1. HELPERS
# ==========================================================================================

def sha256_file(path, chunk_size=1024 * 1024):

    h = hashlib.sha256()

    with open(path, "rb") as f:

        while True:

            chunk = f.read(chunk_size)

            if not chunk:
                break

            h.update(chunk)

    return h.hexdigest()


# ==========================================================================================
# 2. PRE-RUN SCIENTIFIC LOCK AUDIT
# ==========================================================================================

print("=" * 126)
print("ASD-CONTEXTFUSION — STAGE 8B GENERIC-ONLY STRICT ROBUSTNESS")
print("=" * 126)

print("\n[1/7] PRE-RUN SCIENTIFIC LOCK AUDIT")


required = [
    REPO,
    LEGACY_PYTHON,
    MASTER_FILE,
    NT_FILE,
    PARTITION_DIR,
    GENERIC_FEATURE_FILE,
    FEATURE_MANIFEST_FILE,
    SCIENTIFIC_LOCK_FILE,
    FULL_STRICT_CHECKPOINT_DIR,
    FULL_STRICT_RESULT_DIR
]


for path in required:

    assert path.exists(), (
        f"\nMISSING REQUIRED INPUT:\n{path}"
    )

    print(
        "[OK]",
        path
    )


with open(
    SCIENTIFIC_LOCK_FILE,
    "r",
    encoding="utf-8"
) as f:

    scientific_lock = json.load(f)


assert scientific_lock[
    "full_context_feature_count"
] == 316

assert scientific_lock[
    "generic_only_sensitivity_feature_count"
] == 266

assert scientific_lock[
    "filtered_by_disease_removed_feature_count"
] == 50

assert scientific_lock[
    "source_overlap"
] == 0

assert scientific_lock[
    "source_unresolved"
] == 0

assert scientific_lock[
    "keyword_based_predictor_classification"
] is False

assert scientific_lock[
    "outcome_based_feature_selection"
] is False


generic_features = (
    pd.read_csv(
        GENERIC_FEATURE_FILE,
        sep="\t"
    )[
        "feature"
    ]
    .astype(str)
    .tolist()
)


assert len(
    generic_features
) == 266

assert len(
    set(
        generic_features
    )
) == 266


feature_manifest = pd.read_csv(
    FEATURE_MANIFEST_FILE,
    sep="\t"
)


manifest_generic = (
    feature_manifest.loc[
        feature_manifest[
            "authors_source_class"
        ] == "generic",
        "feature"
    ]
    .astype(str)
    .tolist()
)


manifest_filtered = (
    feature_manifest.loc[
        feature_manifest[
            "authors_source_class"
        ] == "filtered_by_disease",
        "feature"
    ]
    .astype(str)
    .tolist()
)


assert len(
    manifest_generic
) == 266

assert len(
    manifest_filtered
) == 50

assert set(
    generic_features
) == set(
    manifest_generic
)


print("\nScientific feature lock:")
print("Full benchmark          : 316")
print("Generic-only sensitivity: 266")
print("Removed disease-context : 50")
print("Source overlap          : 0")
print("Source unresolved       : 0")
print("Feature lock            : VERIFIED [OK]")


GENERIC_FEATURE_SHA = sha256_file(
    GENERIC_FEATURE_FILE
)


# ==========================================================================================
# 3. VERIFY ALL 790 ORIGINAL STRICT TASKS EXIST
# ==========================================================================================

print("\n[2/7] VERIFYING FROZEN NT OUTER-PREDICTION SOURCE")


source_task_count = 0


for iteration in range(
    1,
    11
):

    for balanced_dataset in range(
        79
    ):

        source_file = (
            FULL_STRICT_CHECKPOINT_DIR /
            (
                f"iteration_{iteration:02d}_balanced_"
                f"{balanced_dataset:02d}_fusion_predictions.tsv.gz"
            )
        )

        assert source_file.exists(), (
            f"Missing original strict source checkpoint:\n{source_file}"
        )

        source_task_count += 1


assert source_task_count == 790


print(
    "Verified original strict source tasks:",
    source_task_count,
    "/ 790 [OK]"
)

print(
    "Only NT outer-test probability will be reused."
)

print(
    "Full Context/Fusion predictions will NOT be reused "
    "as robustness predictions."
)


# ==========================================================================================
# 4. BUILD DEDICATED LEGACY-PYTHON STAGE 8B RUNNER
# ==========================================================================================

print("\n[3/7] BUILDING DEDICATED STAGE 8B SCIENTIFIC RUNNER")


runner_code = r"""
from __future__ import print_function

from pathlib import Path
from itertools import combinations

import os
import json
import time
import hashlib
import warnings

warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd

import xgboost as xgb

from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression

from sklearn.metrics import (
    roc_auc_score,
    average_precision_score
)


# ==========================================================================================
# CONSTANTS
# ==========================================================================================

ROOT = Path(r"E:\ASD_ContextFusion_Q1")

RUNNER_VERSION = "ASD_CF_STAGE8B_GENERIC266_STRICT_V1"

MASTER_FILE = (
    ROOT /
    "03_features" /
    "multimodal_master" /
    "dhindsa_NTv2_row_preserved_master.tsv"
)

NT_FILE = (
    ROOT /
    "03_features" /
    "nt_v2" /
    "NTv2_50m_full_transcript_embeddings_18252x512.npy"
)

GENERIC_FEATURE_FILE = (
    ROOT /
    "02_interim" /
    "dhindsa_source_feature_family_lock" /
    "GENERIC_ONLY_266_context_predictors_LOCKED.tsv"
)

SCIENTIFIC_LOCK_FILE = (
    ROOT /
    "00_admin" /
    "ASD_ContextFusion_STAGE8A1_SOURCE_DEFINED_ROBUSTNESS_LOCK.json"
)

PARTITION_DIR = (
    ROOT /
    "04_models" /
    "checkpoints" /
    "dhindsa_PU_common_universe"
)

FULL_STRICT_CHECKPOINT_DIR = (
    ROOT /
    "04_models" /
    "checkpoints" /
    "strict_context_NTv2_fusion"
)

FULL_STRICT_RESULT_DIR = (
    ROOT /
    "05_results" /
    "proposed" /
    "strict_context_NTv2_fusion"
)

CHECKPOINT_DIR = (
    ROOT /
    "04_models" /
    "checkpoints" /
    "stage8B_generic_context_strict_fusion"
)

RESULT_DIR = (
    ROOT /
    "05_results" /
    "proposed" /
    "stage8B_generic_context_strict_fusion"
)

CHECKPOINT_DIR.mkdir(
    parents=True,
    exist_ok=True
)

RESULT_DIR.mkdir(
    parents=True,
    exist_ok=True
)


# ==========================================================================================
# HELPERS
# ==========================================================================================

def sha256_file(
    path,
    chunk_size=1024 * 1024
):

    h = hashlib.sha256()

    with open(
        str(path),
        "rb"
    ) as f:

        while True:

            chunk = f.read(
                chunk_size
            )

            if not chunk:
                break

            h.update(
                chunk
            )

    return h.hexdigest()


RUNNER_SHA256 = sha256_file(
    Path(
        __file__
    )
)

FEATURE_LIST_SHA256 = sha256_file(
    GENERIC_FEATURE_FILE
)


def atomic_tsv(
    df,
    path,
    compression=None
):

    tmp = Path(
        str(path) +
        ".tmp"
    )

    df.to_csv(
        str(tmp),
        sep="\t",
        index=False,
        compression=compression
    )

    os.replace(
        str(tmp),
        str(path)
    )


def task_prediction_file(
    iteration,
    balanced_dataset
):

    return (
        CHECKPOINT_DIR /
        (
            "iteration_%02d_balanced_%02d_robust_predictions.tsv.gz"
            %
            (
                iteration,
                balanced_dataset
            )
        )
    )


def task_metric_file(
    iteration,
    balanced_dataset
):

    return (
        CHECKPOINT_DIR /
        (
            "iteration_%02d_balanced_%02d_robust_metrics.tsv"
            %
            (
                iteration,
                balanced_dataset
            )
        )
    )


def task_coef_file(
    iteration,
    balanced_dataset
):

    return (
        CHECKPOINT_DIR /
        (
            "iteration_%02d_balanced_%02d_meta_coefficients.tsv"
            %
            (
                iteration,
                balanced_dataset
            )
        )
    )


# ==========================================================================================
# LOCKED SCIENTIFIC MODEL SPECIFICATIONS
# ==========================================================================================

XGB_PARAMS = {
    "learning_rate":
        0.01,

    "n_estimators":
        300,

    "max_depth":
        5,

    "min_child_weight":
        3,

    "gamma":
        0,

    "subsample":
        0.8,

    "colsample_bytree":
        0.8,

    "objective":
        "binary:logistic",

    "nthread":
        1,

    "scale_pos_weight":
        1
}


NT_PARAMS = {
    "penalty":
        "l2",

    "C":
        1.0,

    "solver":
        "lbfgs",

    "max_iter":
        2000
}


META_PARAMS = {
    "penalty":
        "l2",

    "C":
        1.0,

    "solver":
        "lbfgs",

    "max_iter":
        2000
}


def fit_context_predict(
    X_train,
    y_train,
    X_predict
):

    model = xgb.XGBClassifier(
        **XGB_PARAMS
    )

    model.fit(
        X_train,
        y_train
    )

    prob = model.predict_proba(
        X_predict
    )[
        :,
        1
    ]

    return np.asarray(
        prob,
        dtype=np.float64
    )


def fit_nt_predict(
    X_train,
    y_train,
    X_predict
):

    scaler = StandardScaler()

    X_train_scaled = scaler.fit_transform(
        X_train
    )

    X_predict_scaled = scaler.transform(
        X_predict
    )

    model = LogisticRegression(
        **NT_PARAMS
    )

    model.fit(
        X_train_scaled,
        y_train
    )

    prob = model.predict_proba(
        X_predict_scaled
    )[
        :,
        1
    ]

    return np.asarray(
        prob,
        dtype=np.float64
    )


# ==========================================================================================
# TASK MODE
# ==========================================================================================

def run_task(
    iteration,
    balanced_dataset
):

    task_start = time.time()

    iteration = int(
        iteration
    )

    balanced_dataset = int(
        balanced_dataset
    )


    print(
        "=" * 118
    )

    print(
        "STAGE 8B TASK | ITER %02d | BAL %02d"
        %
        (
            iteration,
            balanced_dataset
        )
    )

    print(
        "=" * 118
    )


    # --------------------------------------------------------------------------------------
    # Scientific lock
    # --------------------------------------------------------------------------------------

    with open(
        str(
            SCIENTIFIC_LOCK_FILE
        ),
        "r"
    ) as f:

        lock = json.load(
            f
        )


    assert lock[
        "generic_only_sensitivity_feature_count"
    ] == 266

    assert lock[
        "filtered_by_disease_removed_feature_count"
    ] == 50

    assert lock[
        "source_overlap"
    ] == 0

    assert lock[
        "source_unresolved"
    ] == 0


    feature_df = pd.read_csv(
        str(
            GENERIC_FEATURE_FILE
        ),
        sep="\t"
    )


    context_features = (
        feature_df[
            "feature"
        ]
        .astype(str)
        .tolist()
    )


    assert len(
        context_features
    ) == 266

    assert len(
        set(
            context_features
        )
    ) == 266


    # --------------------------------------------------------------------------------------
    # Frozen membership
    # --------------------------------------------------------------------------------------

    membership_file = (
        PARTITION_DIR /
        (
            "iteration_%02d_test_fold_membership.tsv.gz"
            %
            iteration
        )
    )


    membership = pd.read_csv(
        str(
            membership_file
        ),
        sep="\t",
        compression="gzip"
    )


    membership = (
        membership.loc[
            membership[
                "balanced_dataset"
            ] == balanced_dataset
        ]
        .copy()
        .reset_index(
            drop=True
        )
    )


    assert len(
        membership
    ) >= 380

    assert len(
        membership
    ) <= 383

    assert membership[
        "dhindsa_row_id"
    ].is_unique

    assert membership[
        "fold"
    ].nunique() == 10

    assert set(
        membership[
            "fold"
        ].astype(int)
    ) == set(
        range(
            10
        )
    )


    membership[
        "dhindsa_row_id"
    ] = (
        membership[
            "dhindsa_row_id"
        ]
        .astype(int)
    )

    membership[
        "fold"
    ] = (
        membership[
            "fold"
        ]
        .astype(int)
    )


    # --------------------------------------------------------------------------------------
    # Locked master
    # --------------------------------------------------------------------------------------

    master = pd.read_csv(
        str(
            MASTER_FILE
        ),
        sep="\t",
        low_memory=False
    )


    assert len(
        master
    ) == 18253

    assert master[
        "Gene_Name"
    ].nunique() == 18252

    assert master[
        "dhindsa_row_id"
    ].is_unique


    master[
        "dhindsa_row_id"
    ] = (
        master[
            "dhindsa_row_id"
        ]
        .astype(int)
    )

    master[
        "NT_row_index"
    ] = (
        master[
            "NT_row_index"
        ]
        .astype(int)
    )

    master[
        "known_gene"
    ] = (
        master[
            "known_gene"
        ]
        .astype(int)
    )


    assert all(
        feature
        in master.columns
        for feature
        in context_features
    )


    master_idx = master.set_index(
        "dhindsa_row_id",
        drop=False
    )


    balanced_ids = (
        membership[
            "dhindsa_row_id"
        ]
        .astype(int)
        .tolist()
    )


    bal_df = (
        master_idx.loc[
            balanced_ids
        ]
        .copy()
        .reset_index(
            drop=True
        )
    )


    assert len(
        bal_df
    ) == len(
        membership
    )


    assert np.array_equal(
        bal_df[
            "dhindsa_row_id"
        ].to_numpy(
            dtype=np.int64
        ),
        membership[
            "dhindsa_row_id"
        ].to_numpy(
            dtype=np.int64
        )
    )


    bal_df[
        "fold"
    ] = (
        membership[
            "fold"
        ]
        .to_numpy(
            dtype=np.int32
        )
    )


    # Optional identity verification against membership columns.
    if "known_gene" in membership.columns:

        assert np.array_equal(
            bal_df[
                "known_gene"
            ].to_numpy(
                dtype=np.int32
            ),
            membership[
                "known_gene"
            ].to_numpy(
                dtype=np.int32
            )
        )


    if "Gene_Name" in membership.columns:

        assert np.array_equal(
            bal_df[
                "Gene_Name"
            ].astype(str)
            .to_numpy(),
            membership[
                "Gene_Name"
            ].astype(str)
            .to_numpy()
        )


    # --------------------------------------------------------------------------------------
    # Matrices
    # --------------------------------------------------------------------------------------

    X_context = (
        bal_df[
            context_features
        ]
        .to_numpy(
            dtype=np.float32
        )
    )


    assert X_context.shape == (
        len(
            bal_df
        ),
        266
    )


    assert np.isfinite(
        X_context
    ).all()


    X_nt_all = np.load(
        str(
            NT_FILE
        ),
        mmap_mode="r"
    )


    assert X_nt_all.shape == (
        18252,
        512
    )


    nt_indices = (
        bal_df[
            "NT_row_index"
        ]
        .to_numpy(
            dtype=np.int64
        )
    )


    X_nt = np.asarray(
        X_nt_all[
            nt_indices,
            :
        ],
        dtype=np.float32
    )


    assert X_nt.shape == (
        len(
            bal_df
        ),
        512
    )


    assert np.isfinite(
        X_nt
    ).all()


    y = (
        bal_df[
            "known_gene"
        ]
        .to_numpy(
            dtype=np.int32
        )
    )


    folds = (
        bal_df[
            "fold"
        ]
        .to_numpy(
            dtype=np.int32
        )
    )


    # --------------------------------------------------------------------------------------
    # Frozen original outer NT probabilities
    # --------------------------------------------------------------------------------------

    original_task_pred_file = (
        FULL_STRICT_CHECKPOINT_DIR /
        (
            "iteration_%02d_balanced_%02d_fusion_predictions.tsv.gz"
            %
            (
                iteration,
                balanced_dataset
            )
        )
    )


    assert original_task_pred_file.exists()


    frozen_outer = pd.read_csv(
        str(
            original_task_pred_file
        ),
        sep="\t",
        compression="gzip"
    )


    required_outer_cols = {
        "dhindsa_row_id",
        "fold",
        "Gene_Name",
        "known_gene",
        "NT_probability"
    }


    assert required_outer_cols.issubset(
        set(
            frozen_outer.columns
        )
    )


    frozen_outer[
        "dhindsa_row_id"
    ] = (
        frozen_outer[
            "dhindsa_row_id"
        ]
        .astype(int)
    )


    frozen_outer[
        "fold"
    ] = (
        frozen_outer[
            "fold"
        ]
        .astype(int)
    )


    assert frozen_outer[
        "dhindsa_row_id"
    ].is_unique


    assert set(
        frozen_outer[
            "dhindsa_row_id"
        ]
    ) == set(
        balanced_ids
    )


    frozen_outer_idx = frozen_outer.set_index(
        "dhindsa_row_id"
    )


    # Validate source fold / label identity
    frozen_check = (
        membership[
            [
                "dhindsa_row_id",
                "fold"
            ]
        ]
        .merge(
            frozen_outer[
                [
                    "dhindsa_row_id",
                    "fold"
                ]
            ],
            on="dhindsa_row_id",
            how="inner",
            suffixes=(
                "_membership",
                "_source"
            ),
            validate="one_to_one"
        )
    )


    assert len(
        frozen_check
    ) == len(
        membership
    )


    assert np.array_equal(
        frozen_check[
            "fold_membership"
        ].to_numpy(
            dtype=np.int32
        ),
        frozen_check[
            "fold_source"
        ].to_numpy(
            dtype=np.int32
        )
    )


    # ======================================================================================
    # STRICT INNER CROSS-FITTING
    #
    # For every unordered excluded pair {a,b}:
    #
    #   train base models on folds other than a and b
    #   predict a
    #   predict b
    #
    # These predictions are then reused in the mathematically equivalent
    # ordered outer/inner roles.
    # ======================================================================================

    pair_context_predictions = {}
    pair_nt_predictions = {}


    pair_counter = 0


    for a, b in combinations(
        range(
            10
        ),
        2
    ):

        pair_counter += 1


        pair_train_mask = (
            (folds != a)
            &
            (folds != b)
        )


        assert pair_train_mask.sum() > 0


        Xc_train = X_context[
            pair_train_mask
        ]

        Xn_train = X_nt[
            pair_train_mask
        ]

        y_train = y[
            pair_train_mask
        ]


        assert len(
            np.unique(
                y_train
            )
        ) == 2


        # ------------------------------------------------------------------
        # Context pair model — same XGBoost, only feature family changed
        # ------------------------------------------------------------------

        context_model = xgb.XGBClassifier(
            **XGB_PARAMS
        )


        context_model.fit(
            Xc_train,
            y_train
        )


        # ------------------------------------------------------------------
        # NT pair model — unchanged frozen probe specification
        # ------------------------------------------------------------------

        nt_scaler = StandardScaler()

        Xn_train_scaled = nt_scaler.fit_transform(
            Xn_train
        )


        nt_model = LogisticRegression(
            **NT_PARAMS
        )


        nt_model.fit(
            Xn_train_scaled,
            y_train
        )


        for target_fold in [
            a,
            b
        ]:

            target_mask = (
                folds
                ==
                target_fold
            )


            context_prob = (
                context_model
                .predict_proba(
                    X_context[
                        target_mask
                    ]
                )[
                    :,
                    1
                ]
            )


            nt_prob = (
                nt_model
                .predict_proba(
                    nt_scaler.transform(
                        X_nt[
                            target_mask
                        ]
                    )
                )[
                    :,
                    1
                ]
            )


            key = (
                min(
                    a,
                    b
                ),
                max(
                    a,
                    b
                ),
                target_fold
            )


            pair_context_predictions[
                key
            ] = np.asarray(
                context_prob,
                dtype=np.float64
            )


            pair_nt_predictions[
                key
            ] = np.asarray(
                nt_prob,
                dtype=np.float64
            )


        if (
            pair_counter % 15
            == 0
        ):

            print(
                "Leave-two-fold-out pairs:",
                pair_counter,
                "/ 45"
            )


    assert pair_counter == 45

    assert len(
        pair_context_predictions
    ) == 90

    assert len(
        pair_nt_predictions
    ) == 90


    # ======================================================================================
    # OUTER FOLDS
    # ======================================================================================

    prediction_frames = []
    metric_rows = []
    coefficient_rows = []


    for outer_fold in range(
        10
    ):

        outer_test_mask = (
            folds
            ==
            outer_fold
        )

        outer_train_mask = (
            folds
            !=
            outer_fold
        )


        assert outer_test_mask.sum() > 0
        assert outer_train_mask.sum() > 0


        y_outer_train = y[
            outer_train_mask
        ]

        y_outer_test = y[
            outer_test_mask
        ]


        assert len(
            np.unique(
                y_outer_train
            )
        ) == 2

        assert len(
            np.unique(
                y_outer_test
            )
        ) == 2


        # ----------------------------------------------------------------------------------
        # OUTER GENERIC-ONLY CONTEXT
        # train WITHOUT outer fold
        # ----------------------------------------------------------------------------------

        outer_context_model = xgb.XGBClassifier(
            **XGB_PARAMS
        )


        outer_context_model.fit(
            X_context[
                outer_train_mask
            ],
            y_outer_train
        )


        outer_context_prob = (
            outer_context_model
            .predict_proba(
                X_context[
                    outer_test_mask
                ]
            )[
                :,
                1
            ]
        )


        outer_context_prob = np.asarray(
            outer_context_prob,
            dtype=np.float64
        )


        # ----------------------------------------------------------------------------------
        # OUTER NT
        # frozen probability from already-verified primary/strict analysis
        # ----------------------------------------------------------------------------------

        outer_test_ids = (
            bal_df.loc[
                outer_test_mask,
                "dhindsa_row_id"
            ]
            .astype(int)
            .tolist()
        )


        outer_nt_prob = (
            frozen_outer_idx.loc[
                outer_test_ids,
                "NT_probability"
            ]
            .to_numpy(
                dtype=np.float64
            )
        )


        assert np.isfinite(
            outer_nt_prob
        ).all()


        # ==================================================================================
        # LEAKAGE-FREE META-TRAINING SET
        #
        # For outer f, each inner validation g receives predictions
        # from base models trained excluding BOTH f and g.
        # ==================================================================================

        meta_X_parts = []
        meta_y_parts = []


        for inner_fold in range(
            10
        ):

            if inner_fold == outer_fold:
                continue


            inner_mask = (
                folds
                ==
                inner_fold
            )


            pair_key = (
                min(
                    outer_fold,
                    inner_fold
                ),
                max(
                    outer_fold,
                    inner_fold
                ),
                inner_fold
            )


            context_inner_prob = (
                pair_context_predictions[
                    pair_key
                ]
            )


            nt_inner_prob = (
                pair_nt_predictions[
                    pair_key
                ]
            )


            assert len(
                context_inner_prob
            ) == int(
                inner_mask.sum()
            )

            assert len(
                nt_inner_prob
            ) == int(
                inner_mask.sum()
            )


            meta_X_parts.append(
                np.column_stack(
                    [
                        context_inner_prob,
                        nt_inner_prob
                    ]
                )
            )


            meta_y_parts.append(
                y[
                    inner_mask
                ]
            )


        meta_X = np.vstack(
            meta_X_parts
        )

        meta_y = np.concatenate(
            meta_y_parts
        )


        assert meta_X.shape == (
            int(
                outer_train_mask.sum()
            ),
            2
        )

        assert len(
            meta_y
        ) == int(
            outer_train_mask.sum()
        )

        assert len(
            np.unique(
                meta_y
            )
        ) == 2

        assert np.isfinite(
            meta_X
        ).all()


        # ----------------------------------------------------------------------------------
        # LOCKED META LEARNER
        # ----------------------------------------------------------------------------------

        meta_model = LogisticRegression(
            **META_PARAMS
        )


        meta_model.fit(
            meta_X,
            meta_y
        )


        outer_meta_X = np.column_stack(
            [
                outer_context_prob,
                outer_nt_prob
            ]
        )


        fusion_prob = (
            meta_model
            .predict_proba(
                outer_meta_X
            )[
                :,
                1
            ]
        )


        fusion_prob = np.asarray(
            fusion_prob,
            dtype=np.float64
        )


        # ----------------------------------------------------------------------------------
        # Metrics
        # ----------------------------------------------------------------------------------

        context_auc = roc_auc_score(
            y_outer_test,
            outer_context_prob
        )

        context_ap = average_precision_score(
            y_outer_test,
            outer_context_prob
        )

        nt_auc = roc_auc_score(
            y_outer_test,
            outer_nt_prob
        )

        nt_ap = average_precision_score(
            y_outer_test,
            outer_nt_prob
        )

        fusion_auc = roc_auc_score(
            y_outer_test,
            fusion_prob
        )

        fusion_ap = average_precision_score(
            y_outer_test,
            fusion_prob
        )


        metric_rows.append(
            {
                "iteration":
                    iteration,

                "balanced_dataset":
                    balanced_dataset,

                "fold":
                    outer_fold,

                "GenericContext_AUROC":
                    context_auc,

                "GenericContext_AP":
                    context_ap,

                "NT_AUROC":
                    nt_auc,

                "NT_AP":
                    nt_ap,

                "GenericFusion_AUROC":
                    fusion_auc,

                "GenericFusion_AP":
                    fusion_ap,

                "runner_version":
                    RUNNER_VERSION,

                "runner_sha256":
                    RUNNER_SHA256,

                "feature_list_sha256":
                    FEATURE_LIST_SHA256
            }
        )


        coef = (
            meta_model.coef_[
                0
            ]
        )


        coefficient_rows.append(
            {
                "iteration":
                    iteration,

                "balanced_dataset":
                    balanced_dataset,

                "fold":
                    outer_fold,

                "GenericContext_coefficient":
                    float(
                        coef[
                            0
                        ]
                    ),

                "NT_coefficient":
                    float(
                        coef[
                            1
                        ]
                    ),

                "Intercept":
                    float(
                        meta_model.intercept_[
                            0
                        ]
                    ),

                "runner_version":
                    RUNNER_VERSION,

                "runner_sha256":
                    RUNNER_SHA256,

                "feature_list_sha256":
                    FEATURE_LIST_SHA256
            }
        )


        test_frame = (
            bal_df.loc[
                outer_test_mask,
                [
                    "dhindsa_row_id",
                    "Gene_Name",
                    "known_gene"
                ]
            ]
            .copy()
            .reset_index(
                drop=True
            )
        )


        test_frame.insert(
            0,
            "iteration",
            iteration
        )

        test_frame.insert(
            1,
            "balanced_dataset",
            balanced_dataset
        )

        test_frame.insert(
            2,
            "fold",
            outer_fold
        )


        test_frame[
            "GenericContext_probability"
        ] = outer_context_prob


        test_frame[
            "NT_probability"
        ] = outer_nt_prob


        test_frame[
            "GenericFusion_probability"
        ] = fusion_prob


        prediction_frames.append(
            test_frame
        )


        print(
            "Outer fold %02d/10 | Generic AP %.4f | NT AP %.4f | Fusion AP %.4f"
            %
            (
                outer_fold + 1,
                context_ap,
                nt_ap,
                fusion_ap
            )
        )


    # ======================================================================================
    # TASK CHECKPOINT
    # ======================================================================================

    task_predictions = pd.concat(
        prediction_frames,
        ignore_index=True
    )

    task_metrics = pd.DataFrame(
        metric_rows
    )

    task_coefficients = pd.DataFrame(
        coefficient_rows
    )


    assert len(
        task_metrics
    ) == 10

    assert task_metrics[
        "fold"
    ].nunique() == 10

    assert len(
        task_coefficients
    ) == 10

    assert task_coefficients[
        "fold"
    ].nunique() == 10

    assert len(
        task_predictions
    ) == len(
        membership
    )

    assert task_predictions[
        "dhindsa_row_id"
    ].is_unique

    assert set(
        task_predictions[
            "dhindsa_row_id"
        ].astype(int)
    ) == set(
        balanced_ids
    )


    for col in [
        "GenericContext_probability",
        "NT_probability",
        "GenericFusion_probability"
    ]:

        values = (
            task_predictions[
                col
            ]
            .to_numpy(
                dtype=np.float64
            )
        )

        assert np.isfinite(
            values
        ).all()

        assert (
            values >= 0
        ).all()

        assert (
            values <= 1
        ).all()


    pred_path = task_prediction_file(
        iteration,
        balanced_dataset
    )

    metric_path = task_metric_file(
        iteration,
        balanced_dataset
    )

    coef_path = task_coef_file(
        iteration,
        balanced_dataset
    )


    atomic_tsv(
        task_predictions,
        pred_path,
        compression="gzip"
    )

    atomic_tsv(
        task_coefficients,
        coef_path
    )

    # Write metrics LAST.
    # This acts as the final task completion marker.
    atomic_tsv(
        task_metrics,
        metric_path
    )


    elapsed = (
        time.time()
        -
        task_start
    ) / 60.0


    print("")
    print("TASK COMPLETE [OK]")
    print("Minutes:", "%.2f" % elapsed)

    print(
        "Generic Context mean AP:",
        "%.6f"
        %
        task_metrics[
            "GenericContext_AP"
        ].mean()
    )

    print(
        "NT mean AP:",
        "%.6f"
        %
        task_metrics[
            "NT_AP"
        ].mean()
    )

    print(
        "Generic Fusion mean AP:",
        "%.6f"
        %
        task_metrics[
            "GenericFusion_AP"
        ].mean()
    )


# ==========================================================================================
# AGGREGATION MODE
# ==========================================================================================

def aggregate_all():

    print(
        "=" * 118
    )

    print(
        "STAGE 8B AGGREGATION-ONLY MODE"
    )

    print(
        "=" * 118
    )


    metric_frames = []
    prediction_frames = []
    coefficient_frames = []


    for iteration in range(
        1,
        11
    ):

        for balanced_dataset in range(
            79
        ):

            mf = task_metric_file(
                iteration,
                balanced_dataset
            )

            pf = task_prediction_file(
                iteration,
                balanced_dataset
            )

            cf = task_coef_file(
                iteration,
                balanced_dataset
            )


            assert mf.exists(), mf
            assert pf.exists(), pf
            assert cf.exists(), cf


            metrics = pd.read_csv(
                str(
                    mf
                ),
                sep="\t"
            )

            predictions = pd.read_csv(
                str(
                    pf
                ),
                sep="\t",
                compression="gzip"
            )

            coefficients = pd.read_csv(
                str(
                    cf
                ),
                sep="\t"
            )


            assert len(
                metrics
            ) == 10

            assert len(
                coefficients
            ) == 10

            assert metrics[
                "fold"
            ].nunique() == 10

            assert coefficients[
                "fold"
            ].nunique() == 10

            assert predictions[
                "dhindsa_row_id"
            ].is_unique


            assert (
                metrics[
                    "runner_version"
                ] == RUNNER_VERSION
            ).all()

            assert (
                metrics[
                    "runner_sha256"
                ] == RUNNER_SHA256
            ).all()

            assert (
                metrics[
                    "feature_list_sha256"
                ] == FEATURE_LIST_SHA256
            ).all()


            metric_frames.append(
                metrics
            )

            prediction_frames.append(
                predictions
            )

            coefficient_frames.append(
                coefficients
            )


    all_metrics = pd.concat(
        metric_frames,
        ignore_index=True
    )

    all_predictions = pd.concat(
        prediction_frames,
        ignore_index=True
    )

    all_coefficients = pd.concat(
        coefficient_frames,
        ignore_index=True
    )


    assert len(
        all_metrics
    ) == 7900

    assert len(
        all_coefficients
    ) == 7900

    assert all_metrics[
        [
            "iteration",
            "balanced_dataset",
            "fold"
        ]
    ].duplicated().sum() == 0


    # --------------------------------------------------------------------------------------
    # SAVE ALL SPLIT OUTPUTS
    # --------------------------------------------------------------------------------------

    ALL_METRICS_FILE = (
        RESULT_DIR /
        "Stage8B_all_7900_fold_metrics.tsv"
    )

    ALL_PREDICTIONS_FILE = (
        RESULT_DIR /
        "Stage8B_all_OOF_predictions.tsv.gz"
    )

    ALL_COEFFICIENTS_FILE = (
        RESULT_DIR /
        "Stage8B_all_7900_meta_coefficients.tsv"
    )


    atomic_tsv(
        all_metrics,
        ALL_METRICS_FILE
    )

    atomic_tsv(
        all_predictions,
        ALL_PREDICTIONS_FILE,
        compression="gzip"
    )

    atomic_tsv(
        all_coefficients,
        ALL_COEFFICIENTS_FILE
    )


    # ======================================================================================
    # SPLIT-LEVEL SUMMARY
    # ======================================================================================

    split_summary = pd.DataFrame(
        [
            {
                "Model":
                    "Dhindsa_GenericOnly_Context_XGBoost",

                "AUROC_mean":
                    all_metrics[
                        "GenericContext_AUROC"
                    ].mean(),

                "AUROC_sd":
                    all_metrics[
                        "GenericContext_AUROC"
                    ].std(),

                "AP_mean":
                    all_metrics[
                        "GenericContext_AP"
                    ].mean(),

                "AP_sd":
                    all_metrics[
                        "GenericContext_AP"
                    ].std()
            },

            {
                "Model":
                    "NTv2_L2_Logistic_FrozenOuter",

                "AUROC_mean":
                    all_metrics[
                        "NT_AUROC"
                    ].mean(),

                "AUROC_sd":
                    all_metrics[
                        "NT_AUROC"
                    ].std(),

                "AP_mean":
                    all_metrics[
                        "NT_AP"
                    ].mean(),

                "AP_sd":
                    all_metrics[
                        "NT_AP"
                    ].std()
            },

            {
                "Model":
                    "Strict_GenericContext_NTv2_LateFusion",

                "AUROC_mean":
                    all_metrics[
                        "GenericFusion_AUROC"
                    ].mean(),

                "AUROC_sd":
                    all_metrics[
                        "GenericFusion_AUROC"
                    ].std(),

                "AP_mean":
                    all_metrics[
                        "GenericFusion_AP"
                    ].mean(),

                "AP_sd":
                    all_metrics[
                        "GenericFusion_AP"
                    ].std()
            }
        ]
    )


    SPLIT_SUMMARY_FILE = (
        RESULT_DIR /
        "Stage8B_split_level_summary.tsv"
    )


    atomic_tsv(
        split_summary,
        SPLIT_SUMMARY_FILE
    )


    # ======================================================================================
    # GENE-LEVEL AGGREGATION
    # ======================================================================================

    gene_scores = (
        all_predictions
        .groupby(
            [
                "Gene_Name",
                "known_gene"
            ],
            as_index=False
        )
        .agg(
            GenericContext_probability=(
                "GenericContext_probability",
                "mean"
            ),

            NT_probability=(
                "NT_probability",
                "mean"
            ),

            GenericFusion_probability=(
                "GenericFusion_probability",
                "mean"
            ),

            OOF_prediction_count=(
                "GenericContext_probability",
                "size"
            )
        )
    )


    assert len(
        gene_scores
    ) == 18252

    assert gene_scores[
        "Gene_Name"
    ].nunique() == 18252

    assert gene_scores.loc[
        gene_scores[
            "known_gene"
        ] == 1,
        "Gene_Name"
    ].nunique() == 190


    y_gene = (
        gene_scores[
            "known_gene"
        ]
        .to_numpy(
            dtype=np.int32
        )
    )


    generic_context_gene_auc = roc_auc_score(
        y_gene,
        gene_scores[
            "GenericContext_probability"
        ]
    )

    generic_context_gene_ap = average_precision_score(
        y_gene,
        gene_scores[
            "GenericContext_probability"
        ]
    )

    nt_gene_auc = roc_auc_score(
        y_gene,
        gene_scores[
            "NT_probability"
        ]
    )

    nt_gene_ap = average_precision_score(
        y_gene,
        gene_scores[
            "NT_probability"
        ]
    )

    generic_fusion_gene_auc = roc_auc_score(
        y_gene,
        gene_scores[
            "GenericFusion_probability"
        ]
    )

    generic_fusion_gene_ap = average_precision_score(
        y_gene,
        gene_scores[
            "GenericFusion_probability"
        ]
    )


    gene_summary = pd.DataFrame(
        [
            {
                "Model":
                    "Dhindsa_GenericOnly_Context_XGBoost",

                "Gene_level_AUROC":
                    generic_context_gene_auc,

                "Gene_level_AP":
                    generic_context_gene_ap
            },

            {
                "Model":
                    "NTv2_L2_Logistic_FrozenOuter",

                "Gene_level_AUROC":
                    nt_gene_auc,

                "Gene_level_AP":
                    nt_gene_ap
            },

            {
                "Model":
                    "Strict_GenericContext_NTv2_LateFusion",

                "Gene_level_AUROC":
                    generic_fusion_gene_auc,

                "Gene_level_AP":
                    generic_fusion_gene_ap
            }
        ]
    )


    GENE_SCORE_FILE = (
        RESULT_DIR /
        "Stage8B_gene_level_OOF_scores.tsv"
    )

    GENE_SUMMARY_FILE = (
        RESULT_DIR /
        "Stage8B_gene_level_summary.tsv"
    )


    atomic_tsv(
        gene_scores,
        GENE_SCORE_FILE
    )

    atomic_tsv(
        gene_summary,
        GENE_SUMMARY_FILE
    )


    # ======================================================================================
    # VERIFY FROZEN NT IS EXACTLY PRESERVED AGAINST ORIGINAL STRICT RUN
    # ======================================================================================

    FULL_METRIC_FILE = (
        FULL_STRICT_RESULT_DIR /
        "StrictFusion_all_7900_fold_metrics.tsv"
    )

    FULL_GENE_FILE = (
        FULL_STRICT_RESULT_DIR /
        "StrictFusion_gene_level_OOF_scores.tsv"
    )


    assert FULL_METRIC_FILE.exists()
    assert FULL_GENE_FILE.exists()


    full_metrics = pd.read_csv(
        str(
            FULL_METRIC_FILE
        ),
        sep="\t"
    )


    nt_check = (
        all_metrics[
            [
                "iteration",
                "balanced_dataset",
                "fold",
                "NT_AUROC",
                "NT_AP"
            ]
        ]
        .merge(
            full_metrics[
                [
                    "iteration",
                    "balanced_dataset",
                    "fold",
                    "NT_AUROC",
                    "NT_AP"
                ]
            ],
            on=[
                "iteration",
                "balanced_dataset",
                "fold"
            ],
            suffixes=(
                "_robust",
                "_full"
            ),
            validate="one_to_one"
        )
    )


    assert len(
        nt_check
    ) == 7900


    assert np.allclose(
        nt_check[
            "NT_AUROC_robust"
        ],
        nt_check[
            "NT_AUROC_full"
        ],
        rtol=0.0,
        atol=1e-12
    )


    assert np.allclose(
        nt_check[
            "NT_AP_robust"
        ],
        nt_check[
            "NT_AP_full"
        ],
        rtol=0.0,
        atol=1e-12
    )


    # Gene-level NT must reproduce frozen result.
    assert np.isclose(
        nt_gene_auc,
        0.899628,
        rtol=0.0,
        atol=5e-6
    )

    assert np.isclose(
        nt_gene_ap,
        0.099593,
        rtol=0.0,
        atol=5e-6
    )


    # ======================================================================================
    # FULL vs GENERIC-ONLY ROBUSTNESS COMPARISON
    # ======================================================================================

    full_gene = pd.read_csv(
        str(
            FULL_GENE_FILE
        ),
        sep="\t"
    )


    y_full_gene = (
        full_gene[
            "known_gene"
        ]
        .to_numpy(
            dtype=np.int32
        )
    )


    full_context_gene_auc = roc_auc_score(
        y_full_gene,
        full_gene[
            "Context_probability"
        ]
    )

    full_context_gene_ap = average_precision_score(
        y_full_gene,
        full_gene[
            "Context_probability"
        ]
    )

    full_fusion_gene_auc = roc_auc_score(
        y_full_gene,
        full_gene[
            "Fusion_probability"
        ]
    )

    full_fusion_gene_ap = average_precision_score(
        y_full_gene,
        full_gene[
            "Fusion_probability"
        ]
    )


    split_comparison = pd.DataFrame(
        [
            {
                "Model":
                    "Full_Context_316",

                "AUROC_mean":
                    full_metrics[
                        "Context_AUROC"
                    ].mean(),

                "AP_mean":
                    full_metrics[
                        "Context_AP"
                    ].mean()
            },

            {
                "Model":
                    "Full_Strict_Fusion",

                "AUROC_mean":
                    full_metrics[
                        "Fusion_AUROC"
                    ].mean(),

                "AP_mean":
                    full_metrics[
                        "Fusion_AP"
                    ].mean()
            },

            {
                "Model":
                    "GenericOnly_Context_266",

                "AUROC_mean":
                    all_metrics[
                        "GenericContext_AUROC"
                    ].mean(),

                "AP_mean":
                    all_metrics[
                        "GenericContext_AP"
                    ].mean()
            },

            {
                "Model":
                    "GenericOnly_Strict_Fusion",

                "AUROC_mean":
                    all_metrics[
                        "GenericFusion_AUROC"
                    ].mean(),

                "AP_mean":
                    all_metrics[
                        "GenericFusion_AP"
                    ].mean()
            },

            {
                "Model":
                    "NTv2",

                "AUROC_mean":
                    all_metrics[
                        "NT_AUROC"
                    ].mean(),

                "AP_mean":
                    all_metrics[
                        "NT_AP"
                    ].mean()
            }
        ]
    )


    gene_comparison = pd.DataFrame(
        [
            {
                "Model":
                    "Full_Context_316",

                "Gene_level_AUROC":
                    full_context_gene_auc,

                "Gene_level_AP":
                    full_context_gene_ap
            },

            {
                "Model":
                    "Full_Strict_Fusion",

                "Gene_level_AUROC":
                    full_fusion_gene_auc,

                "Gene_level_AP":
                    full_fusion_gene_ap
            },

            {
                "Model":
                    "GenericOnly_Context_266",

                "Gene_level_AUROC":
                    generic_context_gene_auc,

                "Gene_level_AP":
                    generic_context_gene_ap
            },

            {
                "Model":
                    "GenericOnly_Strict_Fusion",

                "Gene_level_AUROC":
                    generic_fusion_gene_auc,

                "Gene_level_AP":
                    generic_fusion_gene_ap
            },

            {
                "Model":
                    "NTv2",

                "Gene_level_AUROC":
                    nt_gene_auc,

                "Gene_level_AP":
                    nt_gene_ap
            }
        ]
    )


    SPLIT_COMPARISON_FILE = (
        RESULT_DIR /
        "Stage8B_full_vs_generic_robustness_split_comparison.tsv"
    )

    GENE_COMPARISON_FILE = (
        RESULT_DIR /
        "Stage8B_full_vs_generic_robustness_gene_comparison.tsv"
    )


    atomic_tsv(
        split_comparison,
        SPLIT_COMPARISON_FILE
    )

    atomic_tsv(
        gene_comparison,
        GENE_COMPARISON_FILE
    )


    # ======================================================================================
    # META COEFFICIENT SUMMARY
    # ======================================================================================

    coefficient_summary = pd.DataFrame(
        [
            {
                "Parameter":
                    "GenericContext_coefficient",

                "Mean":
                    all_coefficients[
                        "GenericContext_coefficient"
                    ].mean(),

                "SD":
                    all_coefficients[
                        "GenericContext_coefficient"
                    ].std(),

                "Median":
                    all_coefficients[
                        "GenericContext_coefficient"
                    ].median(),

                "Fraction_positive":
                    (
                        all_coefficients[
                            "GenericContext_coefficient"
                        ] > 0
                    ).mean()
            },

            {
                "Parameter":
                    "NT_coefficient",

                "Mean":
                    all_coefficients[
                        "NT_coefficient"
                    ].mean(),

                "SD":
                    all_coefficients[
                        "NT_coefficient"
                    ].std(),

                "Median":
                    all_coefficients[
                        "NT_coefficient"
                    ].median(),

                "Fraction_positive":
                    (
                        all_coefficients[
                            "NT_coefficient"
                        ] > 0
                    ).mean()
            },

            {
                "Parameter":
                    "Intercept",

                "Mean":
                    all_coefficients[
                        "Intercept"
                    ].mean(),

                "SD":
                    all_coefficients[
                        "Intercept"
                    ].std(),

                "Median":
                    all_coefficients[
                        "Intercept"
                    ].median(),

                "Fraction_positive":
                    (
                        all_coefficients[
                            "Intercept"
                        ] > 0
                    ).mean()
            }
        ]
    )


    COEF_SUMMARY_FILE = (
        RESULT_DIR /
        "Stage8B_meta_coefficient_summary.tsv"
    )


    atomic_tsv(
        coefficient_summary,
        COEF_SUMMARY_FILE
    )


    # ======================================================================================
    # DELTA SUMMARY
    # ======================================================================================

    delta_summary = pd.DataFrame(
        [
            {
                "Comparison":
                    "GenericFusion_minus_GenericContext",

                "Split_AUROC_delta":
                    (
                        all_metrics[
                            "GenericFusion_AUROC"
                        ].mean()
                        -
                        all_metrics[
                            "GenericContext_AUROC"
                        ].mean()
                    ),

                "Split_AP_delta":
                    (
                        all_metrics[
                            "GenericFusion_AP"
                        ].mean()
                        -
                        all_metrics[
                            "GenericContext_AP"
                        ].mean()
                    ),

                "Gene_AUROC_delta":
                    (
                        generic_fusion_gene_auc
                        -
                        generic_context_gene_auc
                    ),

                "Gene_AP_delta":
                    (
                        generic_fusion_gene_ap
                        -
                        generic_context_gene_ap
                    )
            },

            {
                "Comparison":
                    "GenericContext_minus_FullContext",

                "Split_AUROC_delta":
                    (
                        all_metrics[
                            "GenericContext_AUROC"
                        ].mean()
                        -
                        full_metrics[
                            "Context_AUROC"
                        ].mean()
                    ),

                "Split_AP_delta":
                    (
                        all_metrics[
                            "GenericContext_AP"
                        ].mean()
                        -
                        full_metrics[
                            "Context_AP"
                        ].mean()
                    ),

                "Gene_AUROC_delta":
                    (
                        generic_context_gene_auc
                        -
                        full_context_gene_auc
                    ),

                "Gene_AP_delta":
                    (
                        generic_context_gene_ap
                        -
                        full_context_gene_ap
                    )
            },

            {
                "Comparison":
                    "GenericFusion_minus_FullFusion",

                "Split_AUROC_delta":
                    (
                        all_metrics[
                            "GenericFusion_AUROC"
                        ].mean()
                        -
                        full_metrics[
                            "Fusion_AUROC"
                        ].mean()
                    ),

                "Split_AP_delta":
                    (
                        all_metrics[
                            "GenericFusion_AP"
                        ].mean()
                        -
                        full_metrics[
                            "Fusion_AP"
                        ].mean()
                    ),

                "Gene_AUROC_delta":
                    (
                        generic_fusion_gene_auc
                        -
                        full_fusion_gene_auc
                    ),

                "Gene_AP_delta":
                    (
                        generic_fusion_gene_ap
                        -
                        full_fusion_gene_ap
                    )
            }
        ]
    )


    DELTA_FILE = (
        RESULT_DIR /
        "Stage8B_robustness_delta_summary.tsv"
    )


    atomic_tsv(
        delta_summary,
        DELTA_FILE
    )


    # ======================================================================================
    # MANIFEST
    # ======================================================================================

    manifest = {
        "project":
            "ASD-ContextFusion",

        "stage":
            "8B_source_defined_disease_context_reduced_robustness",

        "runner_version":
            RUNNER_VERSION,

        "runner_sha256":
            RUNNER_SHA256,

        "full_context_feature_count":
            316,

        "generic_only_context_feature_count":
            266,

        "filtered_by_disease_features_removed":
            50,

        "robustness_interpretation":
            (
                "Generic-only / disease-context-reduced sensitivity; "
                "not described as completely leakage-free."
            ),

        "frozen_outer_folds":
            7900,

        "balanced_dataset_tasks":
            790,

        "new_PU_sampling":
            False,

        "full_model_rerun":
            False,

        "NT_outer_probabilities_reused":
            True,

        "NT_inner_models":
            (
                "Recomputed only as required for strict inner cross-fitting "
                "using unchanged StandardScaler + L2 logistic specification."
            ),

        "generic_context_model":
            XGB_PARAMS,

        "NT_model":
            NT_PARAMS,

        "meta_model":
            META_PARAMS,

        "raw_feature_concatenation":
            False,

        "sequence_controls_fused":
            False,

        "strict_nested_crossfit":
            True,

        "feature_list":
            str(
                GENERIC_FEATURE_FILE
            ),

        "feature_list_sha256":
            FEATURE_LIST_SHA256,

        "gene_count":
            int(
                len(
                    gene_scores
                )
            ),

        "ASD_gene_count":
            int(
                gene_scores.loc[
                    gene_scores[
                        "known_gene"
                    ] == 1,
                    "Gene_Name"
                ].nunique()
            )
    }


    MANIFEST_FILE = (
        RESULT_DIR /
        "Stage8B_robustness_manifest.json"
    )


    with open(
        str(
            MANIFEST_FILE
        ),
        "w"
    ) as f:

        json.dump(
            manifest,
            f,
            indent=2
        )


    # ======================================================================================
    # FINAL REPORT
    # ======================================================================================

    print("")
    print("=" * 118)
    print("STAGE 8B GENERIC-ONLY STRICT ROBUSTNESS COMPLETE")
    print("=" * 118)

    print("")
    print("SPLIT-LEVEL PERFORMANCE")
    print(
        split_summary.to_string(
            index=False
        )
    )

    print("")
    print("GENE-LEVEL PERFORMANCE")
    print(
        gene_summary.to_string(
            index=False
        )
    )

    print("")
    print("FULL vs GENERIC-ONLY COMPARISON — SPLIT")
    print(
        split_comparison.to_string(
            index=False
        )
    )

    print("")
    print("FULL vs GENERIC-ONLY COMPARISON — GENE")
    print(
        gene_comparison.to_string(
            index=False
        )
    )

    print("")
    print("ROBUSTNESS DELTAS")
    print(
        delta_summary.to_string(
            index=False
        )
    )

    print("")
    print("META COEFFICIENTS")
    print(
        coefficient_summary.to_string(
            index=False
        )
    )

    print("")
    print("INTEGRITY")
    print("Robustness tasks        : 790/790")
    print("Frozen outer folds      : 7900")
    print("Genes scored            : 18252")
    print("ASD genes               : 190")
    print("Generic features        : 266")
    print("Disease-filtered removed: 50")
    print("Frozen NT reproduced    : YES")
    print("New PU sampling         : NO")
    print("Full Context rerun      : NO")
    print("Hyperparameter tuning   : NO")
    print("Raw concatenation       : NO")
    print("Sequence controls fused : NO")
    print("Strict nested cross-fit : YES")

    print("")
    print("NEXT:")
    print(
        "Paired robustness inference: "
        "GenericFusion vs GenericContext, "
        "plus full-vs-generic sensitivity contrasts."
    )

    print(
        "=" * 118
    )


# ==========================================================================================
# ENTRY POINT
# ==========================================================================================

task_iteration = os.environ.get(
    "ASD_CF_ROBUST_ITERATION"
)

task_balanced_dataset = os.environ.get(
    "ASD_CF_ROBUST_BALANCED_DATASET"
)


if (
    task_iteration is not None
    and
    task_balanced_dataset is not None
):

    run_task(
        int(
            task_iteration
        ),
        int(
            task_balanced_dataset
        )
    )


elif (
    task_iteration is None
    and
    task_balanced_dataset is None
):

    aggregate_all()


else:

    raise RuntimeError(
        "Both ASD_CF_ROBUST_ITERATION and "
        "ASD_CF_ROBUST_BALANCED_DATASET must be provided together."
    )
"""


RUNNER.write_text(
    textwrap.dedent(
        runner_code
    ),
    encoding="utf-8"
)


RUNNER_SHA = sha256_file(
    RUNNER
)


print(
    "Runner:"
)

print(
    RUNNER
)

print(
    "\nRunner SHA256:"
)

print(
    RUNNER_SHA
)


# Save runner creation provenance
runner_manifest = {
    "project":
        "ASD-ContextFusion",

    "stage":
        "8B",

    "runner_version":
        RUNNER_VERSION,

    "runner":
        str(
            RUNNER
        ),

    "runner_sha256":
        RUNNER_SHA,

    "generic_feature_file":
        str(
            GENERIC_FEATURE_FILE
        ),

    "generic_feature_sha256":
        GENERIC_FEATURE_SHA,

    "generic_feature_count":
        266,

    "removed_filtered_by_disease_count":
        50,

    "methodological_change":
        (
            "Source-defined removal of the already-frozen "
            "filtered-by-disease feature family only."
        ),

    "models_or_hyperparameters_changed":
        False,

    "new_PU_sampling":
        False,

    "strict_nested_crossfit":
        True
}


with open(
    RUNNER_MANIFEST,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        runner_manifest,
        f,
        indent=2
    )


# ==========================================================================================
# 5. LOAD FROZEN TASK IDENTITIES + STRICT CHECKPOINT VALIDATOR
# ==========================================================================================

print("\n[4/7] AUDITING EXISTING STAGE 8B CHECKPOINTS")


expected_ids = {}


for iteration in range(
    1,
    11
):

    membership_file = (
        PARTITION_DIR /
        f"iteration_{iteration:02d}_test_fold_membership.tsv.gz"
    )


    membership = pd.read_csv(
        membership_file,
        sep="\t",
        compression="gzip",
        usecols=[
            "balanced_dataset",
            "fold",
            "dhindsa_row_id"
        ]
    )


    for balanced_dataset, df in membership.groupby(
        "balanced_dataset",
        sort=True
    ):

        expected_ids[
            (
                iteration,
                int(
                    balanced_dataset
                )
            )
        ] = set(
            df[
                "dhindsa_row_id"
            ]
            .astype(int)
            .tolist()
        )


assert len(
    expected_ids
) == 790


def pred_file(
    iteration,
    balanced_dataset
):

    return (
        CHECKPOINT_DIR /
        (
            f"iteration_{iteration:02d}_balanced_"
            f"{balanced_dataset:02d}_robust_predictions.tsv.gz"
        )
    )


def metric_file(
    iteration,
    balanced_dataset
):

    return (
        CHECKPOINT_DIR /
        (
            f"iteration_{iteration:02d}_balanced_"
            f"{balanced_dataset:02d}_robust_metrics.tsv"
        )
    )


def coef_file(
    iteration,
    balanced_dataset
):

    return (
        CHECKPOINT_DIR /
        (
            f"iteration_{iteration:02d}_balanced_"
            f"{balanced_dataset:02d}_meta_coefficients.tsv"
        )
    )


def source_nt_file(
    iteration,
    balanced_dataset
):

    return (
        FULL_STRICT_CHECKPOINT_DIR /
        (
            f"iteration_{iteration:02d}_balanced_"
            f"{balanced_dataset:02d}_fusion_predictions.tsv.gz"
        )
    )


def task_valid(
    iteration,
    balanced_dataset
):

    pf = pred_file(
        iteration,
        balanced_dataset
    )

    mf = metric_file(
        iteration,
        balanced_dataset
    )

    cf = coef_file(
        iteration,
        balanced_dataset
    )


    if not (
        pf.exists()
        and
        mf.exists()
        and
        cf.exists()
    ):

        return False


    try:

        metrics = pd.read_csv(
            mf,
            sep="\t"
        )

        coefficients = pd.read_csv(
            cf,
            sep="\t"
        )

        predictions = pd.read_csv(
            pf,
            sep="\t",
            compression="gzip"
        )


        if len(
            metrics
        ) != 10:

            return False


        if metrics[
            "fold"
        ].nunique() != 10:

            return False


        if len(
            coefficients
        ) != 10:

            return False


        if coefficients[
            "fold"
        ].nunique() != 10:

            return False


        required_pred_cols = {
            "iteration",
            "balanced_dataset",
            "fold",
            "dhindsa_row_id",
            "Gene_Name",
            "known_gene",
            "GenericContext_probability",
            "NT_probability",
            "GenericFusion_probability"
        }


        if not required_pred_cols.issubset(
            set(
                predictions.columns
            )
        ):

            return False


        if not predictions[
            "dhindsa_row_id"
        ].is_unique:

            return False


        observed_ids = set(
            predictions[
                "dhindsa_row_id"
            ]
            .astype(int)
            .tolist()
        )


        if observed_ids != expected_ids[
            (
                iteration,
                balanced_dataset
            )
        ]:

            return False


        for col in [
            "GenericContext_probability",
            "NT_probability",
            "GenericFusion_probability"
        ]:

            values = predictions[
                col
            ].to_numpy(
                dtype=float
            )


            if not np.isfinite(
                values
            ).all():

                return False


            if not (
                (
                    values >= 0
                )
                &
                (
                    values <= 1
                )
            ).all():

                return False


        # Provenance must match CURRENT runner + CURRENT locked feature list.
        if "runner_sha256" not in metrics.columns:

            return False

        if "feature_list_sha256" not in metrics.columns:

            return False


        if not (
            metrics[
                "runner_sha256"
            ] == RUNNER_SHA
        ).all():

            return False


        if not (
            metrics[
                "feature_list_sha256"
            ] == GENERIC_FEATURE_SHA
        ).all():

            return False


        # Frozen NT source must remain EXACTLY identical.
        sf = source_nt_file(
            iteration,
            balanced_dataset
        )


        if not sf.exists():

            return False


        source = pd.read_csv(
            sf,
            sep="\t",
            compression="gzip",
            usecols=[
                "dhindsa_row_id",
                "NT_probability"
            ]
        )


        check = predictions[
            [
                "dhindsa_row_id",
                "NT_probability"
            ]
        ].merge(
            source,
            on="dhindsa_row_id",
            how="inner",
            suffixes=(
                "_robust",
                "_source"
            ),
            validate="one_to_one"
        )


        if len(
            check
        ) != len(
            predictions
        ):

            return False


        if not np.allclose(
            check[
                "NT_probability_robust"
            ],
            check[
                "NT_probability_source"
            ],
            rtol=0.0,
            atol=1e-12
        ):

            return False


        return True


    except Exception:

        return False


all_tasks = [
    (
        iteration,
        balanced_dataset
    )
    for iteration in range(
        1,
        11
    )
    for balanced_dataset in range(
        79
    )
]


completed = []

remaining = []


for task in all_tasks:

    iteration, balanced_dataset = task

    if task_valid(
        iteration,
        balanced_dataset
    ):

        completed.append(
            task
        )

    else:

        remaining.append(
            task
        )


print(
    "Completed valid Stage 8B tasks:",
    len(
        completed
    ),
    "/ 790"
)

print(
    "Remaining:",
    len(
        remaining
    )
)


# ==========================================================================================
# 6. PARALLEL CHECKPOINT-SAFE EXECUTION
# ==========================================================================================

print("\n[5/7] RUNNING REMAINING STAGE 8B TASKS")


env_base = os.environ.copy()

env_base[
    "PYTHONIOENCODING"
] = "utf-8"

env_base[
    "PYTHONUTF8"
] = "1"


# Engineering-only thread controls.
for variable in [
    "OMP_NUM_THREADS",
    "MKL_NUM_THREADS",
    "OPENBLAS_NUM_THREADS",
    "NUMEXPR_NUM_THREADS"
]:

    env_base[
        variable
    ] = "1"


print(
    "Parallel workers:",
    MAX_WORKERS
)

print(
    "Each XGBoost model remains nthread=1."
)

print(
    "Scientific models/hyperparameters unchanged."
)

print(
    "Only independent task scheduling is parallelized."
)


def run_one_task(
    task
):

    iteration, balanced_dataset = task


    if task_valid(
        iteration,
        balanced_dataset
    ):

        return {
            "iteration":
                iteration,

            "balanced_dataset":
                balanced_dataset,

            "status":
                "SKIPPED"
        }


    env = env_base.copy()

    env[
        "ASD_CF_ROBUST_ITERATION"
    ] = str(
        iteration
    )

    env[
        "ASD_CF_ROBUST_BALANCED_DATASET"
    ] = str(
        balanced_dataset
    )


    log_file = (
        LOG_DIR /
        (
            f"iteration_{iteration:02d}_"
            f"balanced_{balanced_dataset:02d}.log"
        )
    )


    start = time.time()


    with open(
        log_file,
        "w",
        encoding="utf-8"
    ) as log:

        process = subprocess.Popen(
            [
                str(
                    LEGACY_PYTHON
                ),
                "-u",
                str(
                    RUNNER
                )
            ],
            cwd=str(
                REPO
            ),
            stdout=log,
            stderr=subprocess.STDOUT,
            env=env
        )


        return_code = process.wait()


    elapsed = (
        time.time()
        -
        start
    )


    if return_code != 0:

        return {
            "iteration":
                iteration,

            "balanced_dataset":
                balanced_dataset,

            "status":
                "FAILED",

            "minutes":
                elapsed / 60.0,

            "log":
                str(
                    log_file
                )
        }


    if not task_valid(
        iteration,
        balanced_dataset
    ):

        return {
            "iteration":
                iteration,

            "balanced_dataset":
                balanced_dataset,

            "status":
                "INVALID_CHECKPOINT",

            "minutes":
                elapsed / 60.0,

            "log":
                str(
                    log_file
                )
        }


    metrics = pd.read_csv(
        metric_file(
            iteration,
            balanced_dataset
        ),
        sep="\t"
    )


    return {
        "iteration":
            iteration,

        "balanced_dataset":
            balanced_dataset,

        "status":
            "COMPLETE",

        "minutes":
            elapsed / 60.0,

        "GenericContext_AP":
            metrics[
                "GenericContext_AP"
            ].mean(),

        "NT_AP":
            metrics[
                "NT_AP"
            ].mean(),

        "GenericFusion_AP":
            metrics[
                "GenericFusion_AP"
            ].mean(),

        "log":
            str(
                log_file
            )
    }


failed_tasks = []


if remaining:

    start_parallel = time.time()


    with ThreadPoolExecutor(
        max_workers=MAX_WORKERS
    ) as executor:

        future_to_task = {
            executor.submit(
                run_one_task,
                task
            ):
                task
            for task
            in remaining
        }


        done_total = len(
            completed
        )


        for future in as_completed(
            future_to_task
        ):

            result = future.result()

            done_total += 1


            if result[
                "status"
            ] == "COMPLETE":

                elapsed_hours = (
                    time.time()
                    -
                    start_parallel
                ) / 3600.0


                print(
                    f"\n[{done_total}/790] "
                    f"ITER {result['iteration']:02d} | "
                    f"BAL {result['balanced_dataset']:02d} "
                    f"COMPLETE [OK]"
                )

                print(
                    "   Task time        :",
                    f"{result['minutes']:.1f} min"
                )

                print(
                    "   Generic Context AP:",
                    f"{result['GenericContext_AP']:.4f}"
                )

                print(
                    "   NT AP             :",
                    f"{result['NT_AP']:.4f}"
                )

                print(
                    "   Generic Fusion AP :",
                    f"{result['GenericFusion_AP']:.4f}"
                )

                print(
                    "   Parallel elapsed  :",
                    f"{elapsed_hours:.2f} h"
                )


            elif result[
                "status"
            ] == "SKIPPED":

                print(
                    f"\nITER {result['iteration']:02d} | "
                    f"BAL {result['balanced_dataset']:02d}: "
                    "already valid -> SKIPPED"
                )


            else:

                failed_tasks.append(
                    result
                )

                print(
                    "\nTASK ERROR:",
                    result
                )


else:

    print(
        "All 790 Stage 8B tasks already valid -> no task rerun."
    )


# ==========================================================================================
# 7. FINAL AUDIT
# ==========================================================================================

print("\n[6/7] FINAL 790-TASK STAGE 8B AUDIT")


valid_tasks = []


for iteration in range(
    1,
    11
):

    for balanced_dataset in range(
        79
    ):

        if task_valid(
            iteration,
            balanced_dataset
        ):

            valid_tasks.append(
                (
                    iteration,
                    balanced_dataset
                )
            )


print(
    "Valid Stage 8B tasks:",
    len(
        valid_tasks
    ),
    "/ 790"
)


if failed_tasks:

    print(
        "\nFailed/invalid tasks:",
        len(
            failed_tasks
        )
    )

    for item in failed_tasks[
        :10
    ]:

        print(
            item
        )


if len(
    valid_tasks
) != 790:

    print(
        "\nNOT ALL TASKS COMPLETE."
    )

    print(
        "Do NOT delete checkpoints."
    )

    print(
        "Rerun THIS SAME CELL."
    )

    print(
        "Valid checkpoints will automatically be skipped."
    )

    raise RuntimeError(
        f"Only {len(valid_tasks)}/790 Stage 8B tasks valid."
    )


print(
    "\nALL 790 STAGE 8B TASKS VERIFIED [OK]"
)


# ==========================================================================================
# 8. FINAL AGGREGATION
# ==========================================================================================

print("\n[7/7] FINAL STAGE 8B AGGREGATION")


aggregation_log = (
    LOG_DIR /
    "FINAL_AGGREGATION.log"
)


final_env = env_base.copy()

final_env.pop(
    "ASD_CF_ROBUST_ITERATION",
    None
)

final_env.pop(
    "ASD_CF_ROBUST_BALANCED_DATASET",
    None
)


with open(
    aggregation_log,
    "w",
    encoding="utf-8"
) as log:

    process = subprocess.Popen(
        [
            str(
                LEGACY_PYTHON
            ),
            "-u",
            str(
                RUNNER
            )
        ],
        cwd=str(
            REPO
        ),
        stdout=subprocess.PIPE,
        stderr=subprocess.STDOUT,
        text=True,
        encoding="utf-8",
        errors="replace",
        bufsize=1,
        env=final_env
    )


    for line in process.stdout:

        log.write(
            line
        )

        log.flush()

        print(
            line,
            end=""
        )


    return_code = process.wait()


if return_code != 0:

    raise RuntimeError(
        "\nStage 8B aggregation failed.\n"
        "All 790 task checkpoints remain safe.\n"
        f"Check:\n{aggregation_log}"
    )


# ==========================================================================================
# 9. FINAL OUTPUT VERIFICATION
# ==========================================================================================

FINAL_METRICS = (
    RESULT_DIR /
    "Stage8B_all_7900_fold_metrics.tsv"
)

FINAL_GENE_SCORES = (
    RESULT_DIR /
    "Stage8B_gene_level_OOF_scores.tsv"
)

FINAL_SPLIT_SUMMARY = (
    RESULT_DIR /
    "Stage8B_split_level_summary.tsv"
)

FINAL_GENE_SUMMARY = (
    RESULT_DIR /
    "Stage8B_gene_level_summary.tsv"
)

FINAL_DELTA = (
    RESULT_DIR /
    "Stage8B_robustness_delta_summary.tsv"
)


for path in [
    FINAL_METRICS,
    FINAL_GENE_SCORES,
    FINAL_SPLIT_SUMMARY,
    FINAL_GENE_SUMMARY,
    FINAL_DELTA
]:

    assert path.exists(), (
        f"Missing final Stage 8B output:\n{path}"
    )


final_metrics = pd.read_csv(
    FINAL_METRICS,
    sep="\t"
)

final_gene_scores = pd.read_csv(
    FINAL_GENE_SCORES,
    sep="\t"
)


assert len(
    final_metrics
) == 7900

assert len(
    final_gene_scores
) == 18252

assert final_gene_scores[
    "Gene_Name"
].nunique() == 18252

assert final_gene_scores.loc[
    final_gene_scores[
        "known_gene"
    ] == 1,
    "Gene_Name"
].nunique() == 190


print("\n" + "=" * 126)
print("STAGE 8B COMPLETE + VERIFIED [OK]")
print("=" * 126)

print("\nFINAL SPLIT SUMMARY:")

print(
    pd.read_csv(
        FINAL_SPLIT_SUMMARY,
        sep="\t"
    ).to_string(
        index=False
    )
)

print("\nFINAL GENE SUMMARY:")

print(
    pd.read_csv(
        FINAL_GENE_SUMMARY,
        sep="\t"
    ).to_string(
        index=False
    )
)

print("\nROBUSTNESS DELTAS:")

print(
    pd.read_csv(
        FINAL_DELTA,
        sep="\t"
    ).to_string(
        index=False
    )
)

print("\nFINAL INTEGRITY:")
print("Stage 8B tasks             : 790/790 [OK]")
print("Frozen outer folds         : 7900 [OK]")
print("Genes                      : 18252 [OK]")
print("ASD genes                  : 190 [OK]")
print("Full Context features      : 316 [frozen; NOT rerun]")
print("Generic-only features      : 266 [locked]")
print("Disease-filtered removed   : 50 [locked]")
print("NT outer predictions       : frozen source [exactly verified]")
print("New PU sampling            : NO")
print("Hyperparameter tuning      : NO")
print("Raw concatenation          : NO")
print("Sequence controls fused    : NO")
print("Strict nested cross-fit    : YES")
print("Feature definitions changed: NO")

print("\nNEXT SCIENTIFIC STAGE:")
print(
    "Stage 8C — paired robustness inference "
    "(Generic Fusion vs Generic Context + full-vs-generic contrasts)."
)

print("=" * 126)
```

```python
# ==========================================================================================
# ASD-ContextFusion
# STAGE 8C — PAIRED ROBUSTNESS INFERENCE
#
# PRIMARY QUESTION
# ----------------
# Does strict Context+NT-v2 fusion retain complementary predictive value
# after removal of the authors' filtered-by-disease contextual feature family?
#
# PRIMARY INFERENCE
# -----------------
# Gene-level paired stratified bootstrap:
#   18,252 unique genes
#   190 ASD positives
#   18,062 unlabeled
#   5,000 bootstrap replicates
#
# SUPPORTING INFERENCE
# --------------------
# Exact paired sign-flip test across the 10 stochastic iterations.
#
# IMPORTANT
# ---------
# - NO model fitting
# - NO feature changes
# - NO new PU sampling
# - NO tuning
# - NO naive 7,900-fold p-values
#
# ==========================================================================================

from pathlib import Path
from itertools import product
import hashlib
import json
import time

import numpy as np
import pandas as pd

from sklearn.metrics import (
    roc_auc_score,
    average_precision_score
)


# ==========================================================================================
# 0. SETTINGS
# ==========================================================================================

ROOT = Path(r"E:\ASD_ContextFusion_Q1")

FULL_DIR = (
    ROOT /
    "05_results" /
    "proposed" /
    "strict_context_NTv2_fusion"
)

ROBUST_DIR = (
    ROOT /
    "05_results" /
    "proposed" /
    "stage8B_generic_context_strict_fusion"
)

LOCK_FILE = (
    ROOT /
    "00_admin" /
    "ASD_ContextFusion_STAGE8A1_SOURCE_DEFINED_ROBUSTNESS_LOCK.json"
)

OUT_DIR = (
    ROOT /
    "05_results" /
    "proposed" /
    "stage8C_paired_robustness_inference"
)

OUT_DIR.mkdir(
    parents=True,
    exist_ok=True
)


FULL_METRICS_FILE = (
    FULL_DIR /
    "StrictFusion_all_7900_fold_metrics.tsv"
)

FULL_GENE_FILE = (
    FULL_DIR /
    "StrictFusion_gene_level_OOF_scores.tsv"
)

ROBUST_METRICS_FILE = (
    ROBUST_DIR /
    "Stage8B_all_7900_fold_metrics.tsv"
)

ROBUST_GENE_FILE = (
    ROBUST_DIR /
    "Stage8B_gene_level_OOF_scores.tsv"
)


N_BOOTSTRAP = 5000
BOOTSTRAP_SEED = 20260830
BOOTSTRAP_BATCH = 100

ANALYSIS_VERSION = (
    "ASD_CF_STAGE8C_PAIRED_ROBUSTNESS_V1"
)


BOOT_FILE = (
    OUT_DIR /
    "Stage8C_gene_level_paired_bootstrap_replicates.tsv.gz"
)

BOOT_SUMMARY_FILE = (
    OUT_DIR /
    "Stage8C_gene_level_paired_bootstrap_summary.tsv"
)

ITER_DELTA_FILE = (
    OUT_DIR /
    "Stage8C_iteration_level_paired_deltas.tsv"
)

ITER_SUMMARY_FILE = (
    OUT_DIR /
    "Stage8C_iteration_exact_signflip_summary.tsv"
)

TASK_DESC_FILE = (
    OUT_DIR /
    "Stage8C_balanced_dataset_descriptive_summary.tsv"
)

OBSERVED_FILE = (
    OUT_DIR /
    "Stage8C_observed_robustness_performance.tsv"
)

MANIFEST_FILE = (
    OUT_DIR /
    "Stage8C_paired_robustness_manifest.json"
)


# ==========================================================================================
# 1. HELPERS
# ==========================================================================================

def sha256_file(
    path,
    chunk_size=1024 * 1024
):

    h = hashlib.sha256()

    with open(
        path,
        "rb"
    ) as f:

        while True:

            chunk = f.read(
                chunk_size
            )

            if not chunk:
                break

            h.update(
                chunk
            )

    return h.hexdigest()


def exact_sign_flip_pvalue(
    differences
):

    d = np.asarray(
        differences,
        dtype=float
    )

    d = d[
        np.isfinite(
            d
        )
    ]

    if len(
        d
    ) == 0:

        return np.nan


    observed = abs(
        np.mean(
            d
        )
    )


    signs = np.asarray(
        list(
            product(
                [
                    -1.0,
                    1.0
                ],
                repeat=len(
                    d
                )
            )
        ),
        dtype=float
    )


    null_stats = np.abs(
        (
            signs *
            d.reshape(
                1,
                -1
            )
        ).mean(
            axis=1
        )
    )


    return float(
        np.mean(
            null_stats
            >=
            observed - 1e-15
        )
    )


def bh_adjust(
    pvalues
):

    p = np.asarray(
        pvalues,
        dtype=float
    )

    m = len(
        p
    )

    order = np.argsort(
        p
    )

    ranked = p[
        order
    ]


    adjusted = (
        ranked *
        m /
        np.arange(
            1,
            m + 1
        )
    )


    adjusted = np.minimum.accumulate(
        adjusted[
            ::-1
        ]
    )[
        ::-1
    ]


    adjusted = np.clip(
        adjusted,
        0,
        1
    )


    result = np.empty_like(
        adjusted
    )

    result[
        order
    ] = adjusted


    return result


def calculate_metrics(
    y,
    score
):

    return {
        "AP":
            average_precision_score(
                y,
                score
            ),

        "AUROC":
            roc_auc_score(
                y,
                score
            )
    }


# ==========================================================================================
# 2. INPUT + SCIENTIFIC LOCK AUDIT
# ==========================================================================================

print(
    "=" * 126
)

print(
    "ASD-CONTEXTFUSION — STAGE 8C PAIRED ROBUSTNESS INFERENCE"
)

print(
    "=" * 126
)


print(
    "\n[1/9] INPUT / SCIENTIFIC LOCK AUDIT"
)


for path in [
    FULL_METRICS_FILE,
    FULL_GENE_FILE,
    ROBUST_METRICS_FILE,
    ROBUST_GENE_FILE,
    LOCK_FILE
]:

    assert path.exists(), (
        f"Missing required file:\n{path}"
    )

    print(
        "[OK]",
        path
    )


with open(
    LOCK_FILE,
    "r",
    encoding="utf-8"
) as f:

    lock = json.load(
        f
    )


assert lock[
    "full_context_feature_count"
] == 316

assert lock[
    "generic_only_sensitivity_feature_count"
] == 266

assert lock[
    "filtered_by_disease_removed_feature_count"
] == 50

assert lock[
    "source_overlap"
] == 0

assert lock[
    "source_unresolved"
] == 0


print(
    "\nFeature robustness lock verified [OK]"
)


# ==========================================================================================
# 3. LOAD FROZEN RESULTS
# ==========================================================================================

print(
    "\n[2/9] LOADING FROZEN FULL + ROBUSTNESS RESULTS"
)


full_metrics = pd.read_csv(
    FULL_METRICS_FILE,
    sep="\t"
)

robust_metrics = pd.read_csv(
    ROBUST_METRICS_FILE,
    sep="\t"
)

full_gene = pd.read_csv(
    FULL_GENE_FILE,
    sep="\t"
)

robust_gene = pd.read_csv(
    ROBUST_GENE_FILE,
    sep="\t"
)


assert len(
    full_metrics
) == 7900

assert len(
    robust_metrics
) == 7900

assert len(
    full_gene
) == 18252

assert len(
    robust_gene
) == 18252


print(
    "Full metrics  :",
    full_metrics.shape
)

print(
    "Robust metrics:",
    robust_metrics.shape
)

print(
    "Full genes    :",
    full_gene.shape
)

print(
    "Robust genes  :",
    robust_gene.shape
)


# ==========================================================================================
# 4. SPLIT IDENTITY + FROZEN NT VERIFICATION
# ==========================================================================================

print(
    "\n[3/9] FROZEN SPLIT + NT IDENTITY AUDIT"
)


KEYS = [
    "iteration",
    "balanced_dataset",
    "fold"
]


for key in KEYS:

    assert key in full_metrics.columns
    assert key in robust_metrics.columns


split = (
    full_metrics[
        KEYS +
        [
            "Context_AUROC",
            "Context_AP",
            "NT_AUROC",
            "NT_AP",
            "Fusion_AUROC",
            "Fusion_AP"
        ]
    ]
    .merge(
        robust_metrics[
            KEYS +
            [
                "GenericContext_AUROC",
                "GenericContext_AP",
                "NT_AUROC",
                "NT_AP",
                "GenericFusion_AUROC",
                "GenericFusion_AP"
            ]
        ],
        on=KEYS,
        how="inner",
        suffixes=(
            "_full",
            "_robust"
        ),
        validate="one_to_one"
    )
)


assert len(
    split
) == 7900


assert np.allclose(
    split[
        "NT_AUROC_full"
    ],
    split[
        "NT_AUROC_robust"
    ],
    rtol=0,
    atol=1e-12
)


assert np.allclose(
    split[
        "NT_AP_full"
    ],
    split[
        "NT_AP_robust"
    ],
    rtol=0,
    atol=1e-12
)


print(
    "7,900 frozen fold identities: MATCH [OK]"
)

print(
    "Frozen NT split results      : EXACT MATCH [OK]"
)


# ==========================================================================================
# 5. GENE IDENTITY + SCORES
# ==========================================================================================

print(
    "\n[4/9] GENE-LEVEL IDENTITY AUDIT"
)


required_full_gene = {
    "Gene_Name",
    "known_gene",
    "Context_probability",
    "NT_probability",
    "Fusion_probability"
}


required_robust_gene = {
    "Gene_Name",
    "known_gene",
    "GenericContext_probability",
    "NT_probability",
    "GenericFusion_probability"
}


assert required_full_gene.issubset(
    full_gene.columns
)

assert required_robust_gene.issubset(
    robust_gene.columns
)


gene = (
    full_gene[
        [
            "Gene_Name",
            "known_gene",
            "Context_probability",
            "NT_probability",
            "Fusion_probability"
        ]
    ]
    .rename(
        columns={
            "Context_probability":
                "FullContext",

            "NT_probability":
                "NT_full",

            "Fusion_probability":
                "FullFusion"
        }
    )
    .merge(
        robust_gene[
            [
                "Gene_Name",
                "known_gene",
                "GenericContext_probability",
                "NT_probability",
                "GenericFusion_probability"
            ]
        ].rename(
            columns={
                "GenericContext_probability":
                    "GenericContext",

                "NT_probability":
                    "NT_robust",

                "GenericFusion_probability":
                    "GenericFusion"
            }
        ),
        on=[
            "Gene_Name",
            "known_gene"
        ],
        how="inner",
        validate="one_to_one"
    )
)


assert len(
    gene
) == 18252

assert gene[
    "Gene_Name"
].nunique() == 18252

assert (
    gene[
        "known_gene"
    ].sum()
    == 190
)


assert np.allclose(
    gene[
        "NT_full"
    ],
    gene[
        "NT_robust"
    ],
    rtol=0,
    atol=1e-12
)


print(
    "Unique genes    :",
    gene[
        "Gene_Name"
    ].nunique()
)

print(
    "ASD genes       :",
    int(
        gene[
            "known_gene"
        ].sum()
    )
)

print(
    "Frozen NT genes : EXACT MATCH [OK]"
)


# ==========================================================================================
# 6. OBSERVED PERFORMANCE
# ==========================================================================================

print(
    "\n[5/9] OBSERVED PERFORMANCE"
)


y_gene = gene[
    "known_gene"
].to_numpy(
    dtype=np.int32
)


MODEL_COLUMNS = {
    "FullContext":
        "FullContext",

    "FullFusion":
        "FullFusion",

    "GenericContext":
        "GenericContext",

    "GenericFusion":
        "GenericFusion",

    "NTv2":
        "NT_full"
}


observed = {}


observed_rows = []


for model, column in MODEL_COLUMNS.items():

    metrics = calculate_metrics(
        y_gene,
        gene[
            column
        ].to_numpy(
            dtype=float
        )
    )


    observed[
        model
    ] = metrics


    observed_rows.append(
        {
            "Model":
                model,

            "Gene_level_AUROC":
                metrics[
                    "AUROC"
                ],

            "Gene_level_AP":
                metrics[
                    "AP"
                ]
        }
    )


observed_df = pd.DataFrame(
    observed_rows
)


EXPECTED = {
    "FullContext":
        (
            0.947076,
            0.182366
        ),

    "FullFusion":
        (
            0.951591,
            0.196525
        ),

    "GenericContext":
        (
            0.943594,
            0.174316
        ),

    "GenericFusion":
        (
            0.949912,
            0.189659
        ),

    "NTv2":
        (
            0.899628,
            0.099593
        )
}


for _, row in observed_df.iterrows():

    model = row[
        "Model"
    ]

    expected_auc, expected_ap = (
        EXPECTED[
            model
        ]
    )


    assert np.isclose(
        row[
            "Gene_level_AUROC"
        ],
        expected_auc,
        atol=5e-6,
        rtol=0
    )


    assert np.isclose(
        row[
            "Gene_level_AP"
        ],
        expected_ap,
        atol=5e-6,
        rtol=0
    )


observed_df.to_csv(
    OBSERVED_FILE,
    sep="\t",
    index=False
)


print(
    observed_df.to_string(
        index=False
    )
)


# ==========================================================================================
# 7. DEFINE PRESPECIFIED ROBUSTNESS CONTRASTS
# ==========================================================================================

COMPARISONS = [
    {
        "name":
            "GenericFusion_vs_GenericContext",

        "role":
            "PRIMARY_ROBUSTNESS",

        "type":
            "simple",

        "A":
            "GenericFusion",

        "B":
            "GenericContext"
    },

    {
        "name":
            "GenericContext_vs_FullContext",

        "role":
            "SENSITIVITY_COST_CONTEXT",

        "type":
            "simple",

        "A":
            "GenericContext",

        "B":
            "FullContext"
    },

    {
        "name":
            "GenericFusion_vs_FullFusion",

        "role":
            "SENSITIVITY_COST_FUSION",

        "type":
            "simple",

        "A":
            "GenericFusion",

        "B":
            "FullFusion"
    },

    {
        "name":
            "IncrementalGain_Generic_vs_Full",

        "role":
            "COMPLEMENTARITY_DIFFERENCE_IN_DIFFERENCES",

        "type":
            "difference_in_differences"
    }
]


def observed_delta(
    comparison,
    metric,
    perf
):

    if (
        comparison[
            "type"
        ]
        ==
        "simple"
    ):

        return (
            perf[
                comparison[
                    "A"
                ]
            ][
                metric
            ]
            -
            perf[
                comparison[
                    "B"
                ]
            ][
                metric
            ]
        )


    if (
        comparison[
            "type"
        ]
        ==
        "difference_in_differences"
    ):

        generic_gain = (
            perf[
                "GenericFusion"
            ][
                metric
            ]
            -
            perf[
                "GenericContext"
            ][
                metric
            ]
        )


        full_gain = (
            perf[
                "FullFusion"
            ][
                metric
            ]
            -
            perf[
                "FullContext"
            ][
                metric
            ]
        )


        return (
            generic_gain
            -
            full_gain
        )


    raise RuntimeError(
        "Unknown comparison type."
    )


print(
    "\nOBSERVED ROBUSTNESS CONTRASTS"
)


for comparison in COMPARISONS:

    for metric in [
        "AP",
        "AUROC"
    ]:

        d = observed_delta(
            comparison,
            metric,
            observed
        )

        print(
            f"{comparison['name']:42s} "
            f"{metric:5s} "
            f"delta = {d:+.6f}"
        )


# ==========================================================================================
# 8. SUPPORTING ITERATION-LEVEL EXACT PAIRED TEST
# ==========================================================================================

print(
    "\n[6/9] SUPPORTING 10-ITERATION EXACT SIGN-FLIP INFERENCE"
)


iteration_mean = (
    split
    .groupby(
        "iteration",
        as_index=False
    )
    .mean(
        numeric_only=True
    )
)


assert len(
    iteration_mean
) == 10


def iteration_delta_array(
    comparison,
    metric
):

    if metric == "AP":

        full_context = (
            iteration_mean[
                "Context_AP"
            ].to_numpy(
                dtype=float
            )
        )

        full_fusion = (
            iteration_mean[
                "Fusion_AP"
            ].to_numpy(
                dtype=float
            )
        )

        generic_context = (
            iteration_mean[
                "GenericContext_AP"
            ].to_numpy(
                dtype=float
            )
        )

        generic_fusion = (
            iteration_mean[
                "GenericFusion_AP"
            ].to_numpy(
                dtype=float
            )
        )


    else:

        full_context = (
            iteration_mean[
                "Context_AUROC"
            ].to_numpy(
                dtype=float
            )
        )

        full_fusion = (
            iteration_mean[
                "Fusion_AUROC"
            ].to_numpy(
                dtype=float
            )
        )

        generic_context = (
            iteration_mean[
                "GenericContext_AUROC"
            ].to_numpy(
                dtype=float
            )
        )

        generic_fusion = (
            iteration_mean[
                "GenericFusion_AUROC"
            ].to_numpy(
                dtype=float
            )
        )


    if (
        comparison[
            "name"
        ]
        ==
        "GenericFusion_vs_GenericContext"
    ):

        return (
            generic_fusion
            -
            generic_context
        )


    if (
        comparison[
            "name"
        ]
        ==
        "GenericContext_vs_FullContext"
    ):

        return (
            generic_context
            -
            full_context
        )


    if (
        comparison[
            "name"
        ]
        ==
        "GenericFusion_vs_FullFusion"
    ):

        return (
            generic_fusion
            -
            full_fusion
        )


    if (
        comparison[
            "name"
        ]
        ==
        "IncrementalGain_Generic_vs_Full"
    ):

        return (
            (
                generic_fusion
                -
                generic_context
            )
            -
            (
                full_fusion
                -
                full_context
            )
        )


    raise RuntimeError(
        "Unknown comparison."
    )


iteration_rows = []

iteration_summary_rows = []


for comparison in COMPARISONS:

    for metric in [
        "AP",
        "AUROC"
    ]:

        delta = iteration_delta_array(
            comparison,
            metric
        )


        for iteration_value, delta_value in zip(
            iteration_mean[
                "iteration"
            ],
            delta
        ):

            iteration_rows.append(
                {
                    "comparison":
                        comparison[
                            "name"
                        ],

                    "role":
                        comparison[
                            "role"
                        ],

                    "metric":
                        metric,

                    "iteration":
                        int(
                            iteration_value
                        ),

                    "delta":
                        float(
                            delta_value
                        )
                }
            )


        p_exact = (
            exact_sign_flip_pvalue(
                delta
            )
        )


        iteration_summary_rows.append(
            {
                "comparison":
                    comparison[
                        "name"
                    ],

                "role":
                    comparison[
                        "role"
                    ],

                "metric":
                    metric,

                "n_iterations":
                    10,

                "mean_delta":
                    float(
                        np.mean(
                            delta
                        )
                    ),

                "sd_delta":
                    float(
                        np.std(
                            delta,
                            ddof=1
                        )
                    ),

                "median_delta":
                    float(
                        np.median(
                            delta
                        )
                    ),

                "positive_iterations":
                    int(
                        np.sum(
                            delta > 0
                        )
                    ),

                "negative_iterations":
                    int(
                        np.sum(
                            delta < 0
                        )
                    ),

                "zero_iterations":
                    int(
                        np.sum(
                            delta == 0
                        )
                    ),

                "min_delta":
                    float(
                        np.min(
                            delta
                        )
                    ),

                "max_delta":
                    float(
                        np.max(
                            delta
                        )
                    ),

                "exact_two_sided_signflip_p":
                    p_exact
            }
        )


iteration_delta_df = pd.DataFrame(
    iteration_rows
)

iteration_summary_df = pd.DataFrame(
    iteration_summary_rows
)


iteration_summary_df[
    "BH_FDR_q_across_8_supporting_tests"
] = bh_adjust(
    iteration_summary_df[
        "exact_two_sided_signflip_p"
    ]
)


iteration_delta_df.to_csv(
    ITER_DELTA_FILE,
    sep="\t",
    index=False
)

iteration_summary_df.to_csv(
    ITER_SUMMARY_FILE,
    sep="\t",
    index=False
)


print(
    iteration_summary_df.to_string(
        index=False
    )
)


# ==========================================================================================
# 9. 790 BALANCED-DATASET DESCRIPTIVE CONSISTENCY
#
# NO p-values from these 790 datasets.
# ==========================================================================================

task_mean = (
    split
    .groupby(
        [
            "iteration",
            "balanced_dataset"
        ],
        as_index=False
    )
    .mean(
        numeric_only=True
    )
)


assert len(
    task_mean
) == 790


task_rows = []


for comparison in COMPARISONS:

    for metric in [
        "AP",
        "AUROC"
    ]:

        if metric == "AP":

            fc = task_mean[
                "Context_AP"
            ].to_numpy()

            ff = task_mean[
                "Fusion_AP"
            ].to_numpy()

            gc = task_mean[
                "GenericContext_AP"
            ].to_numpy()

            gf = task_mean[
                "GenericFusion_AP"
            ].to_numpy()


        else:

            fc = task_mean[
                "Context_AUROC"
            ].to_numpy()

            ff = task_mean[
                "Fusion_AUROC"
            ].to_numpy()

            gc = task_mean[
                "GenericContext_AUROC"
            ].to_numpy()

            gf = task_mean[
                "GenericFusion_AUROC"
            ].to_numpy()


        name = comparison[
            "name"
        ]


        if name == "GenericFusion_vs_GenericContext":

            delta = (
                gf
                -
                gc
            )


        elif name == "GenericContext_vs_FullContext":

            delta = (
                gc
                -
                fc
            )


        elif name == "GenericFusion_vs_FullFusion":

            delta = (
                gf
                -
                ff
            )


        elif name == "IncrementalGain_Generic_vs_Full":

            delta = (
                (
                    gf
                    -
                    gc
                )
                -
                (
                    ff
                    -
                    fc
                )
            )


        else:

            raise RuntimeError(
                "Unknown comparison."
            )


        task_rows.append(
            {
                "comparison":
                    name,

                "role":
                    comparison[
                        "role"
                    ],

                "metric":
                    metric,

                "n_balanced_datasets":
                    len(
                        delta
                    ),

                "mean_delta":
                    float(
                        np.mean(
                            delta
                        )
                    ),

                "median_delta":
                    float(
                        np.median(
                            delta
                        )
                    ),

                "sd_delta":
                    float(
                        np.std(
                            delta,
                            ddof=1
                        )
                    ),

                "fraction_delta_gt_0":
                    float(
                        np.mean(
                            delta > 0
                        )
                    ),

                "fraction_delta_lt_0":
                    float(
                        np.mean(
                            delta < 0
                        )
                    )
            }
        )


task_df = pd.DataFrame(
    task_rows
)


task_df.to_csv(
    TASK_DESC_FILE,
    sep="\t",
    index=False
)


# ==========================================================================================
# 10. PRIMARY GENE-LEVEL PAIRED STRATIFIED BOOTSTRAP
# ==========================================================================================

print(
    "\n[7/9] PRIMARY GENE-LEVEL PAIRED STRATIFIED BOOTSTRAP"
)


positive_idx = np.flatnonzero(
    y_gene == 1
)

unlabeled_idx = np.flatnonzero(
    y_gene == 0
)


assert len(
    positive_idx
) == 190

assert len(
    unlabeled_idx
) == 18062


SCORE_ARRAYS = {
    model:
        gene[
            column
        ].to_numpy(
            dtype=float
        )
    for model, column
    in MODEL_COLUMNS.items()
}


BOOT_COLUMNS = [
    "analysis_version",
    "replicate"
]


for comparison in COMPARISONS:

    for metric in [
        "AP",
        "AUROC"
    ]:

        BOOT_COLUMNS.append(
            (
                comparison[
                    "name"
                ]
                +
                "__"
                +
                metric
                +
                "_delta"
            )
        )


# ------------------------------------------------------------------------------------------
# Checkpoint resume
# ------------------------------------------------------------------------------------------

if BOOT_FILE.exists():

    boot_df = pd.read_csv(
        BOOT_FILE,
        sep="\t",
        compression="gzip"
    )


    if set(
        boot_df.columns
    ) != set(
        BOOT_COLUMNS
    ):

        raise RuntimeError(
            "Existing Stage 8C bootstrap checkpoint "
            "has incompatible columns."
        )


    if not (
        boot_df[
            "analysis_version"
        ] == ANALYSIS_VERSION
    ).all():

        raise RuntimeError(
            "Bootstrap checkpoint analysis version mismatch."
        )


    boot_df = (
        boot_df[
            BOOT_COLUMNS
        ]
        .sort_values(
            "replicate"
        )
        .reset_index(
            drop=True
        )
    )


    assert boot_df[
        "replicate"
    ].is_unique


    assert np.array_equal(
        boot_df[
            "replicate"
        ].to_numpy(
            dtype=int
        ),
        np.arange(
            len(
                boot_df
            )
        )
    )


    start_rep = len(
        boot_df
    )


    print(
        "Existing bootstrap checkpoint:",
        start_rep,
        "/",
        N_BOOTSTRAP
    )


else:

    boot_df = pd.DataFrame(
        columns=BOOT_COLUMNS
    )

    start_rep = 0


    print(
        "Starting fresh bootstrap: 0 /",
        N_BOOTSTRAP
    )


bootstrap_session_start = time.time()


for batch_start in range(
    start_rep,
    N_BOOTSTRAP,
    BOOTSTRAP_BATCH
):

    batch_end = min(
        batch_start
        +
        BOOTSTRAP_BATCH,
        N_BOOTSTRAP
    )


    batch_rows = []


    for b in range(
        batch_start,
        batch_end
    ):

        rng = np.random.default_rng(
            np.random.SeedSequence(
                [
                    BOOTSTRAP_SEED,
                    b
                ]
            )
        )


        sampled_positive = rng.choice(
            positive_idx,
            size=len(
                positive_idx
            ),
            replace=True
        )


        sampled_unlabeled = rng.choice(
            unlabeled_idx,
            size=len(
                unlabeled_idx
            ),
            replace=True
        )


        sampled_idx = np.concatenate(
            [
                sampled_positive,
                sampled_unlabeled
            ]
        )


        y_b = y_gene[
            sampled_idx
        ]


        perf_b = {}


        # Compute each model once per replicate
        for model, score_array in SCORE_ARRAYS.items():

            scores_b = score_array[
                sampled_idx
            ]


            perf_b[
                model
            ] = calculate_metrics(
                y_b,
                scores_b
            )


        row = {
            "analysis_version":
                ANALYSIS_VERSION,

            "replicate":
                b
        }


        for comparison in COMPARISONS:

            for metric in [
                "AP",
                "AUROC"
            ]:

                d = observed_delta(
                    comparison,
                    metric,
                    perf_b
                )


                column = (
                    comparison[
                        "name"
                    ]
                    +
                    "__"
                    +
                    metric
                    +
                    "_delta"
                )


                row[
                    column
                ] = d


        batch_rows.append(
            row
        )


    boot_df = pd.concat(
        [
            boot_df,
            pd.DataFrame(
                batch_rows
            )
        ],
        ignore_index=True
    )


    boot_df.to_csv(
        BOOT_FILE,
        sep="\t",
        index=False,
        compression="gzip"
    )


    elapsed_minutes = (
        time.time()
        -
        bootstrap_session_start
    ) / 60


    print(
        f"Bootstrap {batch_end:4d}/{N_BOOTSTRAP} "
        f"| session elapsed {elapsed_minutes:.1f} min"
    )


assert len(
    boot_df
) == N_BOOTSTRAP


# ==========================================================================================
# 11. BOOTSTRAP SUMMARY
# ==========================================================================================

summary_rows = []


for comparison in COMPARISONS:

    for metric in [
        "AP",
        "AUROC"
    ]:

        observed_d = observed_delta(
            comparison,
            metric,
            observed
        )


        column = (
            comparison[
                "name"
            ]
            +
            "__"
            +
            metric
            +
            "_delta"
        )


        delta = boot_df[
            column
        ].to_numpy(
            dtype=float
        )


        ci_low = float(
            np.quantile(
                delta,
                0.025
            )
        )

        ci_high = float(
            np.quantile(
                delta,
                0.975
            )
        )


        summary_rows.append(
            {
                "comparison":
                    comparison[
                        "name"
                    ],

                "role":
                    comparison[
                        "role"
                    ],

                "metric":
                    metric,

                "observed_delta":
                    float(
                        observed_d
                    ),

                "bootstrap_mean_delta":
                    float(
                        np.mean(
                            delta
                        )
                    ),

                "bootstrap_median_delta":
                    float(
                        np.median(
                            delta
                        )
                    ),

                "bootstrap_SD":
                    float(
                        np.std(
                            delta,
                            ddof=1
                        )
                    ),

                "CI95_percentile_low":
                    ci_low,

                "CI95_percentile_high":
                    ci_high,

                "bootstrap_fraction_delta_gt_0":
                    float(
                        np.mean(
                            delta > 0
                        )
                    ),

                "bootstrap_fraction_delta_lt_0":
                    float(
                        np.mean(
                            delta < 0
                        )
                    ),

                "CI_excludes_zero":
                    bool(
                        (
                            ci_low > 0
                        )
                        or
                        (
                            ci_high < 0
                        )
                    ),

                "n_bootstrap":
                    N_BOOTSTRAP,

                "bootstrap_seed":
                    BOOTSTRAP_SEED
            }
        )


bootstrap_summary = pd.DataFrame(
    summary_rows
)


bootstrap_summary.to_csv(
    BOOT_SUMMARY_FILE,
    sep="\t",
    index=False
)


# ==========================================================================================
# 12. ADD KEY OBSERVED EFFECT SIZES
# ==========================================================================================

generic_gene_ap_gain = (
    observed[
        "GenericFusion"
    ][
        "AP"
    ]
    -
    observed[
        "GenericContext"
    ][
        "AP"
    ]
)


generic_gene_auc_gain = (
    observed[
        "GenericFusion"
    ][
        "AUROC"
    ]
    -
    observed[
        "GenericContext"
    ][
        "AUROC"
    ]
)


full_gene_ap_gain = (
    observed[
        "FullFusion"
    ][
        "AP"
    ]
    -
    observed[
        "FullContext"
    ][
        "AP"
    ]
)


full_gene_auc_gain = (
    observed[
        "FullFusion"
    ][
        "AUROC"
    ]
    -
    observed[
        "FullContext"
    ][
        "AUROC"
    ]
)


relative_generic_ap_gain = (
    generic_gene_ap_gain
    /
    observed[
        "GenericContext"
    ][
        "AP"
    ]
)


generic_fusion_vs_full_context_ap = (
    observed[
        "GenericFusion"
    ][
        "AP"
    ]
    -
    observed[
        "FullContext"
    ][
        "AP"
    ]
)


generic_fusion_vs_full_context_auc = (
    observed[
        "GenericFusion"
    ][
        "AUROC"
    ]
    -
    observed[
        "FullContext"
    ][
        "AUROC"
    ]
)


# ==========================================================================================
# 13. REPRODUCIBILITY MANIFEST
# ==========================================================================================

print(
    "\n[8/9] SAVING REPRODUCIBILITY MANIFEST"
)


manifest = {
    "project":
        "ASD-ContextFusion",

    "stage":
        "8C_paired_robustness_inference",

    "analysis_version":
        ANALYSIS_VERSION,

    "scientific_status":
        "analysis-only; no model fitting",

    "primary_metric":
        "Average Precision / AUPRC",

    "secondary_metric":
        "AUROC",

    "primary_comparison":
        "GenericFusion_vs_GenericContext",

    "primary_inference":
        (
            "Paired stratified gene-level bootstrap "
            "on 18,252 unique genes."
        ),

    "supporting_inference":
        (
            "Exact paired sign-flip test across "
            "10 stochastic iterations."
        ),

    "difference_in_differences":
        (
            "(GenericFusion - GenericContext) - "
            "(FullFusion - FullContext)"
        ),

    "naive_7900_fold_inference":
        False,

    "bootstrap_replicates":
        N_BOOTSTRAP,

    "bootstrap_seed":
        BOOTSTRAP_SEED,

    "gene_count":
        18252,

    "ASD_gene_count":
        190,

    "unlabeled_gene_count":
        18062,

    "full_context_features":
        316,

    "generic_context_features":
        266,

    "filtered_by_disease_removed":
        50,

    "new_PU_sampling":
        False,

    "model_fitting":
        False,

    "hyperparameter_tuning":
        False,

    "input_hashes":
        {
            str(
                FULL_METRICS_FILE
            ):
                sha256_file(
                    FULL_METRICS_FILE
                ),

            str(
                FULL_GENE_FILE
            ):
                sha256_file(
                    FULL_GENE_FILE
                ),

            str(
                ROBUST_METRICS_FILE
            ):
                sha256_file(
                    ROBUST_METRICS_FILE
                ),

            str(
                ROBUST_GENE_FILE
            ):
                sha256_file(
                    ROBUST_GENE_FILE
                ),

            str(
                LOCK_FILE
            ):
                sha256_file(
                    LOCK_FILE
                )
        }
}


with open(
    MANIFEST_FILE,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        manifest,
        f,
        indent=2
    )


# ==========================================================================================
# 14. FINAL REPORT
# ==========================================================================================

print(
    "\n[9/9] FINAL STAGE 8C REPORT"
)


print(
    "\n" +
    "=" * 126
)

print(
    "STAGE 8C PAIRED ROBUSTNESS INFERENCE COMPLETE [OK]"
)

print(
    "=" * 126
)


print(
    "\nPRIMARY + SENSITIVITY GENE-LEVEL BOOTSTRAP"
)


print(
    bootstrap_summary[
        [
            "comparison",
            "role",
            "metric",
            "observed_delta",
            "CI95_percentile_low",
            "CI95_percentile_high",
            "bootstrap_fraction_delta_gt_0",
            "CI_excludes_zero"
        ]
    ].to_string(
        index=False
    )
)


print(
    "\nSUPPORTING ITERATION-LEVEL EXACT TEST"
)


print(
    iteration_summary_df[
        [
            "comparison",
            "metric",
            "mean_delta",
            "positive_iterations",
            "negative_iterations",
            "exact_two_sided_signflip_p",
            "BH_FDR_q_across_8_supporting_tests"
        ]
    ].to_string(
        index=False
    )
)


print(
    "\nKEY OBSERVED ROBUSTNESS EFFECTS"
)


print(
    "Generic Fusion - Generic Context | Gene AP:",
    f"{generic_gene_ap_gain:+.6f}"
)

print(
    "Generic Fusion - Generic Context | Gene AUROC:",
    f"{generic_gene_auc_gain:+.6f}"
)

print(
    "Relative Generic gene AP gain:",
    f"{relative_generic_ap_gain * 100:.2f}%"
)

print(
    "Full Fusion - Full Context | Gene AP:",
    f"{full_gene_ap_gain:+.6f}"
)

print(
    "Full Fusion - Full Context | Gene AUROC:",
    f"{full_gene_auc_gain:+.6f}"
)

print(
    "Generic Fusion - Full Context | Gene AP:",
    f"{generic_fusion_vs_full_context_ap:+.6f}"
)

print(
    "Generic Fusion - Full Context | Gene AUROC:",
    f"{generic_fusion_vs_full_context_auc:+.6f}"
)


print(
    "\nINTEGRITY"
)

print(
    "Models fitted                 : NO"
)

print(
    "New PU sampling              : NO"
)

print(
    "Feature definitions changed  : NO"
)

print(
    "Full Context features        : 316"
)

print(
    "Generic Context features     : 266"
)

print(
    "Disease-filtered removed     : 50"
)

print(
    "Frozen outer folds           : 7900"
)

print(
    "Genes                        : 18252"
)

print(
    "ASD genes                    : 190"
)

print(
    "Naive 7900-fold p-values     : NOT USED"
)

print(
    "Primary inference            : paired gene bootstrap"
)


print(
    "\nSAVED TO:"
)

print(
    OUT_DIR
)


print(
    "\nNEXT SCIENTIFIC STAGE:"
)

print(
    "Source-defined developmental-expression / genetic-context "
    "family characterization, followed by temporal/external ASD validation."
)

print(
    "=" * 126
)
```

```python
# ==========================================================================================
# ASD-ContextFusion
# STAGE 9A — FINAL LOCAL EXTERNAL-SOURCE PROVENANCE LOCK
#
# Acquisition:
#   Official Nature supplementary XLSX files downloaded manually in browser.
#
# Scientific status:
#   SAME prespecified external sources.
#   Acquisition route only changed for engineering/access reasons.
#
# IMPORTANT:
#   - NO Context scores accessed
#   - NO NT-v2 scores accessed
#   - NO Fusion scores accessed
#   - NO model ranks accessed
#   - NO external genes selected yet
#   - NO top-k threshold chosen
#
# Purpose:
#   1. Verify both downloads are genuine XLSX
#   2. SHA256-lock exact files
#   3. Inventory all sheets
#   4. Detect likely header rows containing gene-related fields
#   5. Print candidate sheets/columns for Stage 9B
#
# ==========================================================================================

from pathlib import Path
import hashlib
import json
import zipfile
import re

import pandas as pd


# ==========================================================================================
# 0. PATHS
# ==========================================================================================

ROOT = Path(r"E:\ASD_ContextFusion_Q1")

RAW_DIR = (
    ROOT /
    "01_raw" /
    "external_validation_2026"
)

ADMIN_DIR = (
    ROOT /
    "00_admin"
)

RAW_DIR.mkdir(
    parents=True,
    exist_ok=True
)


SOURCES = {

    "Tan_MolecularPsychiatry_2026": {

        "role":
            "PRIMARY_TEMPORAL_EXTERNAL",

        "publication_date":
            "2026-07-16",

        "doi":
            "10.1038/s41380-026-03754-6",

        "article":
            (
                "Genomic landscape of rare variants in a Chinese "
                "autism cohort and discovery of novel risk genes"
            ),

        "article_url":
            "https://www.nature.com/articles/s41380-026-03754-6",

        "official_supplement":
            "Supplementary Table 1-16",

        "official_media_file":
            "41380_2026_3754_MOESM2_ESM.xlsx",

        "local_file":
            (
                RAW_DIR /
                "Tan_2026_MolecularPsychiatry_"
                "Supplementary_Tables_1_16.xlsx"
            )
    },


    "Caballero_CommunBiol_2026": {

        "role":
            "SECONDARY_EXTERNAL_MES",

        "publication_date":
            "2026-05-30",

        "doi":
            "10.1038/s42003-026-10380-z",

        "article":
            (
                "Identification of moderate effect size genes in "
                "autism spectrum disorder through a novel gene "
                "pairing approach"
            ),

        "article_url":
            "https://www.nature.com/articles/s42003-026-10380-z",

        "official_supplement":
            "Supplementary Data S1-S7",

        "official_media_file":
            "42003_2026_10380_MOESM3_ESM.xlsx",

        "local_file":
            (
                RAW_DIR /
                "Caballero_2026_CommunBiol_"
                "Supplementary_Data_S1_S7.xlsx"
            )
    }
}


# ==========================================================================================
# 1. HELPERS
# ==========================================================================================

def sha256_file(
    path,
    chunk_size=1024 * 1024
):

    h = hashlib.sha256()

    with open(
        path,
        "rb"
    ) as f:

        while True:

            chunk = f.read(
                chunk_size
            )

            if not chunk:
                break

            h.update(
                chunk
            )

    return h.hexdigest()


def validate_xlsx(
    path
):

    if not path.exists():

        return False, "FILE_MISSING"


    try:

        with zipfile.ZipFile(
            path,
            "r"
        ) as z:

            names = set(
                z.namelist()
            )


            required = {
                "[Content_Types].xml",
                "xl/workbook.xml"
            }


            if not required.issubset(
                names
            ):

                return False, (
                    "ZIP_PRESENT_BUT_NOT_VALID_XLSX"
                )


        return True, "VALID_XLSX"


    except zipfile.BadZipFile:

        return False, "NOT_A_ZIP_XLSX"


    except Exception as e:

        return False, (
            "ERROR: " +
            str(
                e
            )
        )


def normalize_cell(
    value
):

    if pd.isna(
        value
    ):

        return ""


    value = str(
        value
    ).strip().lower()


    value = re.sub(
        r"[^a-z0-9]+",
        "_",
        value
    )


    return value.strip(
        "_"
    )


# These tokens are ONLY for locating likely table headers.
# They DO NOT define inclusion into the validation set.

GENE_HEADER_LOCATORS = [
    "gene",
    "genes",
    "gene_name",
    "gene_symbol",
    "symbol",
    "hgnc"
]


EVIDENCE_HEADER_LOCATORS = [
    "prior",
    "evidence",
    "novel",
    "association",
    "associated",
    "risk",
    "sfari",
    "category",
    "classification",
    "significant",
    "significance",
    "p_value",
    "pvalue",
    "q_value",
    "qvalue",
    "fdr"
]


def header_locator_score(
    row_values
):

    normalized = [
        normalize_cell(
            x
        )
        for x in row_values
    ]


    gene_hits = []


    for value in normalized:

        for locator in GENE_HEADER_LOCATORS:

            if (
                value == locator
                or
                value.startswith(
                    locator + "_"
                )
                or
                value.endswith(
                    "_" + locator
                )
                or
                ("_" + locator + "_")
                in (
                    "_" +
                    value +
                    "_"
                )
            ):

                gene_hits.append(
                    value
                )

                break


    evidence_hits = []


    for value in normalized:

        for locator in EVIDENCE_HEADER_LOCATORS:

            if locator in value:

                evidence_hits.append(
                    value
                )

                break


    return (
        len(
            gene_hits
        ),
        len(
            evidence_hits
        ),
        gene_hits,
        evidence_hits
    )


# ==========================================================================================
# 2. FILE VALIDATION
# ==========================================================================================

print(
    "=" * 126
)

print(
    "ASD-CONTEXTFUSION — STAGE 9A FINAL LOCAL SOURCE LOCK"
)

print(
    "=" * 126
)


print(
    "\n[1/6] VERIFYING OFFICIAL MANUAL DOWNLOADS"
)


source_rows = []


for source_name, info in SOURCES.items():

    path = info[
        "local_file"
    ]


    valid, status = validate_xlsx(
        path
    )


    print(
        "\nSOURCE:",
        source_name
    )

    print(
        "File:",
        path
    )

    print(
        "Status:",
        status
    )


    if not valid:

        raise RuntimeError(
            "\nRequired official supplementary file "
            "is missing or invalid.\n\n"
            f"Source: {source_name}\n"
            f"Expected local file:\n{path}\n\n"
            "Download the official Nature supplementary "
            "XLSX in your normal browser, save/rename it "
            "to this exact path, then rerun this cell."
        )


    file_hash = sha256_file(
        path
    )


    size_bytes = path.stat().st_size


    print(
        "Bytes:",
        size_bytes
    )

    print(
        "SHA256:",
        file_hash
    )


    source_rows.append(
        {
            "source":
                source_name,

            "role":
                info[
                    "role"
                ],

            "publication_date":
                info[
                    "publication_date"
                ],

            "doi":
                info[
                    "doi"
                ],

            "article":
                info[
                    "article"
                ],

            "article_url":
                info[
                    "article_url"
                ],

            "official_supplement":
                info[
                    "official_supplement"
                ],

            "official_media_file":
                info[
                    "official_media_file"
                ],

            "acquisition_method":
                "manual_browser_official_Nature_link",

            "local_file":
                str(
                    path
                ),

            "size_bytes":
                int(
                    size_bytes
                ),

            "sha256":
                file_hash,

            "xlsx_validation":
                status
        }
    )


source_manifest = pd.DataFrame(
    source_rows
)


# ==========================================================================================
# 3. WORKBOOK SHEET INVENTORY
# ==========================================================================================

print(
    "\n[2/6] WORKBOOK / SHEET INVENTORY"
)


inventory_rows = []

candidate_rows = []

preview_rows = []


for source_name, info in SOURCES.items():

    path = info[
        "local_file"
    ]


    excel = pd.ExcelFile(
        path,
        engine="openpyxl"
    )


    print(
        "\nSOURCE:",
        source_name
    )

    print(
        "Sheets:",
        len(
            excel.sheet_names
        )
    )


    for sheet_index, sheet in enumerate(
        excel.sheet_names
    ):

        print(
            f"  [{sheet_index}] {sheet}"
        )


        # ------------------------------------------------------------------
        # Read first 20 rows WITHOUT assuming where header starts.
        # ------------------------------------------------------------------

        raw_preview = pd.read_excel(
            path,
            sheet_name=sheet,
            header=None,
            nrows=20,
            engine="openpyxl"
        )


        best_header_row = None

        best_gene_hits = []

        best_evidence_hits = []

        best_score = (
            0,
            0
        )


        max_rows_to_scan = min(
            15,
            len(
                raw_preview
            )
        )


        for row_number in range(
            max_rows_to_scan
        ):

            row_values = (
                raw_preview.iloc[
                    row_number
                ]
                .tolist()
            )


            (
                gene_score,
                evidence_score,
                gene_hits,
                evidence_hits
            ) = header_locator_score(
                row_values
            )


            score = (
                gene_score,
                evidence_score
            )


            if score > best_score:

                best_score = score

                best_header_row = (
                    row_number
                )

                best_gene_hits = (
                    gene_hits
                )

                best_evidence_hits = (
                    evidence_hits
                )


        # ------------------------------------------------------------------
        # If a gene-like header row exists, read using that exact row.
        # ------------------------------------------------------------------

        if (
            best_header_row is not None
            and
            best_score[
                0
            ] > 0
        ):

            parsed = pd.read_excel(
                path,
                sheet_name=sheet,
                header=best_header_row,
                engine="openpyxl"
            )


            columns = [
                str(
                    x
                ).strip()
                for x in parsed.columns
            ]


            read_status = (
                "GENE_HEADER_CANDIDATE"
            )


            candidate_rows.append(
                {
                    "source":
                        source_name,

                    "role":
                        info[
                            "role"
                        ],

                    "sheet_index":
                        sheet_index,

                    "sheet":
                        sheet,

                    "detected_header_excel_row":
                        best_header_row + 1,

                    "data_rows":
                        int(
                            len(
                                parsed
                            )
                        ),

                    "columns":
                        int(
                            len(
                                parsed.columns
                            )
                        ),

                    "gene_header_hits":
                        " | ".join(
                            best_gene_hits
                        ),

                    "evidence_header_hits":
                        " | ".join(
                            best_evidence_hits
                        ),

                    "column_names":
                        " | ".join(
                            columns
                        )
                }
            )


            # Save first 5 parsed rows for source-structure inspection only.
            preview = (
                parsed
                .head(
                    5
                )
                .copy()
            )


            preview.insert(
                0,
                "__sheet",
                sheet
            )

            preview.insert(
                0,
                "__source",
                source_name
            )


            preview_rows.append(
                preview.astype(
                    str
                )
            )


        else:

            columns = []

            read_status = (
                "NO_GENE_HEADER_DETECTED"
            )


        inventory_rows.append(
            {
                "source":
                    source_name,

                "sheet_index":
                    sheet_index,

                "sheet":
                    sheet,

                "preview_rows_scanned":
                    max_rows_to_scan,

                "detected_header_excel_row":
                    (
                        best_header_row + 1
                        if best_header_row
                        is not None
                        else
                        None
                    ),

                "gene_header_score":
                    int(
                        best_score[
                            0
                        ]
                    ),

                "evidence_header_score":
                    int(
                        best_score[
                            1
                        ]
                    ),

                "gene_header_hits":
                    " | ".join(
                        best_gene_hits
                    ),

                "evidence_header_hits":
                    " | ".join(
                        best_evidence_hits
                    ),

                "status":
                    read_status
            }
        )


inventory = pd.DataFrame(
    inventory_rows
)


candidates = pd.DataFrame(
    candidate_rows
)


if preview_rows:

    candidate_preview = pd.concat(
        preview_rows,
        ignore_index=True,
        sort=False
    )

else:

    candidate_preview = pd.DataFrame()


# ==========================================================================================
# 4. SAVE SOURCE-LOCK FILES
# ==========================================================================================

print(
    "\n[3/6] SAVING SOURCE PROVENANCE + INVENTORY"
)


SOURCE_MANIFEST_FILE = (
    RAW_DIR /
    "Stage9A_external_source_manifest.tsv"
)

SHEET_INVENTORY_FILE = (
    RAW_DIR /
    "Stage9A_supplementary_sheet_inventory.tsv"
)

CANDIDATE_FILE = (
    RAW_DIR /
    "Stage9A_candidate_gene_table_sheets.tsv"
)

PREVIEW_FILE = (
    RAW_DIR /
    "Stage9A_candidate_sheet_preview_first5.tsv"
)


source_manifest.to_csv(
    SOURCE_MANIFEST_FILE,
    sep="\t",
    index=False
)


inventory.to_csv(
    SHEET_INVENTORY_FILE,
    sep="\t",
    index=False
)


candidates.to_csv(
    CANDIDATE_FILE,
    sep="\t",
    index=False
)


candidate_preview.to_csv(
    PREVIEW_FILE,
    sep="\t",
    index=False
)


# ==========================================================================================
# 5. SCIENTIFIC SOURCE LOCK
# ==========================================================================================

print(
    "\n[4/6] WRITING SCIENTIFIC SOURCE LOCK"
)


LOCK_FILE = (
    ADMIN_DIR /
    "ASD_ContextFusion_STAGE9A_EXTERNAL_VALIDATION_SOURCE_LOCK.json"
)


lock = {

    "project":
        "ASD-ContextFusion",

    "stage":
        "9A_external_validation_source_lock",

    "status":
        "SOURCE_FILES_LOCKED_BEFORE_MODEL_RANK_EVALUATION",

    "primary_source":
        "Tan_MolecularPsychiatry_2026",

    "secondary_source":
        "Caballero_CommunBiol_2026",

    "source_selection_changed":
        False,

    "acquisition_route":
        (
            "manual browser download from official "
            "Nature supplementary links"
        ),

    "acquisition_route_change_is_engineering_only":
        True,

    "model_scores_accessed":
        False,

    "model_ranks_accessed":
        False,

    "external_validation_genes_selected":
        False,

    "validation_categories_finalized":
        False,

    "top_percent_thresholds_finalized":
        False,

    "source_files":
        source_manifest.to_dict(
            orient="records"
        )
}


with open(
    LOCK_FILE,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        lock,
        f,
        indent=2
    )


# ==========================================================================================
# 6. HASH THE GENERATED LOCK MATERIALS
# ==========================================================================================

print(
    "\n[5/6] HASHING STAGE 9A LOCK MATERIALS"
)


hash_files = [
    SOURCE_MANIFEST_FILE,
    SHEET_INVENTORY_FILE,
    CANDIDATE_FILE,
    PREVIEW_FILE,
    LOCK_FILE
]


hash_rows = []


for path in hash_files:

    hash_rows.append(
        {
            "file":
                str(
                    path
                ),

            "sha256":
                sha256_file(
                    path
                )
        }
    )


HASH_FILE = (
    RAW_DIR /
    "Stage9A_lock_material_SHA256.tsv"
)


pd.DataFrame(
    hash_rows
).to_csv(
    HASH_FILE,
    sep="\t",
    index=False
)


# ==========================================================================================
# 7. FINAL REPORT
# ==========================================================================================

print(
    "\n[6/6] FINAL STAGE 9A REPORT"
)


print(
    "\n" +
    "=" * 126
)

print(
    "STAGE 9A EXTERNAL VALIDATION SOURCE LOCK COMPLETE [OK]"
)

print(
    "=" * 126
)


print(
    "\nSOURCE FILES"
)


print(
    source_manifest[
        [
            "source",
            "role",
            "publication_date",
            "size_bytes",
            "sha256"
        ]
    ].to_string(
        index=False
    )
)


print(
    "\nCANDIDATE GENE-TABLE SHEETS"
)


if len(
    candidates
) == 0:

    print(
        "NO GENE-LIKE HEADER AUTOMATICALLY DETECTED."
    )

    print(
        "Do NOT select genes manually yet."
    )


else:

    display_columns = [
        "source",
        "sheet_index",
        "sheet",
        "detected_header_excel_row",
        "data_rows",
        "gene_header_hits",
        "evidence_header_hits",
        "column_names"
    ]


    print(
        candidates[
            display_columns
        ].to_string(
            index=False
        )
    )


print(
    "\nINTEGRITY"
)

print(
    "Scientific sources changed       : NO"
)

print(
    "Download route changed           : YES — engineering/access only"
)

print(
    "Official XLSX validated          : YES"
)

print(
    "SHA256 source lock               : YES"
)

print(
    "Context scores accessed          : NO"
)

print(
    "NT-v2 scores accessed            : NO"
)

print(
    "Fusion scores accessed           : NO"
)

print(
    "Model ranks accessed             : NO"
)

print(
    "Validation genes selected        : NO"
)

print(
    "Top-k threshold chosen           : NO"
)


print(
    "\nSAVED:"
)

print(
    SOURCE_MANIFEST_FILE
)

print(
    SHEET_INVENTORY_FILE
)

print(
    CANDIDATE_FILE
)

print(
    PREVIEW_FILE
)

print(
    LOCK_FILE
)

print(
    HASH_FILE
)


print(
    "\nNEXT SCIENTIFIC STAGE:"
)

print(
    "Stage 9B — source-defined external gene-set lock "
    "BEFORE reading Context / NT-v2 / Fusion rankings."
)

print(
    "=" * 126
)
```

```python
# ==========================================================================================
# ASD-ContextFusion
# STAGE 9B.0 — TAN 2026 TABLE 9 SOURCE-STRUCTURE / STATISTICAL-BLOCK AUDIT
#
# PURPOSE
# -------
# Identify the exact authors' combined-ASD statistical block in
# Supplementary Table 9 BEFORE freezing any external validation genes.
#
# IMPORTANT
# ---------
# - NO Context scores
# - NO NT scores
# - NO Fusion scores
# - NO model ranks
# - NO external validation genes finalized
# - NO top-k thresholds
#
# We do NOT choose an FDR column merely because it produces 245 genes.
# We first inspect its multi-row source header / block identity.
# ==========================================================================================

from pathlib import Path
import hashlib
import json
import re

import numpy as np
import pandas as pd


# ==========================================================================================
# 0. PATHS
# ==========================================================================================

ROOT = Path(r"E:\ASD_ContextFusion_Q1")

RAW_DIR = (
    ROOT /
    "01_raw" /
    "external_validation_2026"
)

TAN_FILE = (
    RAW_DIR /
    "Tan_2026_MolecularPsychiatry_Supplementary_Tables_1_16.xlsx"
)

SOURCE_LOCK = (
    ROOT /
    "00_admin" /
    "ASD_ContextFusion_STAGE9A_EXTERNAL_VALIDATION_SOURCE_LOCK.json"
)

OUT_DIR = (
    ROOT /
    "02_interim" /
    "external_validation_source_lock"
)

OUT_DIR.mkdir(
    parents=True,
    exist_ok=True
)

SHEET = "Supplementary Table 9"

# Stage 9A detected Excel row 4 as parsed header.
HEADER_ROW_ZERO_BASED = 3


# ==========================================================================================
# 1. HELPERS
# ==========================================================================================

def sha256_file(path, chunk_size=1024 * 1024):

    h = hashlib.sha256()

    with open(path, "rb") as f:

        while True:

            chunk = f.read(chunk_size)

            if not chunk:
                break

            h.update(chunk)

    return h.hexdigest()


def clean_text(x):

    if pd.isna(x):
        return ""

    return str(x).strip()


def numeric_series(s):

    return pd.to_numeric(
        s,
        errors="coerce"
    )


# ==========================================================================================
# 2. INPUT LOCK AUDIT
# ==========================================================================================

print("=" * 128)
print("ASD-CONTEXTFUSION — STAGE 9B.0 TAN TABLE 9 STATISTICAL-BLOCK AUDIT")
print("=" * 128)

print("\n[1/7] INPUT / SOURCE LOCK AUDIT")

assert TAN_FILE.exists(), TAN_FILE
assert SOURCE_LOCK.exists(), SOURCE_LOCK

with open(
    SOURCE_LOCK,
    "r",
    encoding="utf-8"
) as f:

    lock = json.load(f)

assert lock[
    "model_scores_accessed"
] is False

assert lock[
    "model_ranks_accessed"
] is False

assert lock[
    "external_validation_genes_selected"
] is False


print("Tan XLSX:", TAN_FILE)
print("SHA256 :", sha256_file(TAN_FILE))
print("Stage 9A source lock: VERIFIED [OK]")


# ==========================================================================================
# 3. READ RAW MULTI-ROW HEADER + PARSED TABLE
# ==========================================================================================

print("\n[2/7] READING RAW MULTI-ROW HEADER")

raw = pd.read_excel(
    TAN_FILE,
    sheet_name=SHEET,
    header=None,
    engine="openpyxl"
)

parsed = pd.read_excel(
    TAN_FILE,
    sheet_name=SHEET,
    header=HEADER_ROW_ZERO_BASED,
    engine="openpyxl"
)


print("Raw shape   :", raw.shape)
print("Parsed shape:", parsed.shape)

print("\nFIRST 6 RAW EXCEL ROWS — first 25 columns:")
print(
    raw.iloc[
        :6,
        :25
    ].to_string(
        index=True,
        header=True
    )
)


# ==========================================================================================
# 4. COLUMN INDEX + MULTI-ROW SOURCE LINEAGE
# ==========================================================================================

print("\n[3/7] STATISTICAL / EVIDENCE COLUMN LINEAGE")


columns = list(
    parsed.columns
)


interesting_patterns = [
    r"^P$",
    r"^P\.",
    r"FDR",
    r"FWER",
    r"MIN_P",
    r"DNE",
    r"DNN",
    r"Genesymbol",
    r"HGNC",
    r"diagnostic",
    r"Evidence",
    r"SFARI",
    r"confidence",
    r"membership",
    r"MSSNG",
    r"SPARK"
]


interesting_indices = []


for idx, col in enumerate(columns):

    col_text = str(col)

    if any(
        re.search(
            pattern,
            col_text,
            flags=re.I
        )
        for pattern in interesting_patterns
    ):

        interesting_indices.append(idx)


lineage_rows = []


for idx in interesting_indices:

    lineage_rows.append(
        {
            "column_index":
                idx,

            "parsed_column":
                str(
                    columns[idx]
                ),

            "raw_excel_row1":
                clean_text(
                    raw.iloc[
                        0,
                        idx
                    ]
                ),

            "raw_excel_row2":
                clean_text(
                    raw.iloc[
                        1,
                        idx
                    ]
                ),

            "raw_excel_row3":
                clean_text(
                    raw.iloc[
                        2,
                        idx
                    ]
                ),

            "raw_excel_row4":
                clean_text(
                    raw.iloc[
                        3,
                        idx
                    ]
                )
        }
    )


lineage = pd.DataFrame(
    lineage_rows
)


print(
    lineage.to_string(
        index=False
    )
)


LINEAGE_FILE = (
    OUT_DIR /
    "Stage9B0_Tan_Table9_column_lineage.tsv"
)


lineage.to_csv(
    LINEAGE_FILE,
    sep="\t",
    index=False
)


# ==========================================================================================
# 5. IDENTIFY EXACT GENE SYMBOL COLUMN
# ==========================================================================================

print("\n[4/7] GENE SYMBOL COLUMN AUDIT")


preferred_gene_columns = [
    "Genesymbol",
    "GeneSymbol",
    "Gene_symbol",
    "Gene",
    "HGNC"
]


gene_column = None


for candidate in preferred_gene_columns:

    if candidate in parsed.columns:

        gene_column = candidate
        break


if gene_column is None:

    raise RuntimeError(
        "Could not identify a source-defined gene-symbol column."
    )


genes = (
    parsed[
        gene_column
    ]
    .astype(str)
    .str.strip()
)


valid_gene_mask = (
    parsed[
        gene_column
    ].notna()
    &
    genes.ne("")
    &
    genes.ne("nan")
)


print("Selected gene column:", gene_column)
print("Rows with gene symbol:", int(valid_gene_mask.sum()))
print(
    "Unique gene symbols:",
    parsed.loc[
        valid_gene_mask,
        gene_column
    ].astype(str).str.strip().nunique()
)


# ==========================================================================================
# 6. AUDIT ALL FDR / FWER COLUMNS WITHOUT SELECTING ONE
# ==========================================================================================

print("\n[5/7] FDR / FWER THRESHOLD COUNTS")


stat_columns = [
    col
    for col in parsed.columns
    if re.search(
        r"(^FDR($|\.)|^FWER($|\.)|^P($|\.)|MIN_P)",
        str(col),
        flags=re.I
    )
]


stat_rows = []


for col in stat_columns:

    values = numeric_series(
        parsed[
            col
        ]
    )


    finite_gene_mask = (
        valid_gene_mask
        &
        values.notna()
    )


    row = {
        "column":
            str(col),

        "numeric_gene_rows":
            int(
                finite_gene_mask.sum()
            ),

        "lt_0.10":
            int(
                (
                    finite_gene_mask
                    &
                    (values < 0.10)
                ).sum()
            ),

        "lt_0.05":
            int(
                (
                    finite_gene_mask
                    &
                    (values < 0.05)
                ).sum()
            ),

        "lt_0.01":
            int(
                (
                    finite_gene_mask
                    &
                    (values < 0.01)
                ).sum()
            ),

        "lt_0.001":
            int(
                (
                    finite_gene_mask
                    &
                    (values < 0.001)
                ).sum()
            )
    }


    stat_rows.append(
        row
    )


stat_audit = pd.DataFrame(
    stat_rows
)


print(
    stat_audit.to_string(
        index=False
    )
)


STAT_FILE = (
    OUT_DIR /
    "Stage9B0_Tan_Table9_statistical_threshold_counts.tsv"
)


stat_audit.to_csv(
    STAT_FILE,
    sep="\t",
    index=False
)


# ==========================================================================================
# 7. FIND "CANDIDATE-SIZED" FDR SETS FOR SOURCE AUDIT ONLY
#
# IMPORTANT:
# This is NOT feature/gene selection.
# It only identifies which statistical columns need source-header inspection.
# The paper reports 245 ASD risk genes.
# ==========================================================================================

print("\n[6/7] CANDIDATE-SIZED FDR SET AUDIT")


candidate_stat_sets = []


fdr_columns = [
    col
    for col in parsed.columns
    if re.search(
        r"^FDR($|\.)",
        str(col),
        flags=re.I
    )
]


for col in fdr_columns:

    values = numeric_series(
        parsed[
            col
        ]
    )


    mask = (
        valid_gene_mask
        &
        values.notna()
        &
        (values < 0.05)
    )


    count = int(
        mask.sum()
    )


    unique_count = int(
        parsed.loc[
            mask,
            gene_column
        ]
        .astype(str)
        .str.strip()
        .nunique()
    )


    if (
        150 <= unique_count <= 350
    ):

        genes_here = (
            parsed.loc[
                mask,
                gene_column
            ]
            .astype(str)
            .str.strip()
            .tolist()
        )


        candidate_stat_sets.append(
            {
                "FDR_column":
                    str(col),

                "rows_FDR_lt_0.05":
                    count,

                "unique_genes_FDR_lt_0.05":
                    unique_count,

                "first_20_genes_SOURCE_ORDER":
                    " | ".join(
                        genes_here[:20]
                    )
            }
        )


candidate_stat_df = pd.DataFrame(
    candidate_stat_sets
)


if len(candidate_stat_df):

    print(
        candidate_stat_df.to_string(
            index=False
        )
    )

else:

    print(
        "No FDR<0.05 block produced 150–350 unique genes."
    )


CANDIDATE_STAT_FILE = (
    OUT_DIR /
    "Stage9B0_Tan_candidate_sized_FDR_blocks.tsv"
)


candidate_stat_df.to_csv(
    CANDIDATE_STAT_FILE,
    sep="\t",
    index=False
)


# ==========================================================================================
# 8. PRIOR-EVIDENCE FIELD DISTRIBUTIONS
#
# Still NO model ranks.
# We inspect source-defined evidence fields only.
# ==========================================================================================

print("\n[7/7] SOURCE PRIOR-EVIDENCE FIELD AUDIT")


evidence_candidates = [
    col
    for col in parsed.columns
    if any(
        token.lower()
        in str(col).lower()
        for token in [
            "diagnostic_category",
            "Evidence from PubMed",
            "SFARI_gene-score",
            "SFARI_syndromic",
            "SFARI_eagle",
            "confidence category"
        ]
    )
]


print(
    "Evidence-related columns:"
)


for col in evidence_candidates:

    print(
        " -",
        col
    )


evidence_summary_rows = []


for col in evidence_candidates:

    series = (
        parsed.loc[
            valid_gene_mask,
            col
        ]
    )


    counts = (
        series
        .fillna(
            "<NA>"
        )
        .astype(str)
        .str.strip()
        .value_counts(
            dropna=False
        )
        .head(
            30
        )
    )


    print(
        "\nCOLUMN:",
        col
    )

    print(
        counts.to_string()
    )


    for value, count in counts.items():

        evidence_summary_rows.append(
            {
                "column":
                    str(col),

                "value":
                    str(value),

                "count":
                    int(count)
            }
        )


evidence_summary = pd.DataFrame(
    evidence_summary_rows
)


EVIDENCE_FILE = (
    OUT_DIR /
    "Stage9B0_Tan_prior_evidence_value_counts.tsv"
)


evidence_summary.to_csv(
    EVIDENCE_FILE,
    sep="\t",
    index=False
)


# ==========================================================================================
# 9. SAVE AUDIT MANIFEST
# ==========================================================================================

MANIFEST_FILE = (
    ROOT /
    "00_admin" /
    "ASD_ContextFusion_STAGE9B0_TAN_TABLE9_AUDIT.json"
)


manifest = {

    "project":
        "ASD-ContextFusion",

    "stage":
        "9B.0_Tan_Table9_source_structure_audit",

    "Tan_source_file":
        str(
            TAN_FILE
        ),

    "Tan_source_sha256":
        sha256_file(
            TAN_FILE
        ),

    "sheet":
        SHEET,

    "parsed_header_excel_row":
        HEADER_ROW_ZERO_BASED + 1,

    "gene_symbol_column":
        gene_column,

    "model_scores_accessed":
        False,

    "model_ranks_accessed":
        False,

    "validation_genes_finalized":
        False,

    "FDR_column_finalized":
        False,

    "prior_evidence_category_finalized":
        False,

    "selection_rule":
        (
            "No final gene-set selection in Stage 9B.0. "
            "Exact statistical block and prior-evidence categories "
            "must be source-resolved first."
        )
}


with open(
    MANIFEST_FILE,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        manifest,
        f,
        indent=2
    )


print("\n" + "=" * 128)
print("STAGE 9B.0 COMPLETE — AUDIT ONLY [OK]")
print("=" * 128)

print("\nNO validation gene set has been finalized.")
print("NO Context / NT-v2 / Fusion ranking was accessed.")

print("\nSAVED:")
print(LINEAGE_FILE)
print(STAT_FILE)
print(CANDIDATE_STAT_FILE)
print(EVIDENCE_FILE)
print(MANIFEST_FILE)

print("\nPASTE BACK THESE THREE SECTIONS:")
print("1. STATISTICAL / EVIDENCE COLUMN LINEAGE")
print("2. FDR / FWER THRESHOLD COUNTS")
print("3. SOURCE PRIOR-EVIDENCE FIELD AUDIT")

print("=" * 128)
```

```python
# ==========================================================================================
# ASD-ContextFusion
# STAGE 9B.0b — TAN 2026 EXACT 245-GENE RULE RESOLUTION
#
# PURPOSE
# -------
# Test whether the paper's reported 245 ASD risk genes correspond exactly to:
#
#       FDR.2 < 0.05  OR  FDR.3 < 0.05
#
# while simultaneously verifying the raw source headers and inspecting
# prior-evidence annotations WITHIN that candidate set.
#
# IMPORTANT
# ---------
# - NO Context scores
# - NO NT scores
# - NO Fusion scores
# - NO model ranks
# - NO final validation gene-set lock yet
# ==========================================================================================

from pathlib import Path
import hashlib
import json
import re

import numpy as np
import pandas as pd


# ==========================================================================================
# 0. PATHS
# ==========================================================================================

ROOT = Path(r"E:\ASD_ContextFusion_Q1")

TAN_FILE = (
    ROOT /
    "01_raw" /
    "external_validation_2026" /
    "Tan_2026_MolecularPsychiatry_Supplementary_Tables_1_16.xlsx"
)

OUT_DIR = (
    ROOT /
    "02_interim" /
    "external_validation_source_lock"
)

OUT_DIR.mkdir(
    parents=True,
    exist_ok=True
)

SHEET = "Supplementary Table 9"

HEADER_ROW_ZERO_BASED = 3


# ==========================================================================================
# 1. HELPERS
# ==========================================================================================

def sha256_file(path, chunk_size=1024 * 1024):

    h = hashlib.sha256()

    with open(path, "rb") as f:

        while True:

            chunk = f.read(chunk_size)

            if not chunk:
                break

            h.update(chunk)

    return h.hexdigest()


def clean(x):

    if pd.isna(x):
        return ""

    return str(x).strip()


# ==========================================================================================
# 2. LOAD SOURCE ONLY
# ==========================================================================================

print("=" * 130)
print("ASD-CONTEXTFUSION — STAGE 9B.0b TAN EXACT 245-GENE RULE RESOLUTION")
print("=" * 130)

print("\n[1/6] SOURCE AUDIT")

assert TAN_FILE.exists()

print("Tan XLSX:")
print(TAN_FILE)

print("\nSHA256:")
print(sha256_file(TAN_FILE))


raw = pd.read_excel(
    TAN_FILE,
    sheet_name=SHEET,
    header=None,
    engine="openpyxl"
)

df = pd.read_excel(
    TAN_FILE,
    sheet_name=SHEET,
    header=HEADER_ROW_ZERO_BASED,
    engine="openpyxl"
)


assert "Genesymbol" in df.columns
assert "FDR.2" in df.columns
assert "FDR.3" in df.columns


print("\nRaw shape   :", raw.shape)
print("Parsed shape:", df.shape)


# ==========================================================================================
# 3. EXACT RAW HEADER AROUND SECOND ASD STATISTICAL BLOCK
# ==========================================================================================

print("\n[2/6] RAW MULTI-ROW HEADER — COLUMNS 65 TO 89")


header_slice = raw.iloc[
    0:4,
    65:90
].copy()


print(
    header_slice.to_string(
        index=True,
        header=True
    )
)


print("\nPARSED COLUMNS 65 TO 89")

for idx in range(65, 90):

    print(
        f"{idx:3d} | {df.columns[idx]}"
    )


# Save exact header provenance
HEADER_FILE = (
    OUT_DIR /
    "Stage9B0b_Tan_columns65_89_raw_header.tsv"
)

header_slice.to_csv(
    HEADER_FILE,
    sep="\t",
    index=True
)


# ==========================================================================================
# 4. EXACT FDR.2 / FDR.3 SET RELATIONSHIP
# ==========================================================================================

print("\n[3/6] TESTING THE TWO SOURCE FDR BLOCKS")


gene = (
    df[
        "Genesymbol"
    ]
    .astype(str)
    .str.strip()
)


valid_gene = (
    df[
        "Genesymbol"
    ].notna()
    &
    gene.ne("")
    &
    gene.ne("nan")
)


fdr2 = pd.to_numeric(
    df[
        "FDR.2"
    ],
    errors="coerce"
)

fdr3 = pd.to_numeric(
    df[
        "FDR.3"
    ],
    errors="coerce"
)


mask_fdr2 = (
    valid_gene
    &
    fdr2.notna()
    &
    (fdr2 < 0.05)
)


mask_fdr3 = (
    valid_gene
    &
    fdr3.notna()
    &
    (fdr3 < 0.05)
)


set_fdr2 = set(
    gene.loc[
        mask_fdr2
    ]
)


set_fdr3 = set(
    gene.loc[
        mask_fdr3
    ]
)


set_union = (
    set_fdr2
    |
    set_fdr3
)


set_intersection = (
    set_fdr2
    &
    set_fdr3
)


only_fdr2 = (
    set_fdr2
    -
    set_fdr3
)


only_fdr3 = (
    set_fdr3
    -
    set_fdr2
)


print(
    "FDR.2 < 0.05 genes :",
    len(
        set_fdr2
    )
)

print(
    "FDR.3 < 0.05 genes :",
    len(
        set_fdr3
    )
)

print(
    "Intersection       :",
    len(
        set_intersection
    )
)

print(
    "FDR.2 only        :",
    len(
        only_fdr2
    )
)

print(
    "FDR.3 only        :",
    len(
        only_fdr3
    )
)

print(
    "UNION             :",
    len(
        set_union
    )
)


expected_union = (
    len(
        set_fdr2
    )
    +
    len(
        set_fdr3
    )
    -
    len(
        set_intersection
    )
)


assert expected_union == len(
    set_union
)


if len(
    set_union
) != 245:

    raise RuntimeError(
        "\nFDR.2/FDR.3 union does NOT reproduce the "
        "published 245-gene count.\n"
        "Do NOT finalize the validation set."
    )


print(
    "\nFDR.2/FDR.3 UNION EXACTLY REPRODUCES "
    "THE PUBLISHED 245-GENE COUNT [OK]"
)


# ==========================================================================================
# 5. CREATE SOURCE-ORDER CANDIDATE 245 SET — NOT YET FINAL LOCK
# ==========================================================================================

print("\n[4/6] BUILDING SOURCE-ORDER 245-GENE CANDIDATE SET")


mask_union = (
    valid_gene
    &
    gene.isin(
        set_union
    )
)


candidate = df.loc[
    mask_union
].copy()


candidate[
    "Genesymbol"
] = (
    candidate[
        "Genesymbol"
    ]
    .astype(str)
    .str.strip()
)


assert len(
    candidate
) == 245

assert candidate[
    "Genesymbol"
].nunique() == 245


candidate.insert(
    0,
    "Tan_245_membership_rule",
    "FDR.2<0.05 OR FDR.3<0.05"
)


candidate.insert(
    1,
    "significant_FDR2",
    (
        pd.to_numeric(
            candidate[
                "FDR.2"
            ],
            errors="coerce"
        )
        <
        0.05
    )
)


candidate.insert(
    2,
    "significant_FDR3",
    (
        pd.to_numeric(
            candidate[
                "FDR.3"
            ],
            errors="coerce"
        )
        <
        0.05
    )
)


print(
    "Candidate rows       :",
    len(
        candidate
    )
)

print(
    "Unique candidate genes:",
    candidate[
        "Genesymbol"
    ].nunique()
)


print("\nSTATISTICAL CONTRIBUTION")

print(
    candidate[
        [
            "significant_FDR2",
            "significant_FDR3"
        ]
    ]
    .value_counts()
    .to_string()
)


# ==========================================================================================
# 6. PRIOR-EVIDENCE STRUCTURE WITHIN THE 245
# ==========================================================================================

print("\n[5/6] PRIOR-EVIDENCE STRUCTURE WITHIN THE 245 CANDIDATE GENES")


evidence_columns = [
    "diagnostic_category",
    "Evidence from PubMed",
    "SFARI_gene-score",
    "SFARI_syndromic",
    "SFARI_eagle score",
    "confidence category",
    "SPARK162 gene list",
    "SPARK404(P<0.001 in denovowest or TDT statistic≥1)",
    "membership in TADA ASD FDR < 0.001 list of 72",
    "membership in TADA ASD FDR < 0.05 list of 185",
    "MSSNG134(FDR<0.1 in TADA)",
    "MSSNG106(FDR<0.05 in TADA)"
]


available_evidence_columns = [
    col
    for col in evidence_columns
    if col in candidate.columns
]


print(
    "Evidence columns found:",
    len(
        available_evidence_columns
    )
)


for col in available_evidence_columns:

    print(
        "\n" +
        "-" * 110
    )

    print(
        "COLUMN:",
        col
    )


    counts = (
        candidate[
            col
        ]
        .fillna(
            "<NA>"
        )
        .astype(str)
        .str.strip()
        .value_counts(
            dropna=False
        )
    )


    print(
        counts.to_string()
    )


# ----------------------------------------------------------------------
# Missing/nonmissing profile of the main ASD-prior-evidence fields.
# This is descriptive only; NO category is defined yet.
# ----------------------------------------------------------------------

prior_profile = pd.DataFrame(
    {
        "Gene":
            candidate[
                "Genesymbol"
            ],

        "FDR2":
            pd.to_numeric(
                candidate[
                    "FDR.2"
                ],
                errors="coerce"
            ),

        "FDR3":
            pd.to_numeric(
                candidate[
                    "FDR.3"
                ],
                errors="coerce"
            )
    }
)


for col in available_evidence_columns:

    prior_profile[
        col
    ] = (
        candidate[
            col
        ]
        .reset_index(
            drop=True
        )
    )


# ==========================================================================================
# 7. SAVE CANDIDATE AUDIT — NOT FINAL GENE-SET LOCK
# ==========================================================================================

print("\n[6/6] SAVING AUDIT-ONLY OUTPUT")


CANDIDATE_FILE = (
    OUT_DIR /
    "Stage9B0b_Tan_245_candidate_UNION_NOT_FINAL.tsv"
)


PROFILE_FILE = (
    OUT_DIR /
    "Stage9B0b_Tan_245_prior_evidence_profile_NOT_FINAL.tsv"
)


SUMMARY_FILE = (
    OUT_DIR /
    "Stage9B0b_Tan_FDR2_FDR3_union_summary.tsv"
)


candidate.to_csv(
    CANDIDATE_FILE,
    sep="\t",
    index=False
)


prior_profile.to_csv(
    PROFILE_FILE,
    sep="\t",
    index=False
)


summary = pd.DataFrame(
    [
        {
            "FDR2_lt_0.05":
                len(
                    set_fdr2
                ),

            "FDR3_lt_0.05":
                len(
                    set_fdr3
                ),

            "intersection":
                len(
                    set_intersection
                ),

            "FDR2_only":
                len(
                    only_fdr2
                ),

            "FDR3_only":
                len(
                    only_fdr3
                ),

            "union":
                len(
                    set_union
                ),

            "published_target":
                245,

            "exact_match":
                len(
                    set_union
                ) == 245
        }
    ]
)


summary.to_csv(
    SUMMARY_FILE,
    sep="\t",
    index=False
)


MANIFEST_FILE = (
    ROOT /
    "00_admin" /
    "ASD_ContextFusion_STAGE9B0b_TAN_245_RULE_AUDIT.json"
)


manifest = {
    "project":
        "ASD-ContextFusion",

    "stage":
        "9B.0b",

    "source":
        "Tan 2026 Supplementary Table 9",

    "candidate_rule_tested":
        "FDR.2 < 0.05 OR FDR.3 < 0.05",

    "FDR2_count":
        len(
            set_fdr2
        ),

    "FDR3_count":
        len(
            set_fdr3
        ),

    "intersection_count":
        len(
            set_intersection
        ),

    "union_count":
        len(
            set_union
        ),

    "matches_published_245":
        len(
            set_union
        ) == 245,

    "validation_gene_set_finalized":
        False,

    "prior_evidence_subsets_finalized":
        False,

    "model_scores_accessed":
        False,

    "model_ranks_accessed":
        False
}


with open(
    MANIFEST_FILE,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        manifest,
        f,
        indent=2
    )


print("\n" + "=" * 130)
print("STAGE 9B.0b COMPLETE — SOURCE RULE AUDIT ONLY [OK]")
print("=" * 130)

print(
    "\nCandidate 245 rule:",
    "FDR.2 < 0.05 OR FDR.3 < 0.05"
)

print(
    "Candidate count:",
    len(
        set_union
    )
)

print("\nValidation gene set finalized : NO")
print("Prior-evidence subsets finalized: NO")
print("Model rankings accessed        : NO")

print("\nPASTE BACK:")
print("1. RAW MULTI-ROW HEADER — COLUMNS 65 TO 89")
print("2. TESTING THE TWO SOURCE FDR BLOCKS")
print("3. PRIOR-EVIDENCE STRUCTURE WITHIN THE 245 CANDIDATE GENES")

print("=" * 130)
```

```python
# ==========================================================================================
# ASD-ContextFusion
# STAGE 9B.0b — TAN 2026 EXACT 245-GENE RULE RESOLUTION
#
# PURPOSE
# -------
# Test whether the paper's reported 245 ASD risk genes correspond exactly to:
#
#       FDR.2 < 0.05  OR  FDR.3 < 0.05
#
# while simultaneously verifying the raw source headers and inspecting
# prior-evidence annotations WITHIN that candidate set.
#
# IMPORTANT
# ---------
# - NO Context scores
# - NO NT scores
# - NO Fusion scores
# - NO model ranks
# - NO final validation gene-set lock yet
# ==========================================================================================

from pathlib import Path
import hashlib
import json
import re

import numpy as np
import pandas as pd


# ==========================================================================================
# 0. PATHS
# ==========================================================================================

ROOT = Path(r"E:\ASD_ContextFusion_Q1")

TAN_FILE = (
    ROOT /
    "01_raw" /
    "external_validation_2026" /
    "Tan_2026_MolecularPsychiatry_Supplementary_Tables_1_16.xlsx"
)

OUT_DIR = (
    ROOT /
    "02_interim" /
    "external_validation_source_lock"
)

OUT_DIR.mkdir(
    parents=True,
    exist_ok=True
)

SHEET = "Supplementary Table 9"

HEADER_ROW_ZERO_BASED = 3


# ==========================================================================================
# 1. HELPERS
# ==========================================================================================

def sha256_file(path, chunk_size=1024 * 1024):

    h = hashlib.sha256()

    with open(path, "rb") as f:

        while True:

            chunk = f.read(chunk_size)

            if not chunk:
                break

            h.update(chunk)

    return h.hexdigest()


def clean(x):

    if pd.isna(x):
        return ""

    return str(x).strip()


# ==========================================================================================
# 2. LOAD SOURCE ONLY
# ==========================================================================================

print("=" * 130)
print("ASD-CONTEXTFUSION — STAGE 9B.0b TAN EXACT 245-GENE RULE RESOLUTION")
print("=" * 130)

print("\n[1/6] SOURCE AUDIT")

assert TAN_FILE.exists()

print("Tan XLSX:")
print(TAN_FILE)

print("\nSHA256:")
print(sha256_file(TAN_FILE))


raw = pd.read_excel(
    TAN_FILE,
    sheet_name=SHEET,
    header=None,
    engine="openpyxl"
)

df = pd.read_excel(
    TAN_FILE,
    sheet_name=SHEET,
    header=HEADER_ROW_ZERO_BASED,
    engine="openpyxl"
)


assert "Genesymbol" in df.columns
assert "FDR.2" in df.columns
assert "FDR.3" in df.columns


print("\nRaw shape   :", raw.shape)
print("Parsed shape:", df.shape)


# ==========================================================================================
# 3. EXACT RAW HEADER AROUND SECOND ASD STATISTICAL BLOCK
# ==========================================================================================

print("\n[2/6] RAW MULTI-ROW HEADER — COLUMNS 65 TO 89")


header_slice = raw.iloc[
    0:4,
    65:90
].copy()


print(
    header_slice.to_string(
        index=True,
        header=True
    )
)


print("\nPARSED COLUMNS 65 TO 89")

for idx in range(65, 90):

    print(
        f"{idx:3d} | {df.columns[idx]}"
    )


# Save exact header provenance
HEADER_FILE = (
    OUT_DIR /
    "Stage9B0b_Tan_columns65_89_raw_header.tsv"
)

header_slice.to_csv(
    HEADER_FILE,
    sep="\t",
    index=True
)


# ==========================================================================================
# 4. EXACT FDR.2 / FDR.3 SET RELATIONSHIP
# ==========================================================================================

print("\n[3/6] TESTING THE TWO SOURCE FDR BLOCKS")


gene = (
    df[
        "Genesymbol"
    ]
    .astype(str)
    .str.strip()
)


valid_gene = (
    df[
        "Genesymbol"
    ].notna()
    &
    gene.ne("")
    &
    gene.ne("nan")
)


fdr2 = pd.to_numeric(
    df[
        "FDR.2"
    ],
    errors="coerce"
)

fdr3 = pd.to_numeric(
    df[
        "FDR.3"
    ],
    errors="coerce"
)


mask_fdr2 = (
    valid_gene
    &
    fdr2.notna()
    &
    (fdr2 < 0.05)
)


mask_fdr3 = (
    valid_gene
    &
    fdr3.notna()
    &
    (fdr3 < 0.05)
)


set_fdr2 = set(
    gene.loc[
        mask_fdr2
    ]
)


set_fdr3 = set(
    gene.loc[
        mask_fdr3
    ]
)


set_union = (
    set_fdr2
    |
    set_fdr3
)


set_intersection = (
    set_fdr2
    &
    set_fdr3
)


only_fdr2 = (
    set_fdr2
    -
    set_fdr3
)


only_fdr3 = (
    set_fdr3
    -
    set_fdr2
)


print(
    "FDR.2 < 0.05 genes :",
    len(
        set_fdr2
    )
)

print(
    "FDR.3 < 0.05 genes :",
    len(
        set_fdr3
    )
)

print(
    "Intersection       :",
    len(
        set_intersection
    )
)

print(
    "FDR.2 only        :",
    len(
        only_fdr2
    )
)

print(
    "FDR.3 only        :",
    len(
        only_fdr3
    )
)

print(
    "UNION             :",
    len(
        set_union
    )
)


expected_union = (
    len(
        set_fdr2
    )
    +
    len(
        set_fdr3
    )
    -
    len(
        set_intersection
    )
)


assert expected_union == len(
    set_union
)


if len(
    set_union
) != 245:

    raise RuntimeError(
        "\nFDR.2/FDR.3 union does NOT reproduce the "
        "published 245-gene count.\n"
        "Do NOT finalize the validation set."
    )


print(
    "\nFDR.2/FDR.3 UNION EXACTLY REPRODUCES "
    "THE PUBLISHED 245-GENE COUNT [OK]"
)


# ==========================================================================================
# 5. CREATE SOURCE-ORDER CANDIDATE 245 SET — NOT YET FINAL LOCK
# ==========================================================================================

print("\n[4/6] BUILDING SOURCE-ORDER 245-GENE CANDIDATE SET")


mask_union = (
    valid_gene
    &
    gene.isin(
        set_union
    )
)


candidate = df.loc[
    mask_union
].copy()


candidate[
    "Genesymbol"
] = (
    candidate[
        "Genesymbol"
    ]
    .astype(str)
    .str.strip()
)


assert len(
    candidate
) == 245

assert candidate[
    "Genesymbol"
].nunique() == 245


candidate.insert(
    0,
    "Tan_245_membership_rule",
    "FDR.2<0.05 OR FDR.3<0.05"
)


candidate.insert(
    1,
    "significant_FDR2",
    (
        pd.to_numeric(
            candidate[
                "FDR.2"
            ],
            errors="coerce"
        )
        <
        0.05
    )
)


candidate.insert(
    2,
    "significant_FDR3",
    (
        pd.to_numeric(
            candidate[
                "FDR.3"
            ],
            errors="coerce"
        )
        <
        0.05
    )
)


print(
    "Candidate rows       :",
    len(
        candidate
    )
)

print(
    "Unique candidate genes:",
    candidate[
        "Genesymbol"
    ].nunique()
)


print("\nSTATISTICAL CONTRIBUTION")

print(
    candidate[
        [
            "significant_FDR2",
            "significant_FDR3"
        ]
    ]
    .value_counts()
    .to_string()
)


# ==========================================================================================
# 6. PRIOR-EVIDENCE STRUCTURE WITHIN THE 245
# ==========================================================================================

print("\n[5/6] PRIOR-EVIDENCE STRUCTURE WITHIN THE 245 CANDIDATE GENES")


evidence_columns = [
    "diagnostic_category",
    "Evidence from PubMed",
    "SFARI_gene-score",
    "SFARI_syndromic",
    "SFARI_eagle score",
    "confidence category",
    "SPARK162 gene list",
    "SPARK404(P<0.001 in denovowest or TDT statistic≥1)",
    "membership in TADA ASD FDR < 0.001 list of 72",
    "membership in TADA ASD FDR < 0.05 list of 185",
    "MSSNG134(FDR<0.1 in TADA)",
    "MSSNG106(FDR<0.05 in TADA)"
]


available_evidence_columns = [
    col
    for col in evidence_columns
    if col in candidate.columns
]


print(
    "Evidence columns found:",
    len(
        available_evidence_columns
    )
)


for col in available_evidence_columns:

    print(
        "\n" +
        "-" * 110
    )

    print(
        "COLUMN:",
        col
    )


    counts = (
        candidate[
            col
        ]
        .fillna(
            "<NA>"
        )
        .astype(str)
        .str.strip()
        .value_counts(
            dropna=False
        )
    )


    print(
        counts.to_string()
    )


# ----------------------------------------------------------------------
# Missing/nonmissing profile of the main ASD-prior-evidence fields.
# This is descriptive only; NO category is defined yet.
# ----------------------------------------------------------------------

prior_profile = pd.DataFrame(
    {
        "Gene":
            candidate[
                "Genesymbol"
            ],

        "FDR2":
            pd.to_numeric(
                candidate[
                    "FDR.2"
                ],
                errors="coerce"
            ),

        "FDR3":
            pd.to_numeric(
                candidate[
                    "FDR.3"
                ],
                errors="coerce"
            )
    }
)


for col in available_evidence_columns:

    prior_profile[
        col
    ] = (
        candidate[
            col
        ]
        .reset_index(
            drop=True
        )
    )


# ==========================================================================================
# 7. SAVE CANDIDATE AUDIT — NOT FINAL GENE-SET LOCK
# ==========================================================================================

print("\n[6/6] SAVING AUDIT-ONLY OUTPUT")


CANDIDATE_FILE = (
    OUT_DIR /
    "Stage9B0b_Tan_245_candidate_UNION_NOT_FINAL.tsv"
)


PROFILE_FILE = (
    OUT_DIR /
    "Stage9B0b_Tan_245_prior_evidence_profile_NOT_FINAL.tsv"
)


SUMMARY_FILE = (
    OUT_DIR /
    "Stage9B0b_Tan_FDR2_FDR3_union_summary.tsv"
)


candidate.to_csv(
    CANDIDATE_FILE,
    sep="\t",
    index=False
)


prior_profile.to_csv(
    PROFILE_FILE,
    sep="\t",
    index=False
)


summary = pd.DataFrame(
    [
        {
            "FDR2_lt_0.05":
                len(
                    set_fdr2
                ),

            "FDR3_lt_0.05":
                len(
                    set_fdr3
                ),

            "intersection":
                len(
                    set_intersection
                ),

            "FDR2_only":
                len(
                    only_fdr2
                ),

            "FDR3_only":
                len(
                    only_fdr3
                ),

            "union":
                len(
                    set_union
                ),

            "published_target":
                245,

            "exact_match":
                len(
                    set_union
                ) == 245
        }
    ]
)


summary.to_csv(
    SUMMARY_FILE,
    sep="\t",
    index=False
)


MANIFEST_FILE = (
    ROOT /
    "00_admin" /
    "ASD_ContextFusion_STAGE9B0b_TAN_245_RULE_AUDIT.json"
)


manifest = {
    "project":
        "ASD-ContextFusion",

    "stage":
        "9B.0b",

    "source":
        "Tan 2026 Supplementary Table 9",

    "candidate_rule_tested":
        "FDR.2 < 0.05 OR FDR.3 < 0.05",

    "FDR2_count":
        len(
            set_fdr2
        ),

    "FDR3_count":
        len(
            set_fdr3
        ),

    "intersection_count":
        len(
            set_intersection
        ),

    "union_count":
        len(
            set_union
        ),

    "matches_published_245":
        len(
            set_union
        ) == 245,

    "validation_gene_set_finalized":
        False,

    "prior_evidence_subsets_finalized":
        False,

    "model_scores_accessed":
        False,

    "model_ranks_accessed":
        False
}


with open(
    MANIFEST_FILE,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        manifest,
        f,
        indent=2
    )


print("\n" + "=" * 130)
print("STAGE 9B.0b COMPLETE — SOURCE RULE AUDIT ONLY [OK]")
print("=" * 130)

print(
    "\nCandidate 245 rule:",
    "FDR.2 < 0.05 OR FDR.3 < 0.05"
)

print(
    "Candidate count:",
    len(
        set_union
    )
)

print("\nValidation gene set finalized : NO")
print("Prior-evidence subsets finalized: NO")
print("Model rankings accessed        : NO")

print("\nPASTE BACK:")
print("1. RAW MULTI-ROW HEADER — COLUMNS 65 TO 89")
print("2. TESTING THE TWO SOURCE FDR BLOCKS")
print("3. PRIOR-EVIDENCE STRUCTURE WITHIN THE 245 CANDIDATE GENES")

print("=" * 130)
```

```python
# ==========================================================================================
# ASD-ContextFusion
# STAGE 9B.1F — FINAL SOURCE-TRANSPARENT TAN EXTERNAL GENE-SET LOCK
#
# WHY "1F"?
# ---------
# The paper abstract reports:
#     245 ASD risk genes
#      45 with limited prior evidence
#      32 with no prior association
#
# Exact Supplementary Table 9 reproducibly contains, among those same 245:
#     167 "well-known"
#      46 "suggestive"
#      32 "no evidence"
#
# The LGD-recessive column DOES NOT reconcile 46 -> 45.
#
# Therefore:
#   - DO NOT force a 45-gene list.
#   - Preserve the paper-vs-supplement discrepancy explicitly.
#   - Use exact supplement-defined gene sets for reproducible validation.
#
# FINAL EXTERNAL SETS:
#   1. Tan_All245                       -> broad primary external set
#   2. Tan_NoEvidence32                -> primary novelty-focused set
#   3. Tan_Suggestive46                -> secondary source-defined set
#   4. Tan_SuggestiveOrNoEvidence78    -> secondary combined low-evidence set
#
# NO model scores or ranks are accessed in this stage.
# ==========================================================================================

from pathlib import Path
import pandas as pd
import numpy as np
import hashlib
import json
import re


# ==========================================================================================
# 0. PATHS / CONSTANTS
# ==========================================================================================

ROOT = Path(r"E:\ASD_ContextFusion_Q1")

TAN_FILE = (
    ROOT /
    "01_raw" /
    "external_validation_2026" /
    "Tan_2026_MolecularPsychiatry_Supplementary_Tables_1_16.xlsx"
)

MASTER_FILE = (
    ROOT /
    "03_features" /
    "multimodal_master" /
    "dhindsa_NTv2_row_preserved_master.tsv"
)

SEED_FILE = (
    ROOT /
    "00_admin" /
    "dhindsa_2025_monoallelic_ASD_190.tsv"
)

STAGE9A_LOCK = (
    ROOT /
    "00_admin" /
    "ASD_ContextFusion_STAGE9A_EXTERNAL_VALIDATION_SOURCE_LOCK.json"
)

OUT_DIR = (
    ROOT /
    "02_interim" /
    "external_validation_source_lock"
)

ADMIN_DIR = (
    ROOT /
    "00_admin"
)

OUT_DIR.mkdir(
    parents=True,
    exist_ok=True
)

EXPECTED_UNIVERSE = 18252
EXPECTED_SEEDS = 190

EXPECTED_ALL = 245
EXPECTED_SUPPLEMENT_WELLKNOWN = 167
EXPECTED_SUPPLEMENT_SUGGESTIVE = 46
EXPECTED_SUPPLEMENT_NOEVIDENCE = 32

PAPER_ABSTRACT_LIMITED_PRIOR = 45
PAPER_ABSTRACT_NO_PRIOR = 32

BOOTSTRAP_SEED = 20260830
N_BOOTSTRAP = 5000


# ==========================================================================================
# 1. HELPERS
# ==========================================================================================

def sha256_file(
    path,
    chunk_size=1024 * 1024
):

    h = hashlib.sha256()

    with open(
        path,
        "rb"
    ) as f:

        while True:

            chunk = f.read(
                chunk_size
            )

            if not chunk:
                break

            h.update(
                chunk
            )

    return h.hexdigest()


def normalize_group(x):

    if pd.isna(x):
        return ""

    x = str(
        x
    ).strip().lower()

    x = (
        x.replace("–", "-")
         .replace("—", "-")
         .replace("_", " ")
    )

    x = re.sub(
        r"\s+",
        " ",
        x
    )

    return x.strip()


def identify_seed_column(
    df
):

    preferred = [
        "Gene_Name",
        "Gene",
        "gene",
        "Genesymbol",
        "GeneSymbol",
        "gene_symbol",
        "symbol"
    ]

    for col in preferred:

        if col in df.columns:

            return col

    if df.shape[1] == 1:

        return df.columns[0]

    raise RuntimeError(
        "Cannot identify seed-gene column."
    )


def save_gene_set(
    genes,
    path,
    set_name,
    source_order
):

    genes = set(
        genes
    )

    order = {
        gene: i
        for i, gene
        in enumerate(
            source_order
        )
    }

    ordered = sorted(
        genes,
        key=lambda x:
            order.get(
                x,
                10**12
            )
    )

    out = pd.DataFrame(
        {
            "Gene_Name":
                ordered,

            "external_set":
                set_name
        }
    )

    out.to_csv(
        path,
        sep="\t",
        index=False
    )

    return out


# ==========================================================================================
# 2. INPUT PROVENANCE AUDIT
# ==========================================================================================

print(
    "=" * 138
)

print(
    "ASD-CONTEXTFUSION — STAGE 9B.1F FINAL SOURCE-TRANSPARENT EXTERNAL LOCK"
)

print(
    "=" * 138
)

print(
    "\n[1/9] INPUT / PROVENANCE AUDIT"
)


for path in [
    TAN_FILE,
    MASTER_FILE,
    SEED_FILE,
    STAGE9A_LOCK
]:

    assert path.exists(), (
        f"Missing required file:\n{path}"
    )

    print(
        "[OK]",
        path
    )


with open(
    STAGE9A_LOCK,
    "r",
    encoding="utf-8"
) as f:

    stage9a = json.load(
        f
    )


assert stage9a[
    "model_scores_accessed"
] is False

assert stage9a[
    "model_ranks_accessed"
] is False


print(
    "\nTan XLSX SHA256:"
)

print(
    sha256_file(
        TAN_FILE
    )
)


# ==========================================================================================
# 3. RECONSTRUCT EXACT 245 RISK GENES
# ==========================================================================================

print(
    "\n[2/9] RECONSTRUCTING EXACT 245 ASD RISK GENES"
)


df = pd.read_excel(
    TAN_FILE,
    sheet_name="Supplementary Table 9",
    header=3,
    engine="openpyxl"
)


required = {
    "Genesymbol",
    "FDR.2",
    "FDR.3",
    "Unnamed: 88",
    "Unnamed: 89"
}


assert required.issubset(
    set(
        df.columns
    )
)


gene = (
    df[
        "Genesymbol"
    ]
    .fillna("")
    .astype(str)
    .str.strip()
)


valid_gene = (
    gene.ne("")
    &
    gene.ne("nan")
)


fdr_tada = pd.to_numeric(
    df[
        "FDR.2"
    ],
    errors="coerce"
)


fdr_denovowest = pd.to_numeric(
    df[
        "FDR.3"
    ],
    errors="coerce"
)


sig_tada = (
    valid_gene
    &
    fdr_tada.notna()
    &
    (
        fdr_tada
        <
        0.05
    )
)


sig_denovowest = (
    valid_gene
    &
    fdr_denovowest.notna()
    &
    (
        fdr_denovowest
        <
        0.05
    )
)


risk_mask = (
    sig_tada
    |
    sig_denovowest
)


risk = df.loc[
    risk_mask
].copy()


risk.insert(
    0,
    "Gene_Name",
    gene.loc[
        risk_mask
    ].values
)


risk.insert(
    1,
    "TADA_FDR_lt_0.05",
    sig_tada.loc[
        risk_mask
    ].values
)


risk.insert(
    2,
    "DeNovoWEST_ASD_all_FDR_lt_0.05",
    sig_denovowest.loc[
        risk_mask
    ].values
)


assert len(
    risk
) == EXPECTED_ALL


assert risk[
    "Gene_Name"
].nunique() == EXPECTED_ALL


print(
    "TADA FDR<0.05       :",
    int(
        sig_tada.sum()
    )
)

print(
    "DeNovoWEST FDR<0.05 :",
    int(
        sig_denovowest.sum()
    )
)

print(
    "Union               :",
    risk[
        "Gene_Name"
    ].nunique()
)


# ==========================================================================================
# 4. PRESERVE EXACT SUPPLEMENT SOURCE GROUPS
# ==========================================================================================

print(
    "\n[3/9] SUPPLEMENT-DEFINED PRIOR-EVIDENCE GROUPS"
)


risk[
    "dominant_group_source"
] = (
    risk[
        "Unnamed: 88"
    ]
    .map(
        normalize_group
    )
)


risk[
    "lgd_recessive_group_source"
] = (
    risk[
        "Unnamed: 89"
    ]
    .map(
        normalize_group
    )
)


dominant_counts = (
    risk[
        "dominant_group_source"
    ]
    .replace(
        "",
        "<NA>"
    )
    .value_counts()
)


print(
    "\nSupplementary Table 9 dominant group:"
)

print(
    dominant_counts.to_string()
)


n_wellknown = int(
    (
        risk[
            "dominant_group_source"
        ]
        ==
        "well-known"
    ).sum()
)


n_suggestive = int(
    (
        risk[
            "dominant_group_source"
        ]
        ==
        "suggestive"
    ).sum()
)


n_noevidence = int(
    risk[
        "dominant_group_source"
    ].isin(
        [
            "no evidence",
            "no-evidence"
        ]
    ).sum()
)


assert n_wellknown == EXPECTED_SUPPLEMENT_WELLKNOWN
assert n_suggestive == EXPECTED_SUPPLEMENT_SUGGESTIVE
assert n_noevidence == EXPECTED_SUPPLEMENT_NOEVIDENCE


assert (
    n_wellknown
    +
    n_suggestive
    +
    n_noevidence
    ==
    EXPECTED_ALL
)


print(
    "\nExact supplementary partition [OK]"
)

print(
    "Well-known :",
    n_wellknown
)

print(
    "Suggestive :",
    n_suggestive
)

print(
    "No evidence:",
    n_noevidence
)


# ==========================================================================================
# 5. EXPLICIT PAPER-vs-SUPPLEMENT DISCREPANCY LOCK
# ==========================================================================================

print(
    "\n[4/9] PAPER-vs-SUPPLEMENT DISCREPANCY AUDIT"
)


limited_count_matches = (
    n_suggestive
    ==
    PAPER_ABSTRACT_LIMITED_PRIOR
)


no_prior_count_matches = (
    n_noevidence
    ==
    PAPER_ABSTRACT_NO_PRIOR
)


print(
    "Paper abstract limited-prior count :",
    PAPER_ABSTRACT_LIMITED_PRIOR
)

print(
    "Supplement suggestive count        :",
    n_suggestive
)

print(
    "Exact match                        :",
    limited_count_matches
)


print(
    "\nPaper abstract no-prior count       :",
    PAPER_ABSTRACT_NO_PRIOR
)

print(
    "Supplement no-evidence count       :",
    n_noevidence
)

print(
    "Exact match                        :",
    no_prior_count_matches
)


assert limited_count_matches is False

assert no_prior_count_matches is True


print(
    "\n[LOCKED INTERPRETATION]"
)

print(
    "Do NOT force a 45-gene list."
)

print(
    "Do NOT relabel the 46 supplement-defined suggestive genes "
    "as an exact reproduction of the abstract's 45 limited-prior genes."
)

print(
    "No-evidence32 exactly reproduces the abstract's no-prior count."
)


# ==========================================================================================
# 6. PROJECT UNIVERSE / 190-SEED AUDIT
# ==========================================================================================

print(
    "\n[5/9] PROJECT UNIVERSE / LOCKED SEED AUDIT"
)


master = pd.read_csv(
    MASTER_FILE,
    sep="\t"
)


assert "Gene_Name" in master.columns
assert "known_gene" in master.columns


universe = set(
    master[
        "Gene_Name"
    ]
    .astype(str)
    .str.strip()
)


assert len(
    universe
) == EXPECTED_UNIVERSE


master_seeds = set(
    master.loc[
        master[
            "known_gene"
        ] == 1,
        "Gene_Name"
    ]
    .astype(str)
    .str.strip()
)


assert len(
    master_seeds
) == EXPECTED_SEEDS


seed_df = pd.read_csv(
    SEED_FILE,
    sep="\t"
)


seed_col = identify_seed_column(
    seed_df
)


file_seeds = set(
    seed_df[
        seed_col
    ]
    .astype(str)
    .str.strip()
)


assert len(
    file_seeds
) == EXPECTED_SEEDS


assert file_seeds == master_seeds


print(
    "Unique project universe:",
    len(
        universe
    )
)

print(
    "Locked seed genes      :",
    len(
        master_seeds
    )
)

print(
    "Seed identity          : EXACT MATCH [OK]"
)


# ==========================================================================================
# 7. DEFINE FINAL REPRODUCIBLE EXTERNAL SETS
# ==========================================================================================

print(
    "\n[6/9] DEFINING FINAL REPRODUCIBLE EXTERNAL SETS"
)


ALL_245 = set(
    risk[
        "Gene_Name"
    ]
)


SUGGESTIVE_46 = set(
    risk.loc[
        risk[
            "dominant_group_source"
        ]
        ==
        "suggestive",
        "Gene_Name"
    ]
)


NO_EVIDENCE_32 = set(
    risk.loc[
        risk[
            "dominant_group_source"
        ].isin(
            [
                "no evidence",
                "no-evidence"
            ]
        ),
        "Gene_Name"
    ]
)


SUGGESTIVE_OR_NO_EVIDENCE_78 = (
    SUGGESTIVE_46
    |
    NO_EVIDENCE_32
)


assert len(
    ALL_245
) == 245

assert len(
    SUGGESTIVE_46
) == 46

assert len(
    NO_EVIDENCE_32
) == 32

assert len(
    SUGGESTIVE_OR_NO_EVIDENCE_78
) == 78


assert SUGGESTIVE_46.isdisjoint(
    NO_EVIDENCE_32
)


SETS = {

    "Tan_All245":
        ALL_245,

    "Tan_SupplementSuggestive46":
        SUGGESTIVE_46,

    "Tan_SupplementNoEvidence32":
        NO_EVIDENCE_32,

    "Tan_SupplementSuggestiveOrNoEvidence78":
        SUGGESTIVE_OR_NO_EVIDENCE_78
}


source_order = (
    risk[
        "Gene_Name"
    ].tolist()
)


summary_rows = []


for set_name, geneset in SETS.items():

    in_universe = (
        geneset
        &
        universe
    )


    absent = (
        geneset
        -
        universe
    )


    seed_overlap = (
        in_universe
        &
        master_seeds
    )


    nonseed = (
        in_universe
        -
        master_seeds
    )


    summary_rows.append(
        {
            "set":
                set_name,

            "source_count":
                len(
                    geneset
                ),

            "in_project_universe":
                len(
                    in_universe
                ),

            "absent_from_project_universe":
                len(
                    absent
                ),

            "overlap_locked_190_seeds":
                len(
                    seed_overlap
                ),

            "nonseed_external_validation":
                len(
                    nonseed
                )
        }
    )


summary = pd.DataFrame(
    summary_rows
)


print(
    summary.to_string(
        index=False
    )
)


# ==========================================================================================
# 8. WRITE ALL LOCKED LISTS + SHA256
# ==========================================================================================

print(
    "\n[7/9] WRITING FINAL LOCKED GENE LISTS"
)


hash_rows = []


for set_name, geneset in SETS.items():

    in_universe = (
        geneset
        &
        universe
    )


    seed_overlap = (
        in_universe
        &
        master_seeds
    )


    nonseed = (
        in_universe
        -
        master_seeds
    )


    absent = (
        geneset
        -
        universe
    )


    file_sets = {

        "SOURCE_LOCKED":
            geneset,

        "IN_PROJECT_UNIVERSE":
            in_universe,

        "SEED_OVERLAP":
            seed_overlap,

        "NONSEED_VALIDATION_LOCKED":
            nonseed,

        "ABSENT_FROM_PROJECT_UNIVERSE":
            absent
    }


    for file_type, genes in file_sets.items():

        path = (
            OUT_DIR /
            f"Stage9B1F_{set_name}_{file_type}.tsv"
        )


        save_gene_set(
            genes=genes,
            path=path,
            set_name=set_name,
            source_order=source_order
        )


        hash_rows.append(
            {
                "set":
                    set_name,

                "file_type":
                    file_type,

                "gene_count":
                    len(
                        genes
                    ),

                "path":
                    str(
                        path
                    ),

                "sha256":
                    sha256_file(
                        path
                    )
            }
        )


# Complete source rows for provenance.
SOURCE_EVIDENCE_FILE = (
    OUT_DIR /
    "Stage9B1F_Tan245_complete_source_evidence_LOCKED.tsv.gz"
)


risk.to_csv(
    SOURCE_EVIDENCE_FILE,
    sep="\t",
    index=False,
    compression="gzip"
)


SUMMARY_FILE = (
    OUT_DIR /
    "Stage9B1F_external_validation_set_summary.tsv"
)


summary.to_csv(
    SUMMARY_FILE,
    sep="\t",
    index=False
)


HASH_FILE = (
    OUT_DIR /
    "Stage9B1F_external_gene_set_SHA256.tsv"
)


hash_df = pd.DataFrame(
    hash_rows
)


hash_df.to_csv(
    HASH_FILE,
    sep="\t",
    index=False
)


# Save discrepancy record separately.
DISCREPANCY_FILE = (
    OUT_DIR /
    "Stage9B1F_paper_vs_supplement_count_discrepancy.tsv"
)


discrepancy = pd.DataFrame(
    [
        {
            "category":
                "limited_prior_vs_suggestive",

            "paper_abstract_count":
                45,

            "supplement_source_label":
                "suggestive",

            "supplement_count":
                46,

            "exact_match":
                False,

            "resolution":
                (
                    "Unresolved source discrepancy. "
                    "No gene removed or reclassified."
                )
        },

        {
            "category":
                "no_prior_vs_no_evidence",

            "paper_abstract_count":
                32,

            "supplement_source_label":
                "no evidence",

            "supplement_count":
                32,

            "exact_match":
                True,

            "resolution":
                (
                    "Exact count agreement."
                )
        }
    ]
)


discrepancy.to_csv(
    DISCREPANCY_FILE,
    sep="\t",
    index=False
)


# ==========================================================================================
# 9. PRESPECIFY STAGE 9C BEFORE MODEL SCORES ARE OPENED
# ==========================================================================================

print(
    "\n[8/9] PRESPECIFYING EXTERNAL-VALIDATION ANALYSIS"
)


VALIDATION_PLAN = {

    "models_primary":
        [
            "Full_Context_316",
            "NTv2",
            "Strict_Fusion"
        ],

    "primary_model_comparison":
        "Strict_Fusion_vs_Full_Context_316",

    "primary_broad_external_set":
        "Tan_All245_NONSEED_VALIDATION_LOCKED",

    "primary_novelty_focused_set":
        "Tan_SupplementNoEvidence32_NONSEED_VALIDATION_LOCKED",

    "secondary_external_sets":
        [
            "Tan_SupplementSuggestive46_NONSEED_VALIDATION_LOCKED",
            "Tan_SupplementSuggestiveOrNoEvidence78_NONSEED_VALIDATION_LOCKED"
        ],

    "abstract_limited_prior_45_set_used":
        False,

    "reason_45_not_used":
        (
            "Exact 45-gene list was not reproducible from "
            "Supplementary Table 9 without arbitrary reconciliation."
        ),

    "rank_direction":
        (
            "Higher model probability = stronger ASD prioritization; "
            "100 percentile = highest-ranked gene."
        ),

    "prespecified_top_percent_thresholds":
        [
            1,
            5,
            10
        ],

    "continuous_metric_primary":
        "rank_percentile",

    "recovery_metrics":
        [
            "top_1_percent_recovery",
            "top_5_percent_recovery",
            "top_10_percent_recovery"
        ],

    "enrichment_background":
        (
            "Frozen 18,252-gene project universe."
        ),

    "paired_continuous_inference":
        (
            "Paired gene-level bootstrap of "
            "Fusion-minus-Context rank-percentile difference."
        ),

    "paired_bootstrap_replicates":
        N_BOOTSTRAP,

    "paired_bootstrap_seed":
        BOOTSTRAP_SEED,

    "primary_external_population_rule":
        (
            "External source genes present in the frozen "
            "18,252-gene universe after removal of the "
            "locked 190 ASD seed genes."
        ),

    "model_selection_or_tuning_allowed":
        False
}


PLAN_FILE = (
    ADMIN_DIR /
    "ASD_ContextFusion_STAGE9C_EXTERNAL_VALIDATION_PRESPECIFICATION.json"
)


with open(
    PLAN_FILE,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        VALIDATION_PLAN,
        f,
        indent=2
    )


# ==========================================================================================
# 10. FINAL SCIENTIFIC LOCK
# ==========================================================================================

LOCK_FILE = (
    ADMIN_DIR /
    "ASD_ContextFusion_STAGE9B1F_EXTERNAL_GENE_SET_FINAL_LOCK.json"
)


manifest = {

    "project":
        "ASD-ContextFusion",

    "stage":
        "9B.1F_source_transparent_final_external_lock",

    "status":
        "FINAL_LOCKED_BEFORE_MODEL_RANK_ACCESS",

    "source":
        "Tan et al. Molecular Psychiatry 2026",

    "source_file":
        str(
            TAN_FILE
        ),

    "source_sha256":
        sha256_file(
            TAN_FILE
        ),

    "source_sheet":
        "Supplementary Table 9",

    "risk_gene_rule":
        (
            "TADA FDR<0.05 OR "
            "DeNovoWEST_ASD_all FDR<0.05"
        ),

    "risk_gene_count":
        245,

    "supplement_dominant_partition":
        {
            "well_known":
                167,

            "suggestive":
                46,

            "no_evidence":
                32
        },

    "paper_abstract_counts":
        {
            "limited_prior_evidence":
                45,

            "no_prior_association":
                32
        },

    "paper_vs_supplement_limited_count_discrepancy":
        True,

    "limited_prior_45_gene_list_claimed":
        False,

    "arbitrary_reconciliation_performed":
        False,

    "recessive_override_used":
        False,

    "primary_broad_external_set":
        "Tan_All245",

    "primary_novelty_focused_set":
        "Tan_SupplementNoEvidence32",

    "secondary_suggestive_set":
        "Tan_SupplementSuggestive46",

    "combined_secondary_set":
        "Tan_SupplementSuggestiveOrNoEvidence78",

    "project_universe_count":
        18252,

    "locked_seed_count":
        190,

    "model_scores_accessed_before_lock":
        False,

    "model_ranks_accessed_before_lock":
        False,

    "top_k_thresholds_prespecified_before_rank_access":
        [
            1,
            5,
            10
        ],

    "new_PU_sampling":
        False,

    "result_driven_gene_removal":
        False,

    "summary":
        summary.to_dict(
            orient="records"
        ),

    "locked_gene_files":
        hash_df.to_dict(
            orient="records"
        )
}


with open(
    LOCK_FILE,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        manifest,
        f,
        indent=2
    )


# ==========================================================================================
# 11. FINAL REPORT
# ==========================================================================================

print(
    "\n[9/9] FINAL REPORT"
)


print(
    "\n" +
    "=" * 138
)

print(
    "STAGE 9B.1F FINAL SOURCE-TRANSPARENT EXTERNAL LOCK COMPLETE [OK]"
)

print(
    "=" * 138
)


print(
    "\nSOURCE COUNTS"
)

print(
    "All significant ASD genes     : 245"
)

print(
    "Supplement well-known         : 167"
)

print(
    "Supplement suggestive         : 46"
)

print(
    "Supplement no evidence        : 32"
)


print(
    "\nPAPER-vs-SUPPLEMENT"
)

print(
    "Paper limited-prior count     : 45"
)

print(
    "Supplement suggestive count   : 46"
)

print(
    "Forced reconciliation         : NO"
)

print(
    "Exact 45-gene list claimed    : NO"
)

print(
    "Paper no-prior count          : 32"
)

print(
    "Supplement no-evidence count  : 32 [EXACT]"
)


print(
    "\nFINAL EXTERNAL SET SUMMARY"
)

print(
    summary.to_string(
        index=False
    )
)


print(
    "\nINTEGRITY"
)

print(
    "Model scores accessed         : NO"
)

print(
    "Model ranks accessed          : NO"
)

print(
    "Result-driven exclusions      : NO"
)

print(
    "Arbitrary gene reconciliation : NO"
)

print(
    "New PU sampling               : NO"
)

print(
    "Gene-set SHA256 lock          : YES"
)

print(
    "Stage 9C plan prespecified    : YES"
)


print(
    "\nFINAL LOCK:"
)

print(
    LOCK_FILE
)


print(
    "\nSTAGE 9C PRESPECIFICATION:"
)

print(
    PLAN_FILE
)


print(
    "\nNEXT SCIENTIFIC STAGE:"
)

print(
    "Stage 9C — FIRST model-rank access."
)

print(
    "Primary broad set    : non-seed Tan All245."
)

print(
    "Primary novelty set  : non-seed supplement-defined NoEvidence32."
)

print(
    "Secondary set        : non-seed supplement-defined Suggestive46."
)

print(
    "Top thresholds       : 1%, 5%, 10% [LOCKED BEFORE SCORE ACCESS]."
)

print(
    "=" * 138
)
```

```python
# ==========================================================================================
# ASD-ContextFusion
# STAGE 9B.2 — EXTERNAL-GENE IDENTITY / ABSENT-GENE AUDIT
#
# PURPOSE
# -------
# Resolve whether the 10 Tan source genes absent from the 18,252-gene
# common universe are:
#
#   A) genuinely outside the locked project universe, OR
#   B) recoverable through exact source/project identity lineage
#      (e.g. Ensembl ID / GENCODE symbol rename).
#
# IMPORTANT
# ---------
# - NO Context probability
# - NO NT probability
# - NO Fusion probability
# - NO model rank
# - NO result-driven mapping
# - NO fuzzy gene-name matching
#
# Only exact stable-ID / project-metadata identity evidence is allowed.
# ==========================================================================================

from pathlib import Path
import pandas as pd
import numpy as np
import hashlib
import json
import re


ROOT = Path(r"E:\ASD_ContextFusion_Q1")

SOURCE_EVIDENCE = (
    ROOT /
    "02_interim" /
    "external_validation_source_lock" /
    "Stage9B1F_Tan245_complete_source_evidence_LOCKED.tsv.gz"
)

MASTER_FILE = (
    ROOT /
    "03_features" /
    "multimodal_master" /
    "dhindsa_NTv2_row_preserved_master.tsv"
)

NT_META_FILE = (
    ROOT /
    "03_features" /
    "nt_v2" /
    "NTv2_embedding_row_metadata.tsv"
)

FINAL_LOCK = (
    ROOT /
    "00_admin" /
    "ASD_ContextFusion_STAGE9B1F_EXTERNAL_GENE_SET_FINAL_LOCK.json"
)

OUT_DIR = (
    ROOT /
    "02_interim" /
    "external_validation_source_lock"
)

OUT_DIR.mkdir(
    parents=True,
    exist_ok=True
)


# ==========================================================================================
# HELPERS
# ==========================================================================================

def sha256_file(path, chunk_size=1024 * 1024):

    h = hashlib.sha256()

    with open(path, "rb") as f:

        while True:

            chunk = f.read(chunk_size)

            if not chunk:
                break

            h.update(chunk)

    return h.hexdigest()


def clean_symbol(x):

    if pd.isna(x):
        return ""

    return str(x).strip()


def clean_ensembl(x):

    if pd.isna(x):
        return ""

    x = str(x).strip()

    # Remove transcript/gene version suffix only.
    x = re.sub(
        r"\.\d+$",
        "",
        x
    )

    return x


def normalize_numeric_id(x):

    if pd.isna(x):
        return ""

    x = str(x).strip()

    if re.fullmatch(
        r"\d+\.0",
        x
    ):
        x = x[:-2]

    return x


# ==========================================================================================
# 1. INPUT AUDIT
# ==========================================================================================

print("=" * 132)
print("ASD-CONTEXTFUSION — STAGE 9B.2 ABSENT-GENE IDENTITY AUDIT")
print("=" * 132)


for path in [
    SOURCE_EVIDENCE,
    MASTER_FILE,
    NT_META_FILE,
    FINAL_LOCK
]:

    assert path.exists(), path
    print("[OK]", path)


with open(
    FINAL_LOCK,
    "r",
    encoding="utf-8"
) as f:

    lock = json.load(f)


assert lock[
    "model_scores_accessed_before_lock"
] is False

assert lock[
    "model_ranks_accessed_before_lock"
] is False


# ==========================================================================================
# 2. LOAD SOURCE / PROJECT IDENTITY TABLES
# ==========================================================================================

risk = pd.read_csv(
    SOURCE_EVIDENCE,
    sep="\t",
    compression="gzip"
)

master = pd.read_csv(
    MASTER_FILE,
    sep="\t"
)

nt = pd.read_csv(
    NT_META_FILE,
    sep="\t"
)


assert len(risk) == 245
assert risk["Gene_Name"].nunique() == 245

assert master["Gene_Name"].nunique() == 18252


project_genes = set(
    master[
        "Gene_Name"
    ]
    .astype(str)
    .str.strip()
)


source_genes = set(
    risk[
        "Gene_Name"
    ]
    .astype(str)
    .str.strip()
)


absent_genes = (
    source_genes
    -
    project_genes
)


print("\nSOURCE GENES :", len(source_genes))
print("PROJECT      :", len(project_genes))
print("ABSENT       :", len(absent_genes))

assert len(absent_genes) == 10


# ==========================================================================================
# 3. SOURCE DETAILS FOR THE 10 ABSENT GENES
# ==========================================================================================

source_id_columns = [
    col
    for col in [
        "Gene_Name",
        "ENSENBL",
        "Entrez",
        "HGNC",
        "Genesymbol",
        "dominant_group_source",
        "lgd_recessive_group_source",
        "FDR.2",
        "FDR.3"
    ]
    if col in risk.columns
]


absent_source = (
    risk.loc[
        risk["Gene_Name"].isin(
            absent_genes
        ),
        source_id_columns
    ]
    .copy()
)


print("\n" + "=" * 132)
print("10 SOURCE GENES ABSENT FROM EXACT Gene_Name UNIVERSE")
print("=" * 132)

print(
    absent_source.to_string(
        index=False
    )
)


# ==========================================================================================
# 4. AUDIT AVAILABLE NT METADATA IDENTITY COLUMNS
# ==========================================================================================

print("\nNT METADATA COLUMNS")

for i, col in enumerate(nt.columns):

    print(
        f"{i:3d} | {col}"
    )


candidate_symbol_cols = [
    col
    for col in nt.columns
    if any(
        token in col.lower()
        for token in [
            "symbol",
            "gene_name"
        ]
    )
]


candidate_ensembl_cols = [
    col
    for col in nt.columns
    if (
        "ensembl" in col.lower()
        or
        "gene_id" in col.lower()
    )
]


candidate_hgnc_cols = [
    col
    for col in nt.columns
    if "hgnc" in col.lower()
]


candidate_entrez_cols = [
    col
    for col in nt.columns
    if (
        "entrez" in col.lower()
        or
        "ncbi_gene" in col.lower()
    )
]


print("\nCandidate symbol columns :", candidate_symbol_cols)
print("Candidate Ensembl columns:", candidate_ensembl_cols)
print("Candidate HGNC columns   :", candidate_hgnc_cols)
print("Candidate Entrez columns :", candidate_entrez_cols)


# ==========================================================================================
# 5. EXACT METADATA IDENTITY MATCHING
# ==========================================================================================

audit_rows = []


for _, src in absent_source.iterrows():

    source_gene = clean_symbol(
        src.get(
            "Gene_Name",
            ""
        )
    )


    source_ens = clean_ensembl(
        src.get(
            "ENSENBL",
            ""
        )
    )


    source_hgnc = normalize_numeric_id(
        src.get(
            "HGNC",
            ""
        )
    )


    source_entrez = normalize_numeric_id(
        src.get(
            "Entrez",
            ""
        )
    )


    evidence = []


    # ------------------------------------------------------------------
    # Exact symbol matches in NT metadata.
    # ------------------------------------------------------------------

    for col in candidate_symbol_cols:

        vals = (
            nt[col]
            .fillna("")
            .astype(str)
            .str.strip()
        )


        hit_idx = nt.index[
            vals == source_gene
        ].tolist()


        for idx in hit_idx:

            evidence.append(
                (
                    "symbol",
                    col,
                    idx
                )
            )


    # ------------------------------------------------------------------
    # Exact stable Ensembl matches.
    # ------------------------------------------------------------------

    if source_ens:

        for col in candidate_ensembl_cols:

            vals = (
                nt[col]
                .map(
                    clean_ensembl
                )
            )


            hit_idx = nt.index[
                vals == source_ens
            ].tolist()


            for idx in hit_idx:

                evidence.append(
                    (
                        "ensembl",
                        col,
                        idx
                    )
                )


    # ------------------------------------------------------------------
    # Exact HGNC matches if metadata provides them.
    # ------------------------------------------------------------------

    if source_hgnc:

        for col in candidate_hgnc_cols:

            vals = (
                nt[col]
                .map(
                    normalize_numeric_id
                )
            )


            hit_idx = nt.index[
                vals == source_hgnc
            ].tolist()


            for idx in hit_idx:

                evidence.append(
                    (
                        "hgnc",
                        col,
                        idx
                    )
                )


    # ------------------------------------------------------------------
    # Exact Entrez matches if metadata provides them.
    # ------------------------------------------------------------------

    if source_entrez:

        for col in candidate_entrez_cols:

            vals = (
                nt[col]
                .map(
                    normalize_numeric_id
                )
            )


            hit_idx = nt.index[
                vals == source_entrez
            ].tolist()


            for idx in hit_idx:

                evidence.append(
                    (
                        "entrez",
                        col,
                        idx
                    )
                )


    # ------------------------------------------------------------------
    # Resolve corresponding PROJECT symbols conservatively.
    # ------------------------------------------------------------------

    matched_project_symbols = set()


    for match_type, col, idx in evidence:

        row = nt.loc[idx]


        possible_symbols = []


        for symbol_col in [
            "original_gene_symbol",
            "gencode_gene_symbol",
            "Gene_Name"
        ]:

            if symbol_col in nt.columns:

                value = clean_symbol(
                    row[
                        symbol_col
                    ]
                )


                if value:

                    possible_symbols.append(
                        value
                    )


        for symbol in possible_symbols:

            if symbol in project_genes:

                matched_project_symbols.add(
                    symbol
                )


    if len(
        matched_project_symbols
    ) == 1:

        status = (
            "EXACT_PROJECT_IDENTITY_MATCH"
        )

        proposed_project_gene = list(
            matched_project_symbols
        )[0]


    elif len(
        matched_project_symbols
    ) == 0:

        status = (
            "NO_PROJECT_IDENTITY_MATCH"
        )

        proposed_project_gene = ""


    else:

        status = (
            "AMBIGUOUS_MULTIPLE_PROJECT_MATCHES"
        )

        proposed_project_gene = " | ".join(
            sorted(
                matched_project_symbols
            )
        )


    audit_rows.append(
        {
            "source_gene":
                source_gene,

            "source_ensembl":
                source_ens,

            "source_hgnc":
                source_hgnc,

            "source_entrez":
                source_entrez,

            "source_group":
                src.get(
                    "dominant_group_source",
                    ""
                ),

            "identity_status":
                status,

            "project_gene_candidate":
                proposed_project_gene,

            "number_project_candidates":
                len(
                    matched_project_symbols
                ),

            "metadata_evidence_count":
                len(
                    evidence
                ),

            "metadata_evidence":
                " || ".join(
                    [
                        f"{x[0]}:{x[1]}:row{x[2]}"
                        for x in evidence
                    ]
                )
        }
    )


audit = pd.DataFrame(
    audit_rows
)


print("\n" + "=" * 132)
print("IDENTITY AUDIT RESULT")
print("=" * 132)

print(
    audit.to_string(
        index=False
    )
)


print("\nSTATUS COUNTS")

print(
    audit[
        "identity_status"
    ]
    .value_counts()
    .to_string()
)


# ==========================================================================================
# 6. SAVE — AUDIT ONLY
# ==========================================================================================

AUDIT_FILE = (
    OUT_DIR /
    "Stage9B2_Tan10_absent_gene_identity_audit.tsv"
)


audit.to_csv(
    AUDIT_FILE,
    sep="\t",
    index=False
)


MANIFEST_FILE = (
    ROOT /
    "00_admin" /
    "ASD_ContextFusion_STAGE9B2_ABSENT_GENE_IDENTITY_AUDIT.json"
)


manifest = {

    "project":
        "ASD-ContextFusion",

    "stage":
        "9B.2_absent_gene_identity_audit",

    "source_gene_count":
        245,

    "exact_project_gene_overlap":
        235,

    "exact_symbol_absent":
        10,

    "identity_matching":
        (
            "Exact project metadata / stable-ID evidence only. "
            "No fuzzy matching."
        ),

    "model_scores_accessed":
        False,

    "model_ranks_accessed":
        False,

    "automatic_gene_remapping_performed":
        False,

    "audit_file":
        str(
            AUDIT_FILE
        ),

    "audit_sha256":
        sha256_file(
            AUDIT_FILE
        )
}


with open(
    MANIFEST_FILE,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        manifest,
        f,
        indent=2
    )


print("\n" + "=" * 132)
print("STAGE 9B.2 ABSENT-GENE IDENTITY AUDIT COMPLETE [OK]")
print("=" * 132)

print("\nAutomatic remapping performed : NO")
print("Model scores accessed         : NO")
print("Model ranks accessed          : NO")
print("Fuzzy matching                : NO")

print("\nSAVED:")
print(AUDIT_FILE)
print(MANIFEST_FILE)

print("\nNEXT:")
print(
    "Resolve only exact identity-supported mappings, if any; "
    "otherwise retain genuine out-of-universe exclusions."
)

print("=" * 132)
```

```python
# ==========================================================================================
# ASD-ContextFusion
# STAGE 9B.3 — FINAL PROJECT-IDENTITY RECONCILIATION
#              + STAGE 9C PRESPECIFICATION V2
#
# IMPORTANT SCIENTIFIC DISTINCTION
# --------------------------------
# External source identity remains the Tan gene symbol.
# Project scoring identity may be an older/historical symbol ONLY when
# exact stable-ID evidence already proved the two identities equivalent.
#
# Example:
#   source_gene = ADGRB1
#   project_scoring_gene = BAI1
#
# This is NOT fuzzy mapping and NOT result-driven remapping.
#
# NO model probabilities or ranks are accessed here.
# ==========================================================================================

from pathlib import Path
import pandas as pd
import numpy as np
import hashlib
import json
import math
import re


# ==========================================================================================
# 0. PATHS
# ==========================================================================================

ROOT = Path(r"E:\ASD_ContextFusion_Q1")

SOURCE_EVIDENCE = (
    ROOT /
    "02_interim" /
    "external_validation_source_lock" /
    "Stage9B1F_Tan245_complete_source_evidence_LOCKED.tsv.gz"
)

IDENTITY_AUDIT = (
    ROOT /
    "02_interim" /
    "external_validation_source_lock" /
    "Stage9B2_Tan10_absent_gene_identity_audit.tsv"
)

MASTER_FILE = (
    ROOT /
    "03_features" /
    "multimodal_master" /
    "dhindsa_NTv2_row_preserved_master.tsv"
)

SEED_FILE = (
    ROOT /
    "00_admin" /
    "dhindsa_2025_monoallelic_ASD_190.tsv"
)

STAGE9B1F_LOCK = (
    ROOT /
    "00_admin" /
    "ASD_ContextFusion_STAGE9B1F_EXTERNAL_GENE_SET_FINAL_LOCK.json"
)

STAGE9B2_MANIFEST = (
    ROOT /
    "00_admin" /
    "ASD_ContextFusion_STAGE9B2_ABSENT_GENE_IDENTITY_AUDIT.json"
)

OUT_DIR = (
    ROOT /
    "02_interim" /
    "external_validation_source_lock"
)

ADMIN_DIR = ROOT / "00_admin"

OUT_DIR.mkdir(
    parents=True,
    exist_ok=True
)


EXPECTED_SOURCE = 245
EXPECTED_PROJECT_UNIVERSE = 18252
EXPECTED_SEEDS = 190

EXPECTED_EXACT_SYMBOL = 235
EXPECTED_EXACT_ALIAS = 7
EXPECTED_UNMATCHED = 3

NONSEED_UNIVERSE_EXPECTED = 18062

BOOTSTRAP_SEED = 20260830
BOOTSTRAP_REPS = 5000


# ==========================================================================================
# 1. HELPERS
# ==========================================================================================

def sha256_file(
    path,
    chunk_size=1024 * 1024
):

    h = hashlib.sha256()

    with open(path, "rb") as f:

        while True:

            chunk = f.read(chunk_size)

            if not chunk:
                break

            h.update(chunk)

    return h.hexdigest()


def norm_group(x):

    if pd.isna(x):
        return ""

    x = str(x).strip().lower()

    x = (
        x.replace("–", "-")
         .replace("—", "-")
         .replace("_", " ")
    )

    x = re.sub(
        r"\s+",
        " ",
        x
    )

    return x.strip()


def identify_seed_column(df):

    candidates = [
        "Gene_Name",
        "Gene",
        "gene",
        "Genesymbol",
        "GeneSymbol",
        "gene_symbol",
        "symbol"
    ]

    for col in candidates:

        if col in df.columns:
            return col

    if df.shape[1] == 1:
        return df.columns[0]

    raise RuntimeError(
        "Unable to identify seed gene column."
    )


# ==========================================================================================
# 2. INPUT AUDIT
# ==========================================================================================

print("=" * 140)
print("ASD-CONTEXTFUSION — STAGE 9B.3 FINAL IDENTITY RECONCILIATION")
print("=" * 140)

print("\n[1/9] INPUT / LOCK AUDIT")


for path in [
    SOURCE_EVIDENCE,
    IDENTITY_AUDIT,
    MASTER_FILE,
    SEED_FILE,
    STAGE9B1F_LOCK,
    STAGE9B2_MANIFEST
]:

    assert path.exists(), path

    print(
        "[OK]",
        path
    )


with open(
    STAGE9B1F_LOCK,
    "r",
    encoding="utf-8"
) as f:

    lock9b1f = json.load(f)


with open(
    STAGE9B2_MANIFEST,
    "r",
    encoding="utf-8"
) as f:

    lock9b2 = json.load(f)


assert lock9b1f[
    "model_scores_accessed_before_lock"
] is False

assert lock9b1f[
    "model_ranks_accessed_before_lock"
] is False

assert lock9b2[
    "model_scores_accessed"
] is False

assert lock9b2[
    "model_ranks_accessed"
] is False


# ==========================================================================================
# 3. LOAD SOURCE / PROJECT UNIVERSE
# ==========================================================================================

print("\n[2/9] LOADING SOURCE + PROJECT IDENTITIES")


risk = pd.read_csv(
    SOURCE_EVIDENCE,
    sep="\t",
    compression="gzip"
)


audit = pd.read_csv(
    IDENTITY_AUDIT,
    sep="\t"
)


master = pd.read_csv(
    MASTER_FILE,
    sep="\t"
)


seed_df = pd.read_csv(
    SEED_FILE,
    sep="\t"
)


assert len(risk) == EXPECTED_SOURCE
assert risk["Gene_Name"].nunique() == EXPECTED_SOURCE

assert master["Gene_Name"].nunique() == EXPECTED_PROJECT_UNIVERSE


project_genes = set(
    master[
        "Gene_Name"
    ]
    .astype(str)
    .str.strip()
)


master_seeds = set(
    master.loc[
        master["known_gene"] == 1,
        "Gene_Name"
    ]
    .astype(str)
    .str.strip()
)


assert len(project_genes) == EXPECTED_PROJECT_UNIVERSE
assert len(master_seeds) == EXPECTED_SEEDS


seed_col = identify_seed_column(
    seed_df
)


file_seeds = set(
    seed_df[
        seed_col
    ]
    .astype(str)
    .str.strip()
)


assert file_seeds == master_seeds


nonseed_universe = (
    project_genes
    -
    master_seeds
)


assert len(
    nonseed_universe
) == NONSEED_UNIVERSE_EXPECTED


print("Project universe :", len(project_genes))
print("ASD seeds        :", len(master_seeds))
print("Non-seed universe:", len(nonseed_universe))


# ==========================================================================================
# 4. BUILD 245 SOURCE -> PROJECT CROSSWALK
# ==========================================================================================

print("\n[3/9] BUILDING SOURCE -> PROJECT IDENTITY CROSSWALK")


audit_lookup = {
    str(row["source_gene"]).strip():
        row
    for _, row
    in audit.iterrows()
}


crosswalk_rows = []


for _, row in risk.iterrows():

    source_gene = str(
        row[
            "Gene_Name"
        ]
    ).strip()


    if source_gene in project_genes:

        mapping_mode = (
            "EXACT_SYMBOL"
        )

        project_gene = (
            source_gene
        )

        identity_evidence = (
            "Exact project Gene_Name"
        )


    else:

        if source_gene not in audit_lookup:

            raise RuntimeError(
                f"Absent source gene missing from Stage9B2 audit: "
                f"{source_gene}"
            )


        a = audit_lookup[
            source_gene
        ]


        if (
            a[
                "identity_status"
            ]
            ==
            "EXACT_PROJECT_IDENTITY_MATCH"
        ):

            project_gene = str(
                a[
                    "project_gene_candidate"
                ]
            ).strip()


            assert project_gene in project_genes


            mapping_mode = (
                "EXACT_STABLE_ID_ALIAS"
            )


            identity_evidence = str(
                a[
                    "metadata_evidence"
                ]
            )


        elif (
            a[
                "identity_status"
            ]
            ==
            "NO_PROJECT_IDENTITY_MATCH"
        ):

            project_gene = ""

            mapping_mode = (
                "OUT_OF_PROJECT_UNIVERSE"
            )

            identity_evidence = (
                "No exact project metadata/stable-ID match"
            )


        else:

            raise RuntimeError(
                f"Ambiguous mapping status for {source_gene}: "
                f"{a['identity_status']}"
            )


    source_group = norm_group(
        row[
            "dominant_group_source"
        ]
    )


    if source_group == "well-known":

        set_group = (
            "well-known"
        )


    elif source_group == "suggestive":

        set_group = (
            "suggestive"
        )


    elif source_group in [
        "no evidence",
        "no-evidence"
    ]:

        set_group = (
            "no-evidence"
        )


    else:

        raise RuntimeError(
            f"Unexpected source category for {source_gene}: "
            f"{source_group}"
        )


    is_seed = (
        project_gene in master_seeds
        if project_gene
        else False
    )


    crosswalk_rows.append(
        {
            "source_gene":
                source_gene,

            "source_group":
                set_group,

            "project_scoring_gene":
                project_gene,

            "mapping_mode":
                mapping_mode,

            "identity_evidence":
                identity_evidence,

            "project_gene_is_locked_ASD_seed":
                bool(
                    is_seed
                ),

            "eligible_nonseed_external_validation":
                bool(
                    project_gene
                    and
                    project_gene
                    in nonseed_universe
                )
        }
    )


crosswalk = pd.DataFrame(
    crosswalk_rows
)


assert len(crosswalk) == 245
assert crosswalk["source_gene"].nunique() == 245


mapping_counts = (
    crosswalk[
        "mapping_mode"
    ]
    .value_counts()
)


print(
    mapping_counts.to_string()
)


assert mapping_counts.get(
    "EXACT_SYMBOL",
    0
) == EXPECTED_EXACT_SYMBOL


assert mapping_counts.get(
    "EXACT_STABLE_ID_ALIAS",
    0
) == EXPECTED_EXACT_ALIAS


assert mapping_counts.get(
    "OUT_OF_PROJECT_UNIVERSE",
    0
) == EXPECTED_UNMATCHED


# ==========================================================================================
# 5. COLLISION AUDIT
# ==========================================================================================

print("\n[4/9] PROJECT-IDENTITY COLLISION AUDIT")


represented = crosswalk.loc[
    crosswalk[
        "project_scoring_gene"
    ].ne("")
].copy()


duplicates = (
    represented[
        "project_scoring_gene"
    ]
    .value_counts()
)


collisions = duplicates[
    duplicates > 1
]


if len(
    collisions
) > 0:

    print(
        collisions.to_string()
    )

    raise RuntimeError(
        "\nTwo or more Tan source genes map to the same project gene. "
        "Manual source-provenance review required BEFORE Stage 9C."
    )


print(
    "No source->project identity collisions [OK]"
)


assert represented[
    "project_scoring_gene"
].nunique() == len(
    represented
)


# ==========================================================================================
# 6. EXACT ALIAS + OUT-OF-UNIVERSE REPORT
# ==========================================================================================

print("\n[5/9] EXACT ALIASES / GENUINELY OUT-OF-UNIVERSE")


alias_rows = crosswalk.loc[
    crosswalk[
        "mapping_mode"
    ]
    ==
    "EXACT_STABLE_ID_ALIAS",
    [
        "source_gene",
        "source_group",
        "project_scoring_gene",
        "project_gene_is_locked_ASD_seed",
        "eligible_nonseed_external_validation"
    ]
].copy()


absent_rows = crosswalk.loc[
    crosswalk[
        "mapping_mode"
    ]
    ==
    "OUT_OF_PROJECT_UNIVERSE",
    [
        "source_gene",
        "source_group"
    ]
].copy()


print(
    "\n7 EXACT STABLE-ID ALIASES"
)

print(
    alias_rows.to_string(
        index=False
    )
)


print(
    "\n3 GENUINELY OUT-OF-UNIVERSE GENES"
)

print(
    absent_rows.to_string(
        index=False
    )
)


# ==========================================================================================
# 7. FINAL SET COUNTS AFTER IDENTITY RECONCILIATION
# ==========================================================================================

print("\n[6/9] FINAL EXTERNAL VALIDATION COUNTS")


SET_DEFINITIONS = {

    "Tan_All245":
        crosswalk.index == crosswalk.index,

    "Tan_SupplementSuggestive46":
        (
            crosswalk[
                "source_group"
            ]
            ==
            "suggestive"
        ),

    "Tan_SupplementNoEvidence32":
        (
            crosswalk[
                "source_group"
            ]
            ==
            "no-evidence"
        ),

    "Tan_SupplementSuggestiveOrNoEvidence78":
        (
            crosswalk[
                "source_group"
            ]
            .isin(
                [
                    "suggestive",
                    "no-evidence"
                ]
            )
        )
}


summary_rows = []


for set_name, mask in SET_DEFINITIONS.items():

    sub = crosswalk.loc[
        mask
    ].copy()


    represented_sub = sub.loc[
        sub[
            "project_scoring_gene"
        ].ne("")
    ]


    seed_sub = represented_sub.loc[
        represented_sub[
            "project_gene_is_locked_ASD_seed"
        ]
    ]


    nonseed_sub = represented_sub.loc[
        represented_sub[
            "eligible_nonseed_external_validation"
        ]
    ]


    exact_alias_sub = sub.loc[
        sub[
            "mapping_mode"
        ]
        ==
        "EXACT_STABLE_ID_ALIAS"
    ]


    absent_sub = sub.loc[
        sub[
            "mapping_mode"
        ]
        ==
        "OUT_OF_PROJECT_UNIVERSE"
    ]


    summary_rows.append(
        {
            "set":
                set_name,

            "source_count":
                len(
                    sub
                ),

            "exact_symbol_matches":
                int(
                    (
                        sub[
                            "mapping_mode"
                        ]
                        ==
                        "EXACT_SYMBOL"
                    ).sum()
                ),

            "exact_stable_id_aliases":
                len(
                    exact_alias_sub
                ),

            "genuinely_out_of_universe":
                len(
                    absent_sub
                ),

            "project_representable_source_genes":
                len(
                    represented_sub
                ),

            "unique_project_scoring_genes":
                represented_sub[
                    "project_scoring_gene"
                ].nunique(),

            "locked_seed_overlap":
                len(
                    seed_sub
                ),

            "final_nonseed_external_validation":
                len(
                    nonseed_sub
                )
        }
    )


summary = pd.DataFrame(
    summary_rows
)


print(
    summary.to_string(
        index=False
    )
)


# ==========================================================================================
# 8. WRITE FINAL RECONCILED FILES
# ==========================================================================================

print("\n[7/9] WRITING FINAL IDENTITY-RECONCILED LOCK FILES")


CROSSWALK_FILE = (
    OUT_DIR /
    "Stage9B3_Tan245_source_to_project_identity_crosswalk_LOCKED.tsv"
)


ALIAS_FILE = (
    OUT_DIR /
    "Stage9B3_exact_stable_ID_aliases_LOCKED.tsv"
)


ABSENT_FILE = (
    OUT_DIR /
    "Stage9B3_genuinely_out_of_project_universe_LOCKED.tsv"
)


SUMMARY_FILE = (
    OUT_DIR /
    "Stage9B3_external_validation_identity_reconciled_summary.tsv"
)


crosswalk.to_csv(
    CROSSWALK_FILE,
    sep="\t",
    index=False
)


alias_rows.to_csv(
    ALIAS_FILE,
    sep="\t",
    index=False
)


absent_rows.to_csv(
    ABSENT_FILE,
    sep="\t",
    index=False
)


summary.to_csv(
    SUMMARY_FILE,
    sep="\t",
    index=False
)


# ----------------------------------------------------------------------
# Write scoring-ready files preserving BOTH source and project identities.
# ----------------------------------------------------------------------

scoring_file_records = []


for set_name, mask in SET_DEFINITIONS.items():

    sub = crosswalk.loc[
        mask
        &
        crosswalk[
            "eligible_nonseed_external_validation"
        ]
    ].copy()


    sub = sub[
        [
            "source_gene",
            "source_group",
            "project_scoring_gene",
            "mapping_mode",
            "project_gene_is_locked_ASD_seed"
        ]
    ]


    assert not sub[
        "project_gene_is_locked_ASD_seed"
    ].any()


    assert sub[
        "project_scoring_gene"
    ].is_unique


    path = (
        OUT_DIR /
        f"Stage9B3_{set_name}_NONSEED_PROJECT_IDENTITIES_LOCKED.tsv"
    )


    sub.to_csv(
        path,
        sep="\t",
        index=False
    )


    scoring_file_records.append(
        {
            "set":
                set_name,

            "gene_count":
                len(
                    sub
                ),

            "path":
                str(
                    path
                ),

            "sha256":
                sha256_file(
                    path
                )
        }
    )


SCORING_HASH_FILE = (
    OUT_DIR /
    "Stage9B3_scoring_ready_external_sets_SHA256.tsv"
)


pd.DataFrame(
    scoring_file_records
).to_csv(
    SCORING_HASH_FILE,
    sep="\t",
    index=False
)


# ==========================================================================================
# 9. STAGE 9C PRESPECIFICATION V2
# ==========================================================================================

print("\n[8/9] FREEZING STAGE 9C PRESPECIFICATION V2")


N = len(
    nonseed_universe
)


TOP1 = math.ceil(
    N * 0.01
)

TOP5 = math.ceil(
    N * 0.05
)

TOP10 = math.ceil(
    N * 0.10
)


assert N == 18062
assert TOP1 == 181
assert TOP5 == 904
assert TOP10 == 1807


PLAN_V2_FILE = (
    ADMIN_DIR /
    "ASD_ContextFusion_STAGE9C_EXTERNAL_VALIDATION_PRESPECIFICATION_V2.json"
)


plan_v2 = {

    "project":
        "ASD-ContextFusion",

    "stage":
        "9C_external_validation_prespecification_V2",

    "status":
        "LOCKED_BEFORE_ANY_MODEL_SCORE_OR_RANK_ACCESS",

    "supersedes":
        (
            "ASD_ContextFusion_STAGE9C_EXTERNAL_VALIDATION_"
            "PRESPECIFICATION.json"
        ),

    "reason_for_V2":
        (
            "Pre-result methodological cleanup after exact external-gene "
            "identity audit: scoring background is restricted to the "
            "same non-seed candidate population used for independent "
            "external validation, and exact stable-ID aliases are "
            "reconciled before rank access."
        ),

    "scientific_source_gene_sets_changed":
        False,

    "identity_mapping_rule":
        (
            "Exact project symbol when available; otherwise only exact "
            "stable-ID-supported project identity from Stage 9B.2. "
            "No fuzzy matching."
        ),

    "ranking_universe":
        (
            "Frozen project genes excluding all 190 locked ASD seeds."
        ),

    "ranking_universe_count":
        N,

    "why_training_seeds_removed_from_background":
        (
            "External-validation question concerns prioritization among "
            "genes not used as locked ASD positives. Training seeds are "
            "therefore excluded from both evaluation targets and the "
            "candidate ranking universe."
        ),

    "models":
        [
            "Context",
            "NTv2",
            "StrictFusion"
        ],

    "primary_model_comparison":
        "StrictFusion_minus_Context",

    "primary_external_set":
        "Tan_All245_nonseed_identity_reconciled",

    "key_novelty_focused_set":
        "Tan_SupplementNoEvidence32_nonseed_identity_reconciled",

    "secondary_sets":
        [
            "Tan_SupplementSuggestive46_nonseed_identity_reconciled",
            (
                "Tan_SupplementSuggestiveOrNoEvidence78_"
                "nonseed_identity_reconciled"
            )
        ],

    "paper_45_limited_prior_list_used":
        False,

    "rank_order":
        (
            "Descending model probability; deterministic Gene_Name "
            "ascending tie-breaker."
        ),

    "rank_percentile_formula":
        (
            "100 * (N - ordinal_rank) / (N - 1), where rank 1 is best; "
            "thus best=100 and worst=0."
        ),

    "top_percent_rule":
        (
            "Exact first K genes under the prespecified deterministic "
            "total ordering."
        ),

    "top_thresholds":
        {
            "top_1_percent_K":
                TOP1,

            "top_5_percent_K":
                TOP5,

            "top_10_percent_K":
                TOP10
        },

    "primary_continuous_statistic":
        (
            "Mean paired external-gene rank-percentile difference: "
            "StrictFusion - Context."
        ),

    "additional_continuous_summaries":
        [
            "median rank percentile",
            "mean rank percentile",
            "bootstrap 95% CI"
        ],

    "paired_bootstrap":
        {
            "replicates":
                BOOTSTRAP_REPS,

            "seed":
                BOOTSTRAP_SEED,

            "resampling_unit":
                "external gene",

            "same_gene_resample_across_models":
                True
        },

    "top_k_outputs":
        [
            "recovered_gene_count",
            "recovery_fraction",
            "enrichment_fold",
            "hypergeometric_p"
        ],

    "enrichment_background":
        N,

    "enrichment_multiple_testing":
        (
            "BH FDR across all model × external-set × top-threshold "
            "hypergeometric enrichment tests reported in Stage 9C."
        ),

    "result_based_threshold_selection":
        False,

    "model_tuning":
        False,

    "new_PU_sampling":
        False,

    "model_scores_accessed_before_this_lock":
        False,

    "model_ranks_accessed_before_this_lock":
        False
}


with open(
    PLAN_V2_FILE,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        plan_v2,
        f,
        indent=2
    )


# ==========================================================================================
# 10. FINAL STAGE 9B.3 LOCK
# ==========================================================================================

FINAL_LOCK_FILE = (
    ADMIN_DIR /
    "ASD_ContextFusion_STAGE9B3_IDENTITY_RECONCILIATION_FINAL_LOCK.json"
)


final_lock = {

    "project":
        "ASD-ContextFusion",

    "stage":
        "9B.3_identity_reconciliation_final_lock",

    "status":
        "FINAL_BEFORE_MODEL_RANK_ACCESS",

    "source_gene_count":
        245,

    "exact_symbol_matches":
        235,

    "exact_stable_ID_aliases":
        7,

    "genuinely_out_of_universe":
        3,

    "project_universe":
        18252,

    "locked_ASD_seeds":
        190,

    "nonseed_ranking_universe":
        18062,

    "fuzzy_matching":
        False,

    "manual_alias_guessing":
        False,

    "result_driven_mapping":
        False,

    "model_scores_accessed":
        False,

    "model_ranks_accessed":
        False,

    "crosswalk_file":
        str(
            CROSSWALK_FILE
        ),

    "crosswalk_sha256":
        sha256_file(
            CROSSWALK_FILE
        ),

    "summary":
        summary.to_dict(
            orient="records"
        ),

    "stage9C_prespecification_V2":
        str(
            PLAN_V2_FILE
        ),

    "stage9C_prespecification_V2_sha256":
        sha256_file(
            PLAN_V2_FILE
        )
}


with open(
    FINAL_LOCK_FILE,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        final_lock,
        f,
        indent=2
    )


# ==========================================================================================
# 11. FINAL REPORT
# ==========================================================================================

print("\n[9/9] FINAL REPORT")

print("\n" + "=" * 140)
print("STAGE 9B.3 FINAL IDENTITY RECONCILIATION COMPLETE [OK]")
print("=" * 140)


print("\nMAPPING SUMMARY")
print(
    mapping_counts.to_string()
)


print("\nEXACT STABLE-ID ALIASES")
print(
    alias_rows.to_string(
        index=False
    )
)


print("\nGENUINELY OUT-OF-UNIVERSE")
print(
    absent_rows.to_string(
        index=False
    )
)


print("\nFINAL EXTERNAL SET SUMMARY")
print(
    summary.to_string(
        index=False
    )
)


print("\nSTAGE 9C BACKGROUND")
print(
    "Non-seed ranking universe:",
    N
)

print(
    "Top 1% K :",
    TOP1
)

print(
    "Top 5% K :",
    TOP5
)

print(
    "Top 10% K:",
    TOP10
)


print("\nINTEGRITY")
print("Source external gene definitions changed : NO")
print("Exact stable-ID mappings                : 7")
print("Fuzzy mappings                          : 0")
print("Genuinely absent                        : 3")
print("Model scores accessed                   : NO")
print("Model ranks accessed                    : NO")
print("Result-driven exclusions                : NO")
print("Stage 9C V2 prespecified                : YES")


print("\nFINAL IDENTITY LOCK:")
print(
    FINAL_LOCK_FILE
)


print("\nSTAGE 9C V2 PLAN:")
print(
    PLAN_V2_FILE
)


print("\nNEXT:")
print(
    "Stage 9C — FIRST model-score/rank access."
)

print("=" * 140)
```

```python
# ==========================================================================================
# ASD-ContextFusion
# STAGE 9C — PRESPECIFIED TEMPORAL / EXTERNAL VALIDATION
#
# FIRST MODEL-SCORE / MODEL-RANK ACCESS
#
# LOCKED BEFORE THIS STAGE
# ------------------------
# Ranking universe:
#     18,062 project genes after excluding all 190 locked ASD seeds
#
# External sets:
#     Broad non-seed Tan set                    = 134
#     Supplement-defined suggestive             = 45
#     Supplement-defined no-evidence            = 30
#     Suggestive OR no-evidence                 = 75
#
# Top-K:
#     top 1%  = 181
#     top 5%  = 904
#     top 10% = 1807
#
# Models:
#     Context
#     NT-v2
#     Strict Fusion
#
# PRIMARY MODEL COMPARISON:
#     Strict Fusion - Context
#
# PRIMARY CONTINUOUS STATISTIC:
#     paired mean external-gene rank-percentile difference
#
# BOOTSTRAP:
#     5,000 paired gene resamples
#     seed = 20260830
#
# IMPORTANT:
#     - NO model fitting
#     - NO tuning
#     - NO new PU sampling
#     - NO threshold selection
#     - NO gene-set changes
# ==========================================================================================

from pathlib import Path

import hashlib
import json
import math

import numpy as np
import pandas as pd

from scipy.stats import hypergeom


# ==========================================================================================
# 0. PATHS / CONSTANTS
# ==========================================================================================

ROOT = Path(r"E:\ASD_ContextFusion_Q1")

SCORE_FILE = (
    ROOT /
    "05_results" /
    "proposed" /
    "strict_context_NTv2_fusion" /
    "StrictFusion_gene_level_OOF_scores.tsv"
)

IDENTITY_LOCK = (
    ROOT /
    "00_admin" /
    "ASD_ContextFusion_STAGE9B3_IDENTITY_RECONCILIATION_FINAL_LOCK.json"
)

PLAN_FILE = (
    ROOT /
    "00_admin" /
    "ASD_ContextFusion_STAGE9C_EXTERNAL_VALIDATION_PRESPECIFICATION_V2.json"
)

SCORING_HASH_FILE = (
    ROOT /
    "02_interim" /
    "external_validation_source_lock" /
    "Stage9B3_scoring_ready_external_sets_SHA256.tsv"
)

EXT_DIR = (
    ROOT /
    "02_interim" /
    "external_validation_source_lock"
)

OUT_DIR = (
    ROOT /
    "05_results" /
    "proposed" /
    "stage9C_external_validation"
)

OUT_DIR.mkdir(
    parents=True,
    exist_ok=True
)


N_TOTAL = 18252
N_SEEDS = 190
N_NONSEED = 18062

TOP_K = {
    "Top1pct": 181,
    "Top5pct": 904,
    "Top10pct": 1807
}

BOOTSTRAP_REPS = 5000
BOOTSTRAP_SEED = 20260830

ANALYSIS_VERSION = (
    "ASD_CF_STAGE9C_EXTERNAL_VALIDATION_V1"
)


SET_CONFIG = {

    "Tan_All245_nonseed":
        {
            "path":
                EXT_DIR /
                (
                    "Stage9B3_Tan_All245_"
                    "NONSEED_PROJECT_IDENTITIES_LOCKED.tsv"
                ),

            "expected":
                134,

            "role":
                "PRIMARY_BROAD_EXTERNAL"
        },


    "Tan_Suggestive46_nonseed":
        {
            "path":
                EXT_DIR /
                (
                    "Stage9B3_Tan_SupplementSuggestive46_"
                    "NONSEED_PROJECT_IDENTITIES_LOCKED.tsv"
                ),

            "expected":
                45,

            "role":
                "SECONDARY_SUGGESTIVE"
        },


    "Tan_NoEvidence32_nonseed":
        {
            "path":
                EXT_DIR /
                (
                    "Stage9B3_Tan_SupplementNoEvidence32_"
                    "NONSEED_PROJECT_IDENTITIES_LOCKED.tsv"
                ),

            "expected":
                30,

            "role":
                "PRIMARY_NOVELTY_FOCUSED"
        },


    "Tan_SuggestiveOrNoEvidence78_nonseed":
        {
            "path":
                EXT_DIR /
                (
                    "Stage9B3_"
                    "Tan_SupplementSuggestiveOrNoEvidence78_"
                    "NONSEED_PROJECT_IDENTITIES_LOCKED.tsv"
                ),

            "expected":
                75,

            "role":
                "SECONDARY_LOW_EVIDENCE_COMBINED"
        }
}


MODEL_COLUMNS = {
    "Context":
        "Context_probability",

    "NTv2":
        "NT_probability",

    "StrictFusion":
        "Fusion_probability"
}


# ==========================================================================================
# 1. HELPERS
# ==========================================================================================

def sha256_file(
    path,
    chunk_size=1024 * 1024
):

    h = hashlib.sha256()

    with open(
        path,
        "rb"
    ) as f:

        while True:

            chunk = f.read(
                chunk_size
            )

            if not chunk:
                break

            h.update(
                chunk
            )

    return h.hexdigest()


def bh_adjust(
    pvalues
):

    p = np.asarray(
        pvalues,
        dtype=float
    )

    m = len(
        p
    )

    order = np.argsort(
        p
    )

    ranked = p[
        order
    ]


    adjusted = (
        ranked *
        m /
        np.arange(
            1,
            m + 1
        )
    )


    adjusted = np.minimum.accumulate(
        adjusted[
            ::-1
        ]
    )[
        ::-1
    ]


    adjusted = np.minimum(
        adjusted,
        1.0
    )


    result = np.empty_like(
        adjusted
    )

    result[
        order
    ] = adjusted


    return result


# ==========================================================================================
# 2. PRE-RUN LOCK AUDIT
# ==========================================================================================

print(
    "=" * 140
)

print(
    "ASD-CONTEXTFUSION — STAGE 9C EXTERNAL VALIDATION"
)

print(
    "=" * 140
)


print(
    "\n[1/10] PRE-RUN SCIENTIFIC LOCK AUDIT"
)


for path in [
    SCORE_FILE,
    IDENTITY_LOCK,
    PLAN_FILE,
    SCORING_HASH_FILE
]:

    assert path.exists(), (
        f"Missing required file:\n{path}"
    )

    print(
        "[OK]",
        path
    )


with open(
    IDENTITY_LOCK,
    "r",
    encoding="utf-8"
) as f:

    identity_lock = json.load(
        f
    )


with open(
    PLAN_FILE,
    "r",
    encoding="utf-8"
) as f:

    plan = json.load(
        f
    )


assert identity_lock[
    "model_scores_accessed"
] is False

assert identity_lock[
    "model_ranks_accessed"
] is False

assert identity_lock[
    "exact_symbol_matches"
] == 235

assert identity_lock[
    "exact_stable_ID_aliases"
] == 7

assert identity_lock[
    "genuinely_out_of_universe"
] == 3

assert identity_lock[
    "nonseed_ranking_universe"
] == 18062


assert plan[
    "ranking_universe_count"
] == 18062

assert plan[
    "top_thresholds"
][
    "top_1_percent_K"
] == 181

assert plan[
    "top_thresholds"
][
    "top_5_percent_K"
] == 904

assert plan[
    "top_thresholds"
][
    "top_10_percent_K"
] == 1807


assert (
    sha256_file(
        PLAN_FILE
    )
    ==
    identity_lock[
        "stage9C_prespecification_V2_sha256"
    ]
)


print(
    "\nStage 9C V2 prespecification hash: VERIFIED [OK]"
)


# ==========================================================================================
# 3. VERIFY EXTERNAL SCORING SET HASHES
# ==========================================================================================

print(
    "\n[2/10] VERIFYING LOCKED EXTERNAL SCORING SETS"
)


hash_table = pd.read_csv(
    SCORING_HASH_FILE,
    sep="\t"
)


external_sets = {}


for set_name, config in SET_CONFIG.items():

    path = config[
        "path"
    ]

    assert path.exists(), path


    ext = pd.read_csv(
        path,
        sep="\t"
    )


    assert len(
        ext
    ) == config[
        "expected"
    ]


    assert ext[
        "project_scoring_gene"
    ].is_unique


    assert not ext[
        "project_gene_is_locked_ASD_seed"
    ].astype(
        bool
    ).any()


    hash_row = hash_table.loc[
        hash_table[
            "set"
        ].eq(
            set_name.replace(
                "_nonseed",
                ""
            )
        )
    ]


    # The set names in the hash table are the original Stage 9B.3 names.
    if len(
        hash_row
    ) == 0:

        expected_source_name = {

            "Tan_All245_nonseed":
                "Tan_All245",

            "Tan_Suggestive46_nonseed":
                "Tan_SupplementSuggestive46",

            "Tan_NoEvidence32_nonseed":
                "Tan_SupplementNoEvidence32",

            "Tan_SuggestiveOrNoEvidence78_nonseed":
                "Tan_SupplementSuggestiveOrNoEvidence78"

        }[
            set_name
        ]


        hash_row = hash_table.loc[
            hash_table[
                "set"
            ].eq(
                expected_source_name
            )
        ]


    assert len(
        hash_row
    ) == 1


    expected_hash = str(
        hash_row.iloc[
            0
        ][
            "sha256"
        ]
    )


    actual_hash = sha256_file(
        path
    )


    assert actual_hash == expected_hash


    external_sets[
        set_name
    ] = ext


    print(
        f"{set_name:42s}",
        len(
            ext
        ),
        "genes [HASH OK]"
    )


# ==========================================================================================
# 4. FIRST MODEL-SCORE ACCESS
# ==========================================================================================

print(
    "\n[3/10] FIRST MODEL-SCORE ACCESS"
)


scores = pd.read_csv(
    SCORE_FILE,
    sep="\t"
)


required_score_columns = {
    "Gene_Name",
    "known_gene",
    "Context_probability",
    "NT_probability",
    "Fusion_probability"
}


assert required_score_columns.issubset(
    set(
        scores.columns
    )
)


assert len(
    scores
) == N_TOTAL


assert scores[
    "Gene_Name"
].nunique() == N_TOTAL


scores[
    "known_gene"
] = pd.to_numeric(
    scores[
        "known_gene"
    ],
    errors="raise"
).astype(
    int
)


assert int(
    scores[
        "known_gene"
    ].sum()
) == N_SEEDS


for model, column in MODEL_COLUMNS.items():

    x = pd.to_numeric(
        scores[
            column
        ],
        errors="raise"
    )


    assert np.isfinite(
        x
    ).all()


    assert (
        (
            x >= 0
        )
        &
        (
            x <= 1
        )
    ).all()


    scores[
        column
    ] = x


print(
    "Genes:",
    len(
        scores
    )
)

print(
    "Locked ASD seeds:",
    int(
        scores[
            "known_gene"
        ].sum()
    )
)


# ==========================================================================================
# 5. CREATE LOCKED 18,062-GENE NON-SEED RANKING UNIVERSE
# ==========================================================================================

print(
    "\n[4/10] BUILDING PRESPECIFIED NON-SEED RANKING UNIVERSE"
)


nonseed_scores = (
    scores.loc[
        scores[
            "known_gene"
        ] == 0
    ]
    .copy()
)


assert len(
    nonseed_scores
) == N_NONSEED


assert nonseed_scores[
    "Gene_Name"
].nunique() == N_NONSEED


rank_table = nonseed_scores[
    [
        "Gene_Name",
        "Context_probability",
        "NT_probability",
        "Fusion_probability"
    ]
].copy()


for model, column in MODEL_COLUMNS.items():

    ordered = (
        rank_table[
            [
                "Gene_Name",
                column
            ]
        ]
        .sort_values(
            by=[
                column,
                "Gene_Name"
            ],
            ascending=[
                False,
                True
            ],
            kind="mergesort"
        )
        .reset_index(
            drop=True
        )
    )


    ordered[
        f"{model}_rank"
    ] = np.arange(
        1,
        N_NONSEED + 1,
        dtype=int
    )


    ordered[
        f"{model}_rank_percentile"
    ] = (
        100.0
        *
        (
            N_NONSEED
            -
            ordered[
                f"{model}_rank"
            ]
        )
        /
        (
            N_NONSEED
            -
            1
        )
    )


    mapping_rank = dict(
        zip(
            ordered[
                "Gene_Name"
            ],
            ordered[
                f"{model}_rank"
            ]
        )
    )


    mapping_pct = dict(
        zip(
            ordered[
                "Gene_Name"
            ],
            ordered[
                f"{model}_rank_percentile"
            ]
        )
    )


    rank_table[
        f"{model}_rank"
    ] = (
        rank_table[
            "Gene_Name"
        ].map(
            mapping_rank
        )
    )


    rank_table[
        f"{model}_rank_percentile"
    ] = (
        rank_table[
            "Gene_Name"
        ].map(
            mapping_pct
        )
    )


    assert rank_table[
        f"{model}_rank"
    ].notna().all()


    assert sorted(
        rank_table[
            f"{model}_rank"
        ].astype(
            int
        ).tolist()
    ) == list(
        range(
            1,
            N_NONSEED + 1
        )
    )


    tied_values = int(
        rank_table[
            column
        ].duplicated(
            keep=False
        ).sum()
    )


    print(
        f"{model:12s} tied-score rows:",
        tied_values,
        "| deterministic Gene_Name tie-breaker applied"
    )


# ==========================================================================================
# 6. EXTERNAL GENE RANK DETAILS
# ==========================================================================================

print(
    "\n[5/10] EXTERNAL GENE RANK EVALUATION"
)


detail_rows = []


for set_name, ext in external_sets.items():

    merged = (
        ext
        .merge(
            rank_table,
            left_on="project_scoring_gene",
            right_on="Gene_Name",
            how="left",
            validate="one_to_one"
        )
    )


    assert len(
        merged
    ) == len(
        ext
    )


    assert merged[
        "Gene_Name"
    ].notna().all()


    merged[
        "external_set"
    ] = set_name


    merged[
        "Fusion_minus_Context_rank_percentile"
    ] = (
        merged[
            "StrictFusion_rank_percentile"
        ]
        -
        merged[
            "Context_rank_percentile"
        ]
    )


    merged[
        "NT_minus_Context_rank_percentile"
    ] = (
        merged[
            "NTv2_rank_percentile"
        ]
        -
        merged[
            "Context_rank_percentile"
        ]
    )


    detail_rows.append(
        merged
    )


external_details = pd.concat(
    detail_rows,
    ignore_index=True
)


# ==========================================================================================
# 7. CONTINUOUS RANK SUMMARY
# ==========================================================================================

rank_summary_rows = []


for set_name, ext in external_sets.items():

    d = external_details.loc[
        external_details[
            "external_set"
        ].eq(
            set_name
        )
    ]


    for model in [
        "Context",
        "NTv2",
        "StrictFusion"
    ]:

        pct = d[
            f"{model}_rank_percentile"
        ].to_numpy(
            dtype=float
        )


        ranks = d[
            f"{model}_rank"
        ].to_numpy(
            dtype=int
        )


        rank_summary_rows.append(
            {
                "external_set":
                    set_name,

                "role":
                    SET_CONFIG[
                        set_name
                    ][
                        "role"
                    ],

                "n_external_genes":
                    len(
                        d
                    ),

                "model":
                    model,

                "mean_rank_percentile":
                    float(
                        np.mean(
                            pct
                        )
                    ),

                "median_rank_percentile":
                    float(
                        np.median(
                            pct
                        )
                    ),

                "Q1_rank_percentile":
                    float(
                        np.quantile(
                            pct,
                            0.25
                        )
                    ),

                "Q3_rank_percentile":
                    float(
                        np.quantile(
                            pct,
                            0.75
                        )
                    ),

                "mean_ordinal_rank":
                    float(
                        np.mean(
                            ranks
                        )
                    ),

                "median_ordinal_rank":
                    float(
                        np.median(
                            ranks
                        )
                    )
            }
        )


rank_summary = pd.DataFrame(
    rank_summary_rows
)


# ==========================================================================================
# 8. TOP-K RECOVERY + HYPERGEOMETRIC ENRICHMENT
# ==========================================================================================

print(
    "\n[6/10] TOP-K RECOVERY / ENRICHMENT"
)


topk_rows = []


for set_name, ext in external_sets.items():

    d = external_details.loc[
        external_details[
            "external_set"
        ].eq(
            set_name
        )
    ]


    M = len(
        d
    )


    for model in [
        "Context",
        "NTv2",
        "StrictFusion"
    ]:

        for threshold_name, K in TOP_K.items():

            ranks = d[
                f"{model}_rank"
            ].to_numpy(
                dtype=int
            )


            recovered = int(
                np.sum(
                    ranks <= K
                )
            )


            recovery_fraction = (
                recovered
                /
                M
            )


            expected_random_overlap = (
                M
                *
                K
                /
                N_NONSEED
            )


            baseline_fraction = (
                K
                /
                N_NONSEED
            )


            enrichment_fold = (
                recovery_fraction
                /
                baseline_fraction
            )


            p_hyper = float(
                hypergeom.sf(
                    recovered - 1,
                    N_NONSEED,
                    M,
                    K
                )
            )


            topk_rows.append(
                {
                    "external_set":
                        set_name,

                    "role":
                        SET_CONFIG[
                            set_name
                        ][
                            "role"
                        ],

                    "model":
                        model,

                    "threshold":
                        threshold_name,

                    "K":
                        K,

                    "background_N":
                        N_NONSEED,

                    "external_set_M":
                        M,

                    "recovered_genes":
                        recovered,

                    "recovery_fraction":
                        recovery_fraction,

                    "expected_random_overlap":
                        expected_random_overlap,

                    "baseline_fraction":
                        baseline_fraction,

                    "enrichment_fold":
                        enrichment_fold,

                    "hypergeometric_p":
                        p_hyper
                }
            )


topk = pd.DataFrame(
    topk_rows
)


topk[
    "BH_FDR_q_across_all_prespecified_enrichment_tests"
] = bh_adjust(
    topk[
        "hypergeometric_p"
    ].to_numpy(
        dtype=float
    )
)


assert len(
    topk
) == (
    len(
        SET_CONFIG
    )
    *
    3
    *
    3
)


# ==========================================================================================
# 9. PAIRED BOOTSTRAP — FUSION vs CONTEXT
# ==========================================================================================

print(
    "\n[7/10] PAIRED EXTERNAL-GENE BOOTSTRAP: FUSION - CONTEXT"
)


bootstrap_rows = []

bootstrap_summary_rows = []


for set_index, set_name in enumerate(
    SET_CONFIG.keys()
):

    d = external_details.loc[
        external_details[
            "external_set"
        ].eq(
            set_name
        )
    ].copy()


    delta = d[
        "Fusion_minus_Context_rank_percentile"
    ].to_numpy(
        dtype=float
    )


    n = len(
        delta
    )


    observed_mean = float(
        np.mean(
            delta
        )
    )


    observed_median = float(
        np.median(
            delta
        )
    )


    boot_means = np.empty(
        BOOTSTRAP_REPS,
        dtype=float
    )


    for b in range(
        BOOTSTRAP_REPS
    ):

        rng = np.random.default_rng(
            np.random.SeedSequence(
                [
                    BOOTSTRAP_SEED,
                    set_index,
                    b
                ]
            )
        )


        idx = rng.integers(
            0,
            n,
            size=n
        )


        boot_mean = float(
            np.mean(
                delta[
                    idx
                ]
            )
        )


        boot_means[
            b
        ] = boot_mean


        bootstrap_rows.append(
            {
                "analysis_version":
                    ANALYSIS_VERSION,

                "external_set":
                    set_name,

                "replicate":
                    b,

                "mean_Fusion_minus_Context_rank_percentile":
                    boot_mean
            }
        )


    ci_low = float(
        np.quantile(
            boot_means,
            0.025
        )
    )


    ci_high = float(
        np.quantile(
            boot_means,
            0.975
        )
    )


    bootstrap_summary_rows.append(
        {
            "external_set":
                set_name,

            "role":
                SET_CONFIG[
                    set_name
                ][
                    "role"
                ],

            "n_external_genes":
                n,

            "observed_mean_Fusion_minus_Context_rank_percentile":
                observed_mean,

            "observed_median_Fusion_minus_Context_rank_percentile":
                observed_median,

            "bootstrap_mean":
                float(
                    np.mean(
                        boot_means
                    )
                ),

            "bootstrap_SD":
                float(
                    np.std(
                        boot_means,
                        ddof=1
                    )
                ),

            "CI95_low":
                ci_low,

            "CI95_high":
                ci_high,

            "bootstrap_fraction_delta_gt_0":
                float(
                    np.mean(
                        boot_means > 0
                    )
                ),

            "CI_excludes_zero":
                bool(
                    (
                        ci_low > 0
                    )
                    or
                    (
                        ci_high < 0
                    )
                ),

            "bootstrap_replicates":
                BOOTSTRAP_REPS,

            "bootstrap_seed":
                BOOTSTRAP_SEED
        }
    )


bootstrap_replicates = pd.DataFrame(
    bootstrap_rows
)


bootstrap_summary = pd.DataFrame(
    bootstrap_summary_rows
)


# ==========================================================================================
# 10. DESCRIPTIVE TOP-K FUSION vs CONTEXT DELTAS
# ==========================================================================================

recovery_comparison_rows = []


for set_name in SET_CONFIG.keys():

    for threshold_name in TOP_K.keys():

        c = topk.loc[
            (
                topk[
                    "external_set"
                ] == set_name
            )
            &
            (
                topk[
                    "threshold"
                ] == threshold_name
            )
            &
            (
                topk[
                    "model"
                ] == "Context"
            )
        ].iloc[
            0
        ]


        f = topk.loc[
            (
                topk[
                    "external_set"
                ] == set_name
            )
            &
            (
                topk[
                    "threshold"
                ] == threshold_name
            )
            &
            (
                topk[
                    "model"
                ] == "StrictFusion"
            )
        ].iloc[
            0
        ]


        recovery_comparison_rows.append(
            {
                "external_set":
                    set_name,

                "threshold":
                    threshold_name,

                "Context_recovered":
                    int(
                        c[
                            "recovered_genes"
                        ]
                    ),

                "Fusion_recovered":
                    int(
                        f[
                            "recovered_genes"
                        ]
                    ),

                "Fusion_minus_Context_recovered":
                    int(
                        f[
                            "recovered_genes"
                        ]
                        -
                        c[
                            "recovered_genes"
                        ]
                    ),

                "Context_recovery_fraction":
                    float(
                        c[
                            "recovery_fraction"
                        ]
                    ),

                "Fusion_recovery_fraction":
                    float(
                        f[
                            "recovery_fraction"
                        ]
                    ),

                "Fusion_minus_Context_recovery_fraction":
                    float(
                        f[
                            "recovery_fraction"
                        ]
                        -
                        c[
                            "recovery_fraction"
                        ]
                    )
            }
        )


recovery_comparison = pd.DataFrame(
    recovery_comparison_rows
)


# ==========================================================================================
# 11. SAVE RESULTS
# ==========================================================================================

print(
    "\n[8/10] SAVING STAGE 9C RESULTS"
)


RANK_UNIVERSE_FILE = (
    OUT_DIR /
    "Stage9C_nonseed_18062_model_ranks.tsv.gz"
)


DETAIL_FILE = (
    OUT_DIR /
    "Stage9C_external_gene_rank_details.tsv"
)


RANK_SUMMARY_FILE = (
    OUT_DIR /
    "Stage9C_external_set_rank_summary.tsv"
)


TOPK_FILE = (
    OUT_DIR /
    "Stage9C_topk_recovery_enrichment.tsv"
)


RECOVERY_COMPARISON_FILE = (
    OUT_DIR /
    "Stage9C_Fusion_vs_Context_topk_recovery_deltas.tsv"
)


BOOT_REPLICATE_FILE = (
    OUT_DIR /
    "Stage9C_paired_rank_bootstrap_replicates.tsv.gz"
)


BOOT_SUMMARY_FILE = (
    OUT_DIR /
    "Stage9C_paired_rank_bootstrap_summary.tsv"
)


rank_table.to_csv(
    RANK_UNIVERSE_FILE,
    sep="\t",
    index=False,
    compression="gzip"
)


external_details.to_csv(
    DETAIL_FILE,
    sep="\t",
    index=False
)


rank_summary.to_csv(
    RANK_SUMMARY_FILE,
    sep="\t",
    index=False
)


topk.to_csv(
    TOPK_FILE,
    sep="\t",
    index=False
)


recovery_comparison.to_csv(
    RECOVERY_COMPARISON_FILE,
    sep="\t",
    index=False
)


bootstrap_replicates.to_csv(
    BOOT_REPLICATE_FILE,
    sep="\t",
    index=False,
    compression="gzip"
)


bootstrap_summary.to_csv(
    BOOT_SUMMARY_FILE,
    sep="\t",
    index=False
)


# ==========================================================================================
# 12. MANIFEST
# ==========================================================================================

print(
    "\n[9/10] WRITING REPRODUCIBILITY MANIFEST"
)


result_files = [
    RANK_UNIVERSE_FILE,
    DETAIL_FILE,
    RANK_SUMMARY_FILE,
    TOPK_FILE,
    RECOVERY_COMPARISON_FILE,
    BOOT_REPLICATE_FILE,
    BOOT_SUMMARY_FILE
]


manifest = {

    "project":
        "ASD-ContextFusion",

    "stage":
        "9C_external_validation",

    "analysis_version":
        ANALYSIS_VERSION,

    "first_model_score_access_stage":
        "9C",

    "models":
        [
            "Context",
            "NTv2",
            "StrictFusion"
        ],

    "primary_comparison":
        "StrictFusion_minus_Context",

    "ranking_universe":
        "nonseed_project_genes",

    "ranking_universe_N":
        N_NONSEED,

    "topK":
        TOP_K,

    "external_sets":
        {
            key:
                {
                    "n":
                        config[
                            "expected"
                        ],

                    "role":
                        config[
                            "role"
                        ],

                    "file":
                        str(
                            config[
                                "path"
                            ]
                        ),

                    "sha256":
                        sha256_file(
                            config[
                                "path"
                            ]
                        )
                }
            for key, config
            in SET_CONFIG.items()
        },

    "rank_direction":
        (
            "Higher probability is better."
        ),

    "tie_break":
        (
            "Gene_Name ascending."
        ),

    "rank_percentile":
        (
            "100*(N-rank)/(N-1), "
            "best=100 and worst=0."
        ),

    "paired_bootstrap_replicates":
        BOOTSTRAP_REPS,

    "paired_bootstrap_seed":
        BOOTSTRAP_SEED,

    "bootstrap_fraction_gt_zero_is_pvalue":
        False,

    "enrichment_test":
        "upper-tail hypergeometric",

    "enrichment_FDR":
        (
            "BH across all prespecified "
            "model x set x threshold tests."
        ),

    "model_fitting":
        False,

    "model_tuning":
        False,

    "new_PU_sampling":
        False,

    "external_gene_sets_changed":
        False,

    "topK_changed":
        False,

    "score_file":
        str(
            SCORE_FILE
        ),

    "score_file_sha256":
        sha256_file(
            SCORE_FILE
        ),

    "prespecification_file":
        str(
            PLAN_FILE
        ),

    "prespecification_sha256":
        sha256_file(
            PLAN_FILE
        ),

    "result_files":
        [
            {
                "path":
                    str(
                        path
                    ),

                "sha256":
                    sha256_file(
                        path
                    )
            }
            for path in result_files
        ]
}


MANIFEST_FILE = (
    OUT_DIR /
    "Stage9C_external_validation_manifest.json"
)


with open(
    MANIFEST_FILE,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        manifest,
        f,
        indent=2
    )


# ==========================================================================================
# 13. FINAL REPORT
# ==========================================================================================

print(
    "\n[10/10] FINAL STAGE 9C REPORT"
)


print(
    "\n" +
    "=" * 140
)

print(
    "STAGE 9C EXTERNAL VALIDATION COMPLETE [OK]"
)

print(
    "=" * 140
)


print(
    "\nCONTINUOUS RANK-PERCENTILE PERFORMANCE"
)


print(
    rank_summary[
        [
            "external_set",
            "n_external_genes",
            "model",
            "mean_rank_percentile",
            "median_rank_percentile"
        ]
    ].to_string(
        index=False
    )
)


print(
    "\nPAIRED FUSION-vs-CONTEXT BOOTSTRAP"
)


print(
    bootstrap_summary[
        [
            "external_set",
            "n_external_genes",
            "observed_mean_Fusion_minus_Context_rank_percentile",
            "observed_median_Fusion_minus_Context_rank_percentile",
            "CI95_low",
            "CI95_high",
            "bootstrap_fraction_delta_gt_0",
            "CI_excludes_zero"
        ]
    ].to_string(
        index=False
    )
)


print(
    "\nTOP-K RECOVERY"
)


print(
    topk[
        [
            "external_set",
            "model",
            "threshold",
            "recovered_genes",
            "recovery_fraction",
            "enrichment_fold",
            "hypergeometric_p",
            "BH_FDR_q_across_all_prespecified_enrichment_tests"
        ]
    ].to_string(
        index=False
    )
)


print(
    "\nFUSION-vs-CONTEXT TOP-K RECOVERY DELTAS"
)


print(
    recovery_comparison.to_string(
        index=False
    )
)


print(
    "\nINTEGRITY"
)

print(
    "Ranking universe           : 18062 non-seed genes"
)

print(
    "Top 1%                    : 181"
)

print(
    "Top 5%                    : 904"
)

print(
    "Top 10%                   : 1807"
)

print(
    "Models fitted             : NO"
)

print(
    "Models tuned              : NO"
)

print(
    "New PU sampling           : NO"
)

print(
    "External sets changed     : NO"
)

print(
    "Post-result thresholding  : NO"
)


print(
    "\nSAVED TO:"
)

print(
    OUT_DIR
)


print(
    "\nPASTE BACK:"
)

print(
    "1. CONTINUOUS RANK-PERCENTILE PERFORMANCE"
)

print(
    "2. PAIRED FUSION-vs-CONTEXT BOOTSTRAP"
)

print(
    "3. FUSION-vs-CONTEXT TOP-K RECOVERY DELTAS"
)

print(
    "4. TOP-K RECOVERY rows for Tan_NoEvidence32_nonseed"
)


print(
    "=" * 140
)
```

```python
# ==========================================================================================
# ASD-ContextFusion
# STAGE 9D - EXTERNAL SEQUENCE-REPRESENTATION VALIDATION
#
# SCIENTIFIC QUESTION
# -------------------
# On the already locked 2026 Tan external ASD gene sets, does the pretrained
# NT-v2 transcript representation prioritize external ASD genes better than
# simple sequence baselines?
#
# Models compared:
#   1. NT-v2
#   2. Length + GC
#   3. 4-mer
#
# IMPORTANT
# ---------
# - NO model fitting
# - NO 7,900-fold rerun
# - NO embedding generation
# - NO hyperparameter tuning
# - NO new PU sampling
# - NO new external gene selection
# - NO threshold selection from results
#
# Ranking universe:
#   18,062 frozen non-seed project genes
#
# External sets:
#   Broad Tan non-seed       = 134
#   Suggestive               = 45
#   No-evidence              = 30
#   Suggestive/no-evidence   = 75
#
# Prespecified top-K:
#   top 1%  = 181
#   top 5%  = 904
#   top 10% = 1807
#
# Primary comparisons:
#   NT-v2 - LengthGC
#   NT-v2 - 4-mer
#
# Continuous inference:
#   paired external-gene bootstrap of mean rank-percentile difference
#   5,000 replicates, seed 20260830
# ==========================================================================================

from pathlib import Path
import pandas as pd
import numpy as np
import hashlib
import json

from scipy.stats import hypergeom


# ==========================================================================================
# 0. PATHS / CONSTANTS
# ==========================================================================================

ROOT = Path(r"E:\ASD_ContextFusion_Q1")

SCORE_FILE = (
    ROOT /
    "05_results" /
    "proposed" /
    "sequence_controls" /
    "Context_NTv2_LengthGC_4mer_gene_level_OOF_scores.tsv"
)

MASTER_FILE = (
    ROOT /
    "03_features" /
    "multimodal_master" /
    "dhindsa_NTv2_row_preserved_master.tsv"
)

STAGE9B3_LOCK = (
    ROOT /
    "00_admin" /
    "ASD_ContextFusion_STAGE9B3_IDENTITY_RECONCILIATION_FINAL_LOCK.json"
)

STAGE9C_PLAN = (
    ROOT /
    "00_admin" /
    "ASD_ContextFusion_STAGE9C_EXTERNAL_VALIDATION_PRESPECIFICATION_V2.json"
)

SCORING_HASH_FILE = (
    ROOT /
    "02_interim" /
    "external_validation_source_lock" /
    "Stage9B3_scoring_ready_external_sets_SHA256.tsv"
)

EXT_DIR = (
    ROOT /
    "02_interim" /
    "external_validation_source_lock"
)

OUT_DIR = (
    ROOT /
    "05_results" /
    "proposed" /
    "stage9D_external_sequence_controls"
)

ADMIN_DIR = ROOT / "00_admin"

OUT_DIR.mkdir(
    parents=True,
    exist_ok=True
)


N_PROJECT = 18252
N_SEEDS = 190
N_NONSEED = 18062

TOP_K = {
    "Top1pct": 181,
    "Top5pct": 904,
    "Top10pct": 1807
}

BOOT_REPS = 5000
BOOT_SEED = 20260830

ANALYSIS_VERSION = "ASD_CF_STAGE9D_EXTERNAL_SEQUENCE_CONTROLS_V1"


SET_CONFIG = {

    "Tan_All245_nonseed": {
        "stage9b3_name": "Tan_All245",
        "file": (
            EXT_DIR /
            "Stage9B3_Tan_All245_NONSEED_PROJECT_IDENTITIES_LOCKED.tsv"
        ),
        "expected_n": 134,
        "role": "PRIMARY_BROAD"
    },

    "Tan_Suggestive46_nonseed": {
        "stage9b3_name": "Tan_SupplementSuggestive46",
        "file": (
            EXT_DIR /
            "Stage9B3_Tan_SupplementSuggestive46_"
            "NONSEED_PROJECT_IDENTITIES_LOCKED.tsv"
        ),
        "expected_n": 45,
        "role": "SECONDARY_SUGGESTIVE"
    },

    "Tan_NoEvidence32_nonseed": {
        "stage9b3_name": "Tan_SupplementNoEvidence32",
        "file": (
            EXT_DIR /
            "Stage9B3_Tan_SupplementNoEvidence32_"
            "NONSEED_PROJECT_IDENTITIES_LOCKED.tsv"
        ),
        "expected_n": 30,
        "role": "PRIMARY_NOVELTY_FOCUSED"
    },

    "Tan_SuggestiveOrNoEvidence78_nonseed": {
        "stage9b3_name": "Tan_SupplementSuggestiveOrNoEvidence78",
        "file": (
            EXT_DIR /
            "Stage9B3_Tan_SupplementSuggestiveOrNoEvidence78_"
            "NONSEED_PROJECT_IDENTITIES_LOCKED.tsv"
        ),
        "expected_n": 75,
        "role": "SECONDARY_LOW_EVIDENCE_COMBINED"
    }
}


# ==========================================================================================
# 1. HELPERS
# ==========================================================================================

def sha256_file(path, chunk_size=1024 * 1024):

    h = hashlib.sha256()

    with open(path, "rb") as f:

        while True:

            chunk = f.read(chunk_size)

            if not chunk:
                break

            h.update(chunk)

    return h.hexdigest()


def bh_adjust(pvalues):

    p = np.asarray(
        pvalues,
        dtype=float
    )

    m = len(p)

    order = np.argsort(p)

    ranked = p[order]

    adjusted = (
        ranked *
        m /
        np.arange(
            1,
            m + 1
        )
    )

    adjusted = np.minimum.accumulate(
        adjusted[::-1]
    )[::-1]

    adjusted = np.minimum(
        adjusted,
        1.0
    )

    out = np.empty_like(
        adjusted
    )

    out[order] = adjusted

    return out


def resolve_column(
    columns,
    exact_candidates,
    contains_tokens,
    model_name
):

    columns = list(columns)

    # Exact preferred matches.
    for candidate in exact_candidates:

        if candidate in columns:
            return candidate

    # Conservative fallback.
    hits = []

    for col in columns:

        lower = str(col).lower()

        if all(
            token.lower() in lower
            for token in contains_tokens
        ):

            hits.append(col)

    if len(hits) == 1:
        return hits[0]

    raise RuntimeError(
        f"\nCould not uniquely resolve score column for {model_name}.\n"
        f"Candidate matches: {hits}\n"
        f"Available columns:\n{columns}"
    )


# ==========================================================================================
# 2. INPUT / LOCK AUDIT
# ==========================================================================================

print("=" * 140)
print("ASD-CONTEXTFUSION - STAGE 9D EXTERNAL SEQUENCE-CONTROL VALIDATION")
print("=" * 140)

print("\n[1/11] INPUT / SCIENTIFIC LOCK AUDIT")


for path in [
    SCORE_FILE,
    MASTER_FILE,
    STAGE9B3_LOCK,
    STAGE9C_PLAN,
    SCORING_HASH_FILE
]:

    assert path.exists(), (
        f"Missing required file:\n{path}"
    )

    print(
        "[OK]",
        path
    )


with open(
    STAGE9B3_LOCK,
    "r",
    encoding="utf-8"
) as f:

    lock9b3 = json.load(f)


with open(
    STAGE9C_PLAN,
    "r",
    encoding="utf-8"
) as f:

    plan9c = json.load(f)


assert lock9b3["nonseed_ranking_universe"] == N_NONSEED

assert plan9c["ranking_universe_count"] == N_NONSEED

assert plan9c["top_thresholds"]["top_1_percent_K"] == 181
assert plan9c["top_thresholds"]["top_5_percent_K"] == 904
assert plan9c["top_thresholds"]["top_10_percent_K"] == 1807


# ==========================================================================================
# 3. WRITE STAGE 9D PRESPECIFICATION BEFORE READING SCORES
# ==========================================================================================

print("\n[2/11] WRITING STAGE 9D PRESPECIFICATION BEFORE SCORE ACCESS")


PRESPEC_FILE = (
    ADMIN_DIR /
    "ASD_ContextFusion_STAGE9D_EXTERNAL_SEQUENCE_CONTROL_PRESPECIFICATION.json"
)


prespec = {

    "project":
        "ASD-ContextFusion",

    "stage":
        "9D_external_sequence_representation_validation",

    "status":
        "LOCKED_BEFORE_STAGE9D_SEQUENCE_CONTROL_SCORE_ACCESS",

    "scientific_question":
        (
            "Do frozen NT-v2 transcript embeddings prioritize "
            "post-benchmark external ASD genes better than simple "
            "Length+GC and 4-mer sequence representations?"
        ),

    "models":
        [
            "NTv2",
            "LengthGC",
            "Fourmer"
        ],

    "primary_pairwise_comparisons":
        [
            "NTv2_minus_LengthGC",
            "NTv2_minus_Fourmer"
        ],

    "ranking_universe":
        "18,062 locked non-seed project genes",

    "ranking_universe_count":
        N_NONSEED,

    "external_sets":
        {
            key: {
                "expected_n":
                    value["expected_n"],
                "role":
                    value["role"]
            }
            for key, value
            in SET_CONFIG.items()
        },

    "rank_direction":
        "Higher probability = stronger ASD prioritization",

    "rank_tie_break":
        "Gene_Name ascending",

    "rank_percentile_formula":
        "100*(N-rank)/(N-1); best=100, worst=0",

    "top_thresholds":
        TOP_K,

    "primary_continuous_statistic":
        (
            "Mean paired rank-percentile difference "
            "between NT-v2 and each simple sequence control"
        ),

    "bootstrap_replicates":
        BOOT_REPS,

    "bootstrap_seed":
        BOOT_SEED,

    "bootstrap_unit":
        "external gene",

    "topk_outputs":
        [
            "recovery_count",
            "recovery_fraction",
            "enrichment_fold",
            "hypergeometric_p",
            "BH_FDR_q"
        ],

    "model_fitting":
        False,

    "model_tuning":
        False,

    "new_PU_sampling":
        False,

    "new_embedding_generation":
        False,

    "external_gene_sets_changed":
        False,

    "post_result_threshold_selection":
        False
}


with open(
    PRESPEC_FILE,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        prespec,
        f,
        indent=2
    )


print("[OK]", PRESPEC_FILE)


# ==========================================================================================
# 4. VERIFY LOCKED EXTERNAL SETS
# ==========================================================================================

print("\n[3/11] VERIFYING LOCKED EXTERNAL SET HASHES")


hash_table = pd.read_csv(
    SCORING_HASH_FILE,
    sep="\t"
)


external_sets = {}


for set_name, config in SET_CONFIG.items():

    path = config["file"]

    assert path.exists(), path

    ext = pd.read_csv(
        path,
        sep="\t"
    )


    assert len(ext) == config["expected_n"]

    assert "project_scoring_gene" in ext.columns

    assert ext[
        "project_scoring_gene"
    ].is_unique


    if "project_gene_is_locked_ASD_seed" in ext.columns:

        assert not (
            ext[
                "project_gene_is_locked_ASD_seed"
            ].astype(bool)
        ).any()


    row = hash_table.loc[
        hash_table["set"].eq(
            config["stage9b3_name"]
        )
    ]


    assert len(row) == 1, (
        f"Could not uniquely find Stage9B3 hash entry for "
        f"{config['stage9b3_name']}"
    )


    expected_hash = str(
        row.iloc[0]["sha256"]
    )


    actual_hash = sha256_file(
        path
    )


    assert actual_hash == expected_hash


    external_sets[
        set_name
    ] = ext


    print(
        f"{set_name:44s}",
        len(ext),
        "genes [HASH OK]"
    )


# ==========================================================================================
# 5. LOAD PROJECT UNIVERSE
# ==========================================================================================

print("\n[4/11] PROJECT NON-SEED UNIVERSE AUDIT")


master = pd.read_csv(
    MASTER_FILE,
    sep="\t"
)


assert "Gene_Name" in master.columns
assert "known_gene" in master.columns


project_genes = set(
    master[
        "Gene_Name"
    ]
    .astype(str)
    .str.strip()
)


seed_genes = set(
    master.loc[
        master["known_gene"] == 1,
        "Gene_Name"
    ]
    .astype(str)
    .str.strip()
)


nonseed_genes = (
    project_genes
    -
    seed_genes
)


assert len(project_genes) == N_PROJECT
assert len(seed_genes) == N_SEEDS
assert len(nonseed_genes) == N_NONSEED


print("Project genes :", len(project_genes))
print("ASD seeds     :", len(seed_genes))
print("Non-seed      :", len(nonseed_genes))


# ==========================================================================================
# 6. FIRST STAGE 9D SEQUENCE-CONTROL SCORE ACCESS
# ==========================================================================================

print("\n[5/11] FIRST STAGE 9D SEQUENCE-CONTROL SCORE ACCESS")


scores = pd.read_csv(
    SCORE_FILE,
    sep="\t"
)


print(
    "Score table shape:",
    scores.shape
)


print(
    "\nAVAILABLE SCORE COLUMNS:"
)

for i, col in enumerate(
    scores.columns
):

    print(
        f"{i:3d} | {col}"
    )


assert "Gene_Name" in scores.columns


NT_COL = resolve_column(
    scores.columns,
    exact_candidates=[
        "NT_probability",
        "NTv2_probability",
        "NT_v2_probability"
    ],
    contains_tokens=[
        "nt",
        "prob"
    ],
    model_name="NT-v2"
)


LENGTHGC_COL = resolve_column(
    scores.columns,
    exact_candidates=[
        "LengthGC_probability",
        "Length_GC_probability",
        "LengthGC_prob"
    ],
    contains_tokens=[
        "length",
        "gc"
    ],
    model_name="Length+GC"
)


FOURMER_COL = resolve_column(
    scores.columns,
    exact_candidates=[
        "Fourmer_probability",
        "FourMer_probability",
        "4mer_probability",
        "Fourmer_prob",
        "4mer_prob"
    ],
    contains_tokens=[
        "mer"
    ],
    model_name="4-mer"
)


MODEL_COLUMNS = {
    "NTv2":
        NT_COL,

    "LengthGC":
        LENGTHGC_COL,

    "Fourmer":
        FOURMER_COL
}


print("\nRESOLVED MODEL COLUMNS")

for model, col in MODEL_COLUMNS.items():

    print(
        f"{model:10s} -> {col}"
    )


assert scores[
    "Gene_Name"
].nunique() == N_PROJECT


score_genes = set(
    scores[
        "Gene_Name"
    ]
    .astype(str)
    .str.strip()
)


assert score_genes == project_genes


for model, col in MODEL_COLUMNS.items():

    scores[col] = pd.to_numeric(
        scores[col],
        errors="raise"
    )

    assert np.isfinite(
        scores[col]
    ).all()


# ==========================================================================================
# 7. BUILD 18,062-GENE NON-SEED RANKS
# ==========================================================================================

print("\n[6/11] BUILDING NON-SEED SEQUENCE-MODEL RANKS")


nonseed_scores = (
    scores.loc[
        scores[
            "Gene_Name"
        ].isin(
            nonseed_genes
        )
    ]
    .copy()
)


assert len(nonseed_scores) == N_NONSEED


rank_table = nonseed_scores[
    [
        "Gene_Name",
        NT_COL,
        LENGTHGC_COL,
        FOURMER_COL
    ]
].copy()


for model, col in MODEL_COLUMNS.items():

    ordered = (
        rank_table[
            [
                "Gene_Name",
                col
            ]
        ]
        .sort_values(
            by=[
                col,
                "Gene_Name"
            ],
            ascending=[
                False,
                True
            ],
            kind="mergesort"
        )
        .reset_index(
            drop=True
        )
    )


    ordered[
        f"{model}_rank"
    ] = np.arange(
        1,
        N_NONSEED + 1
    )


    ordered[
        f"{model}_rank_percentile"
    ] = (
        100.0
        *
        (
            N_NONSEED
            -
            ordered[
                f"{model}_rank"
            ]
        )
        /
        (
            N_NONSEED
            -
            1
        )
    )


    rank_map = dict(
        zip(
            ordered["Gene_Name"],
            ordered[f"{model}_rank"]
        )
    )


    pct_map = dict(
        zip(
            ordered["Gene_Name"],
            ordered[f"{model}_rank_percentile"]
        )
    )


    rank_table[
        f"{model}_rank"
    ] = rank_table[
        "Gene_Name"
    ].map(
        rank_map
    )


    rank_table[
        f"{model}_rank_percentile"
    ] = rank_table[
        "Gene_Name"
    ].map(
        pct_map
    )


    tied_rows = int(
        rank_table[
            col
        ].duplicated(
            keep=False
        ).sum()
    )


    print(
        f"{model:10s}",
        "tied-score rows:",
        tied_rows
    )


# ==========================================================================================
# 8. EXTERNAL GENE RANK DETAILS
# ==========================================================================================

print("\n[7/11] EXTERNAL SEQUENCE-REPRESENTATION EVALUATION")


detail_frames = []


for set_name, ext in external_sets.items():

    merged = (
        ext
        .merge(
            rank_table,
            left_on="project_scoring_gene",
            right_on="Gene_Name",
            how="left",
            validate="one_to_one"
        )
    )


    assert len(merged) == len(ext)

    assert merged[
        "Gene_Name"
    ].notna().all()


    merged[
        "external_set"
    ] = set_name


    merged[
        "NTv2_minus_LengthGC_rank_percentile"
    ] = (
        merged[
            "NTv2_rank_percentile"
        ]
        -
        merged[
            "LengthGC_rank_percentile"
        ]
    )


    merged[
        "NTv2_minus_Fourmer_rank_percentile"
    ] = (
        merged[
            "NTv2_rank_percentile"
        ]
        -
        merged[
            "Fourmer_rank_percentile"
        ]
    )


    detail_frames.append(
        merged
    )


details = pd.concat(
    detail_frames,
    ignore_index=True
)


# ==========================================================================================
# 9. CONTINUOUS RANK SUMMARY
# ==========================================================================================

rank_summary_rows = []


for set_name in SET_CONFIG.keys():

    d = details.loc[
        details[
            "external_set"
        ] == set_name
    ]


    for model in [
        "NTv2",
        "LengthGC",
        "Fourmer"
    ]:

        pct = d[
            f"{model}_rank_percentile"
        ].to_numpy(
            dtype=float
        )


        ranks = d[
            f"{model}_rank"
        ].to_numpy(
            dtype=int
        )


        rank_summary_rows.append(
            {
                "external_set":
                    set_name,

                "role":
                    SET_CONFIG[
                        set_name
                    ][
                        "role"
                    ],

                "n_external_genes":
                    len(d),

                "model":
                    model,

                "mean_rank_percentile":
                    float(
                        np.mean(pct)
                    ),

                "median_rank_percentile":
                    float(
                        np.median(pct)
                    ),

                "Q1_rank_percentile":
                    float(
                        np.quantile(
                            pct,
                            0.25
                        )
                    ),

                "Q3_rank_percentile":
                    float(
                        np.quantile(
                            pct,
                            0.75
                        )
                    ),

                "mean_ordinal_rank":
                    float(
                        np.mean(ranks)
                    ),

                "median_ordinal_rank":
                    float(
                        np.median(ranks)
                    )
            }
        )


rank_summary = pd.DataFrame(
    rank_summary_rows
)


# ==========================================================================================
# 10. TOP-K RECOVERY / ENRICHMENT
# ==========================================================================================

topk_rows = []


for set_name in SET_CONFIG.keys():

    d = details.loc[
        details[
            "external_set"
        ] == set_name
    ]


    M = len(d)


    for model in [
        "NTv2",
        "LengthGC",
        "Fourmer"
    ]:

        for threshold, K in TOP_K.items():

            ranks = d[
                f"{model}_rank"
            ].to_numpy(
                dtype=int
            )


            recovered = int(
                np.sum(
                    ranks <= K
                )
            )


            recovery_fraction = (
                recovered
                /
                M
            )


            random_fraction = (
                K
                /
                N_NONSEED
            )


            enrichment = (
                recovery_fraction
                /
                random_fraction
            )


            expected_random = (
                M *
                K /
                N_NONSEED
            )


            p_hyper = float(
                hypergeom.sf(
                    recovered - 1,
                    N_NONSEED,
                    M,
                    K
                )
            )


            topk_rows.append(
                {
                    "external_set":
                        set_name,

                    "role":
                        SET_CONFIG[
                            set_name
                        ][
                            "role"
                        ],

                    "model":
                        model,

                    "threshold":
                        threshold,

                    "K":
                        K,

                    "background_N":
                        N_NONSEED,

                    "external_set_M":
                        M,

                    "recovered_genes":
                        recovered,

                    "recovery_fraction":
                        recovery_fraction,

                    "expected_random_overlap":
                        expected_random,

                    "enrichment_fold":
                        enrichment,

                    "hypergeometric_p":
                        p_hyper
                }
            )


topk = pd.DataFrame(
    topk_rows
)


topk[
    "BH_FDR_q_all_prespecified_tests"
] = bh_adjust(
    topk[
        "hypergeometric_p"
    ].to_numpy()
)


# ==========================================================================================
# 11. PAIRED BOOTSTRAP: NT-v2 vs SIMPLE CONTROLS
# ==========================================================================================

print("\n[8/11] PAIRED BOOTSTRAP: NT-v2 vs LengthGC / 4-mer")


comparisons = {

    "NTv2_vs_LengthGC":
        (
            "NTv2_rank_percentile",
            "LengthGC_rank_percentile"
        ),

    "NTv2_vs_Fourmer":
        (
            "NTv2_rank_percentile",
            "Fourmer_rank_percentile"
        )
}


boot_summary_rows = []
boot_rep_rows = []


for set_index, set_name in enumerate(
    SET_CONFIG.keys()
):

    d = details.loc[
        details[
            "external_set"
        ] == set_name
    ].copy()


    for comparison_index, (
        comparison_name,
        columns
    ) in enumerate(
        comparisons.items()
    ):

        model_a_col, model_b_col = columns


        delta = (
            d[
                model_a_col
            ].to_numpy(
                dtype=float
            )
            -
            d[
                model_b_col
            ].to_numpy(
                dtype=float
            )
        )


        n = len(delta)


        observed_mean = float(
            np.mean(delta)
        )


        observed_median = float(
            np.median(delta)
        )


        boot_means = np.empty(
            BOOT_REPS,
            dtype=float
        )


        for b in range(
            BOOT_REPS
        ):

            rng = np.random.default_rng(
                np.random.SeedSequence(
                    [
                        BOOT_SEED,
                        set_index,
                        comparison_index,
                        b
                    ]
                )
            )


            idx = rng.integers(
                0,
                n,
                size=n
            )


            value = float(
                np.mean(
                    delta[idx]
                )
            )


            boot_means[b] = value


            boot_rep_rows.append(
                {
                    "external_set":
                        set_name,

                    "comparison":
                        comparison_name,

                    "replicate":
                        b,

                    "mean_rank_percentile_delta":
                        value
                }
            )


        ci_low = float(
            np.quantile(
                boot_means,
                0.025
            )
        )


        ci_high = float(
            np.quantile(
                boot_means,
                0.975
            )
        )


        boot_summary_rows.append(
            {
                "external_set":
                    set_name,

                "role":
                    SET_CONFIG[
                        set_name
                    ][
                        "role"
                    ],

                "comparison":
                    comparison_name,

                "n_external_genes":
                    n,

                "observed_mean_rank_percentile_delta":
                    observed_mean,

                "observed_median_rank_percentile_delta":
                    observed_median,

                "bootstrap_mean":
                    float(
                        np.mean(
                            boot_means
                        )
                    ),

                "bootstrap_SD":
                    float(
                        np.std(
                            boot_means,
                            ddof=1
                        )
                    ),

                "CI95_low":
                    ci_low,

                "CI95_high":
                    ci_high,

                "bootstrap_fraction_delta_gt_0":
                    float(
                        np.mean(
                            boot_means > 0
                        )
                    ),

                "CI_excludes_zero":
                    bool(
                        ci_low > 0
                        or
                        ci_high < 0
                    ),

                "bootstrap_replicates":
                    BOOT_REPS,

                "bootstrap_seed":
                    BOOT_SEED
            }
        )


boot_summary = pd.DataFrame(
    boot_summary_rows
)


boot_replicates = pd.DataFrame(
    boot_rep_rows
)


# ==========================================================================================
# 12. DESCRIPTIVE TOP-K NT-v2 vs CONTROL DELTAS
# ==========================================================================================

topk_delta_rows = []


for set_name in SET_CONFIG.keys():

    for threshold in TOP_K.keys():

        nt = topk.loc[
            (
                topk[
                    "external_set"
                ] == set_name
            )
            &
            (
                topk[
                    "threshold"
                ] == threshold
            )
            &
            (
                topk[
                    "model"
                ] == "NTv2"
            )
        ].iloc[0]


        for control in [
            "LengthGC",
            "Fourmer"
        ]:

            ctrl = topk.loc[
                (
                    topk[
                        "external_set"
                    ] == set_name
                )
                &
                (
                    topk[
                        "threshold"
                    ] == threshold
                )
                &
                (
                    topk[
                        "model"
                    ] == control
                )
            ].iloc[0]


            topk_delta_rows.append(
                {
                    "external_set":
                        set_name,

                    "threshold":
                        threshold,

                    "comparison":
                        f"NTv2_vs_{control}",

                    "NTv2_recovered":
                        int(
                            nt[
                                "recovered_genes"
                            ]
                        ),

                    "control_recovered":
                        int(
                            ctrl[
                                "recovered_genes"
                            ]
                        ),

                    "NTv2_minus_control_recovered":
                        int(
                            nt[
                                "recovered_genes"
                            ]
                            -
                            ctrl[
                                "recovered_genes"
                            ]
                        ),

                    "NTv2_recovery_fraction":
                        float(
                            nt[
                                "recovery_fraction"
                            ]
                        ),

                    "control_recovery_fraction":
                        float(
                            ctrl[
                                "recovery_fraction"
                            ]
                        ),

                    "NTv2_minus_control_recovery_fraction":
                        float(
                            nt[
                                "recovery_fraction"
                            ]
                            -
                            ctrl[
                                "recovery_fraction"
                            ]
                        )
                }
            )


topk_deltas = pd.DataFrame(
    topk_delta_rows
)


# ==========================================================================================
# 13. SAVE RESULTS
# ==========================================================================================

print("\n[9/11] SAVING STAGE 9D RESULTS")


RANK_FILE = (
    OUT_DIR /
    "Stage9D_nonseed_18062_sequence_model_ranks.tsv.gz"
)

DETAIL_FILE = (
    OUT_DIR /
    "Stage9D_external_gene_sequence_rank_details.tsv"
)

RANK_SUMMARY_FILE = (
    OUT_DIR /
    "Stage9D_external_sequence_rank_summary.tsv"
)

TOPK_FILE = (
    OUT_DIR /
    "Stage9D_external_sequence_topk_enrichment.tsv"
)

TOPK_DELTA_FILE = (
    OUT_DIR /
    "Stage9D_NTv2_vs_controls_topk_deltas.tsv"
)

BOOT_SUMMARY_FILE = (
    OUT_DIR /
    "Stage9D_NTv2_vs_controls_bootstrap_summary.tsv"
)

BOOT_REP_FILE = (
    OUT_DIR /
    "Stage9D_NTv2_vs_controls_bootstrap_replicates.tsv.gz"
)


rank_table.to_csv(
    RANK_FILE,
    sep="\t",
    index=False,
    compression="gzip"
)


details.to_csv(
    DETAIL_FILE,
    sep="\t",
    index=False
)


rank_summary.to_csv(
    RANK_SUMMARY_FILE,
    sep="\t",
    index=False
)


topk.to_csv(
    TOPK_FILE,
    sep="\t",
    index=False
)


topk_deltas.to_csv(
    TOPK_DELTA_FILE,
    sep="\t",
    index=False
)


boot_summary.to_csv(
    BOOT_SUMMARY_FILE,
    sep="\t",
    index=False
)


boot_replicates.to_csv(
    BOOT_REP_FILE,
    sep="\t",
    index=False,
    compression="gzip"
)


# ==========================================================================================
# 14. MANIFEST
# ==========================================================================================

print("\n[10/11] WRITING REPRODUCIBILITY MANIFEST")


result_files = [
    RANK_FILE,
    DETAIL_FILE,
    RANK_SUMMARY_FILE,
    TOPK_FILE,
    TOPK_DELTA_FILE,
    BOOT_SUMMARY_FILE,
    BOOT_REP_FILE
]


MANIFEST_FILE = (
    OUT_DIR /
    "Stage9D_external_sequence_controls_manifest.json"
)


manifest = {

    "project":
        "ASD-ContextFusion",

    "stage":
        "9D_external_sequence_controls",

    "analysis_version":
        ANALYSIS_VERSION,

    "scientific_question":
        (
            "External temporal generalization of NT-v2 "
            "versus simple sequence controls."
        ),

    "models":
        MODEL_COLUMNS,

    "ranking_universe_N":
        N_NONSEED,

    "topK":
        TOP_K,

    "external_sets":
        {
            key: {
                "n":
                    value[
                        "expected_n"
                    ],

                "role":
                    value[
                        "role"
                    ],

                "file":
                    str(
                        value[
                            "file"
                        ]
                    ),

                "sha256":
                    sha256_file(
                        value[
                            "file"
                        ]
                    )
            }
            for key, value
            in SET_CONFIG.items()
        },

    "primary_comparisons":
        list(
            comparisons.keys()
        ),

    "paired_bootstrap_replicates":
        BOOT_REPS,

    "paired_bootstrap_seed":
        BOOT_SEED,

    "bootstrap_fraction_gt_zero_is_pvalue":
        False,

    "model_fitting":
        False,

    "model_tuning":
        False,

    "new_PU_sampling":
        False,

    "new_embedding_generation":
        False,

    "external_gene_sets_changed":
        False,

    "score_file":
        str(
            SCORE_FILE
        ),

    "score_file_sha256":
        sha256_file(
            SCORE_FILE
        ),

    "prespecification_file":
        str(
            PRESPEC_FILE
        ),

    "prespecification_sha256":
        sha256_file(
            PRESPEC_FILE
        ),

    "result_files":
        [
            {
                "path":
                    str(path),

                "sha256":
                    sha256_file(path)
            }
            for path in result_files
        ]
}


with open(
    MANIFEST_FILE,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        manifest,
        f,
        indent=2
    )


# ==========================================================================================
# 15. FINAL REPORT
# ==========================================================================================

print("\n[11/11] FINAL STAGE 9D REPORT")

print("\n" + "=" * 140)
print("STAGE 9D EXTERNAL SEQUENCE-CONTROL VALIDATION COMPLETE [OK]")
print("=" * 140)


print("\nCONTINUOUS RANK-PERCENTILE PERFORMANCE")

print(
    rank_summary[
        [
            "external_set",
            "n_external_genes",
            "model",
            "mean_rank_percentile",
            "median_rank_percentile"
        ]
    ].to_string(
        index=False
    )
)


print("\nPAIRED NT-v2 vs CONTROL BOOTSTRAP")

print(
    boot_summary[
        [
            "external_set",
            "comparison",
            "n_external_genes",
            "observed_mean_rank_percentile_delta",
            "observed_median_rank_percentile_delta",
            "CI95_low",
            "CI95_high",
            "bootstrap_fraction_delta_gt_0",
            "CI_excludes_zero"
        ]
    ].to_string(
        index=False
    )
)


print("\nTOP-K RECOVERY")

print(
    topk[
        [
            "external_set",
            "model",
            "threshold",
            "recovered_genes",
            "recovery_fraction",
            "enrichment_fold",
            "hypergeometric_p",
            "BH_FDR_q_all_prespecified_tests"
        ]
    ].to_string(
        index=False
    )
)


print("\nNT-v2 vs CONTROL TOP-K DELTAS")

print(
    topk_deltas.to_string(
        index=False
    )
)


print("\nINTEGRITY")

print("Models fitted            : NO")
print("Models tuned             : NO")
print("7,900 folds rerun        : NO")
print("New PU sampling          : NO")
print("New embeddings generated : NO")
print("External genes changed   : NO")
print("Thresholds changed       : NO")
print("Ranking universe         : 18,062 non-seed genes")


print("\nSAVED TO:")
print(OUT_DIR)


print("\nPASTE BACK THESE 3 SECTIONS:")
print("1. CONTINUOUS RANK-PERCENTILE PERFORMANCE")
print("2. PAIRED NT-v2 vs CONTROL BOOTSTRAP")
print("3. NT-v2 vs CONTROL TOP-K DELTAS")

print("=" * 140)
```

```python
# ==========================================================================================
# ASD-ContextFusion
# STAGE 9E.0 — HISTORICAL ASD COMPARATOR SOURCE LOCK
#
# SOURCE:
# DeepND 2022 Supplementary Table S1
#
# PURPOSE:
# - validate official XLSX
# - SHA256 lock
# - inspect sheets / columns
# - identify available genome-wide ASD comparator scores/ranks
#
# NO:
# - model training
# - 7900-fold rerun
# - external-set change
# - comparator cherry-picking
# - performance analysis yet
# ==========================================================================================

from pathlib import Path
import pandas as pd
import hashlib
import json
import zipfile


ROOT = Path(r"E:\ASD_ContextFusion_Q1")

SOURCE_FILE = (
    ROOT /
    "01_raw" /
    "historical_comparators" /
    "DeepND_2022_TableS1_genomewide_ASD_rankings.xlsx"
)

OUT_DIR = (
    ROOT /
    "02_interim" /
    "historical_comparator_lock"
)

ADMIN_DIR = ROOT / "00_admin"

OUT_DIR.mkdir(
    parents=True,
    exist_ok=True
)


def sha256_file(path, chunk_size=1024 * 1024):

    h = hashlib.sha256()

    with open(path, "rb") as f:

        while True:
            chunk = f.read(chunk_size)

            if not chunk:
                break

            h.update(chunk)

    return h.hexdigest()


print("=" * 130)
print("ASD-CONTEXTFUSION — STAGE 9E.0 HISTORICAL COMPARATOR SOURCE LOCK")
print("=" * 130)


# ==========================================================================================
# 1. FILE VALIDATION
# ==========================================================================================

print("\n[1/6] VALIDATING DEEPND OFFICIAL XLSX")

assert SOURCE_FILE.exists(), (
    f"File not found:\n{SOURCE_FILE}"
)

assert zipfile.is_zipfile(SOURCE_FILE), (
    "File is not a genuine XLSX ZIP container."
)

with zipfile.ZipFile(SOURCE_FILE, "r") as z:

    names = set(z.namelist())

    assert "[Content_Types].xml" in names
    assert "xl/workbook.xml" in names


source_hash = sha256_file(
    SOURCE_FILE
)


print("File   :", SOURCE_FILE)
print("Bytes  :", SOURCE_FILE.stat().st_size)
print("SHA256 :", source_hash)
print("Status : VALID_XLSX [OK]")


# ==========================================================================================
# 2. WORKBOOK INVENTORY
# ==========================================================================================

print("\n[2/6] WORKBOOK INVENTORY")


xls = pd.ExcelFile(
    SOURCE_FILE,
    engine="openpyxl"
)


print(
    "Number of sheets:",
    len(xls.sheet_names)
)


for i, sheet in enumerate(
    xls.sheet_names
):

    print(
        f"[{i}] {sheet}"
    )


# ==========================================================================================
# 3. INSPECT EACH SHEET WITHOUT ASSUMING HEADER
# ==========================================================================================

print("\n[3/6] RAW SHEET PREVIEW")


preview_rows = []


for sheet_index, sheet in enumerate(
    xls.sheet_names
):

    raw = pd.read_excel(
        SOURCE_FILE,
        sheet_name=sheet,
        header=None,
        nrows=12,
        engine="openpyxl"
    )


    print("\n" + "-" * 120)
    print(
        f"SHEET {sheet_index}: {sheet}"
    )
    print("-" * 120)

    print(
        raw.to_string(
            index=True,
            header=True,
            max_cols=60
        )
    )


    for row_index in range(
        min(
            len(raw),
            12
        )
    ):

        values = [
            str(x).strip()
            for x in raw.iloc[
                row_index
            ].tolist()
            if pd.notna(x)
        ]


        joined = " | ".join(
            values
        )


        preview_rows.append(
            {
                "sheet_index":
                    sheet_index,

                "sheet":
                    sheet,

                "excel_row":
                    row_index + 1,

                "content":
                    joined
            }
        )


preview_df = pd.DataFrame(
    preview_rows
)


PREVIEW_FILE = (
    OUT_DIR /
    "Stage9E0_DeepND_raw_first12_rows.tsv"
)


preview_df.to_csv(
    PREVIEW_FILE,
    sep="\t",
    index=False
)


# ==========================================================================================
# 4. HEADER-CANDIDATE AUDIT
# ==========================================================================================

print("\n[4/6] HEADER / COMPARATOR COLUMN AUDIT")


locator_terms = [
    "gene",
    "symbol",
    "asd",
    "deepnd",
    "forecasd",
    "krishnan",
    "dawn",
    "damages",
    "tada",
    "rank",
    "score",
    "probability",
    "prediction",
    "pli"
]


candidate_rows = []


for sheet_index, sheet in enumerate(
    xls.sheet_names
):

    raw = pd.read_excel(
        SOURCE_FILE,
        sheet_name=sheet,
        header=None,
        nrows=20,
        engine="openpyxl"
    )


    for row_index in range(
        len(raw)
    ):

        cells = [
            str(x).strip()
            for x in raw.iloc[
                row_index
            ].tolist()
            if pd.notna(x)
        ]


        lower = [
            x.lower()
            for x in cells
        ]


        hits = sorted(
            {
                term
                for term in locator_terms
                if any(
                    term in cell
                    for cell in lower
                )
            }
        )


        if (
            "gene" in hits
            or
            "symbol" in hits
        ) and len(hits) >= 2:

            candidate_rows.append(
                {
                    "sheet_index":
                        sheet_index,

                    "sheet":
                        sheet,

                    "excel_row":
                        row_index + 1,

                    "locator_hits":
                        " | ".join(
                            hits
                        ),

                    "row_content":
                        " | ".join(
                            cells
                        )
                }
            )


candidate_headers = pd.DataFrame(
    candidate_rows
)


print(
    candidate_headers.to_string(
        index=False
    )
)


HEADER_FILE = (
    OUT_DIR /
    "Stage9E0_DeepND_candidate_header_rows.tsv"
)


candidate_headers.to_csv(
    HEADER_FILE,
    sep="\t",
    index=False
)


# ==========================================================================================
# 5. TRY PLAUSIBLE HEADER ROWS AND PRINT COLUMN NAMES
# ==========================================================================================

print("\n[5/6] CANDIDATE PARSED COLUMN NAMES")


column_inventory_rows = []


for _, candidate in candidate_headers.iterrows():

    sheet = candidate[
        "sheet"
    ]

    excel_row = int(
        candidate[
            "excel_row"
        ]
    )


    try:

        parsed = pd.read_excel(
            SOURCE_FILE,
            sheet_name=sheet,
            header=excel_row - 1,
            engine="openpyxl"
        )

    except Exception as e:

        continue


    print("\n" + "=" * 120)

    print(
        f"SHEET: {sheet} | HEADER EXCEL ROW: {excel_row}"
    )

    print(
        f"Parsed shape: {parsed.shape}"
    )


    for i, col in enumerate(
        parsed.columns
    ):

        print(
            f"{i:3d} | {col}"
        )


        column_inventory_rows.append(
            {
                "sheet":
                    sheet,

                "header_excel_row":
                    excel_row,

                "parsed_rows":
                    len(
                        parsed
                    ),

                "column_index":
                    i,

                "column_name":
                    str(
                        col
                    )
            }
        )


column_inventory = pd.DataFrame(
    column_inventory_rows
)


COLUMN_FILE = (
    OUT_DIR /
    "Stage9E0_DeepND_candidate_column_inventory.tsv"
)


column_inventory.to_csv(
    COLUMN_FILE,
    sep="\t",
    index=False
)


# ==========================================================================================
# 6. SOURCE LOCK MANIFEST
# ==========================================================================================

print("\n[6/6] WRITING SOURCE LOCK")


LOCK_FILE = (
    ADMIN_DIR /
    "ASD_ContextFusion_STAGE9E0_DEEPND_COMPARATOR_SOURCE_LOCK.json"
)


manifest = {

    "project":
        "ASD-ContextFusion",

    "stage":
        "9E.0_historical_comparator_source_lock",

    "source":
        (
            "DeepND 2022 Supplementary Table S1 — "
            "Genome-wide risk-probability predictions and rankings "
            "for ASD and ID"
        ),

    "source_file":
        str(
            SOURCE_FILE
        ),

    "source_sha256":
        source_hash,

    "source_file_size_bytes":
        SOURCE_FILE.stat().st_size,

    "official_description":
        (
            "DeepND Supplementary Table S1 contains genome-wide "
            "DeepND risk predictions/rankings and rankings from "
            "other published gene discovery algorithms."
        ),

    "historical_comparators_selected":
        False,

    "external_validation_results_recomputed":
        False,

    "model_fitting":
        False,

    "new_PU_sampling":
        False,

    "our_model_scores_accessed_in_stage9E":
        False,

    "comparator_performance_accessed":
        False,

    "candidate_header_file":
        str(
            HEADER_FILE
        ),

    "candidate_column_inventory_file":
        str(
            COLUMN_FILE
        )
}


with open(
    LOCK_FILE,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        manifest,
        f,
        indent=2
    )


print("\n" + "=" * 130)
print("STAGE 9E.0 DEEPND COMPARATOR SOURCE LOCK COMPLETE [OK]")
print("=" * 130)

print("\nComparator selected        : NO")
print("Comparator performance read: NO")
print("Models fitted              : NO")
print("7,900-fold rerun           : NO")
print("New PU sampling            : NO")

print("\nSAVED:")
print(PREVIEW_FILE)
print(HEADER_FILE)
print(COLUMN_FILE)
print(LOCK_FILE)

print("\nPASTE BACK:")
print("1. HEADER / COMPARATOR COLUMN AUDIT")
print("2. CANDIDATE PARSED COLUMN NAMES")

print("=" * 130)
```

```python
# ==========================================================================================
# ASD-ContextFusion
# STAGE 9E.1 — HISTORICALLY-UNSEEN TEMPORAL BENCHMARK AGAINST PUBLISHED ASD PREDICTORS
#
# PURPOSE
# -------
# Test whether frozen ASD-ContextFusion rankings generalize to 2026 Tan ASD genes
# relative to published historical ASD-gene prioritization methods / generic constraint.
#
# OUR MODELS:
#   Context
#   Strict Fusion
#
# HISTORICAL COMPARATORS FROM LOCKED DeepND Table S1:
#   DeepND Probability          (higher = better)
#   Krishnan et al. Ranking    (lower  = better)
#   Dawn Ranking               (lower  = better)
#   DAMAGES Ranking            (lower  = better)
#   pLI                        (higher = better; generic constraint baseline)
#
# PRIMARY FAIRNESS RULE
# ---------------------
# Evaluate on genes that were HISTORICALLY UNLABELED in DeepND:
#   Evidence Level == "-"
#
# Thus DeepND/Krishnan training labels are not counted as novel temporal successes.
#
# For each comparator:
#   - use pair-specific common universe
#   - never impute missing comparator scores
#   - rerank OUR model and comparator inside exactly the SAME universe
#   - evaluate the same external genes available in that universe
#
# INFERENCE:
#   - paired bootstrap of rank-percentile difference
#   - 5,000 replicates
#   - seed 20260830
#
# NO:
#   - model fitting
#   - 7,900-fold rerun
#   - tuning
#   - new PU sampling
#   - new external gene selection
#   - result-driven comparator selection
# ==========================================================================================

from pathlib import Path
import pandas as pd
import numpy as np
import hashlib
import json
import math
from scipy.stats import hypergeom


# ==========================================================================================
# 0. PATHS
# ==========================================================================================

ROOT = Path(r"E:\ASD_ContextFusion_Q1")

DEEPND_FILE = (
    ROOT /
    "01_raw" /
    "historical_comparators" /
    "DeepND_2022_TableS1_genomewide_ASD_rankings.xlsx"
)

DEEPND_LOCK = (
    ROOT /
    "00_admin" /
    "ASD_ContextFusion_STAGE9E0_DEEPND_COMPARATOR_SOURCE_LOCK.json"
)

STAGE9C_DIR = (
    ROOT /
    "05_results" /
    "proposed" /
    "stage9C_external_validation"
)

EXT_DIR = (
    ROOT /
    "02_interim" /
    "external_validation_source_lock"
)

OUT_DIR = (
    ROOT /
    "05_results" /
    "proposed" /
    "stage9E_historical_comparator_benchmark"
)

ADMIN_DIR = ROOT / "00_admin"

OUT_DIR.mkdir(
    parents=True,
    exist_ok=True
)

BOOT_REPS = 5000
BOOT_SEED = 20260830

N_EXPECTED_NONSEED = 18062


# ==========================================================================================
# 1. LOCKED EXTERNAL SETS
# ==========================================================================================

SET_CONFIG = {

    "Tan_All245_nonseed": {
        "file":
            EXT_DIR /
            "Stage9B3_Tan_All245_NONSEED_PROJECT_IDENTITIES_LOCKED.tsv",
        "expected_n": 134,
        "role": "PRIMARY_BROAD"
    },

    "Tan_Suggestive46_nonseed": {
        "file":
            EXT_DIR /
            "Stage9B3_Tan_SupplementSuggestive46_"
            "NONSEED_PROJECT_IDENTITIES_LOCKED.tsv",
        "expected_n": 45,
        "role": "SECONDARY_SUGGESTIVE"
    },

    "Tan_NoEvidence32_nonseed": {
        "file":
            EXT_DIR /
            "Stage9B3_Tan_SupplementNoEvidence32_"
            "NONSEED_PROJECT_IDENTITIES_LOCKED.tsv",
        "expected_n": 30,
        "role": "PRIMARY_NOVELTY_FOCUSED"
    },

    "Tan_SuggestiveOrNoEvidence78_nonseed": {
        "file":
            EXT_DIR /
            "Stage9B3_Tan_SupplementSuggestiveOrNoEvidence78_"
            "NONSEED_PROJECT_IDENTITIES_LOCKED.tsv",
        "expected_n": 75,
        "role": "SECONDARY_LOW_EVIDENCE_COMBINED"
    }
}


# ==========================================================================================
# 2. COMPARATOR LOCK — DEFINED BEFORE TEMPORAL PERFORMANCE ACCESS
# ==========================================================================================

COMPARATORS = {

    "DeepND": {
        "column": "Probability",
        "direction": "higher"
    },

    "Krishnan": {
        "column": "Krishnan et al. Ranking",
        "direction": "lower"
    },

    "DAWN": {
        "column": "Dawn Ranking",
        "direction": "lower"
    },

    "DAMAGES": {
        "column": "DAMAGES Ranking",
        "direction": "lower"
    },

    "pLI": {
        "column": "pLI",
        "direction": "higher"
    }
}

OUR_MODELS = [
    "Context",
    "Fusion"
]


# ==========================================================================================
# 3. HELPERS
# ==========================================================================================

def sha256_file(path, chunk_size=1024 * 1024):

    h = hashlib.sha256()

    with open(path, "rb") as f:

        while True:

            chunk = f.read(chunk_size)

            if not chunk:
                break

            h.update(chunk)

    return h.hexdigest()


def numeric_clean(series):

    s = (
        series
        .astype(str)
        .str.strip()
        .str.replace(",", ".", regex=False)
    )

    s = s.replace(
        {
            "-": np.nan,
            "nan": np.nan,
            "None": np.nan,
            "": np.nan
        }
    )

    return pd.to_numeric(
        s,
        errors="coerce"
    )


def bh_adjust(pvalues):

    p = np.asarray(
        pvalues,
        dtype=float
    )

    m = len(p)

    order = np.argsort(p)
    ranked = p[order]

    adjusted = (
        ranked *
        m /
        np.arange(
            1,
            m + 1
        )
    )

    adjusted = np.minimum.accumulate(
        adjusted[::-1]
    )[::-1]

    adjusted = np.minimum(
        adjusted,
        1.0
    )

    out = np.empty_like(
        adjusted
    )

    out[order] = adjusted

    return out


def resolve_rank_column(columns, model):

    hits = [
        c for c in columns
        if model.lower() in str(c).lower()
        and "rank" in str(c).lower()
        and "percent" not in str(c).lower()
    ]

    if len(hits) == 1:
        return hits[0]

    raise RuntimeError(
        f"Could not uniquely resolve {model} rank column.\n"
        f"Hits: {hits}\n"
        f"Columns: {list(columns)}"
    )


def rerank_percentile(
    df,
    value_col,
    direction,
    output_prefix
):

    ascending = (
        direction == "lower"
    )

    ordered = (
        df[
            [
                "Gene_Name",
                value_col
            ]
        ]
        .sort_values(
            by=[
                value_col,
                "Gene_Name"
            ],
            ascending=[
                ascending,
                True
            ],
            kind="mergesort"
        )
        .reset_index(
            drop=True
        )
    )

    N = len(ordered)

    ordered[
        f"{output_prefix}_rank"
    ] = np.arange(
        1,
        N + 1
    )

    ordered[
        f"{output_prefix}_rank_percentile"
    ] = (
        100.0 *
        (
            N -
            ordered[
                f"{output_prefix}_rank"
            ]
        )
        /
        (
            N - 1
        )
    )

    return ordered[
        [
            "Gene_Name",
            f"{output_prefix}_rank",
            f"{output_prefix}_rank_percentile"
        ]
    ]


# ==========================================================================================
# 4. SOURCE INTEGRITY
# ==========================================================================================

print("=" * 145)
print("ASD-CONTEXTFUSION — STAGE 9E.1 HISTORICALLY-UNSEEN TEMPORAL BENCHMARK")
print("=" * 145)

print("\n[1/12] SOURCE INTEGRITY")


assert DEEPND_FILE.exists()
assert DEEPND_LOCK.exists()
assert STAGE9C_DIR.exists()


with open(
    DEEPND_LOCK,
    "r",
    encoding="utf-8"
) as f:

    deep_lock = json.load(f)


current_hash = sha256_file(
    DEEPND_FILE
)


assert (
    current_hash
    ==
    deep_lock["source_sha256"]
)


print("DeepND SHA256:", current_hash)
print("DeepND source lock: MATCH [OK]")


# ==========================================================================================
# 5. WRITE PRESPECIFICATION BEFORE TEMPORAL COMPARATOR ANALYSIS
# ==========================================================================================

print("\n[2/12] WRITING STAGE 9E.1 PRESPECIFICATION")


PRESPEC_FILE = (
    ADMIN_DIR /
    "ASD_ContextFusion_STAGE9E1_HISTORICAL_TEMPORAL_BENCHMARK_PRESPECIFICATION.json"
)


prespec = {

    "project":
        "ASD-ContextFusion",

    "stage":
        "9E.1_historical_temporal_comparator_benchmark",

    "source_file":
        str(DEEPND_FILE),

    "source_sha256":
        current_hash,

    "our_models":
        OUR_MODELS,

    "historical_comparators":
        COMPARATORS,

    "primary_historical_unseen_definition":
        "DeepND ASD Evidence Level == '-'",

    "rationale":
        (
            "Genes labeled positive or negative in the historical "
            "DeepND benchmark are excluded from the primary temporal "
            "candidate universe so that historical training labels "
            "cannot count as novel temporal successes."
        ),

    "missing_comparator_scores":
        "Never imputed; comparator-specific common universe used.",

    "pairwise_fairness":
        (
            "Our model and comparator are reranked within exactly "
            "the same comparator-specific historically-unlabeled universe."
        ),

    "rank_percentile":
        "100*(N-rank)/(N-1), best=100, worst=0",

    "top_thresholds":
        [
            0.01,
            0.05,
            0.10
        ],

    "bootstrap_replicates":
        BOOT_REPS,

    "bootstrap_seed":
        BOOT_SEED,

    "primary_pair":
        "Fusion_vs_DeepND",

    "supporting_pair":
        "Context_vs_DeepND",

    "secondary_comparators":
        [
            "Krishnan",
            "DAWN",
            "DAMAGES",
            "pLI"
        ],

    "no_model_training":
        True,

    "no_model_tuning":
        True,

    "no_new_PU_sampling":
        True,

    "no_external_gene_reselection":
        True,

    "no_performance_based_comparator_selection":
        True
}


with open(
    PRESPEC_FILE,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        prespec,
        f,
        indent=2
    )


print("[OK]", PRESPEC_FILE)


# ==========================================================================================
# 6. LOCATE FROZEN STAGE 9C NON-SEED RANK TABLE
# ==========================================================================================

print("\n[3/12] LOCATING FROZEN STAGE 9C RANK TABLE")


rank_candidates = []


for p in STAGE9C_DIR.rglob("*.tsv*"):

    try:

        head = pd.read_csv(
            p,
            sep="\t",
            nrows=5
        )

    except Exception:

        continue


    cols_lower = [
        str(c).lower()
        for c in head.columns
    ]


    if (
        "gene_name" in cols_lower
        and any(
            "context" in c
            and "rank" in c
            for c in cols_lower
        )
        and any(
            "fusion" in c
            and "rank" in c
            for c in cols_lower
        )
    ):

        try:

            tmp = pd.read_csv(
                p,
                sep="\t"
            )

        except Exception:

            continue


        if (
            len(tmp) == N_EXPECTED_NONSEED
            and "Gene_Name" in tmp.columns
            and tmp["Gene_Name"].is_unique
        ):

            rank_candidates.append(
                (
                    p,
                    tmp
                )
            )


assert len(rank_candidates) == 1, (
    "Expected exactly one Stage9C 18,062-gene rank table.\n"
    f"Found: {[str(x[0]) for x in rank_candidates]}"
)


OUR_RANK_FILE, our_ranks = rank_candidates[0]


print("Frozen rank file:", OUR_RANK_FILE)
print("Rows            :", len(our_ranks))


CONTEXT_RANK_COL = resolve_rank_column(
    our_ranks.columns,
    "Context"
)

FUSION_RANK_COL = resolve_rank_column(
    our_ranks.columns,
    "Fusion"
)


print("Context order column:", CONTEXT_RANK_COL)
print("Fusion order column :", FUSION_RANK_COL)


# ==========================================================================================
# 7. LOAD DEEPND ASD SHEET + IDENTITY HARMONIZATION
# ==========================================================================================

print("\n[4/12] LOADING HISTORICAL COMPARATOR TABLE")


deep = pd.read_excel(
    DEEPND_FILE,
    sheet_name="DeepND ASD Risk Ranking",
    header=0,
    engine="openpyxl"
)


required_cols = [
    "Gene Name",
    "Probability",
    "Krishnan et al. Ranking",
    "Dawn Ranking",
    "DAMAGES Ranking",
    "Evidence Level",
    "pLI"
]


for col in required_cols:

    assert col in deep.columns, col


deep["Gene Name"] = (
    deep["Gene Name"]
    .astype(str)
    .str.strip()
)


deep["Evidence Level"] = (
    deep["Evidence Level"]
    .astype(str)
    .str.strip()
)


print("DeepND rows:", len(deep))


print("\nEvidence Level counts:")

print(
    deep[
        "Evidence Level"
    ]
    .value_counts(
        dropna=False
    )
    .to_string()
)


# ------------------------------------------------------------------
# Locked Stage9B identity aliases:
# Source/current symbol -> project scoring identity.
# These were established BEFORE model-rank inspection in Stage9B.
# ------------------------------------------------------------------

LOCKED_ALIAS_TO_PROJECT = {

    "ADGRB1": "BAI1",
    "BICRA": "GLTSCR1",
    "EMSY": "C11orf30",
    "ICE1": "KIAA0947",
    "KMT5B": "SUV420H1",
    "NSD2": "WHSC1",
    "SRPRA": "SRPR"
}


project_gene_set = set(
    our_ranks[
        "Gene_Name"
    ]
    .astype(str)
    .str.strip()
)


def map_to_project_identity(gene):

    if gene in project_gene_set:
        return gene

    mapped = LOCKED_ALIAS_TO_PROJECT.get(
        gene,
        gene
    )

    return mapped


deep[
    "Project_Gene"
] = deep[
    "Gene Name"
].map(
    map_to_project_identity
)


# Deterministic identity collision handling:
# Prefer a row whose original DeepND symbol exactly equals project identity.
deep[
    "_exact_project_symbol"
] = (
    deep[
        "Gene Name"
    ]
    ==
    deep[
        "Project_Gene"
    ]
).astype(int)


deep = (
    deep
    .sort_values(
        [
            "Project_Gene",
            "_exact_project_symbol",
            "Gene Name"
        ],
        ascending=[
            True,
            False,
            True
        ],
        kind="mergesort"
    )
    .drop_duplicates(
        subset=[
            "Project_Gene"
        ],
        keep="first"
    )
    .reset_index(
        drop=True
    )
)


for comp, cfg in COMPARATORS.items():

    deep[
        f"{comp}_value"
    ] = numeric_clean(
        deep[
            cfg[
                "column"
            ]
        ]
    )


# ==========================================================================================
# 8. HISTORICALLY UNLABELED UNIVERSE
# ==========================================================================================

print("\n[5/12] DEFINING HISTORICALLY-UNLABELED CANDIDATE UNIVERSE")


deep[
    "historically_unlabeled"
] = (
    deep[
        "Evidence Level"
    ]
    ==
    "-"
)


deep_unlabeled = deep.loc[
    deep[
        "historically_unlabeled"
    ]
].copy()


base = (
    our_ranks[
        [
            "Gene_Name",
            CONTEXT_RANK_COL,
            FUSION_RANK_COL
        ]
    ]
    .merge(
        deep_unlabeled,
        left_on="Gene_Name",
        right_on="Project_Gene",
        how="inner",
        validate="one_to_one"
    )
)


print(
    "Our non-seed universe:",
    len(our_ranks)
)

print(
    "DeepND historically unlabeled rows:",
    len(deep_unlabeled)
)

print(
    "Historically unlabeled ∩ project non-seed:",
    len(base)
)


assert len(base) > 10000, (
    "Unexpectedly small historical common universe."
)


# ==========================================================================================
# 9. LOAD EXTERNAL SETS
# ==========================================================================================

print("\n[6/12] LOADING LOCKED TAN EXTERNAL SETS")


external_sets = {}


for set_name, cfg in SET_CONFIG.items():

    ext = pd.read_csv(
        cfg["file"],
        sep="\t"
    )


    assert len(ext) == cfg["expected_n"]
    assert "project_scoring_gene" in ext.columns
    assert ext["project_scoring_gene"].is_unique


    external_sets[
        set_name
    ] = set(
        ext[
            "project_scoring_gene"
        ]
        .astype(str)
        .str.strip()
    )


    print(
        f"{set_name:45s}",
        len(
            external_sets[
                set_name
            ]
        )
    )


# ==========================================================================================
# 10. COMPARATOR-SPECIFIC FAIR RANKING UNIVERSES
# ==========================================================================================

print("\n[7/12] BUILDING COMPARATOR-SPECIFIC FAIR UNIVERSES")


coverage_rows = []
rank_detail_frames = []


for comparator, cfg in COMPARATORS.items():

    comp_value_col = (
        f"{comparator}_value"
    )


    pair = base.loc[
        base[
            comp_value_col
        ].notna()
    ].copy()


    N = len(pair)


    assert N >= 1000, (
        f"Comparator {comparator} has unexpectedly small coverage: {N}"
    )


    # ------------------------------------------------------------------
    # Re-rank Context inside identical comparator-specific universe.
    # Original Context rank is monotonic with frozen Context probability.
    # ------------------------------------------------------------------

    context_tmp = pair[
        [
            "Gene_Name",
            CONTEXT_RANK_COL
        ]
    ].copy()


    context_tmp = (
        context_tmp
        .sort_values(
            [
                CONTEXT_RANK_COL,
                "Gene_Name"
            ],
            ascending=[
                True,
                True
            ],
            kind="mergesort"
        )
        .reset_index(
            drop=True
        )
    )


    context_tmp[
        "Context_pair_rank"
    ] = np.arange(
        1,
        N + 1
    )


    context_tmp[
        "Context_pair_rank_percentile"
    ] = (
        100.0 *
        (
            N -
            context_tmp[
                "Context_pair_rank"
            ]
        )
        /
        (
            N - 1
        )
    )


    # ------------------------------------------------------------------
    # Re-rank Fusion inside identical comparator-specific universe.
    # ------------------------------------------------------------------

    fusion_tmp = pair[
        [
            "Gene_Name",
            FUSION_RANK_COL
        ]
    ].copy()


    fusion_tmp = (
        fusion_tmp
        .sort_values(
            [
                FUSION_RANK_COL,
                "Gene_Name"
            ],
            ascending=[
                True,
                True
            ],
            kind="mergesort"
        )
        .reset_index(
            drop=True
        )
    )


    fusion_tmp[
        "Fusion_pair_rank"
    ] = np.arange(
        1,
        N + 1
    )


    fusion_tmp[
        "Fusion_pair_rank_percentile"
    ] = (
        100.0 *
        (
            N -
            fusion_tmp[
                "Fusion_pair_rank"
            ]
        )
        /
        (
            N - 1
        )
    )


    # ------------------------------------------------------------------
    # Comparator ranking.
    # ------------------------------------------------------------------

    comparator_tmp = rerank_percentile(
        pair.rename(
            columns={
                comp_value_col:
                    "_comparator_value"
            }
        ),
        value_col="_comparator_value",
        direction=cfg["direction"],
        output_prefix=comparator
    )


    ranked = (
        pair[
            [
                "Gene_Name",
                "Evidence Level",
                comp_value_col
            ]
        ]
        .merge(
            context_tmp[
                [
                    "Gene_Name",
                    "Context_pair_rank",
                    "Context_pair_rank_percentile"
                ]
            ],
            on="Gene_Name",
            how="inner",
            validate="one_to_one"
        )
        .merge(
            fusion_tmp[
                [
                    "Gene_Name",
                    "Fusion_pair_rank",
                    "Fusion_pair_rank_percentile"
                ]
            ],
            on="Gene_Name",
            how="inner",
            validate="one_to_one"
        )
        .merge(
            comparator_tmp,
            on="Gene_Name",
            how="inner",
            validate="one_to_one"
        )
    )


    ranked[
        "comparator"
    ] = comparator


    ranked[
        "comparison_universe_N"
    ] = N


    rank_detail_frames.append(
        ranked
    )


    for set_name, gene_set in external_sets.items():

        n_available = int(
            ranked[
                "Gene_Name"
            ].isin(
                gene_set
            ).sum()
        )


        coverage_rows.append(
            {
                "comparator":
                    comparator,

                "comparison_universe_N":
                    N,

                "external_set":
                    set_name,

                "locked_external_n":
                    len(gene_set),

                "historically_unlabeled_and_comparator_evaluable_n":
                    n_available,

                "external_evaluable_fraction":
                    (
                        n_available
                        /
                        len(gene_set)
                    )
            }
        )


rank_details = pd.concat(
    rank_detail_frames,
    ignore_index=True
)


coverage = pd.DataFrame(
    coverage_rows
)


print(
    coverage.to_string(
        index=False
    )
)


# ==========================================================================================
# 11. CONTINUOUS PERFORMANCE + PAIRED BOOTSTRAP
# ==========================================================================================

print("\n[8/12] CONTINUOUS TEMPORAL PERFORMANCE + PAIRED BOOTSTRAP")


summary_rows = []
bootstrap_rows = []
bootstrap_rep_rows = []


for comparator, cfg in COMPARATORS.items():

    ranked = rank_details.loc[
        rank_details[
            "comparator"
        ] == comparator
    ].copy()


    comp_pct_col = (
        f"{comparator}_rank_percentile"
    )


    for set_index, (
        set_name,
        gene_set
    ) in enumerate(
        external_sets.items()
    ):

        ext = ranked.loc[
            ranked[
                "Gene_Name"
            ].isin(
                gene_set
            )
        ].copy()


        if len(ext) == 0:
            continue


        # --------------------------------------------------------------
        # Descriptive performance.
        # --------------------------------------------------------------

        model_pct_cols = {

            "Context":
                "Context_pair_rank_percentile",

            "Fusion":
                "Fusion_pair_rank_percentile",

            comparator:
                comp_pct_col
        }


        for model, pct_col in model_pct_cols.items():

            values = ext[
                pct_col
            ].to_numpy(
                dtype=float
            )


            summary_rows.append(
                {
                    "comparator_framework":
                        comparator,

                    "external_set":
                        set_name,

                    "n_evaluable_external":
                        len(ext),

                    "comparison_universe_N":
                        int(
                            ext[
                                "comparison_universe_N"
                            ].iloc[0]
                        ),

                    "model":
                        model,

                    "mean_rank_percentile":
                        float(
                            np.mean(
                                values
                            )
                        ),

                    "median_rank_percentile":
                        float(
                            np.median(
                                values
                            )
                        )
                }
            )


        # --------------------------------------------------------------
        # Paired bootstrap: Context/Fusion minus comparator.
        # --------------------------------------------------------------

        for model_index, model in enumerate(
            OUR_MODELS
        ):

            our_col = (
                f"{model}_pair_rank_percentile"
            )


            delta = (
                ext[
                    our_col
                ].to_numpy(
                    dtype=float
                )
                -
                ext[
                    comp_pct_col
                ].to_numpy(
                    dtype=float
                )
            )


            observed_mean = float(
                np.mean(
                    delta
                )
            )


            observed_median = float(
                np.median(
                    delta
                )
            )


            n = len(delta)


            boot_means = np.empty(
                BOOT_REPS,
                dtype=float
            )


            comp_index = list(
                COMPARATORS.keys()
            ).index(
                comparator
            )


            for b in range(
                BOOT_REPS
            ):

                rng = np.random.default_rng(
                    np.random.SeedSequence(
                        [
                            BOOT_SEED,
                            comp_index,
                            set_index,
                            model_index,
                            b
                        ]
                    )
                )


                idx = rng.integers(
                    0,
                    n,
                    size=n
                )


                value = float(
                    np.mean(
                        delta[
                            idx
                        ]
                    )
                )


                boot_means[b] = value


                bootstrap_rep_rows.append(
                    {
                        "comparator":
                            comparator,

                        "external_set":
                            set_name,

                        "our_model":
                            model,

                        "replicate":
                            b,

                        "mean_rank_percentile_delta":
                            value
                    }
                )


            ci_low = float(
                np.quantile(
                    boot_means,
                    0.025
                )
            )


            ci_high = float(
                np.quantile(
                    boot_means,
                    0.975
                )
            )


            bootstrap_rows.append(
                {
                    "comparator":
                        comparator,

                    "external_set":
                        set_name,

                    "our_model":
                        model,

                    "comparison":
                        f"{model}_minus_{comparator}",

                    "n_evaluable_external":
                        n,

                    "observed_mean_rank_percentile_delta":
                        observed_mean,

                    "observed_median_rank_percentile_delta":
                        observed_median,

                    "CI95_low":
                        ci_low,

                    "CI95_high":
                        ci_high,

                    "bootstrap_fraction_delta_gt_0":
                        float(
                            np.mean(
                                boot_means > 0
                            )
                        ),

                    "CI_excludes_zero":
                        bool(
                            ci_low > 0
                            or
                            ci_high < 0
                        )
                }
            )


continuous_summary = pd.DataFrame(
    summary_rows
)


bootstrap_summary = pd.DataFrame(
    bootstrap_rows
)


bootstrap_replicates = pd.DataFrame(
    bootstrap_rep_rows
)


# ==========================================================================================
# 12. TOP 1 / 5 / 10% RECOVERY
# ==========================================================================================

print("\n[9/12] TOP-K TEMPORAL RECOVERY")


topk_rows = []


for comparator in COMPARATORS.keys():

    ranked = rank_details.loc[
        rank_details[
            "comparator"
        ] == comparator
    ].copy()


    comp_rank_col = (
        f"{comparator}_rank"
    )


    N = int(
        ranked[
            "comparison_universe_N"
        ].iloc[0]
    )


    for set_name, gene_set in external_sets.items():

        ext = ranked.loc[
            ranked[
                "Gene_Name"
            ].isin(
                gene_set
            )
        ].copy()


        M = len(ext)


        if M == 0:
            continue


        models_rank_cols = {

            "Context":
                "Context_pair_rank",

            "Fusion":
                "Fusion_pair_rank",

            comparator:
                comp_rank_col
        }


        for pct in [
            0.01,
            0.05,
            0.10
        ]:

            K = int(
                math.ceil(
                    N * pct
                )
            )


            for model, rank_col in models_rank_cols.items():

                recovered = int(
                    (
                        ext[
                            rank_col
                        ]
                        <= K
                    ).sum()
                )


                recovery_fraction = (
                    recovered /
                    M
                )


                expected_fraction = (
                    K /
                    N
                )


                enrichment = (
                    recovery_fraction
                    /
                    expected_fraction
                )


                p_hyper = float(
                    hypergeom.sf(
                        recovered - 1,
                        N,
                        M,
                        K
                    )
                )


                topk_rows.append(
                    {
                        "comparator_framework":
                            comparator,

                        "external_set":
                            set_name,

                        "model":
                            model,

                        "comparison_universe_N":
                            N,

                        "external_M":
                            M,

                        "threshold_percent":
                            int(
                                pct * 100
                            ),

                        "K":
                            K,

                        "recovered":
                            recovered,

                        "recovery_fraction":
                            recovery_fraction,

                        "enrichment_fold":
                            enrichment,

                        "hypergeometric_p":
                            p_hyper
                    }
                )


topk = pd.DataFrame(
    topk_rows
)


topk[
    "BH_FDR_q_all_prespecified_enrichment_tests"
] = bh_adjust(
    topk[
        "hypergeometric_p"
    ].to_numpy()
)


# ==========================================================================================
# 13. DIRECT RECOVERY DELTA: OUR MODEL MINUS COMPARATOR
# ==========================================================================================

recovery_delta_rows = []


for comparator in COMPARATORS.keys():

    for set_name in external_sets.keys():

        for threshold in [
            1,
            5,
            10
        ]:

            sub = topk.loc[
                (
                    topk[
                        "comparator_framework"
                    ]
                    ==
                    comparator
                )
                &
                (
                    topk[
                        "external_set"
                    ]
                    ==
                    set_name
                )
                &
                (
                    topk[
                        "threshold_percent"
                    ]
                    ==
                    threshold
                )
            ]


            if len(sub) == 0:
                continue


            comp_row = sub.loc[
                sub[
                    "model"
                ]
                ==
                comparator
            ]


            if len(comp_row) != 1:
                continue


            comp_row = comp_row.iloc[0]


            for our_model in OUR_MODELS:

                our_row = sub.loc[
                    sub[
                        "model"
                    ]
                    ==
                    our_model
                ]


                if len(our_row) != 1:
                    continue


                our_row = our_row.iloc[0]


                recovery_delta_rows.append(
                    {
                        "comparator":
                            comparator,

                        "external_set":
                            set_name,

                        "threshold_percent":
                            threshold,

                        "our_model":
                            our_model,

                        "our_recovered":
                            int(
                                our_row[
                                    "recovered"
                                ]
                            ),

                        "comparator_recovered":
                            int(
                                comp_row[
                                    "recovered"
                                ]
                            ),

                        "our_minus_comparator_recovered":
                            int(
                                our_row[
                                    "recovered"
                                ]
                                -
                                comp_row[
                                    "recovered"
                                ]
                            ),

                        "our_enrichment_fold":
                            float(
                                our_row[
                                    "enrichment_fold"
                                ]
                            ),

                        "comparator_enrichment_fold":
                            float(
                                comp_row[
                                    "enrichment_fold"
                                ]
                            )
                    }
                )


recovery_deltas = pd.DataFrame(
    recovery_delta_rows
)


# ==========================================================================================
# 14. SAVE
# ==========================================================================================

print("\n[10/12] SAVING STAGE 9E.1 RESULTS")


files_to_save = {

    "Stage9E1_historical_comparator_coverage.tsv":
        coverage,

    "Stage9E1_continuous_rank_percentile_summary.tsv":
        continuous_summary,

    "Stage9E1_paired_bootstrap_summary.tsv":
        bootstrap_summary,

    "Stage9E1_topk_enrichment.tsv":
        topk,

    "Stage9E1_topk_recovery_deltas.tsv":
        recovery_deltas,

    "Stage9E1_rank_details.tsv.gz":
        rank_details
}


saved_files = []


for filename, df in files_to_save.items():

    path = OUT_DIR / filename


    if str(path).endswith(
        ".gz"
    ):

        df.to_csv(
            path,
            sep="\t",
            index=False,
            compression="gzip"
        )

    else:

        df.to_csv(
            path,
            sep="\t",
            index=False
        )


    saved_files.append(
        path
    )


BOOT_REP_FILE = (
    OUT_DIR /
    "Stage9E1_bootstrap_replicates.tsv.gz"
)


bootstrap_replicates.to_csv(
    BOOT_REP_FILE,
    sep="\t",
    index=False,
    compression="gzip"
)


saved_files.append(
    BOOT_REP_FILE
)


# ==========================================================================================
# 15. FINAL MANIFEST
# ==========================================================================================

print("\n[11/12] WRITING FINAL MANIFEST")


MANIFEST_FILE = (
    OUT_DIR /
    "Stage9E1_manifest.json"
)


manifest = {

    "project":
        "ASD-ContextFusion",

    "stage":
        "9E.1",

    "analysis":
        "Historically-unseen temporal benchmark",

    "DeepND_source":
        str(DEEPND_FILE),

    "DeepND_SHA256":
        current_hash,

    "our_frozen_rank_file":
        str(OUR_RANK_FILE),

    "our_frozen_rank_SHA256":
        sha256_file(
            OUR_RANK_FILE
        ),

    "historical_unseen_rule":
        "DeepND Evidence Level == '-'",

    "comparators":
        COMPARATORS,

    "our_models":
        OUR_MODELS,

    "bootstrap_replicates":
        BOOT_REPS,

    "bootstrap_seed":
        BOOT_SEED,

    "model_fitting":
        False,

    "model_tuning":
        False,

    "PU_resampling":
        False,

    "external_gene_reselection":
        False,

    "missing_score_imputation":
        False,

    "result_files":
        [
            {
                "path":
                    str(p),

                "sha256":
                    sha256_file(p)
            }
            for p in saved_files
        ]
}


with open(
    MANIFEST_FILE,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        manifest,
        f,
        indent=2
    )


# ==========================================================================================
# 16. FINAL REPORT
# ==========================================================================================

print("\n[12/12] FINAL REPORT")

print("\n" + "=" * 145)
print("STAGE 9E.1 HISTORICALLY-UNSEEN TEMPORAL BENCHMARK COMPLETE [OK]")
print("=" * 145)


print("\nA. COMPARATOR COVERAGE / HISTORICALLY-UNSEEN EXTERNAL SET SIZES")

print(
    coverage.to_string(
        index=False
    )
)


print("\nB. MEAN / MEDIAN RANK-PERCENTILE PERFORMANCE")

print(
    continuous_summary[
        [
            "comparator_framework",
            "external_set",
            "n_evaluable_external",
            "model",
            "mean_rank_percentile",
            "median_rank_percentile"
        ]
    ].to_string(
        index=False
    )
)


print("\nC. PAIRED BOOTSTRAP — OUR MODEL MINUS HISTORICAL COMPARATOR")

print(
    bootstrap_summary[
        [
            "comparator",
            "external_set",
            "our_model",
            "n_evaluable_external",
            "observed_mean_rank_percentile_delta",
            "CI95_low",
            "CI95_high",
            "bootstrap_fraction_delta_gt_0",
            "CI_excludes_zero"
        ]
    ].to_string(
        index=False
    )
)


print("\nD. TOP-K RECOVERY DELTAS — OUR MODEL MINUS COMPARATOR")

print(
    recovery_deltas.to_string(
        index=False
    )
)


print("\nINTEGRITY")

print("Model fitting                    : NO")
print("7,900-fold rerun                 : NO")
print("Hyperparameter tuning            : NO")
print("New PU sampling                  : NO")
print("External gene reselection        : NO")
print("Missing comparator score imputed : NO")
print("Historical labels counted        : NO")
print("Primary historical-unseen rule   : Evidence Level == '-'")


print("\nSAVED TO:")
print(OUT_DIR)


print("\nPASTE BACK THESE FOUR SECTIONS:")
print("A. COMPARATOR COVERAGE / HISTORICALLY-UNSEEN EXTERNAL SET SIZES")
print("B. MEAN / MEDIAN RANK-PERCENTILE PERFORMANCE")
print("C. PAIRED BOOTSTRAP — OUR MODEL MINUS HISTORICAL COMPARATOR")
print("D. TOP-K RECOVERY DELTAS — OUR MODEL MINUS COMPARATOR")

print("=" * 145)
```

```python
# ==========================================================================================
# ASD-ContextFusion
# STAGE 10A + 10B
# PUBLICATION-READY TABLES + FIGURES
#
# CURRENT PROJECT STATE:
#   Stage 9E.1 COMPLETE
#
# PURPOSE
# -------
# Assemble all frozen results into:
#
#   MAIN TABLES
#   Table 1  - Study/framework characteristics
#   Table 2  - Primary internal five-model performance
#   Table 3  - Paired inference and disease-context robustness
#   Table 4  - 2026 temporal external validation
#   Table 5  - External NT-v2 vs sequence controls
#   Table 6  - DeepND-unlabeled historical comparator benchmark
#
#   MAIN FIGURES
#   Figure 1 - ASD-ContextFusion framework schematic
#   Figure 2 - Internal five-model performance
#   Figure 3 - Paired complementarity / robustness forest plot
#   Figure 4 - 2026 temporal external validation
#   Figure 5 - External NT-v2 vs sequence controls
#   Figure 6 - Historical comparator temporal benchmark
#
# OUTPUT FORMATS
# --------------
# Figures:
#   PDF       = vector
#   TIFF      = 600 DPI
#   PNG       = 300 DPI
#
# Tables:
#   XLSX formatted workbook
#   TSV source tables
#
# IMPORTANT
# ---------
# NO model fitting
# NO XGBoost
# NO logistic regression
# NO 7,900-fold rerun
# NO tuning
# NO PU resampling
# NO threshold changes
# NO gene reselection
# NO result-driven model changes
#
# Engineering/output assembly only.
# Scientific behavior unchanged.
# ==========================================================================================


# ==========================================================================================
# 0. IMPORTS
# ==========================================================================================

from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import FancyBboxPatch, FancyArrowPatch
import hashlib
import json
import textwrap
import warnings

from openpyxl import load_workbook
from openpyxl.styles import (
    Font,
    PatternFill,
    Border,
    Side,
    Alignment
)
from openpyxl.utils import get_column_letter


warnings.filterwarnings("ignore", category=UserWarning)


# ==========================================================================================
# 1. PATHS
# ==========================================================================================

ROOT = Path(r"E:\ASD_ContextFusion_Q1")

STAGE9D_DIR = (
    ROOT /
    "05_results" /
    "proposed" /
    "stage9D_external_sequence_controls"
)

STAGE9E_DIR = (
    ROOT /
    "05_results" /
    "proposed" /
    "stage9E_historical_comparator_benchmark"
)

OUT_ROOT = (
    ROOT /
    "06_publication" /
    "stage10_tables_figures"
)

TABLE_DIR = OUT_ROOT / "tables"
FIGURE_DIR = OUT_ROOT / "figures"
SOURCE_DIR = OUT_ROOT / "figure_source_data"

for d in [
    OUT_ROOT,
    TABLE_DIR,
    FIGURE_DIR,
    SOURCE_DIR
]:
    d.mkdir(
        parents=True,
        exist_ok=True
    )


# ==========================================================================================
# 2. REQUIRED FROZEN INPUTS
# ==========================================================================================

STAGE9D_RANK = (
    STAGE9D_DIR /
    "Stage9D_external_sequence_rank_summary.tsv"
)

STAGE9D_BOOT = (
    STAGE9D_DIR /
    "Stage9D_NTv2_vs_controls_bootstrap_summary.tsv"
)

STAGE9D_TOPK = (
    STAGE9D_DIR /
    "Stage9D_external_sequence_topk_enrichment.tsv"
)

STAGE9D_TOPK_DELTA = (
    STAGE9D_DIR /
    "Stage9D_NTv2_vs_controls_topk_deltas.tsv"
)


STAGE9E_COVERAGE = (
    STAGE9E_DIR /
    "Stage9E1_historical_comparator_coverage.tsv"
)

STAGE9E_CONT = (
    STAGE9E_DIR /
    "Stage9E1_continuous_rank_percentile_summary.tsv"
)

STAGE9E_BOOT = (
    STAGE9E_DIR /
    "Stage9E1_paired_bootstrap_summary.tsv"
)

STAGE9E_TOPK = (
    STAGE9E_DIR /
    "Stage9E1_topk_enrichment.tsv"
)

STAGE9E_DELTA = (
    STAGE9E_DIR /
    "Stage9E1_topk_recovery_deltas.tsv"
)


required_files = [
    STAGE9D_RANK,
    STAGE9D_BOOT,
    STAGE9D_TOPK,
    STAGE9D_TOPK_DELTA,
    STAGE9E_COVERAGE,
    STAGE9E_CONT,
    STAGE9E_BOOT,
    STAGE9E_TOPK,
    STAGE9E_DELTA
]


print("=" * 150)
print("ASD-CONTEXTFUSION — STAGE 10A + 10B PUBLICATION ASSEMBLY")
print("=" * 150)

print("\n[1/14] INPUT INTEGRITY")


for p in required_files:

    assert p.exists(), (
        f"Required frozen result file missing:\n{p}"
    )

    print(
        "[OK]",
        p
    )


# ==========================================================================================
# 3. HELPERS
# ==========================================================================================

def sha256_file(path, chunk_size=1024 * 1024):

    h = hashlib.sha256()

    with open(path, "rb") as f:

        while True:

            chunk = f.read(chunk_size)

            if not chunk:
                break

            h.update(chunk)

    return h.hexdigest()


def safe_float(x):

    if pd.isna(x):
        return np.nan

    return float(x)


def save_pub_figure(fig, stem):

    pdf_path = FIGURE_DIR / f"{stem}.pdf"
    png_path = FIGURE_DIR / f"{stem}.png"
    tif_path = FIGURE_DIR / f"{stem}.tiff"

    fig.savefig(
        pdf_path,
        bbox_inches="tight"
    )

    fig.savefig(
        png_path,
        dpi=300,
        bbox_inches="tight"
    )

    try:

        fig.savefig(
            tif_path,
            dpi=600,
            bbox_inches="tight",
            pil_kwargs={
                "compression": "tiff_lzw"
            }
        )

    except Exception:

        fig.savefig(
            tif_path,
            dpi=600,
            bbox_inches="tight"
        )

    plt.close(fig)

    return [
        pdf_path,
        png_path,
        tif_path
    ]


def grouped_bar(
    ax,
    categories,
    series_dict,
    ylabel,
    ylim=None
):

    n_series = len(series_dict)

    x = np.arange(
        len(categories)
    )

    total_width = 0.78

    width = (
        total_width /
        n_series
    )

    offsets = (
        np.arange(n_series)
        -
        (n_series - 1) / 2
    ) * width

    for offset, (
        label,
        values
    ) in zip(
        offsets,
        series_dict.items()
    ):

        ax.bar(
            x + offset,
            values,
            width=width * 0.92,
            label=label
        )

    ax.set_xticks(x)
    ax.set_xticklabels(categories)
    ax.set_ylabel(ylabel)

    if ylim is not None:
        ax.set_ylim(*ylim)

    ax.legend(
        frameon=False
    )

    ax.spines[
        "top"
    ].set_visible(False)

    ax.spines[
        "right"
    ].set_visible(False)


# ==========================================================================================
# 4. LOAD STAGE 9D + 9E FROZEN RESULTS
# ==========================================================================================

print("\n[2/14] LOADING STAGE 9D / 9E FROZEN RESULTS")


seq_rank = pd.read_csv(
    STAGE9D_RANK,
    sep="\t"
)

seq_boot = pd.read_csv(
    STAGE9D_BOOT,
    sep="\t"
)

seq_topk = pd.read_csv(
    STAGE9D_TOPK,
    sep="\t"
)

seq_delta = pd.read_csv(
    STAGE9D_TOPK_DELTA,
    sep="\t"
)


hist_cov = pd.read_csv(
    STAGE9E_COVERAGE,
    sep="\t"
)

hist_cont = pd.read_csv(
    STAGE9E_CONT,
    sep="\t"
)

hist_boot = pd.read_csv(
    STAGE9E_BOOT,
    sep="\t"
)

hist_topk = pd.read_csv(
    STAGE9E_TOPK,
    sep="\t"
)

hist_delta = pd.read_csv(
    STAGE9E_DELTA,
    sep="\t"
)


assert set(
    seq_rank["model"]
) >= {
    "NTv2",
    "LengthGC",
    "Fourmer"
}


assert set(
    hist_cov["comparator"]
) >= {
    "DeepND",
    "Krishnan",
    "DAWN",
    "DAMAGES",
    "pLI"
}


print("Stage 9D rows:", len(seq_rank))
print("Stage 9E continuous rows:", len(hist_cont))


# ==========================================================================================
# 5. DISPLAY LABELS
# ==========================================================================================

SET_LABELS = {

    "Tan_All245_nonseed":
        "Broad non-seed (n=134)",

    "Tan_Suggestive46_nonseed":
        "Suggestive (n=45)",

    "Tan_NoEvidence32_nonseed":
        "No-evidence (n=30)",

    "Tan_SuggestiveOrNoEvidence78_nonseed":
        "Suggestive/no-evidence (n=75)"
}


SET_SHORT = {

    "Tan_All245_nonseed":
        "Broad134",

    "Tan_Suggestive46_nonseed":
        "Suggestive45",

    "Tan_NoEvidence32_nonseed":
        "NoEvidence30",

    "Tan_SuggestiveOrNoEvidence78_nonseed":
        "Combined75"
}


SEQ_MODEL_LABELS = {

    "NTv2":
        "NT-v2",

    "LengthGC":
        "Length+GC",

    "Fourmer":
        "4-mer"
}


# ==========================================================================================
# 6. TABLE 1 — FRAMEWORK CHARACTERISTICS
# ==========================================================================================

print("\n[3/14] BUILDING MAIN TABLES")


table1 = pd.DataFrame(
    [
        [
            "Project gene universe",
            "18,252 genes",
            "Exact multimodal project universe"
        ],
        [
            "Locked ASD positive seeds",
            "190 genes",
            "Monoallelic ASD seed set"
        ],
        [
            "Non-seed ranking universe",
            "18,062 genes",
            "Used for temporal ranking"
        ],
        [
            "Context predictors",
            "316",
            "Frozen Dhindsa-derived contextual predictors"
        ],
        [
            "Generic context predictors",
            "266",
            "Disease-context-reduced sensitivity set"
        ],
        [
            "Disease-filtered predictors",
            "50",
            "Excluded in generic-only robustness analysis"
        ],
        [
            "NT-v2 representation",
            "512 dimensions",
            "Full-transcript pretrained nucleotide representation"
        ],
        [
            "NT-v2 model",
            "nucleotide-transformer-v2-50m-multi-species",
            "Pretrained model; no fine-tuning"
        ],
        [
            "PU stochastic iterations",
            "10",
            "Frozen"
        ],
        [
            "Balanced PU datasets per iteration",
            "79",
            "Frozen"
        ],
        [
            "Outer folds per balanced dataset",
            "10",
            "Frozen"
        ],
        [
            "Total outer train/test splits",
            "7,900",
            "Not treated as independent inferential units"
        ],
        [
            "Primary external broad set",
            "134 non-seed genes",
            "Tan et al. 2026"
        ],
        [
            "Suggestive external set",
            "45 genes",
            "Project-representable non-seed subset"
        ],
        [
            "No-evidence external set",
            "30 genes",
            "Primary novelty-focused subset"
        ],
        [
            "Combined low-evidence set",
            "75 genes",
            "Suggestive or no-evidence"
        ]
    ],
    columns=[
        "Characteristic",
        "Value",
        "Definition / role"
    ]
)


# ==========================================================================================
# 7. TABLE 2 — PRIMARY INTERNAL FIVE-MODEL PERFORMANCE
# ==========================================================================================

table2 = pd.DataFrame(
    [
        [
            "Context",
            0.942180,
            0.037068,
            0.909122,
            0.064823,
            0.947076,
            0.182366
        ],
        [
            "NT-v2",
            0.862039,
            0.059357,
            0.808142,
            0.086136,
            0.899628,
            0.099593
        ],
        [
            "Length+GC",
            0.831980,
            0.064687,
            0.769483,
            0.092269,
            0.832049,
            0.052922
        ],
        [
            "4-mer",
            0.705966,
            0.085251,
            0.601583,
            0.101046,
            0.779947,
            0.039135
        ],
        [
            "Strict Fusion",
            0.944043,
            0.035867,
            0.912769,
            0.062831,
            0.951591,
            0.196525
        ]
    ],
    columns=[
        "Model",
        "Split AUROC",
        "Split AUROC SD",
        "Split AP",
        "Split AP SD",
        "Gene AUROC",
        "Gene AP"
    ]
)


# ==========================================================================================
# 8. TABLE 3 — PAIRED INTERNAL / ROBUSTNESS INFERENCE
# ==========================================================================================

table3 = pd.DataFrame(
    [
        [
            "Primary full-context",
            "Fusion − Context",
            "AP",
            0.014159,
            -0.012178,
            0.039833,
            False,
            0.001953,
            "Gene-level bootstrap CI crosses zero; sign-flip supportive"
        ],
        [
            "Primary full-context",
            "Fusion − Context",
            "AUROC",
            0.004515,
            0.000206,
            0.009281,
            True,
            0.001953,
            "Gene-level bootstrap CI excludes zero"
        ],
        [
            "Disease-context-reduced",
            "Generic Fusion − Generic Context",
            "AP",
            0.015343,
            -0.010172,
            0.040478,
            False,
            0.001953,
            "Directional AP gain"
        ],
        [
            "Disease-context-reduced",
            "Generic Fusion − Generic Context",
            "AUROC",
            0.006318,
            0.000715,
            0.013274,
            True,
            0.001953,
            "Positive paired bootstrap CI"
        ],
        [
            "Context robustness",
            "Generic Context − Full Context",
            "AP",
            -0.008050,
            -0.018187,
            0.001257,
            False,
            np.nan,
            "AP CI crosses zero"
        ],
        [
            "Context robustness",
            "Generic Context − Full Context",
            "AUROC",
            -0.003481,
            -0.006772,
            -0.001246,
            True,
            np.nan,
            "Full Context retains modest AUROC advantage"
        ],
        [
            "Fusion robustness",
            "Generic Fusion − Full Fusion",
            "AP",
            -0.006867,
            -0.014501,
            -0.000407,
            True,
            np.nan,
            "Full Fusion retains modest AP advantage"
        ],
        [
            "Fusion robustness",
            "Generic Fusion − Full Fusion",
            "AUROC",
            -0.001678,
            -0.003080,
            -0.000562,
            True,
            np.nan,
            "Full Fusion retains modest AUROC advantage"
        ],
        [
            "Difference-in-differences",
            "Generic incremental gain − Full incremental gain",
            "AP",
            0.001184,
            -0.006335,
            0.008569,
            False,
            np.nan,
            "Incremental AP gain stable to source reduction"
        ],
        [
            "Difference-in-differences",
            "Generic incremental gain − Full incremental gain",
            "AUROC",
            0.001803,
            0.000081,
            0.004646,
            True,
            np.nan,
            "Incremental AUROC gain slightly larger after source reduction"
        ]
    ],
    columns=[
        "Analysis",
        "Comparison",
        "Metric",
        "Delta",
        "95% CI low",
        "95% CI high",
        "95% CI excludes zero",
        "Supporting sign-flip p",
        "Interpretation"
    ]
)


# ==========================================================================================
# 9. TABLE 4 — STAGE 9C TEMPORAL EXTERNAL VALIDATION
# ==========================================================================================

stage9c_rows = [

    # Broad134
    [
        "Broad134",
        134,
        "Context",
        82.084263,
        92.029788,
        14,
        51,
        76
    ],
    [
        "Broad134",
        134,
        "NT-v2",
        74.754336,
        85.283207,
        12,
        31,
        51
    ],
    [
        "Broad134",
        134,
        "Strict Fusion",
        82.101700,
        91.974420,
        14,
        50,
        77
    ],

    # Suggestive45
    [
        "Suggestive45",
        45,
        "Context",
        78.004048,
        90.089142,
        6,
        16,
        23
    ],
    [
        "Suggestive45",
        45,
        "NT-v2",
        71.313758,
        78.633520,
        3,
        5,
        14
    ],
    [
        "Suggestive45",
        45,
        "Strict Fusion",
        79.242936,
        89.015005,
        3,
        14,
        22
    ],

    # NoEvidence30
    [
        "NoEvidence30",
        30,
        "Context",
        65.283207,
        72.255135,
        0,
        4,
        7
    ],
    [
        "NoEvidence30",
        30,
        "NT-v2",
        60.109444,
        68.996733,
        0,
        4,
        6
    ],
    [
        "NoEvidence30",
        30,
        "Strict Fusion",
        63.568462,
        74.707934,
        0,
        4,
        7
    ],

    # Combined75
    [
        "Combined75",
        75,
        "Context",
        72.915712,
        84.297658,
        6,
        20,
        30
    ],
    [
        "Combined75",
        75,
        "NT-v2",
        66.832032,
        75.499695,
        3,
        9,
        20
    ],
    [
        "Combined75",
        75,
        "Strict Fusion",
        72.973147,
        83.854715,
        3,
        18,
        29
    ]
]


table4 = pd.DataFrame(
    stage9c_rows,
    columns=[
        "External set",
        "N genes",
        "Model",
        "Mean rank percentile",
        "Median rank percentile",
        "Top 1% recovered",
        "Top 5% recovered",
        "Top 10% recovered"
    ]
)


table4b = pd.DataFrame(
    [
        [
            "Broad134",
            134,
            0.017437,
            -1.277108,
            1.400825,
            0.5136,
            False
        ],
        [
            "Suggestive45",
            45,
            1.238888,
            -1.344748,
            4.095239,
            0.8072,
            False
        ],
        [
            "NoEvidence30",
            30,
            -1.714744,
            -5.442380,
            2.643292,
            0.2020,
            False
        ],
        [
            "Combined75",
            75,
            0.057435,
            -2.208194,
            2.351433,
            0.5090,
            False
        ]
    ],
    columns=[
        "External set",
        "N genes",
        "Fusion − Context mean rank-percentile delta",
        "95% CI low",
        "95% CI high",
        "Bootstrap fraction > 0",
        "95% CI excludes zero"
    ]
)


# ==========================================================================================
# 10. TABLE 5 — STAGE 9D EXTERNAL SEQUENCE VALIDATION
# ==========================================================================================

seq_rank2 = seq_rank.copy()

seq_rank2[
    "External set"
] = seq_rank2[
    "external_set"
].map(
    SET_SHORT
)

seq_rank2[
    "Model display"
] = seq_rank2[
    "model"
].map(
    SEQ_MODEL_LABELS
)


seq_topk2 = seq_topk.copy()

seq_topk2[
    "External set"
] = seq_topk2[
    "external_set"
].map(
    SET_SHORT
)

seq_topk2[
    "Model display"
] = seq_topk2[
    "model"
].map(
    SEQ_MODEL_LABELS
)


topk_count_pivot = (
    seq_topk2
    .pivot_table(
        index=[
            "External set",
            "Model display"
        ],
        columns="threshold",
        values="recovered_genes",
        aggfunc="first"
    )
    .reset_index()
)


table5 = (
    seq_rank2[
        [
            "External set",
            "n_external_genes",
            "Model display",
            "mean_rank_percentile",
            "median_rank_percentile"
        ]
    ]
    .merge(
        topk_count_pivot,
        on=[
            "External set",
            "Model display"
        ],
        how="left",
        validate="one_to_one"
    )
    .rename(
        columns={
            "n_external_genes":
                "N genes",

            "Model display":
                "Model",

            "mean_rank_percentile":
                "Mean rank percentile",

            "median_rank_percentile":
                "Median rank percentile",

            "Top1pct":
                "Top 1% recovered",

            "Top5pct":
                "Top 5% recovered",

            "Top10pct":
                "Top 10% recovered"
        }
    )
)


table5b = (
    seq_boot[
        [
            "external_set",
            "comparison",
            "n_external_genes",
            "observed_mean_rank_percentile_delta",
            "CI95_low",
            "CI95_high",
            "bootstrap_fraction_delta_gt_0",
            "CI_excludes_zero"
        ]
    ]
    .copy()
)


table5b[
    "External set"
] = table5b[
    "external_set"
].map(
    SET_SHORT
)


table5b = (
    table5b
    .drop(
        columns=[
            "external_set"
        ]
    )
    .rename(
        columns={
            "comparison":
                "Comparison",

            "n_external_genes":
                "N genes",

            "observed_mean_rank_percentile_delta":
                "Mean rank-percentile delta",

            "CI95_low":
                "95% CI low",

            "CI95_high":
                "95% CI high",

            "bootstrap_fraction_delta_gt_0":
                "Bootstrap fraction > 0",

            "CI_excludes_zero":
                "95% CI excludes zero"
        }
    )
)


# ==========================================================================================
# 11. TABLE 6 — HISTORICAL COMPARATOR BENCHMARK
#     Main compact table = Broad set only
# ==========================================================================================

comparators = [
    "DeepND",
    "Krishnan",
    "DAWN",
    "DAMAGES",
    "pLI"
]


hist_main_rows = []


for comp in comparators:

    cov = hist_cov.loc[
        (
            hist_cov[
                "comparator"
            ] == comp
        )
        &
        (
            hist_cov[
                "external_set"
            ] == "Tan_All245_nonseed"
        )
    ].iloc[0]


    cont = hist_cont.loc[
        (
            hist_cont[
                "comparator_framework"
            ] == comp
        )
        &
        (
            hist_cont[
                "external_set"
            ] == "Tan_All245_nonseed"
        )
    ]


    context_mean = float(
        cont.loc[
            cont["model"] == "Context",
            "mean_rank_percentile"
        ].iloc[0]
    )


    fusion_mean = float(
        cont.loc[
            cont["model"] == "Fusion",
            "mean_rank_percentile"
        ].iloc[0]
    )


    comp_mean = float(
        cont.loc[
            cont["model"] == comp,
            "mean_rank_percentile"
        ].iloc[0]
    )


    boot = hist_boot.loc[
        (
            hist_boot[
                "comparator"
            ] == comp
        )
        &
        (
            hist_boot[
                "external_set"
            ] == "Tan_All245_nonseed"
        )
        &
        (
            hist_boot[
                "our_model"
            ] == "Fusion"
        )
    ].iloc[0]


    delta_rows = hist_delta.loc[
        (
            hist_delta[
                "comparator"
            ] == comp
        )
        &
        (
            hist_delta[
                "external_set"
            ] == "Tan_All245_nonseed"
        )
        &
        (
            hist_delta[
                "our_model"
            ] == "Fusion"
        )
    ]


    def topk_value(threshold, column):

        row = delta_rows.loc[
            delta_rows[
                "threshold_percent"
            ] == threshold
        ]

        if len(row) == 0:
            return np.nan

        return row.iloc[0][column]


    hist_main_rows.append(
        [
            comp,
            int(
                cov[
                    "comparison_universe_N"
                ]
            ),
            int(
                cov[
                    "historically_unlabeled_and_comparator_evaluable_n"
                ]
            ),
            context_mean,
            fusion_mean,
            comp_mean,
            float(
                boot[
                    "observed_mean_rank_percentile_delta"
                ]
            ),
            float(
                boot[
                    "CI95_low"
                ]
            ),
            float(
                boot[
                    "CI95_high"
                ]
            ),
            int(
                topk_value(
                    1,
                    "our_recovered"
                )
            ),
            int(
                topk_value(
                    1,
                    "comparator_recovered"
                )
            ),
            int(
                topk_value(
                    5,
                    "our_recovered"
                )
            ),
            int(
                topk_value(
                    5,
                    "comparator_recovered"
                )
            ),
            int(
                topk_value(
                    10,
                    "our_recovered"
                )
            ),
            int(
                topk_value(
                    10,
                    "comparator_recovered"
                )
            )
        ]
    )


table6 = pd.DataFrame(
    hist_main_rows,
    columns=[
        "Comparator",
        "Comparison universe N",
        "Evaluable broad external genes",
        "Context mean rank percentile",
        "Fusion mean rank percentile",
        "Comparator mean rank percentile",
        "Fusion − comparator mean delta",
        "95% CI low",
        "95% CI high",
        "Fusion top 1%",
        "Comparator top 1%",
        "Fusion top 5%",
        "Comparator top 5%",
        "Fusion top 10%",
        "Comparator top 10%"
    ]
)


table6.insert(
    1,
    "Comparator type",
    [
        "ASD deep-learning predictor",
        "ASD gene-prioritization method",
        "ASD gene-prioritization method",
        "ASD gene-prioritization method",
        "Generic LoF-constraint baseline"
    ]
)


# ==========================================================================================
# 12. SAVE SOURCE TSV TABLES
# ==========================================================================================

print("\n[4/14] SAVING PUBLICATION SOURCE TABLES")


main_tables = {

    "Table1_framework_characteristics.tsv":
        table1,

    "Table2_primary_internal_performance.tsv":
        table2,

    "Table3_paired_inference_robustness.tsv":
        table3,

    "Table4_temporal_external_validation.tsv":
        table4,

    "Table4b_fusion_vs_context_external.tsv":
        table4b,

    "Table5_external_sequence_validation.tsv":
        table5,

    "Table5b_sequence_paired_inference.tsv":
        table5b,

    "Table6_historical_comparator_broad.tsv":
        table6
}


for filename, df in main_tables.items():

    path = TABLE_DIR / filename

    df.to_csv(
        path,
        sep="\t",
        index=False
    )

    print(
        "[SAVED]",
        path
    )


# Also preserve complete frozen Stage9D / Stage9E tables
# as supplementary source data.

supplementary_tables = {

    "Supplement_Stage9D_topk.tsv":
        seq_topk,

    "Supplement_Stage9D_topk_deltas.tsv":
        seq_delta,

    "Supplement_Stage9E_coverage.tsv":
        hist_cov,

    "Supplement_Stage9E_continuous.tsv":
        hist_cont,

    "Supplement_Stage9E_bootstrap.tsv":
        hist_boot,

    "Supplement_Stage9E_topk.tsv":
        hist_topk,

    "Supplement_Stage9E_topk_deltas.tsv":
        hist_delta
}


for filename, df in supplementary_tables.items():

    df.to_csv(
        TABLE_DIR / filename,
        sep="\t",
        index=False
    )


# ==========================================================================================
# 13. FORMATTED EXCEL WORKBOOK
# ==========================================================================================

print("\n[5/14] BUILDING FORMATTED EXCEL WORKBOOK")


XLSX_FILE = (
    TABLE_DIR /
    "ASD_ContextFusion_Stage10_Publication_Tables.xlsx"
)


readme = pd.DataFrame(
    [
        [
            "Project",
            "ASD-ContextFusion"
        ],
        [
            "Current scientific state",
            "Stage 9E.1 complete; core modelling and validation frozen"
        ],
        [
            "Stage 10 purpose",
            "Publication table and figure assembly only"
        ],
        [
            "Model fitting in Stage 10",
            "NO"
        ],
        [
            "7,900-fold rerun",
            "NO"
        ],
        [
            "New PU sampling",
            "NO"
        ],
        [
            "External gene reselection",
            "NO"
        ],
        [
            "Primary external source",
            "Tan et al. Molecular Psychiatry 2026"
        ],
        [
            "Historical comparator terminology",
            "DeepND-unlabeled temporal comparator benchmark"
        ],
        [
            "Important interpretation",
            (
                "NT-v2 contains externally generalizable sequence signal; "
                "Fusion improves internal metrics modestly but does not "
                "consistently outperform Context temporally."
            )
        ]
    ],
    columns=[
        "Field",
        "Value"
    ]
)


with pd.ExcelWriter(
    XLSX_FILE,
    engine="openpyxl"
) as writer:

    readme.to_excel(
        writer,
        sheet_name="README",
        index=False
    )

    table1.to_excel(
        writer,
        sheet_name="T1_Framework",
        index=False
    )

    table2.to_excel(
        writer,
        sheet_name="T2_InternalPerf",
        index=False
    )

    table3.to_excel(
        writer,
        sheet_name="T3_Paired",
        index=False
    )

    table4.to_excel(
        writer,
        sheet_name="T4_Temporal",
        index=False
    )

    table4b.to_excel(
        writer,
        sheet_name="T4b_FusionCtx",
        index=False
    )

    table5.to_excel(
        writer,
        sheet_name="T5_SeqExternal",
        index=False
    )

    table5b.to_excel(
        writer,
        sheet_name="T5b_SeqPaired",
        index=False
    )

    table6.to_excel(
        writer,
        sheet_name="T6_Historical",
        index=False
    )

    seq_topk.to_excel(
        writer,
        sheet_name="S1_9D_TopK",
        index=False
    )

    hist_cov.to_excel(
        writer,
        sheet_name="S2_9E_Coverage",
        index=False
    )

    hist_boot.to_excel(
        writer,
        sheet_name="S3_9E_Boot",
        index=False
    )

    hist_topk.to_excel(
        writer,
        sheet_name="S4_9E_TopK",
        index=False
    )


# ------------------------------------------------------------------
# Style workbook
# ------------------------------------------------------------------

wb = load_workbook(
    XLSX_FILE
)


header_fill = PatternFill(
    "solid",
    fgColor="17365D"
)

header_font = Font(
    color="FFFFFF",
    bold=True
)

static_font = Font(
    color="666666"
)

imported_font = Font(
    color="008000"
)

thin_gray = Side(
    style="thin",
    color="D9E2F3"
)


for ws in wb.worksheets:

    ws.sheet_view.showGridLines = False

    ws.freeze_panes = "A2"

    for cell in ws[1]:

        cell.fill = header_fill
        cell.font = header_font

        cell.alignment = Alignment(
            horizontal="center",
            vertical="center",
            wrap_text=True
        )

    ws.row_dimensions[
        1
    ].height = 34

    for row in ws.iter_rows(
        min_row=2
    ):

        for cell in row:

            cell.alignment = Alignment(
                vertical="top",
                wrap_text=True
            )

    # Strategic bottom border under header only.
    for cell in ws[1]:

        cell.border = Border(
            bottom=thin_gray
        )

    # Auto-width with sensible limits.
    for col_idx, column_cells in enumerate(
        ws.columns,
        start=1
    ):

        max_len = 0

        for cell in column_cells:

            if cell.value is None:
                continue

            value = str(
                cell.value
            )

            max_len = max(
                max_len,
                len(value)
            )

        width = min(
            max(
                max_len + 2,
                11
            ),
            44
        )

        ws.column_dimensions[
            get_column_letter(
                col_idx
            )
        ].width = width


# Numeric formatting
for sheet_name in [
    "T2_InternalPerf",
    "T3_Paired",
    "T4_Temporal",
    "T4b_FusionCtx",
    "T5_SeqExternal",
    "T5b_SeqPaired",
    "T6_Historical"
]:

    ws = wb[
        sheet_name
    ]

    for row in ws.iter_rows(
        min_row=2
    ):

        for cell in row:

            if isinstance(
                cell.value,
                float
            ):

                cell.number_format = "0.0000"


wb.save(
    XLSX_FILE
)


print(
    "[SAVED]",
    XLSX_FILE
)


# ==========================================================================================
# 14. FIGURE STYLE
# ==========================================================================================

print("\n[6/14] CONFIGURING PUBLICATION FIGURE STYLE")


plt.rcParams.update(
    {
        "font.family":
            "Arial",

        "font.size":
            9,

        "axes.titlesize":
            10,

        "axes.labelsize":
            9,

        "xtick.labelsize":
            8,

        "ytick.labelsize":
            8,

        "legend.fontsize":
            8,

        "figure.titlesize":
            11,

        "pdf.fonttype":
            42,

        "ps.fonttype":
            42
    }
)


# ==========================================================================================
# 15. FIGURE 1 — FRAMEWORK SCHEMATIC
# ==========================================================================================

print("\n[7/14] FIGURE 1 — FRAMEWORK")


fig, ax = plt.subplots(
    figsize=(
        7.2,
        8.0
    )
)

ax.set_xlim(
    0,
    10
)

ax.set_ylim(
    0,
    12
)

ax.axis(
    "off"
)


def box(
    x,
    y,
    w,
    h,
    text,
    fontsize=8.5,
    linewidth=1.2
):

    patch = FancyBboxPatch(
        (
            x,
            y
        ),
        w,
        h,
        boxstyle="round,pad=0.04,rounding_size=0.08",
        facecolor="white",
        edgecolor="black",
        linewidth=linewidth
    )

    ax.add_patch(
        patch
    )

    ax.text(
        x + w / 2,
        y + h / 2,
        text,
        ha="center",
        va="center",
        fontsize=fontsize,
        wrap=True
    )


def arrow(
    x1,
    y1,
    x2,
    y2
):

    ax.add_patch(
        FancyArrowPatch(
            (
                x1,
                y1
            ),
            (
                x2,
                y2
            ),
            arrowstyle="-|>",
            mutation_scale=11,
            linewidth=1.0,
            color="black"
        )
    )


box(
    2.0,
    10.5,
    6.0,
    1.0,
    (
        "Project universe: 18,252 genes\n"
        "190 locked ASD seeds + 18,062 non-seed genes"
    )
)

box(
    2.0,
    8.9,
    6.0,
    1.0,
    (
        "Frozen positive-unlabeled design\n"
        "10 iterations × 79 balanced datasets × 10 outer folds\n"
        "= 7,900 outer train/test splits"
    )
)

arrow(
    5.0,
    10.5,
    5.0,
    9.9
)


# Branches
box(
    0.4,
    6.9,
    4.1,
    1.2,
    (
        "Context branch\n"
        "316 contextual predictors\n"
        "XGBoost"
    )
)

box(
    5.5,
    6.9,
    4.1,
    1.2,
    (
        "Sequence branch\n"
        "Full-transcript NT-v2, 512-D\n"
        "train-only scaling + L2 logistic probe"
    )
)

arrow(
    4.3,
    8.9,
    2.5,
    8.1
)

arrow(
    5.7,
    8.9,
    7.5,
    8.1
)


box(
    5.5,
    5.2,
    4.1,
    1.0,
    (
        "Prespecified sequence controls\n"
        "Length+GC and all 256 oriented 4-mers"
    ),
    fontsize=8.0
)

arrow(
    7.55,
    6.9,
    7.55,
    6.2
)


box(
    2.0,
    3.4,
    6.0,
    1.2,
    (
        "Strict leakage-controlled late fusion\n"
        "Nested cross-fitted Context + NT-v2 probabilities\n"
        "L2 logistic meta learner"
    )
)

arrow(
    2.5,
    6.9,
    4.2,
    4.6
)

arrow(
    7.5,
    6.9,
    5.8,
    4.6
)


box(
    0.5,
    1.6,
    4.1,
    1.0,
    (
        "Robustness and uncertainty\n"
        "Disease-context-reduced analysis\n"
        "Paired gene-level bootstrap"
    ),
    fontsize=8.0
)

box(
    5.4,
    1.6,
    4.1,
    1.0,
    (
        "Temporal / historical evaluation\n"
        "Tan et al. 2026 + NT controls\n"
        "DeepND-unlabeled comparator benchmark"
    ),
    fontsize=8.0
)

arrow(
    4.3,
    3.4,
    2.6,
    2.6
)

arrow(
    5.7,
    3.4,
    7.4,
    2.6
)


ax.text(
    5.0,
    0.45,
    (
        "Primary question: does pretrained nucleotide sequence representation "
        "add information beyond strong ASD context?"
    ),
    ha="center",
    va="center",
    fontsize=9,
    fontweight="bold"
)


fig.suptitle(
    "ASD-ContextFusion analytical framework",
    y=0.99,
    fontweight="bold"
)


fig1_files = save_pub_figure(
    fig,
    "Figure1_ASD_ContextFusion_framework"
)


# ==========================================================================================
# 16. FIGURE 2 — INTERNAL FIVE-MODEL PERFORMANCE
# ==========================================================================================

print("\n[8/14] FIGURE 2 — INTERNAL PERFORMANCE")


fig, axes = plt.subplots(
    1,
    2,
    figsize=(
        7.2,
        3.5
    )
)


model_order = [
    "Context",
    "NT-v2",
    "Length+GC",
    "4-mer",
    "Strict Fusion"
]


tmp = (
    table2
    .set_index(
        "Model"
    )
    .loc[
        model_order
    ]
)


x = np.arange(
    len(model_order)
)


axes[0].bar(
    x,
    tmp[
        "Gene AP"
    ].values
)

axes[0].set_xticks(
    x
)

axes[0].set_xticklabels(
    model_order,
    rotation=35,
    ha="right"
)

axes[0].set_ylabel(
    "Average precision"
)

axes[0].set_title(
    "A. Gene-level average precision"
)

axes[0].set_ylim(
    0,
    0.22
)


axes[1].bar(
    x,
    tmp[
        "Gene AUROC"
    ].values
)

axes[1].set_xticks(
    x
)

axes[1].set_xticklabels(
    model_order,
    rotation=35,
    ha="right"
)

axes[1].set_ylabel(
    "AUROC"
)

axes[1].set_title(
    "B. Gene-level AUROC"
)

axes[1].set_ylim(
    0.70,
    0.98
)


for axx in axes:

    axx.spines[
        "top"
    ].set_visible(False)

    axx.spines[
        "right"
    ].set_visible(False)


fig.suptitle(
    "Internal performance of Context, sequence representations and strict fusion",
    fontweight="bold"
)

fig.tight_layout()


fig2_files = save_pub_figure(
    fig,
    "Figure2_internal_five_model_performance"
)


# ==========================================================================================
# 17. FIGURE 3 — PAIRED COMPLEMENTARITY / ROBUSTNESS FOREST
# ==========================================================================================

print("\n[9/14] FIGURE 3 — PAIRED INFERENCE FOREST")


plot3 = table3.copy()


fig, axes = plt.subplots(
    1,
    2,
    figsize=(
        8.2,
        4.5
    )
)


for axx, metric in zip(
    axes,
    [
        "AP",
        "AUROC"
    ]
):

    d = (
        plot3.loc[
            plot3[
                "Metric"
            ] == metric
        ]
        .reset_index(
            drop=True
        )
    )


    y = np.arange(
        len(d)
    )


    values = d[
        "Delta"
    ].to_numpy(
        dtype=float
    )


    lows = d[
        "95% CI low"
    ].to_numpy(
        dtype=float
    )


    highs = d[
        "95% CI high"
    ].to_numpy(
        dtype=float
    )


    xerr = np.vstack(
        [
            values - lows,
            highs - values
        ]
    )


    axx.errorbar(
        values,
        y,
        xerr=xerr,
        fmt="o",
        capsize=3,
        linewidth=1.2
    )


    axx.axvline(
        0,
        linewidth=1,
        linestyle="--"
    )


    labels = [
        textwrap.fill(
            x,
            28
        )
        for x in d[
            "Comparison"
        ]
    ]


    axx.set_yticks(
        y
    )

    axx.set_yticklabels(
        labels
    )

    axx.invert_yaxis()

    axx.set_xlabel(
        f"Paired Δ {metric}"
    )

    axx.set_title(
        f"{metric}"
    )

    axx.spines[
        "top"
    ].set_visible(False)

    axx.spines[
        "right"
    ].set_visible(False)


fig.suptitle(
    "Paired internal complementarity and source-defined robustness",
    fontweight="bold"
)

fig.tight_layout()


fig3_files = save_pub_figure(
    fig,
    "Figure3_paired_complementarity_robustness_forest"
)


# ==========================================================================================
# 18. FIGURE 4 — 2026 TEMPORAL EXTERNAL VALIDATION
# ==========================================================================================

print("\n[10/14] FIGURE 4 — 2026 TEMPORAL VALIDATION")


set_order = [
    "Broad134",
    "Suggestive45",
    "NoEvidence30",
    "Combined75"
]


model_order_external = [
    "Context",
    "NT-v2",
    "Strict Fusion"
]


plot4 = table4.copy()


series4 = {}


for model in model_order_external:

    d = (
        plot4.loc[
            plot4[
                "Model"
            ] == model
        ]
        .set_index(
            "External set"
        )
        .loc[
            set_order
        ]
    )


    series4[
        model
    ] = d[
        "Mean rank percentile"
    ].values


fig, ax = plt.subplots(
    figsize=(
        7.2,
        4.2
    )
)


grouped_bar(
    ax,
    [
        "Broad\n(n=134)",
        "Suggestive\n(n=45)",
        "No-evidence\n(n=30)",
        "Combined\n(n=75)"
    ],
    series4,
    "Mean rank percentile",
    ylim=(
        50,
        86
    )
)


ax.axhline(
    50,
    linestyle="--",
    linewidth=0.9
)


ax.set_title(
    "2026 post-benchmark temporal validation",
    fontweight="bold"
)


ax.text(
    0.99,
    0.02,
    (
        "Higher percentile = stronger prioritization.\n"
        "Fusion does not consistently outperform Context."
    ),
    transform=ax.transAxes,
    ha="right",
    va="bottom",
    fontsize=7.5
)


fig.tight_layout()


fig4_files = save_pub_figure(
    fig,
    "Figure4_2026_temporal_external_validation"
)


# ==========================================================================================
# 19. FIGURE 5 — NT-v2 VS SIMPLE SEQUENCE CONTROLS EXTERNALLY
# ==========================================================================================

print("\n[11/14] FIGURE 5 — EXTERNAL SEQUENCE CONTROL VALIDATION")


seq_set_order = [
    "Tan_All245_nonseed",
    "Tan_Suggestive46_nonseed",
    "Tan_NoEvidence32_nonseed",
    "Tan_SuggestiveOrNoEvidence78_nonseed"
]


seq_model_order = [
    "NTv2",
    "LengthGC",
    "Fourmer"
]


series5 = {}


for model in seq_model_order:

    d = (
        seq_rank.loc[
            seq_rank[
                "model"
            ] == model
        ]
        .set_index(
            "external_set"
        )
        .loc[
            seq_set_order
        ]
    )


    series5[
        SEQ_MODEL_LABELS[
            model
        ]
    ] = d[
        "mean_rank_percentile"
    ].values


fig, ax = plt.subplots(
    figsize=(
        7.2,
        4.2
    )
)


grouped_bar(
    ax,
    [
        "Broad\n(n=134)",
        "Suggestive\n(n=45)",
        "No-evidence\n(n=30)",
        "Combined\n(n=75)"
    ],
    series5,
    "Mean rank percentile",
    ylim=(
        45,
        80
    )
)


ax.axhline(
    50,
    linestyle="--",
    linewidth=0.9
)


ax.set_title(
    "External generalization of NT-v2 versus simple sequence controls",
    fontweight="bold"
)


fig.tight_layout()


fig5_files = save_pub_figure(
    fig,
    "Figure5_external_NTv2_sequence_control_validation"
)


# ==========================================================================================
# 20. FIGURE 6 — HISTORICAL COMPARATOR BROAD TEMPORAL BENCHMARK
# ==========================================================================================

print("\n[12/14] FIGURE 6 — HISTORICAL COMPARATOR BENCHMARK")


fig, ax = plt.subplots(
    figsize=(
        7.5,
        4.5
    )
)


x = np.arange(
    len(table6)
)


offset = 0.18


ax.scatter(
    x - offset,
    table6[
        "Context mean rank percentile"
    ],
    marker="o",
    s=45,
    label="Context"
)


ax.scatter(
    x,
    table6[
        "Fusion mean rank percentile"
    ],
    marker="s",
    s=45,
    label="Strict Fusion"
)


ax.scatter(
    x + offset,
    table6[
        "Comparator mean rank percentile"
    ],
    marker="^",
    s=45,
    label="Historical comparator / pLI"
)


for i, row in table6.iterrows():

    ax.text(
        i,
        46.2,
        (
            f"n={int(row['Evaluable broad external genes'])}"
        ),
        ha="center",
        va="top",
        fontsize=7
    )


ax.set_xticks(
    x
)

ax.set_xticklabels(
    table6[
        "Comparator"
    ]
)

ax.set_ylabel(
    "Mean rank percentile"
)

ax.set_ylim(
    45,
    85
)

ax.axhline(
    50,
    linestyle="--",
    linewidth=0.9
)


ax.legend(
    frameon=False,
    ncol=3,
    loc="upper center",
    bbox_to_anchor=(
        0.5,
        1.16
    )
)


ax.set_title(
    "DeepND-unlabeled broad temporal comparator benchmark",
    fontweight="bold"
)


ax.spines[
    "top"
].set_visible(False)

ax.spines[
    "right"
].set_visible(False)


fig.text(
    0.5,
    0.005,
    (
        "Comparator-specific common universes were used; "
        "n denotes evaluable broad external genes."
    ),
    ha="center",
    fontsize=7.5
)


fig.tight_layout(
    rect=[
        0,
        0.04,
        1,
        0.94
    ]
)


fig6_files = save_pub_figure(
    fig,
    "Figure6_historical_comparator_temporal_benchmark"
)


# ==========================================================================================
# 21. SAVE FIGURE SOURCE DATA
# ==========================================================================================

print("\n[13/14] SAVING FIGURE SOURCE DATA")


figure_sources = {

    "Figure2_source.tsv":
        table2,

    "Figure3_source.tsv":
        table3,

    "Figure4_source.tsv":
        table4,

    "Figure5_source.tsv":
        seq_rank,

    "Figure6_source.tsv":
        table6
}


for filename, df in figure_sources.items():

    df.to_csv(
        SOURCE_DIR / filename,
        sep="\t",
        index=False
    )


# ==========================================================================================
# 22. DRAFT FIGURE CAPTIONS
# ==========================================================================================

captions = """
FIGURE 1. ASD-ContextFusion analytical framework.
The framework combines a frozen positive-unlabeled contextual ASD gene-prioritization
branch with full-transcript NT-v2 nucleotide foundation-model representations.
Sequence controls include transcript Length+GC and all 256 oriented 4-mers.
Context and NT-v2 probabilities are combined using strict nested cross-fitted late
fusion. Robustness analyses include source-defined disease-context reduction and
paired gene-level inference, followed by source-locked 2026 temporal validation
and a DeepND-unlabeled historical comparator benchmark.

FIGURE 2. Internal gene-level performance of the five prespecified models.
Context was the strongest individual predictor, while strict Context+NT-v2 fusion
achieved the highest internal gene-level AUROC and average precision. NT-v2
outperformed both Length+GC and 4-mer sequence controls.

FIGURE 3. Paired complementarity and disease-context robustness.
Points indicate observed paired gene-level metric differences and horizontal
intervals indicate 95% paired bootstrap confidence intervals. Strict fusion showed
a small positive AUROC gain over Context, whereas the corresponding AP interval
included zero. Complementarity persisted after source-defined disease-context
reduction.

FIGURE 4. Temporal prioritization of post-benchmark ASD genes reported in 2026.
Context and strict fusion strongly prioritized the broad non-seed external gene set.
However, the incremental temporal advantage of fusion over Context was not
consistently reproduced, particularly in the no-evidence subset.

FIGURE 5. External sequence-representation validation.
NT-v2 achieved higher mean rank percentiles than simple sequence controls in the
broad temporal ASD set. The NT-v2 advantage over 4-mer representation persisted
in several low-evidence analyses, whereas superiority over Length+GC was not
consistent in the hardest subsets.

FIGURE 6. DeepND-unlabeled temporal comparison with historical ASD predictors and pLI.
Context and strict fusion were broadly competitive with DeepND, exceeded Krishnan
and DAWN in several comparisons, and substantially outperformed the generic pLI
constraint baseline. Each comparator was evaluated within its own score-complete,
DeepND-unlabeled common universe; therefore sample sizes differ across comparisons.
"""


CAPTION_FILE = (
    OUT_ROOT /
    "Stage10_draft_figure_captions.txt"
)


CAPTION_FILE.write_text(
    captions.strip(),
    encoding="utf-8"
)


# ==========================================================================================
# 23. REPRODUCIBILITY MANIFEST
# ==========================================================================================

all_output_files = []


# Tables
for p in TABLE_DIR.rglob("*"):

    if p.is_file():
        all_output_files.append(
            p
        )


# Figures
for p in FIGURE_DIR.rglob("*"):

    if p.is_file():
        all_output_files.append(
            p
        )


# Figure sources
for p in SOURCE_DIR.rglob("*"):

    if p.is_file():
        all_output_files.append(
            p
        )


all_output_files.append(
    CAPTION_FILE
)


MANIFEST_FILE = (
    OUT_ROOT /
    "ASD_ContextFusion_STAGE10_publication_manifest.json"
)


manifest = {

    "project":
        "ASD-ContextFusion",

    "stage":
        "10A_10B_publication_tables_figures",

    "scientific_state":
        "Stage9E1_complete_core_analysis_frozen",

    "model_fitting":
        False,

    "model_tuning":
        False,

    "PU_resampling":
        False,

    "embedding_generation":
        False,

    "external_gene_reselection":
        False,

    "threshold_change":
        False,

    "stage10_role":
        "Engineering/output assembly only; scientific behavior unchanged",

    "input_files":
        [
            {
                "path":
                    str(p),

                "sha256":
                    sha256_file(p)
            }
            for p in required_files
        ],

    "output_files":
        [
            {
                "path":
                    str(p),

                "sha256":
                    sha256_file(p)
            }
            for p in sorted(
                all_output_files,
                key=lambda x: str(x)
            )
        ]
}


with open(
    MANIFEST_FILE,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        manifest,
        f,
        indent=2
    )


# ==========================================================================================
# 24. FINAL QA
# ==========================================================================================

print("\n[14/14] FINAL QA")


# Locked key-value assertions.
assert np.isclose(
    table2.loc[
        table2[
            "Model"
        ] == "Strict Fusion",
        "Gene AP"
    ].iloc[0],
    0.196525
)


assert np.isclose(
    table2.loc[
        table2[
            "Model"
        ] == "Context",
        "Gene AUROC"
    ].iloc[0],
    0.947076
)


broad_nt = seq_rank.loc[
    (
        seq_rank[
            "external_set"
        ] == "Tan_All245_nonseed"
    )
    &
    (
        seq_rank[
            "model"
        ] == "NTv2"
    ),
    "mean_rank_percentile"
].iloc[0]


assert np.isclose(
    broad_nt,
    74.754336,
    atol=1e-6
)


deepnd_broad = table6.loc[
    table6[
        "Comparator"
    ] == "DeepND"
].iloc[0]


assert int(
    deepnd_broad[
        "Evaluable broad external genes"
    ]
) == 100


assert np.isclose(
    deepnd_broad[
        "Comparator mean rank percentile"
    ],
    77.322655,
    atol=1e-6
)


print("\n" + "=" * 150)
print("STAGE 10A + 10B PUBLICATION ASSEMBLY COMPLETE [OK]")
print("=" * 150)

print("\nSCIENTIFIC INTEGRITY")
print("Model fitting                 : NO")
print("7,900-fold rerun              : NO")
print("Hyperparameter tuning         : NO")
print("New PU sampling               : NO")
print("New embeddings                : NO")
print("External gene reselection     : NO")
print("External threshold changes    : NO")
print("Scientific design changed     : NO")

print("\nMAIN TABLES")
print("Table 1 : Framework characteristics")
print("Table 2 : Internal five-model performance")
print("Table 3 : Paired inference / robustness")
print("Table 4 : 2026 temporal validation")
print("Table 5 : External NT-v2 sequence-control validation")
print("Table 6 : Historical comparator benchmark")

print("\nMAIN FIGURES")
print("Figure 1 : Framework schematic")
print("Figure 2 : Internal performance")
print("Figure 3 : Paired inference forest")
print("Figure 4 : Temporal external validation")
print("Figure 5 : External NT-v2 sequence controls")
print("Figure 6 : Historical comparator benchmark")

print("\nFORMATS")
print("Figures : PDF vector + PNG 300 DPI + TIFF 600 DPI")
print("Tables  : XLSX + TSV source tables")

print("\nOUTPUT ROOT:")
print(OUT_ROOT)

print("\nEXCEL WORKBOOK:")
print(XLSX_FILE)

print("\nFIGURE DIRECTORY:")
print(FIGURE_DIR)

print("\nMANIFEST:")
print(MANIFEST_FILE)

print("\nNEXT SCIENTIFIC STEP:")
print(
    "Review Tables/Figures -> Stage 10C rule-based biological interpretation -> "
    "final supplementary package -> manuscript."
)

print("=" * 150)
```

```python
# ======================================================================================
# ASD-ContextFusion
# FINAL PUBLICATION-QUALITY FIGURE SUITE
#
# PURPOSE:
#   Rebuild ALL final manuscript figures from FROZEN/CANONICAL results only.
#
# IMPORTANT:
#   - NO model fitting
#   - NO hyperparameter tuning
#   - NO PU resampling
#   - NO external-set redefinition
#   - NO post-result thresholding
#   - NO new biological analysis
#
# OUTPUT:
#   5 MAIN FIGURES
#   3 SUPPLEMENTARY FIGURES
#
#   PDF  = vector publication master
#   SVG  = editable vector
#   PNG  = 600 dpi
#
# USE:
#   Run in normal modern Python/Jupyter kernel, NOT legacy mantis37.
# ======================================================================================

from pathlib import Path
import re
import gc
import warnings

import numpy as np
import pandas as pd

import matplotlib
matplotlib.use("Agg")

import matplotlib.pyplot as plt
import matplotlib as mpl
from matplotlib import font_manager as fm
from matplotlib.patches import FancyBboxPatch, FancyArrowPatch
from matplotlib.lines import Line2D

warnings.filterwarnings("ignore")


# ======================================================================================
# 0. PATHS
# ======================================================================================

ROOT = Path(r"E:\ASD_ContextFusion_Q1")

STRICT_DIR = (
    ROOT /
    "05_results" /
    "proposed" /
    "strict_context_NTv2_fusion"
)

STAGE9C_DIR = (
    ROOT /
    "05_results" /
    "proposed" /
    "stage9C_external_validation"
)

STAGE9D_DIR = (
    ROOT /
    "05_results" /
    "proposed" /
    "stage9D_external_sequence_controls"
)

STAGE9E_DIR = (
    ROOT /
    "05_results" /
    "proposed" /
    "stage9E_historical_comparator_benchmark"
)

OUT = (
    ROOT /
    "06_publication" /
    "stage10_tables_figures" /
    "FINAL_PUBLICATION_FIGURES"
)

OUT.mkdir(
    parents=True,
    exist_ok=True
)

for p in [
    ROOT,
    STRICT_DIR,
    STAGE9C_DIR,
    STAGE9D_DIR,
    STAGE9E_DIR
]:
    assert p.exists(), f"MISSING REQUIRED PATH:\n{p}"


# ======================================================================================
# 1. GLOBAL FIGURE STYLE
# ======================================================================================

available_fonts = {
    f.name
    for f in fm.fontManager.ttflist
}

FONT = (
    "Arial"
    if "Arial" in available_fonts
    else "DejaVu Sans"
)

mpl.rcParams.update(
    {
        "font.family": FONT,
        "font.size": 9,
        "axes.titlesize": 10.5,
        "axes.labelsize": 9.5,
        "xtick.labelsize": 8.5,
        "ytick.labelsize": 8.5,
        "legend.fontsize": 8.2,
        "figure.titlesize": 12,
        "axes.linewidth": 0.8,
        "xtick.major.width": 0.8,
        "ytick.major.width": 0.8,
        "xtick.major.size": 3.5,
        "ytick.major.size": 3.5,
        "pdf.fonttype": 42,
        "ps.fonttype": 42,
        "svg.fonttype": "none",
        "savefig.facecolor": "white",
        "figure.facecolor": "white",
        "axes.facecolor": "white",
    }
)

# Color-blind-friendly, journal-safe palette
COLORS = {
    "Context":   "#0072B2",
    "NT-v2":     "#D55E00",
    "Fusion":    "#009E73",
    "Length+GC": "#7A7A7A",
    "4-mer":     "#CC79A7",
    "Reference": "#444444",
    "Light":     "#D9D9D9",
    "Grid":      "#E8E8E8",
}

SAVE_TIFF = False   # Change to True ONLY if target journal explicitly requires TIFF.
PNG_DPI = 600


# ======================================================================================
# 2. HELPERS
# ======================================================================================

def save_figure(
    fig,
    stem,
    dpi=PNG_DPI
):
    pdf = OUT / f"{stem}.pdf"
    svg = OUT / f"{stem}.svg"
    png = OUT / f"{stem}.png"

    fig.savefig(
        pdf,
        bbox_inches="tight",
        pad_inches=0.05
    )

    fig.savefig(
        svg,
        bbox_inches="tight",
        pad_inches=0.05
    )

    fig.savefig(
        png,
        dpi=dpi,
        bbox_inches="tight",
        pad_inches=0.05
    )

    if SAVE_TIFF:
        tif = OUT / f"{stem}.tiff"

        fig.savefig(
            tif,
            dpi=600,
            bbox_inches="tight",
            pad_inches=0.05,
            pil_kwargs={
                "compression": "tiff_lzw"
            }
        )

    plt.close(fig)
    gc.collect()

    print(f"[OK] {stem}")


def panel_label(
    ax,
    label
):
    ax.text(
        -0.10,
        1.06,
        label,
        transform=ax.transAxes,
        fontsize=12,
        fontweight="bold",
        va="top",
        ha="left"
    )


def clean_axis(
    ax,
    grid_axis=None
):
    ax.spines["top"].set_visible(False)
    ax.spines["right"].set_visible(False)

    ax.tick_params(
        direction="out"
    )

    if grid_axis == "x":
        ax.grid(
            axis="x",
            linestyle="-",
            linewidth=0.5,
            color=COLORS["Grid"],
            zorder=0
        )

    if grid_axis == "y":
        ax.grid(
            axis="y",
            linestyle="-",
            linewidth=0.5,
            color=COLORS["Grid"],
            zorder=0
        )


def read_tabular(
    path,
    nrows=None
):
    name = path.name.lower()

    if ".tsv" in name:
        sep = "\t"
    elif ".csv" in name:
        sep = ","
    else:
        raise ValueError(
            f"Unsupported file:\n{path}"
        )

    return pd.read_csv(
        path,
        sep=sep,
        compression="infer",
        nrows=nrows,
        low_memory=False
    )


def locate_table(
    directory,
    required_columns,
    prefer_tokens=None
):
    """
    Finds an EXISTING saved result table by required columns.
    This is file discovery only; no scientific calculation.
    """

    if prefer_tokens is None:
        prefer_tokens = []

    candidates = []

    for pattern in [
        "*.tsv",
        "*.tsv.gz",
        "*.csv",
        "*.csv.gz"
    ]:
        candidates.extend(
            directory.rglob(pattern)
        )

    matches = []

    required_lower = {
        x.lower()
        for x in required_columns
    }

    for path in candidates:

        try:
            head = read_tabular(
                path,
                nrows=3
            )
        except Exception:
            continue

        cols_lower = {
            str(c).lower()
            for c in head.columns
        }

        if required_lower.issubset(
            cols_lower
        ):

            score = sum(
                token.lower()
                in path.name.lower()
                for token
                in prefer_tokens
            )

            matches.append(
                (
                    score,
                    len(str(path)),
                    path
                )
            )

    if not matches:
        raise FileNotFoundError(
            "\nNo table found with required columns:\n"
            f"{required_columns}\n"
            f"Directory:\n{directory}"
        )

    matches.sort(
        key=lambda x: (
            -x[0],
            x[1],
            str(x[2])
        )
    )

    selected = matches[0][2]

    print(
        "Resolved:",
        selected
    )

    return selected


def find_col(
    df,
    candidates
):
    lower_map = {
        str(c).lower(): c
        for c in df.columns
    }

    for candidate in candidates:
        if candidate.lower() in lower_map:
            return lower_map[
                candidate.lower()
            ]

    for c in df.columns:
        cl = str(c).lower()

        for candidate in candidates:
            if candidate.lower() in cl:
                return c

    raise KeyError(
        f"Could not resolve column from:\n{candidates}\n"
        f"Available:\n{list(df.columns)}"
    )


def normalize_model(
    value
):
    s = str(value).strip().lower()
    compact = (
        s.replace("_", "")
        .replace("-", "")
        .replace(" ", "")
    )

    if (
        "fusion" in s
        or "strictfusion" in compact
    ):
        return "Fusion"

    if (
        "lengthgc" in compact
        or (
            "length" in s
            and "gc" in s
        )
    ):
        return "Length+GC"

    if (
        "fourmer" in compact
        or "4mer" in compact
    ):
        return "4-mer"

    if (
        "ntv2" in compact
        or "nucleotidetransformer" in compact
    ):
        return "NT-v2"

    if "context" in s:
        return "Context"

    return str(value)


def parse_threshold(
    value
):
    s = str(value)

    nums = re.findall(
        r"\d+",
        s
    )

    if not nums:
        raise ValueError(
            f"Cannot parse threshold: {value}"
        )

    return int(
        nums[0]
    )


def ci_error(
    estimate,
    lo,
    hi
):
    return np.array(
        [
            [
                estimate - lo
            ],
            [
                hi - estimate
            ]
        ]
    )


def annotate_value(
    ax,
    x,
    y,
    text,
    dx=0.005,
    fontsize=8
):
    ax.text(
        x + dx,
        y,
        text,
        va="center",
        ha="left",
        fontsize=fontsize
    )


# ======================================================================================
# 3. LOAD CANONICAL FINAL TABLES
# ======================================================================================

GENE_SUMMARY_FILE = (
    STRICT_DIR /
    "FiveModel_gene_level_summary.tsv"
)

SPLIT_SUMMARY_FILE = (
    STRICT_DIR /
    "FiveModel_split_level_summary.tsv"
)

assert GENE_SUMMARY_FILE.exists(), (
    f"Missing canonical file:\n"
    f"{GENE_SUMMARY_FILE}"
)

assert SPLIT_SUMMARY_FILE.exists(), (
    f"Missing canonical file:\n"
    f"{SPLIT_SUMMARY_FILE}"
)

gene_summary = read_tabular(
    GENE_SUMMARY_FILE
)

split_summary = read_tabular(
    SPLIT_SUMMARY_FILE
)


# --------------------------------------------------------------------------------------
# Resolve internal gene-level columns
# --------------------------------------------------------------------------------------

GENE_MODEL_COL = find_col(
    gene_summary,
    [
        "Model"
    ]
)

GENE_AUROC_COL = find_col(
    gene_summary,
    [
        "Gene_level_AUROC",
        "gene auroc"
    ]
)

GENE_AP_COL = find_col(
    gene_summary,
    [
        "Gene_level_AP",
        "gene ap"
    ]
)

gene_summary[
    "Model_clean"
] = gene_summary[
    GENE_MODEL_COL
].map(
    normalize_model
)

gene_perf = (
    gene_summary
    .set_index(
        "Model_clean"
    )
)


# --------------------------------------------------------------------------------------
# HARD AUDIT AGAINST LOCKED FINAL VALUES
# Prevents accidental use of old 55-feature / preliminary outputs.
# --------------------------------------------------------------------------------------

EXPECTED_GENE = {
    "Context": (
        0.947076,
        0.182366
    ),
    "NT-v2": (
        0.899628,
        0.099593
    ),
    "Length+GC": (
        0.832049,
        0.052922
    ),
    "4-mer": (
        0.779947,
        0.039135
    ),
    "Fusion": (
        0.951591,
        0.196525
    )
}

for model, (
    expected_auc,
    expected_ap
) in EXPECTED_GENE.items():

    assert model in gene_perf.index, (
        f"{model} missing from canonical gene summary."
    )

    actual_auc = float(
        gene_perf.loc[
            model,
            GENE_AUROC_COL
        ]
    )

    actual_ap = float(
        gene_perf.loc[
            model,
            GENE_AP_COL
        ]
    )

    assert np.isclose(
        actual_auc,
        expected_auc,
        atol=5e-5
    ), (
        f"AUROC mismatch for {model}: "
        f"{actual_auc} vs {expected_auc}"
    )

    assert np.isclose(
        actual_ap,
        expected_ap,
        atol=5e-5
    ), (
        f"AP mismatch for {model}: "
        f"{actual_ap} vs {expected_ap}"
    )

print(
    "\nCanonical five-model gene-level summary: VERIFIED [OK]"
)


# ======================================================================================
# 4. LOAD STAGE 9C TABLES
# ======================================================================================

STAGE9C_CONT_FILE = locate_table(
    STAGE9C_DIR,
    required_columns=[
        "external_set",
        "n_external_genes",
        "model",
        "mean_rank_percentile",
        "median_rank_percentile"
    ],
    prefer_tokens=[
        "continuous",
        "performance"
    ]
)

STAGE9C_BOOT_FILE = locate_table(
    STAGE9C_DIR,
    required_columns=[
        "external_set",
        "n_external_genes",
        "CI95_low",
        "CI95_high",
        "bootstrap_fraction_delta_gt_0"
    ],
    prefer_tokens=[
        "bootstrap",
        "fusion"
    ]
)

STAGE9C_TOPK_FILE = locate_table(
    STAGE9C_DIR,
    required_columns=[
        "external_set",
        "model",
        "threshold",
        "recovered_genes",
        "recovery_fraction"
    ],
    prefer_tokens=[
        "top",
        "recovery"
    ]
)

stage9c_cont = read_tabular(
    STAGE9C_CONT_FILE
)

stage9c_boot = read_tabular(
    STAGE9C_BOOT_FILE
)

stage9c_topk = read_tabular(
    STAGE9C_TOPK_FILE
)

stage9c_cont[
    "model_clean"
] = stage9c_cont[
    "model"
].map(
    normalize_model
)

stage9c_topk[
    "model_clean"
] = stage9c_topk[
    "model"
].map(
    normalize_model
)

stage9c_topk[
    "threshold_pct"
] = stage9c_topk[
    "threshold"
].map(
    parse_threshold
)


# ======================================================================================
# 5. LOAD STAGE 9D TABLES
# ======================================================================================

STAGE9D_CONT_FILE = locate_table(
    STAGE9D_DIR,
    required_columns=[
        "external_set",
        "n_external_genes",
        "model",
        "mean_rank_percentile",
        "median_rank_percentile"
    ],
    prefer_tokens=[
        "continuous",
        "performance"
    ]
)

STAGE9D_BOOT_FILE = locate_table(
    STAGE9D_DIR,
    required_columns=[
        "external_set",
        "comparison",
        "observed_mean_rank_percentile_delta",
        "CI95_low",
        "CI95_high"
    ],
    prefer_tokens=[
        "bootstrap",
        "sequence"
    ]
)

stage9d_cont = read_tabular(
    STAGE9D_CONT_FILE
)

stage9d_boot = read_tabular(
    STAGE9D_BOOT_FILE
)

stage9d_cont[
    "model_clean"
] = stage9d_cont[
    "model"
].map(
    normalize_model
)


# ======================================================================================
# 6. LOAD STAGE 9E HISTORICAL-COMPARATOR BOOTSTRAP
# ======================================================================================

STAGE9E_BOOT_FILE = locate_table(
    STAGE9E_DIR,
    required_columns=[
        "comparator",
        "external_set",
        "our_model",
        "observed_mean_rank_percentile_delta",
        "CI95_low",
        "CI95_high"
    ],
    prefer_tokens=[
        "bootstrap",
        "continuous"
    ]
)

stage9e_boot = read_tabular(
    STAGE9E_BOOT_FILE
)

stage9e_boot[
    "our_model_clean"
] = stage9e_boot[
    "our_model"
].map(
    normalize_model
)


# ======================================================================================
# 7. LOCKED INTERNAL INFERENCE VALUES
#
# These are FINAL frozen Stage-7/8 results.
# They are plotting constants only; nothing is recomputed.
# ======================================================================================

INTERNAL_EFFECTS = pd.DataFrame(
    {
        "metric": [
            "AUROC",
            "AP"
        ],
        "estimate": [
            0.004515,
            0.014159
        ],
        "low": [
            0.000206,
            -0.012178
        ],
        "high": [
            0.009281,
            0.039833
        ]
    }
)

ROBUSTNESS_EFFECTS = pd.DataFrame(
    {
        "setting": [
            "Full Context",
            "Full Context",
            "Disease-context-reduced",
            "Disease-context-reduced"
        ],
        "metric": [
            "AUROC",
            "AP",
            "AUROC",
            "AP"
        ],
        "estimate": [
            0.004515,
            0.014159,
            0.006318,
            0.015343
        ],
        "low": [
            0.000206,
            -0.012178,
            0.000715,
            -0.010172
        ],
        "high": [
            0.009281,
            0.039833,
            0.013274,
            0.040478
        ]
    }
)

META_COEFS = pd.DataFrame(
    {
        "branch": [
            "Context",
            "NT-v2"
        ],
        "mean": [
            4.318359,
            1.468889
        ],
        "sd": [
            0.250752,
            0.325536
        ]
    }
)

GENERIC_ABSOLUTE = pd.DataFrame(
    {
        "setting": [
            "Full Context",
            "Full Fusion",
            "Reduced Context",
            "Reduced Fusion"
        ],
        "AUROC": [
            0.947076,
            0.951591,
            0.943594,
            0.949912
        ],
        "AP": [
            0.182366,
            0.196525,
            0.174316,
            0.189659
        ]
    }
)


# ======================================================================================
# 8. LABEL DEFINITIONS
# ======================================================================================

EXT_ORDER = [
    "Tan_All245_nonseed",
    "Tan_Suggestive46_nonseed",
    "Tan_NoEvidence32_nonseed",
    "Tan_SuggestiveOrNoEvidence78_nonseed"
]

EXT_LABEL = {
    "Tan_All245_nonseed":
        "Broad\n(n=134)",

    "Tan_Suggestive46_nonseed":
        "Suggestive\n(n=45)",

    "Tan_NoEvidence32_nonseed":
        "No evidence\n(n=30)",

    "Tan_SuggestiveOrNoEvidence78_nonseed":
        "Suggestive or\nno evidence (n=75)"
}

MODEL_ORDER = [
    "Context",
    "NT-v2",
    "Fusion"
]

SEQ_ORDER = [
    "NT-v2",
    "Length+GC",
    "4-mer"
]


# ======================================================================================
# FIGURE 1
# STUDY DESIGN / SCIENTIFIC QUESTION
# ======================================================================================

fig, ax = plt.subplots(
    figsize=(
        13.0,
        7.1
    )
)

ax.set_xlim(
    0,
    1
)

ax.set_ylim(
    0,
    1
)

ax.axis(
    "off"
)


def workflow_box(
    x,
    y,
    w,
    h,
    text,
    fc,
    ec="#333333",
    fs=9,
    weight="normal"
):

    patch = FancyBboxPatch(
        (
            x,
            y
        ),
        w,
        h,
        boxstyle=(
            "round,pad=0.012,"
            "rounding_size=0.015"
        ),
        linewidth=1.1,
        edgecolor=ec,
        facecolor=fc,
        transform=ax.transAxes,
        clip_on=False
    )

    ax.add_patch(
        patch
    )

    ax.text(
        x + w / 2,
        y + h / 2,
        text,
        transform=ax.transAxes,
        ha="center",
        va="center",
        fontsize=fs,
        fontweight=weight,
        linespacing=1.25
    )


def workflow_arrow(
    x1,
    y1,
    x2,
    y2
):

    arr = FancyArrowPatch(
        (
            x1,
            y1
        ),
        (
            x2,
            y2
        ),
        transform=ax.transAxes,
        arrowstyle="-|>",
        mutation_scale=13,
        linewidth=1.15,
        color="#555555",
        shrinkA=2,
        shrinkB=2
    )

    ax.add_patch(
        arr
    )


# ----------------------------------------------------------------------
# Input universe
# ----------------------------------------------------------------------

workflow_box(
    0.025,
    0.68,
    0.17,
    0.18,
    (
        "Common gene universe\n"
        "18,252 genes\n"
        "190 ASD seeds\n"
        "18,062 unlabeled"
    ),
    "#EEF5FA",
    ec=COLORS["Context"],
    fs=9.2,
    weight="bold"
)

# Context branch
workflow_box(
    0.255,
    0.78,
    0.16,
    0.13,
    (
        "Context branch\n"
        "316 predictors\n"
        "XGBoost"
    ),
    "#E4F1F8",
    ec=COLORS["Context"],
    fs=9
)

# Transcript branch
workflow_box(
    0.255,
    0.55,
    0.16,
    0.13,
    (
        "Representative transcript\n"
        "GENCODE v50\n"
        "one transcript/gene"
    ),
    "#FFF0E9",
    ec=COLORS["NT-v2"],
    fs=8.8
)

# NT
workflow_box(
    0.475,
    0.63,
    0.16,
    0.14,
    (
        "NT-v2\n"
        "full-transcript embedding\n"
        "512 dimensions"
    ),
    "#FCE8DE",
    ec=COLORS["NT-v2"],
    fs=9,
    weight="bold"
)

# Controls
workflow_box(
    0.475,
    0.43,
    0.16,
    0.12,
    (
        "Sequence controls\n"
        "Length + GC\n"
        "256 oriented 4-mers"
    ),
    "#F3F3F3",
    ec=COLORS["Length+GC"],
    fs=8.8
)

# PU
workflow_box(
    0.255,
    0.22,
    0.38,
    0.12,
    (
        "Frozen matched positive-unlabeled evaluation\n"
        "10 iterations × 79 balanced sets × 10 folds = 7,900 test splits"
    ),
    "#F7F7F7",
    ec="#555555",
    fs=9
)

# Fusion
workflow_box(
    0.695,
    0.70,
    0.17,
    0.15,
    (
        "Strict cross-fitted\n"
        "late fusion\n"
        "Context + NT-v2"
    ),
    "#E4F5EE",
    ec=COLORS["Fusion"],
    fs=9.2,
    weight="bold"
)

# Inference
workflow_box(
    0.695,
    0.48,
    0.17,
    0.14,
    (
        "Gene-level inference\n"
        "paired bootstrap\n"
        "5,000 replicates"
    ),
    "#EEF7F3",
    ec=COLORS["Fusion"],
    fs=8.8
)

# External
workflow_box(
    0.695,
    0.25,
    0.17,
    0.14,
    (
        "Post-benchmark\n"
        "Tan 2026 validation\n"
        "source-locked genes"
    ),
    "#FFF7DE",
    ec="#C28A00",
    fs=8.8
)

workflow_box(
    0.695,
    0.04,
    0.17,
    0.13,
    (
        "Historical benchmarking\n"
        "DeepND / Krishnan / DAWN\n"
        "DAMAGES / pLI"
    ),
    "#F5EEFA",
    ec="#7E57A6",
    fs=8.5
)

# Main scientific answer
workflow_box(
    0.885,
    0.34,
    0.10,
    0.40,
    (
        "QUESTION\n\n"
        "Does NT-v2\n"
        "contain ASD-\n"
        "relevant signal\n"
        "beyond simple\n"
        "sequence and\n"
        "strong Context?\n\n"
        "Does it improve\n"
        "future-gene\n"
        "prioritization?"
    ),
    "#FFFFFF",
    ec="#222222",
    fs=8.6,
    weight="bold"
)

# Arrows
workflow_arrow(
    0.195,
    0.77,
    0.255,
    0.84
)

workflow_arrow(
    0.195,
    0.73,
    0.255,
    0.61
)

workflow_arrow(
    0.415,
    0.615,
    0.475,
    0.69
)

workflow_arrow(
    0.415,
    0.60,
    0.475,
    0.49
)

workflow_arrow(
    0.415,
    0.84,
    0.695,
    0.79
)

workflow_arrow(
    0.635,
    0.70,
    0.695,
    0.77
)

workflow_arrow(
    0.635,
    0.28,
    0.695,
    0.55
)

workflow_arrow(
    0.78,
    0.70,
    0.78,
    0.62
)

workflow_arrow(
    0.78,
    0.48,
    0.78,
    0.39
)

workflow_arrow(
    0.78,
    0.25,
    0.78,
    0.17
)

workflow_arrow(
    0.86,
    0.55,
    0.885,
    0.55
)

ax.text(
    0.025,
    0.955,
    "A",
    transform=ax.transAxes,
    fontsize=13,
    fontweight="bold"
)

ax.text(
    0.060,
    0.955,
    (
        "ASD-ContextFusion study design: "
        "controlled evaluation of incremental foundation-model value"
    ),
    transform=ax.transAxes,
    fontsize=11.5,
    fontweight="bold",
    va="top"
)

save_figure(
    fig,
    "Figure1_ASD_ContextFusion_Study_Design"
)


# ======================================================================================
# FIGURE 2
# DOES NT-v2 CONTAIN SIGNAL BEYOND SIMPLE SEQUENCE COMPOSITION?
# ======================================================================================

fig = plt.figure(
    figsize=(
        12.8,
        8.6
    )
)

gs = fig.add_gridspec(
    2,
    2,
    width_ratios=[
        0.92,
        1.08
    ],
    height_ratios=[
        1,
        1.13
    ],
    wspace=0.34,
    hspace=0.38
)

axA = fig.add_subplot(
    gs[0, 0]
)

axB = fig.add_subplot(
    gs[0, 1]
)

axC = fig.add_subplot(
    gs[1, 0]
)

axD = fig.add_subplot(
    gs[1, 1]
)


# ----------------------------------------------------------------------
# 2A Internal AUROC
# ----------------------------------------------------------------------

models_seq_internal = [
    "Context",
    "NT-v2",
    "Length+GC",
    "4-mer"
]

y = np.arange(
    len(
        models_seq_internal
    )
)[::-1]

for yi, model in zip(
    y,
    models_seq_internal
):

    val = float(
        gene_perf.loc[
            model,
            GENE_AUROC_COL
        ]
    )

    axA.scatter(
        val,
        yi,
        s=70,
        color=COLORS[
            model
        ],
        edgecolor="white",
        linewidth=0.8,
        zorder=3
    )

    axA.text(
        val + 0.005,
        yi,
        f"{val:.3f}",
        va="center",
        fontsize=8.4
    )

axA.set_yticks(
    y
)

axA.set_yticklabels(
    models_seq_internal
)

axA.set_xlim(
    0.74,
    0.98
)

axA.set_xlabel(
    "Gene-level AUROC"
)

axA.set_title(
    "Internal discrimination"
)

clean_axis(
    axA,
    grid_axis="x"
)

panel_label(
    axA,
    "A"
)


# ----------------------------------------------------------------------
# 2B Internal AP
# ----------------------------------------------------------------------

for yi, model in zip(
    y,
    models_seq_internal
):

    val = float(
        gene_perf.loc[
            model,
            GENE_AP_COL
        ]
    )

    axB.scatter(
        val,
        yi,
        s=70,
        color=COLORS[
            model
        ],
        edgecolor="white",
        linewidth=0.8,
        zorder=3
    )

    axB.text(
        val + 0.004,
        yi,
        f"{val:.3f}",
        va="center",
        fontsize=8.4
    )

axB.set_yticks(
    y
)

axB.set_yticklabels(
    models_seq_internal
)

axB.set_xlim(
    0.02,
    0.205
)

axB.set_xlabel(
    "Gene-level average precision"
)

axB.set_title(
    "Internal precision–recall performance"
)

clean_axis(
    axB,
    grid_axis="x"
)

panel_label(
    axB,
    "B"
)


# ----------------------------------------------------------------------
# 2C Broad temporal sequence-control performance
# ----------------------------------------------------------------------

broad9d = (
    stage9d_cont.loc[
        stage9d_cont[
            "external_set"
        ].astype(str)
        == "Tan_All245_nonseed"
    ]
    .copy()
)

broad9d = broad9d.loc[
    broad9d[
        "model_clean"
    ].isin(
        SEQ_ORDER
    )
]

broad_map = (
    broad9d
    .set_index(
        "model_clean"
    )[
        "mean_rank_percentile"
    ]
    .to_dict()
)

yc = np.arange(
    len(
        SEQ_ORDER
    )
)[::-1]

for yi, model in zip(
    yc,
    SEQ_ORDER
):

    value = float(
        broad_map[
            model
        ]
    )

    axC.scatter(
        value,
        yi,
        s=78,
        color=COLORS[
            model
        ],
        edgecolor="white",
        linewidth=0.8,
        zorder=3
    )

    axC.text(
        value + 0.9,
        yi,
        f"{value:.1f}",
        va="center",
        fontsize=8.4
    )

axC.set_yticks(
    yc
)

axC.set_yticklabels(
    SEQ_ORDER
)

axC.set_xlim(
    60,
    82
)

axC.set_xlabel(
    "Mean rank percentile"
)

axC.set_title(
    "Temporal validation: broad 2026 set (n=134)"
)

clean_axis(
    axC,
    grid_axis="x"
)

panel_label(
    axC,
    "C"
)


# ----------------------------------------------------------------------
# 2D External paired NT-v2 vs sequence-control effects
# ----------------------------------------------------------------------

boot9d = stage9d_boot.copy()

def normalize_sequence_comparison(
    value
):
    s = (
        str(value)
        .lower()
        .replace("_", "")
        .replace("-", "")
        .replace(" ", "")
    )

    if "lengthgc" in s:
        return "NT-v2 − Length+GC"

    if (
        "fourmer" in s
        or "4mer" in s
    ):
        return "NT-v2 − 4-mer"

    return str(value)


boot9d[
    "comparison_clean"
] = boot9d[
    "comparison"
].map(
    normalize_sequence_comparison
)

set_base_y = {
    ext: yval
    for ext, yval
    in zip(
        EXT_ORDER,
        np.arange(
            len(
                EXT_ORDER
            )
        )[::-1]
    )
}

comparison_styles = {
    "NT-v2 − Length+GC": {
        "offset": 0.12,
        "color": COLORS[
            "Length+GC"
        ],
        "marker": "o"
    },
    "NT-v2 − 4-mer": {
        "offset": -0.12,
        "color": COLORS[
            "4-mer"
        ],
        "marker": "s"
    }
}

for comparison, style in comparison_styles.items():

    dfc = boot9d.loc[
        boot9d[
            "comparison_clean"
        ] == comparison
    ]

    for _, row in dfc.iterrows():

        ext = str(
            row[
                "external_set"
            ]
        )

        if ext not in set_base_y:
            continue

        yy = (
            set_base_y[
                ext
            ]
            + style[
                "offset"
            ]
        )

        est = float(
            row[
                "observed_mean_rank_percentile_delta"
            ]
        )

        lo = float(
            row[
                "CI95_low"
            ]
        )

        hi = float(
            row[
                "CI95_high"
            ]
        )

        axD.errorbar(
            est,
            yy,
            xerr=ci_error(
                est,
                lo,
                hi
            ),
            fmt=style[
                "marker"
            ],
            color=style[
                "color"
            ],
            ecolor=style[
                "color"
            ],
            markersize=5.5,
            capsize=2.8,
            linewidth=1.2,
            zorder=3
        )

axD.axvline(
    0,
    color="#333333",
    linestyle="--",
    linewidth=0.9,
    zorder=1
)

axD.set_yticks(
    [
        set_base_y[
            e
        ]
        for e in EXT_ORDER
    ]
)

axD.set_yticklabels(
    [
        EXT_LABEL[
            e
        ]
        for e in EXT_ORDER
    ]
)

axD.set_xlabel(
    "Mean rank-percentile difference\n(positive favors NT-v2)"
)

axD.set_title(
    "Paired temporal sequence-control comparison"
)

clean_axis(
    axD,
    grid_axis="x"
)

panel_label(
    axD,
    "D"
)

handles = [
    Line2D(
        [0],
        [0],
        marker="o",
        linestyle="none",
        markerfacecolor=COLORS[
            "Length+GC"
        ],
        markeredgecolor=COLORS[
            "Length+GC"
        ],
        label="NT-v2 − Length+GC"
    ),
    Line2D(
        [0],
        [0],
        marker="s",
        linestyle="none",
        markerfacecolor=COLORS[
            "4-mer"
        ],
        markeredgecolor=COLORS[
            "4-mer"
        ],
        label="NT-v2 − 4-mer"
    )
]

axD.legend(
    handles=handles,
    frameon=False,
    loc="lower right"
)

save_figure(
    fig,
    "Figure2_NTv2_Sequence_Signal_and_Controls"
)


# ======================================================================================
# FIGURE 3
# HOW MUCH DOES NT-v2 ADD BEYOND CONTEXT?
# ======================================================================================

fig = plt.figure(
    figsize=(
        12.8,
        8.0
    )
)

gs = fig.add_gridspec(
    2,
    2,
    wspace=0.34,
    hspace=0.42
)

axA = fig.add_subplot(
    gs[0, 0]
)

axB = fig.add_subplot(
    gs[0, 1]
)

axC = fig.add_subplot(
    gs[1, 0]
)

axD = fig.add_subplot(
    gs[1, 1]
)


# ----------------------------------------------------------------------
# 3A AUROC dumbbell
# ----------------------------------------------------------------------

context_auc = float(
    gene_perf.loc[
        "Context",
        GENE_AUROC_COL
    ]
)

fusion_auc = float(
    gene_perf.loc[
        "Fusion",
        GENE_AUROC_COL
    ]
)

axA.plot(
    [
        context_auc,
        fusion_auc
    ],
    [
        0,
        0
    ],
    color="#777777",
    linewidth=2.1,
    zorder=1
)

axA.scatter(
    context_auc,
    0,
    s=105,
    color=COLORS[
        "Context"
    ],
    edgecolor="white",
    linewidth=1,
    zorder=3
)

axA.scatter(
    fusion_auc,
    0,
    s=105,
    color=COLORS[
        "Fusion"
    ],
    edgecolor="white",
    linewidth=1,
    zorder=3
)

axA.text(
    context_auc,
    0.11,
    f"Context\n{context_auc:.4f}",
    ha="center",
    fontsize=8.5
)

axA.text(
    fusion_auc,
    -0.16,
    f"Fusion\n{fusion_auc:.4f}",
    ha="center",
    fontsize=8.5
)

axA.set_ylim(
    -0.35,
    0.35
)

axA.set_yticks(
    []
)

axA.set_xlim(
    0.944,
    0.954
)

axA.set_xlabel(
    "Gene-level AUROC"
)

axA.set_title(
    "Internal AUROC gain (expanded axis)"
)

clean_axis(
    axA,
    grid_axis="x"
)

panel_label(
    axA,
    "A"
)


# ----------------------------------------------------------------------
# 3B AP dumbbell
# ----------------------------------------------------------------------

context_ap = float(
    gene_perf.loc[
        "Context",
        GENE_AP_COL
    ]
)

fusion_ap = float(
    gene_perf.loc[
        "Fusion",
        GENE_AP_COL
    ]
)

axB.plot(
    [
        context_ap,
        fusion_ap
    ],
    [
        0,
        0
    ],
    color="#777777",
    linewidth=2.1,
    zorder=1
)

axB.scatter(
    context_ap,
    0,
    s=105,
    color=COLORS[
        "Context"
    ],
    edgecolor="white",
    linewidth=1,
    zorder=3
)

axB.scatter(
    fusion_ap,
    0,
    s=105,
    color=COLORS[
        "Fusion"
    ],
    edgecolor="white",
    linewidth=1,
    zorder=3
)

axB.text(
    context_ap,
    0.11,
    f"Context\n{context_ap:.4f}",
    ha="center",
    fontsize=8.5
)

axB.text(
    fusion_ap,
    -0.16,
    f"Fusion\n{fusion_ap:.4f}",
    ha="center",
    fontsize=8.5
)

axB.set_ylim(
    -0.35,
    0.35
)

axB.set_yticks(
    []
)

axB.set_xlim(
    0.175,
    0.203
)

axB.set_xlabel(
    "Gene-level average precision"
)

axB.set_title(
    "Internal AP gain (expanded axis)"
)

clean_axis(
    axB,
    grid_axis="x"
)

panel_label(
    axB,
    "B"
)


# ----------------------------------------------------------------------
# 3C Internal paired inference
# ----------------------------------------------------------------------

metric_y = {
    "AUROC": 1,
    "AP": 0
}

for _, row in INTERNAL_EFFECTS.iterrows():

    metric = row[
        "metric"
    ]

    est = float(
        row[
            "estimate"
        ]
    )

    lo = float(
        row[
            "low"
        ]
    )

    hi = float(
        row[
            "high"
        ]
    )

    axC.errorbar(
        est,
        metric_y[
            metric
        ],
        xerr=ci_error(
            est,
            lo,
            hi
        ),
        fmt="o",
        color=COLORS[
            "Fusion"
        ],
        ecolor=COLORS[
            "Fusion"
        ],
        markersize=6.5,
        capsize=3,
        linewidth=1.35,
        zorder=3
    )

    axC.text(
        hi + 0.0018,
        metric_y[
            metric
        ],
        (
            f"Δ={est:+.4f}\n"
            f"[{lo:+.4f}, {hi:+.4f}]"
        ),
        va="center",
        fontsize=7.8
    )

axC.axvline(
    0,
    color="#333333",
    linestyle="--",
    linewidth=0.9
)

axC.set_yticks(
    [
        1,
        0
    ]
)

axC.set_yticklabels(
    [
        "AUROC",
        "Average precision"
    ]
)

axC.set_xlim(
    -0.018,
    0.052
)

axC.set_xlabel(
    "Fusion − Context"
)

axC.set_title(
    "Gene-level paired bootstrap"
)

clean_axis(
    axC,
    grid_axis="x"
)

panel_label(
    axC,
    "C"
)


# ----------------------------------------------------------------------
# 3D Full vs disease-context-reduced robustness
# ----------------------------------------------------------------------

metric_base = {
    "AUROC": 1,
    "AP": 0
}

setting_style = {
    "Full Context": {
        "offset": 0.12,
        "color": COLORS[
            "Context"
        ],
        "marker": "o"
    },
    "Disease-context-reduced": {
        "offset": -0.12,
        "color": "#8C6BB1",
        "marker": "s"
    }
}

for setting, style in setting_style.items():

    dfset = ROBUSTNESS_EFFECTS.loc[
        ROBUSTNESS_EFFECTS[
            "setting"
        ] == setting
    ]

    for _, row in dfset.iterrows():

        metric = row[
            "metric"
        ]

        yy = (
            metric_base[
                metric
            ]
            + style[
                "offset"
            ]
        )

        est = float(
            row[
                "estimate"
            ]
        )

        lo = float(
            row[
                "low"
            ]
        )

        hi = float(
            row[
                "high"
            ]
        )

        axD.errorbar(
            est,
            yy,
            xerr=ci_error(
                est,
                lo,
                hi
            ),
            fmt=style[
                "marker"
            ],
            color=style[
                "color"
            ],
            ecolor=style[
                "color"
            ],
            markersize=6,
            capsize=3,
            linewidth=1.3,
            zorder=3
        )

axD.axvline(
    0,
    color="#333333",
    linestyle="--",
    linewidth=0.9
)

axD.set_yticks(
    [
        1,
        0
    ]
)

axD.set_yticklabels(
    [
        "AUROC",
        "Average precision"
    ]
)

axD.set_xlim(
    -0.018,
    0.052
)

axD.set_xlabel(
    "Fusion − corresponding Context model"
)

axD.set_title(
    "Complementarity after disease-context reduction"
)

clean_axis(
    axD,
    grid_axis="x"
)

panel_label(
    axD,
    "D"
)

robust_handles = [
    Line2D(
        [0],
        [0],
        marker="o",
        linestyle="none",
        color=COLORS[
            "Context"
        ],
        label="Full Context"
    ),
    Line2D(
        [0],
        [0],
        marker="s",
        linestyle="none",
        color="#8C6BB1",
        label="Disease-context-reduced"
    )
]

axD.legend(
    handles=robust_handles,
    frameon=False,
    loc="lower right"
)

save_figure(
    fig,
    "Figure3_Incremental_Value_and_Robustness"
)


# ======================================================================================
# FIGURE 4
# INTERNAL COMPLEMENTARITY vs TEMPORAL REALITY
# ======================================================================================

fig = plt.figure(
    figsize=(
        13.0,
        9.0
    )
)

gs = fig.add_gridspec(
    2,
    2,
    height_ratios=[
        1.08,
        0.92
    ],
    wspace=0.34,
    hspace=0.40
)

axA = fig.add_subplot(
    gs[0, 0]
)

axB = fig.add_subplot(
    gs[0, 1]
)

axC = fig.add_subplot(
    gs[1, 0]
)

axD = fig.add_subplot(
    gs[1, 1]
)


# ----------------------------------------------------------------------
# 4A Mean rank percentile across temporal sets
# ----------------------------------------------------------------------

base_positions = {
    ext: yval
    for ext, yval
    in zip(
        EXT_ORDER,
        np.arange(
            len(
                EXT_ORDER
            )
        )[::-1]
    )
}

model_offsets = {
    "Context": 0.18,
    "NT-v2": 0.00,
    "Fusion": -0.18
}

for model in MODEL_ORDER:

    dmodel = stage9c_cont.loc[
        stage9c_cont[
            "model_clean"
        ] == model
    ]

    for _, row in dmodel.iterrows():

        ext = str(
            row[
                "external_set"
            ]
        )

        if ext not in base_positions:
            continue

        yy = (
            base_positions[
                ext
            ]
            + model_offsets[
                model
            ]
        )

        value = float(
            row[
                "mean_rank_percentile"
            ]
        )

        axA.scatter(
            value,
            yy,
            s=50,
            color=COLORS[
                model
            ],
            edgecolor="white",
            linewidth=0.7,
            zorder=3
        )

axA.set_yticks(
    [
        base_positions[
            e
        ]
        for e in EXT_ORDER
    ]
)

axA.set_yticklabels(
    [
        EXT_LABEL[
            e
        ]
        for e in EXT_ORDER
    ]
)

axA.set_xlim(
    52,
    88
)

axA.set_xlabel(
    "Mean rank percentile"
)

axA.set_title(
    "Post-benchmark ASD gene prioritization"
)

clean_axis(
    axA,
    grid_axis="x"
)

panel_label(
    axA,
    "A"
)

temporal_handles = [
    Line2D(
        [0],
        [0],
        marker="o",
        linestyle="none",
        color=COLORS[
            model
        ],
        label=model
    )
    for model in MODEL_ORDER
]

axA.legend(
    handles=temporal_handles,
    frameon=False,
    loc="lower right"
)


# ----------------------------------------------------------------------
# 4B Fusion vs Context temporal effect
# ----------------------------------------------------------------------

delta_col = find_col(
    stage9c_boot,
    [
        "observed_mean_Fusion_minus_Context_rank_percentile",
        "observed_mean_rank_percentile_delta"
    ]
)

for ext in EXT_ORDER:

    d = stage9c_boot.loc[
        stage9c_boot[
            "external_set"
        ].astype(str)
        == ext
    ]

    assert len(d) >= 1, (
        f"Missing Stage9C bootstrap row: {ext}"
    )

    row = d.iloc[0]

    est = float(
        row[
            delta_col
        ]
    )

    lo = float(
        row[
            "CI95_low"
        ]
    )

    hi = float(
        row[
            "CI95_high"
        ]
    )

    yy = base_positions[
        ext
    ]

    axB.errorbar(
        est,
        yy,
        xerr=ci_error(
            est,
            lo,
            hi
        ),
        fmt="o",
        color=COLORS[
            "Fusion"
        ],
        ecolor=COLORS[
            "Fusion"
        ],
        markersize=6.5,
        capsize=3,
        linewidth=1.3,
        zorder=3
    )

axB.axvline(
    0,
    color="#333333",
    linestyle="--",
    linewidth=0.9
)

axB.set_yticks(
    [
        base_positions[
            e
        ]
        for e in EXT_ORDER
    ]
)

axB.set_yticklabels(
    [
        EXT_LABEL[
            e
        ]
        for e in EXT_ORDER
    ]
)

axB.set_xlabel(
    "Fusion − Context mean rank percentile"
)

axB.set_title(
    "Incremental temporal value of fusion"
)

clean_axis(
    axB,
    grid_axis="x"
)

panel_label(
    axB,
    "B"
)


# ----------------------------------------------------------------------
# Function for Top-k temporal recovery panels
# ----------------------------------------------------------------------

def plot_topk_recovery(
    ax,
    external_set,
    title
):

    subset = stage9c_topk.loc[
        stage9c_topk[
            "external_set"
        ].astype(str)
        == external_set
    ].copy()

    xvals = np.array(
        [
            1,
            5,
            10
        ],
        dtype=float
    )

    for model in MODEL_ORDER:

        d = subset.loc[
            subset[
                "model_clean"
            ] == model
        ].copy()

        d = (
            d.sort_values(
                "threshold_pct"
            )
        )

        vals = []

        for threshold in [
            1,
            5,
            10
        ]:

            dt = d.loc[
                d[
                    "threshold_pct"
                ] == threshold
            ]

            assert len(dt) >= 1, (
                f"Missing {external_set}, "
                f"{model}, top {threshold}%"
            )

            vals.append(
                float(
                    dt.iloc[0][
                        "recovery_fraction"
                    ]
                )
            )

        ax.plot(
            xvals,
            vals,
            marker="o",
            markersize=5.5,
            linewidth=1.7,
            color=COLORS[
                model
            ],
            label=model
        )

    # Random expectation:
    # If X% of all genes are selected,
    # expected recovery fraction under random ranking = X%.
    ax.plot(
        xvals,
        xvals / 100.0,
        linestyle="--",
        linewidth=1.1,
        color="#777777",
        label="Random expectation"
    )

    ax.set_xticks(
        xvals
    )

    ax.set_xticklabels(
        [
            "Top 1%",
            "Top 5%",
            "Top 10%"
        ]
    )

    ax.set_ylabel(
        "External-gene recovery fraction"
    )

    ax.set_title(
        title
    )

    clean_axis(
        ax,
        grid_axis="y"
    )


# ----------------------------------------------------------------------
# 4C Broad Top-k
# ----------------------------------------------------------------------

plot_topk_recovery(
    axC,
    "Tan_All245_nonseed",
    "Broad temporal set (n=134)"
)

axC.set_ylim(
    0,
    0.64
)

panel_label(
    axC,
    "C"
)

axC.legend(
    frameon=False,
    loc="upper left"
)


# ----------------------------------------------------------------------
# 4D No-evidence Top-k
# ----------------------------------------------------------------------

plot_topk_recovery(
    axD,
    "Tan_NoEvidence32_nonseed",
    "No-evidence temporal set (n=30)"
)

axD.set_ylim(
    0,
    0.30
)

panel_label(
    axD,
    "D"
)

# avoid duplicated legend
leg = axD.get_legend()
if leg is not None:
    leg.remove()

save_figure(
    fig,
    "Figure4_Temporal_Validation_and_External_Recovery"
)


# ======================================================================================
# FIGURE 5
# HISTORICAL COMPARATOR BENCHMARK
# ======================================================================================

fig, axes = plt.subplots(
    1,
    2,
    figsize=(
        13.0,
        6.3
    )
)

comparators = [
    "DeepND",
    "Krishnan",
    "DAWN",
    "DAMAGES",
    "pLI"
]


def historical_forest(
    ax,
    external_set,
    panel_title,
    label
):

    d = stage9e_boot.loc[
        stage9e_boot[
            "external_set"
        ].astype(str)
        == external_set
    ].copy()

    ybase = {
        comparator: y
        for comparator, y
        in zip(
            comparators,
            np.arange(
                len(
                    comparators
                )
            )[::-1]
        )
    }

    offsets = {
        "Context": 0.12,
        "Fusion": -0.12
    }

    for model in [
        "Context",
        "Fusion"
    ]:

        dm = d.loc[
            d[
                "our_model_clean"
            ] == model
        ]

        for comparator in comparators:

            rowdf = dm.loc[
                dm[
                    "comparator"
                ].astype(str)
                == comparator
            ]

            if len(
                rowdf
            ) == 0:
                continue

            row = rowdf.iloc[0]

            est = float(
                row[
                    "observed_mean_rank_percentile_delta"
                ]
            )

            lo = float(
                row[
                    "CI95_low"
                ]
            )

            hi = float(
                row[
                    "CI95_high"
                ]
            )

            yy = (
                ybase[
                    comparator
                ]
                + offsets[
                    model
                ]
            )

            ax.errorbar(
                est,
                yy,
                xerr=ci_error(
                    est,
                    lo,
                    hi
                ),
                fmt=(
                    "o"
                    if model
                    == "Context"
                    else "s"
                ),
                color=COLORS[
                    model
                ],
                ecolor=COLORS[
                    model
                ],
                markersize=5.8,
                capsize=2.8,
                linewidth=1.2,
                zorder=3
            )

    labels = []

    for comparator in comparators:

        dc = d.loc[
            d[
                "comparator"
            ].astype(str)
            == comparator
        ]

        n_text = ""

        if (
            len(dc)
            and
            "n_evaluable_external"
            in dc.columns
        ):

            n = int(
                dc.iloc[0][
                    "n_evaluable_external"
                ]
            )

            n_text = f" (n={n})"

        labels.append(
            comparator
            + n_text
        )

    ax.axvline(
        0,
        color="#333333",
        linestyle="--",
        linewidth=0.9
    )

    ax.set_yticks(
        [
            ybase[
                c
            ]
            for c in comparators
        ]
    )

    ax.set_yticklabels(
        labels
    )

    ax.set_xlabel(
        "Our model − historical comparator\n"
        "mean rank percentile"
    )

    ax.set_title(
        panel_title
    )

    clean_axis(
        ax,
        grid_axis="x"
    )

    panel_label(
        ax,
        label
    )

    # Dynamic symmetric-ish limits
    low_min = float(
        d[
            "CI95_low"
        ].min()
    )

    high_max = float(
        d[
            "CI95_high"
        ].max()
    )

    margin = (
        high_max
        - low_min
    ) * 0.09

    ax.set_xlim(
        low_min - margin,
        high_max + margin
    )


historical_forest(
    axes[0],
    "Tan_All245_nonseed",
    "Broad post-benchmark ASD genes",
    "A"
)

historical_forest(
    axes[1],
    "Tan_NoEvidence32_nonseed",
    "No-evidence post-benchmark ASD genes",
    "B"
)

hist_handles = [
    Line2D(
        [0],
        [0],
        marker="o",
        linestyle="none",
        color=COLORS[
            "Context"
        ],
        label="Context"
    ),
    Line2D(
        [0],
        [0],
        marker="s",
        linestyle="none",
        color=COLORS[
            "Fusion"
        ],
        label="Fusion"
    )
]

axes[0].legend(
    handles=hist_handles,
    frameon=False,
    loc="lower right"
)

fig.text(
    0.50,
    0.012,
    (
        "Positive values indicate higher ranking of later ASD genes "
        "by Context/Fusion than by the historical comparator."
    ),
    ha="center",
    fontsize=8.2
)

save_figure(
    fig,
    "Figure5_Historical_Comparator_Benchmark"
)


# ======================================================================================
# SUPPLEMENTARY FIGURE S1
# FUSION META-MODEL COEFFICIENTS
# ======================================================================================

fig, ax = plt.subplots(
    figsize=(
        6.5,
        4.3
    )
)

y = np.arange(
    len(
        META_COEFS
    )
)[::-1]

for yi, (_, row) in zip(
    y,
    META_COEFS.iterrows()
):

    branch = row[
        "branch"
    ]

    mean = float(
        row[
            "mean"
        ]
    )

    sd = float(
        row[
            "sd"
        ]
    )

    ax.errorbar(
        mean,
        yi,
        xerr=sd,
        fmt="o",
        markersize=8,
        capsize=4,
        linewidth=1.4,
        color=COLORS[
            branch
        ],
        ecolor=COLORS[
            branch
        ]
    )

    ax.text(
        mean + sd + 0.10,
        yi,
        f"{mean:.2f} ± {sd:.2f}",
        va="center",
        fontsize=8.5
    )

ax.set_yticks(
    y
)

ax.set_yticklabels(
    META_COEFS[
        "branch"
    ]
)

ax.set_xlabel(
    "Late-fusion logistic coefficient"
)

ax.set_title(
    "Relative contribution to strict late fusion"
)

clean_axis(
    ax,
    grid_axis="x"
)

save_figure(
    fig,
    "Supplementary_Figure_S1_Fusion_Meta_Coefficients"
)


# ======================================================================================
# SUPPLEMENTARY FIGURE S2
# SPLIT-LEVEL FIVE-MODEL PERFORMANCE
# ======================================================================================

SPLIT_MODEL_COL = find_col(
    split_summary,
    [
        "Model"
    ]
)

SPLIT_AUC_COL = find_col(
    split_summary,
    [
        "AUROC_mean",
        "AUROC"
    ]
)

SPLIT_AP_COL = find_col(
    split_summary,
    [
        "AP_mean",
        "average precision"
    ]
)

split_summary[
    "Model_clean"
] = split_summary[
    SPLIT_MODEL_COL
].map(
    normalize_model
)

split_perf = (
    split_summary
    .set_index(
        "Model_clean"
    )
)

all_models = [
    "Context",
    "NT-v2",
    "Length+GC",
    "4-mer",
    "Fusion"
]

fig, axes = plt.subplots(
    1,
    2,
    figsize=(
        10.5,
        4.8
    )
)

ys = np.arange(
    len(
        all_models
    )
)[::-1]

for yi, model in zip(
    ys,
    all_models
):

    auc = float(
        split_perf.loc[
            model,
            SPLIT_AUC_COL
        ]
    )

    ap = float(
        split_perf.loc[
            model,
            SPLIT_AP_COL
        ]
    )

    axes[0].scatter(
        auc,
        yi,
        s=60,
        color=COLORS[
            model
        ],
        edgecolor="white",
        linewidth=0.7
    )

    axes[1].scatter(
        ap,
        yi,
        s=60,
        color=COLORS[
            model
        ],
        edgecolor="white",
        linewidth=0.7
    )

axes[0].set_yticks(
    ys
)

axes[0].set_yticklabels(
    all_models
)

axes[1].set_yticks(
    ys
)

axes[1].set_yticklabels(
    all_models
)

axes[0].set_xlabel(
    "Mean split-level AUROC"
)

axes[1].set_xlabel(
    "Mean split-level average precision"
)

axes[0].set_title(
    "AUROC across 7,900 frozen test folds"
)

axes[1].set_title(
    "AP across 7,900 frozen test folds"
)

for axx in axes:
    clean_axis(
        axx,
        grid_axis="x"
    )

panel_label(
    axes[0],
    "A"
)

panel_label(
    axes[1],
    "B"
)

save_figure(
    fig,
    "Supplementary_Figure_S2_Split_Level_Performance"
)


# ======================================================================================
# SUPPLEMENTARY FIGURE S3
# FULL vs DISEASE-CONTEXT-REDUCED ABSOLUTE PERFORMANCE
# ======================================================================================

fig, axes = plt.subplots(
    1,
    2,
    figsize=(
        10.5,
        4.8
    )
)

settings = [
    "Full Context",
    "Full Fusion",
    "Reduced Context",
    "Reduced Fusion"
]

setting_colors = {
    "Full Context":
        COLORS[
            "Context"
        ],
    "Full Fusion":
        COLORS[
            "Fusion"
        ],
    "Reduced Context":
        "#6BAED6",
    "Reduced Fusion":
        "#74C69D"
}

ys = np.arange(
    len(
        settings
    )
)[::-1]

for yi, setting in zip(
    ys,
    settings
):

    row = GENERIC_ABSOLUTE.loc[
        GENERIC_ABSOLUTE[
            "setting"
        ] == setting
    ].iloc[0]

    axes[0].scatter(
        float(
            row[
                "AUROC"
            ]
        ),
        yi,
        s=65,
        color=setting_colors[
            setting
        ],
        edgecolor="white",
        linewidth=0.8
    )

    axes[1].scatter(
        float(
            row[
                "AP"
            ]
        ),
        yi,
        s=65,
        color=setting_colors[
            setting
        ],
        edgecolor="white",
        linewidth=0.8
    )

axes[0].set_yticks(
    ys
)

axes[0].set_yticklabels(
    settings
)

axes[1].set_yticks(
    ys
)

axes[1].set_yticklabels(
    settings
)

axes[0].set_xlabel(
    "Gene-level AUROC"
)

axes[1].set_xlabel(
    "Gene-level average precision"
)

axes[0].set_title(
    "Absolute AUROC"
)

axes[1].set_title(
    "Absolute average precision"
)

axes[0].set_xlim(
    0.940,
    0.954
)

axes[1].set_xlim(
    0.165,
    0.205
)

for axx in axes:
    clean_axis(
        axx,
        grid_axis="x"
    )

panel_label(
    axes[0],
    "A"
)

panel_label(
    axes[1],
    "B"
)

save_figure(
    fig,
    "Supplementary_Figure_S3_Disease_Context_Reduced_Performance"
)


# ======================================================================================
# SUPPLEMENTARY FIGURE S4
# FULL EXTERNAL SEQUENCE-CONTROL RANK PERFORMANCE
#
# Useful if journal allows >3 supplementary figures.
# ======================================================================================

fig, ax = plt.subplots(
    figsize=(
        8.5,
        5.8
    )
)

positions = {
    ext: y
    for ext, y
    in zip(
        EXT_ORDER,
        np.arange(
            len(
                EXT_ORDER
            )
        )[::-1]
    )
}

seq_offsets = {
    "NT-v2": 0.18,
    "Length+GC": 0.00,
    "4-mer": -0.18
}

for model in SEQ_ORDER:

    dm = stage9d_cont.loc[
        stage9d_cont[
            "model_clean"
        ] == model
    ]

    for _, row in dm.iterrows():

        ext = str(
            row[
                "external_set"
            ]
        )

        if ext not in positions:
            continue

        ax.scatter(
            float(
                row[
                    "mean_rank_percentile"
                ]
            ),
            (
                positions[
                    ext
                ]
                + seq_offsets[
                    model
                ]
            ),
            s=55,
            color=COLORS[
                model
            ],
            edgecolor="white",
            linewidth=0.7,
            zorder=3
        )

ax.set_yticks(
    [
        positions[
            e
        ]
        for e in EXT_ORDER
    ]
)

ax.set_yticklabels(
    [
        EXT_LABEL[
            e
        ]
        for e in EXT_ORDER
    ]
)

ax.set_xlabel(
    "Mean external-gene rank percentile"
)

ax.set_title(
    "Temporal performance of sequence representations"
)

clean_axis(
    ax,
    grid_axis="x"
)

seq_handles = [
    Line2D(
        [0],
        [0],
        marker="o",
        linestyle="none",
        color=COLORS[
            m
        ],
        label=m
    )
    for m in SEQ_ORDER
]

ax.legend(
    handles=seq_handles,
    frameon=False,
    loc="lower right"
)

save_figure(
    fig,
    "Supplementary_Figure_S4_External_Sequence_Control_Performance"
)


# ======================================================================================
# 9. WRITE FIGURE MANIFEST
# ======================================================================================

manifest = pd.DataFrame(
    [
        {
            "figure":
                "Figure 1",
            "file_stem":
                "Figure1_ASD_ContextFusion_Study_Design",
            "scientific_question":
                "Overall study design and incremental-value evaluation framework"
        },
        {
            "figure":
                "Figure 2",
            "file_stem":
                "Figure2_NTv2_Sequence_Signal_and_Controls",
            "scientific_question":
                "Does NT-v2 contain ASD-relevant sequence signal beyond simple sequence composition?"
        },
        {
            "figure":
                "Figure 3",
            "file_stem":
                "Figure3_Incremental_Value_and_Robustness",
            "scientific_question":
                "How much does NT-v2 add beyond strong Context, and is complementarity robust?"
        },
        {
            "figure":
                "Figure 4",
            "file_stem":
                "Figure4_Temporal_Validation_and_External_Recovery",
            "scientific_question":
                "Does the internal fusion advantage translate to subsequently implicated ASD genes?"
        },
        {
            "figure":
                "Figure 5",
            "file_stem":
                "Figure5_Historical_Comparator_Benchmark",
            "scientific_question":
                "How competitive are Context and Fusion against historical ASD prioritization scores?"
        },
        {
            "figure":
                "Supplementary Figure S1",
            "file_stem":
                "Supplementary_Figure_S1_Fusion_Meta_Coefficients",
            "scientific_question":
                "Relative Context and NT-v2 contribution to the fusion meta-model"
        },
        {
            "figure":
                "Supplementary Figure S2",
            "file_stem":
                "Supplementary_Figure_S2_Split_Level_Performance",
            "scientific_question":
                "Descriptive split-level performance across the 7,900 frozen test folds"
        },
        {
            "figure":
                "Supplementary Figure S3",
            "file_stem":
                "Supplementary_Figure_S3_Disease_Context_Reduced_Performance",
            "scientific_question":
                "Absolute performance after source-defined disease-context reduction"
        },
        {
            "figure":
                "Supplementary Figure S4",
            "file_stem":
                "Supplementary_Figure_S4_External_Sequence_Control_Performance",
            "scientific_question":
                "External sequence-control ranking across all temporal evidence subsets"
        }
    ]
)

manifest.to_csv(
    OUT /
    "FINAL_FIGURE_MANIFEST.tsv",
    sep="\t",
    index=False
)


# ======================================================================================
# 10. FINAL REPORT
# ======================================================================================

print(
    "\n" +
    "=" * 100
)

print(
    "FINAL PUBLICATION FIGURE SUITE COMPLETE [OK]"
)

print(
    "=" * 100
)

print(
    "\nOutput directory:"
)

print(
    OUT
)

print(
    "\nMAIN FIGURES:"
)

print(
    "  Figure 1 -> Study design"
)

print(
    "  Figure 2 -> NT-v2 sequence signal vs explicit sequence controls"
)

print(
    "  Figure 3 -> Incremental value beyond Context + robustness"
)

print(
    "  Figure 4 -> Temporal validation / internal-vs-external reality"
)

print(
    "  Figure 5 -> Historical comparator benchmark"
)

print(
    "\nSUPPLEMENTARY:"
)

print(
    "  S1 -> Fusion meta coefficients"
)

print(
    "  S2 -> Split-level five-model performance"
)

print(
    "  S3 -> Disease-context-reduced absolute performance"
)

print(
    "  S4 -> External sequence-control performance"
)

print(
    "\nFormats:"
)

print(
    "  PDF : vector publication master"
)

print(
    "  SVG : editable vector"
)

print(
    f"  PNG : {PNG_DPI} dpi"
)

print(
    "\nSCIENTIFIC BEHAVIOR:"
)

print(
    "  New model fitting          : NO"
)

print(
    "  New PU sampling            : NO"
)

print(
    "  Hyperparameter tuning      : NO"
)

print(
    "  External gene reselection  : NO"
)

print(
    "  New analysis               : NO"
)

print(
    "  Plotting existing results  : YES"
)

print(
    "=" * 100
)
```

```python
# ======================================================================================
# 5. LOAD EXACT LOCKED STAGE 9B3 NoEvidence30 FILE
#
# ENGINEERING/RUNTIME-ONLY FIX
# Scientific behavior unchanged.
#
# We already know the canonical Stage9B3 file.
# Do NOT rediscover/rederive the external set from Tan Table 9.
#
# This block:
#   1. loads the exact locked Stage9B3 file
#   2. identifies the PROJECT identity column by overlap with the already-frozen
#      Stage9C 18,062-gene non-seed universe
#   3. requires EXACTLY 30 locked project identities
#   4. stops if this cannot be established uniquely/safely
# ======================================================================================

NOEVIDENCE_FILE = (
    SOURCE_LOCK_DIR /
    "Stage9B3_Tan_SupplementNoEvidence32_NONSEED_PROJECT_IDENTITIES_LOCKED.tsv"
)

assert NOEVIDENCE_FILE.exists(), (
    "Canonical Stage9B3 NoEvidence project-identity lock is missing:\n"
    f"{NOEVIDENCE_FILE}"
)

noevidence_locked = pd.read_csv(
    NOEVIDENCE_FILE,
    sep="\t",
    low_memory=False
)

print(
    "\nLOCKED STAGE9B3 NoEvidence FILE:"
)

print(
    NOEVIDENCE_FILE
)

print(
    "Shape:",
    noevidence_locked.shape
)

print(
    "\nColumns:"
)

for i, col in enumerate(
    noevidence_locked.columns
):
    print(
        f"{i:>3} | {col}"
    )


# ----------------------------------------------------------------------
# Frozen Stage9C non-seed universe already loaded above
# ----------------------------------------------------------------------

rank_gene_set = set(
    ranks[
        gene_col
    ]
    .map(clean_gene)
    .dropna()
)


# ----------------------------------------------------------------------
# Audit EVERY Stage9B3 column against frozen project identities
#
# We are NOT constructing the gene set.
# We are identifying which column stores the already-locked PROJECT identity.
# ----------------------------------------------------------------------

column_audit = []

for col in noevidence_locked.columns:

    vals = (
        noevidence_locked[
            col
        ]
        .map(clean_gene)
        .dropna()
    )

    unique_vals = set(
        vals
    )

    overlap = (
        unique_vals
        &
        rank_gene_set
    )

    name = str(
        col
    ).lower()

    # Prefer explicit project/gene identity columns;
    # penalize obvious source-only columns.
    name_score = 0

    if "project" in name:
        name_score += 20

    if "gene" in name:
        name_score += 10

    if "identity" in name:
        name_score += 8

    if "resolved" in name:
        name_score += 6

    if "mapped" in name:
        name_score += 5

    if "symbol" in name:
        name_score += 4

    if "source" in name:
        name_score -= 10

    if "tan" in name:
        name_score -= 5

    column_audit.append(
        {
            "column":
                col,

            "unique_nonmissing":
                len(
                    unique_vals
                ),

            "project_universe_overlap":
                len(
                    overlap
                ),

            "name_score":
                name_score
        }
    )


column_audit = pd.DataFrame(
    column_audit
).sort_values(
    [
        "project_universe_overlap",
        "name_score"
    ],
    ascending=[
        False,
        False
    ]
)

print(
    "\nCOLUMN -> FROZEN PROJECT-UNIVERSE AUDIT:"
)

print(
    column_audit.to_string(
        index=False
    )
)


# ----------------------------------------------------------------------
# Valid project-identity columns MUST recover exactly the locked 30 genes.
# ----------------------------------------------------------------------

valid_columns = column_audit.loc[
    column_audit[
        "project_universe_overlap"
    ] == 30
].copy()

if len(
    valid_columns
) == 0:

    raise RuntimeError(
        "\nNo Stage9B3 column contains exactly 30 identities from the "
        "frozen 18,062-gene non-seed universe.\n"
        "STOPPING rather than altering/reconstructing the external set."
    )


# Highest semantic score among columns with exact 30-project-gene overlap
valid_columns = valid_columns.sort_values(
    [
        "name_score",
        "unique_nonmissing"
    ],
    ascending=[
        False,
        True
    ]
)

NOEVIDENCE_GENE_COL = valid_columns.iloc[
    0
][
    "column"
]


# ----------------------------------------------------------------------
# Extract ONLY identities already represented in frozen Stage9C universe.
# ----------------------------------------------------------------------

raw_locked_values = (
    noevidence_locked[
        NOEVIDENCE_GENE_COL
    ]
    .map(clean_gene)
    .dropna()
)

noevidence_genes = sorted(
    set(
        raw_locked_values
    )
    &
    rank_gene_set
)


# ----------------------------------------------------------------------
# HARD SCIENTIFIC QC
# ----------------------------------------------------------------------

assert len(
    noevidence_genes
) == 30, (
    f"Expected exactly 30 locked NoEvidence project genes; "
    f"found {len(noevidence_genes)}"
)

assert len(
    set(
        noevidence_genes
    )
) == 30

assert set(
    noevidence_genes
).issubset(
    rank_gene_set
)


print(
    "\n" +
    "=" * 110
)

print(
    "LOCKED NoEvidence30 PROJECT IDENTITY RECOVERY — PASS"
)

print(
    "=" * 110
)

print(
    "Canonical file :",
    NOEVIDENCE_FILE
)

print(
    "Selected column:",
    NOEVIDENCE_GENE_COL
)

print(
    "Locked genes   :",
    len(
        noevidence_genes
    )
)

print(
    "All in frozen Stage9C universe:",
    set(
        noevidence_genes
    ).issubset(
        rank_gene_set
    )
)

print(
    "\nLocked project identities:"
)

print(
    ", ".join(
        noevidence_genes
    )
)

print(
    "\nScientific behavior changed : NO"
)

print(
    "External set redefined      : NO"
)

print(
    "Genes selected from scores   : NO"
)

print(
    "Proceed to Section 6."
)
```

```python
# ======================================================================================
# ASD-ContextFusion
# STAGE 10C — TARGETED BIOLOGICAL / GENETIC INTERPRETATION
# CORRECTED FINAL VERSION
#
# SCIENTIFIC PURPOSE
# ------------------
# Focused interpretation of the already-locked Tan 2026 NoEvidence30 genes.
#
# PRIMARY:
#   - Context rank percentile
#   - NT-v2 rank percentile
#   - Strict Fusion rank percentile
#   - Tan TADA FDR
#   - Tan DeNovoWEST ASD-all FDR
#   - genetic support route
#   - optional gnomAD v4 pLI
#
# EXPLORATORY, LOCKED BEFORE LITERATURE CURATION:
#   NT-v2 sequence reprioritization:
#
#       delta = NT-v2 rank percentile - Context rank percentile
#
#   retain delta > 0
#   sort descending delta
#   tie-break Gene_Name ascending
#   keep at most 10 genes
#
# IMPORTANT
# ---------
# NO:
#   - model fitting
#   - model tuning
#   - PU resampling
#   - new embedding
#   - external-set redefinition
#   - pathway/GO enrichment
#   - literature-based candidate selection
#
# The selected candidates are exploratory sequence-reprioritized genes,
# NOT newly discovered or validated causal ASD genes.
# ======================================================================================


# ======================================================================================
# 0. IMPORTS
# ======================================================================================

from pathlib import Path
import json
import hashlib
import gc
import warnings

import numpy as np
import pandas as pd

import matplotlib
matplotlib.use("Agg")

import matplotlib.pyplot as plt
import matplotlib as mpl
from matplotlib import font_manager as fm
from matplotlib.lines import Line2D

warnings.filterwarnings("ignore")


# ======================================================================================
# 1. PATHS
# ======================================================================================

ROOT = Path(
    r"E:\ASD_ContextFusion_Q1"
)

STAGE9C_DIR = (
    ROOT /
    "05_results" /
    "proposed" /
    "stage9C_external_validation"
)

SOURCE_LOCK_DIR = (
    ROOT /
    "02_interim" /
    "external_validation_source_lock"
)

ADMIN_DIR = (
    ROOT /
    "00_admin"
)

TAN_XLSX = (
    ROOT /
    "01_raw" /
    "external_validation_2026" /
    "Tan_2026_MolecularPsychiatry_Supplementary_Tables_1_16.xlsx"
)

RANK_FILE = (
    STAGE9C_DIR /
    "Stage9C_nonseed_18062_model_ranks.tsv.gz"
)

NOEVIDENCE_FILE = (
    SOURCE_LOCK_DIR /
    "Stage9B3_Tan_SupplementNoEvidence32_NONSEED_PROJECT_IDENTITIES_LOCKED.tsv"
)

OUT = (
    ROOT /
    "06_publication" /
    "stage10_tables_figures" /
    "stage10C_biology"
)

TABLE_DIR = (
    OUT /
    "tables"
)

FIG_DIR = (
    OUT /
    "figures"
)

LOCK_DIR = (
    OUT /
    "locks"
)

for directory in [
    OUT,
    TABLE_DIR,
    FIG_DIR,
    LOCK_DIR
]:
    directory.mkdir(
        parents=True,
        exist_ok=True
    )


# ======================================================================================
# 2. INPUT EXISTENCE QC
# ======================================================================================

required_inputs = [
    RANK_FILE,
    NOEVIDENCE_FILE,
    TAN_XLSX
]

for path in required_inputs:

    assert path.exists(), (
        f"MISSING REQUIRED INPUT:\n{path}"
    )

print(
    "=" * 120
)

print(
    "ASD-CONTEXTFUSION — STAGE 10C"
)

print(
    "TARGETED BIOLOGICAL / GENETIC INTERPRETATION"
)

print(
    "=" * 120
)


# ======================================================================================
# 3. HELPERS
# ======================================================================================

def sha256_file(
    path,
    chunk_size=1024 * 1024
):

    h = hashlib.sha256()

    with open(
        path,
        "rb"
    ) as f:

        while True:

            chunk = f.read(
                chunk_size
            )

            if not chunk:
                break

            h.update(
                chunk
            )

    return h.hexdigest()


def clean_gene(x):

    if pd.isna(x):
        return np.nan

    x = str(
        x
    ).strip()

    if (
        x == ""
        or
        x.lower() == "nan"
    ):
        return np.nan

    return x


def find_column(
    df,
    candidates
):

    exact_map = {
        str(col).strip().lower():
            col

        for col
        in df.columns
    }

    for candidate in candidates:

        key = str(
            candidate
        ).lower()

        if key in exact_map:
            return exact_map[
                key
            ]

    for col in df.columns:

        col_low = str(
            col
        ).lower()

        for candidate in candidates:

            if (
                str(
                    candidate
                ).lower()
                in
                col_low
            ):
                return col

    return None


def save_figure(
    fig,
    stem
):

    pdf = (
        FIG_DIR /
        f"{stem}.pdf"
    )

    svg = (
        FIG_DIR /
        f"{stem}.svg"
    )

    png = (
        FIG_DIR /
        f"{stem}.png"
    )

    fig.savefig(
        pdf,
        bbox_inches="tight",
        pad_inches=0.05
    )

    fig.savefig(
        svg,
        bbox_inches="tight",
        pad_inches=0.05
    )

    fig.savefig(
        png,
        dpi=600,
        bbox_inches="tight",
        pad_inches=0.05
    )

    plt.close(
        fig
    )

    gc.collect()


# ======================================================================================
# 4. WRITE EXPLORATORY INTERPRETATION LOCK
#
# This is NOT claimed to have been prespecified before Stage 9 results.
# It is locked BEFORE independent literature curation.
# ======================================================================================

INTERPRETATION_LOCK = {

    "project":
        "ASD-ContextFusion",

    "stage":
        "Stage10C",

    "analysis_type":
        "targeted exploratory biological/genetic interpretation",

    "primary_external_set":
        "Tan 2026 source-locked NoEvidence32 represented nonseed set",

    "expected_project_gene_count":
        30,

    "candidate_selection_rule": {

        "quantity":
            "NTv2_rank_percentile - Context_rank_percentile",

        "eligibility":
            "delta > 0",

        "ordering": [
            "delta descending",
            "Gene_Name ascending"
        ],

        "maximum_candidates":
            10
    },

    "candidate_selection_timing":
        "locked before independent literature curation",

    "selection_does_not_use": [
        "literature evidence",
        "gene biological function",
        "TADA FDR magnitude",
        "DeNovoWEST FDR magnitude",
        "pLI",
        "Fusion rank"
    ],

    "interpretation":
        (
            "Exploratory sequence-reprioritized genes only; "
            "not newly discovered causal ASD genes."
        ),

    "prohibited_steps": [
        "model refitting",
        "hyperparameter tuning",
        "new PU sampling",
        "new embeddings",
        "external set modification",
        "pathway enrichment fishing"
    ]
}

INTERPRETATION_LOCK_FILE = (
    LOCK_DIR /
    "ASD_ContextFusion_STAGE10C_BIOLOGICAL_INTERPRETATION_LOCK.json"
)

with open(
    INTERPRETATION_LOCK_FILE,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        INTERPRETATION_LOCK,
        f,
        indent=2
    )

print(
    "\nInterpretation rule lock:"
)

print(
    INTERPRETATION_LOCK_FILE
)


# ======================================================================================
# 5. LOAD FROZEN STAGE 9C RANK TABLE
# ======================================================================================

ranks = pd.read_csv(
    RANK_FILE,
    sep="\t",
    compression="gzip",
    low_memory=False
)

print(
    "\nStage9C rank table shape:",
    ranks.shape
)

assert len(
    ranks
) == 18062, (
    f"Expected 18,062 rows; found {len(ranks)}"
)


GENE_COL = find_column(
    ranks,
    [
        "Gene_Name",
        "Gene"
    ]
)

assert GENE_COL is not None

ranks[
    GENE_COL
] = ranks[
    GENE_COL
].map(
    clean_gene
)

assert (
    ranks[
        GENE_COL
    ].nunique()
    == 18062
)


# ======================================================================================
# 6. RESOLVE SAVED RANK-PERCENTILE COLUMNS
# ======================================================================================

CONTEXT_PCT_COL = find_column(
    ranks,
    [
        "Context_rank_percentile"
    ]
)

NT_PCT_COL = find_column(
    ranks,
    [
        "NTv2_rank_percentile",
        "NT_rank_percentile"
    ]
)

FUSION_PCT_COL = find_column(
    ranks,
    [
        "StrictFusion_rank_percentile",
        "Fusion_rank_percentile"
    ]
)

assert CONTEXT_PCT_COL is not None
assert NT_PCT_COL is not None
assert FUSION_PCT_COL is not None

print(
    "\nSaved rank-percentile columns:"
)

print(
    "Context:",
    CONTEXT_PCT_COL
)

print(
    "NT-v2 :",
    NT_PCT_COL
)

print(
    "Fusion:",
    FUSION_PCT_COL
)


ranks[
    "Context_rank_percentile"
] = pd.to_numeric(
    ranks[
        CONTEXT_PCT_COL
    ],
    errors="coerce"
)

ranks[
    "NTv2_rank_percentile"
] = pd.to_numeric(
    ranks[
        NT_PCT_COL
    ],
    errors="coerce"
)

ranks[
    "Fusion_rank_percentile"
] = pd.to_numeric(
    ranks[
        FUSION_PCT_COL
    ],
    errors="coerce"
)


for col in [
    "Context_rank_percentile",
    "NTv2_rank_percentile",
    "Fusion_rank_percentile"
]:

    assert ranks[
        col
    ].notna().all()

    assert ranks[
        col
    ].between(
        0,
        100
    ).all()


# ======================================================================================
# 7. LOAD EXACT LOCKED STAGE 9B3 NoEvidence30 IDENTITY TABLE
#
# VERIFIED COLUMN STRUCTURE:
#   source_gene
#   source_group
#   project_scoring_gene
#   mapping_mode
#   project_gene_is_locked_ASD_seed
#
# IMPORTANT:
# We use the exact source_gene -> project_scoring_gene mapping already locked
# during Stage9B3. No alias guessing or reconstruction is needed.
# ======================================================================================

identity = pd.read_csv(
    NOEVIDENCE_FILE,
    sep="\t",
    low_memory=False
)

print(
    "\nLOCKED STAGE9B3 NoEvidence FILE:"
)

print(
    NOEVIDENCE_FILE
)

print(
    "Shape:",
    identity.shape
)

EXPECTED_ID_COLUMNS = [
    "source_gene",
    "source_group",
    "project_scoring_gene",
    "mapping_mode",
    "project_gene_is_locked_ASD_seed"
]

for col in EXPECTED_ID_COLUMNS:

    assert col in identity.columns, (
        f"Expected Stage9B3 column missing: {col}"
    )


identity[
    "source_gene"
] = identity[
    "source_gene"
].map(
    clean_gene
)

identity[
    "project_scoring_gene"
] = identity[
    "project_scoring_gene"
].map(
    clean_gene
)


# Hard QC
assert len(
    identity
) == 30

assert (
    identity[
        "source_gene"
    ].nunique()
    == 30
)

assert (
    identity[
        "project_scoring_gene"
    ].nunique()
    == 30
)


rank_gene_set = set(
    ranks[
        GENE_COL
    ]
)

locked_project_genes = set(
    identity[
        "project_scoring_gene"
    ]
)

assert locked_project_genes.issubset(
    rank_gene_set
)

assert len(
    locked_project_genes
) == 30


# No locked ASD seed should remain in a NONSEED validation table.
seed_flag_numeric = pd.to_numeric(
    identity[
        "project_gene_is_locked_ASD_seed"
    ],
    errors="coerce"
).fillna(
    0
)

assert (
    seed_flag_numeric
    == 0
).all()


print(
    "\n" +
    "=" * 120
)

print(
    "LOCKED NoEvidence30 PROJECT IDENTITY QC — PASS"
)

print(
    "=" * 120
)

print(
    "Locked source genes :",
    identity[
        "source_gene"
    ].nunique()
)

print(
    "Locked project genes:",
    identity[
        "project_scoring_gene"
    ].nunique()
)

print(
    "All project genes in Stage9C universe:",
    locked_project_genes.issubset(
        rank_gene_set
    )
)

print(
    "Locked ASD seeds remaining:",
    int(
        (
            seed_flag_numeric
            != 0
        ).sum()
    )
)

print(
    "\nScientific behavior changed: NO"
)

print(
    "External set redefined     : NO"
)


# ======================================================================================
# 8. EXTRACT FROZEN MODEL RANKS FOR EXACT NoEvidence30
# ======================================================================================

bio = (
    ranks.loc[
        ranks[
            GENE_COL
        ].isin(
            locked_project_genes
        ),
        [
            GENE_COL,
            "Context_rank_percentile",
            "NTv2_rank_percentile",
            "Fusion_rank_percentile"
        ]
    ]
    .copy()
    .rename(
        columns={
            GENE_COL:
                "Gene_Name"
        }
    )
)

assert len(
    bio
) == 30

assert (
    bio[
        "Gene_Name"
    ].nunique()
    == 30
)


bio[
    "NT_minus_Context_rank_percentile"
] = (
    bio[
        "NTv2_rank_percentile"
    ]
    -
    bio[
        "Context_rank_percentile"
    ]
)

bio[
    "Fusion_minus_Context_rank_percentile"
] = (
    bio[
        "Fusion_rank_percentile"
    ]
    -
    bio[
        "Context_rank_percentile"
    ]
)


# ======================================================================================
# 9. ADD LOCKED SOURCE IDENTITY MAPPING
# ======================================================================================

identity_map = (
    identity[
        [
            "source_gene",
            "project_scoring_gene",
            "source_group",
            "mapping_mode"
        ]
    ]
    .rename(
        columns={
            "project_scoring_gene":
                "Gene_Name"
        }
    )
)

bio = bio.merge(
    identity_map,
    on="Gene_Name",
    how="left",
    validate="one_to_one"
)

assert bio[
    "source_gene"
].notna().all()


# ======================================================================================
# 10. LOAD TAN 2026 SUPPLEMENTARY TABLE 9
#
# Previous source audit established:
#
#   Genesymbol = source gene
#   FDR.2      = TADA ASD FDR
#   FDR.3      = DeNovoWEST ASD-all FDR
#
# These values are used ONLY to annotate the already-locked 30 genes.
# They do NOT define or select the gene set here.
# ======================================================================================

tan = pd.read_excel(
    TAN_XLSX,
    sheet_name="Supplementary Table 9",
    header=3
)

print(
    "\nTan Supplementary Table 9 shape:",
    tan.shape
)

for required_col in [
    "Genesymbol",
    "FDR.2",
    "FDR.3"
]:

    assert required_col in tan.columns, (
        f"Expected Tan Table 9 column missing: {required_col}"
    )


tan[
    "source_gene"
] = tan[
    "Genesymbol"
].map(
    clean_gene
)

tan[
    "Tan_TADA_FDR"
] = pd.to_numeric(
    tan[
        "FDR.2"
    ],
    errors="coerce"
)

tan[
    "Tan_DeNovoWEST_ASD_all_FDR"
] = pd.to_numeric(
    tan[
        "FDR.3"
    ],
    errors="coerce"
)


# ======================================================================================
# 11. OPTIONAL pLI FROM TAN TABLE
# ======================================================================================

PLI_CANDIDATES = [
    "gnomAD_v4_lof.pLI",
    "gnomAD_v4_lof_pLI",
    "pLI"
]

PLI_COL = None

for candidate in PLI_CANDIDATES:

    if candidate in tan.columns:

        PLI_COL = candidate

        break


if PLI_COL is not None:

    tan[
        "gnomAD_v4_pLI"
    ] = pd.to_numeric(
        tan[
            PLI_COL
        ],
        errors="coerce"
    )

    print(
        "pLI column:",
        PLI_COL
    )

else:

    tan[
        "gnomAD_v4_pLI"
    ] = np.nan

    print(
        "pLI column not found; continuing without pLI."
    )


# ======================================================================================
# 12. EXTRACT TAN ANNOTATION USING EXACT LOCKED SOURCE GENES
#
# This is safer than manually applying aliases because Stage9B3 already contains
# the exact source_gene -> project_scoring_gene mapping.
# ======================================================================================

locked_source_genes = set(
    identity[
        "source_gene"
    ]
)

tan_locked = (
    tan.loc[
        tan[
            "source_gene"
        ].isin(
            locked_source_genes
        ),
        [
            "source_gene",
            "Tan_TADA_FDR",
            "Tan_DeNovoWEST_ASD_all_FDR",
            "gnomAD_v4_pLI"
        ]
    ]
    .copy()
)


# Duplicate protection
dup_source = tan_locked.loc[
    tan_locked[
        "source_gene"
    ].duplicated(
        keep=False
    )
]

if len(
    dup_source
) > 0:

    print(
        "\nDUPLICATE TAN SOURCE ROWS:"
    )

    print(
        dup_source.to_string(
            index=False
        )
    )

    raise RuntimeError(
        "Duplicate relevant source-gene rows found in Tan Table 9. "
        "Stopping rather than resolving post hoc."
    )


assert (
    tan_locked[
        "source_gene"
    ].nunique()
    == 30
), (
    f"Mapped only "
    f"{tan_locked['source_gene'].nunique()} / 30 "
    f"locked Tan source genes."
)


# ======================================================================================
# 13. MERGE TAN GENETIC EVIDENCE
# ======================================================================================

bio = bio.merge(
    tan_locked,
    on="source_gene",
    how="left",
    validate="one_to_one"
)

assert len(
    bio
) == 30


# ======================================================================================
# 14. DEFINE INDEPENDENT GENETIC-EVIDENCE ROUTE
#
# Original source-defined Tan union:
#
#   TADA FDR < 0.05
#       OR
#   DeNovoWEST ASD-all FDR < 0.05
#
# Here we simply annotate which evidence route supported each already-locked gene.
# ======================================================================================

bio[
    "TADA_supported"
] = (
    bio[
        "Tan_TADA_FDR"
    ] < 0.05
)

bio[
    "DeNovoWEST_supported"
] = (
    bio[
        "Tan_DeNovoWEST_ASD_all_FDR"
    ] < 0.05
)


def evidence_route(
    row
):

    tada = bool(
        row[
            "TADA_supported"
        ]
    )

    dnw = bool(
        row[
            "DeNovoWEST_supported"
        ]
    )

    if tada and dnw:
        return "Both"

    if tada:
        return "TADA only"

    if dnw:
        return "DeNovoWEST only"

    return "Neither"


bio[
    "Independent_genetic_evidence_route"
] = bio.apply(
    evidence_route,
    axis=1
)


neither = bio.loc[
    bio[
        "Independent_genetic_evidence_route"
    ] == "Neither"
]


if len(
    neither
) > 0:

    print(
        "\nGENES FAILING ORIGINAL TAN UNION RULE:"
    )

    print(
        neither[
            [
                "source_gene",
                "Gene_Name",
                "Tan_TADA_FDR",
                "Tan_DeNovoWEST_ASD_all_FDR"
            ]
        ].to_string(
            index=False
        )
    )

    raise RuntimeError(
        "At least one locked gene does not satisfy "
        "TADA FDR<0.05 OR DeNovoWEST FDR<0.05. "
        "Audit the Tan Table 9 header before proceeding."
    )


bio[
    "Minimum_genetic_FDR"
] = bio[
    [
        "Tan_TADA_FDR",
        "Tan_DeNovoWEST_ASD_all_FDR"
    ]
].min(
    axis=1,
    skipna=True
)

bio[
    "Minus_log10_minimum_genetic_FDR"
] = (
    -np.log10(
        bio[
            "Minimum_genetic_FDR"
        ].clip(
            lower=np.finfo(
                float
            ).tiny
        )
    )
)


# ======================================================================================
# 15. LOCK SEQUENCE-REPRIORITIZED CANDIDATES
#
# Candidate selection uses ONLY:
#
#   locked NoEvidence30 membership
#   frozen NT-v2 percentile
#   frozen Context percentile
#
# It does NOT use:
#   Tan FDR
#   pLI
#   Fusion score
#   literature
# ======================================================================================

sequence_candidates = (
    bio.loc[
        bio[
            "NT_minus_Context_rank_percentile"
        ] > 0
    ]
    .sort_values(
        [
            "NT_minus_Context_rank_percentile",
            "Gene_Name"
        ],
        ascending=[
            False,
            True
        ],
        kind="mergesort"
    )
    .head(
        10
    )
    .copy()
)


sequence_candidates.insert(
    0,
    "Sequence_reprioritization_rank",
    np.arange(
        1,
        len(
            sequence_candidates
        ) + 1
    )
)


CANDIDATE_LOCK_FILE = (
    LOCK_DIR /
    "Stage10C_LOCKED_sequence_reprioritized_NoEvidence_candidates.tsv"
)

sequence_candidates.to_csv(
    CANDIDATE_LOCK_FILE,
    sep="\t",
    index=False
)

CANDIDATE_SHA256 = sha256_file(
    CANDIDATE_LOCK_FILE
)


CANDIDATE_METADATA = {

    "candidate_file":
        str(
            CANDIDATE_LOCK_FILE
        ),

    "sha256":
        CANDIDATE_SHA256,

    "n_candidates":
        int(
            len(
                sequence_candidates
            )
        ),

    "selection_rule":
        (
            "NTv2_rank_percentile - Context_rank_percentile > 0; "
            "descending delta; Gene_Name ascending tie-break; maximum 10"
        ),

    "selection_used": [
        "locked NoEvidence30 membership",
        "frozen Context rank percentile",
        "frozen NT-v2 rank percentile"
    ],

    "selection_did_not_use": [
        "Tan TADA FDR",
        "Tan DeNovoWEST FDR",
        "pLI",
        "Fusion rank",
        "literature evidence",
        "biological function"
    ]
}


CANDIDATE_METADATA_FILE = (
    LOCK_DIR /
    "Stage10C_sequence_candidate_selection_SHA256_LOCK.json"
)

with open(
    CANDIDATE_METADATA_FILE,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        CANDIDATE_METADATA,
        f,
        indent=2
    )


# ======================================================================================
# 16. FINAL NoEvidence30 TABLE
# ======================================================================================

FINAL_COLS = [

    "Gene_Name",
    "source_gene",
    "source_group",
    "mapping_mode",

    "Context_rank_percentile",
    "NTv2_rank_percentile",
    "Fusion_rank_percentile",

    "NT_minus_Context_rank_percentile",
    "Fusion_minus_Context_rank_percentile",

    "Tan_TADA_FDR",
    "Tan_DeNovoWEST_ASD_all_FDR",

    "TADA_supported",
    "DeNovoWEST_supported",

    "Independent_genetic_evidence_route",

    "Minimum_genetic_FDR",
    "Minus_log10_minimum_genetic_FDR",

    "gnomAD_v4_pLI"
]


bio = bio[
    FINAL_COLS
].copy()


BIO_TABLE = (
    TABLE_DIR /
    "Table7_Tan2026_NoEvidence30_model_and_genetic_interpretation.tsv"
)

bio.to_csv(
    BIO_TABLE,
    sep="\t",
    index=False
)


# ======================================================================================
# 17. CREATE LITERATURE-CURATION TEMPLATE
#
# BLANK by design.
#
# Fill only AFTER candidate identities are locked.
# ======================================================================================

candidate_display_cols = [

    "Sequence_reprioritization_rank",

    "Gene_Name",
    "source_gene",

    "Context_rank_percentile",
    "NTv2_rank_percentile",
    "NT_minus_Context_rank_percentile",

    "Fusion_rank_percentile",

    "Independent_genetic_evidence_route",

    "Tan_TADA_FDR",
    "Tan_DeNovoWEST_ASD_all_FDR",

    "gnomAD_v4_pLI"
]


curation = sequence_candidates[
    candidate_display_cols
].copy()


curation[
    "Molecular_function"
] = ""

curation[
    "Neuronal_or_neurodevelopmental_evidence"
] = ""

curation[
    "Human_NDD_or_neurological_evidence"
] = ""

curation[
    "Relevant_brain_or_neuronal_expression_evidence"
] = ""

curation[
    "Key_independent_reference_PMID_or_DOI"
] = ""

curation[
    "Brief_evidence_summary"
] = ""

curation[
    "Curation_status"
] = "NOT_CURATED"


CURATION_FILE = (
    TABLE_DIR /
    "Stage10C_LOCKED_candidate_independent_literature_curation_template.tsv"
)

curation.to_csv(
    CURATION_FILE,
    sep="\t",
    index=False
)


# ======================================================================================
# 18. MANUSCRIPT SUMMARY
# ======================================================================================

summary_rows = []


route_counts = (
    bio[
        "Independent_genetic_evidence_route"
    ]
    .value_counts()
)


for route in [
    "TADA only",
    "DeNovoWEST only",
    "Both"
]:

    summary_rows.append(
        {
            "measure":
                f"Genetic evidence route: {route}",

            "value":
                int(
                    route_counts.get(
                        route,
                        0
                    )
                )
        }
    )


summary_rows.extend(
    [
        {
            "measure":
                "Represented locked NoEvidence genes",

            "value":
                30
        },

        {
            "measure":
                "Top 10% by Context",

            "value":
                int(
                    (
                        bio[
                            "Context_rank_percentile"
                        ] >= 90
                    ).sum()
                )
        },

        {
            "measure":
                "Top 10% by NT-v2",

            "value":
                int(
                    (
                        bio[
                            "NTv2_rank_percentile"
                        ] >= 90
                    ).sum()
                )
        },

        {
            "measure":
                "Top 10% by Fusion",

            "value":
                int(
                    (
                        bio[
                            "Fusion_rank_percentile"
                        ] >= 90
                    ).sum()
                )
        },

        {
            "measure":
                "NT-v2 ranked higher than Context",

            "value":
                int(
                    (
                        bio[
                            "NT_minus_Context_rank_percentile"
                        ] > 0
                    ).sum()
                )
        },

        {
            "measure":
                "Locked sequence-reprioritized candidates",

            "value":
                int(
                    len(
                        sequence_candidates
                    )
                )
        }
    ]
)


summary_df = pd.DataFrame(
    summary_rows
)


SUMMARY_FILE = (
    TABLE_DIR /
    "Stage10C_manuscript_biology_summary.tsv"
)

summary_df.to_csv(
    SUMMARY_FILE,
    sep="\t",
    index=False
)


# ======================================================================================
# 19. EXCEL WORKBOOK
# ======================================================================================

EXCEL_FILE = (
    TABLE_DIR /
    "ASD_ContextFusion_Stage10C_Biological_Interpretation.xlsx"
)


with pd.ExcelWriter(
    EXCEL_FILE,
    engine="openpyxl"
) as writer:

    bio.sort_values(
        [
            "Minimum_genetic_FDR",
            "Gene_Name"
        ],
        ascending=[
            True,
            True
        ]
    ).to_excel(
        writer,
        sheet_name="NoEvidence30",
        index=False
    )

    sequence_candidates.to_excel(
        writer,
        sheet_name="SequenceReprioritized",
        index=False
    )

    curation.to_excel(
        writer,
        sheet_name="LiteratureCuration",
        index=False
    )

    summary_df.to_excel(
        writer,
        sheet_name="Summary",
        index=False
    )


# ======================================================================================
# 20. PUBLICATION FIGURE STYLE
# ======================================================================================

available_fonts = {
    f.name
    for f in fm.fontManager.ttflist
}

FONT = (
    "Arial"
    if "Arial" in available_fonts
    else "DejaVu Sans"
)


mpl.rcParams.update(
    {
        "font.family":
            FONT,

        "font.size":
            9,

        "axes.titlesize":
            10.5,

        "axes.labelsize":
            9.5,

        "xtick.labelsize":
            8,

        "ytick.labelsize":
            8,

        "legend.fontsize":
            8,

        "axes.spines.top":
            False,

        "axes.spines.right":
            False,

        "pdf.fonttype":
            42,

        "svg.fonttype":
            "none",

        "figure.facecolor":
            "white",

        "axes.facecolor":
            "white"
    }
)


COLOR_CONTEXT = "#0072B2"
COLOR_NT = "#D55E00"
COLOR_FUSION = "#009E73"


def panel_label(
    ax,
    label
):

    ax.text(
        -0.12,
        1.06,
        label,
        transform=ax.transAxes,
        fontsize=12,
        fontweight="bold",
        va="top"
    )


# ======================================================================================
# 21. FIGURE 6
#
# A = all locked NoEvidence30 genes
# B = sequence-reprioritized candidates
# C = independent Tan genetic support
# ======================================================================================

fig = plt.figure(
    figsize=(
        13.5,
        9.0
    )
)


gs = fig.add_gridspec(
    2,
    2,
    height_ratios=[
        1.10,
        0.90
    ],
    width_ratios=[
        1.05,
        0.95
    ],
    hspace=0.42,
    wspace=0.42
)


axA = fig.add_subplot(
    gs[:, 0]
)

axB = fig.add_subplot(
    gs[0, 1]
)

axC = fig.add_subplot(
    gs[1, 1]
)


# --------------------------------------------------------------------------------------
# PANEL A
# All 30 genes sorted by independent genetic FDR
# --------------------------------------------------------------------------------------

plot_all = (
    bio.sort_values(
        [
            "Minimum_genetic_FDR",
            "Gene_Name"
        ],
        ascending=[
            True,
            True
        ],
        kind="mergesort"
    )
    .reset_index(
        drop=True
    )
)


yA = np.arange(
    len(
        plot_all
    )
)[::-1]


for yi, (_, row) in zip(
    yA,
    plot_all.iterrows()
):

    values = [

        float(
            row[
                "Context_rank_percentile"
            ]
        ),

        float(
            row[
                "NTv2_rank_percentile"
            ]
        ),

        float(
            row[
                "Fusion_rank_percentile"
            ]
        )
    ]

    axA.plot(
        values,
        [
            yi,
            yi,
            yi
        ],
        color="#D7D7D7",
        linewidth=0.8,
        zorder=1
    )

    axA.scatter(
        values[
            0
        ],
        yi,
        s=26,
        color=COLOR_CONTEXT,
        zorder=3
    )

    axA.scatter(
        values[
            1
        ],
        yi,
        s=26,
        color=COLOR_NT,
        zorder=3
    )

    axA.scatter(
        values[
            2
        ],
        yi,
        s=26,
        color=COLOR_FUSION,
        zorder=3
    )


axA.axvline(
    90,
    linestyle="--",
    linewidth=0.9,
    color="#777777"
)

axA.set_yticks(
    yA
)

axA.set_yticklabels(
    plot_all[
        "Gene_Name"
    ]
)

axA.set_xlim(
    0,
    101
)

axA.set_xlabel(
    "Genome-wide non-seed rank percentile"
)

axA.set_title(
    "Model prioritization of Tan 2026 NoEvidence genes"
)

axA.grid(
    axis="x",
    linewidth=0.45,
    color="#EAEAEA"
)

axA.legend(
    handles=[
        Line2D(
            [0],
            [0],
            marker="o",
            linestyle="none",
            color=COLOR_CONTEXT,
            label="Context"
        ),
        Line2D(
            [0],
            [0],
            marker="o",
            linestyle="none",
            color=COLOR_NT,
            label="NT-v2"
        ),
        Line2D(
            [0],
            [0],
            marker="o",
            linestyle="none",
            color=COLOR_FUSION,
            label="Fusion"
        )
    ],
    frameon=False,
    loc="lower right"
)

panel_label(
    axA,
    "A"
)


# --------------------------------------------------------------------------------------
# PANEL B
# Sequence reprioritization
# --------------------------------------------------------------------------------------

cand_plot = (
    sequence_candidates.sort_values(
        [
            "NT_minus_Context_rank_percentile",
            "Gene_Name"
        ],
        ascending=[
            True,
            True
        ]
    )
    .reset_index(
        drop=True
    )
)


yB = np.arange(
    len(
        cand_plot
    )
)


for yi, (_, row) in zip(
    yB,
    cand_plot.iterrows()
):

    context = float(
        row[
            "Context_rank_percentile"
        ]
    )

    nt = float(
        row[
            "NTv2_rank_percentile"
        ]
    )

    delta = float(
        row[
            "NT_minus_Context_rank_percentile"
        ]
    )

    axB.plot(
        [
            context,
            nt
        ],
        [
            yi,
            yi
        ],
        color="#B8B8B8",
        linewidth=1.5
    )

    axB.scatter(
        context,
        yi,
        s=40,
        color=COLOR_CONTEXT,
        zorder=3
    )

    axB.scatter(
        nt,
        yi,
        s=40,
        color=COLOR_NT,
        zorder=3
    )

    axB.text(
        min(
            nt + 1.0,
            100
        ),
        yi,
        f"+{delta:.1f}",
        va="center",
        fontsize=7.5,
        color="#444444"
    )


axB.set_yticks(
    yB
)

axB.set_yticklabels(
    cand_plot[
        "Gene_Name"
    ]
)

axB.set_xlim(
    0,
    103
)

axB.set_xlabel(
    "Rank percentile"
)

axB.set_title(
    "Locked NT-v2 sequence reprioritization"
)

axB.grid(
    axis="x",
    linewidth=0.45,
    color="#EAEAEA"
)

axB.legend(
    handles=[
        Line2D(
            [0],
            [0],
            marker="o",
            linestyle="none",
            color=COLOR_CONTEXT,
            label="Context"
        ),
        Line2D(
            [0],
            [0],
            marker="o",
            linestyle="none",
            color=COLOR_NT,
            label="NT-v2"
        )
    ],
    frameon=False,
    loc="lower right"
)

panel_label(
    axB,
    "B"
)


# --------------------------------------------------------------------------------------
# PANEL C
# Independent genetic support
# --------------------------------------------------------------------------------------

for yi, (_, row) in zip(
    yB,
    cand_plot.iterrows()
):

    tada = row[
        "Tan_TADA_FDR"
    ]

    dnw = row[
        "Tan_DeNovoWEST_ASD_all_FDR"
    ]

    if pd.notna(
        tada
    ):

        axC.scatter(
            -np.log10(
                max(
                    float(
                        tada
                    ),
                    np.finfo(
                        float
                    ).tiny
                )
            ),
            yi + 0.10,
            marker="o",
            s=40,
            color="#0072B2",
            zorder=3
        )

    if pd.notna(
        dnw
    ):

        axC.scatter(
            -np.log10(
                max(
                    float(
                        dnw
                    ),
                    np.finfo(
                        float
                    ).tiny
                )
            ),
            yi - 0.10,
            marker="s",
            s=40,
            color="#D55E00",
            zorder=3
        )


FDR05_LINE = (
    -np.log10(
        0.05
    )
)

axC.axvline(
    FDR05_LINE,
    linestyle="--",
    linewidth=0.9,
    color="#555555"
)

axC.set_yticks(
    yB
)

axC.set_yticklabels(
    cand_plot[
        "Gene_Name"
    ]
)

axC.set_xlabel(
    "−log10(FDR)"
)

axC.set_title(
    "Independent Tan 2026 genetic support"
)

axC.grid(
    axis="x",
    linewidth=0.45,
    color="#EAEAEA"
)

axC.legend(
    handles=[
        Line2D(
            [0],
            [0],
            marker="o",
            linestyle="none",
            color="#0072B2",
            label="TADA"
        ),
        Line2D(
            [0],
            [0],
            marker="s",
            linestyle="none",
            color="#D55E00",
            label="DeNovoWEST ASD-all"
        )
    ],
    frameon=False,
    loc="lower right"
)

panel_label(
    axC,
    "C"
)


fig.suptitle(
    (
        "Biological and genetic interpretation of post-benchmark "
        "ASD genes with no prior evidence"
    ),
    fontsize=12,
    fontweight="bold",
    y=0.995
)


FIG6_STEM = (
    "Figure6_Tan2026_NoEvidence_Biological_Genetic_Interpretation"
)

save_figure(
    fig,
    FIG6_STEM
)


# ======================================================================================
# 22. SUPPLEMENTARY FIGURE S5
# NT-v2 minus Context rank-percentile delta for all NoEvidence30
# ======================================================================================

all_delta = (
    bio.sort_values(
        [
            "NT_minus_Context_rank_percentile",
            "Gene_Name"
        ],
        ascending=[
            True,
            True
        ],
        kind="mergesort"
    )
    .reset_index(
        drop=True
    )
)


fig, ax = plt.subplots(
    figsize=(
        8.2,
        8.3
    )
)


yS = np.arange(
    len(
        all_delta
    )
)


bar_colors = [

    (
        COLOR_NT
        if value > 0
        else "#888888"
    )

    for value
    in all_delta[
        "NT_minus_Context_rank_percentile"
    ]
]


ax.barh(
    yS,
    all_delta[
        "NT_minus_Context_rank_percentile"
    ],
    color=bar_colors,
    height=0.68
)

ax.axvline(
    0,
    linewidth=0.9,
    color="#222222"
)

ax.set_yticks(
    yS
)

ax.set_yticklabels(
    all_delta[
        "Gene_Name"
    ]
)

ax.set_xlabel(
    "NT-v2 − Context rank percentile"
)

ax.set_title(
    "Sequence-driven reprioritization across Tan NoEvidence30"
)

ax.grid(
    axis="x",
    linewidth=0.45,
    color="#EAEAEA"
)

save_figure(
    fig,
    "Supplementary_Figure_S5_NoEvidence30_NTv2_minus_Context"
)


# ======================================================================================
# 23. SHA256 MANIFEST
# ======================================================================================

files_to_hash = [

    INTERPRETATION_LOCK_FILE,
    CANDIDATE_LOCK_FILE,
    CANDIDATE_METADATA_FILE,

    BIO_TABLE,
    CURATION_FILE,
    SUMMARY_FILE,
    EXCEL_FILE,

    FIG_DIR /
    f"{FIG6_STEM}.pdf",

    FIG_DIR /
    f"{FIG6_STEM}.png",

    FIG_DIR /
    "Supplementary_Figure_S5_NoEvidence30_NTv2_minus_Context.pdf"
]


hash_rows = []


for path in files_to_hash:

    assert path.exists()

    hash_rows.append(
        {
            "file":
                str(
                    path
                ),

            "sha256":
                sha256_file(
                    path
                )
        }
    )


HASH_MANIFEST = (
    OUT /
    "Stage10C_SHA256_manifest.tsv"
)


pd.DataFrame(
    hash_rows
).to_csv(
    HASH_MANIFEST,
    sep="\t",
    index=False
)


# ======================================================================================
# 24. FINAL REPORT
# ======================================================================================

print(
    "\n" +
    "=" * 120
)

print(
    "ASD-CONTEXTFUSION — STAGE 10C COMPLETE [OK]"
)

print(
    "=" * 120
)


print(
    "\nLOCKED NoEvidence genes:",
    len(
        bio
    )
)


print(
    "\nGENETIC EVIDENCE ROUTES:"
)

print(
    bio[
        "Independent_genetic_evidence_route"
    ]
    .value_counts()
    .to_string()
)


print(
    "\nTOP-10% RECOVERY CHECK:"
)

print(
    "Context:",
    int(
        (
            bio[
                "Context_rank_percentile"
            ] >= 90
        ).sum()
    ),
    "/ 30"
)

print(
    "NT-v2 :",
    int(
        (
            bio[
                "NTv2_rank_percentile"
            ] >= 90
        ).sum()
    ),
    "/ 30"
)

print(
    "Fusion:",
    int(
        (
            bio[
                "Fusion_rank_percentile"
            ] >= 90
        ).sum()
    ),
    "/ 30"
)


print(
    "\nNT-v2 > Context genes:",
    int(
        (
            bio[
                "NT_minus_Context_rank_percentile"
            ] > 0
        ).sum()
    )
)


print(
    "\n" +
    "=" * 120
)

print(
    "LOCKED SEQUENCE-REPRIORITIZED CANDIDATES"
)

print(
    "=" * 120
)


PRINT_COLUMNS = [

    "Sequence_reprioritization_rank",

    "Gene_Name",
    "source_gene",

    "Context_rank_percentile",
    "NTv2_rank_percentile",
    "NT_minus_Context_rank_percentile",

    "Fusion_rank_percentile",

    "Independent_genetic_evidence_route",

    "Tan_TADA_FDR",
    "Tan_DeNovoWEST_ASD_all_FDR",

    "gnomAD_v4_pLI"
]


print(
    sequence_candidates[
        PRINT_COLUMNS
    ].to_string(
        index=False
    )
)


print(
    "\nCandidate lock SHA256:"
)

print(
    CANDIDATE_SHA256
)


print(
    "\nOUTPUT DIRECTORY:"
)

print(
    OUT
)


print(
    "\nKEY FILES:"
)

print(
    "Full NoEvidence30 table:"
)

print(
    BIO_TABLE
)

print(
    "\nLocked candidate table:"
)

print(
    CANDIDATE_LOCK_FILE
)

print(
    "\nLiterature-curation template:"
)

print(
    CURATION_FILE
)

print(
    "\nExcel workbook:"
)

print(
    EXCEL_FILE
)

print(
    "\nFigure 6:"
)

print(
    FIG_DIR /
    f"{FIG6_STEM}.pdf"
)

print(
    "\nSHA manifest:"
)

print(
    HASH_MANIFEST
)


print(
    "\nSCIENTIFIC INTEGRITY:"
)

print(
    "Model fitting                  : NO"
)

print(
    "Hyperparameter tuning          : NO"
)

print(
    "PU resampling                  : NO"
)

print(
    "New embeddings                 : NO"
)

print(
    "External gene-set redefinition : NO"
)

print(
    "Literature used for selection  : NO"
)

print(
    "GO/pathway enrichment          : NO"
)


print(
    "\nNEXT:"
)

print(
    "Paste back the complete "
    "'LOCKED SEQUENCE-REPRIORITIZED CANDIDATES' table."
)

print(
    "=" * 120
)
```
